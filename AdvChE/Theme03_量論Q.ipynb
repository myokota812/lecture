## 1. 設計した反応系

反応数4、成分数5（A, B, C, D, E）とする。

$$
\left\{
\begin{aligned}
A+B &\rightarrow C \\
C &\rightarrow D \\
D &\rightarrow E \\
A+B &\rightarrow E
\end{aligned}
\right.
$$

この反応系には

### 逐次反応

$$
A+B \rightarrow C \rightarrow D \rightarrow E
$$

### 総括反応

$$
A+B \rightarrow E
$$

が含まれている。

倍率反応は含まれていない。

## 3. 行列の Rank

反応①〜③を加える。

$$
(A+B\rightarrow C)
+
(C\rightarrow D)
+
(D\rightarrow E)
$$

左辺と右辺に現れる中間体

$$
C,\;D
$$

を消去すると

$$
A+B\rightarrow E
$$

となる。

したがって

$$
\boxed{
\nu_4=\nu_1+\nu_2+\nu_3
}
$$

である。

一方，

$$
\nu_1,\quad \nu_2,\quad \nu_3
$$

は互いに線形独立である。

よって

$$
\boxed{
\operatorname{rank}(\nu)=3
}
$$
