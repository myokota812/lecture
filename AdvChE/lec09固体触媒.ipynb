## 1. 固定床触媒反応器（PBR）
流体が触媒充填層を通過しながら反応する。

## 2. PBR設計方程式
$$\frac{dF_A}{dW}=r_A'$$

$$W=F_{A0}\int_0^X \frac{dX}{-r_A'}$$

## 3. 触媒重量基準速度
$$-r_A'=kC_A$$

単位例
$$mol\,kg_{cat}^{-1}s^{-1}$$

## 4. 一次反応
$$W=\frac{F_{A0}}{kC_{A0}}\ln\frac{1}{1-X}$$

## 5. 圧力降下
Ergun式
$$\frac{dP}{dz}=-\left[\frac{150(1-\varepsilon)^2\mu u}{\varepsilon^3d_p^2}+\frac{1.75(1-\varepsilon)\rho u^2}{\varepsilon^3d_p}\right]$$

## 6. 複合反応への適用
$$A\rightarrow B\rightarrow C$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

## 7. PBRの支配方程式
$$\frac{dF_A}{dW}=-r_1$$
$$\frac{dF_B}{dW}=r_1-r_2$$
$$\frac{dF_C}{dW}=r_2$$

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
k1=1.0;k2=0.3
def pbr(W,y):
 CA,CB,CC=y
 return [-k1*CA,k1*CA-k2*CB,k2*CB]
sol=solve_ivp(pbr,[0,10],[1,0,0],dense_output=True)
W=np.linspace(0,10,300)
CA,CB,CC=sol.sol(W)
plt.plot(W,CA,label="A")
plt.plot(W,CB,label="B")
plt.plot(W,CC,label="C")
plt.legend();plt.grid();plt.show()
```

## 8. PFRとの比較
PFR:
$$V=F_{A0}\int_0^X\frac{dX}{-r_A}$$

PBR:
$$W=F_{A0}\int_0^X\frac{dX}{-r_A'}$$

## 解答
