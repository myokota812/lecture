## 4. 逐次反応
$$A\rightarrow B\rightarrow C$$

目的生成物Bの収率最大化を考える。

```python
import numpy as np
k1=1.0
k2=0.3
CA0=1.0
t=np.linspace(0,10,1000)
CB=k1*CA0/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
idx=np.argmax(CB)
print("Optimal time =",t[idx])
print("Maximum CB =",CB[idx])
```

## 6. リサイクル系の最適化
$$X_o=\frac{(1+R)X_s}{1+RX_s}$$

Rを設計変数として最適化する。

```python
import numpy as np
R=np.linspace(0,10,200)
Xs=0.4
Xo=(1+R)*Xs/(1+R*Xs)
print("Best conversion in range =",Xo.max())
```

## 8. Pythonによる最適化例

```python
import numpy as np
from scipy.optimize import minimize_scalar

def objective(t):
    k1=1.0;k2=0.3
    CB=k1/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
    return -CB
res=minimize_scalar(objective,bounds=(0,10),method="bounded")
print(res.x,-res.fun)
```
