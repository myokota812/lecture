# <span style="color:rgb(213,80,0)">1. 固体触媒反応速度式</span>

液相均一反応の場合、反応速度式は、 $-r_{\mathrm{A}} =kC_{\mathrm{A}}$ のように表せましたが、固体触媒表面での反応（不均一系反応）については、別の取り扱いをします。様々なモデルが提案されておりますが、ここでは Langmuir–Hinshelwood（L–H）速度式を対象とします。

#### ① 発想：反応は触媒表面で起こる

不均一系触媒反応では、反応は触媒表面の活性点で進行する（とします）。活性点を"\*"で表すことにします。例えば　 $\mathbf{A}+\mathbf{B}\to \mathbf{C}$ 　の場合、

気相分子 A, B はまず表面に吸着する

&#160;&#160;&#160;&#160; A(g) + \* ↔ A\*

&#160;&#160;&#160;&#160; B(g) + \* ↔ B\*

A\*が吸着した状態です。吸着した分子同士が反応し、吸着生成物(C\*)ができます。

&#160;&#160;&#160;&#160; A\* + B\* → C\*

通常この仮定が律速となります。

吸着生成物C\*が離脱して気体生成物Cができあがります。

&#160;&#160;&#160;&#160; C\* → C(g) + \*

② Langmuirの仮定（競合吸着）

1. 単層吸着（1つのサイトに1つの分子）
2. サイトはすべて等価
3. 吸着分子同士の相互作用なし
4. 平衡吸着

## 吸着平衡

例：A の吸着

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; A(g)+∗⇌A∗

平衡定数 $K_A$ を用いる,表面被覆率 ( $\theta_A$ ) は

$$\theta_A =\frac{K_A C_A }{(1+K_A C_A +K_B C_B +K_C C_C )}$$

同様に

$$\theta_B =\frac{K_B C_B }{(1+K_A C_A +K_B C_B +K_C C_C )}$$

空のサイト（反応サイト）

$$\theta_* =\frac{1}{(1+K_A C_A +K_B C_B +K_C C_C )}$$

\-\-\-

#### 律速段階を設定

反応が A と B の吸着種の反応で律速なら、：

&#160;&#160;&#160;&#160;&#160;
$$r=k\theta_A \theta_B$$

水素化のように H₂ が解離吸着する場合は

&#160;&#160;&#160;&#160;&#160;
$$r=k\theta_A (\theta_H )^3$$

これが表面反応律速の基本です。

\-\-\-

#### 被覆率を濃度で表し速度式に代入

吸着等温式を導入して

$$\begin{array}{l} \theta_A =\frac{K_A C_A }{D}\newline \theta_H =\frac{K_H C_H }{D}\newline  \end{array}$$

ただし、 $D=1+K_A C_A +K_H C_H +K_C C_C$

#### *反応速度（律速が A* *+ 3H* → C\* の場合）

$$r=k\left(\frac{K_A C_A }{D}\right){\left(\frac{K_H C_H }{D}\right)}^3$$

整理すると

$$r=\frac{k(K_A C_A )(K_H C_H )^3 }{(1+K_H C_H +K_A C_A +K_C C_C )^4 }$$

これが**Langmuir–Hinshelwood（L–H）**です。

#### L–H 式の物理的意味

\- 分子＝吸着種が出会う確率

\- 分母＝表面の奪い合い（競合吸着）

\- 温度上昇→吸着弱化→速度低下する場合もある。

# <span style="color:rgb(213,80,0)">2. 触媒充填層反応器（PBR)設計</span>

## 目的
- 触媒充填層反応器（Packed Bed Reactor, PBR）とは何か
- なぜ「非等温」「非等圧」を考える必要があるのか
- 物質収支・エネルギー収支・圧力損失（Ergun式）がどのようにつながるのか

## 1. 対象とする反応系

$$\mathrm{C_6 H_6 +3H_2 \to C_6 H_{12} }$$

- A：ベンゼン
- B：水素
- C：シクロヘキサン

## 2. 反応器
- 気相
- 触媒充填層反応器（PBR）
- 定常状態
- 軸方向一次元モデル
- 非等温、非等圧

教科書的にはPBR モデルでは：

\- 温度一定（等温）

\- 圧力一定（等圧）

と仮定することが多い。しかし実際には：

\- 水素化反応は\*\*強い発熱反応\*\*

\- 充填層では\*\*摩擦により圧力が低下\*\*

するため、

$$T=T(z),~~P=P(z)$$

を同時に解く必要がある。

### 3. 独立変数としての触媒重量 W

PBR では以下の関係が成り立つ：

$$W=\rho_b Az$$

- $W$ ：触媒質量 \[kg\]
- $\rho_b$ ：触媒層見かけ密度 \[kg/m³\]
- $A$ ：反応器断面積 \[m²\]

PBR設計式は 座標 zではなく、W 基準で解く。

## 4. 状態変数（未知量）

このモデルで同時に求める量：

- $x_A (W)$ ベンゼン転化率
- $T(W)$  反応器内温度
- $P(W)$  反応器内圧力

## 5. 物質量流速(転化率表現）

$$x_A =\frac{F_{A0} -F_A }{F_{A0} }$$

より、

$$\begin{array}{l} F_A =F_{A0} (1-x_A )\newline F_B =F_{B0} -3F_{A0} x_A \newline F_C =F_{C0} +F_{A0} x_A  \end{array}$$

## 6. 濃度計算（非等圧・非等温）

理想気体を仮定して、状態方程式より：

$$c_{tot} =\frac{P}{RT}$$

$$C_i =y_i c_{tot}$$

 圧力・温度が変わると\*\*反応速度も変わる\*\*。

## 7. 反応速度式（Langmuir–Hinshelwood）

冒頭で説明した触媒表面反応律速モデル：

$$-r_A^{\prime } =\frac{k(T)K_H^3 K_A C_H^3 C_A }{(1+K_H C_H +K_A C_A +K_C C_C )^4 }$$

\- 分子の吸着競合を表現

\- 定数は Arrhenius 型で温度依存

## 8. PBR の設計方程式（物質収支）

$$\frac{dx_A }{dW}=\frac{-r_A^{\prime } }{F_{A0} }$$

(導出）



PBRモデル図（簡単のため成分Aのみ表記）

 **PBR では長さ** $z$ **ではなく触媒質量** $W$ **を独立変数にします。**

触媒層の幾何学量には、**見かけ（バルク）密度** $\rho_{\mathrm{b}}$ と断面積 $A$ を使い、微小長さ $dz$ と微小質量 $dW$ の間に

$$dW=\rho_{\mathrm{b}} Adz$$

の関係があります（ $\rho_{\mathrm{b}}$ は空隙を含む見かけ密度）。

\-\-\-

反応器内の微小触媒重量 $dW$ に対して、定常における物質：

&#160;&#160;&#160;&#160;&#160; 流入 \- 流出 + 生成(消費) = 0

$$F_{\mathrm{A}} (W)-F_{\mathrm{A}} (W+dW)+r_{\mathrm{A}} \prime dW=0$$

- $F_{\mathrm{A}} (W)$ ：累積触媒量 $W$ (入口からの位置にもなる） での A のモル流量（mol/s）
- $r_{\mathrm{A}} \prime$ ：**触媒基準反応速度**（mol/kg\-cat/s）：触媒質量単位での反応速度であり、通常の反応速度と区別するために'プライム'を記号に付しております。

これを変形して

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$\frac{F_{\mathrm{A}} (W+dW)-F_{\mathrm{A}} (W)}{dW}=r_{\mathrm{A}} \prime$$

 $dW\to 0$ の極限をとると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$\frac{dF_{\mathrm{A}} }{dW}=r_{\mathrm{A}} \prime$$

これが設計PBRの設計方程式になる。

転化率 $X$ を使う場合は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$F_{\mathrm{A}} (W)=F_{\mathrm{A0}} (1-X)$$

これを $W$ で微分すると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$\frac{dF_{\mathrm{A}} }{dW}=-F_{\mathrm{A0}} \frac{dX}{dW}$$

従って設計方程式は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$-F_{\mathrm{A0}} \frac{dX}{dW}=r_{\mathrm{A}} \prime$$

## 9. エネルギー収支（非等温）

$$\frac{dT}{dW}=\frac{-\Delta Hr_A^{\prime } -\frac{Ua}{\rho_b }(T-T_a )}{F_{tot} C_p }$$

\- 第1項：反応発熱

\- 第2項：外部熱交換

a：単位体積あたり伝熱面積 \[m²/m³\]

## 10. 圧力損失（Ergun 式）

これは前回説明しました

$$\frac{\Delta P}{L}=\frac{150(1-\varepsilon )^2 \,\mu \,u}{\varepsilon^3 \,d_p^2 }\;+\;\frac{1.75(1-\varepsilon )\,\rho \,u^2 }{\varepsilon^3 \,d_p }$$

の微分系を使います。

$$\frac{dP}{dz}=-\left\lbrack \frac{150\mu (1-\varepsilon )^2 }{\varepsilon^3 d_p^2 }u+\frac{1.75(1-\varepsilon )}{\varepsilon^3 d_p }\rho_g u^2 \right\rbrack$$

$$\frac{dP}{dW}=\frac{1}{\rho_b A}\frac{dP}{dz}$$

(空隙率 ε補足）\-\-\-\-\-\-

空隙率 ε（ベッド空隙率）は、

ε=空隙体積/充填層全体積

充填層の体積を $V$ とすると：

- 固体触媒の体積： $V_{\mathrm{s}} =(1-\varepsilon )V$
- 触媒質量： $W=\rho_{\mathrm{p}} V_{\mathrm{s}} =\rho_{\mathrm{p}} (1-\varepsilon )V$

よって、**みかけ密度（バルク密度）** は

$$\rho_{\mathrm{b}} =\frac{W}{V}=\rho_{\mathrm{p}} (1-\varepsilon )$$

 **具体例**

粒子密度（固体） $\rho_{\mathrm{p}}$ =2500 $\mathrm{kg}/{\mathrm{m}}^3$

空隙率 $\varepsilon$ = 0.40 の場合、

 $\rho_b$ =(1−0.40)×2500=1500 kg/m3

この $\rho_{\mathrm{b}}$ が、PBR 設計式

$$dW=\rho_b A\,dz$$

にそのまま使われます。


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# ==========================================================
# 非等温・非等圧 PBR
# Ergun圧力損失
# 外部熱交換
# Langmuir-Hinshelwood速度式
# ==========================================================

# ----------------------------------------------------------
# 反応管
# ----------------------------------------------------------

D = 0.05
L = 0.1

A = np.pi * (D / 2) ** 2

# ----------------------------------------------------------
# 触媒充填層
# ----------------------------------------------------------

rho_p = 2500.0
epsilon = 0.30
dp = 3e-3

rho_b = (1.0 - epsilon) * rho_p
W_total = A * L * rho_b

# ----------------------------------------------------------
# 操作条件
# ----------------------------------------------------------

R = 8.314

P0 = 126.7e3
T0 = 398.15

FA0 = 2.48e-3
FB0 = 7.44e-2
FC0 = 0.0

U = 75.4
a = 4.0 / D

Ta = 373.15

DeltaH = -206000.0

Cp_mol = 32.5
mu = 2e-5

# ----------------------------------------------------------
# 分子量
# ----------------------------------------------------------

MWA = 78.11e-3
MWB = 2.016e-3
MWC = 84.16e-3

params = dict(
    R=R,
    FA0=FA0,
    FB0=FB0,
    FC0=FC0,
    A=A,
    rho_b=rho_b,
    U=U,
    a=a,
    Ta=Ta,
    DeltaH=DeltaH,
    Cp_mol=Cp_mol,
    epsilon=epsilon,
    dp=dp,
    mu=mu,
    MWA=MWA,
    MWB=MWB,
    MWC=MWC
)

# ==========================================================
# Langmuir-Hinshelwood速度式
# ==========================================================

def rateLH(T, CA, CH, CC):

    k = 3.214e6 * np.exp(-6093.0 / T)

    KH = 3.88e-9 * np.exp(7805.0 / T)
    KA = 3.26e-7 * np.exp(5640.0 / T)
    KC = 1.84e-6 * np.exp(4481.0 / T)

    denom = 1.0 + KH * CH + KA * CA + KC * CC
    denom = max(denom, 1e-20)

    rA_abs = (
        k
        * KH**3
        * KA
        * CH**3
        * CA
        / denom**4
    )

    return rA_abs


# ==========================================================
# ODE
# ==========================================================

def rhs_noniso_pbr_ergun(W, y):

    xA, T, P = y

    T = np.clip(T, 200.0, 2000.0)
    P = max(P, 1e3)

    # --------------------------------------
    # 化学量論
    # --------------------------------------

    FA = params["FA0"] * (1.0 - xA)

    FB = params["FB0"] - 3.0 * params["FA0"] * xA
    FB = max(FB, 1e-16)

    FC = params["FC0"] + params["FA0"] * xA

    Ftot = FA + FB + FC

    # --------------------------------------
    # 濃度
    # --------------------------------------

    cTot = P / (params["R"] * T)

    yA = FA / max(Ftot, np.finfo(float).eps)
    yH = FB / max(Ftot, np.finfo(float).eps)
    yC = FC / max(Ftot, np.finfo(float).eps)

    CA = yA * cTot
    CH = yH * cTot
    CC = yC * cTot

    # --------------------------------------
    # 反応速度
    # --------------------------------------

    rA_abs = rateLH(T, CA, CH, CC)

    # --------------------------------------
    # 熱収支
    # --------------------------------------

    num = (
        abs(params["DeltaH"]) * rA_abs
        - (params["U"] * params["a"] / params["rho_b"])
        * (T - params["Ta"])
    )

    den = max(
        Ftot * params["Cp_mol"],
        1e-12
    )

    dT_dW = num / den

    # --------------------------------------
    # Ergun
    # --------------------------------------

    Qv = Ftot * params["R"] * T / P

    u = Qv / params["A"]

    MWmix = (
        yA * params["MWA"]
        + yH * params["MWB"]
        + yC * params["MWC"]
    )

    rho_g = cTot * MWmix

    dP_dz = (
        -(150
          * params["mu"]
          * (1 - params["epsilon"])**2
          / (params["epsilon"]**3 * params["dp"]**2))
        * u
        -
        (1.75
         * rho_g
         * (1 - params["epsilon"])
         / (params["epsilon"]**3 * params["dp"]))
        * u**2
    )

    dP_dW = dP_dz / (
        params["rho_b"]
        * params["A"]
    )

    # --------------------------------------
    # 転化率
    # --------------------------------------

    dxA_dW = rA_abs / params["FA0"]

    return [dxA_dW, dT_dW, dP_dW]


# ==========================================================
# 後処理
# ==========================================================

def post_profiles(W, xA, T, P, p):

    FA = p["FA0"] * (1 - xA)

    FB = p["FB0"] - 3 * p["FA0"] * xA
    FB = np.maximum(FB, 1e-16)

    FC = p["FC0"] + p["FA0"] * xA

    Ftot = FA + FB + FC

    cTot = P / (p["R"] * T)

    yA = FA / Ftot
    yH = FB / Ftot
    yC = FC / Ftot

    CA = yA * cTot
    CH = yH * cTot
    CC = yC * cTot

    rA_abs = np.array([
        rateLH(Ti, CAi, CHi, CCi)
        for Ti, CAi, CHi, CCi
        in zip(T, CA, CH, CC)
    ])

    return (
        FA, FB, FC,
        Ftot,
        cTot,
        CA, CH, CC,
        rA_abs
    )


# ==========================================================
# ODE計算
# ==========================================================

y0 = [0.0, T0, P0]

W_eval = np.linspace(
    0.0,
    W_total,
    500
)

sol = solve_ivp(
    rhs_noniso_pbr_ergun,
    (0.0, W_total),
    y0,
    method="BDF",
    t_eval=W_eval,
    rtol=1e-8,
    atol=[1e-10, 1e-10, 1e-6]
)

Wsol = sol.t

xA = sol.y[0]
T = sol.y[1]
P = sol.y[2]

# ==========================================================
# 後処理
# ==========================================================

(
    FA,
    FB,
    FC,
    Ftot,
    cTot,
    CA,
    CH,
    CC,
    rA_abs
) = post_profiles(
    Wsol,
    xA,
    T,
    P,
    params
)

# ==========================================================
# 可視化
# ==========================================================

fig, axs = plt.subplots(
    2, 3,
    figsize=(12, 7)
)

axs = axs.ravel()

axs[0].plot(Wsol, xA, 'b')
axs[0].set_title('Conversion')
axs[0].set_xlabel('W [kg_cat]')
axs[0].set_ylabel('xA')
axs[0].grid()

axs[1].plot(Wsol, T, 'r')
axs[1].set_title('Temperature')
axs[1].set_xlabel('W [kg_cat]')
axs[1].set_ylabel('T [K]')
axs[1].grid()

axs[2].plot(Wsol, rA_abs, 'k')
axs[2].set_title('Rate of reaction')
axs[2].set_xlabel('W [kg_cat]')
axs[2].set_ylabel('-rA')

axs[3].plot(Wsol, CA, label='CA')
axs[3].plot(Wsol, CC, label='CC')
axs[3].legend()
axs[3].grid()
axs[3].set_title('Concentration profile')

axs[4].plot(Wsol, Ftot)
axs[4].grid()
axs[4].set_title('Total mole flow rate')

axs[5].plot(Wsol, cTot)
axs[5].grid()
axs[5].set_title('Total concentration')

plt.tight_layout()

# ----------------------------------------------------------
# 圧力
# ----------------------------------------------------------

plt.figure(figsize=(6,4))
plt.plot(Wsol, P / 1e5, linewidth=2)

plt.xlabel('W [kg_cat]')
plt.ylabel('P [bar]')
plt.title('Pressure profile')
plt.grid()

plt.show()

# ==========================================================
# 出口サマリ
# ==========================================================

print("\n=== Outlet summary（Ergun） ===")

print(f"W_end      = {Wsol[-1]:.6f} kg_cat")
print(f"xA_out     = {xA[-1]:.6f}")
print(f"T_out      = {T[-1]:.2f} K")
print(f"P_out      = {P[-1]/1000:.2f} kPa")
print(f"FA_out     = {FA[-1]:.5e} mol/s")
print(f"FB_out     = {FB[-1]:.5e} mol/s")
print(f"FC_out     = {FC[-1]:.5e} mol/s")
print(f"Ftot_out   = {Ftot[-1]:.5e} mol/s")
print(f"cTot_out   = {cTot[-1]:.3f} mol/m³")
```
