#### 1. 単反応と複合反応

単一反応

$$A \rightarrow B$$

では

$$
-r_A=r
$$
のみを考えればよい。

しかし実際の化学プロセスでは複数の反応が同時に進行することが多い。

例
$$
A \rightarrow B\\ 
A \rightarrow C
$$
あるいは
$$
A \rightarrow B \rightarrow C
$$
このような系を
複合反応 (complex reactions)
という。

---

<h3>2. 複合反応の分類</h3>

<h4>2.1 並列反応</h4>

複数の反応が競争的に進行
$$
A \rightarrow B\\
A \rightarrow C
$$

速度式

$$
r_B=k_1 C_A^{n_1}\\ 
r_C=k_2 C_A^{n_2}
$$


<h4>2.2 逐次反応</h4>

中間生成物を経由して進行

$$
A \rightarrow B \rightarrow C
$$

Bが目的生成物である場合，いかにBを取り出すかが重要となる。

<h4>2.3 並列逐次反応</h4>

現実のプロセスの多くは

$$
A \rightarrow B \rightarrow D\\ 
A \rightarrow C \rightarrow D
$$

のようなネットワーク構造を持つ。

---

<h3>3. 転化率</h3>

Aの転化率

$$
X_A = \frac{F_{A0}-F_A}{F_{A0}}\\
0 \le X_A \le 1
$$

例

$F_{A0}=100\, ,  F_A=40$
なら

$X_A=0.60$

である。

<h3>4. 収率（Yield）</h3>

AからBへの収率

$$
Y_B = \frac{\text{生成した } B} {\text{供給した } A}
$$

モル基準では
$$
Y_B = \frac{F_B}{F_{A0}}
$$

収率は
「投入した原料に対してどれだけ目的物が得られたか」

を表す。


<h3>5. 選択率（Selectivity）</h3>

<h5>並列反応</h5>

$$
A \rightarrow B\\
A \rightarrow C
$$
を考える。

- 瞬間選択率
$$
S_{B/C} = \frac{r_B}{r_C}
$$

- 総合選択率
$$
S_{B/C} = \frac{F_B}{F_C}
$$

意味：

副生成物Cに対してどれだけBを作れたか。

<h3>6. 並列反応の選択率</h3>

反応速度

$$
r_B=k_1C_A^{n_1}\\
r_C=k_2C_A^{n_2}
$$

瞬間選択率

$$
S_{B/C} = \frac{k_1}{k_2} C_A^{n_1-n_2}
$$
この式は極めて重要である。

- ケース1
$$
n_1 > n_2
$$
なら
$$
S_{B/C} \propto C_A^{(n_1-n_2)}
$$
したがって

$C_A$

を高くする。

高濃度が有利

→ PFRが有利

- ケース2
$n_1<n_2$

なら
$C_A$

を下げる。

低濃度が有利

→ CSTRが有利

<h3>7. CSTRとPFRの比較</h3>
PFR

入口濃度が高い

$$C_A \approx C_{A0}$$

の領域が存在

~~~
高濃度
 ↓
入口
 ↓
出口
低濃度
~~~

CSTR

槽内濃度一定

$$
C_A=C_{A,\mathrm{out}}
$$

CSTR

   均一濃度


したがって

$$
n_1>n_2
$$
なら ##PFR##

が有利である。

<h3>8. 例題</h3>
<h5>問題</h5>

$$
A \rightarrow B\\
A \rightarrow C
$$

速度式
$$
r_B=2C_A^2\\ 
r_C=4C_A
$$

$$

$$

<h3>9. 逐次反応</h3>

$$
A \xrightarrow{\;r_1\;} B \xrightarrow{\;r_2\;} C 
$$

<h5>速度式</h5>

$$
r_1=k_1C_A\\
r_2=k_2C_B
$$

<h5>物質収支</h5>

$$
\begin{aligned}
\frac{dC_A}{dt} &= -k_1C_A\\
\frac{dC_B}{dt} &= k_1C_A-k_2C_B\\
\frac{dC_C}{dt} &= k_2C_B
\end{aligned}
$$

<h5>解析解</h5>

初期条件

$$
C_A(0)=C_{A0},\qquad C_B(0)=0,\qquad C_C(0)=0
$$
の解析解は以下のようになります。

<h5>Aの濃度</h5>

最初の式は単独で解けるので

$$
\frac{dC_A}{dt}=-k_1C_A
$$

より
$$
\boxed{ C_A(t)=C_{A0}e^{-k_1t} }
$$
◇◇◇
<h5>Bの濃度</h5>

$$
\frac{dC_B}{dt}+k_2C_B = k_1C_{A0}e^{-k_1t}　\tag{B}
$$

<h5>解析解</h5>
線形微分方程式なので、まず斉次方程式の解を求め、その後に特殊解を求める方法で一般解を導出していく(未定係数法）。

1. 斉次方程式の解

まず(B)式の右辺をゼロにした斉次方程式

$$
\frac{dC_B}{dt}+k_2C_B=0
$$
は、変数分離して

$$
\frac{dC_B}{C_B}=-k_2dt
$$
積分すると
$$
\ln C_B=-k_2t+\text{const.}
$$
したがって

$$
C_{Bh}=Ke^{-k_2t} 
$$
が斉次解です。

2. 特殊解を求める

右辺が
$$
k_1C_{A0}e^{-k_1t}
$$
なので、特殊解を

$$
C_{Bp}=Ae^{-k_1t} 
$$
と仮定すると
$$
\frac{dC_{Bp}}{dt} = -k_1Ae^{-k_1t}
$$
なので、元の方程式へ代入して
$$
-k_1Ae^{-k_1t} +k_2Ae^{-k_1t} = k_1C_{A0}e^{-k_1t} \\
A(k_2-k_1)e^{-k_1t} = k_1C_{A0}e^{-k_1t}
$$
よって

$$
A= \frac{k_1C_{A0}}{k_2-k_1} 
$$
となり、
$$
C_{Bp} = \frac{k_1C_{A0}}{k_2-k_1}e^{-k_1t} 
$$
を得る。

したがって一般解は
$$
C_B=C_{Bh}+C_{Bp}
$$
なので
$$
C_B = Ke^{-k_2t} + \frac{k_1C_{A0}}{k_2-k_1}e^{-k_1t}
$$

初期条件 $C_B(0)=0$
$$
0 = K + \frac{k_1C_{A0}}{k_2-k_1}
$$
したがって
$$
K = -\frac{k_1C_{A0}}{k_2-k_1}.
$$
これを代入すると
$$
C_B = -\frac{k_1C_{A0}}{k_2-k_1}e^{-k_2t} + \frac{k_1C_{A0}}{k_2-k_1}e^{-k_1t}.
$$
整理して
$$
\boxed{C_B(t) = \frac{k_1C_{A0}}{k_2-k_1} \left( e^{-k_1t} - e^{-k_2t} \right)} 
$$

◇◇◇




$k_1=k_2$のとき

$$
C_B(t)= \frac{k_1C_{A0}}{k_2-k_1} \left(e^{-k_1t}-e^{-k_2t}\right)
$$
で $k_2\to k_1$ の極限を考えます。

$k_1=k$ を固定して
$$
C_B= kC_{A0} \lim_{k_2\to k} \frac{e^{-kt}-e^{-k_2t}}{k_2-k}
$$
と書きます。

ここで
$$
f(x)=e^{-xt}
$$
とおくと
$$
C_B= kC_{A0} \lim_{k_2\to k} \frac{f(k)-f(k_2)}{k_2-k}.
$$
これは差分商の極限なので
$$
\lim_{k_2\to k} \frac{f(k)-f(k_2)}{k_2-k} = -f'(k).
$$
一方
$$
f'(x)=-te^{-xt}
$$
ですから
$$
-f'(k)=te^{-kt}.
$$
したがって
$$
\lim_{k_2\to k_1} \frac{k_1}{k_2-k_1} \left(e^{-k_1t}-e^{-k_2t}\right) = kte^{-kt} 
$$
となり、
$$
\boxed{ C_B(t)=C_{A0}\,kt\,e^{-kt} }
$$
◇◇◇

<h5>Cの濃度</h5>

物質保存

$$
C_A​+C_B​+C_C​=C_{A0}
$$

を用いると

$$
C_C = C_{A0}-C_A-C_B
$$

なので
$$
\boxed{C_C(t)= C_{A0} \left[ 1 -\frac{k_2}{k_2-k_1}e^{-k_1t} +\frac{k_1}{k_2-k_1}e^{-k_2t} \right] }
$$

となります。


特別な場合 $k_1=k_2=k$のときは
$$C_C=C_{A0}-C_A-C_B$$

から
$$
C_A=C_{A0}e^{-kt}, \qquad C_B=C_{A0}kte^{-kt}
$$
を代入し
$$
\boxed{C_C =} C_{A0} -C_{A0}e^{-kt} -C_{A0}kte^{-kt} = \boxed{C_{A0}\left[1-(1+kt)e^{-kt}\right]}.
$$

<h5>10. B最大条件</h5>

極大点では

$$
\frac{dC_B}{dt}=0
$$
したがって
$$
k_1C_A = k_2C_B
$$
この条件が

中間生成物Bの最大条件

である。

<h3>演習問題

<h5>問1</h5>

並列反応
$$
A \rightarrow B
$$

$$
A \rightarrow C
$$

において，生成速度式が

$$
r_B = 3C_A^2,\qquad r_C = 2C_A
$$

で与えられる。

目的生成物 \(B\) をできるだけ多く得るためには，CSTR と PFR のどちらが有利か答えよ。

---

## 解答

生成物 \(B\) と副生成物 \(C\) の瞬間選択率を考える。

$$
S_{B/C}
=
\frac{r_B}{r_C}
=
\frac{3C_A^2}{2C_A}
=
\frac{3}{2}C_A
$$

したがって，

$$
S_{B/C} \propto C_A
$$

となる。

すなわち，反応物濃度 \(C_A\) が高いほど，目的生成物 \(B\) に対する選択率が高くなる。

### PFRの場合

PFRでは入口付近で反応物濃度が高く保たれるため，

$$
S_{B/C}
=
\frac{3}{2}C_A
$$

も大きい。

したがって，目的生成物 \(B\) が優先的に生成される。

### CSTRの場合

CSTRでは反応器全体の濃度が出口濃度に等しい。

$$
C_A = C_{A,\mathrm{out}}
$$

したがって反応器全体が比較的低い濃度で反応するため，

$$
S_{B/C}
=
\frac{3}{2}C_A
$$

も小さくなり，副生成物 \(C\) が相対的に多く生成される。

---

## 結論

$$
\boxed{\text{PFR が有利}}
$$

---

## 補足

この問題では

$$
S_{B/C}
=
\frac{k_B}{k_C}
C_A^{\,2-1}
$$

となり，選択率は \(C_A\) の増加とともに大きくなる。

一般に並列反応

$$
A \rightarrow B
$$

$$
A \rightarrow C
$$

について，

$$
r_B = k_B C_A^\alpha,
\qquad
r_C = k_C C_A^\beta
$$

であるとき，

$$
S_{B/C}
=
\frac{r_B}{r_C}
=
\frac{k_B}{k_C}
C_A^{\alpha-\beta}
$$

である。

- \( \alpha > \beta \) のとき：高濃度が有利 → PFR
- \( \alpha < \beta \) のとき：低濃度が有利 → CSTR
- \( \alpha = \beta \) のとき：反応器形式による差は小さい

本問では

$$
\alpha = 2,\qquad \beta = 1
$$

であるため，

$$
\boxed{\text{PFR が有利}}
$$

となる。

---

## 問2

並列反応

$$
A \rightarrow B
$$

$$
A \rightarrow C
$$

において，生成速度式が

$$
r_B = 2C_A,\qquad r_C = 5C_A^3
$$

で与えられる。

目的生成物 \(B\) をできるだけ多く得るためには，CSTR と PFR のどちらが有利か答えよ。また，その理由を説明せよ。

---



## 解答

生成物 \(B\) と副生成物 \(C\) の瞬間選択率を考える。

$$
S_{B/C}
=
\frac{r_B}{r_C}
=
\frac{2C_A}{5C_A^3}
=
\frac{2}{5C_A^2}
$$

したがって，

$$
S_{B/C}
\propto
\frac{1}{C_A^2}
$$

となる。

すなわち，反応物濃度 \(C_A\) が低いほど，目的生成物 \(B\) に対する選択率が高くなる。

### PFRの場合

PFRでは入口付近で反応物濃度が高い。

したがって，

$$
S_{B/C}
=
\frac{2}{5C_A^2}
$$

は小さくなり，副生成物 \(C\) が相対的に多く生成される。

### CSTRの場合

CSTRでは反応器全体の濃度が出口濃度に等しく，

$$
C_A = C_{A,\mathrm{out}}
$$

となる。

反応器全体が比較的低い濃度で反応するため，

$$
S_{B/C}
=
\frac{2}{5C_A^2}
$$

は大きくなり，目的生成物 \(B\) が優先的に生成される。

---

## 結論

$$
\boxed{\text{CSTR が有利}}
$$

---

S_{B/C}
=
\frac{k_B}{k_C}
C_A^{\,1-3}
=
\frac{k_B}{k_C}
C_A^{-2}
$$

となる。

一般に並列反応

$$

$$
A \rightarrow C
$$

に対して，

\qquad

であれば，

<h3>問3</h3>

$$
A\rightarrow B\rightarrow C 
$$

で $k_1=1\,\mathrm{h^{-1}}, k_2=0.5\,\mathrm{h^{-1}}$ の場合，$C_B$ が最大となる条件を導け。
「転化率が高いほど目的生成物収率も高い」とは限らないことを説明せよ。

大学院向けの次回講義としては、この内容の発展として「複合反応系のPFR・CSTR設計方程式」「最適滞留時間」「中間生成物収率の最大化」「シリーズCSTRとPFRの組合せ設計」に進むのが自然です。

解答例

この問題は並列反応

$$
\mathrm{A} \rightarrow \mathrm{B},\qquad \mathrm{A} \rightarrow \mathrm{C}
$$

において、目的生成物 $\mathrm{B}$ の選択率を最大にする反応器を選ぶ問題です。

模範解答

与えられた生成速度式は
$$
r_B=3C_A^2,\qquad r_C=2C_A
$$
である。

目的生成物 \mathrm{B} に対する瞬間選択率は

$$
S_{B/C}=\frac{r_B}{r_C}
$$
より、
$$
S_{B/C} =\frac{3C_A^2}{2C_A} =\frac{3}{2}C_A
$$
となる。

したがって、

$$
S_{B/C}\propto C_A
$$
であり、反応物濃度 $C_A$ が高いほど \mathrm{B} が有利に生成される。

<h5>CSTR と PFR の比較</h5>

CSTR では槽内濃度は出口濃度と等しく、反応器全体で低い $C_A$ に保たれる。
PFR では入口付近で高い $C_A$ が存在し、反応器内の平均濃度も CSTR より高い。

本系では高濃度ほど \mathrm{B} の選択率が高くなるため、

$$
\boxed{\text{PFR（管型反応器）が有利}}
$$
である。
◇◇◇

Levenspiel流の判定法

一般に並列反応

$$
A\rightarrow B,\qquad A\rightarrow C
$$

において

$$
\frac{r_B}{r_C}\propto C_A^{\,n_B-n_C}
$$
である。

本問では
$$
n_B=2,\qquad n_C=1
$$
なので
$$
n_B-n_C=1>0
$$
となり、高濃度が有利である。

よって

$$\boxed{\text{PFR}}$$

を選択する。

答：PFR（管型反応器）。



逐次反応の場合は、Levenspiel の考え方が少し変わります。

並列反応では
$$
A \rightarrow B\\
A \rightarrow C
$$

に対して
$$
\frac{r_B}{r_C}
$$

を見ればよかったのですが、逐次反応では

$$
A \rightarrow B \rightarrow C
$$

となり、目的物 $\mathrm{B}$ が中間生成物になるため、

$$
\text{Bを作る速さ}
\quad \text{vs} \quad
\text{Bを消費する速さ}
$$

の競争を考えます。

<h5>Levenspiel流の判定</h5>

逐次反応

$$
A \xrightarrow{r_1} B \xrightarrow{r_2} C
$$

では

$$
r_1=k_1C_A^\alpha\\
r_2=k_2C_B^\beta
$$


とすると、

$$
\frac{r_1}{r_2}
$$

を大きくしたいことになります。

一般則

<h5>B生成反応の次数の方が大きい</h5>

例

$$
r_1=k_1C_A^2\\
r_2=k_2C_B
$$

このとき高濃度が有利です。

PFRの入口では

$C_A$


が非常に高いので、

$r_1$


が大きくなり、まず$\mathrm{B}$を大量に作れます。

したがって
$$
\boxed{\mathrm{PFR\ が有利}}
$$

<h5>B消費反応の次数の方が大きい</h5>

例
$$
r_1=k_1C_A\\
r_2=k_2C_B^2
$$

この場合、

$C_B$


が高くなると$\mathrm{B}$が急速に消費されます。

CSTRでは槽内全体で$\mathrm{B}$濃度が高くなりやすく、

$r_2$

が大きくなります。

PFRではBが徐々に生成して徐々に消費されるため、

$C_B$

のピークが低く抑えられます。

したがって

$$
\boxed{\mathrm{PFR\ が有利}}
$$

になることが多いです。

Levenspielが強調している重要な結論

逐次反応
$$
A \rightarrow B \rightarrow C
$$

で

$\mathrm{B}$


を最大化したい場合は、

一般に
$$
\boxed{\mathrm{PFRの方が有利}}
$$

になることが多いです。

理由は、PFRでは滞留時間分布がなく、

入った流体は同じ時間だけ反応する
過反応して $B\rightarrow C$ が進む部分が少ない

からです。

一方CSTRでは

一部はすぐ出る
一部は長時間滞留する

ため、
$$
B \rightarrow C
$$

が進みやすくなります。

一次逐次反応の有名な結果
$$
A \xrightarrow{k_1} B \xrightarrow{k_2} C
$$

なら、

同じ平均滞留時間では

$$
\boxed{\text{中間生成物Bの収率はPFR > CSTR}}
$$

となります。

このため化学反応工学では

$$
A \rightarrow B \rightarrow C
$$

でBを製造したい場合、

まず

$$
\boxed{\text{PFRが有利ではないか？}}
$$

と考えるのが定石です。

授業や演習で扱う Levenspiel の範囲なら、

並列反応 → 選択率 $r_B/r_C$ の濃度依存性を見る
逐次反応 → 中間生成物 $\mathrm{B}$ を守りたいので基本的に PFR が有利

と整理しておくと分かりやすいです。
