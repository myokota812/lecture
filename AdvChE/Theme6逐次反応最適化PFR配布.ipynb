### 【課題1】回分反応器（BR)から管型反応器（PFR）への展開

これまでの講義で、閉鎖系である「回分反応器（Batch Reactor:BR）」の物質収支および微分型設計方程式を学びました。定容反応に限定されますが、実は、連続式反応器の代表例である管型反応器（Plug Flow Reactor: PFR）の設計方程式も形だけを見れば、回分の設計方程式の反応時間$t$を滞留時間$\tau$に置き換えただけで、基本的に同じような設計方程式になります。

$$\frac{dC_i}{dt} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N) \tag{BR}$$

$$\frac{dC_i}{d\tau} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N) \tag{PFR}$$

以下の導入部分につづいて、PFRの設計方程式（PFR)を導出してください。ただし、定容反応とします。

### 【課題2】PFRにおける逐次反応の収率改善に関するプロセス設計提案

一次の不可逆逐次反応 $A \xrightarrow{k_1} B \xrightarrow{k_2} C$ をPFRで実施する場合、空間時間 $\tau$ を最適化しても、単パス（1回通過）の条件だけでは目的物質 $B$ の収率 $Y_B$ には理論的な頭打ち（限界）が存在します。<br>
【補足】：BatchからPFRに移行するのは、今後のつながりの関係です。基本となる設計方程式をみてもらってもわかるとおり、BatchからPFRに変えても、特に工夫しなければ、改善は期待できません。

単一の均一条件での運転の限界を打破し、目的物質 $B$ の生産効率（収率・経済性）を劇的に高めるためのプロセス改善策について、以下の**選択肢の軸**を組み合わせて、あなたの独自のアイデアを論理的に提案しなさい。

【付録】逐次反応（$\text{A} \to \text{B} \to \text{C}$）における主な式
基本的にBRと同じ式です、BRにおける時間$t$をPFRにおける空間時間$\tau$に置換したものになります。PFRの結果をグラフにすると、装置内の成分濃度や転化率等の分布と解釈してください。

各成分濃度の軸方向($\tau$)分布
$$\left\{ \begin{aligned}
C_A &= C_{A0} e^{-k_1 \tau}\\
C_B &= C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)\\
C_C &= C_{A0} \left( 1 - \frac{k_2 e^{-k_1 \tau} - k_1 e^{-k_2 \tau}}{k_2 - k_1} \right)
\end{aligned} \right.$$
<br>

転化率、収率、選択率
$$\left\{ \begin{aligned}
 X_A(\tau) &= 1 - e^{-k_1 \tau}\\
 Y_B(\tau) &= \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right) \\
 S_B(\tau) &= \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 \tau} - e^{-k_2 \tau}}{1 - e^{-k_1 \tau}}
\end{aligned} \right.$$
