## 1. 複合反応とは

逐次反応: A→B→C

並列反応: A→B, A→C

目的生成物を高選択的に得ることが重要である。

## 2. 逐次反応

$$A\rightarrow B\rightarrow C$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

## 3. 量論行列

$$S=\begin{bmatrix}-1&0\\1&-1\\0&1\end{bmatrix}$$

$$\frac{d\mathbf C}{dt}=S\mathbf r$$

## 4. 支配方程式

$$\frac{dC_A}{dt}=-k_1C_A$$

$$\frac{dC_B}{dt}=k_1C_A-k_2C_B$$

$$\frac{dC_C}{dt}=k_2C_B$$

## 5. 解析解

$$C_A=C_{A0}e^{-k_1t}$$

$$C_B=\frac{k_1C_{A0}}{k_2-k_1}(e^{-k_1t}-e^{-k_2t})$$

$$C_C=C_{A0}-C_A-C_B$$

## 6. 特殊ケース k1=k2=k

$$C_A=C_{A0}e^{-kt}$$

$$C_B=C_{A0}kte^{-kt}$$

$$C_C=C_{A0}[1-(1+kt)e^{-kt}]$$

## 7. Bの極大条件

$$\frac{dC_B}{dt}=0$$

より

$$k_1C_A=k_2C_B$$

極大時間

$$t_{max}=\frac{1}{k_2-k_1}\ln\left(\frac{k_2}{k_1}\right)$$

```python
import numpy as np
import matplotlib.pyplot as plt
k1=1.0;k2=0.3;CA0=1.0
t=np.linspace(0,10,500)
CA=CA0*np.exp(-k1*t)
CB=k1*CA0/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
CC=CA0-CA-CB
plt.plot(t,CA,label="A")
plt.plot(t,CB,label="B")
plt.plot(t,CC,label="C")
plt.legend();plt.grid();plt.show()
```

## 8. 並列反応

$$A\rightarrow B$$
$$A\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 9. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C}$$

$$S_{B/C}=\frac{k_B}{k_C}C_A^{m-n}$$

- m>n : 高濃度有利
- m<n : 低濃度有利

## 10. 転化率・収率・選択率

$$X=\frac{C_{A0}-C_A}{C_{A0}}$$

$$Y_B=\frac{B生成量}{A供給量}$$

$$S_{B/C}=\frac{B生成量}{C生成量}$$

```python
import numpy as np
kB=3
kC=2
CA=np.linspace(0.01,2,200)
S=(kB/kC)*CA
print("S at CA=1 =",1.5)
```

## 11. 反応器選定との関係

高濃度有利ならPFR・回分反応器が有利。
低濃度有利ならCSTRが有利。
