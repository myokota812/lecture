## 1. 複合反応とは

逐次反応: $$\mathrm{A}\rightarrow \mathrm{B}\rightarrow \mathrm{C}$$

並列反応: $$\mathrm{A}\rightarrow \mathrm{B}\\ \mathrm{A}\rightarrow \mathrm{C}$$

◇◇◇

実際の反応では、一つの反応だけが起こるとは限りません。複数の反応が関係する系を複合反応と呼びます。代表例として、A→B→Cのような逐次反応と、A→BおよびA→Cのような並列反応があります。このことは前回の講義でも触れました。本講義では、それぞれの反応速度式の立て方と濃度変化の解析方法を学びます。

---

## 2. 逐次反応

$$ A \xrightarrow{r_1} B \xrightarrow{r_2} C \tag{1}$$

$$r_1=k_1C_A \tag{2}$$
$$r_2=k_2C_B \tag{3}$$

◇◇◇

まず、逐次反応を考えます。

この反応系では、Aが反応してBを生成し、さらに生成したBが反応してCになります。

それぞれの反応速度を一次反応と仮定すると、

第1段階および第2段階の速度はそれぞれ、(2),(3)式で与えられます。

この反応ではBが中間生成物となるため、Bは生成される一方で消費もされます。そのため、Bの濃度は一般に一度増加した後、減少するという特徴的な挙動を示します。これが単純反応との大きな違いです。

---


## 3. 物質収支式（支配方程式）

逐次反応(1),(2),(3)式おいて、定積回分反応器の物質収支式は

$$
\left\{ 
    \begin{aligned} 
        \frac{dC_A}{dt} &= -k_1C_A\\ 
        \frac{dC_B}{dt} &= k_1C_A-k_2C_B \tag{4}\\ 
        \frac{dC_C}{dt} &= k_2C_B
    \end{aligned}
\right. 
$$
となります。

◇◇◇

それでは、体積一定の回分反応器を例にして、物質収支式を導出してみます。

まずAは反応1でのみ消費されるため、
$$
\frac{dC_A}{dt}=-k_1C_A
$$
となります。

一方Bは、反応1で生成される一方、反応2で消費されます。

したがって、
$$
\frac{dC_B}{dt}=k_1C_A-k_2C_B
$$
となります。

この式の第1項が生成速度、第2項が消費速度を表しています。

つまりBの濃度変化は「生成量と消費量の差」で決まります。

Cは反応2でのみ生成されるため、

$$
\frac{dC_C}{dt}=k_2C_B
$$

以上が、同じ装置内で同時進行しますので、式の上では(4)式の連立微分方程式で表現できます。

---



## 4. 量論行列

（４）式を反応速度に基づく量論行列で表していきます。

まず濃度ベクトルを$\mathbf C$とおきます。
$$
\mathbf C= \begin{bmatrix} C_A\\ C_B\\ C_C \end{bmatrix}
$$

そうすると先程の物質収支式を行列で表すと次のとおりです。

$$
\frac{d\mathbf C}{dt} =  
\begin{bmatrix} -1 & 0 \\ 1 & -1 \\ 0 & 1 \end{bmatrix} 
\begin{bmatrix} k_1C_A\\ k_2C_B \end{bmatrix}.
$$

ここで
$$
S= \begin{bmatrix} -1 & 0 \\ 1 & -1 \\ 0 & 1 \end{bmatrix}
$$
$$
\mathbf r= \begin{bmatrix} r_1\\ r_2 \end{bmatrix} = \begin{bmatrix} k_1C_A\\ k_2C_B \end{bmatrix}
$$
とおくと,速度に基づく量論方程式は次のように完結に表現できます。

$$
\frac{d\mathbf C}{dt}
=
S\mathbf r
$$

この形にしておくと、多数の成分や反応を含む複雑な系でも統一的に扱うことができます。

---

## 5. 解法


それでは、連立微分方程式を解いてみましょう。ここではあえて３種類の解法を勉強します。

<h5>5.1 直接微分方程式を解く方法</h5>

初期条件を $t = 0$ において $C_A(0) = C_{A0}$, $C_B(0) = 0$, $C_C(0) = 0$ として各成分の濃度 $C_A(t)$, $C_B(t)$, $C_C(t)$ を解析的に求めます。

**1. $C_A(t)$ の導出**

第1式は変数分離形です。

$$\frac{dC_A}{C_A} = -k_1 dt$$

両辺を積分して初期条件を適用すると：

$$C_A(t) = C_{A0} e^{-k_1 t}$$



**2. $C_B(t)$ の導出**

定数変化法（微分方程式の標準的な解法）を用いて、第2式の $C_B(t)$ について解く手順を詳しく解説します。

解く対象の微分方程式：


$$\frac{dC_B}{dt} + k_2 C_B = k_1 C_{A0} e^{-k_1 t}$$

---

**Step 1: 斉次方程式の一般解を求める**

まず、右辺を $0$ と置いた斉次方程式（同次方程式）を解きます。

$$\frac{dC_B}{dt} + k_2 C_B = 0$$

変数分離を行います。

$$\frac{dC_B}{C_B} = -k_2 dt$$

両辺を積分します。

$$\ln \vert{}C_B\vert{} = -k_2 t + C_1 \implies C_B(t) = A e^{-k_2 t}$$

（ここで $A$ は任意の定数です）

---

**Step 2: 任意定数 $A$ を関数 $A(t)$ に置き換える（定数変化法）**

定数 $A$ を時間の関数 $A(t)$ と仮定し、求める非斉次方程式の特解の形を次のように置きます。

$$C_B(t) = A(t) e^{-k_2 t}$$

これを微分します（積の微分公式を使用）。

$$\frac{dC_B}{dt} = \frac{dA(t)}{dt} e^{-k_2 t} - k_2 A(t) e^{-k_2 t}$$

---

**Step 3: 元の微分方程式に代入する**

得られた $C_B$ と $\frac{dC_B}{dt}$ を元の式 $\frac{dC_B}{dt} + k_2 C_B = k_1 C_{A0} e^{-k_1 t}$ に代入します。

$$\left( \frac{dA(t)}{dt} e^{-k_2 t} - k_2 A(t) e^{-k_2 t} \right) + k_2 \left( A(t) e^{-k_2 t} \right) = k_1 C_{A0} e^{-k_1 t}$$

左辺の $k_2 A(t) e^{-k_2 t}$ の項同士が打ち消し合って消去されます。

$$\frac{dA(t)}{dt} e^{-k_2 t} = k_1 C_{A0} e^{-k_1 t}$$

両辺に $e^{k_2 t}$ を掛けます。

$$\frac{dA(t)}{dt} = k_1 C_{A0} e^{(k_2 - k_1)t}$$

---

**Step 4: $A(t)$ を求める**

両辺を $t$ で積分します。ここからは $k_1 \neq k_2$ と $k_1 = k_2$ で場合分けします。

* **$k_1 \neq k_2$ の場合：**

$$A(t) = \int k_1 C_{A0} e^{(k_2 - k_1)t} dt = \frac{k_1 C_{A0}}{k_2 - k_1} e^{(k_2 - k_1)t} + B$$

（$B$ は積分定数）

これを $C_B(t) = A(t) e^{-k_2 t}$ に代入すると、一般解が得られます。

$$C_B(t) = \left( \frac{k_1 C_{A0}}{k_2 - k_1} e^{(k_2 - k_1)t} + B \right) e^{-k_2 t} = \frac{k_1 C_{A0}}{k_2 - k_1} e^{-k_1 t} + B e^{-k_2 t}$$

* **$k_1 = k_2$ の場合：**

$$\frac{dA(t)}{dt} = k_1 C_{A0} \implies A(t) = k_1 C_{A0} t + B$$

$$C_B(t) = (k_1 C_{A0} t + B) e^{-k_1 t}$$

---

**Step 5: 初期条件 $C_B(0) = 0$ を適用する**

* **$k_1 \neq k_2$ の場合：**

$$C_B(0) = \frac{k_1 C_{A0}}{k_2 - k_1} + B = 0 \implies B = -\frac{k_1 C_{A0}}{k_2 - k_1}$$

これを戻すと、目的の解が得られます。

$$C_B(t) = \frac{k_1 C_{A0}}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

* **$k_1 = k_2$ の場合：**

$$C_B(0) = B = 0 \implies C_B(t) = k_1 C_{A0} t e^{-k_1 t}$$

---



**3. $C_C(t)$ の導出**
物質収支の関係式 $C_A(t) + C_B(t) + C_C(t) = C_{A0}$（または第3式を積分）より、$C_C(t) = C_{A0} - C_A(t) - C_B(t)$ として求めます。

* $k_1 \neq k_2$ の場合：

$$C_C(t) = C_{A0} \left[ 1 - \frac{k_2}{k_2 - k_1} e^{-k_1 t} + \frac{k_1}{k_2 - k_1} e^{-k_2 t} \right]$$

* $k_1 = k_2 = k$ の場合：

$$C_C(t) = C_{A0} \left[ 1 - (1 + kt) e^{-kt} \right]$$

---


<h5>5.2 行列微分方程式を解く方法</h5>


先ほどの連立微分方程式は、（化学量論行列） $S$ と反応速度ベクトル $\mathbf{r}$ を用いて 
$$\frac{d\mathbf{C}}{dt} = S\mathbf{r}$$
 と表されていますが、解く際はこれを**係数行列 $K$ による線形微分方程式 
 $$\frac{d\mathbf{C}}{dt} = K\mathbf{C}$$
  に書き直して行列指数関数（または固有値分解）を用います。
  
  ---


**1. 微分方程式を標準形式 $\frac{d\mathbf{C}}{dt} = K\mathbf{C}$ に変形**

反応速度ベクトル $\mathbf{r}$ は次のように表現できます。

$$\mathbf{r} = \begin{bmatrix} k_1 C_A \\ k_2 C_B \end{bmatrix} = \begin{bmatrix} k_1 & 0 & 0 \\ 0 & k_2 & 0 \end{bmatrix} \begin{bmatrix} C_A \\ C_B \\ C_C \end{bmatrix}$$

これを $\frac{d\mathbf{C}}{dt} = S\mathbf{r}$ に代入すると：

$$\frac{d\mathbf{C}}{dt} = \begin{bmatrix} -1 & 0 \\ 1 & -1 \\ 0 & 1 \end{bmatrix} \begin{bmatrix} k_1 & 0 & 0 \\ 0 & k_2 & 0 \end{bmatrix} \mathbf{C} = \begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix} \mathbf{C}$$

この $3 \times 3$ 行列を **$K$** と置きます。

$$K = \begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix}$$

---


**2. 行列微分方程式の一般解の形**

1階定数係数行列微分方程式 $\frac{d\mathbf{C}}{dt} = K\mathbf{C}$ を満たす解ですが、直感的に予想できます。行列であることは無視して、

**１階微分すると元の関数に戻る**

とみれば、指数関数が解となることは明白です。

そこで

1階定数係数行列微分方程式 $\frac{d\mathbf{C}}{dt} = K\mathbf{C}$を満たす解は、**行列指数関数（Matrix Exponential）** を用いて直接以下のように書けます。

$$\mathbf{C}(t) = e^{Kt} \mathbf{C}(0)$$

ここで、初期状態は $\mathbf{C}(0) = \begin{bmatrix} C_{A0} \\ 0 \\ 0 \end{bmatrix}$ です。

---

**3. $e^{Kt}$ の計算（固有値・対角化を用いる方法）** 

行列 $P$ を使った対角化の手順を具体的に書き下し、$e^{Kt} \mathbf{C}(0)$ の計算プロセスをステップ順に解説します。

---

**Step 1. 固有ベクトルと変換行列 $P$ の作成**

係数行列 $K$ の固有方程式 $(K - \lambda I)\mathbf{v} = \mathbf{0}$ を解いて、各固有値に対応する固有ベクトルを求めます。

* **$\lambda_1 = -k_1$ に対する固有ベクトル $\mathbf{v}_1$**

$$\begin{bmatrix} 0 & 0 & 0 \\ k_1 & k_1 - k_2 & 0 \\ 0 & k_2 & k_1 \end{bmatrix} \begin{bmatrix} v_{11} \\ v_{21} \\ v_{31} \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \implies \mathbf{v}_1 = \begin{bmatrix} k_2 - k_1 \\ k_1 \\ -k_2 \end{bmatrix}$$

* **$\lambda_2 = -k_2$ に対する固有ベクトル $\mathbf{v}_2$**

$$\begin{bmatrix} k_2 - k_1 & 0 & 0 \\ k_1 & 0 & 0 \\ 0 & k_2 & k_2 \end{bmatrix} \begin{bmatrix} v_{12} \\ v_{22} \\ v_{32} \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \implies \mathbf{v}_2 = \begin{bmatrix} 0 \\ -1 \\ 1 \end{bmatrix}$$

* **$\lambda_3 = 0$ に対する固有ベクトル $\mathbf{v}_3$**

$$\begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix} \begin{bmatrix} v_{13} \\ v_{23} \\ v_{33} \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \\ 0 \end{bmatrix} \implies \mathbf{v}_3 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}$$

固有ベクトルを列として並べた行列 $P$ は次のようになります。

$$P = \begin{bmatrix} \mathbf{v}_1 & \mathbf{v}_2 & \mathbf{v}_3 \end{bmatrix} = \begin{bmatrix} k_2 - k_1 & 0 & 0 \\ k_1 & -1 & 0 \\ -k_2 & 1 & 1 \end{bmatrix}$$

---

**Step 2. 逆行列 $P^{-1}$ の導出**

$P$ は下三角行列なので、逆行列 $P^{-1}$ も容易に求まります。

$$P^{-1} = \begin{bmatrix} \frac{1}{k_2 - k_1} & 0 & 0 \\ \frac{k_1}{k_2 - k_1} & -1 & 0 \\ 1 & 1 & 1 \end{bmatrix}$$

---

**Step 3. 行列の掛け算 $e^{Kt} = P e^{Dt} P^{-1}$ の実行**

対角行列の指数関数 $e^{Dt} = \text{diag}(e^{-k_1 t}, e^{-k_2 t}, 1)$ を用いて計算します。

まず $e^{Dt} P^{-1}$ を計算します。

$$e^{Dt} P^{-1} = \begin{bmatrix} e^{-k_1 t} & 0 & 0 \\ 0 & e^{-k_2 t} & 0 \\ 0 & 0 & 1 \end{bmatrix} \begin{bmatrix} \frac{1}{k_2 - k_1} & 0 & 0 \\ \frac{k_1}{k_2 - k_1} & -1 & 0 \\ 1 & 1 & 1 \end{bmatrix} = \begin{bmatrix} \frac{e^{-k_1 t}}{k_2 - k_1} & 0 & 0 \\ \frac{k_1 e^{-k_2 t}}{k_2 - k_1} & -e^{-k_2 t} & 0 \\ 1 & 1 & 1 \end{bmatrix}$$

次に、左から $P$ を掛けます。

$$e^{Kt} = \begin{bmatrix} k_2 - k_1 & 0 & 0 \\ k_1 & -1 & 0 \\ -k_2 & 1 & 1 \end{bmatrix} \begin{bmatrix} \frac{e^{-k_1 t}}{k_2 - k_1} & 0 & 0 \\ \frac{k_1 e^{-k_2 t}}{k_2 - k_1} & -e^{-k_2 t} & 0 \\ 1 & 1 & 1 \end{bmatrix}$$

$$e^{Kt} = \begin{bmatrix} e^{-k_1 t} & 0 & 0 \\ \frac{k_1}{k_2 - k_1}(e^{-k_1 t} - e^{-k_2 t}) & e^{-k_2 t} & 0 \\ 1 - \frac{k_2 e^{-k_1 t} - k_1 e^{-k_2 t}}{k_2 - k_1} & 1 - e^{-k_2 t} & 1 \end{bmatrix}$$

---

**Step 4. 初期条件 $\mathbf{C}(0)$ を掛けて最終解を算出**

初期濃度 $\mathbf{C}(0) = \begin{bmatrix} C_{A0} \\ 0 \\ 0 \end{bmatrix}$ を掛けます。

$$\mathbf{C}(t) = e^{Kt} \begin{bmatrix} C_{A0} \\ 0 \\ 0 \end{bmatrix}$$

右辺のベクトルは第1成分以外が $0$ なので、行列 $e^{Kt}$ の**第1列目のみ**が抽出されます。

$$\mathbf{C}(t) = \begin{bmatrix} e^{-k_1 t} \\ \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) \\ 1 - \frac{k_2}{k_2 - k_1} e^{-k_1 t} + \frac{k_1}{k_2 - k_1} e^{-k_2 t} \end{bmatrix} C_{A0}$$

これで、前回のスカラー微分方程式の連立解と完全に一致することが確認できます。

---

> **Point:** $S$（化学量論行列）と $\mathbf{r}$（速度ベクトル）に分けた表記は**系の構造と反応速度則を分離して記述する際に便利**ですが、解を解析的に求める際は $K = S \times (\text{速度定数行列})$ の形にして固有値分解や行列指数関数を用いるのが一般的な手順です。

<h5>5.3 数値計算により解く方法</h5>

逐次反応（$A \xrightarrow{k_1} B \xrightarrow{k_2} C$）の連立微分方程式を数値的に解くには、常微分方程式（ODE）の標準的な数値解法であるルンゲ＝クッタ法（Runge-Kutta method, 4次）や、Pythonの科学計算ライブラリ `scipy.integrate.solve_ivp` を用いるのが一般的です。

---

### 数値解法の基本的な考え方

時間刻みを $\Delta t$ とし、$t_n = n \Delta t$ における濃度を $\mathbf{C}_n = [C_A(t_n), C_B(t_n), C_C(t_n)]^T$ とします。

1. **オイラー法（概念）**:
勾配を使って直進的に近似します：

$$\mathbf{C}_{n+1} \approx \mathbf{C}_n + \left( \frac{d\mathbf{C}}{dt} \right)_n \Delta t$$


2. **4次ルンゲ＝クッタ法（RK4）**:
1ステップ間における勾配を4点（始点・中間点2つ・終点）で計算して加重平均を取るため、誤差が $\mathcal{O}(\Delta t^4)$ と非常に小さく高精度です。

---

### Pythonコード（数値解法と解析解の比較グラフ）

Pythonの `scipy` と `matplotlib` を用いた実装例です。数値解には `solve_ivp`（内部でRK45等を使用）を利用し、事前に求めた解析解と並べて比較グラフを描画します。



---



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# --- 1. パラメータと初期条件の設定 ---
k1 = 0.5   # 反応速度定数 k1 [1/min]
k2 = 0.2   # 反応速度定数 k2 [1/min]
CA0 = 1.0  # 初期濃度 C_A(0) [mol/L]

t_span = (0, 20)            # 計算時間範囲 [0, 20]
t_eval = np.linspace(0, 20, 200) # グラフ描画用の時間点

# --- 2. 微分方程式の定義（数値計算用） ---
def reaction_system(t, C):
    CA, CB, CC = C
    dCA_dt = -k1 * CA
    dCB_dt = k1 * CA - k2 * CB
    dCC_dt = k2 * CB
    return [dCA_dt, dCB_dt, dCC_dt]

# --- 3. 数値解の計算 (Scipy solve_ivp) ---
C0 = [CA0, 0.0, 0.0]  # 初期条件 [CA0, CB0, CC0]
sol = solve_ivp(reaction_system, t_span, C0, t_eval=t_eval, method='RK45')

# --- 4. 解析解の計算 ---
t_exact = t_eval
CA_exact = CA0 * np.exp(-k1 * t_exact)
CB_exact = (k1 * CA0 / (k2 - k1)) * (np.exp(-k1 * t_exact) - np.exp(-k2 * t_exact))
CC_exact = CA0 * (1 - (k2 / (k2 - k1)) * np.exp(-k1 * t_exact) + (k1 / (k2 - k1)) * np.exp(-k2 * t_exact))

# --- 5. 比較グラフの描画 ---
plt.figure(figsize=(10, 6))

# 数値解のプロット（点）
plt.plot(sol.t, sol.y[0], 'o', label='$C_A$ (Numerical)', markersize=4, color='tab:blue')
plt.plot(sol.t, sol.y[1], 's', label='$C_B$ (Numerical)', markersize=4, color='tab:orange')
plt.plot(sol.t, sol.y[2], '^', label='$C_C$ (Numerical)', markersize=4, color='tab:green')

# 解析解のプロット（実線）
plt.plot(t_exact, CA_exact, '-', label='$C_A$ (Analytical)', color='navy', linewidth=1.5)
plt.plot(t_exact, CB_exact, '-', label='$C_B$ (Analytical)', color='darkorange', linewidth=1.5)
plt.plot(t_exact, CC_exact, '-', label='$C_C$ (Analytical)', color='darkgreen', linewidth=1.5)

# グラフ装飾
plt.xlabel('Time $t$ [min]', fontsize=12)
plt.ylabel('Concentration [mol/L]', fontsize=12)
plt.title('Comparison between Numerical and Analytical Solutions', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# 表示
plt.show()
```

### コードの解説

1. **`solve_ivp` 関数**:
`reaction_system` で定義した微分方程式を数値的に解きます。デフォルトで `RK45`（4次・5次のルンゲ＝クッタ適応ステップ法）が使われており、高精度な数値解が得られます。
2. **比較プロット**:
プロット上の「マーカー（丸・四角・三角）」が**数値解**、「実線」が**解析解**を表します。2つの結果が完全に重なることで、数値解法の正確性と解析解の導出が正しいことを視覚的に検証できます。

## 7. Bの極大条件

$$\frac{dC_B}{dt}=0$$

より

$$k_1C_A=k_2C_B$$

極大時間

$$t_{max}=\frac{1}{k_2-k_1}\ln\left(\frac{k_2}{k_1}\right)$$

## 8. 並列反応

$$A\rightarrow B$$
$$A\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 9. 瞬間選択率
