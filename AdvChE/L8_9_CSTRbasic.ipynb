　　　　　　　　　CSTRモデル

成分 $i$ に対して、反応器の物質収支（一般式）は

&#160;&#160;&#160;&#160;
$$\begin{array}{l} \frac{dN_i }{dt}~~=~~(\textrm{成分}\;i\;\textrm{の流入速度})-(\textrm{成分}\;i\;\textrm{の流出速度})+(\textrm{生成速度})\newline ~~\;　=FC_{i0} -FC_i +Vr_i  \end{array}$$

$$\begin{array}{ll} N_i  & :\textrm{反応器内の成分}\;i\;\textrm{のモル数}\;\textrm{[mol]}\newline C_{i0}  & :\textrm{入口の成分}\;i{\;\textrm{の濃度}\;\textrm{[mol/m}}^3 \textrm{]}\newline C_i  & :\textrm{出口の成分}\;i{\;\textrm{の濃度}\;\textrm{[mol/m}}^3 \textrm{]}\newline F & :{\textrm{体積流量}\;\textrm{[m}}^3 \textrm{/s]}\newline r_i  & :{\textrm{反応による生成速度}\;\textrm{[mol/(m}}^3 \cdot \textrm{s)]（生成なら＋，消費なら}\textrm{-}\textrm{）}\newline V & :{\textrm{反応器体積　[m}}^3 ] \end{array}$$

定常状態では $\frac{dN_i }{dt}=0$ であるため、

&#160;&#160;&#160;&#160;
$$F(C_{i0} -C_i )+Vr_i =0$$

簡単ですが、これがCSTRの設計方程式です。さらに進めるには、反応の種類を選択する必要があります。

#### 例１：一次反応 A → B の場合

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{A}} =-kC_{\textrm{A}}$$

&#160;&#160;&#160;&#160; であるから

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$F(C_{A0} -C_A )+V(-kC_A )=0$$

&#160;&#160;&#160;&#160;&#160; $\tau =V/Fとして$ 書き直すと

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_A =\frac{C_{A0} }{1+k\tau }$$

 $\tau$ は時間の次元を持つことがわかります。滞留時間もしくは空間時間と呼ばれています。

#### 例２： $A\;\overset{k_1 }{\to} \;D\;\overset{k_2 }{\to} \;U$ の場合

　成分A：先ほどの式がそのまま使える。

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_A =\frac{C_{A0} }{1+k\tau }$$

&#160;&#160;&#160;&#160; 成分D：

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; 反応速度　 $r_D =k_1 C_A -k_2 C_D$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; より


$$C_{D0} -C_D +\tau (k_1 C_A -k_2 C_D )=0$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; 供給反応液は成分Aのみとすると、 $C_{D0} =0$

　　従って、

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_D =\frac{k_1 \tau C_A }{1+k_2 \tau }$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_A =\frac{C_{A0} }{1+k\tau }$ を代入すると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_D =C_{A0} \frac{k_1 \tau }{(1+k_1 \tau )(1+k_2 \tau )}$       Eq. (A)

max関数による、CSTR（A \-> D \-> U）の最大値求解コードは以下の通りです。

実行してみましょう（セクションの実行）。


```matlab
%% CSTR optimal residence time for A -> D -> U
clear; clc;


% Parameters
k1 = 1.0e0;   % 1/s
k2 = 0.1e0;   % 1/s
CA0 = 1.0;  % mol/L


% Define function for CD vs tau
CD = @(tau) (k1 .* tau .* CA0) ./ ((1 + k1 .* tau) .* (1 + k2 .* tau));


% Search range for tau
tau = linspace(0, 20, 500);


% Compute concentration profile
CD_values = CD(tau);


% Find maximum (optimum)
[CDmax, idx] = max(CD_values);
tau_opt = tau(idx);


% Display results
fprintf('Optimal residence time = %.3f s\n', tau_opt);
```

```matlab
fprintf('Maximum CD = %.4f mol/L\n', CDmax);
```

```matlab


% Plot
figure;
plot(tau, CD_values, 'LineWidth', 2);
hold on;
plot(tau_opt, CDmax, 'ro', 'MarkerSize', 8, 'LineWidth', 2);
xlabel('\tau (s)');
ylabel('C_D (mol/L)');
title('CSTR: Intermediate D concentration vs Residence Time');
grid on;
legend('C_D', sprintf('\\tau_{opt} = %.2f s', tau_opt));
```

```matlab
tau_opt_num = 1 / sqrt(k1 * k2);
fprintf('Analytical optimum tau = %.3f s\n', tau_opt_num);
```

# <span style="color:rgb(213,80,0)">課題４(提出不要）</span>

 $C_{A0} \frac{k_1 \tau }{(1+k_1 \tau )(1+k_2 \tau )}$ *を* $\tau$ *について微分し、最適滞留時間* $\tau_{\textrm{opt}}$ *が*

$$\tau_{\textrm{opt}} =\frac{1}{\sqrt{k_1 k_2 }}$$

*となることを、自分のノートに計算し、確認してください（提出不要）。*

MATLABでシンボリックに解を求めると以下のようになります。　MATLABのPC版を使っている人はSymbolic Math Toolboxが必要です。


```matlab
clear; clc;
syms tau k1 k2 CA0 real
assume(k1 > 0);
assume(k2 > 0);
assume(tau > 0);


CD = CA0 * (k1 * tau) / ((1 + k1 * tau) * (1 + k2 * tau));
dCD_dtau = diff(CD, tau);
tau_opt_theory = simplify(solve(dCD_dtau == 0, tau));


disp('Analytical expression for tau_opt:');
```

```matlab
pretty(tau_opt_theory)
```

```matlab


% 数値代入
k1_val = 1.0; k2_val = 0.1; CA0_val = 1.0;
tau_opt_theory = double(subs(tau_opt_theory, {k1, k2, CA0}, {k1_val, k2_val, CA0_val}));
fprintf('\nτ_opt (theory) = %.5f s\n', tau_opt_theory);
```

### 例３：逐次反応（ A\->D\->U）に並行反応( 2A\->P)が加わった場合

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$\begin{array}{l} A\;\overset{k_1 }{\to} \;D\;\overset{k_2 }{\to} \;U\newline A+A\;\overset{k_3 }{\to} \;P \end{array}$$

定常状態における成分Aの物質収支式は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_{\textrm{A0}} -C_{\textrm{A}} +\tau r_{\textrm{A}} =0$$

ここで

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{A}} =-k_1 C_{\textrm{A}} -2k_3 C_{\textrm{A}}^2$$

を代入して整理すると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$2\tau k_3 C_{\textrm{A}}^2 +(1+k_1 \tau )C_{\textrm{A}} -C_{\textrm{A0}} =0$$

の２次方程式となり、解の公式から $C_{\textrm{A}}$ は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_{\textrm{A}} =\frac{-(1+k_1 \tau ).+\sqrt{(1+k_1 \tau )^2 +8k_3 \tau C_{\textrm{A0}} }}{4k_3 \tau }$  \-\-\- Eq. (B)

一方、成分Dの物質収支式は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_{\textrm{D0}} -C_{\textrm{D}} +\tau r_{\textrm{D}} =0$$

 で、 $r_{\textrm{D}}$ は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{D}} =k_1 C_{\textrm{A}} -k_2 \textrm{D}$$

これを整理すると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_{\textrm{D}} =\frac{k_1 \tau }{1+k_2 \tau }C_{\textrm{A}}$   \-\-\-\-  Eq.(C)

結局Eqs. (B),  (C)を連立して、MATLABのmax関数で $\tau_{\textrm{opt}}$ を求めることができる。（はずである）
