## 1. PFRとは
ピストン流れを仮定する連続反応器である。

仮定:
- 半径方向完全混合
- 軸方向混合なし
- 定常状態

## 2. 微小体積の物質収支

$$dF_A=r_A dV$$

## 3. PFR設計方程式

$$\frac{dF_A}{dV}=r_A$$

## 4. 転化率表示

$$F_A=F_{A0}(1-X)$$

$$dF_A=-F_{A0}dX$$

$$V=F_{A0}\int_0^X\frac{dX}{-r_A}$$

## 5. 一次反応

$$-r_A=kC_A$$

$$V=\frac{F_{A0}}{kC_{A0}}\ln\frac{1}{1-X}$$

## 6. 空間時間

$$\tau=\frac{V}{v_0}$$

$$\tau=\frac1k\ln\frac1{1-X}$$

## 7. 回分反応器との対応

回分反応器

$$t=\int_0^X\frac{C_{A0}dX}{-r_A}$$

PFR

$$\tau=\int_0^X\frac{C_{A0}dX}{-r_A}$$

したがって

$$t\leftrightarrow\tau$$

## 8. 逐次反応

$$A\rightarrow B\rightarrow C$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

## 9. PFRの支配方程式

$$\frac{dC_A}{d\tau}=-k_1C_A$$

$$\frac{dC_B}{d\tau}=k_1C_A-k_2C_B$$

$$\frac{dC_C}{d\tau}=k_2C_B$$

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

k1=1.0
k2=0.3

def pfr(tau,y):
    CA,CB,CC=y
    r1=k1*CA
    r2=k2*CB
    return [-r1,r1-r2,r2]

sol=solve_ivp(pfr,[0,10],[1,0,0],dense_output=True)
tau=np.linspace(0,10,400)
CA,CB,CC=sol.sol(tau)
plt.plot(tau,CA,label="A")
plt.plot(tau,CB,label="B")
plt.plot(tau,CC,label="C")
plt.xlabel("Space Time")
plt.ylabel("Concentration")
plt.legend()
plt.grid()
plt.show()
```

## 10. 並列反応と選択率

$$A\rightarrow B$$
$$A\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

$$S_{B/C}=\frac{k_B}{k_C}C_A^{m-n}$$

## 11. Levenspiel流の判定
- m>n : 高濃度有利 → PFR有利
- m<n : 低濃度有利 → CSTR有利

## 12. 例題

$$r_B=3C_A^2$$
$$r_C=2C_A$$

$$S_{B/C}=\frac32C_A$$

高濃度有利なのでPFRまたは回分反応器が有利。

## 演習問題
1. 一次反応のPFR設計式を導出せよ。
2. $$X=0.8$$, $$k=0.5\,min^{-1}$$ のとき空間時間を求めよ。
3. $$r_B=2C_A$$, $$r_C=5C_A^3$$ のとき有利な反応器を答えよ。

## 解答

$$\tau=\frac1k\ln\frac1{1-X}$$

$$\tau=\frac1{0.5}\ln(5)=3.22\,min$$

低濃度有利なのでCSTR。
