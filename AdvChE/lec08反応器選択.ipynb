## 1. 3種類の理想反応器

- Batch Reactor
- PFR
- CSTR

それぞれ濃度履歴が異なるため選択率も異なる。

## 2. 設計式の比較

Batch
$$t=\frac{C_{A0}dX}{-r_A}$$

PFR
$$V=F_{A0}
\int_0^X \frac{dX}{-r_A}$$

CSTR
$$V=\frac{F_{A0}X}{(-r_A)_{out}}$$

## 3. 濃度履歴

Batch：高濃度から低濃度へ変化

PFR：入口から出口へ徐々に変化

CSTR：全体が出口濃度

## 4. 並列反応

$$A\rightarrow B$$
$$A\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 5. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C}
=\frac{k_B}{k_C}C_A^{m-n}$$

## 6. Levenspiel流判定法

m>n ：高濃度有利

→ Batch, PFR有利

m<n ：低濃度有利

→ CSTR有利

## 7. 例1

$$r_B=3C_A^2$$
$$r_C=2C_A$$

$$S_{B/C}=\frac32 C_A$$

高濃度有利 → PFRまたはBatch

## 8. 例2

$$r_B=2C_A$$
$$r_C=5C_A^3$$

$$S_{B/C}=\frac25 C_A^{-2}$$

低濃度有利 → CSTR

## 9. 逐次反応

$$A\rightarrow B\rightarrow C$$

Bは中間生成物である。

## 10. B最大条件

$$\frac{dC_B}{dt}=0$$

$$k_1C_A=k_2C_B$$

## 11. 逐次反応に有利な反応器

中間生成物Bを回収したい場合、BatchやPFRが有利なことが多い。

CSTRではBがさらにCへ変換されやすい。

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.05,2,300)
S=1.5*CA
plt.plot(CA,S)
plt.xlabel("C_A")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
```

## 12. 反応器選定フローチャート

1. 選択率式を求める
2. 濃度依存性を調べる
3. 高濃度有利か低濃度有利か判定
4. 反応器を選択

## 演習問題

1. $$r_B=4C_A^2,\ r_C=C_A$$ のとき有利な反応器を答えよ。

2. $$r_B=C_A,\ r_C=4C_A^3$$ のとき有利な反応器を答えよ。

3. 逐次反応でB最大条件を導出せよ。

4. Batch・PFR・CSTRの濃度履歴の違いを説明せよ。

## 解答

1. PFRまたはBatch

2. CSTR

3. $$k_1C_A=k_2C_B$$

4. BatchとPFRは高濃度から低濃度へ変化，CSTRは出口濃度で一様。

## 理論
瞬間選択率 $$S_{B/C}=r_B/r_C$$ を用いて反応器を選定する。

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.05,2.0,200)
S=(3/2)*CA
plt.plot(CA,S)
plt.xlabel("CA")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
print("High concentration is favorable")
print("Recommended reactor: PFR or Batch")
```

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.05,2.0,200)
S=(2/5)*CA**(-2)
plt.plot(CA,S)
plt.xlabel("CA")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
print("Low concentration is favorable")
print("Recommended reactor: CSTR")
```

```python
def recommend_reactor(m,n):
    if m>n:
        return "PFR or Batch"
    elif m<n:
        return "CSTR"
    return "No preference"

for m,n in [(2,1),(1,3),(2,2)]:
    print(f"m={m}, n={n} --> {recommend_reactor(m,n)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.1,2.0,200)
plt.plot(CA,CA,label="m>n")
plt.plot(CA,CA**(-1),label="m<n")
plt.xlabel("CA")
plt.ylabel("S_B/C")
plt.legend()
plt.grid()
plt.show()
```

```python
## 逐次反応
```

```python
import numpy as np
import matplotlib.pyplot as plt
k1=1.0
k2=0.3
CA0=1.0
t=np.linspace(0,10,400)
CA=CA0*np.exp(-k1*t)
CB=k1*CA0/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
plt.plot(t,CB)
plt.xlabel("time")
plt.ylabel("CB")
plt.grid()
plt.show()
imax=np.argmax(CB)
print("Maximum B concentration=",CB[imax])
print("at time=",t[imax])
```
