## 1. 転化率

$$X=\frac{F_{A0}-F_A}{F_{A0}}$$

回分系では

$$X=\frac{C_{A0}-C_A}{C_{A0}}$$

## 2. 収率

目的生成物Bの収率

$$Y_B=\frac{	\text{生成したB}}{	\text{供給したA}}$$

## 3. 総合選択率

$$\bar S_{B/C}=\frac{N_B}{N_C}$$

または

$$\bar S_{B/C}=\frac{F_B}{F_C}$$

## 4. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C}$$

瞬間的な生成速度比を表す。

## 5. 並列反応

$$A
\rightarrow B$$
$$A
\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 6. 選択率と濃度

$$S_{B/C}=\frac{k_B}{k_C}C_A^{m-n}$$

- m>n : 高濃度有利
- m<n : 低濃度有利

## 7. Levenspiel流の反応器選定

高濃度有利 → PFR・回分反応器

低濃度有利 → CSTR

## 8. 例題1

$$r_B=3C_A^2$$
$$r_C=2C_A$$

$$S_{B/C}=\frac{3}{2}C_A$$

高濃度有利なのでPFRまたは回分反応器が有利。

## 9. 例題2

$$r_B=2C_A$$
$$r_C=5C_A^3$$

$$S_{B/C}=\frac{2}{5}C_A^{-2}$$

低濃度有利なのでCSTRが有利。

## 10. 逐次反応

$$A
\rightarrow B
\rightarrow C$$

目的生成物Bは中間生成物である。

## 11. B最大条件

$$\frac{dC_B}{dt}=0$$

$$k_1C_A=k_2C_B$$

この時Bは極大値を持つ。

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.1,2,200)
S=1.5*CA
plt.plot(CA,S)
plt.xlabel("CA")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
```

## 演習問題
1. 転化率・収率・選択率の違いを説明せよ。
2. r_B=4C_A^2, r_C=C_A の場合の有利な反応器を答えよ。
3. r_B=C_A, r_C=10C_A^3 の場合の有利な反応器を答えよ。
4. A→B→C においてB最大条件を導出せよ。

## 解答
1. 転化率はAの消費割合、収率は目的生成物生成量、選択率は生成物比。
2. PFRまたは回分反応器。
3. CSTR。
4. k1CA=k2CB。
