# 第4回 転化率・収率・選択率と反応器選定



第4回では、反応工学において非常に重要な3つの指標である転化率・収率・選択率について学びます。<br>
化学の大学院生には説明するまでもない記述がいくつかありますが、定義をきちんとしておきたいという思いで書いております。<br>
その点は我慢してみてください。

- 転化率は「反応物がどれだけ反応したか」、
- 収率は「目的生成物がどれだけ得られたか」、
- 選択率は「目的生成物をどれだけ優先的に作れたか」

を表します。<br>
特に実際の化学プロセスでは、副反応によって不要な生成物が生じることが多く、単に転化率を高くするだけでは十分ではありません。<br>
転化率99%でも、収率が例えば0.5%ではプロセスとしてなりたちません。<br>
目的物を効率よく生産するためには、収率や選択率の評価が不可欠です。<br>
また、並列反応や逐次反応を例に、それぞれの反応系で収率や選択率がどのように変化するかを確認します。<br>
さらに後半では、これらの考え方を反応器設計に応用し、反応器選定法について学びます。<br>

今日の講義を通して、『どれだけ反応したか』だけでなく、『何をどれだけ作れたか』という視点で反応を評価できるようになることを目標とします。

## 学習目標
- 転化率・収率・選択率を厳密に定義できる
- 瞬間選択率と総合選択率を区別できる
- 並列反応・逐次反応で評価できる
- Levenspiel流の反応器選定法を理解する

<h1>並列反応の場合</h1>

---

$$
\text{A} \xrightarrow{r_1} \text{B}
$$

$$
\text{A} \xrightarrow{r_2} \text{C}
$$
- $\text{A}$:反応物質
- $\text{B}$:目的生成物質
- $\text{C}$:副生成物

## 1. 転化率
流通系(PFRやCSTR)：
$$X=\frac{F_{A0}-F_A}{F_{A0}} \tag{1}$$

回分系：

$$X=\frac{N_{A0}-N_A}{N_{A0}} \tag{2}$$

◇◇◇


<details><summary>📝🖱️説明文はこちら</summary>

まず、反応工学で最も基本的な指標である転化率を定義します。<br>
簡単のため、上式の並列反応を対象とします。<br>
ここでは$\text{A}$：原料、$\text{B}$：目的生成物、$\text{C}$:副生成物、として説明していきます。

**転化率**とは、「反応物がどれだけ反応したか」を表す指標です。<br>
例えば、100 molのAを反応器に投入し、そのうち80 molが反応した場合、転化率は80％になります。<br>
流通式反応器では、入口のモル流量を $F_{A0}$、出口のモル流量を $F_A$ とすると、(1)式で定義されます。<br>
一方、回分反応器では流入や流出がないため、モル流量の代わりに物質量を使い(2)で表されます。<br>
つまり転化率とは、「最初にあったAのうち、どれだけが反応したか」を割合で示したものです。

</details>

---

## 2. 収率と選択率

<h3>2.1. 流通系の収率と選択率</h3>

<h5>表１：流通系（連続流れ反応器）における収率・選択率</h5>

| | 瞬間 (Instantaneous) | 総括・供給原料基準 (Overall, Feed) | 総括・反応原料もしくは副生成物基準 (Overall, Reacted) |
| :--- | :---: | :---: | :---: |
| **収率 (Yield)** | $Y'_B = \dfrac{r_B}{-r_A}$ | $Y_B = \dfrac{F_B - F_{B0}}{F_{A0}}$ | $\Phi_B = \dfrac{F_B - F_{B0}}{F_{A0} - F_A}$ |
| **選択率 (Selectivity)** | $S'_{B/C} = \dfrac{r_B}{r_C}$ | $-$| $$S_{B/C} = \dfrac{F_B - F_{B0}}{F_C - F_{C0}}$$ |

◇◇◇

ここで転化率、収率、選択率の定義を体系的に整理します。<br>
まず、評価方法は大きく 瞬間値 と 総括値 に分けられます。<br>
瞬間値は「今この瞬間の反応の様子」を表し、総括値は「反応の結果として最終的にどれだけ得られたか」を表します。<br>

**収率**について見ると、瞬間収率
$$
Y'_B=\frac{r_B}{-r_A}
$$
は、反応物Aが消費される速度に対して、Bが生成する速度の割合を表しています。<br>
つまり、消費されたAが今どれだけBになっているかを示す指標です。<br>
一方、総括収率
$$
Y_B=\frac{F_B-F_{B0}}{F_{A0}}
$$
は、**供給したA全体に対してどれだけBが得られたか**を示します。<br>
これは実際の生産量を評価するときに重要な指標です。<br>
分母を$F_{A0}-F_A$とした収率$\Phi_B$、つまり反応した原料基準の収率も定義できます。<br>

次に**選択率**を見てみます。<br>
選択率の定義は教科書によってまちまちですが、本講義では、生成物のうち目的物の割合を選択率とします。

瞬間選択率（参考）
$$
S'_{B/C}=\frac{r_B}{r_C}
$$
は、**ある瞬間にBとCのどちらが優先的に生成されているか**を表します。<br>
並列反応の反応器選定では、この瞬間選択率が重要な役割を果たします。<br>

一方、総括選択率は
$$
S_{B/C} = \frac{F_B-F_{B0}} {F_C-F_{C0}}
$$
で定義されます。<br>
値が大きいほど、目的生成物が優先的に生成されていることを意味します。<br>
このように、瞬間値は反応器や運転条件の選定に利用し、総括値は最終的なプロセス性能の評価に利用します。<br>
つまり、瞬間収率・瞬間選択率は設計のための指標、総括収率・総括選択率は結果を評価する指標と理解しておくと整理しやすいでしょう。

## 例題：瞬間値と総括値の役割を考えてみよう

### 解く必要はありません。読むだけで結構です

---

以下の並列反応系の反応装置を考える。

$$ \begin{aligned} 
A &\xrightarrow{r_1} B\\\ 
A &\xrightarrow{r_2} C 
\end{aligned} 
$$

反応速度式は次のとおりである。

$$ r_1=k_1 C_A,\qquad r_2=k_2 C_A $$

ここで、$r_1,r_2$ は各反応の速度、$C_A$ は原料 $A$ の濃度である。いずれも一次反応とする。

ある条件で反応器を運転したところ、次の結果が得られた。

| 成分                 | 供給量・出口量     |
| ------------------ | ----------- |
| 原料 $A$ の供給量    | 100 mol/min |
| $A$ の出口量       | 20 mol/min  |
| 目的生成物 $B$ の出口量 | 48 mol/min  |
| 副生成物 $C$ の出口量  | 32 mol/min  |

### 問1　総括値による性能評価

まず、反応器全体の性能を評価しなさい。

(1) 原料 $A$ の転化率 $X_A$ を求めなさい。

(2) 目的生成物 $B$ の総括収率 $Y_{B}$ を求めなさい。

$$ Y_{B}=\frac{F_B}{F_{A0}} $$

(3) $B$ の $C$ に対する総括選択率 $S_{B/C}$ を求めなさい。

$$ S_{B/C}=\frac{F_B}{F_C} $$

(4) 転化率が80%であるにもかかわらず、目的生成物の総括収率が48%にとどまっている。プロセスの問題点を説明しなさい。

<details>
<summary>解答例</summary>
(1) 転化率

$$ X_A=\frac{100-20}{100}=0.80 $$

したがって、転化率は 80%。

(2) 総括収率

$$ Y_{B}=\frac{48}{100}=0.48 $$

したがって、総括収率は 48%。

(3) 総括選択率

$$ S\_{B/C}=\frac{48}{32}=1.50 $$

(4) 考察

原料 $A$ は80%転化しているが、供給量に対する目的生成物 $B$ の収率は48%である。一方、副生成物 $C$ が32 mol/min生成している。

したがって、原料を反応させることだけでなく、反応経路を目的生成物 $B$ の生成側へ導くことが課題である。
</details>

### 問2　瞬間値による原因分析

反応器内のある位置で、次の反応速度が得られた。

| 項目                       | 反応速度           |
| ------------------------ | -------------- |
| $A$ の消費速度 $-r_A$ | 10 mol/(L·min) |
| $B$ の生成速度 $r_B$  | 6 mol/(L·min)  |
| $C$ の生成速度 $r_C$  | 4 mol/(L·min)  |

(1) $B$ の瞬間収率を求めなさい。

$$ Y^\prime_{B}=\frac{r_B}{-r_A} $$

(2) $B$ の $C$ に対する瞬間選択率を求めなさい。

$$ S^\prime_{B/C}=\frac{r_B}{r_C} $$

(3) 問1の総括値と比較し、この反応器内の位置では、目的生成物 $B$ と副生成物 $C$ がどのような割合で生成しているか説明しなさい。



<details>
<summary>解答例</summary>
(1) 瞬間収率

$$ Y^\prime_{B}=\frac{6}{10}=0.60 $$

したがって、瞬間収率は 60%。

(2) 瞬間選択率

$$ S^\prime_{B/C}=\frac{6}{4}=1.50 $$

したがって、瞬間選択率は 1.50。

(3) 考察

この位置では、消費された $A$ に対応する反応速度の60%が $B$ の生成、40%が $C$ の生成に相当する。

総括選択率も1.50であるため、この測定値だけでは、反応器全体で選択率が低下しているとは判断できない。ほかの位置や条件での反応速度も調べる必要がある。


</details>

このような評価結果により、改善策について検討することが可能となる。

改善策の例

| 改善策       | 期待される効果                             | 注意点                   |
| --------- | ----------------------------------- | --------------------- |
| ① 温度変更    | 反応速度定数の比 $k_1/k_2$ を変えられる可能性がある | 両反応の温度依存性を確認する        |
| ② 触媒変更    | 目的反応を選択的に促進できる可能性がある                | 触媒活性・寿命・コストも評価する      |
| ③ 反応器形式変更 | 濃度や滞留時間の分布を変えられる                    | 一次並列反応では、選択率改善に限界がある  |
| ④ 原料濃度変更  | 反応速度そのものを変えられる                      | 同じ次数の一次反応では、選択率は変わらない |

この系では、特に温度変更や触媒変更によって $k_1/k_2$ を大きくすることが、選択率改善の基本方針となる。

(4) 総括値で評価する理由

瞬間値は特定の位置・条件での反応速度の比であり、プロセス全体の生産性能を直接表すものではない。改善後に総括収率と総括選択率を測定すれば、実際に目的生成物 $B$ の生産量が増え、副生成物 $C$ が抑制されたかを確認できる。

この演習の要点は、総括値で問題を発見し、瞬間値で原因を分析し、改善策を実施した後、再び総括値で効果を評価することです。

<h3>2.2. 回分系の収率と選択率</h3>

<h5>表２：回分系（バッチ反応器）における収率・選択率</h5>

| | 瞬間 (Instantaneous) | 総括・供給原料基準 (Overall, Feed) | 総括・反応原料もしくは副生成物基準 (Overall, Reacted) |
| :--- | :---: | :---: | :---: |
| **収率 (Yield)** | $Y'_B = \dfrac{r_B}{-r_A}$ | $Y_B = \dfrac{N_B - N_{B0}}{N_{A0}}$ | $\Phi_B = \dfrac{N_B - N_{B0}}{N_{A0} - N_A}$ |
| **選択率 (Selectivity)** | $S'_{B/C} = \dfrac{r_B}{r_C}$ | $-$ | $$S_{B/C} = \dfrac{N_B - N_{B0}}{N_C - N_{C0}}$$ |

この表は、先ほどの流通系の場合を回分系へ置き換えたものです。<br>
流通系では物質量流量 $F$ を用いていましたが、回分系では流入・流出がないため、物質量 $N$ を用います。<br>
基本的な考え方は流通系と同じです。<br>
瞬間収率$Y'_B$および瞬間選択率$S'_{B/C}$は反応器に依存せず、共通です。<br>
一方、総括収率
$$
Y_B=\frac{N_B-N_{B0}}{N_{A0}}
$$
は、仕込んだA全体に対してどれだけBが得られたかを示します。<br>
さらに、
$$
\Phi_B = \frac{N_B-N_{B0}} {N_{A0}-N_A}
$$
は、反応したAのうちどれだけがBになったかを表す指標です。<br>

選択率は、

$$
S_{B/C} = \frac{N_B-N_{B0}} {N_C-N_{C0}}
$$
で定義されます。<br>
つまり回分系でも、瞬間値は反応の進み方を評価する指標、総括値は最終的な反応結果を評価する指標と理解しておけば十分です。

## 3. 転化率、収率の関係式

原料 A の転化率 $X_A$ はモル流量を用いて以下のように表されます。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

これを用いると、2つの式の関係は次のように綺麗に整理できます。

$$\underbrace{\frac{F_B - F_{B0}}{F_{A0}}}_{\text{供給原料基準の収率 } Y_B} = \underbrace{\left( \frac{F_{A0} - F_A}{F_{A0}} \right)}_{\text{転化率 } X_A} \times \underbrace{\left( \frac{F_B - F_{B0}}{F_{A0} - F_A} \right)}_{\text{（反応原料基準の収率）}}$$

$$\therefore Y_B = X_A \cdot \Phi_B$$

以上は量論係数が１の場合ですが、それ以外の場合について、次に説明していきます。

---




## 4. 量論係数が１でない場合

一般化された化学量論係数 $a, b, d, c$ を持つ並列反応系

$$a\text{A} \xrightarrow{r_1} b\text{B} \quad (\text{反応1})$$

$$d\text{A} \xrightarrow{r_2} c\text{C} \quad (\text{反応2})$$

を**連続流通系反応器**で実施する場合の、転化率・選択率・収率の一般的な定義を解説します。

流通系のモル流量を以下のように定義します。

* 供給時のAのモル流量：$F_{A0}$
* 反応器出口のモル流量：$F_A$（未反応A）、$F_B$（生成物B）、$F_C$（生成物C）

---

### 4.1. 転化率（Conversion: $X_A$）

反応物の種類や**化学量論係数に関わらず**、供給されたAのうち実際に反応した割合です。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

---



### 4.2. 収率（Yield: $Y_i$ および $\Phi_i$）

**① 供給原料基準の収率（$Y_i$）** と **② 反応原料基準の収率（$\Phi_i$）** の2つの表し方について説明します。

* **反応1**：$a\text{A} \rightarrow b\text{B}$ （1モルの $\text{B}$ を作るには $\frac{a}{b}$ モルの $\text{A}$ が必要）
* **反応2**：$d\text{A} \rightarrow c\text{C}$ （1モルの $\text{C}$ を作るには $\frac{d}{c}$ モルの $\text{A}$ が必要）

---

#### 4.2.1. 供給原料基準の収率 ($Y_B, Y_C$)

「供給したすべての原料A」に対して、最終的に目的の生成物等に変換された割合を表します（プロセス全体の生産性を評価する指標です）。分母には**供給された総モル流量 $F_{A0}$** を用います。

* **$\text{B}$ の収率 ($Y_B$)**：

$$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$


* **$\text{C}$ の収率 ($Y_C$)**：

$$Y_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0}} = \frac{d (F_C - F_{C0})}{c F_{A0}}$$



---

#### 4.2,2. 反応原料基準の収率（$\Phi_B, \Phi_C$）

「実際に反応した原料A」に対して、どれだけ目的の生成物に変換されたかを表す指標です。分母には実際に反応したモル流量（$F_{A0} - F_A$）を用います。
（※この反応原料基準の収率は、後述する「選択率（反応原料基準 $S_{B/A}$）」と完全に一致します）

* **$\text{B}$ の反応原料基準収率（選択率） ($\Phi_B$ / $S_{B/A}$)**：

$$\Phi_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$


* **$\text{C}$ の反応原料基準収率（選択率） ($\Phi_C$ / $S_{C/A}$)**：

$$\Phi_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0} - F_A} = \frac{d (F_C - F_{C0})}{c (F_{A0} - F_A)}$$



> **💡 補足（収率と転化率・選択率の関係）**
> 供給原料基準の収率（$Y_B$）は、転化率（$X_A$）と反応原料基準の収率（$\Phi_B$）の積として表すことができます。
> 
> $$Y_B = X_A \times \Phi_B \quad (\text{収率} = \text{転化率} \times \text{選択率})$$
> 
>

---


### 4.3. 選択率（Selectivity: $S_{B/C}$）

反応器から得られた出力（生成物）に着目し、目的生成物 $\text{B}$ と副生成物 $\text{C}$ がどのような比率で生成されたかを表す指標です。各反応の化学量論係数（$a, b, d, c$）を考慮し、**消費された原料 $\text{A}$ の当量換算**で比をとります。

* **反応1**：$a\text{A} \xrightarrow{r_1} b\text{B}$ （1モルの $\text{B}$ を作るには $\frac{a}{b}$ モルの $\text{A}$ が必要）
* **反応2**：$d\text{A} \xrightarrow{r_2} c\text{C}$ （1モルの $\text{C}$ を作るには $\frac{d}{c}$ モルの $\text{A}$ が必要）

これらに基づき、$\text{B}$ と $\text{C}$ の選択率の比（生成比 $S_{B/C}$）は次のように定義されます。

$$S_{B/C} = \frac{\frac{a}{b} F_B}{\frac{d}{c} F_C} = \frac{a c F_B}{b d F_C}$$

> **💡 ポイント**
> 原料 $\text{A}$ の消費量（$F_{A0} - F_A$）が分子・分母で相殺されるため、**「最終的な出口のモル流量 $F_B$ と $F_C$ の比」に化学量論係数からの補正項（$\frac{ac}{bd}$）を掛け合わせたもの**としてシンプルに表されます。
> （※ もしすべての係数 $a, b, d, c$ が $1$ の場合は、単純に $S_{B/C} = \frac{F_B}{F_C}$ となります）

### 4.4. まとめ

一般化された化学量論係数 $a, b, d, c$ を持つ並列反応系

$$a\text{A} \xrightarrow{r_1} b\text{B} \quad (\text{反応1})$$

$$d\text{A} \xrightarrow{r_2} c\text{C} \quad (\text{反応2})$$

における、転化率、収率、選択率（連続流通系）

- 転化率
$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

- 供給原料基準の収率 ($Y_B, Y_C$)
  -  **$\text{B}$ の収率 ($Y_B$)**：
    $$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$
  - **$\text{C}$ の収率 ($Y_C$)**：

<h1>逐次反応の場合</h1>

---

化学量論係数を含む**逐次反応）**

$$a\text{A} \xrightarrow{r_1} b\text{B} \xrightarrow{r_2} c\text{C}$$

（$\text{A}$: 原料、$\text{B}$: 中間生成物、$\text{C}$: 最終生成物）を**連続流通系反応器**で実施する場合の、転化率・収率・選択率の定義を解説します。

---

### 1. 転化率（Conversion: $X_A$）

反応の起点となる原料Aの消費割合です。後続の反応（B $\rightarrow$ C）に関わらず、**供給されたAのうち実際に反応した割合**として、並列反応や単一反応と同じ定義になります。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

* $F_{A0}$：供給時のAのモル流量
* $F_A$：出口の未反応Aのモル流量

---

### 2. 収率（Yield: $Y_B, Y_C$）

#### 2.1. 供給原料基準の収率 ($Y_B, Y_C$)

供給されたすべての原料Aの総モル流量 $F_{A0}$ に対し、出口で得られた各生成物の正味のモル流量（$F_B - F_{B0}, F_C - F_{C0}$）の割合です。各段階の化学量論係数を考慮して補正します。

#### ① 中間生成物 B の収率 ($Y_B$)

反応1（$a\text{A} \rightarrow b\text{B}$）より、1モルのBを作るには $\frac{a}{b}$ モルのAが必要です。


$$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$

#### ② 最終生成物 C の収率 ($Y_C$)

反応1（$a\text{A} \rightarrow b\text{B}$）と反応2（$b\text{B} \rightarrow c\text{C}$）の全体の関係から、**1モルのCを作るために必要な原料Aのモル数**は $\frac{a}{c}$ です。<br>
したがって、Cの供給原料基準の収率 $Y_C$ は次のように表されます。

$$Y_C = \frac{\frac{a}{c} (F_C - F_{C0})}{F_{A0}} = \frac{a (F_C - F_{C0})}{c F_{A0}}$$


---

#### 2.2. 反応原料基準の収率（$\Phi_B, \Phi_C$）

反応原料基準の収率 $\Phi$ は、**「実際に反応して消費された原料A」のうち、どれだけの割合がそれぞれの物質（BまたはC）に変換されているか**を表す指標です。
分母には必ず「実際に反応したAのモル流量（$F_{A0} - F_A$）」を用います。

#### ① 中間生成物 B の反応原料基準収率（$\Phi_B$）

反応1（$a\text{A} \rightarrow b\text{B}$）の化学量論係数（1モルのBを作るのに必要なAは $\frac{a}{b}$ モル）を考慮します。


$$\Phi_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$

* **意味**: 実際に反応したAのうち、**現在まだ「Bの状態で残っている」ものの割合**です。

#### ② 最終生成物 C の反応原料基準収率（$\Phi_C$）



反応1（$a\text{A} \rightarrow b\text{B}$）と反応2（$b\text{B} \rightarrow c\text{C}$）の化学量論関係から、「1モルのCを作るのに必要な原料Aのモル数」は $\frac{a}{c}$ となります。
（※Aが $a$ モル反応するとBが $b$ モルでき、そのBがすべて反応するとCが $c$ モルできるため、$a$ モルのAから最終的に $c$ モルのCが生まれる関係になります）

したがって、Cの反応原料基準の収率 $\Phi_C$ は次のように表されます。

$$\Phi_C = \frac{\frac{a}{c} (F_C - F_{C0})}{F_{A0} - F_A} = \frac{a (F_C - F_{C0})}{c (F_{A0} - F_A)}$$


---



### 2. 選択率（Selectivity: $S_{B/C}$ など）

逐次反応において選択率という言葉はピンと来ませんが、ここでは「反応した原料Aのうち、どれだけが目的の段階（B）にとどまっているか」を選択率として用います。

#### 反応原料基準の B の選択率（$S_B$）

実際に反応したAの量（$F_{A0} - F_A$）のうち、どれだけがBの生成・存在に使われているかを表します。


$$S_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$

* **収率との関係**：
並列反応と同様に、以下の関係が成り立ちます。

$$Y_B = X_A \times S_B \quad (\text{収率} = \text{転化率} \times \text{選択率})$$



---

### 💡 逐次反応における重要な特徴

逐次反応の最大の特徴は、中間生成物である **B の収率（$Y_B$）や選択率（$S_B$）が、反応時間（または転化率 $X_A$）に対して「山なりの曲線（極大値）」を描く**点です。

1. **反応初期**：A $\rightarrow$ B（$r_1$）が優勢なため、$Y_B$ は上昇する。
2. **反応後期**：B $\rightarrow$ C（$r_2$）が進むため、Bが消費されて $Y_B$ は低下し、最終的に $C$ が増える。

そのため、逐次反応のプロセス設計では、「Bが最も多く得られる最適滞留時間（または最適転化率）」を見極めることが非常に重要になります。

# 模範解答：逐次反応PFRの転化率・収率・選択率（完全解法版）

## (1) PFRの物質収支式

反応器内の体積流量 $v$ は一定であるため、成分 $i$ のモル流量 $F_i$ と濃度 $C_i$ の間には $F_i = v C_i$ の関係が成り立ちます。

PFRの物質収支の基礎式 $\frac{dF_i}{dV} = r_i^{\mathrm{net}}$ に各成分の生成・消費速度を代入すると、以下の微分方程式が得られます。

* **成分 A:**

$$\frac{dF_A}{dV} = -2 r_1 = \mathbf{-2 k_1 C_A^2}$$


* **成分 B:**

$$\frac{dF_B}{dV} = r_1 - r_2 = \mathbf{k_1 C_A^2 - k_2 C_B}$$


* **成分 C:**

$$\frac{dF_C}{dV} = 3 r_2 = \mathbf{3 k_2 C_B}$$



また、空間時間 $\tau = \frac{V}{v}$（$dV = v\,d\tau$）を用いると、濃度 $C_i = \frac{F_i}{v}$ に関する微分方程式に変換できます。

$$\frac{dC_A}{d\tau} = -2 k_1 C_A^2$$

$$\frac{dC_B}{d\tau} = k_1 C_A^2 - k_2 C_B$$

$$\frac{dC_C}{d\tau} = 3 k_2 C_B$$

---

## (2) Aの転化率 $X_A$ の導出

初期濃度（入口濃度）は $C_{A0} = \frac{F_{A0}}{v} = \frac{10\ \mathrm{mol/min}}{2\ \mathrm{L/min}} = 5.0\ \mathrm{mol/L}$ です。

Aについての微分方程式を標準的な変数分離法で積分します。

$$-\int_{C_{A0}}^{C_A} \frac{dC_A}{C_A^2} = 2 k_1 \int_{0}^{\tau} d\tau \implies \frac{1}{C_A} - \frac{1}{C_{A0}} = 2 k_1 \tau$$

したがって、濃度 $C_A(\tau)$ およびモル流量 $F_A(\tau)$ は以下のようになります。

$$C_A(\tau) = \frac{C_{A0}}{1 + 2 k_1 C_{A0} \tau}$$

$$F_A(\tau) = v C_A(\tau) = \frac{F_{A0}}{1 + 2 k_1 C_{A0} \tau}$$

与えられた値を代入します：

* $2 k_1 C_{A0} \tau = 2 \times 0.10\ \mathrm{L/(mol\cdot min)} \times 5.0\ \mathrm{mol/L} \times 2.0\ \mathrm{min} = 2.0$

出口での A のモル流量：


$$F_A = \frac{10}{1 + 2.0} = \frac{10}{3} \approx \mathbf{3.33\ \mathrm{mol/min}}$$

転化率 $X_A$：


$$X_A = \frac{F_{A0} - F_A}{F_{A0}} = \frac{10 - 3.33}{10} = \mathbf{0.667 \quad (66.7\%)}$$

---

## (3) Bの収率 $Y_B$ の導出

Bについての微分方程式に (2) で求めた $C_A(\tau)$ を代入します。

$$\frac{dC_B}{d\tau} + k_2 C_B = k_1 C_A(\tau)^2 = \frac{k_1 C_{A0}^2}{(1 + 2 k_1 C_{A0} \tau)^2}$$

これは 1 階線形微分方程式です。積分因子 $e^{k_2 \tau}$ を掛けて変形します。

$$\frac{d}{d\tau}\left( C_B e^{k_2 \tau} \right) = \frac{k_1 C_{A0}^2 e^{k_2 \tau}}{(1 + 2 k_1 C_{A0} \tau)^2}$$

両辺を $0$ から $\tau$ まで積分します（初期条件：$\tau = 0$ で $C_B = 0$）。

$$C_B(\tau) = k_1 C_{A0}^2 e^{-k_2 \tau} \int_{0}^{\tau} \frac{e^{k_2 t}}{(1 + 2 k_1 C_{A0} t)^2} \, dt$$

具体的に数値（$k_1 C_{A0} = 0.50\ \mathrm{min^{-1}}$, $k_2 = 0.20\ \mathrm{min^{-1}}$, $k_1 C_{A0}^2 = 2.50\ \mathrm{mol/(L\cdot min)}$）を代入して積分を評価します。

$$C_B(2) = 2.50 \cdot e^{-0.40} \int_{0}^{2} \frac{e^{0.20 t}}{(1 + t)^2} \, dt$$

被積分関数を数値積分（あるいは近似解法）で計算すると：


$$\int_{0}^{2} \frac{e^{0.20 t}}{(1 + t)^2} \, dt \approx 0.7303$$

したがって、出口での B の濃度およびモル流量は：


$$C_B = 2.50 \times 0.6703 \times 0.7303 \approx 1.224\ \mathrm{mol/L}$$

$$F_B = v C_B = 2\ \mathrm{L/min} \times 1.224\ \mathrm{mol/L} \approx \mathbf{2.45\ \mathrm{mol/min}}$$

収率 $Y_B$：


$$Y_B = \frac{F_B - F_{B0}}{F_{A0}} = \frac{2.45 - 0}{10} = \mathbf{0.245 \quad (24.5\%)}$$

---

## (4) Cに対するBの選択率 $S_{B/C}$

A の原子保存則（化学量論関係）より、消費された A から生成した B と C のモル数は以下を満たします。

$$\text{消費された A のモル流量} = F_{A0} - F_A = 2 F_B + \frac{2}{3} F_C$$

これより、出口での C のモル流量 $F_C$ は次のように求まります。

$$F_C = \frac{3}{2} \left( F_{A0} - F_A - 2 F_B \right)$$

求めた値を代入します。


$$F_C = \frac{3}{2} \left( 10 - 3.333 - 2(2.448) \right) = \frac{3}{2} (10 - 3.333 - 4.896) = \frac{3}{2} \times 1.771 \approx \mathbf{2.66\ \mathrm{mol/min}}$$

したがって、選択率 $S_{B/C}$ は：

$$S_{B/C} = \frac{F_B - F_{B0}}{F_C - F_{C0}} = \frac{2.45}{2.66} \approx \mathbf{0.921 \quad (92.1\%)}$$

## (5) 考察

### 1. 転化率・収率・選択率の概念的違い

* **転化率（$X_A$）：** 反応器に供給された原料 A のうち、反応によって**消費された割合**（未反応原料の残り具合の指標）。
* **収率（$Y_B$）：** 供給した原料 A に対し、**目的生成物 B がどれだけ得られたか**の割合（全体的な生産効率の指標）。
* **選択率（$S_{B/C}$）：** 反応で生成した生成物同士（BとC）の比率。**消費された原料が副生成物ではなく目的物にどれだけ効率よく変換されたか**の指標。

### 2. 空間時間が長くなると $S_{B/C}$ が低下する理由

逐次反応（$2\mathrm{A} \to \mathrm{B} \to 3\mathrm{C}$）において、中間体 B は反応が進行するにつれて副生成物 C へと不可逆的に分解・変換されます。
空間時間 $\tau$ が長くなると、反応器内での滞留時間が延びるため B の分解反応が進み、$F_B$ が減少して $F_C$ が増加します。その結果、比率である $S_{B/C} = \frac{F_B}{F_C}$ は低下します。

### 3. 並列反応の「瞬間生成速度比」との違い

* **並列反応の瞬間生成速度比（$\frac{r_B}{r_C}$）：** 反応器内の「ある特定の地点（局所的）」における生成速度の比率。
* **今回の選択率（$S_{B/C}$）：** 反応器入口から出口までの「全行程で蓄積された」最終的な流出モル数の比率（積分量）。
逐次反応では各地点で生成速度比がダイナミックに変化するため、局所的な速度比と出口での累計モル比（選択率）は一致しません。

### 【解答例】

#### 1. 原料 A の転化率 $X_A$ の計算

$$X_A = \frac{F_{A0} - F_A}{F_{A0}} = \frac{100 - 10}{100} = \frac{90}{100} = \mathbf{0.90 \quad (90\%)}$$

---

#### 2. 目的生成物 B の全選択率 $\Phi_B$ の計算

全選択率は「反応して消費された A のうち、B の生成に使われた A の割合」です。
$2\text{A} \longrightarrow \text{B}$ より、B が 1 mol 生成するのに A は 2 mol 消費されるため、分子に量論補正の **2** を掛けます。

$$\Phi_B = \frac{2(F_B - F_{B0})}{F_{A0} - F_A} = \frac{2 \times (30 - 0)}{100 - 10} = \frac{60}{90} = \frac{2}{3} \approx \mathbf{0.667 \quad (66.7\%)}$$

---

#### 3. 関係式 $Y_B = X_A \cdot \Phi_B$ による B の収率 $Y_B$ の算定

関係式 $Y_B = X_A \cdot \Phi_B$ に、問1・問2で求めた値を代入します。

$$Y_B = X_A \cdot \Phi_B = 0.90 \times \frac{2}{3} = \mathbf{0.60 \quad (60\%)}$$

> **【直接定義式による検証】**
> 
> $$Y_B = \frac{2(F_B - F_{B0})}{F_{A0}} = \frac{2 \times 30}{100} = \frac{60}{100} = 0.60 \quad (60\%)$$
> 
> 
> 
> 定義式から直接求めた値と完全一致します。

---

#### 4. 副生成物 C の収率 $Y_C$ の算定と整合性の確認

$3\text{A} \longrightarrow 2\text{C}$（$\frac{3}{2}\text{A} \longrightarrow \text{C}$）より、C が 1 mol 生成するのに A は **$\frac{3}{2}$（1.5）mol** 消費されます。

* **C の全選択率 $\Phi_C$:**

$$\Phi_C = \frac{\frac{3}{2}(F_C - F_{C0})}{F_{A0} - F_A} = \frac{1.5 \times (20 - 0)}{100 - 10} = \frac{30}{90} = \frac{1}{3} \approx 0.333 \quad (33.3\%)$$


* **C の収率 $Y_C$:**

$$Y_C = X_A \cdot \Phi_C = 0.90 \times \frac{1}{3} = \mathbf{0.30 \quad (30\%)}$$



#### 【整合性の確認】

1. **全選択率の和 ($\Phi_B + \Phi_C = 1.0$):**

$$\Phi_B + \Phi_C = \frac{2}{3} + \frac{1}{3} = \mathbf{1.00 \quad (100\%)}$$



「反応した A（90 mol/min）の 66.7%（60 mol/min分）が B に、33.3%（30 mol/min分）が C に使われた」となり、選択率の和がぴったり 1.0 になります。
2. **収率の和 ($Y_B + Y_C = X_A$):**

$$Y_B + Y_C = 0.60 + 0.30 = \mathbf{0.90} = X_A$$



「供給した A（100 mol/min）の 60% が B に、30% が C に変換され、合計 90%（＝転化率 $X_A$）が反応して消費された」となり、収率の和が転化率 $X_A$ に一致することが確認できます。




転化率（いくら反応したか）だけを追うと失敗し、**「選択率（欲しいものがどれだけできたか）」を意識しないと利益や効率が激減する**という実践的な差を体感できるオリジナルの演習問題を作成しました。

前述の並列反応の数式モデル（$2\text{A} \rightarrow \text{B}$, $3\text{A} \rightarrow 2\text{C}$）を踏襲しつつ、定性・定量の両面から理解が深まる構成にしています。

---

### 【演習問題】：化学プラントの設計変更と利益の最大化

ある化学工場で、高価な有用薬品 **B** を合成するために以下の液相並列反応を行っています。副生成物 **C** は廃棄処理費用がかかる産業廃棄物です。

1. **主反応（2次反応）:** $2\text{A} \xrightarrow{k_1} \text{B} \quad \left(r_1 = k_1 C_A^2, \quad k_1 = 0.1 \text{ L/(mol}\cdot\text{min)}\right)$
2. **副反応（1次反応）:** $3\text{A} \xrightarrow{k_2} 2\text{C} \quad \left(r_2 = k_2 C_A, \quad k_2 = 0.1 \text{ min}^{-1}\right)$

**プロセス条件:**

* 原料供給量: $F_{A0} = 100 \text{ mol/min}$
* 供給濃度: $C_{A0} = 2.0 \text{ mol/L}$（したがって供給体積流量 $v_0 = 50 \text{ L/min}$）

現場の若手エンジニアが、「反応器のサイズ（体積 $V$）を大きくして原料 **A** の**転化率 $X_A$ を 50% から 90% に引き上げれば、目的物 B の生産量も増えて売上・利益が跳ね上がるはずだ**」と提案し、反応器を大型の完全混合槽型反応器（**CSTR**）に変更しました。

---

#### 【問い】

1. **【転化率と選択率の計算】**
CSTR において、転化率を **$X_A = 0.50$**（変更前）から **$X_A = 0.90$**（変更後）に引き上げたとき、主生成物 **B** への全選択率 $\Phi_B$（消費された A のうち B の生成に使われた割合）はどのように変化するか求めなさい。

2. **【エンジニアの勘違いの解説】**
転化率を 90% に高めた結果、目的物 B の「1 リットルあたりの生成量（収率）」および「プラントの経済性（効率）」はどうなったか説明し、エンジニアの提案が失敗に終わった定性的な理由を述べなさい。

---

### 解答と解説

#### Step 1: 転化率と選択率の計算（問1の解答）

CSTR における出口の原料濃度は $C_A = C_{A0}(1 - X_A)$ です。
また、CSTR の瞬間選択率（＝全選択率 $\Phi_B$）は以下の式で表されます。

$$\Phi_B = \frac{2 r_1}{-r_A} = \frac{2 k_1 C_A^2}{2 k_1 C_A^2 + 3 k_2 C_A} = \frac{2 C_A}{2 C_A + 3}$$

* **変更前（$X_A = 0.50$ のとき）：**
* 出口濃度: $C_A = 2.0 \times (1 - 0.50) = 1.0 \text{ mol/L}$
* 全選択率: $\Phi_B = \frac{2(1.0)}{2(1.0) + 3} = \frac{2}{5} = \mathbf{0.40 \quad (40\%)}$


* **変更後（$X_A = 0.90$ のとき）：**
* 出口濃度: $C_A = 2.0 \times (1 - 0.90) = 0.20 \text{ mol/L}$
* 全選択率: $\Phi_B = \frac{2(0.20)}{2(0.20) + 3} = \frac{0.40}{3.40} \approx \mathbf{0.118 \quad (11.8\%)}$



---

#### Step 2: 収率とプラント経済性の比較（問2の解答）

原料 A 1 mol あたりから生成する B のモル数（収率 $Y_B$）は、$Y_B = X_A \times \frac{\Phi_B}{2}$（※ A 2モルから B 1モルができるため $\frac{1}{2}$ を掛ける）となります。

| 指標 | 変更前 ($X_A = 50\%$) | 変更後 ($X_A = 90\%$) | 比較結果 |
| --- | --- | --- | --- |
| **転化率 $X_A$** | $50\%$ | **$90\%$** | 大幅アップ（一見成功に見える） |
| **選択率 $\Phi_B$** | **$40\%$** | $11.8\%$ | **激減（約 $1/3.4$ に低下）** |
| **B の収率 $Y_B$** | **$0.10 \text{ mol/mol-A}$** | $0.053 \text{ mol/mol-A}$ | **目的物の獲得量はほぼ半減！** |
| **副生物 C の割合** | 60% | **88.2%** | 大部分が無駄なゴミになる |

##### 【定性的な理由・ポイント】

1. **主反応（2次）と副反応（1次）の「反応次数」の違い:**
主反応は濃度 $C_A$ の2乗に比例して進みます。転化率を 90% にしようとして CSTR 内の濃度を $C_A = 0.20$ まで下げてしまうと、高次である主反応（2次）の速度が極端に低下し、相対的に低次（1次）の副反応ばかりが進むようになります。
2. **「転化率」と「選択率」のトレードオフ:**
単一反応では「転化率を上げる ＝ 目的物がより多く取れる」が成り立ちますが、複合反応では**転化率を追うあまり選択率を犠牲にすると、原料を消費しているにもかかわらず目的物の生成量が減り、大量の副生成物（廃棄物）を生み出してしまう**という逆転現象が起こります。

---

### この問題で学べる本質（まとめ）

* 単一反応 $\implies$ **「転化率（$X_A$）の最大化」** ＝ 装置の最適設計
* 複合反応 $\implies$ **「選択率（$\Phi_B$）と転化率の積（＝収率 $Y_B$）の最大化」** ＝ 装置の最適設計

複合反応においては、「とにかく反応させれば良い（高転化率）」ではなく、「高濃度を維持して欲しい反応だけを優先して進ませる（高選択率）」という**反応特性に応じた設計思考**が不可欠であることが、数値で実感できる問題となっています。



---

### 【問2】：指定の転化率・選択率から反応器体積 $V$ を求める


以下の並列反応が液相（等容系）で行われています。

1. **主反応（2次反応）:** $2\text{A} \xrightarrow{k_1} \text{B} \quad \left(r_1 = k_1 C_A^2, \quad k_1 = 0.1 \text{ L/(mol}\cdot\text{min)}\right)$
2. **副反応（1次反応）:** $3\text{A} \xrightarrow{k_2} 2\text{C} \quad \left(r_2 = k_2 C_A, \quad k_2 = 0.1 \text{ min}^{-1}\right)$

**プロセス条件:**

* 原料供給量: $F_{A0} = 100 \text{ mol/min}$
* 供給濃度: $C_{A0} = 2.0 \text{ mol/L}$（したがって供給体積流量 $v_0 = 50 \text{ L/min}$）

**【設計目標（仕様指定）】**

* **目標転化率:** $X_A = 0.80$ （80%）
* **目標全選択率（主生成物 B）:** $\Phi_B = 0.80$ （80%）


完全混合槽型反応器（**CSTR**）を用いてこの目標（$X_A = 0.80, \Phi_B = 0.80$）を達成できるか検討しなさい。
可能な場合は必要な**反応器体積 $V_{\text{CSTR}}$** を求めなさい。
達成できない場合は、その根拠を示しなさい。

---

### 解答と解説

#### Step 1: 出口での A の濃度 $C_A$ の決定

目標転化率が $X_A = 0.80$ なので、CSTR 出口における A の濃度 $C_A$ は次のようになります。

$$C_A = C_{A0}(1 - X_A) = 2.0 \times (1 - 0.80) = \mathbf{0.40 \text{ mol/L}}$$

CSTR は完全混合であるため、**反応器内部の A の濃度はどこでも均一に $C_A = 0.40 \text{ mol/L}$** となります。

---

#### Step 2: 反応速度式の評価

A の全消費速度 $-r_A$ は以下のように書けます。

$$-r_A = 2r_1 + 3r_2 = 2(k_1 C_A^2) + 3(k_2 C_A)$$

ここで、出口濃度 $C_A = 0.40 \text{ mol/L}$ を代入して各項の反応速度を計算します。

* **主反応による A の消費速度:**

$$2 r_1 = 2 \times 0.1 \times (0.40)^2 = 2 \times 0.016 = \mathbf{0.032 \text{ mol/(L}\cdot\text{min)}}$$


* **副反応による A の消費速度:**

$$3 r_2 = 3 \times 0.1 \times 0.40 = \mathbf{0.120 \text{ mol/(L}\cdot\text{min)}}$$


* **A の全消費速度 $-r_A$:**

$$-r_A = 0.032 + 0.120 = \mathbf{0.152 \text{ mol/(L}\cdot\text{min)}}$$



---

#### Step 3: 全選択率 $\Phi_B$ の検証

（※CSTR 内の濃度条件で、目標の全選択率 $\Phi_B = 0.80$ が達成できているかを確認します）

CSTRにおける瞬間全選択率はそのまま全体の全選択率 $\Phi_B$ になるため：

$$\Phi_B = \frac{2 r_1}{-r_A} = \frac{0.032}{0.152} \approx \mathbf{0.21 \quad (21\%)}$$

> **【重要な気づき・ポイント】**
> CSTR では A の濃度が低く維持されるため（$C_A = 0.40$）、高次元（2次）の主反応速度 $r_1$ が極端に低下し、選択率は **21%** しか達成できません。
> つまり、**「反応速度定数 $k_1, k_2$ と濃度条件が固定されている場合、CSTR では $X_A = 0.80$ と $\Phi_B = 0.80$ を同時指定することは物理的に不可能」**であることが分かります。

---

#### Step 4: PFR（管型反応器）の場合の検証と体積計算

主反応が2次、副反応が1次の場合、**原料 A の濃度を高く保てる PFR の方が選択率を大幅に引き上げることができます。**

PFR における局所（瞬間）全選択率 $\Phi'_B$ は以下のようになります。

$$\Phi'_B = \frac{2 k_1 C_A^2}{2 k_1 C_A^2 + 3 k_2 C_A} = \frac{2 k_1 C_A}{2 k_1 C_A + 3 k_2}$$

* **入口（$C_A = 2.0$）での瞬間選択率:**

$$\Phi'_B = \frac{2(0.1)(2.0)}{2(0.1)(2.0) + 3(0.1)} = \frac{0.4}{0.7} \approx \mathbf{57.1\%}$$


* **出口（$C_A = 0.40$）での瞬間選択率:**

$$\Phi'_B = \frac{2(0.1)(0.4)}{2(0.1)(0.4) + 3(0.1)} = \frac{0.08}{0.38} \approx \mathbf{21.1\%}$$



---

### まとめ：指定された $X_A$ と $\Phi_B$ から体積 $V$ を計算する流れ

1. **目標 $X_A$ から出口濃度 $C_A$ を決める。**
2. **指定の全選択率 $\Phi_B$ が成り立つための濃度・操作条件（または必要な反応器形式）を特定する。**
3. **設計方程式に代入して体積 $V$ を計算する。**
* **CSTR:** $V = \dfrac{F_{A0} X_A}{-r_A(C_A)}$
* **PFR:** $V = F_{A0} \displaystyle\int_{0}^{X_A} \frac{dX_A}{-r_A}$



「指定の転化率と選択率を両立させるために、どの形式の反応器（PFR vs CSTR）を選び、どれだけの体積 $V$ にすべきか」という問題設定は、まさに反応工学の醍醐味と言えます。
