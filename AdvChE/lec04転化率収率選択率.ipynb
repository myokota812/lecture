# 第4回 転化率・収率・選択率と反応器選定

大学院反応工学講義（自習用）

第4回では、反応工学において非常に重要な3つの指標である転化率・収率・選択率について学びます。

- 転化率は「反応物がどれだけ反応したか」、
- 収率は「目的生成物がどれだけ得られたか」、
- 選択率は「目的生成物をどれだけ優先的に作れたか」

を表します。<br>
特に実際の化学プロセスでは、副反応によって不要な生成物が生じることが多く、単に転化率を高くするだけでは十分ではありません。<br>
転化率99%でも、収率が例えば0.5%ではプロセスとしてなりたちません。<br>
目的物を効率よく生産するためには、収率や選択率の評価が不可欠です。<br>
また、並列反応や逐次反応を例に、それぞれの反応系で収率や選択率がどのように変化するかを確認します。<br>
さらに後半では、これらの考え方を反応器設計に応用し、反応器選定法について学びます。<br>

今日の講義を通して、『どれだけ反応したか』だけでなく、『何をどれだけ作れたか』という視点で反応を評価できるようになることを目標とします。

<h1>並列反応の場合</h1>

---

$$
\text{A} \xrightarrow{r_1} \text{B}
$$

$$
\text{A} \xrightarrow{r_2} \text{C}
$$
- $\text{A}$:反応物質
- $\text{B}$:目的生成物質
- $\text{C}$:副生成物

## 1. 転化率
流通系：
$$X=\frac{F_{A0}-F_A}{F_{A0}} \tag{1}$$

回分系：

$$X=\frac{N_{A0}-N_A}{N_{A0}} \tag{2}$$

◇◇◇


<details><summary>📝🖱️説明文はこちら</summary>

まず、反応工学で最も基本的な指標である転化率を定義します。<br>
簡単のため、上式の並列反応を対象とします。<br>
ここでは$\text{A}$：原料、$\text{B}$：目的生成物、$\text{C}$:副生成物、として説明していきます。

**転化率**とは、「反応物がどれだけ反応したか」を表す指標です。<br>
例えば、100 molのAを反応器に投入し、そのうち80 molが反応した場合、転化率は80％になります。<br>
流通式反応器では、入口のモル流量を $F_{A0}$、出口のモル流量を $F_A$ とすると、(1)式で定義されます。<br>
一方、回分反応器では流入や流出がないため、モル流量の代わりに物質量を使い(2)で表されます。<br>
つまり転化率とは、「最初にあったAのうち、どれだけが反応したか」を割合で示したものです。

</details>

## 2. 収率と選択率

<h3>2.1. 流通系の収率と選択率</h3>

<h5>表１：流通系（連続流れ反応器）における収率・選択率</h5>

| | 瞬間 (Instantaneous) | 総括・供給原料基準 (Overall, Feed) | 総括・反応原料もしくは副生成物基準 (Overall, Reacted) |
| :--- | :---: | :---: | :---: |
| **収率 (Yield)** | $Y'_B = \dfrac{r_B}{-r_A}$ | $Y_B = \dfrac{F_B - F_{B0}}{F_{A0}}$ | $\Phi_B = \dfrac{F_B - F_{B0}}{F_{A0} - F_A}$ |
| **選択率 (Selectivity)** | $S'_{B/C} = \dfrac{r_B}{r_C}$ | $-$| $$S_{B/C} = \dfrac{F_B - F_{B0}}{F_C - F_{C0}}$$ |

◇◇◇

ここで転化率、収率、選択率の定義を体系的に整理します。<br>
まず、評価方法は大きく 瞬間値 と 総括値 に分けられます。<br>
瞬間値は「今この瞬間の反応の様子」を表し、総括値は「反応の結果として最終的にどれだけ得られたか」を表します。<br>

**収率**について見ると、瞬間収率
$$
Y'_B=\frac{r_B}{-r_A}
$$
は、反応物Aが消費される速度に対して、Bが生成する速度の割合を表しています。<br>
つまり、消費されたAが今どれだけBになっているかを示す指標です。<br>
一方、総括収率
$$
Y_B=\frac{F_B-F_{B0}}{F_{A0}}
$$
は、**供給したA全体に対してどれだけBが得られたか**を示します。<br>
これは実際の生産量を評価するときに重要な指標です。<br>
分母を$F_{A0}-F_A$とした収率$\Phi_B$、つまり反応した原料基準の収率も定義できます。<br>

次に選択率を見てみます。<br>

選択率の定義は教科書によってまちまちですが、本講義では、生成物のうち目的物の割合を選択率とします。

瞬間選択率（参考）
$$
S'_{B/C}=\frac{r_B}{r_C}
$$
は、**ある瞬間にBとCのどちらが優先的に生成されているか**を表します。<br>
並列反応の反応器選定では、この瞬間選択率が重要な役割を果たします。<br>

一方、総括選択率は
$$
S_{B/C} = \frac{F_B-F_{B0}} {F_C-F_{C0}}
$$
で定義されます。<br>
値が大きいほど、目的生成物が優先的に生成されていることを意味します。<br>
このように、瞬間値は反応器や運転条件の選定に利用し、総括値は最終的なプロセス性能の評価に利用します。<br>
つまり、瞬間収率・瞬間選択率は設計のための指標、総括収率・総括選択率は結果を評価する指標と理解しておくと整理しやすいでしょう。

---








<h3>2.2. 回分系の収率と選択率</h3>

<h5>表２：回分系（バッチ反応器）における収率・選択率</h5>

| | 瞬間 (Instantaneous) | 総括・供給原料基準 (Overall, Feed) | 総括・反応原料もしくは副生成物基準 (Overall, Reacted) |
| :--- | :---: | :---: | :---: |
| **収率 (Yield)** | $Y'_B = \dfrac{r_B}{-r_A}$ | $Y_B = \dfrac{N_B - N_{B0}}{N_{A0}}$ | $\Phi_B = \dfrac{N_B - N_{B0}}{N_{A0} - N_A}$ |
| **選択率 (Selectivity)** | $S'_{B/C} = \dfrac{r_B}{r_C}$ | $-$ | $$S_{B/A} = \dfrac{N_B - N_{B0}}{N_{A0} - N_A}= \Phi_B\\ S_{B/C} = \dfrac{N_B - N_{B0}}{N_C - N_{C0}}$$ |

この表は、先ほどの流通系の場合を回分系へ置き換えたものです。<br>
流通系では物質量流量 $F$ を用いていましたが、回分系では流入・流出がないため、物質量 $N$ を用います。<br>
基本的な考え方は流通系と同じです。<br>
瞬間収率$Y'_B$および瞬間選択率$S'_{B/C}$は反応器に依存せず、共通です。<br>
一方、総括収率
$$
Y_B=\frac{N_B-N_{B0}}{N_{A0}}
$$
は、仕込んだA全体に対してどれだけBが得られたかを示します。<br>
さらに、
$$
\Phi_B = \frac{N_B-N_{B0}} {N_{A0}-N_A}
$$
は、反応したAのうちどれだけがBになったかを表す指標です。<br>
選択率についても同様に、
$$
S_{B/A} = \frac{N_B-N_{B0}} {N_{A0}-N_A}
$$
と
$$
S_{B/C} = \frac{N_B-N_{B0}} {N_C-N_{C0}}
$$
が定義されます。<br>
つまり回分系でも、瞬間値は反応の進み方を評価する指標、総括値は最終的な反応結果を評価する指標と理解しておけば十分です。

## 3. 転化率、収率の関係式

原料 A の転化率 $X_A$ はモル流量を用いて以下のように表されます。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

これを用いると、2つの式の関係は次のように綺麗に整理できます。

$$\underbrace{\frac{F_B - F_{B0}}{F_{A0}}}_{\text{供給原料基準の収率 } Y_B} = \underbrace{\left( \frac{F_{A0} - F_A}{F_{A0}} \right)}_{\text{転化率 } X_A} \times \underbrace{\left( \frac{F_B - F_{B0}}{F_{A0} - F_A} \right)}_{\text{（反応原料基準の収率）}}$$

$$\therefore Y_B = X_A \cdot \Phi_B$$

以上は量論係数が１の場合ですが、それ以外の場合について、次に説明していきます。

---




## 4. 量論係数が１でない場合

一般化された化学量論係数 $a, b, d, c$ を持つ並列反応系

$$a\text{A} \xrightarrow{r_1} b\text{B} \quad (\text{反応1})$$

$$d\text{A} \xrightarrow{r_2} c\text{C} \quad (\text{反応2})$$

を**連続流通系反応器**で実施する場合の、転化率・選択率・収率の一般的な定義を解説します。

流通系のモル流量を以下のように定義します。

* 供給時のAのモル流量：$F_{A0}$
* 反応器出口のモル流量：$F_A$（未反応A）、$F_B$（生成物B）、$F_C$（生成物C）

---

### 4.1. 転化率（Conversion: $X_A$）

反応物の種類や**化学量論係数に関わらず**、供給されたAのうち実際に反応した割合です。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

---



### 4.2. 収率（Yield: $Y_i$ および $\Phi_i$）

**① 供給原料基準の収率（$Y_i$）** と **② 反応原料基準の収率（$\Phi_i$）** の2つの表し方について説明します。

* **反応1**：$a\text{A} \rightarrow b\text{B}$ （1モルの $\text{B}$ を作るには $\frac{a}{b}$ モルの $\text{A}$ が必要）
* **反応2**：$d\text{A} \rightarrow c\text{C}$ （1モルの $\text{C}$ を作るには $\frac{d}{c}$ モルの $\text{A}$ が必要）

---

#### 4.2.1. 供給原料基準の収率 ($Y_B, Y_C$)

「供給したすべての原料A」に対して、最終的に目的の生成物等に変換された割合を表します（プロセス全体の生産性を評価する指標です）。分母には**供給された総モル流量 $F_{A0}$** を用います。

* **$\text{B}$ の収率 ($Y_B$)**：

$$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$


* **$\text{C}$ の収率 ($Y_C$)**：

$$Y_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0}} = \frac{d (F_C - F_{C0})}{c F_{A0}}$$



---

#### 4.2,2. 反応原料基準の収率（$\Phi_B, \Phi_C$）

「実際に反応した原料A」に対して、どれだけ目的の生成物に変換されたかを表す指標です。分母には実際に反応したモル流量（$F_{A0} - F_A$）を用います。
（※この反応原料基準の収率は、後述する「選択率（反応原料基準 $S_{B/A}$）」と完全に一致します）

* **$\text{B}$ の反応原料基準収率（選択率） ($\Phi_B$ / $S_{B/A}$)**：

$$\Phi_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$


* **$\text{C}$ の反応原料基準収率（選択率） ($\Phi_C$ / $S_{C/A}$)**：

$$\Phi_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0} - F_A} = \frac{d (F_C - F_{C0})}{c (F_{A0} - F_A)}$$



> **💡 補足（収率と転化率・選択率の関係）**
> 供給原料基準の収率（$Y_B$）は、転化率（$X_A$）と反応原料基準の収率（$\Phi_B$）の積として表すことができます。
> 
> $$Y_B = X_A \times \Phi_B \quad (\text{収率} = \text{転化率} \times \text{選択率})$$
> 
>

---


### 4.3. 選択率（Selectivity: $S_{B/C}$）

反応器から得られた出力（生成物）に着目し、目的生成物 $\text{B}$ と副生成物 $\text{C}$ がどのような比率で生成されたかを表す指標です。各反応の化学量論係数（$a, b, d, c$）を考慮し、**消費された原料 $\text{A}$ の当量換算**で比をとります。

* **反応1**：$a\text{A} \xrightarrow{r_1} b\text{B}$ （1モルの $\text{B}$ を作るには $\frac{a}{b}$ モルの $\text{A}$ が必要）
* **反応2**：$d\text{A} \xrightarrow{r_2} c\text{C}$ （1モルの $\text{C}$ を作るには $\frac{d}{c}$ モルの $\text{A}$ が必要）

これらに基づき、$\text{B}$ と $\text{C}$ の選択率の比（生成比 $S_{B/C}$）は次のように定義されます。

$$S_{B/C} = \frac{\frac{a}{b} F_B}{\frac{d}{c} F_C} = \frac{a c F_B}{b d F_C}$$

> **💡 ポイント**
> 原料 $\text{A}$ の消費量（$F_{A0} - F_A$）が分子・分母で相殺されるため、**「最終的な出口のモル流量 $F_B$ と $F_C$ の比」に化学量論係数からの補正項（$\frac{ac}{bd}$）を掛け合わせたもの**としてシンプルに表されます。
> （※ もしすべての係数 $a, b, d, c$ が $1$ の場合は、単純に $S_{B/C} = \frac{F_B}{F_C}$ となります）

### 4.4. まとめ

一般化された化学量論係数 $a, b, d, c$ を持つ並列反応系

$$a\text{A} \xrightarrow{r_1} b\text{B} \quad (\text{反応1})$$

$$d\text{A} \xrightarrow{r_2} c\text{C} \quad (\text{反応2})$$

における、転化率、収率、選択率（連続流通系）

- 転化率
$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

- 供給原料基準の収率 ($Y_B, Y_C$)
  -  **$\text{B}$ の収率 ($Y_B$)**：
    $$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$
  - **$\text{C}$ の収率 ($Y_C$)**：
    $$Y_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0}} = \frac{d (F_C - F_{C0})}{c F_{A0}}$$

- 反応原料基準の収率（$\Phi_B, \Phi_C$）
  -  **$\text{B}$ の反応原料基準収率（選択率） ($\Phi_B$ / $S_{B/A}$)**：
     $$\Phi_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$
  - **$\text{C}$ の反応原料基準収率（選択率） ($\Phi_C$ / $S_{C/A}$)**：$$\Phi_C = \frac{\frac{d}{c} (F_C - F_{C0})}{F_{A0} - F_A} = \frac{d (F_C - F_{C0})}{c (F_{A0} - F_A)}$$

- 選択率（Selectivity: $S_{B/C}$）

 $$S_{B/C} = \frac{\frac{a}{b} F_B}{\frac{d}{c} F_C} = \frac{a c F_B}{b d F_C}$$

<h1>逐次反応の場合</h1>

---

化学量論係数を含む**逐次反応）**

$$a\text{A} \xrightarrow{r_1} b\text{B} \xrightarrow{r_2} c\text{C}$$

（$\text{A}$: 原料、$\text{B}$: 中間生成物、$\text{C}$: 最終生成物）を**連続流通系反応器**で実施する場合の、転化率・収率・選択率の定義を解説します。

---

### 1. 転化率（Conversion: $X_A$）

反応の起点となる原料Aの消費割合です。後続の反応（B $\rightarrow$ C）に関わらず、**供給されたAのうち実際に反応した割合**として、並列反応や単一反応と同じ定義になります。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}}$$

* $F_{A0}$：供給時のAのモル流量
* $F_A$：出口の未反応Aのモル流量

---

### 2. 収率（Yield: $Y_B, Y_C$）

#### 2.1. 供給原料基準の収率 ($Y_B, Y_C$)

供給されたすべての原料Aの総モル流量 $F_{A0}$ に対し、出口で得られた各生成物の正味のモル流量（$F_B - F_{B0}, F_C - F_{C0}$）の割合です。各段階の化学量論係数を考慮して補正します。

#### ① 中間生成物 B の収率 ($Y_B$)

反応1（$a\text{A} \rightarrow b\text{B}$）より、1モルのBを作るには $\frac{a}{b}$ モルのAが必要です。


$$Y_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0}} = \frac{a (F_B - F_{B0})}{b F_{A0}}$$

#### ② 最終生成物 C の収率 ($Y_C$)

反応1（$a\text{A} \rightarrow b\text{B}$）と反応2（$b\text{B} \rightarrow c\text{C}$）の全体の関係から、**1モルのCを作るために必要な原料Aのモル数**は $\frac{a}{c}$ です。<br>
したがって、Cの供給原料基準の収率 $Y_C$ は次のように表されます。

$$Y_C = \frac{\frac{a}{c} (F_C - F_{C0})}{F_{A0}} = \frac{a (F_C - F_{C0})}{c F_{A0}}$$


---

#### 2.2. 反応原料基準の収率（$\Phi_B, \Phi_C$）

反応原料基準の収率 $\Phi$ は、**「実際に反応して消費された原料A」のうち、どれだけの割合がそれぞれの物質（BまたはC）に変換されているか**を表す指標です。
分母には必ず「実際に反応したAのモル流量（$F_{A0} - F_A$）」を用います。

#### ① 中間生成物 B の反応原料基準収率（$\Phi_B$）

反応1（$a\text{A} \rightarrow b\text{B}$）の化学量論係数（1モルのBを作るのに必要なAは $\frac{a}{b}$ モル）を考慮します。


$$\Phi_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$

* **意味**: 実際に反応したAのうち、**現在まだ「Bの状態で残っている」ものの割合**です。

#### ② 最終生成物 C の反応原料基準収率（$\Phi_C$）



反応1（$a\text{A} \rightarrow b\text{B}$）と反応2（$b\text{B} \rightarrow c\text{C}$）の化学量論関係から、「1モルのCを作るのに必要な原料Aのモル数」は $\frac{a}{c}$ となります。
（※Aが $a$ モル反応するとBが $b$ モルでき、そのBがすべて反応するとCが $c$ モルできるため、$a$ モルのAから最終的に $c$ モルのCが生まれる関係になります）

したがって、Cの反応原料基準の収率 $\Phi_C$ は次のように表されます。

$$\Phi_C = \frac{\frac{a}{c} (F_C - F_{C0})}{F_{A0} - F_A} = \frac{a (F_C - F_{C0})}{c (F_{A0} - F_A)}$$


---

### 2. 並列反応との決定的な違い（和についての注意点）

先ほど解説した**並列反応**（$\text{A} \rightarrow \text{B}$ と $\text{A} \rightarrow \text{C}$）では、反応したAのすべてがBまたはCのどちらかに直接変わるため、反応原料基準の収率（＝選択率）の和は必ず **$\Phi_B + \Phi_C = 1$（100%）** になりました。

しかし、今回の**逐次反応**（$\text{A} \rightarrow \text{B} \rightarrow \text{C}$）では事情が異なります。

* **BがさらにCへ変化してしまう**ため、反応が進むにつれてBは消費されていきます。
* そのため、任意の時点における反応原料基準の収率を足し合わせても、**必ずしも $1$（100%）になるとは限りません**（多くの場合、中間体Bが存在している間は $\Phi_B + \Phi_C < 1$ になるか、あるいはCの定義をどう置くかによって総和の挙動が変わります）。

このように、逐次反応では「反応したAが今どこに存在しているか（A、B、Cのどの状態か）」のバランスが刻々と変化するため、$\Phi_B$ がピークを迎えてから減少していくという特徴的な挙動を示します。

---

### 3. 選択率（Selectivity: $S_{B/C}$ など）

逐次反応における選択率は、「反応した原料Aのうち、どれだけが目的の段階（B）にとどまっているか」を評価するために用いられます。

#### 反応原料基準の B の選択率（$S_B$）

実際に反応したAの量（$F_{A0} - F_A$）のうち、どれだけがBの生成・存在に使われているかを表します。


$$S_B = \frac{\frac{a}{b} (F_B - F_{B0})}{F_{A0} - F_A} = \frac{a (F_B - F_{B0})}{b (F_{A0} - F_A)}$$

* **収率との関係**：
並列反応と同様に、以下の関係が成り立ちます。

$$Y_B = X_A \times S_B \quad (\text{収率} = \text{転化率} \times \text{選択率})$$



---

### 💡 逐次反応における重要な特徴

逐次反応の最大の特徴は、中間生成物である **B の収率（$Y_B$）や選択率（$S_B$）が、反応時間（または転化率 $X_A$）に対して「山なりの曲線（極大値）」を描く**点です。

1. **反応初期**：A $\rightarrow$ B（$r_1$）が優勢なため、$Y_B$ は上昇する。
2. **反応後期**：B $\rightarrow$ C（$r_2$）が進むため、Bが消費されて $Y_B$ は低下し、最終的に $C$ が増える。

そのため、逐次反応のプロセス設計では、「Bが最も多く得られる最適滞留時間（または最適転化率）」を見極めることが非常に重要になります。

````python

import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# 1. パラメータ設定
# ==================================================
k1 = 1.0
k2 = 0.2

FA0 = 10.0
FB0 = 0.0
FC0 = 0.0

# 選択率の化学量論係数
a = 1.0
b = 1.0

# 空間時間
tau_vals = np.linspace(0, 5.0, 200)

# ==================================================
# 2. PFR出口のモル流量（解析解）
# ==================================================
FA = FA0 * np.exp(-k1 * tau_vals)

if not np.isclose(k1, k2):
    FB = (
        FA0 * k1 / (k2 - k1)
        * (np.exp(-k1 * tau_vals)
           - np.exp(-k2 * tau_vals))
    )
else:
    FB = FA0 * k1 * tau_vals * np.exp(-k1 * tau_vals)

# A -> B -> C が各1:1で進行する場合
FC = FA0 - FA - FB

# ==================================================
# 3. 転化率・収率・選択率
# ==================================================

# Aの転化率
XA = (FA0 - FA) / FA0

# 消費されたAのモル流量
reacted_A = FA0 - FA

# Bの正味生成モル流量
delta_FB = FB - FB0

# Phi_B：反応したAを基準とするBの生成割合
# Phi_B：反応したAを基準とするBの生成割合
Phi_B = np.divide(
    FB - FB0,
    FA0 - FA,
    out=np.full_like(FB, np.nan),
    where=(FA0 - FA) > 0
)

# S_B：指定された定義による選択率
S_B = np.divide(
    a * (FB - FB0),
    b * (FC - FC0),
    out=np.full_like(FB, np.nan),
    where=(FA0 - FA) > 0
)

# Y_B：供給したAを基準とするBの収率
Y_B = delta_FB / FA0

# S_B：指定された定義によるBの選択率
S_B = np.divide(
    a * delta_FB,
    b * reacted_A,
    out=np.zeros_like(FB),
    where=reacted_A > 0
)

# ==================================================
# 4. グラフ描画
# ==================================================
fig, (ax1, ax2) = plt.subplots(
    1, 2, figsize=(14, 5.5)
)

# --------------------------------------------------
# 左：各成分のモル流量
# --------------------------------------------------
ax1.plot(tau_vals, FA, label=r"$F_A$", linewidth=2)
ax1.plot(tau_vals, FB, label=r"$F_B$", linewidth=2)
ax1.plot(tau_vals, FC, label=r"$F_C$", linewidth=2)

ax1.set_title("Molar Flow Rates in PFR")
ax1.set_xlabel(r"Space time $\tau$")
ax1.set_ylabel("Molar flow rate [mol/time]")
ax1.grid(True, alpha=0.3)
ax1.legend()

# --------------------------------------------------
# 右：転化率・収率・選択率
# 横軸を空間時間 tau に変更
# --------------------------------------------------
ax2.plot(
    tau_vals, XA,
    label=r"$X_A$ (Conversion)",
    linewidth=2
)
ax2.plot(
    tau_vals, Phi_B,
    label=r"$\Phi_B$",
    linewidth=2
)
ax2.plot(
    tau_vals, Y_B,
    label=r"$Y_B$ (Yield)",
    linewidth=2
)
ax2.plot(
    tau_vals, S_B,
    label=r"$S_B$ (Selectivity)",
    linestyle="--",
    linewidth=2
)

ax2.set_title("Conversion, Yield and Selectivity vs Space Time")
ax2.set_xlabel(r"Space time $\tau$")
ax2.set_ylabel("Fraction")
ax2.set_xlim(0, 5.0)
ax2.set_ylim(-0.05, 1.05)
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

# ==================================================
# 5. Bのモル流量が最大になる空間時間
# ==================================================
if not np.isclose(k1, k2):
    tau_max = np.log(k1 / k2) / (k1 - k2)
else:
    tau_max = 1.0 / k1

FB_max = np.interp(tau_max, tau_vals, FB)

print(f"Bの流量が最大となる空間時間 tau = {tau_max:.4f}")
print(f"そのときのBのモル流量 FB = {FB_max:.4f} mol/time")
````

````python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. 流通系のパラメータ設定 ---
k1 = 1.0  # 反応速度定数 1 (A -> B)
k2 = 0.2  # 反応速度定数 2 (B -> C)

FA0 = 10.0  # 原料Aの供給モル流量 [mol/time]
FB0 = 0.0  # 入口のBのモル流量（初期値は0）
FC0 = 0.0  # 入口のCのモル流量（初期値は0）

# 空間時間（滞留時間） tau = V / v_0 の配列
tau_vals = np.linspace(0, 5.0, 200)

# --- 2. PFR出口のモル流量計算（解析解ベース） ---
# ※ 密度変化がない（体積流量が一定）と仮定し、モル流量ベースで計算
FA = FA0 * np.exp(-k1 * tau_vals)
FB = (
    (FA0 * k1 / (k2 - k1)) * (np.exp(-k1 * tau_vals) - np.exp(-k2 * tau_vals))
    if k1 != k2
    else FA0 * k1 * tau_vals * np.exp(-k1 * tau_vals)
)
FC = FA0 - FA - FB  # 全体のモル保存則（FB0, FC0 = 0 の場合）

# --- 3. 転化率・収率・選択率の計算 ---
# 転化率 X_A
Xa = (FA0 - FA) / FA0

# 実際に消費されたAのモル流量
reacted_A = FA0 - FA

# 反応原料基準の収率（＝反応原料基準の選択率 Phi_B, Phi_C）
Phi_B = np.divide(
    FB - FB0, reacted_A, out=np.ones_like(FB), where=reacted_A > 0
)
Phi_C = np.divide(
    FC - FC0, reacted_A, out=np.zeros_like(FC), where=reacted_A > 0
)

# 供給原料基準の収率 (Y_B, Y_C)
YB = (FB - FB0) / FA0
YC = (FC - FC0) / FA0

# --- 4. グラフの描画 ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# 【左側グラフ】 滞留時間 tau とモル流量 F の変化
ax1.plot(tau_vals, FA, label="FA (Reactant A)", color="blue", linewidth=2)
ax1.plot(
    tau_vals,
    FB,
    label="FB (Intermediate B / Yield Peak)",
    color="orange",
    linewidth=2,
)
ax1.plot(tau_vals, FC, label="FC (Product C)", color="green", linewidth=2)
ax1.set_title(
    "Molar Flow Rates in PFR (Consecutive Reaction)",
    fontsize=12,
    fontweight="bold",
)
ax1.set_xlabel("Space Time / Residence Time (tau)", fontsize=11)
ax1.set_ylabel("Molar Flow Rate [mol/time]", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend()

# 【右側グラフ】 転化率 (Xa) と 収率・選択率の関係
ax2.plot(
    Xa, YB, label="YB: Supply-base Yield of B", color="orange", linewidth=2
)
ax2.plot(
    Xa,
    Phi_B,
    label="Phi_B (SB): Reacted-base Selectivity of B",
    linestyle="--",
    color="orange",
)
ax2.plot(
    Xa,
    Phi_C,
    label="Phi_C (SC): Reacted-base Selectivity of C",
    linestyle="--",
    color="green",
)
ax2.set_title(
    "Yield & Selectivity vs Conversion (Xa) in PFR",
    fontsize=12,
    fontweight="bold",
)
ax2.set_xlabel("Conversion (Xa)", fontsize=11)
ax2.set_ylabel("Fraction", fontsize=11)
ax2.set_ylim(-0.05, 1.05)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()
````

このコードは、逐次反応 $$A \rightarrow B \rightarrow C$$ が起こる押し出し流れ反応器（PFR）について、滞留時間による各成分のモル流量の変化と、転化率・収率・選択率の関係を計算して描画するプログラムです。

反応速度式は、濃度を用いると次のようになります。

$$ A \xrightarrow{k_1} B \xrightarrow{k_2} C $$

$$ r_1=k_1 C_A,\qquad r_2=k_2 C_B $$

ここで、$$C_A,C_B$$ は各成分の濃度、$$r_1,r_2$$ は各反応の速度です。

このコードでは、体積流量が一定であると仮定しているため、空間時間 $$\tau$$ を時間に相当する変数として使い、各成分のモル流量を計算します。

化学反応工学の観点では、特に次のことを理解するためのコードです。

- 原料 $A$ は反応によって減少する。
- 中間生成物 $B$ は、最初は増加するが、さらに反応して $C$ になるため、途中で最大値をとる。
- $C$ は時間とともに増加する。
- $B$ を多く得たい場合、反応をどこまで進めるべきかを考えられる。

| 部分            | 役割                     |

| 1. パラメータ設定    | 反応速度定数、供給流量、空間時間を決める   |
| 2. 解析解の計算     | $$F_A,F_B,F_C$$ を求める |
| 3. 転化率・収率・選択率 | 反応成績を定量化する             |
| 4. グラフ描画      | 流量と反応成績の変化を可視化する       |

それぞれを順番に解説します。

## 3. 第1部分：パラメータの設定

```
import matplotlib.pyplot as pltimport numpy as np
```

- `numpy`：指数関数、数値配列などの計算に使います。
- `matplotlib.pyplot`：グラフを描画します。

続いて、反応条件を設定します。

```
k1 = 1.0k2 = 0.2FA0 = 10.0FB0 = 0.0FC0 = 0.0
```

各変数の意味は次のとおりです。

| 変数    | 意味                         | 値           |
| ----- | -------------------------- | ----------- |
| `k1`  | $$A\rightarrow B$$ の速度定数 | 1.0         |
| `k2`  | $$B\rightarrow C$$ の速度定数 | 0.2         |
| `FA0` | 入口のAのモル流量                  | 10 mol/time |
| `FB0` | 入口のBのモル流量                  | 0           |
| `FC0` | 入口のCのモル流量                  | 0           |

ここでは、入口にはAだけが存在し、BとCは存在しないとしています。

### 空間時間の配列

```
tau_vals = np.linspace(0, 5.0, 200)
```

`np.linspace(開始値, 終了値, 個数)` は、指定範囲を等間隔に分割します。

したがって、ここでは$ 0\leq \tau\leq 5 $
の範囲を200点に分けて計算します。

空間時間は

$$ \boxed{\tau=\frac{V}{v_0}} $$

で定義されます。

- $V$：反応器の体積
- $v_0$：入口の体積流量



## 4. 第2部分：各成分のモル流量を計算する

ここがプログラムの中心部分です。

### 4.1 原料Aのモル流量

```
FA = FA0 * np.exp(-k1 * tau_vals)
```

これは一次反応の解析解です。

PFR（押し出し流れ反応器）の物質収支式は

$$ \frac{dF_A}{dV}=-r_1 $$

です。体積流量を $v_0$ とし、$C_A=F_A/v_0$、$r_1=k_1C_A$ とすると、

$$ \frac{dF_A}{d\tau}=-k_1F_A $$

となります。これを積分すると、

$$ \boxed{F_A=F_{A0}e^{-k_1\tau}} $$

が得られます。<br>
コードの `np.exp(-k1 * tau_vals)` が、指数関数 $e^{-k_1\tau}$ を計算しています。<br>
滞留時間が長くなるほど、Aのモル流量は減少します。

### 4.2 中間生成物Bのモル流量

```
FB = (    (FA0 * k1 / (k2 - k1))    * (np.exp(-k1 * tau_vals) - np.exp(-k2 * tau_vals))    if k1 != k2    else FA0 * k1 * tau_vals * np.exp(-k1 * tau_vals))
```

少し長い式ですが、意味は明確です。<br>
Bは、Aから生成される一方で、Cへと変化して消費されます。<br>
Bの物質収支式は

$$ \frac{dF_B}{d\tau}=k_1F_A-k_2F_B $$

です。<br>
右辺の2項を見てください。

- $k_1F_A$：AからBが生成する速度
- $k_2F_B$：BからCへ反応する速度

Aの解析解を代入してこの微分方程式を解くと、$k_1\neq k_2$ の場合、

$$ \boxed{ F_B=F_{A0}\frac{k_1}{k_2-k_1} \left(e^{-k_1\tau}-e^{-k_2\tau}\right) } $$

となります。<br>
これがコードの通常の計算式です。<br>

#### `if k1 != k2 else ...` の意味

速度定数が等しい場合には、上の式の分母 $k_2-k_1$ がゼロになります。そのため、別の解析解を使用します。

$$ \boxed{ F_B=F_{A0}k_1\tau e^{-k_1\tau} } \qquad (k_1=k_2) $$

Pythonの条件式は、この2つの式を自動的に切り替えています。

### 4.3 最終生成物Cのモル流量

```
FC = FA0 - FA - FB
```

ここでは、反応

$$ A\rightarrow B\rightarrow C $$

の各段階で、A、B、Cが1対1で変換されると仮定しています。<br>
したがって、入口と出口で全成分のモル流量の合計が等しくなります。

$$ F_A+F_B+F_C=F_{A0} $$

これを $F_C$ について整理すると、

$$ \boxed{F_C=F_{A0}-F_A-F_B} $$

となります。<br>
なお、これは今回の反応の化学量論と入口条件に基づく関係です。一般の反応系で、総モル流量が必ず一定になるわけではありません。

## 5. 第3部分：転化率・収率・選択率を計算する

ここでは、反応器の性能を評価する3つの指標を求めています。

### 5.1 転化率 $$X_A$$

```
Xa = (FA0 - FA) / FA0
```

### 5.2 反応したAの流量

```
reacted_A = FA0 - FA
```

これは転化率の分子と同じ量です。

$$ F_{A,\mathrm{reacted}}=F_{A0}-F_A $$

つまり、反応器内で消費されたAのモル流量です。<br>
この量を使うと、実際に反応したAのうち、どれだけがBやCとして得られたかを評価できます。

### 5.3 反応したAを基準とする収率・選択率

```
Phi_B = np.divide(    FB - FB0, reacted_A,    out=np.zeros_like(FB),    where=reacted_A > 0)
```

```
Phi_C = np.divide(    FC - FC0, reacted_A,    out=np.zeros_like(FC),    where=reacted_A > 0)
```

このコードでは、次の量を計算しています。

$$ \boxed{\Phi_B=\frac{F_B-F_{B0}}{F_{A0}-F_A}} $$

$$ \boxed{\Phi_C=\frac{F_C-F_{C0}}{F_{A0}-F_A}} $$

入口にBとCがない場合は、

$$ \Phi_B=\frac{F_B}{F_{A0}-F_A}, \qquad \Phi_C=\frac{F_C}{F_{A0}-F_A} $$

です。

これらは、反応によって消費されたAを基準にした生成割合です。今回の逐次反応では、AがBになった割合と、さらに反応してCになった割合を表します。

`np.divide` は、配列同士の割り算を行う関数です。

- `where=reacted_A > 0`：消費されたAが正の場合に計算する
- `out=np.zeros_like(FB)`：計算しない場所の初期値を0にする

入口では $$\tau=0$$ なので、消費されたAもゼロになります。その点でゼロ除算が発生するのを防いでいます。<br>
今回の反応では、A、B、Cのモル数が1対1で対応するため、

$$ \boxed{\Phi_B+\Phi_C=1} $$

という関係が成立します。ただし、これはこの反応系と入口条件に依存する結果です。

### 5.4 供給原料を基準とする収率

```
YB = (FB - FB0) / FA0YC = (FC - FC0) / FA0
```

今度は、分母が「反応したA」ではなく「供給したA」です。

$$ \boxed{Y_B=\frac{F_B-F_{B0}}{F_{A0}}} $$

$$ \boxed{Y_C=\frac{F_C-F_{C0}}{F_{A0}}} $$

例えば、Aを10 mol/time供給し、出口でBが3 mol/time得られたなら、

$$ Y_B=\frac{3}{10}=0.3 $$

したがって、供給原料基準のB収率は30%です。<br>
転化率との関係は、

$$ \boxed{Y_B=X_A\Phi_B} $$

です。実際に反応したAの割合が $$X_A$$、そのうちBとして得られた割合が $\Phi_B$ なので、供給したA全体に対するBの生成割合は、その積になります。

同様に、

$$ Y_C=X_A\Phi_C $$

です。

## 6. 第4部分：グラフを描画する

```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
```

この命令で、横に2つのグラフを並べて描画する領域を作成します。

- `ax1`：左側のグラフ
- `ax2`：右側のグラフ
- `figsize=(14, 6)`：図全体の大きさ

### 6.1 左側：滞留時間と各成分のモル流量


## 7. このコードから何がわかるのか？

今回の設定では、

$$ k_1=1.0,\qquad k_2=0.2 $$

なので、AからBへの反応は、BからCへの反応より速い設定です。<br>
このとき、Bのモル流量が最大になる滞留時間は、解析的に求められます。<br>
Bの流量が最大となる条件は、

$$ \frac{dF_B}{d\tau}=0 $$

です。物質収支式から、

$$ k_1F_A-k_2F_B=0 $$

すなわち、

$$ k_1F_A=k_2F_B $$

となります。これを解析解に適用すると、

$$ \boxed{ \tau_{\max} =\frac{\ln(k_1/k_2)}{k_1-k_2} } $$

が得られます。今回の数値を代入すると、

$$ \tau_{\max} =\frac{\ln(1.0/0.2)}{1.0-0.2} \approx 2.01 $$

となります。<br>
Bのモル流量が最大になる空間時間

# 約 2.01

速度定数と整合する時間単位

つまり、Bを目的生成物として回収する場合、今回の理想モデルでは、空間時間を約2.01に設定すると出口のB流量が最大になります。

## 8. さらに重要な注意点：収率と選択率の定義

化学反応工学では、「収率」と「選択率」は文献や授業によって定義の表し方が異なることがあります。このコードでは、明確に次のように定義しています。

$$ Y_B=\frac{F_B-F_{B0}}{F_{A0}} $$

$$ \Phi_B=\frac{F_B-F_{B0}}{F_{A0}-F_A} $$

ここで `Phi_B` は、消費されたAに対するBの生成量の比です。今回の逐次反応では、反応したAのうちBとして残っている割合を表すため、選択率に相当する指標として使っています。

ただし、一般的な選択率は、競争する複数の反応によって目的生成物と副生成物ができる場合など、異なる定義で使われることもあります。講義資料にする場合は、必ず定義式を明記するのが大切です。

また、今回の `FC = FA0 - FA - FB` は、A・B・Cが1対1で変換され、入口にBとCがないことを前提としています。

このプログラムの最も重要な学習ポイントは、転化率を最大にする操作条件と、目的生成物Bの収率を最大にする操作条件は一致しない場合がある、ということです。

````python
import matplotlib.pyplot as plt
import numpy as np

# --- 1. パラメータの設定 ---
k1 = 1.0  # 反応速度定数 1 (A -> B)
k2 = 0.2  # 反応速度定数 2 (B -> C)
CA0 = 1.0  # 原料Aの初期濃度

# 連続的な滞留時間（tau）の配列を作成
tau_vals = np.linspace(0, 5.0, 200)

# --- 2. 各濃度の計算（解析解） ---
CA = CA0 * np.exp(-k1 * tau_vals)
CB = (
    (CA0 * k1 / (k2 - k1)) * (np.exp(-k1 * tau_vals) - np.exp(-k2 * tau_vals))
    if k1 != k2
    else CA0 * k1 * tau_vals * np.exp(-k1 * tau_vals)
)
CC = CA0 - CA - CB

# --- 3. 各指標の計算 ---
Xa = (CA0 - CA) / CA0
reacted_A = CA0 - CA
Phi_B = np.divide(
    CB, reacted_A, out=np.zeros_like(CB), where=reacted_A > 0
)  # Bの反応原料基準収率（選択率）
Phi_C = np.divide(
    CC, reacted_A, out=np.zeros_like(CC), where=reacted_A > 0
)  # Cの反応原料基準収率
YB = CB / CA0  # Bの供給基準収率（実質収率）

# --- 4. グラフの描画 ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

# 【左側グラフ】 滞留時間と濃度の変化
ax1.plot(tau_vals, CA, label="A (Reactant)", color="blue", linewidth=2)
ax1.plot(
    tau_vals,
    CB,
    label="B (Intermediate / Yield Peak)",
    color="orange",
    linewidth=2,
)
ax1.plot(tau_vals, CC, label="C (Product)", color="green", linewidth=2)
ax1.set_title(
    "Concentration Profiles in Consecutive Reaction", fontsize=12, fontweight="bold"
)
ax1.set_xlabel("Residence Time (tau)", fontsize=11)
ax1.set_ylabel("Concentration", fontsize=11)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend()

# 【右側グラフ】 転化率と収率・選択率の関係（ピークの確認）
ax2.plot(
    Xa, YB, label="YB: B Yield (Supply Base)", color="orange", linewidth=2
)
ax2.plot(
    Xa, Phi_B, label="Phi_B: B Selectivity (Reacted Base)", linestyle="--", color="orange"
)
ax2.plot(
    Xa, Phi_C, label="Phi_C: C Selectivity (Reacted Base)", linestyle="--", color="green"
)
ax2.set_title(
    "Yield & Selectivity vs Conversion (Xa)", fontsize=12, fontweight="bold"
)
ax2.set_xlabel("Conversion (Xa)", fontsize=11)
ax2.set_ylabel("Fraction", fontsize=11)
ax2.set_ylim(-0.05, 1.05)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()
````

「**CSTRでは瞬間収率$Y'_B $ と　全選択率 $\Phi_B$ が等しくなる**」


* **瞬間収率 $Y'_B$（ある局所・ある濃度における選択率）：**
ある「一瞬間」または「特定の位置」において、反応して消えた A のうち、どれだけが B の生成に使われているかを示す**反応速度の比**です。

$$Y'_B = \frac{\text{B の生成に使われる A の消費速度}}{\text{A の全消費速度}} = \frac{2 r_1}{-r_A}$$


* **全選択率（出口での全体の選択率） $\Phi_B$：**
反応器全体（入口から出口まで）を通じて、実際に反応して消えた A の総モル数のうち、B に変換された A の総モル数の割合です。

$$\Phi_B = \frac{\text{出口で B の生成に使われた A の総流量}}{\text{出口までに反応して消えた A の総流量}} = \frac{2(F_B - F_{B0})}{F_{A0} - F_A}$$

---
### 数式による証明

CSTR の物質収支式（設計方程式）を用いて、これを数式で確かめます。

* **B の生成に関する物質収支:**

$$F_B - F_{B0} = r_B \cdot V = r_1 \cdot V$$

* **A の消費に関する物質収支:**

$$F_{A0} - F_A = (-r_A) \cdot V$$

---

選択率 $\Phi_B$ は「消費された A」のうち「B の生成に使われた A」の割合であるため、反応量論（$\text{2A} \rightarrow \text{B}$）より、B の生成量の 2 倍の A が消費されていることに注意して 2 つの式の比をとります。

$$
\begin{aligned}
\Phi_B &= \frac{2(F_B - F_{B0})}{F_{A0} - F_A}\\ 
       &= \frac{2 (r_1 \cdot V)}{(-r_A) \cdot V}\\ 
       &= \frac{2 r_1}{-r_A}\\
       &= Y'_B
\end{aligned}
$$

これにより、反応器体積 $V$ や流量によらず、CSTR における全選択率 $\Phi_B$ は「反応速度の比 $\frac{2 r_1}{-r_A}$」そのものになることが証明されます。

### 【問1】

反応器内で以下の並列反応が進行しています。

1. 主反応： $2\text{A} \longrightarrow \text{B}$
2. 副反応： $3\text{A} \longrightarrow 2\text{C}$

原料 A をモル流量 $F_{A0} = 100 \text{ mol/min}$ で連続流れ反応器（PFRまたはCSTR）に供給したところ、反応器出口における各成分のモル流量は以下の通りになりました。

* $F_A = 10 \text{ mol/min}$（残存する A）
* $F_B = 30 \text{ mol/min}$（生成した B）
* $F_C = 20 \text{ mol/min}$（生成した C）

なお、供給液中に B および C は含まれていません（$F_{B0} = 0, \ F_{C0} = 0$）。


1. **原料 A の転化率 $X_A$** を求めなさい。
2. **目的生成物 B の全選択率 $\Phi_B$** を求めなさい。
3. **関係式 $Y_B = X_A \cdot \Phi_B$** を用いて、**目的生成物 B の収率 $Y_B$** を算出しなさい。
4. **副生成物 C の収率 $Y_C$** を算出し、$Y_B + Y_C = X_A$（または $\Phi_B + \Phi_C = 1.0$）が成立することを確認しなさい。

---


### 【解答例】

#### 1. 原料 A の転化率 $X_A$ の計算

転化率は「仕込んだ A のうち反応して減った割合」であり、定義式より以下のように求められます。

$$X_A = \frac{F_{A0} - F_A}{F_{A0}} = \frac{100 - 10}{100} = \frac{90}{100} = \mathbf{0.90 \quad (90\%)}$$

---

#### 2. 目的生成物 B の全選択率 $\Phi_B$ の計算

全選択率は「反応して消費された A のうち、B の生成に使われた A の割合」です。
$2\text{A} \longrightarrow \text{B}$ より、B が 1 mol 生成するのに A は 2 mol 消費されるため、分子に量論補正の **2** を掛けます。

$$\Phi_B = \frac{2(F_B - F_{B0})}{F_{A0} - F_A} = \frac{2 \times (30 - 0)}{100 - 10} = \frac{60}{90} = \frac{2}{3} \approx \mathbf{0.667 \quad (66.7\%)}$$

---

#### 3. 関係式 $Y_B = X_A \cdot \Phi_B$ による B の収率 $Y_B$ の算定

関係式 $Y_B = X_A \cdot \Phi_B$ に、問1・問2で求めた値を代入します。

$$Y_B = X_A \cdot \Phi_B = 0.90 \times \frac{2}{3} = \mathbf{0.60 \quad (60\%)}$$

> **【直接定義式による検証】**
> 
> $$Y_B = \frac{2(F_B - F_{B0})}{F_{A0}} = \frac{2 \times 30}{100} = \frac{60}{100} = 0.60 \quad (60\%)$$
> 
> 
> 
> 定義式から直接求めた値と完全一致します。

---

#### 4. 副生成物 C の収率 $Y_C$ の算定と整合性の確認

$3\text{A} \longrightarrow 2\text{C}$（$\frac{3}{2}\text{A} \longrightarrow \text{C}$）より、C が 1 mol 生成するのに A は **$\frac{3}{2}$（1.5）mol** 消費されます。

* **C の全選択率 $\Phi_C$:**

$$\Phi_C = \frac{\frac{3}{2}(F_C - F_{C0})}{F_{A0} - F_A} = \frac{1.5 \times (20 - 0)}{100 - 10} = \frac{30}{90} = \frac{1}{3} \approx 0.333 \quad (33.3\%)$$


* **C の収率 $Y_C$:**

$$Y_C = X_A \cdot \Phi_C = 0.90 \times \frac{1}{3} = \mathbf{0.30 \quad (30\%)}$$



#### 【整合性の確認】

1. **全選択率の和 ($\Phi_B + \Phi_C = 1.0$):**

$$\Phi_B + \Phi_C = \frac{2}{3} + \frac{1}{3} = \mathbf{1.00 \quad (100\%)}$$



「反応した A（90 mol/min）の 66.7%（60 mol/min分）が B に、33.3%（30 mol/min分）が C に使われた」となり、選択率の和がぴったり 1.0 になります。
2. **収率の和 ($Y_B + Y_C = X_A$):**

$$Y_B + Y_C = 0.60 + 0.30 = \mathbf{0.90} = X_A$$



「供給した A（100 mol/min）の 60% が B に、30% が C に変換され、合計 90%（＝転化率 $X_A$）が反応して消費された」となり、収率の和が転化率 $X_A$ に一致することが確認できます。




転化率（いくら反応したか）だけを追うと失敗し、**「選択率（欲しいものがどれだけできたか）」を意識しないと利益や効率が激減する**という実践的な差を体感できるオリジナルの演習問題を作成しました。

前述の並列反応の数式モデル（$2\text{A} \rightarrow \text{B}$, $3\text{A} \rightarrow 2\text{C}$）を踏襲しつつ、定性・定量の両面から理解が深まる構成にしています。

---

### 【演習問題】：化学プラントの設計変更と利益の最大化

ある化学工場で、高価な有用薬品 **B** を合成するために以下の液相並列反応を行っています。副生成物 **C** は廃棄処理費用がかかる産業廃棄物です。

1. **主反応（2次反応）:** $2\text{A} \xrightarrow{k_1} \text{B} \quad \left(r_1 = k_1 C_A^2, \quad k_1 = 0.1 \text{ L/(mol}\cdot\text{min)}\right)$
2. **副反応（1次反応）:** $3\text{A} \xrightarrow{k_2} 2\text{C} \quad \left(r_2 = k_2 C_A, \quad k_2 = 0.1 \text{ min}^{-1}\right)$

**プロセス条件:**

* 原料供給量: $F_{A0} = 100 \text{ mol/min}$
* 供給濃度: $C_{A0} = 2.0 \text{ mol/L}$（したがって供給体積流量 $v_0 = 50 \text{ L/min}$）

現場の若手エンジニアが、「反応器のサイズ（体積 $V$）を大きくして原料 **A** の**転化率 $X_A$ を 50% から 90% に引き上げれば、目的物 B の生産量も増えて売上・利益が跳ね上がるはずだ**」と提案し、反応器を大型の完全混合槽型反応器（**CSTR**）に変更しました。

---

#### 【問い】

1. **【転化率と選択率の計算】**
CSTR において、転化率を **$X_A = 0.50$**（変更前）から **$X_A = 0.90$**（変更後）に引き上げたとき、主生成物 **B** への全選択率 $\Phi_B$（消費された A のうち B の生成に使われた割合）はどのように変化するか求めなさい。

2. **【エンジニアの勘違いの解説】**
転化率を 90% に高めた結果、目的物 B の「1 リットルあたりの生成量（収率）」および「プラントの経済性（効率）」はどうなったか説明し、エンジニアの提案が失敗に終わった定性的な理由を述べなさい。

---

### 解答と解説

#### Step 1: 転化率と選択率の計算（問1の解答）

CSTR における出口の原料濃度は $C_A = C_{A0}(1 - X_A)$ です。
また、CSTR の瞬間選択率（＝全選択率 $\Phi_B$）は以下の式で表されます。

$$\Phi_B = \frac{2 r_1}{-r_A} = \frac{2 k_1 C_A^2}{2 k_1 C_A^2 + 3 k_2 C_A} = \frac{2 C_A}{2 C_A + 3}$$

* **変更前（$X_A = 0.50$ のとき）：**
* 出口濃度: $C_A = 2.0 \times (1 - 0.50) = 1.0 \text{ mol/L}$
* 全選択率: $\Phi_B = \frac{2(1.0)}{2(1.0) + 3} = \frac{2}{5} = \mathbf{0.40 \quad (40\%)}$


* **変更後（$X_A = 0.90$ のとき）：**
* 出口濃度: $C_A = 2.0 \times (1 - 0.90) = 0.20 \text{ mol/L}$
* 全選択率: $\Phi_B = \frac{2(0.20)}{2(0.20) + 3} = \frac{0.40}{3.40} \approx \mathbf{0.118 \quad (11.8\%)}$



---

#### Step 2: 収率とプラント経済性の比較（問2の解答）

原料 A 1 mol あたりから生成する B のモル数（収率 $Y_B$）は、$Y_B = X_A \times \frac{\Phi_B}{2}$（※ A 2モルから B 1モルができるため $\frac{1}{2}$ を掛ける）となります。

| 指標 | 変更前 ($X_A = 50\%$) | 変更後 ($X_A = 90\%$) | 比較結果 |
| --- | --- | --- | --- |
| **転化率 $X_A$** | $50\%$ | **$90\%$** | 大幅アップ（一見成功に見える） |
| **選択率 $\Phi_B$** | **$40\%$** | $11.8\%$ | **激減（約 $1/3.4$ に低下）** |
| **B の収率 $Y_B$** | **$0.10 \text{ mol/mol-A}$** | $0.053 \text{ mol/mol-A}$ | **目的物の獲得量はほぼ半減！** |
| **副生物 C の割合** | 60% | **88.2%** | 大部分が無駄なゴミになる |

##### 【定性的な理由・ポイント】

1. **主反応（2次）と副反応（1次）の「反応次数」の違い:**
主反応は濃度 $C_A$ の2乗に比例して進みます。転化率を 90% にしようとして CSTR 内の濃度を $C_A = 0.20$ まで下げてしまうと、高次である主反応（2次）の速度が極端に低下し、相対的に低次（1次）の副反応ばかりが進むようになります。
2. **「転化率」と「選択率」のトレードオフ:**
単一反応では「転化率を上げる ＝ 目的物がより多く取れる」が成り立ちますが、複合反応では**転化率を追うあまり選択率を犠牲にすると、原料を消費しているにもかかわらず目的物の生成量が減り、大量の副生成物（廃棄物）を生み出してしまう**という逆転現象が起こります。

---

### この問題で学べる本質（まとめ）

* 単一反応 $\implies$ **「転化率（$X_A$）の最大化」** ＝ 装置の最適設計
* 複合反応 $\implies$ **「選択率（$\Phi_B$）と転化率の積（＝収率 $Y_B$）の最大化」** ＝ 装置の最適設計

複合反応においては、「とにかく反応させれば良い（高転化率）」ではなく、「高濃度を維持して欲しい反応だけを優先して進ませる（高選択率）」という**反応特性に応じた設計思考**が不可欠であることが、数値で実感できる問題となっています。



---

### 【問2】：指定の転化率・選択率から反応器体積 $V$ を求める


以下の並列反応が液相（等容系）で行われています。

1. **主反応（2次反応）:** $2\text{A} \xrightarrow{k_1} \text{B} \quad \left(r_1 = k_1 C_A^2, \quad k_1 = 0.1 \text{ L/(mol}\cdot\text{min)}\right)$
2. **副反応（1次反応）:** $3\text{A} \xrightarrow{k_2} 2\text{C} \quad \left(r_2 = k_2 C_A, \quad k_2 = 0.1 \text{ min}^{-1}\right)$

**プロセス条件:**

* 原料供給量: $F_{A0} = 100 \text{ mol/min}$
* 供給濃度: $C_{A0} = 2.0 \text{ mol/L}$（したがって供給体積流量 $v_0 = 50 \text{ L/min}$）

**【設計目標（仕様指定）】**

* **目標転化率:** $X_A = 0.80$ （80%）
* **目標全選択率（主生成物 B）:** $\Phi_B = 0.80$ （80%）


完全混合槽型反応器（**CSTR**）を用いてこの目標（$X_A = 0.80, \Phi_B = 0.80$）を達成できるか検討しなさい。
可能な場合は必要な**反応器体積 $V_{\text{CSTR}}$** を求めなさい。
達成できない場合は、その根拠を示しなさい。

---

### 解答と解説

#### Step 1: 出口での A の濃度 $C_A$ の決定

目標転化率が $X_A = 0.80$ なので、CSTR 出口における A の濃度 $C_A$ は次のようになります。

$$C_A = C_{A0}(1 - X_A) = 2.0 \times (1 - 0.80) = \mathbf{0.40 \text{ mol/L}}$$

CSTR は完全混合であるため、**反応器内部の A の濃度はどこでも均一に $C_A = 0.40 \text{ mol/L}$** となります。

---

#### Step 2: 反応速度式の評価

A の全消費速度 $-r_A$ は以下のように書けます。

$$-r_A = 2r_1 + 3r_2 = 2(k_1 C_A^2) + 3(k_2 C_A)$$

ここで、出口濃度 $C_A = 0.40 \text{ mol/L}$ を代入して各項の反応速度を計算します。

* **主反応による A の消費速度:**

$$2 r_1 = 2 \times 0.1 \times (0.40)^2 = 2 \times 0.016 = \mathbf{0.032 \text{ mol/(L}\cdot\text{min)}}$$


* **副反応による A の消費速度:**

$$3 r_2 = 3 \times 0.1 \times 0.40 = \mathbf{0.120 \text{ mol/(L}\cdot\text{min)}}$$


* **A の全消費速度 $-r_A$:**

$$-r_A = 0.032 + 0.120 = \mathbf{0.152 \text{ mol/(L}\cdot\text{min)}}$$



---

#### Step 3: 全選択率 $\Phi_B$ の検証

（※CSTR 内の濃度条件で、目標の全選択率 $\Phi_B = 0.80$ が達成できているかを確認します）

CSTRにおける瞬間全選択率はそのまま全体の全選択率 $\Phi_B$ になるため：

$$\Phi_B = \frac{2 r_1}{-r_A} = \frac{0.032}{0.152} \approx \mathbf{0.21 \quad (21\%)}$$

> **【重要な気づき・ポイント】**
> CSTR では A の濃度が低く維持されるため（$C_A = 0.40$）、高次元（2次）の主反応速度 $r_1$ が極端に低下し、選択率は **21%** しか達成できません。
> つまり、**「反応速度定数 $k_1, k_2$ と濃度条件が固定されている場合、CSTR では $X_A = 0.80$ と $\Phi_B = 0.80$ を同時指定することは物理的に不可能」**であることが分かります。

---

#### Step 4: PFR（管型反応器）の場合の検証と体積計算

主反応が2次、副反応が1次の場合、**原料 A の濃度を高く保てる PFR の方が選択率を大幅に引き上げることができます。**

PFR における局所（瞬間）全選択率 $\Phi'_B$ は以下のようになります。

$$\Phi'_B = \frac{2 k_1 C_A^2}{2 k_1 C_A^2 + 3 k_2 C_A} = \frac{2 k_1 C_A}{2 k_1 C_A + 3 k_2}$$

* **入口（$C_A = 2.0$）での瞬間選択率:**

$$\Phi'_B = \frac{2(0.1)(2.0)}{2(0.1)(2.0) + 3(0.1)} = \frac{0.4}{0.7} \approx \mathbf{57.1\%}$$


* **出口（$C_A = 0.40$）での瞬間選択率:**

$$\Phi'_B = \frac{2(0.1)(0.4)}{2(0.1)(0.4) + 3(0.1)} = \frac{0.08}{0.38} \approx \mathbf{21.1\%}$$



---

### まとめ：指定された $X_A$ と $\Phi_B$ から体積 $V$ を計算する流れ

1. **目標 $X_A$ から出口濃度 $C_A$ を決める。**
2. **指定の全選択率 $\Phi_B$ が成り立つための濃度・操作条件（または必要な反応器形式）を特定する。**
3. **設計方程式に代入して体積 $V$ を計算する。**
* **CSTR:** $V = \dfrac{F_{A0} X_A}{-r_A(C_A)}$
* **PFR:** $V = F_{A0} \displaystyle\int_{0}^{X_A} \frac{dX_A}{-r_A}$



「指定の転化率と選択率を両立させるために、どの形式の反応器（PFR vs CSTR）を選び、どれだけの体積 $V$ にすべきか」という問題設定は、まさに反応工学の醍醐味と言えます。



###【問】並列反応における選択率・収率と反応器体積の設計

原料 $A$ から目的生成物 $B$ と副生成物 $C$ が生成する、次の不可逆並列反応を考える。

$$
A \xrightarrow{r_1} B
$$

$$
A \xrightarrow{r_2} C
$$

反応速度は次式で表される。

$$
r_1=k_1C_A
$$

$$
r_2=k_2C_A^2
$$

ここで、$C_A$ は $A$ の濃度 $[\mathrm{mol/L}]$、反応速度 $r_1,r_2$ は
$[\mathrm{mol/(L\cdot min)}]$ とする。

反応速度定数は

$$
k_1=0.10\ {\rm min^{-1}}
$$

$$
k_2=0.050\ {\rm L/(mol\cdot min)}
$$

である。

原料 $A$ は

$$
C_{A0}=2.0\ {\rm mol/L}
$$

の濃度で、体積流量

$$
v_0=10\ {\rm L/min}
$$

で連続的に反応器へ供給されるものとする。

液相反応であり、反応中の体積流量は一定とする。

---

## 問1 

目的生成物 $B$ に対する副生成物 $C$ の瞬間選択率を

$$
S=\frac{r_1}{r_2}
$$

と定義する。

$S$ を $C_A$ の関数として表しなさい。

また、$C_A=2.0$、$1.0$、$0.5\ {\rm mol/L}$ のときの瞬間選択率を求めなさい。

◇◇◇

<details>


問1の解答は以下の通りです。

### 1. $S$ を $C_A$ の関数として表す

瞬間選択率 $S$ の定義式に、与えられた反応速度式 $r_1 = k_1 C_A$ および $r_2 = k_2 C_A^2$ を代入します。

$$S = \frac{r_1}{r_2} = \frac{k_1 C_A}{k_2 C_A^2} = \frac{k_1}{k_2 C_A}$$

反応速度定数の値 $k_1 = 0.10 \text{ min}^{-1}$、$k_2 = 0.050 \text{ L/(mol}\cdot\text{min)}$ を代入します。

$$S = \frac{0.10}{0.050 \cdot C_A} = \frac{2}{C_A}$$

したがって、$S$ を $C_A$ の関数として表すと以下のようになります。

$$S = \frac{2}{C_A} \quad (\text{または } 2 C_A^{-1})$$

---

### 2. 各 $C_A$ における瞬間選択率 $S$ の値

得られた $S = \frac{2}{C_A}$ に各濃度を代入して求めます。

* **$C_A = 2.0 \text{ mol/L}$ のとき**

$$S = \frac{2}{2.0} = \mathbf{1.0}$$


* **$C_A = 1.0 \text{ mol/L}$ のとき**

$$S = \frac{2}{1.0} = \mathbf{2.0}$$


* **$C_A = 0.5 \text{ mol/L}$ のとき**

$$S = \frac{2}{0.5} = \mathbf{4.0}$$



---

### 解答のまとめ

* **$S$ の関数表示:** $S = \dfrac{2}{C_A}$
* **$C_A = 2.0 \text{ mol/L}$ のときの $S$:** $1.0$
* **$C_A = 1.0 \text{ mol/L}$ のときの $S$:** $2.0$
* **$C_A = 0.5 \text{ mol/L}$ のときの $S$:** $4.0$

</details>

## 問2
問１の結果から、反応が進んで $C_A$ が小さくなると、Bに対する選択性がどのように変化するか説明しなさい。

さらに、

$$
r_1=k_1C_A,\qquad r_2=k_2C_A^2
$$

を比較して、

> $A$ の濃度を高く保つことと、低くすることのどちらがBの生成に有利か

を説明しなさい。

<details>
問2の解答は以下の通りです。

### 1. $C_A$ が小さくなるときの選択性の変化

問1より、目的生成物 B に対する副生成物 C の瞬間選択率は $S = \dfrac{r_1}{r_2} = \dfrac{2}{C_A}$ と表されます。

反応が進んで原料濃度 $C_A$ が小さくなると、分母の $C_A$ が減少するため、**瞬間選択率 $S$（副生成物 C に対する目的生成物 B の生成速度比 $r_1/r_2$）は大きくなります**。

すなわち、反応が進行して $C_A$ が低下するほど、**目的生成物 B に対する選択性が高くなります（B が相対的に生成しやすくなります）**。

---

### 2. $A$ の濃度と B の生成に対する有利・不利の比較

反応速度式を比較すると、以下のようになります。

* 目的生成物 B の生成速度: $r_1 = k_1 C_A$ （$C_A$ の1次）
* 副生成物 C の生成速度: $r_2 = k_2 C_A^2$ （$C_A$ の2次）

目的生成物 B の反応次数（1次）は、副生成物 C の反応次数（2次）よりも低いため、$C_A$ の変化による影響を比較すると次のようになります。

* **$A$ の高濃度時:** 2次反応である副生成物 C の生成速度 $r_2$ がより急激に増大するため、C の生成が優勢になります。
* **$A$ の低濃度時:** $C_A^2$ の値が著しく小さくなるため、副生成物 C の生成が抑えられ、1次反応である目的生成物 B の生成速度 $r_1$ が相対的に優勢になります。

したがって、**$A$ の濃度を低く保つことのほうが B の生成に有利**です。

---

### 解答のまとめ

* **選択性の変化:** $C_A$ が小さくなるほど瞬間選択率 $S = r_1/r_2$ は増加するため、**Bに対する選択性は高くなる（上昇する）**。
* **濃度の有利・不利:** B の反応次数（1次）が C の反応次数（2次）より低いため、**$A$ の濃度を低く保つことのほうが B の生成に有利**である。

</details>

## 問3

Aの転化率を

$$
X_A=\frac{C_{A0}-C_A}{C_{A0}}
$$

と定義する。

Bの収率を

$$
Y_B=\frac{\text{生成したBの量}}
{\text{供給したAの量}}
$$

と定義するとき、$Y_B$を $X_A$ と選択率を用いて表しなさい。

ただし、この反応では選択率が一定ではないことに注意しなさい。

<details>
目的生成物 B の収率 $Y_B$ を、転化率 $X_A$ および選択率（瞬間選択率または全選択率）を用いて表します。

---

### 1. 瞬時選択率 $\varphi_B$ を用いた表示（一般式）

原料 A の反応速度（消費速度）を $-r_A = r_1 + r_2$ とすると、目的生成物 B への**瞬時選択率 $\varphi_B$**（反応物 A 1 mol 消費あたりに生成する B の mol 数）は次のように定義されます。

$$\varphi_B = \frac{r_1}{-r_A} = \frac{r_1}{r_1 + r_2}$$

* **連続式（PFR / 回分反応器）の場合:**
選択率が濃度 $C_A$（すなわち転化率 $X_A$）に応じて変化するため、収率 $Y_B$ は瞬時選択率 $\varphi_B$ を $0$ から $X_A$ まで積分して表されます。
$$Y_B = \int_0^{X_A} \varphi_B \, dX_A$$


* **完全混合槽型反応器（CSTR）の場合:**
槽内および出口の濃度（転化率 $X_A$）が一様で一定となるため、出口における瞬時選択率を $\varphi_{B,\text{exit}}$ とすると、積分を行わずに次のように表されます。
$$Y_B = \varphi_{B,\text{exit}} \cdot X_A$$



※なお、問1で与えられた副生成物 C に対する速度比 $S = \frac{r_1}{r_2}$ を用いると、$\varphi_B = \frac{S}{1 + S}$ と変換できるため、$Y_B = \int_0^{X_A} \frac{S}{1 + S} \, dX_A$ と書くこともできます。

---

### 2. 全選択率 $\Phi_B$ を用いた表示

反応器全体での**全選択率（統合選択率）$\Phi_B$**（消費された A の総量に対する生成した B の総量の割合）を次のように定義します。

$$\Phi_B = \frac{\text{生成した B の量}}{\text{消費された A の量}}$$

このとき、収率 $Y_B$ の定義式に変形を加えると次のように表せます。

$$Y_B = \frac{\text{生成した B の量}}{\text{供給した A の量}} = \left( \frac{\text{生成した B の量}}{\text{消費された A の量}} \right) \cdot \left( \frac{\text{消費された A の量}}{\text{供給した A の量}} \right)$$

したがって、全選択率 $\Phi_B$ と転化率 $X_A$ を用いると、反応器の形式に関わらず常にシンプルに積の形で表されます。

$$Y_B = \Phi_B \cdot X_A$$

---

### 解答のまとめ

選択率の定義の取り方により、以下のように表されます。

* **瞬時選択率 $\varphi_B = \frac{r_1}{r_1 + r_2}$ を用いる場合（一般式）:**

$$Y_B = \int_0^{X_A} \varphi_B \, dX_A \quad \text{（PFR / 回分反応器）}$$


$$Y_B = \varphi_{B,\text{exit}} \cdot X_A \quad \text{（CSTR）}$$


* **全選択率 $\Phi_B = \frac{\text{生成した B の量}}{\text{消費された A の量}}$ を用いる場合:**

$$Y_B = \Phi_B \cdot X_A$$
</details>

## （3）Bの収率を定義しなさい ― 解答例

AからBとCが生成する並列反応

$$
A\xrightarrow{r_1}B
$$

$$
A\xrightarrow{r_2}C
$$

を考えます。

Aの転化率を

$$
\boxed{
X_A=\frac{C_{A0}-C_A}{C_{A0}}
}
$$

と定義します。

---

### ① Bの収率

Bの収率 \(Y_B\) は、

$$
\boxed{
Y_B=
\frac{\text{生成したBの量}}
{\text{供給したAの量}}
}
$$

と定義します。

液相で体積流量が一定なら、

$$
F_{A0}=v_0C_{A0}
$$

ですから、

$$
\boxed{
Y_B=\frac{F_B}{F_{A0}}
}
$$

また、

$$
F_B=v_0C_B
$$

なので、

$$
\boxed{
Y_B=\frac{C_B}{C_{A0}}
}
$$

となります。

---

### ② 選択率との関係

ここで注意が必要です。

今回の瞬間選択率は

$$
S=\frac{r_1}{r_2}
$$

です。

したがって、BとCの生成量の比も、**選択率が一定なら**

$$
\frac{F_B}{F_C}=S
$$

となります。

しかし今回の反応では、

$$
S=\frac{k_1}{k_2C_A}
$$

なので、\(C_A\) が変化すると選択率も変化します。

したがって、

$$
\boxed{
Y_B=S X_A
}
$$

とは**そのままでは書けません**。

これは今回の問題で重要なポイントです。

---

### ③ Bの収率を求める

Bの生成速度は

$$
r_1=k_1C_A
$$

です。

したがって、反応器内でAが少し消費されたとき、その一部がBになります。

Aの消費速度は

$$
-r_A=r_1+r_2
$$

なので、

$$
-r_A=k_1C_A+k_2C_A^2
$$

です。

したがって、Aが消費される瞬間にBになる割合は

$$
\frac{r_1}{r_1+r_2}
$$

です。

瞬間選択率

$$
S=\frac{r_1}{r_2}
$$

を使えば、

$$
\frac{r_1}{r_1+r_2}
=
\frac{S}{1+S}
$$

となります。

したがって、

$$
\boxed{
\text{瞬間的にAがBになる割合}
=
\frac{S}{1+S}
}
$$

です。

---

### ④ 今回の反応では

$$
S=\frac{k_1}{k_2C_A}
$$

なので、

$$
\frac{S}{1+S}
=
\frac{k_1/(k_2C_A)}
{1+k_1/(k_2C_A)}
$$

整理すると、

$$
\boxed{
\frac{r_1}{r_1+r_2}
=
\frac{k_1}{k_1+k_2C_A}
}
$$

となります。

つまり、

$$
\boxed{
Y_B
=
\frac{\text{Bとして生成したA}}
{\text{供給したA}}
}
$$

であり、今回のように選択率が変化する場合には、**反応器全体についてB生成量を積分して求める必要があります。**

---

### 学生に押さえてほしいこと

今回の問題では、

$$
\boxed{
S=\frac{r_1}{r_2}
}
$$

は**BとCの生成速度の比**です。

一方、

$$
\boxed{
\frac{S}{1+S}
=
\frac{r_1}{r_1+r_2}
}
$$

は、**その瞬間に消費されたAのうちBになった割合**です。

そして、

$$
\boxed{
Y_B\neq SX_A
}
$$

であることに注意します。

これは \(S\) が反応器内で一定ではなく、

$$
\boxed{
S=\frac{k_1}{k_2C_A}
}
$$

として \(C_A\) とともに変化するためです。

この点を理解すると、次の（4）～（6）で**CSTRの物質収支から実際のB収率を求める**意味が明確になります。

# CSTRによる設計

次に、完全混合型連続撹拌槽反応器（CSTR）を用いる。

反応器出口の濃度は反応器内の濃度と等しい。

---

## (4) Aについて物質収支を立てなさい

定常状態において、Aについて

$$
\boxed{
\text{流入}-\text{流出}-\text{反応による消費}=0
}
$$

の物質収支を立てなさい。

さらに、

$$
C_A=C_{A0}(1-X_A)
$$

を用いて、必要反応器体積 $V$ を $X_A$ の関数として表しなさい。

---
<details>
## （4）Aについて物質収支 ― 解答例

反応は

$$
A\xrightarrow{r_1}B
$$

$$
A\xrightarrow{r_2}C
$$

であり、

$$
r_1=k_1C_A,\qquad
r_2=k_2C_A^2
$$

です。

CSTRでは、反応器内の濃度と出口濃度が等しいので、反応速度も反応器内の濃度 \(C_A\) で評価します。

### ① 定常状態の物質収支

Aについて、

$$
\text{流入}-\text{流出}-\text{反応による消費}=0
$$

ですから、

$$
F_{A0}-F_A-V(r_1+r_2)=0
$$

となります。

液相で体積流量 \(v_0\) が一定なので、

$$
F_{A0}=v_0C_{A0}
$$

$$
F_A=v_0C_A
$$

です。

したがって、

$$
v_0C_{A0}-v_0C_A
-
V(r_1+r_2)=0
$$

となります。

ここに

$$
r_1=k_1C_A
$$

$$
r_2=k_2C_A^2
$$

を代入すると、

$$
v_0(C_{A0}-C_A)
=
V(k_1C_A+k_2C_A^2)
$$

したがって、

$$
\boxed{
V=
\frac{v_0(C_{A0}-C_A)}
{k_1C_A+k_2C_A^2}
}
$$

となります。

---

### ② 転化率 \(X_A\) を使う

問題文より、

$$
C_A=C_{A0}(1-X_A)
$$

です。

また、

$$
C_{A0}-C_A
=
C_{A0}-C_{A0}(1-X_A)
=
C_{A0}X_A
$$

です。

したがって、

$$
V=
\frac{v_0C_{A0}X_A}
{k_1C_{A0}(1-X_A)
+k_2C_{A0}^2(1-X_A)^2}
$$

となります。

分母から \(C_{A0}\) をくくると、

$$
\boxed{
V=
\frac{v_0X_A}
{k_1(1-X_A)
+k_2C_{A0}(1-X_A)^2}
}
$$

が求める式です。

---

### ③ 今回の数値を代入すると

$$
v_0=10\ {\rm L/min}
$$

$$
C_{A0}=2.0\ {\rm mol/L}
$$

$$
k_1=0.10\ {\rm min^{-1}}
$$

$$
k_2=0.050\ {\rm L/(mol\cdot min)}
$$

なので、

$$
V=
\frac{10X_A}
{0.10(1-X_A)
+0.050(2.0)(1-X_A)^2}
$$

したがって、

$$
\boxed{
V=
\frac{10X_A}
{0.10(1-X_A)+0.10(1-X_A)^2}
}
$$

です。

さらに整理すると、

$$
\boxed{
V=
\frac{10X_A}
{0.10(1-X_A)(2-X_A)}
}
$$

したがって、

$$
\boxed{
V=
\frac{100X_A}
{(1-X_A)(2-X_A)}
\quad {\rm L}
}
$$

となります。

### 最終答

$$
\boxed{
\displaystyle
V=
\frac{v_0X_A}
{k_1(1-X_A)+k_2C_{A0}(1-X_A)^2}
}
$$

今回の条件では、

$$
\boxed{
\displaystyle
V=
\frac{100X_A}{(1-X_A)(2-X_A)}
\ {\rm L}
}
$$

です。

この式が、次の（5）で**Bの収率 \(Y_B\) を求める際のCSTR設計式**になります。

</details>

## (5) Bについて物質収支を立てなさい

Bについて定常状態の物質収支を立て、

$$
C_B
$$

を $C_A$ の関数として表しなさい。

その結果を用いて、Bの収率

$$
Y_B=\frac{v_0C_B}{v_0C_{A0}}
$$

を $X_A$ の関数として求めなさい。

---
<details>
## （5）Bについて物質収支 ― 解答例

反応

$$
A\xrightarrow{r_1}B
$$

$$
A\xrightarrow{r_2}C
$$

において、**Bは反応1によってのみ生成**します。

したがって、Bの生成速度は

$$
r_1=k_1C_A
$$

です。

---

### ① Bについての物質収支

定常状態のCSTRでは、

$$
\text{流入}-\text{流出}+\text{生成}=0
$$

です。

Bは原料中に含まれていないので、

$$
F_{B0}=0
$$

です。

したがって、

$$
0-F_B+Vr_1=0
$$

となります。

よって、

$$
\boxed{
F_B=Vr_1
}
$$

です。

ここで

$$
r_1=k_1C_A
$$

なので、

$$
\boxed{
F_B=Vk_1C_A
}
$$

となります。

---

### ② Bの濃度を求める

液相で体積流量 \(v_0\) が一定なので、

$$
F_B=v_0C_B
$$

です。

したがって、

$$
v_0C_B=Vk_1C_A
$$

より、

$$
\boxed{
C_B=\frac{Vk_1C_A}{v_0}
}
$$

となります。

---

### ③ Aの物質収支の結果を利用する

（4）で求めたAの物質収支は、

$$
v_0(C_{A0}-C_A)
=
V(k_1C_A+k_2C_A^2)
$$

でした。

したがって、

$$
V=
\frac{v_0(C_{A0}-C_A)}
{k_1C_A+k_2C_A^2}
$$

です。

これを

$$
C_B=\frac{Vk_1C_A}{v_0}
$$

に代入します。

$$
C_B
=
\frac{k_1C_A}{v_0}
\frac{v_0(C_{A0}-C_A)}
{k_1C_A+k_2C_A^2}
$$

整理すると、

$$
C_B
=
\frac{k_1(C_{A0}-C_A)}
{k_1+k_2C_A}
$$

したがって、

$$
\boxed{
C_B=
\frac{k_1(C_{A0}-C_A)}
{k_1+k_2C_A}
}
$$

となります。

---

### ④ Bの収率 \(Y_B\)

Bの収率を

$$
Y_B=\frac{F_B}{F_{A0}}
$$

と定義します。

体積流量一定なので、

$$
Y_B
=
\frac{v_0C_B}{v_0C_{A0}}
=
\frac{C_B}{C_{A0}}
$$

です。

したがって、

$$
Y_B
=
\frac{k_1(C_{A0}-C_A)}
{C_{A0}(k_1+k_2C_A)}
$$

ここで、

$$
C_A=C_{A0}(1-X_A)
$$

を代入すると、

$$
C_{A0}-C_A=C_{A0}X_A
$$

なので、

$$
Y_B
=
\frac{k_1C_{A0}X_A}
{C_{A0}[k_1+k_2C_{A0}(1-X_A)]}
$$

よって、

$$
\boxed{
Y_B=
\frac{k_1X_A}
{k_1+k_2C_{A0}(1-X_A)}
}
$$

が得られます。

---

### ⑤ 今回の数値を代入

$$
k_1=0.10
$$

$$
k_2=0.050
$$

$$
C_{A0}=2.0
$$

なので、

$$
Y_B
=
\frac{0.10X_A}
{0.10+0.050(2.0)(1-X_A)}
$$

分母を整理すると、

$$
Y_B
=
\frac{0.10X_A}
{0.10+0.10(1-X_A)}
$$

したがって、

$$
\boxed{
Y_B=\frac{X_A}{2-X_A}
}
$$

となります。

---

### （5）の最終答

$$
\boxed{
F_B=Vr_1=Vk_1C_A
}
$$

$$
\boxed{
C_B=
\frac{k_1(C_{A0}-C_A)}
{k_1+k_2C_A}
}
$$

そしてBの収率は、

$$
\boxed{
Y_B=
\frac{k_1X_A}
{k_1+k_2C_{A0}(1-X_A)}
}
$$

今回の条件では、

$$
\boxed{
Y_B=\frac{X_A}{2-X_A}
}
$$

です。

ここから（6）では、この

$$
Y_B=\frac{X_A}{2-X_A}
$$

を使って、**「Bの収率が最大になる \(X_A\) は存在するのか？」**を調べることになります。

</details>

---

## (6) Bの収率が最大となる転化率を求めなさい

(5)で求めた $Y_B(X_A)$ を微分し、

$$
\frac{dY_B}{dX_A}=0
$$

となる条件を求めなさい。

この結果から、CSTRにおいてBの収率を最大にする転化率

$$
\boxed{X_{A,\mathrm{opt}}}
$$

を求めなさい。

<details>
## （6）Bの収率が最大となる転化率 ― 解答例

（5）で、CSTRにおけるBの収率は

$$
\boxed{
Y_B=\frac{X_A}{2-X_A}
}
$$

と求めました。

ここで、

$$
0\leq X_A<1
$$

です。

### ① 収率を転化率で微分する

Bの収率

$$
Y_B=\frac{X_A}{2-X_A}
$$

を \(X_A\) で微分します。

商の微分より、

$$
\frac{dY_B}{dX_A}
=
\frac{(2-X_A)-X_A(-1)}
{(2-X_A)^2}
$$

したがって、

$$
\frac{dY_B}{dX_A}
=
\frac{2}
{(2-X_A)^2}
$$

です。

つまり、

$$
\boxed{
\frac{dY_B}{dX_A}>0
}
$$

が \(0\leq X_A<1\) の範囲で常に成立します。

---

### ② したがって収率はどうなるか

$$
\frac{dY_B}{dX_A}>0
$$

なので、\(X_A\) が大きくなるほど \(Y_B\) も大きくなります。

したがって、Bの収率を最大にするには、

$$
\boxed{
X_A\rightarrow1
}
$$

とすればよいことになります。

ただし、\(X_A=1\) をCSTRで実現しようとすると、（4）で求めた

$$
V=
\frac{100X_A}
{(1-X_A)(2-X_A)}
$$

より、

$$
X_A\rightarrow1
$$

のとき

$$
\boxed{
V\rightarrow\infty
}
$$

となります。

つまり、**100%転化率を有限のCSTR体積で達成することはできません。**

---

### ③ \(X_A\rightarrow1\) のときのB収率

$$
Y_B=\frac{X_A}{2-X_A}
$$

に

$$
X_A\rightarrow1
$$

を代入すると、

$$
Y_B\rightarrow
\frac{1}{2-1}=1
$$

したがって、

$$
\boxed{
Y_B\rightarrow100\%
}
$$

となります。

---

## 結論

$$
\boxed{
\text{Bの収率は }X_A\text{ が大きいほど高くなる}
}
$$

したがって理論上、

$$
\boxed{
X_A\rightarrow1
\quad\Rightarrow\quad
Y_B\rightarrow100\%
}
$$

です。

しかし、

$$
\boxed{
X_A\rightarrow1
\quad\Rightarrow\quad
V\rightarrow\infty
}
$$

なので、**「Bの収率最大化」だけを考えると、必要反応器体積が無限大になる**という結果になります。

---

### この問題から分かること

ここがこの演習の重要なポイントです。

$$
\boxed{
\text{収率を最大化}
\quad\Longrightarrow\quad
X_A\text{をできるだけ大きくする}
}
$$

一方、

$$
\boxed{
\text{装置体積を小さくする}
\quad\Longrightarrow\quad
X_A\text{を小さくする}
}
$$

という**トレードオフ**があります。

したがって実際の装置設計では、

> 「Bの収率を100%にする」

という設計ではなく、

$$
\boxed{
Y_B\geq Y_{B,\mathrm{target}}
}
$$

のような**目標収率を設定し、その条件を満たす最小の反応器体積 \(V\) を求める**

という考え方が実用的です。

この問題設定なら、次の（7）で \(V\) を計算する流れがきれいにつながります。

</details>

---

## (7) 最適転化率における必要体積を求めなさい

(6)で求めた

$$
X_{A,\mathrm{opt}}
$$

をCSTRの設計式に代入し、Bの収率を最大にするために必要な反応器体積

$$
\boxed{V_{\mathrm{opt}}}
$$

を求めなさい。

<details>
## （7）最適転化率における必要体積 ― 解答例

（6）では、Bの収率

$$
Y_B=\frac{X_A}{2-X_A}
$$

は \(X_A\) が大きくなるほど増加することが分かりました。

したがって、Bの収率だけを最大化するなら、

$$
\boxed{X_A\rightarrow1}
$$

が最適です。

---

### ① CSTRの設計式

（4）で求めたCSTRの必要体積は、

$$
V=
\frac{100X_A}
{(1-X_A)(2-X_A)}
$$

でした。

したがって、最適転化率

$$
X_A\rightarrow1
$$

を代入すると、

$$
V
=
\frac{100X_A}
{(1-X_A)(2-X_A)}
$$

より、

$$
\lim_{X_A\rightarrow1}V
=
\infty
$$

となります。

したがって、

$$
\boxed{
V_{\mathrm{opt}}\rightarrow\infty
}
$$

です。

---

## ② なぜ無限大になるのか

CSTRでは、反応器内が完全混合されているため、

$$
C_A=C_{A0}(1-X_A)
$$

です。

転化率を100%に近づけると、

$$
X_A\rightarrow1
$$

なので、

$$
C_A\rightarrow0
$$

となります。

反応速度は

$$
r_1+r_2
=
k_1C_A+k_2C_A^2
$$

ですから、

$$
C_A\rightarrow0
$$

では

$$
r_1+r_2\rightarrow0
$$

となります。

つまり、

> **反応をほぼ完全に進めようとすると、反応速度がほぼゼロの状態で反応を続けなければならない**

ため、必要な反応器体積が非常に大きくなります。

---

## ③ 「最適転化率」の意味に注意

ここで重要なのは、

$$
\boxed{\text{B収率の最大化}}
$$

だけを目的関数にした場合、

$$
X_A\rightarrow1
$$

となりますが、

$$
\boxed{V\rightarrow\infty}
$$

となることです。

したがって、**有限の装置体積を求めるという意味では、このままでは実用的な最適解はありません。**

例えば、

$$
\boxed{Y_B=0.40}
$$

以上を達成する、という設計条件に変更すれば、有限の \(V\) を決定できます。

---

### ④ 例えば \(Y_B=40\%\) を目標にすると

（5）より、

$$
Y_B=\frac{X_A}{2-X_A}
$$

です。

目標を

$$
Y_B=0.40
$$

とすると、

$$
0.40=\frac{X_A}{2-X_A}
$$

したがって、

$$
0.40(2-X_A)=X_A
$$

$$
0.80-0.40X_A=X_A
$$

$$
0.80=1.40X_A
$$

よって、

$$
\boxed{
X_A=0.5714
}
$$

です。

この転化率をCSTRの体積式

$$
V=
\frac{100X_A}
{(1-X_A)(2-X_A)}
$$

に代入すると、

$$
V=
\frac{100(0.5714)}
{(1-0.5714)(2-0.5714)}
$$

より、

$$
\boxed{
V\approx333.3\ {\rm L}
}
$$

となります。

---

### 結論

今回の問題を**「Bの収率を最大化する」だけ**として解くと、

$$
\boxed{
X_{A,\mathrm{opt}}\rightarrow1
}
$$

$$
\boxed{
Y_{B,\mathrm{max}}\rightarrow1
}
$$

ですが、

$$
\boxed{
V_{\mathrm{opt}}\rightarrow\infty
}
$$

となります。

したがって、実際の設計問題では、

$$
\boxed{
\text{「目標B収率を達成するための最小体積」}
}
$$

と設定する方が適切です。

なお、この問題はここで一つ重要な設計上の問題が出ています。**「収率最大化」だけでは体積 \(V\) が決まらない**ので、次の段階では「\(Y_B\geq40\%\) を満たす最小 \(V\)」のように問題を設定すると、反応器設計問題としてきれいに完結します。

# PFRによる設計

次に、同じ反応を管型反応器（PFR）で行う。

PFRでは反応器内の濃度が位置によって変化するため、

$$
C_A=C_{A0}(1-X_A)
$$

として、微小体積 $dV$ における物質収支を考える。

---

## (8) PFRの設計式を導きなさい

Aについて物質収支を立て、

$$
\boxed{
dV=
\frac{F_{A0}}{r_1+r_2}dX_A
}
$$

の形の式を導きなさい。

さらに、

$$
F_{A0}=v_0C_{A0}
$$

を用いて、PFRの必要体積

$$
V
$$

を $X_A$ の積分として表しなさい。

---

## (9) PFRにおけるBの生成量を求めなさい

微小体積 $dV$ で生成するBについて物質収支を立て、

$$
dF_B=r_1dV
$$

を用いて、出口におけるBの収率

$$
Y_B
$$

を求める式を導きなさい。

---

## (10) CSTRとPFRを比較しなさい

同じ最終転化率 $X_A$ を達成する場合について、

* CSTRの必要体積
* PFRの必要体積
* Bの収率
* Bの選択率

を比較しなさい。

特に、

> **なぜ反応器内のA濃度の違いが、BとCの生成割合に影響するのか**

を、反応速度式

$$
r_1=k_1C_A
$$

$$
r_2=k_2C_A^2
$$

から説明しなさい。

---

# (11) 設計問題

製品として最低でも

$$
Y_B=0.40
$$

のB収率を得たい。

この条件を満たすCSTRを設計しなさい。

求めるものは、

$$
\boxed{
X_A
}
$$

$$
\boxed{
C_A
}
$$

$$
\boxed{
C_B
}
$$

$$
\boxed{
C_C
}
$$

$$
\boxed{
V
}
$$

である。

さらに、その条件における瞬間選択率

$$
\boxed{
S=\frac{r_1}{r_2}
}
$$

を求めなさい。

---

# (12) 考察

次の問いについて考察しなさい。

### 問1

Bの収率を高くするために、単純に転化率を100%に近づければよいか。

理由を説明しなさい。

### 問2

この反応では

$$
r_1\propto C_A
$$

であるのに対して、

$$
r_2\propto C_A^2
$$

である。

この反応速度の違いは、反応器設計にどのような影響を与えるか。

### 問3

「Bをできるだけ多く作りたい」という目的と、

「Bをできるだけ選択的に作りたい」という目的は、必ずしも同じではない。

この違いを、選択率・収率・転化率を用いて説明しなさい。

### 問4

CSTRとPFRのどちらがBの製造に適しているかを、反応器内の $C_A$ の変化と選択率の観点から説明しなさい。

この問題のポイントは、**一次反応と二次反応を競合させたこと**です。

$$
S=\frac{r_1}{r_2}
=\frac{k_1C_A}{k_2C_A^2}
=\frac{k_1}{k_2C_A}
$$

なので、

$$
\boxed{C_A\downarrow\quad\Rightarrow\quad S\uparrow}
$$

となります。
