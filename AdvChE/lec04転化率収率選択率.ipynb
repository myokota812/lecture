## 1. 転化率
流通系：
$$X=\frac{F_{A0}-F_A}{F_{A0}} \tag{1}$$

回分系：

$$X=\frac{C_{A0}-C_A}{C_{A0}} \tag{2}$$

◇◇◇


<details><summary>📝🖱️説明文はこちら</summary>

まず、反応工学で最も基本的な指標である転化率を定義します。

転化率とは、「反応物がどれだけ反応したか」を表す指標です。

例えば、100 molのAを反応器に投入し、そのうち80 molが反応した場合、転化率は80％になります。

流通式反応器では、入口のモル流量を $F_{A0}$、出口のモル流量を $F_A$ とすると、(1)式で定義されます。


一方、回分反応器では流入や流出がないため、モル流量の代わりに濃度を使い(2)で表されます。

つまり転化率とは、

「最初にあったAのうち、どれだけが反応したか」

を割合で示したものです。

反応器設計では、まず目標転化率を決め、その転化率を達成するために必要な反応器体積を求めていきます。
</details>

---

## 2. 収率

目的生成物Bの収率

$$Y_B=\frac{	\text{生成したB}}{	\text{供給したA}} \tag{3}$$

収率の定義そのものは反応器によらず同じです。

<details><summary>📝🖱️説明文はこちら</summary>

次に、収率です。

収率は、「供給した反応物から、目的生成物がどれだけ得られたか」を表す指標です。

このスライドでは、目的生成物を$\text{B}$としており、その場合の収率は上式のようになります。


例えば、100 molのAを供給し、その結果として80 molのBが得られた場合、収率は0.8、つまり80%になります。

ここで重要なのは、収率は単にAが減った割合ではないという点です。

Aが反応しても、そのすべてが目的物Bになるとは限りません。

副反応によって別の物質が生成される場合、転化率が高くても収率は低くなることがあります。

したがって収率は、

「反応の結果として、目的物をどれだけ効率よく作れたか」

を評価する指標と考えることができます。

化学プロセスでは、単に転化率を上げるだけでなく、目的生成物の収率を高く維持することが重要になります。

</details>

---
CSTR・PFR・Batchでは測定する量が異なるため、実際の式の書き方が変わります。



<h3>2.1. Batch反応器の収率</h3>

　初期時刻に $\text{A}$ を仕込み、

　$N_{A0}$から$N_B$ が生成したとすると、

$$
Y_B = \frac{N_B}{N_{A0}}\tag{4}
$$
　です。
　一定体積なら
$$
Y_B = \frac{C_B}{C_{A0}}\tag{5}
$$
　とも書けます。

<h3>2.2. CSTRの収率</h3>

　流入・流出があるため、モル流量で定義します。

　入口$F_{A0}$, 出口 $F_B$として

$$
Y_B = \frac{F_B}{F_{A0}} \tag{6}
$$
　です。

　一般には
$$
Y_B = \frac{F_B-F_{B0}}{F_{A0}} \tag{6'}
$$
　と書きます。



<h3>2.3. PFRの収率</h3>

　PFRも流通系なので定義はCSTRと同じです。
$$
Y_B = \frac{F_B-F_{B0}}{F_{A0}}
$$
　です。

　PFRとCSTRの違いは収率の定義ではなく、同じ反応条件でも出口組成 $F_B$ が異なるため、結果として収率が変わる点です。

---




## 3. 総合選択率

<h3>副生成物(C)基準の総合選択率</h3>

回分
$$\bar S_{B/C}=\frac{N_B}{N_C} \tag{7}$$

または
流通系
$$\bar S_{B/C}=\frac{F_B}{F_C} \tag{8}$$

<h3>反応物(A)基準の総合選択率</h3>

回分
$$
\bar S_{B/A}=\frac{N_B-N_{B0}}{N_{A0}-N_A} \tag{9}
$$

流通系
$$
\bar S_{B/A}=\frac{F_B-F_{B0}}{F_{A0}-F_A} \tag{10}
$$
◇◇◇

<details><summary>📝🖱️説明文はこちら</summary>

ここでは総合選択率について説明します。

総合選択率とは、反応がある程度進行した後の結果をもとに、目的生成物をどれだけ選択的に作ることができたかを表す指標です。

まず、副生成物Cを基準とした総合選択率を考えます。

回分系では(7)式、流通系では(8)式で定義されます。

これは、目的生成物Bが副生成物Cの何倍生成したかを表しています。


次に、反応物Aを基準とした総合選択率を考えます。

回分系では(9)式、流通系では(10)式となります。

この式は、消費したAのうち、どれだけが目的生成物Bになったかを表しています。


副生成物基準の選択率は「BとCの競争」を評価する指標、

反応物基準の選択率は「消費したAの行き先」を評価する指標と考えると理解しやすいでしょう。

反応器設計では特に反応物基準の総合選択率がよく利用されます。

</details>

---

## 4. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C} \tag{9}$$

瞬間的な生成速度比を表す。

◇◇◇

<details><summary>📝🖱️説明文はこちら</summary>

次に、瞬間選択率について説明します。

先ほど学んだ総合選択率は、ある時点までに生成されたBとCの累積量の比でした。

これに対して瞬間選択率は、

「今この瞬間に、BとCのどちらが優先的に生成されているか」

を表す指標で、(9)式で定義されます。

ここで $r_B$ はBの生成速度、$r_C$ はCの生成速度です。

例えば、ある瞬間に

$$
r_B=2,\qquad r_C=1
$$
であれば、
$$
S_{B/C}=2
$$
となり、BがCの2倍の速さで生成されていることを意味します。

一方、
$$
S_{B/C}<1
$$
であれば、その瞬間にはCの方が優先的に生成されています。

つまり瞬間選択率は、反応速度そのものを比較する指標であり、濃度や温度の変化に応じて時間とともに変化します。

そのため、反応器の選定や運転条件の最適化を考える際には、総合選択率だけでなく瞬間選択率も重要になります。
</details>

---

## 5. 転化率、収率、選択率の関係

転化率
$$
X= \frac{F_{A0}-F_A}{F_{A0}}
$$
収率
$$
Y_B= \frac{F_B-F_{B0}}{F_{A0}}
$$
反応物基準の総合選択率
$$
\bar S_{B/A}=\frac{F_B-F_{B0}}{F_{A0}-F_A} \tag{10}
$$
より、

$$
Y_B = X\,\bar S_{B/A}
$$

が成り立ちます。



◇◇◇

転化率は「Aがどれだけ消えたか」、収率は「Bがどれだけ得られたか」を表します。Batchではモル数や濃度で、CSTRやPFRではモル流量で表しますが、本質的には『供給したAから目的物Bがどれだけ得られたか』という同じ概念です。

## 5. 並列反応

$$A
\rightarrow B$$
$$A
\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 6. 選択率と濃度

$$S_{B/C}=\frac{k_B}{k_C}C_A^{m-n}$$

- m>n : 高濃度有利
- m<n : 低濃度有利

## 7. Levenspiel流の反応器選定

高濃度有利 → PFR・回分反応器

低濃度有利 → CSTR

## 8. 例題1

$$r_B=3C_A^2$$
$$r_C=2C_A$$

$$S_{B/C}=\frac{3}{2}C_A$$

高濃度有利なのでPFRまたは回分反応器が有利。

## 9. 例題2

$$r_B=2C_A$$
$$r_C=5C_A^3$$

$$S_{B/C}=\frac{2}{5}C_A^{-2}$$

低濃度有利なのでCSTRが有利。

## 10. 逐次反応

$$A
\rightarrow B
\rightarrow C$$

目的生成物Bは中間生成物である。

## 11. B最大条件

$$\frac{dC_B}{dt}=0$$

$$k_1C_A=k_2C_B$$

この時Bは極大値を持つ。

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.1,2,200)
S=1.5*CA
plt.plot(CA,S)
plt.xlabel("CA")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
```
