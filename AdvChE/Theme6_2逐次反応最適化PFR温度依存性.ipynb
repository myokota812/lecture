- 濃度基準の設計方程式（定容系の場合）
$$\frac{dC_i}{d\tau} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

- $A \xrightarrow{k_1} B \xrightarrow{k_2} C$
の場合

$$\left\{ \begin{aligned}
\frac{dC_A}{d\tau} &= -k_1 C_A\\
\frac{dC_B}{d\tau} &= k_1 C_A - k_2 C_B\\
\frac{dC_C}{d\tau} &= k_2 C_B
\end{aligned} \right.$$

これより
$$\left\{ \begin{aligned}
C_A &= C_{A0} e^{-k_1 \tau}\\
C_B &= C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)\\
C_C &= C_{A0} \left( 1 - \frac{k_2 e^{-k_1 \tau} - k_1 e^{-k_2 \tau}}{k_2 - k_1} \right)
\end{aligned} \right.$$


(1) 転化率 $X_A$ :

$$X_A(\tau) = 1 - e^{-k_1 \tau} \tag{1}$$

(2) 収率 $Y_B(\tau)$

$$Y_B(\tau) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right) \tag{2}$$

(3) 選択率 $S_B(\tau)$

$$S_B(\tau) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 \tau} - e^{-k_2 \tau}}{1 - e^{-k_1 \tau}} \tag{3}$$

(4) $C_B$ が最大となる最適滞留時間（最適反応時間） $\tau_{\text{opt}}$

$$t_{\text{opt}} = \frac{1}{k_2 - k_1} \ln \left( \frac{k_2}{k_1} \right) \tag{4}$$

(5) 最大収率 $Y_{B,\max}$:

$$Y_{B,\max} = \left(\frac{k_1}{k_2}\right)^{\frac{k_2}{k_2 - k_1}} \tag{5}$$

（以上、復習)

### 2. 本題：反応速度定数の温度依存性（アレニウスの式）の導入

これまでの議論は、速度定数 $k_1, k_2$ の温度依存性を考慮しないで、定数として扱い、その大小関係でPFRの最適化を考えてきました。<br>
ここからは、$k_1, k_2$ にアレニウスの温度依存性の式を導入し、温度変化が転化率 $X_A$ および収率 $Y_B$ に与える影響を考えていきます。

---

### 2.1. アレニウス式による速度定数の表現

一次の不可逆逐次反応 $A \to B \to C$ の各段階の反応速度定数 $k_1, k_2$ は、絶対温度 $T \, [\mathrm{K}]$ の関数としてアレニウスの式（Arrhenius equation）で表されます。

$$k_1(T) = A_1 \exp\left( -\frac{E_1}{RT} \right) \tag{6}$$

$$k_2(T) = A_2 \exp\left( -\frac{E_2}{RT} \right) \tag{7}$$:

* $A_1, A_2$：頻度因子（前指数因子） $[\mathrm{s^{-1}}]$
* $E_1, E_2$：活性化エネルギー $[\mathrm{J/mol}]$
* $R$：気体定数 ($8.314 \, \mathrm{J/(mol \cdot K)}$)
* $T$：反応温度 $[\mathrm{K}]$

---

### 2.2. 温度 $T$ の関数としての転化率 $X_A$ と収率 $Y_B$

空間時間 $\tau$ を一定とした等温PFRにおいて、操作温度 $T$ を変化させたときの転化率 $X_A$ と収率 $Y_B$ は、式 (2) および式 (3) の $k_1, k_2$ に式 (6), (7) を代入することで記述されます。

#### (1) 転化率 $X_A(\tau, T)$

$$X_A(\tau, T) = 1 - \exp\left( -k_1(T)\tau \right) = 1 - \exp\left\{ -A_1 \tau \exp\left( -\frac{E_1}{RT} \right) \right\} \tag{8}$$

* **温度に対する挙動：**
活性化エネルギーは常に正 ($E_1 > 0$) であるため、温度 $T$ が上昇すると $k_1(T)$ は指数関数的に増大します。その結果、**空間時間 $\tau$ が一定であれば、温度 $T$ の上昇に伴い転化率 $X_A$ は単調に増加し $1$（100%）に漸近します。**

#### (2) 収率 $Y_B(\tau, T)$

$$Y_B(\tau, T) = \frac{k_1(T)}{k_2(T) - k_1(T)} \left( e^{-k_1(T)\tau} - e^{-k_2(T)\tau} \right) \tag{9}$$

もしくは

$$Y_B = \frac{A_1 e^{-E_1/RT}}{A_2 e^{-E_2/RT} - A_1 e^{-E_1/RT}} \left[ \exp\left(-A_1 e^{-E_1/RT} \tau\right) - \exp\left(-A_2 e^{-E_2/RT} \tau\right) \right]\tag{9*}$$

* **温度に対する挙動：**
収率 $Y_B$ の温度変化は、$k_1(T)$ と $k_2(T)$ のバランス、すなわち活性化エネルギーの大小関係（$E_1$ と $E_2$ の比較）によって劇的に変化します。

---


$k_1$ と $k_2$ にアレニウスの式（(4)式・(5)式）を代入すると、先ほど求めた最適滞留時間 $t_{\text{opt}}$ や最大収率 $Y_{B,\text{max}}$ が「温度 $T$ の関数」として完全に書き表されます。

この代入を行うことで、先ほど整理した「温度と収率の関係（Case I / Case II）」が数式としても明快に裏付けられます。

---

### 2.3. 最適滞留時間 $\tau_{\text{opt}}$ と最大収率 $Y_{B,\text{max}}$

温度をパラメーターとしますが、操作中は等温であるので、基本的に前回の式が使えます。(4)式に $k_1 = A_1 \exp\left(-\frac{E_1}{RT}\right)$ と $k_2 = A_2 \exp\left(-\frac{E_2}{RT}\right)$ を代入します。

* **対数の部分（分子）：**

$$\ln \left( \frac{k_2}{k_1} \right) = \ln \left( \frac{A_2}{A_1} \exp\left( \frac{E_1 - E_2}{RT} \right) \right) = \ln\left(\frac{A_2}{A_1}\right) + \frac{E_1 - E_2}{RT}$$


* **分母の部分：**

$$k_2 - k_1 = A_2 \exp\left(-\frac{E_2}{RT}\right) - A_1 \exp\left(-\frac{E_1}{RT}\right)$$



したがって、温度 $T$ を変数とした最適滞留時間の式は以下のようになります。

$$\tau_{\text{opt}}\ (T) = \frac{\ln\left(\frac{A_2}{A_1}\right) + \frac{E_1 - E_2}{RT}}{A_2 \exp\left(-\frac{E_2}{RT}\right) - A_1 \exp\left(-\frac{E_1}{RT}\right)} \tag{10}$$



さらに、この $t_{\text{opt}}$ を $C_B$ の式に代入して整理すると、最大収率 $Y_{B,\text{max}}$ は以下の形に表すことができます。

$$Y_{B,\text{max}} = \left( \frac{k_1}{k_2} \right)^{\frac{k_2}{k_2 - k_1}} = \left[ \frac{A_1}{A_2} \exp\left( -\frac{E_1 - E_2}{RT} \right) \right]^{\frac{k_2}{k_2 - k_1}} \tag{11}$$



(11)式をグラフにしてみました。

- 簡単のため$A_1=A_2$
- $E_1-E_2 = \Delta E$とおく

* **$\Delta E > 0$ の場合（Case I）:**
* $A_1 = A_2$ だといくら温度を上げても頭打ちになり、高収率を出すには $A_1 \gg A_2$ のような追加の条件が必要になります。


* **$\Delta E < 0$ の場合（Case II）:**
* $A_1 = A_2$ という対等な条件であっても、**「低温操作」をするだけで、特別な高 $A_1$ 因子なしに極めて高い収率を出すことができます。**

## 3. 反応熱を考慮した解析

反応熱を考慮した場合の解析法を紹介します。


### PFRの微小体積要素における「エンタルピー収支」のモデル図

```text
       [ 微小体積要素 dV （位置 z から z + dz まで） ]
       
                 流入エンタルピー             流出エンタルピー
              ┌─────────────────────┐
              │                     │
   ─────►│  [微小体積要素 dV]  │──────────►
  (位置 z)    │                     │  (位置 z+dz)
  Σ F_i H_i  │   ・反応熱の出入り  │   Σ F_i H_i + d(Σ F_i H_i)
              │   ・外部との熱交換  │
              └──────────┬──────────┘
                         │
                         ▼
               外部との熱のやり取り (Q)
               [ 冷却・加熱ジャケット等 ]

```

---

### 図の各要素の解説（収支式の内訳）

物質収支の図（モル流量 $F_i$ の変化）に、**熱エネルギーの移動**を加えたものがエンタルピー収支になります。

1. **流入エンタルピー（左側）: $\sum F_i H_i \Big\vert{}_z$**
* 微小要素の入り口（位置 $z$）へ、流体が持つすべての成分のエンタルピーが持ち込まれる流量です。


2. **流出エンタルピー（右側）: $\sum F_i H_i \Big\vert{}_{z+dz} = \sum F_i H_i + d\left(\sum F_i H_i\right)$**
* 微小要素の出口（位置 $z+dz$）から、反応を経た流体が持ち出すエンタルピーの流量です。


3. **内部の生成・消費（反応熱）: $\sum (-\Delta H_{Rj}) r_j dV$**
* 体積 $dV$ の中で化学反応（$A \to B \to C$）が起きたことにより、発生あるいは吸収される熱量です（発熱反応ならプラス、吸熱反応ならマイナス）。


4. **外部との熱交換（上下のやり取り）: $\pm Ua(T - T_a) dV$**
* 反応器の壁面を介して、外部のジャケットや冷却水・加熱源とやり取りする熱量です（断熱操作の場合はこの項が $0$ になります）。



---



### 符号・座標・位置に関する記号

* **$z$**: 反応器入口からの距離（長さ） $\mathrm{[m]}$
* **$dz$**: 反応器の微小長さ $\mathrm{[m]}$
* **$V$**: 反応器の体積（容積） $\mathrm{[m^3]}$
* **$dV$**: 微小体積要素 $\mathrm{[m^3]}$
* **$S$**: 反応器の断面積 $\mathrm{[m^2]}$ （$dV = S dz$）

### 物質・流速に関する記号

* **$i$**: 化学種のインデックス（今回の系では $A, B, C$） $[-]$
* **$F_i$**: 成分 $i$ のモル流量 $\mathrm{[mol/s]}$
* **$C_i$**: 成分 $i$ のモル濃度 $\mathrm{[mol/m^3]}$
* **$v$**: 体積流量 $\mathrm{[m^3/s]}$
* **$r_j$**: 反応 $j$ の反応速度（単位体積・単位時間あたりの反応量） $\mathrm{[mol/(m^3 \cdot s)]}$
* **$\nu_{i,j}$**: 化学反応 $j$ における成分 $i$ の化学量論係数 $[-]$（生成物は正、反応物は負）

### 熱力学・エネルギーに関する記号

* **$H_i$**: 成分 $i$ のモルエンタルピー $\mathrm{[J/mol]}$
* **$C_{p,i}$**: 成分 $i$ の定圧モル比熱 $\mathrm{[J/(mol \cdot K)]}$
* **$T$**: 流体の温度 $\mathrm{[K]}$ （または $\mathrm{^\circ C}$）
* **$T_a$**: 外部（ジャケット等）の温度 $\mathrm{[K]}$ （または $\mathrm{^\circ C}$）
* **$\Delta H_{Rj}$**: 反応 $j$ の標準反応エンタルピー（反応熱） $\mathrm{[J/mol]}$
* **$U$**: 総括伝熱係数 $\mathrm{[W/(m^2 \cdot K)]}$ （$\mathrm{J/(s \cdot m^2 \cdot K)}$）
* **$a$**: 単位体積あたりの伝熱面積（比表面積） $\mathrm{[m^2/m^3]}$ （$= \mathrm{m^{-1}}$）
* **$\tau$**: 滞留時間（空間時間、$\tau = V / v$） $\mathrm{[s]}$

---



### ステップ1：エネルギー収支の出発式（式 3.1 の形）の立式



$$\left[ \sum_{i} F_i H_i \right]_z - \left[ \sum_{i} F_i H_i \right]_{z+dz} + \left( \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j \right) dV - U a (T - T_a) dV = 0 \tag{3.1}$$

* 第1項・第2項：位置 $z$ と $z+dz$ における顕熱の出入り【注１】
* 第3項：微小体積 $dV$ 内で発生・吸収される反応熱（発熱ならプラス）
* 第4項：外部の冷却・加熱ジャケット等との熱のやり取り

>【注１】
>各成分 $i$ のモルエンタルピー $H_i$ を、基準温度 $T_{\text{ref}}$ からの**顕熱（比熱による温度変化分）のみ**として定義します。
>
>$$H_i(T) = \int_{T_{\text{ref}}}^{T} C_{p,i} dT$$
>
>（※ここでは生成エンタルピー $\Delta H_{f,i}^\circ$ は含めません）

---
### ステップ2：左辺の変換

位置 $z+dz$ でのエンタルピー流束を、位置 $z$ の周りでテイラー展開の１次項までに近似します。

$$\left[ \sum F_i H_i \right]_{z+dz} \approx \left[ \sum F_i H_i \right]_z + \frac{d}{dz}\left( \sum F_i H_i \right) dz$$

すると(3.1)式は

$$ - \frac{d}{dz}\left( \sum F_i H_i \right) dz + \left( \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j \right) dV - U a (T - T_a) dV = 0 \tag{3.1}$$

従って

$$ - d\left( \sum F_i H_i \right)  + \left( \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j \right) dV - U a (T - T_a) dV = 0 \tag{3.2}$$

### ステップ3：微分形式への変換

(3.2)式の両辺を $dV$ で割ると、微分のかたちが $\frac{d}{dV}$（反応器容積 $V$ に関する微分）になります。

$$- \frac{d\left( \sum F_i H_i \right)}{dV} + \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) = 0$$

これを移項して整理すると、「単位体積あたりのエンタルピー変化率」を表す微分方程式になります。

$$\frac{d}{dV} \left( \sum_{i} F_i H_i \right) = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) \tag{3.3}$$

---

#### $dV = S dz$ （長さ座標 $z$ と断面積 $S$）に置き換える場合

反応器の長さ方向の位置 $z$ を基準にしたい場合は、$dV = S dz$ を代入します。

$$- \frac{1}{S} \frac{d\left( \sum F_i H_i \right)}{dz} + \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) = 0$$

断面積 $S=1$の場合
$$\frac{d}{dz} \left( \sum_{i} F_i H_i \right) = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) \tag{3.4}$$

---

### ステップ4：左辺（顕熱項）の微分展開

積の微分法則（$\frac{d(AB)}{dz} = A\frac{dB}{dz} + B\frac{dA}{dz}$）を使って、左辺を展開します。

$$\sum_{i} \left( F_i \frac{dH_i}{dz} + H_i \frac{dF_i}{dz} \right) = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a)$$

ここで、

1. **第1項の $\frac{dH_i}{dz}$：** 顕熱の定義から $\frac{dH_i}{dz} = C_{p,i} \frac{dT}{dz}$ となります。
2. **第2項の $H_i \frac{dF_i}{dz}$：** 流儀２では $H_i$ が「顕熱のみ」であるため、この第2項は「温度が異なる物質が生成・消滅することに伴う顕熱の補正項」を意味します。しかし、通常の化学反応においてこの第2項は、生成エンタルピーを基準にしていないことに伴う微小な項（あるいは発熱・吸熱そのものとは別の物理量）となり、実用上は非常に小さいため無視されるか、あるいは反応熱の定義の中に綺麗に吸収されます。

最終的に、顕熱の変化（温度変化）が主たる項として残り、次のように整理されます。

$$\sum_{i} F_i C_{p,i} \frac{dT}{dz} = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) \tag{3.5}$$

---

### ステップ5：滞留時間 $\tau$ 基準への変換と最終形

モル流量 $F_i$ を濃度 $C_i$ と体積流量 $v$ の積（$F_i = C_i v$）に置き換え、両辺を $v$ で割ることで、微小長さ $dz$ から滞留時間の微小変化 $d\tau$（$d\tau = dV / v = dz / v$）に変換すると、本講義の解析に使える形になります。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a) \tag{3.6}$$

####連続反応 $A \xrightarrow{k_1} B \xrightarrow{k_2} C$ （$r_1 = k_1 C_A, \ r_2 = k_2 C_B$）の場合

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) k_1(T) C_A + (-\Delta H_{R2}) k_2(T) C_B - U a (T - T_a)$$

左辺はもちろん
$$\sum_{i \in \{A, B, C\}} C_i C_{p,i} \frac{dT}{d\tau} = \left( C_A C_{p,A} + C_B C_{p,B} + C_C C_{p,C} \right) \frac{dT}{d\tau}$$
です。

---



最終的に、PFRにおける逐次反応の解析は以下の連立微分方程式を解くことになります。

$$\begin{cases} \displaystyle\frac{dC_A}{d\tau} = -k_1 C_A \\ \displaystyle\frac{dC_B}{d\tau} = k_1 C_A - k_2 C_B \\ \displaystyle\frac{dC_C}{d\tau} = k_2 C_B \\ \displaystyle\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) k_1 C_A + (-\Delta H_{R2}) k_2 C_B - U a (T - T_a) \end{cases} \tag{3.7}$$

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# --- 1. 物性値・反応パラメータの設定（反応が起きやすい値に調整） ---
# --- 劇的な変化を見るためのパラメータ設定例 ---
A1, E1 = 1.0e4, 50000.0  # 反応1: A -> B
A2, E2 = 2.0e4, 55000.0  # 反応2: B -> C (高温になると副反応が一気に進む設定)
R_gas = 8.314

Cp_A, Cp_B, Cp_C = 150.0, 150.0, 150.0

Delta_H_R1 = -80000.0  # 発熱量を大きく設定 (-80 kJ/mol)
Delta_H_R2 = -50000.0  # (-50 kJ/mol)

Ua = 200.0  # 冷却を弱めてホットスポットが出やすくする
T_a = 400.0

C_A0 = 10.0  # 初期濃度を5倍に増加
C_B0, C_C0 = 0.0, 0.0
T0 = 400.0

tau_max = 10.0


# --- 2. 速度定数 (アレニウスの式) の定義 ---
def get_k1(T):
  return A1 * np.exp(-E1 / (R_gas * T))


def get_k2(T):
  return A2 * np.exp(-E2 / (R_gas * T))


# --- 3. 連立常微分方程式（ODE）の定義 ---
def pfr_nonisothermal_ode(tau, y):
  CA, CB, CC, T = y

  k1 = get_k1(T)
  k2 = get_k2(T)

  # 物質収支
  dCA_dtau = -k1 * CA
  dCB_dtau = k1 * CA - k2 * CB
  dCC_dtau = k2 * CB

  # 系の熱容量 (Σ Ci Cpi)
  sum_Cp = CA * Cp_A + CB * Cp_B + CC * Cp_C
  sum_Cp = max(sum_Cp, 1.0)  # ゼロ割防止

  # 熱の生成・消費
  heat_reaction1 = (-Delta_H_R1) * k1 * CA
  heat_reaction2 = (-Delta_H_R2) * k2 * CB
  heat_exchange = -Ua * (T - T_a)

  # エネルギー収支（dT/dτ）
  dT_dtau = (heat_reaction1 + heat_reaction2 + heat_exchange) / sum_Cp

  return [dCA_dtau, dCB_dtau, dCC_dtau, dT_dtau]


# --- 4. 数値積分（solve_ivp）の実行 ---
y0 = [C_A0, C_B0, C_C0, T0]
tau_span = (0, tau_max)
tau_eval = np.linspace(0, tau_max, 500)

sol = solve_ivp(
    pfr_nonisothermal_ode, tau_span, y0, t_eval=tau_eval, method="Radau"
)

# --- 5. グラフ描画 ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# 濃度プロファイル
ax1.plot(sol.t, sol.y[0], label="C_A (Reactant)", color="blue", linewidth=2)
ax1.plot(
    sol.t, sol.y[1], label="C_B (Desired Product)", color="green", linewidth=2
)
ax1.plot(sol.t, sol.y[2], label="C_C (Byproduct)", color="red", linewidth=2)
ax1.set_xlabel("Residence Time $\\tau$", fontsize=12)
ax1.set_ylabel("Concentration [mol/m³]", fontsize=12)
ax1.set_title("Concentration Profiles along PFR", fontsize=14)
ax1.grid(True, linestyle="--", alpha=0.6)
ax1.legend()

# 温度プロファイル
ax2.plot(
    sol.t, sol.y[3], label="Reactor Temp $T$", color="purple", linewidth=2
)
ax2.axhline(
    y=T_a, color="gray", linestyle="--", label=f"Jacket Temp $T_a$ ({T_a} K)"
)
ax2.set_xlabel("Residence Time $\\tau$", fontsize=12)
ax2.set_ylabel("Temperature [K]", fontsize=12)
ax2.set_title("Temperature Profile along PFR", fontsize=14)
ax2.grid(True, linestyle="--", alpha=0.6)
ax2.legend()

plt.tight_layout()
plt.show()
````

「流儀１」（エンタルピー $H_i$ に**基準状態の生成エンタルピー（化学エネルギー）まですべて含める**アプローチ）を用いた、エネルギー収支式の厳密な導出プロセスです。

この流儀では、出発式の段階で外から「反応熱」という項を足す必要は一切ありません。流体の出入りを表す微分の計算から、**自動的に反応熱が導き出される**のが美しいポイントです。

---

### ステップ1：エンタルピーの定義と出発式

#### ① モルエンタルピー $H_i$ の定義

各成分 $i$ のモルエンタルピーを、基準温度 $T_{\text{ref}}$ における**生成エンタルピー $\Delta H_{f,i}^\circ$** と、温度 $T$ までの**顕熱**の和として定義します。


$$H_i(T) = \Delta H_{f,i}^\circ + \int_{T_{\text{ref}}}^{T} C_{p,i} dT$$

#### ② PFRのエネルギー収支の一般式（出発式）

微小体積 $dV = S dz$ を持つ PFR の定常状態において、全エンタルピー流束の変化が外部熱交換（冷却・加熱）と等しいという式を立てます。


$$\frac{d}{dz} \left( \sum_{i} F_i H_i \right) = - U a (T - T_a)$$


（※断熱の場合は右辺が $0$ になりますが、ここでは一般的な熱交換ありで進めます）

---

### ステップ2：積の微分則による展開

左辺の微分に、積の微分法則（$\frac{d(AB)}{dz} = A\frac{dB}{dz} + B\frac{dA}{dz}$）を適用します。

$$\sum_{i} \left( F_i \frac{dH_i}{dz} + H_i \frac{dF_i}{dz} \right) = - U a (T - T_a) \tag{A}$$

ここで、

* **第1項 ($F_i \frac{dH_i}{dz}$)：** 温度変化に伴う熱の出入り（**顕熱変化**）
* **第2項 ($H_i \frac{dF_i}{dz}$)：** 化学反応による物質の増減に伴うエネルギーの出入り（**化学エネルギー変換 ＝ 反応熱の正体**）

です。ユーザー様が先ほど気付かれた通り、**この第2項こそが鍵**になります。

---

### ステップ3：各項の変形

#### ① 第1項（顕熱項）の変形

定義された $H_i$ を $z$ で微分すると、ライプニッツの微分則により生成エンタルピー（定数）は消え、比熱だけが残ります。


$$\frac{dH_i}{dz} = C_{p,i} \frac{dT}{dz}$$


よって、第1項は $\sum_{i} F_i C_{p,i} \frac{dT}{dz}$ となります。

#### ② 第2項（化学エネルギー・反応熱項）の変形

物質のモル流量の変化率 $\frac{dF_i}{dz}$ は、微小体積 $dV = S dz$ あたりの各素反応の速度 $r_j$ と化学量論係数 $\nu_{i,j}$ を用いて表されます。


$$\frac{dF_i}{dz} = S \sum_{j=1}^{M} \nu_{i,j} r_j = v \sum_{j=1}^{M} \nu_{i,j} r_j$$


（※ $S$ は断面積、$v$ は体積流量、$F_i = C_i v$）

これを第2項の $H_i \frac{dF_i}{dz}$ に代入します。


$$\sum_{i} H_i \left( v \sum_{j=1}^{M} \nu_{i,j} r_j \right) = v \sum_{j=1}^{M} r_j \left( \sum_{i} \nu_{i,j} H_i \right)$$

ここで、括弧の中身 $\sum_{i} \nu_{i,j} H_i$ こそが、**反応 $j$ における反応エンタルピー（反応熱） $\Delta H_{Rj}$** そのものです！

* $\Delta H_{Rj} = \sum_{i} \nu_{i,j} H_i$ （生成物のエンタルピー総和 － 反応物のエンタルピー総和）

したがって、第2項は次のように書き換えられます。


$$v \sum_{j=1}^{M} \Delta H_{Rj} r_j$$

---

### ステップ4：式をまとめて滞留時間 $\tau$ 基準に変換する

変形した第1項と第2項を式 (A) に戻します。

$$\sum_{i} F_i C_{p,i} \frac{dT}{dz} + v \sum_{j=1}^{M} \Delta H_{Rj} r_j = - U a (T - T_a)$$

ここで、

1. $F_i = C_i v$ を代入する
2. 両辺を体積流量 $v$ で割り、 $dz / v = d\tau$ （滞留時間の微小変化）の関係を用いて整理する

すると、次式が得られます。


$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} + \sum_{j=1}^{M} \Delta H_{Rj} r_j = - U a (T - T_a)$$

反応熱の項を右辺に移行すると：


$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = \sum_{j=1}^{M} (-\Delta H_{Rj}) r_j - U a (T - T_a)$$

---

### ステップ5：連続反応 $A \to B \to C$ への適用

ご提示の 2 段階の連続反応（$r_1 = k_1 C_A, \ r_2 = k_2 C_B$）を右辺の総和に代入すれば、私たちが普段使うおなじみの形が完全に導かれます。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) k_1(T) C_A + (-\Delta H_{R2}) k_2(T) C_B - U a (T - T_a)$$

---

### まとめ

* **流儀１**では、出発式の段階で「反応熱」を書き込む必要はありません。
* 全エンタルピー $H_i$ の微分を展開したときに出てくる **$H_i \frac{dF_i}{dz}$（第2項）が、まさに化学反応によるエネルギー変化（反応熱）そのもの**だからです。
* 数学的に変形していくと、この第2項が自然と $\sum (-\Delta H_{Rj}) r_j$ に化けて右辺に移動するため、結果として「流儀２（最初から反応熱を置く方法）」と**完全に全く同じ方程式**になります。

ユーザー様が直感された通り、左辺の展開のなかにすべての答えが内包されているという、非常に一貫性の高い美しい導出方法です！



### これらを数式（定常状態のエネルギー収支）にまとめると：

$$\left[ (\sum F_i H_i)_z \right] - \left[ (\sum F_i H_i)_{z+dz} \right] + (\text{反応熱}) + (\text{外部熱交換}) = 0 \tag{3.1}$$

これを微小長さ $dz$（または微小体積 $dV$）あたりで微分形式に整理していきます。




式を少し変形して、流出から流入を引く形にすると：


$$\left[ (\sum F_i H_i)_{z+dz} \right] - \left[ (\sum F_i H_i)_z \right] = (\text{反応熱}) + (\text{外部熱交換})$$


位置 $z+dz$ でのエンタルピー流束を、位置 $z$ の周りでテイラー展開の１次項までに近似します。により近似します。

$$\left[ \sum F_i H_i \right]_{z+dz} \approx \left[ \sum F_i H_i \right]_z + \frac{d}{dz}\left( \sum F_i H_i \right) dz$$

これを元の式に戻して、整理すると

$$-\frac{d}{dz}\left( \sum F_i H_i \right) dz + (\text{反応熱}) + (\text{外部熱交換}) = 0 \tag{3.2} $$



---

### ステップ1：左辺を展開する


(3.2)式の左辺において、モル流量 $F_i$ と、温度 $T$ の関数であるモルエンタルピー $H_i(T)$ の積の微分に、積の微分法則（$\frac{d(AB)}{dz} = A\frac{dB}{dz} + B\frac{dA}{dz}$）を適用します。

$$\sum_{i} \left( F_i \frac{dH_i}{dz} + H_i \frac{dF_i}{dz} \right) = 0 \tag{3.3}$$

---

### ステップ2：顕熱項（温度変化の項）の変形

モルエンタルピー $H_i(T)$ は、基準温度 $T_{\text{ref}}$ からの顕熱（比熱による温度変化）を用いて次のように表されます。


$$H_i(T) = H_{i}^{\circ}(T_{\text{ref}}) + \int_{T_{\text{ref}}}^{T} C_{p,i} dT$$

これを $z$ で微分すると、ライプニッツの微分則によりエンタルピーの温度微分は定圧モル比熱 $C_{p,i}$ になります。


$$\frac{dH_i}{dz} = C_{p,i} \frac{dT}{dz}$$

これを(3.3)式の第1項に代入します。


$$\sum_{i} \left( F_i C_{p,i} \frac{dT}{dz} + H_i \frac{dF_i}{dz} \right) = 0$$

---

### ステップ3：モル流量の変化（物質収支）と「反応熱」の結びつき

左辺第2項にある $\frac{dF_i}{dz}$ は、位置 $z$ における成分 $i$ のモル流量の変化率であり、化学反応による生成・消費速度（モル基準の反応速度 $R_{i} = \sum_{j} \nu_{i,j} r_j$）に相当します。

微小体積 $dV = S dz$ （$S$ は断面積）を用いて書き直すと、反応工学の定義から：


$$\frac{dF_i}{dV} = \sum_{j} \nu_{i,j} r_j$$


となります（ここで $\nu_{i,j}$ は化学反応 $j$ における成分 $i$ の化学量論係数、$r_j$ は反応速度）。

これを使って第2項を書き換えると、


$$\sum_{i} H_i \frac{dF_i}{dV} = \sum_{i} H_i \left( \sum_{j} \nu_{i,j} r_j \right)$$

和の順序を入れ替えると、括弧の中身は「各反応の反応エンタルピー（反応熱 $-\Delta H_{Rj}$）」そのものになります。


$$\sum_{j} r_j \left( \sum_{i} \nu_{i,j} H_i \right) = \sum_{j} (-\Delta H_{Rj}) r_j$$


（※ 発熱反応の場合、生成物のエンタルピー総和が反応物より低くなるため、$-\Delta H_R > 0$（発熱）となります）

---

### ステップ4：体積座標から滞留時間 $\tau$ への変換

ここまでをまとめると、微小体積 $dV$ あたりのエネルギー収支は次のようになります。

$$\sum_{i} F_i C_{p,i} \frac{dT}{dV} = \sum_{j} (-\Delta H_{Rj}) r_j$$

ここで、

1. モル流量 $F_i$ は、濃度 $C_i$ と体積流量 $v$ の積（$F_i = C_i v$）
2. 滞留時間 $\tau$ は、体積 $V$ と体積流量 $v$ の関係から $d\tau = \frac{dV}{v}$ （または $V = v\tau$ より $\frac{dV}{d\tau} = v$）

これらを代入して両辺を体積流量 $v$ で割ると、左辺の $F_i$ が $C_i$ に変わります。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = \sum_{j} (-\Delta H_{Rj}) r_j$$

---

### ステップ5：ご提示の連続反応 $A \to B \to C$ への適用

今回の系では、反応が 2 つあります。

1. 反応1 ($A \to B$)：速度 $r_1 = k_1(T) C_A$、反応熱 $(-\Delta H_{R1})$
2. 反応2 ($B \to C$)：速度 $r_2 = k_2(T) C_B$、反応熱 $(-\Delta H_{R2})$

これを右辺の総和 $\sum_{j} (-\Delta H_{Rj}) r_j$ に具体的に当てはめると、求める式が完成します。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) k_1(T) C_A + (-\Delta H_{R2}) k_2(T) C_B$$

このように、「顕熱の変化（左辺）」＝「各素反応で発生・吸収される熱量の総和（右辺）」というエネルギーの保存則から、この微分方程式が綺麗に導かれます。




---

### 3.1. 前提条件と基礎方程式

* 反応式：$\text{A} \to \text{B} \to \text{C}$
* 反応器：定常状態のPFR（プラグフロー反応器）
* 座標：反応器入口からの距離 $z$、または滞留時間 $\tau$（あるいは位置 $z$ における空間時間）
* 熱効果：各素反応の反応熱をそれぞれ $\Delta H_{R1}$、$\Delta H_{R2}$ とする。

PFRの微小体積要素 $dV$（微小時間 $d\tau$ に相当）に関する定常状態のエネルギー収支（エンタルピー収支）の一般式は、単位時間あたりに流入・流出するエンタルピーの差と、反応熱の発生・吸収のバランスから以下のように表されます。

$$\frac{d}{dz} \left( \sum_{i} F_i H_i \text{ あるいは全エンタルピー流束} \right) = \text{発熱・吸熱項}$$

これをモル流量 $F_i$ とモル比熱 $C_{p,i}$、および反応速度 $r_j$ を用いて展開します。モル流量 $F_i$ は濃度 $C_i$ と体積流量 $v$ の積（$F_i = C_i v$）ですが、ここでは単位体積あたりの式として整理します。

---

### 2. 反応速度の定義

ご提示の連続反応：

1. 反応1: $A \to B$ （速度 $r_1 = k_1 C_A$）
2. 反応2: $B \to C$ （速度 $r_2 = k_2 C_B$）

各成分の生成速度 $\frac{dC_i}{d\tau}$ は以下のように表されます。


$$\frac{dC_A}{d\tau} = -r_1$$

$$\frac{dC_B}{d\tau} = r_1 - r_2$$

$$\frac{dC_C}{d\tau} = r_2$$

---

### 3. エンタルピー収支式の立式

微小セクションにおける温度 $T$ の変化を追うため、全系の熱容量と反応熱の関係を考えます。
反応器内の流体の見かけの熱容量を $\sum (C_i C_{p,i})$ （または単位体積あたりの熱容量 $\rho C_p$ あるいは $\sum F_i C_{p,i}$）とし、外部との熱交換（伝熱係数 $U$、伝熱面積 $A_{\text{heat}}$、外温 $T_a$）を考慮すると、エネルギー収支式は一般に次式で表されます。

$$\sum_{i} F_i C_{p,i} \frac{dT}{dz} = (- \Delta H_{R1}) r_1 v + (- \Delta H_{R2}) r_2 v \pm (\text{熱交換項})$$

これを滞留時間 $\tau$ （または容積座標）を基準にした微分方程式の形に書き換えます。

#### ① 断熱操作の場合（熱ロス・外部冷却がない場合）

外部との熱の出入りがない場合、反応熱の出入りがそのまま流体の温度変化（顕熱の変化）に使われます。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) r_1 + (-\Delta H_{R2}) r_2$$

ここで、

* $\Delta H_{R1}$ : 反応1 ($A \to B$) の標準反応エンタルピー（発熱反応なら負）
* $\Delta H_{R2}$ : 反応2 ($B \to C$) の標準反応エンタルピー
* $C_{p,i}$ : 成分 $i$ の定圧モル比熱
* $r_1 = k_1(T) C_A$
* $r_2 = k_2(T) C_B$

#### ② 伝熱を伴う場合（ジャケット冷却など）

反応器に冷却・加熱ジャケットがついており、熱伝達係数 $U$、伝熱面積 $a$（単位体積あたりの伝熱面積）、外壁温度 $T_a$ がある場合は、右辺に熱交換の項が加わります。

$$\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) r_1 + (-\Delta H_{R2}) r_2 - U a (T - T_a)$$

---

### まとめ

$A \to B \to C$ の連続反応における PFR のエンタルピー収支式（断熱系）は、成分の物質収支と連立させることで以下のように記述されます。

$$\begin{cases} \displaystyle\frac{dC_A}{d\tau} = -k_1(T) C_A \\ \displaystyle\frac{dC_B}{d\tau} = k_1(T) C_A - k_2(T) C_B \\ \displaystyle\frac{dC_C}{d\tau} = k_2(T) C_B \\ \displaystyle\sum_{i} C_i C_{p,i} \frac{dT}{d\tau} = (-\Delta H_{R1}) k_1(T) C_A + (-\Delta H_{R2}) k_2(T) C_B \end{cases}$$

### 3.3. 活性化エネルギーの大小関係による選択率・収率の温度依存性

逐次反応において目的物質 $B$ を効率よく得るには、主反応（$A \to B$）を速めつつ、逐次的な副反応（$B \to C$）を抑える必要があります。両反応の速度比 $\frac{k_1(T)}{k_2(T)}$ をとると、温度依存性が明確になります。

$$\frac{k_1(T)}{k_2(T)} = \frac{A_1}{A_2} \exp\left( -\frac{E_1 - E_2}{RT} \right) \tag{8}$$

この式 (8) から、$E_1$ と $E_2$ の大小関係によって以下の2つのケースに分類されます。

| ケース | 活性化エネルギーの関係 | 高温にしたときの変化 | 収率 $Y_B$ を高めるための温度戦略 |
| --- | --- | --- | --- |
| **Case I** | **$E_1 > E_2$**<br>（主反応の方が温度に敏感） | $k_1$ の増加率が $k_2$ を上回り、<br>速度比 $k_1/k_2$ が**増大**する。 | **高温操作が有利**<br>高温にするほど選択率・最大収率ともに向上する。 |
| **Case II** | **$E_1 < E_2$**<br>（副反応の方が温度に敏感） | $k_2$ の増加率が $k_1$ を上回り、<br>速度比 $k_1/k_2$ が**減少**する。 | **低温操作（または最適温度の選定）が有利**<br>高温にしすぎると $B \to C$ の分解が急増し収率が低下する。 |

## 4. 最適化の基準と温度・空間時間のトレードオフ

実際のPFRの設計および運転における最適化では、「空間時間 $\tau$（反応器サイズ・流量）」**と**「反応温度 $T$」の2つを組み合わせて目的関数を最大化します。

### 4.1. 目的物質 $B$ の収率 $Y_B$ の最大化

等温操作のもとで収率 $Y_B$ を最大にする**最適空間時間 $\tau_{\max}(T)$** と、そのときの**最大収率 $Y_{B,\max}(T)$** は、温度 $T$ の関数として次のように表されます。

$$\tau_{\max}(T) = \frac{1}{k_2(T) - k_1(T)} \ln\left(\frac{k_2(T)}{k_1(T)}\right) \tag{12}$$

$$Y_{B,\max}(T) = \left(\frac{k_1(T)}{k_2(T)}\right)^{\frac{k_2(T)}{k_2(T) - k_1(T)}} \tag{13}$$

#### (1) 空間時間 $\tau$ を固定したまま温度 $T$ を変化させる場合（既存プラントの運転最適化）

すでに反応器の体積 $V$ と流量 $v_0$（すなわち $\tau$）が決まっている場合、温度 $T$ を変化させると以下の現象が起こります。

* **低温側：** $k_1(T)$ が小さいため反応が十分に進行せず（$X_A$ が低い）、$Y_B$ は小さい。
* **最適温度 $T_{\text{opt}}$：** 与えられた空間時間 $\tau$ が、ちょうどその温度における最適空間時間 $\tau_{\max}(T_{\text{opt}})$ と一致する温度で、**収率 $Y_B$ は極大値**をとります。
* **高温側：** 反応速度が速くなりすぎて滞留時間 $\tau$ の間に $B \to C$ の過剰反応が進行し、転化率 $X_A$ は $100\%$ に近づく一方で、収率 $Y_B$ は急激に低下します（オーバーリアクション）。

> **重要ポイント：**
> 活性化エネルギーの大小（$E_1 > E_2$ か $E_1 < E_2$ か）にかかわらず、**「空間時間 $\tau$ が一定」という条件下では、収率 $Y_B(T)$ は必ずある最適温度 $T_{\text{opt}}$ でピーク（極大値）を持ちます。**

#### (2) 温度 $T$ と空間時間 $\tau$ を同時に設計する場合（新規反応器の設計最適化）

* **$E_1 > E_2$ の場合：**
温度 $T$ を高くするほど理論上の最大収率 $Y_{B,\max}(T)$ が高くなり、同時に必要な最適空間時間 $\tau_{\max}(T)$ も短くなる（＝反応器を小型化できる）ため、**材質や安全性が許す限り高温・短滞留時間**での運転が最適となります。
* **$E_1 < E_2$ の場合：**
温度 $T$ を低くするほど最大収率 $Y_{B,\max}(T)$ は向上しますが、反応速度が低下するため必要な空間時間 $\tau_{\max}(T)$ は長大化（＝反応器が巨大化・コスト増）します。そのため、**「収率（原料コスト）」と「反応器体積（設備コスト）」の経済的トレードオフ**から最適温度 $T_{\text{opt}}$ を決定する必要があります。

---

### 4.2. 温度変化に伴う $X_A$ と $Y_B$ の挙動まとめ（視覚的理解のための概念図）

空間時間 $\tau$ を一定として横軸に温度 $T$ をとった場合、$X_A$ と $Y_B$ の曲線は以下の特徴を示します。

1. **転化率 $X_A(T)$：** S字カーブ（シグモイド状）を描きながら $0$ から $1$ へ単調増加します。
2. **収率 $Y_B(T)$：** 常に $Y_B(T) \le X_A(T)$ の領域内にあり、低温域では $X_A(T)$ に沿って立ち上がりますが、ある温度 $T_{\text{opt}}$ で極大を迎えた後、高温域では $0$ に向かって減少する**山なりの曲線**を描きます。

---

## 1.3. 具体例：逐次反応 $A \to B \to C$ への適用確認

これまでの一般式が、PFRにおける逐次反応にどのように当てはまるかを確認します。

* 化学種 ($N=3$): $A$ ($i=1$), $B$ ($i=2$), $C$ ($i=3$)
* 反応 ($M=2$):
* 反応1 ($j=1$): $A \to B \quad (r_1 = k_1 C_A)$
* 反応2 ($j=2$): $B \to C \quad (r_2 = k_2 C_B)$



化学量論係数 $\nu_{i,j}$ は以下の通りです。

* $A$ について： $\nu_{1,1} = -1, \ \nu_{1,2} = 0$
* $B$ について： $\nu_{2,1} = +1, \ \nu_{2,2} = -1$
* $C$ について： $\nu_{3,1} = 0, \ \nu_{3,2} = +1$

定容系におけるPFRの一般式 $\frac{dC_i}{d\tau} = \sum_{j=1}^{M} \nu_{i,j} r_j$ に代入すると、空間時間 $\tau$ に関する以下の3元連立微分方程式が完全に導出されます。

1. **成分 $A$:** $\frac{dC_A}{d\tau} = (-1)r_1 + (0)r_2 = -k_1 C_A$
2. **成分 $B$:** $\frac{dC_B}{d\tau} = (+1)r_1 + (-1)r_2 = k_1 C_A - k_2 C_B$
3. **成分 $C$:** $\frac{dC_C}{d\tau} = (0)r_1 + (+1)r_2 = k_2 C_B$

## 2. 転化率、収率、選択率

本講義では、次のような一次の不可逆逐次反応を、連続式反応器の一つである管型反応器（Plug Flow Reactor: PFR）で実施する系を考えます。

$$A \xrightarrow{k_1} B \xrightarrow{k_2} C$$

* $A$：原料
* $B$：目的とする中間生成物
* $C$：副生物（過剰反応による分解物など）
* $k_1, k_2$：それぞれの反応速度定数

反応器入口における $A$ の濃度を $C_{A0}$ とし、体積変化は無視できる（定容系）と仮定します。また、反応器入口（空間時間 $\tau = 0$）において $B$ および $C$ は存在しないものとします。

---

## 2.1. 諸定義の確認

反応の進行度や目的物質の生産効率を評価するため、以下の3つの指標を定義します（回分反応器の時間 $t$ が、PFRでは**空間時間 $\tau = \frac{V}{v_0}$** に置き換わります）。

### (1) 転化率 (Conversion) $X_A$

反応した原料 $A$ の割合を示します。

$$X_A(\tau) = \frac{C_{A0} - C_A(\tau)}{C_{A0}} = 1 - \frac{C_A(\tau)}{C_{A0}} \tag{eqX}$$

### (2) 収率 (Yield) $Y_B$

供給した原料 $A$ に対し、目的物質 $B$ がどれだけ得られたかの割合です。

$$Y_B(\tau) = \frac{\text{生成した } B \text{ のモル速度}}{\text{供給した } A \text{ の初期モル速度}} = \frac{C_B(\tau)}{C_{A0}} \tag{eqY}$$

### (3) 選択率 (Selectivity) $S_B$

反応した原料 $A$ のうち、どれだけが目的物質 $B$ に変化したかの割合です。

$$S_B(\tau) = \frac{\text{生成した } B \text{ のモル速度}}{\text{反応した } A \text{ のモル速度}} = \frac{C_B(\tau)}{C_{A0} - C_A(\tau)} = \frac{Y_B(\tau)}{X_A(\tau)} \tag{eqS}$$

---

## 2.2. 定式化

PFRにおける各成分の物質収支式（空間時間 $\tau$ に関する微分方程式）は以下の通りです。

$$\frac{dC_A}{d\tau} = -k_1 C_A$$

$$\frac{dC_B}{d\tau} = k_1 C_A - k_2 C_B$$

$$\frac{dC_C}{d\tau} = k_2 C_B$$

これらを解くことで、反応管の長さ（空間時間 $\tau$）に沿った各変数の濃度変化を明らかにできます。

* **原料 $A$:**

$$C_A = C_{A0} e^{-k_1 \tau}\tag{1}$$

* **中間生成物 $B$ ($k_1 \neq k_2$ の場合):**

$$C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)\tag{2}$$

* **副生物 $C$:**

$$C_C = C_{A0} \left( 1 - \frac{k_2 e^{-k_1 \tau} - k_1 e^{-k_2 \tau}}{k_2 - k_1} \right)\tag{3}$$

---

### (1) 転化率 $X_A$ の空間時間変化

原料 $A$ の濃度は(1)式より、転化率は次のように表されます。

$$X_A(\tau) = 1 - e^{-k_1 \tau} \tag{4}$$

* **特徴:** 転化率の挙動は第2段階の反応速度定数 $k_2$ に依存せず、$k_1$ のみで決定されます。空間時間 $\tau$ が大きくなる（反応管が長くなる、または流速が遅くなる）につれて $X_A \to 1$（100%）に漸近します。

### (2) 収率 $Y_B(\tau)$ の空間時間変化の導出

#### 1. 定義の確認

収率 $Y_B(\tau)$ は、「供給した原料 $A$ の初期濃度に対する、生成した目的物質 $B$ の濃度の割合」として定義されます。定容系を仮定すると、濃度を用いて次のように表されます。

$$Y_B(\tau) = \frac{C_B(\tau)}{C_{A0}}$$

PFRにおける $B$ の濃度 $C_B(\tau)$ は Eq. (2)（$k_1 \neq k_2$ の場合）であるので、これを収率の定義式(eqY)に代入します。

$$Y_B(\tau) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)}{C_{A0}}$$

分子・分母の $C_{A0}$ で約分すると、収率 $Y_B(\tau)$ の空間時間変化を表す一般式が得られます。

$$Y_B(\tau) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right) \tag{5}$$

---

### 補足：挙動の特徴

* **反応器入口 ($\tau = 0$ のとき):**
$e^0 = 1$ より、$Y_B(0) = \frac{k_1}{k_2 - k_1} (1 - 1) = 0$ となり、まだ $B$ は生成していません。
* **空間時間が非常に大きい場合 ($\tau \to \infty$ のとき):**
$e^{-k_1 \tau} \to 0, e^{-k_2 \tau} \to 0$ となるため、$Y_B(\infty) \to 0$ となります。最終的にすべての $B$ が $C$ に変化するため、収率はゼロに戻ります。
* **極大値の存在:**
空間時間 $\tau$ が大きくなるにつれて $Y_B(\tau)$ は一旦増加してピーク（極大値）を迎えた後、減少に転じます。このピークを与える空間時間が「最適空間時間 $\tau_{\max}$」となります。

---

### (3) 選択率 $S_B(\tau)$ の空間時間変化の導出

選択率の定義式(eqS)に $C_A(\tau)$ および $C_B(\tau)$ の式を代入します。

$$S_B(\tau) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)}{C_{A0} - C_{A0} e^{-k_1 \tau}}$$

分子・分母の $C_{A0}$ で約分すると、選択率の空間時間変化を表す一般式が得られます。

$$S_B(\tau) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 \tau} - e^{-k_2 \tau}}{1 - e^{-k_1 \tau}} \tag{6}$$

* **挙動の傾向:**
* **反応器入口付近 ($\tau \to 0$):** 副反応 ($B \to C$) がほとんど起きていないため、$S_B \to 1$（100%）となります。
* **空間時間が非常に大きい場合 ($\tau \to \infty$):** すべての物質が $C$ に変化してしまうため、$S_B \to 0$ となります。すなわち、選択率は空間時間とともに単調減少します。

---

## 3. なにをもって最適化とするか（最適化の基準）

化学工業やプラント運転におけるPFRの設計・運用において「最適化」を行う際、何を基準にするか（目的関数）が極めて重要になります。逐次反応における最適化には、主に以下の視点が存在します。

## 3.1. 目的物質 $B$ の収率（または生産量）の最大化

技術的・化学的な観点における最も一般的な最適化基準です。

* **課題:** 反応管の長さ（空間時間 $\tau$）が短すぎると原料 $A$ が残ってしまい、長すぎると $B$ がさらに反応して $C$ になってしまうため、収率 $Y_B$ には必ず極大値が存在します。
* **最適空間時間 $\tau_{\max}$ の導出:**
$\frac{dC_B}{d\tau} = 0$ となる条件から、収率を最大にする空間時間が求められます。

$$\tau_{\max} = \frac{1}{k_2 - k_1} \ln\left(\frac{k_2}{k_1}\right) \tag{7}$$

* **最大収率 $Y_{B,\max}$:**

$$Y_{B,\max} = \left(\frac{k_1}{k_2}\right)^{\frac{k_2}{k_2 - k_1}} \tag{8}$$

````python
#@title PFR
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 滞留時間（空間時間）の軸設定 (tau = 0 から 5 まで)
tau = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A の計算
X_A = 1 - np.exp(-k1 * tau)

# 2. 収率 (Yield) Y_B の計算
if k1 == k2:
    Y_B = k1 * tau * np.exp(-k1 * tau)
    tau_max = 1 / k1
    Y_B_max = np.exp(-1)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * tau) - np.exp(-k2 * tau))
    tau_max = (1 / (k2 - k1)) * np.log(k2 / k1)
    Y_B_max = (k1 / k2) ** (k2 / (k2 - k1))

# 3. 選択率 (Selectivity) S_B の計算
S_B = np.ones_like(tau)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]

# tau_max における値の計算
X_A_taumax = 1 - np.exp(-k1 * tau_max)
S_B_taumax = Y_B_max / X_A_taumax

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(tau, X_A, label=r'Conversion ($X_A$)', color='blue', linewidth=2)
plt.plot(tau, Y_B, label=r'Yield ($Y_B$)', color='green', linewidth=2)
plt.plot(tau, S_B, label=r'Selectivity ($S_B$)', color='orange', linewidth=2, linestyle='--')

# 最適滞留時間 tau_max の明示
plt.axvline(x=tau_max, color='red', linestyle=':', linewidth=1.5, label=r'Optimal $\tau_{\max}$')
plt.scatter([tau_max], [Y_B_max], color='green', zorder=5)

annotation_text = (
    f'PFR Max Yield Point\n'
    f'  $\tau_{{max}} = {tau_max:.2f}$\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A(\tau_{{max}}) = {X_A_taumax:.2f}$\n'
    f'  $S_B(\tau_{{max}}) = {S_B_taumax:.2f}$'
)

plt.annotate(annotation_text,
             xy=(tau_max, Y_B_max),
             xytext=(tau_max + 0.3, Y_B_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

plt.title(r'PFR Consecutive Reaction ($A \to B \to C$) Analysis ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Space Time / Residence Time ($\tau$)', fontsize=12)
plt.ylabel('Fraction / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.1)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
````

<h1><font color="RoyalBlue">Theme 7：CSTRの設計方程式と最適化</font></h1>


これまでの回分反応器および管型反応器（PFR）に続き、連続式反応器のもう一つの基本型である完全混合槽型反応器（Continuous Stirred-Tank Reactor: CSTR）における複合反応の設計方程式を導出します。

---

### 1.1. CSTRの基本仮定

完全混合槽型反応器は、槽内の流体が攪拌機によって瞬時に、かつ完全に均一に混合されると仮定する連続式反応器です。

* **定常状態（Steady State）**：時間の経過に伴う反応器内の状態変化がありません（$\frac{\partial}{\partial t} = 0$）。
* **完全混合（Complete Mixing）**：槽内のどの場所でも濃度や温度が均一であり、**反応器の出口における組成は、反応器内部の組成と完全に一致**します。

---

### 1.2. 微小体積要素ではなく「系全体」の物質収支

CSTRでは槽内が均一であるため、微小要素ではなく反応器全体（体積 $V$）を対象として物質収支を立てます。「流入 - 流出 + 反応による生成・消費 = 0」の定常状態の式は以下のようになります。

$$F_{i0} - F_i + V \sum_{j=1}^{M} \nu_{i,j} r_j = 0 \quad (i = 1, 2, \dots, N)$$

* $F_{i0}$：化学種 $i$ の流入モル速度 $[\mathrm{mol/s}]$
* $F_i$：化学種 $i$ の流出モル速度（＝反応器内のモル速度） $[\mathrm{mol/s}]$
* $V$：反応器の有効体積 $[\mathrm{m^3}]$

---

### 1.3. 濃度基準の設計方程式への変換（定容系の場合）

体積流量 $v$ が一定（$v = v_0 = \text{constant}$）である定容系を仮定し、モル速度を $F_i = v_0 C_i$（流入側は $F_{i0} = v_0 C_{i0}$）と表します。空間時間 $\tau = \frac{V}{v_0}$ を用いて整理すると、CSTRの設計方程式が得られます。

$$C_i - C_{i0} = \tau \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

微分形ではなく、**代数方程式（Algebraic Equation）の形で表される**のがCSTRの大きな特徴です。

---

## 2. 逐次反応 $A \to B \to C$ への適用と諸定義

次のような一次の不可逆逐次反応をCSTRで実施する系を考えます。

$$A \xrightarrow{k_1} B \xrightarrow{k_2} C$$

化学量論係数を代入すると、CSTRにおける各成分の代数方程式は以下のようになります。

1. **成分 $A$:** $C_A - C_{A0} = \tau (-k_1 C_A) \implies C_A = \frac{C_{A0}}{1 + k_1 \tau}$
2. **成分 $B$:** $C_B - 0 = \tau (k_1 C_A - k_2 C_B) \implies C_B = \frac{k_1 \tau C_A}{1 + k_2 \tau} = \frac{k_1 \tau C_{A0}}{(1 + k_1 \tau)(1 + k_2 \tau)}$
3. **成分 $C$:** $C_C - 0 = \tau (k_2 C_B) \implies C_C = C_{A0} - C_A - C_B$

---

### 2.1. 転化率・収率・選択率の導出

#### (1) 転化率 $X_A$

$$X_A(\tau) = \frac{C_{A0} - C_A}{C_{A0}} = 1 - \frac{1}{1 + k_1 \tau} = \frac{k_1 \tau}{1 + k_1 \tau} \tag{eqX-cstr}$$

#### (2) 収率 $Y_B$

$$Y_B(\tau) = \frac{C_B}{C_{A0}} = \frac{k_1 \tau}{(1 + k_1 \tau)(1 + k_2 \tau)} \tag{eqY-cstr}$$

#### (3) 選択率 $S_B$

$$S_B(\tau) = \frac{Y_B(\tau)}{X_A(\tau)} = \frac{1}{1 + k_2 \tau} \tag{eqS-cstr}$$

---

## 3. CSTRにおける最適化（中間体 $B$ の最大収率）

CSTRにおいて、目的生成物 $B$ の収率 $Y_B(\tau)$ を最大にする最適な空間時間 $\tau_{\max}$ は、収率の式を $\tau$ で微分してゼロとおくことで求められます。

$$\frac{dY_B}{d\tau} = 0$$

これを解くと、CSTR特有の最適空間時間が得られます。

$$\tau_{\max} = \frac{1}{\sqrt{k_1 k_2}} \tag{7-cstr}$$

このときの**CSTRにおける最大収率 $Y_{B,\max}$** は次のように表されます。

$$Y_{B,\max} = \frac{1}{\left(1 + \sqrt{\frac{k_2}{k_1}}\right)^2} \tag{8-cstr}$$

---

### 💡 バッチ/PFR と CSTR の決定的な違い（考察）

同じ速度定数 $k_1, k_2$ で比較した場合、**CSTRの最大収率 $Y_{B,\max}$ は、バッチやPFR（管型反応器）の最大収率よりも必ず低くなります。**

* **理由**: CSTRは完全混合であるため、反応器内に入った瞬間から生成した $B$ が高濃度で存在し、次の副反応（$B \to C$）の環境に晒されてしまいます。そのため、中間体を大量に得たい逐次反応において、**CSTRは構造上非常に不利な反応器**となります。

````python
#@title CODE 2：CSTRにおけるBの収率最大化
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 空間時間（Residence Time / Space Time）軸の設定 (tau = 0 から 5 まで)
tau = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A の計算 (CSTRの式)
X_A = (k1 * tau) / (1 + k1 * tau)

# 2. 収率 (Yield) Y_B の計算 (CSTRの式)
Y_B = (k1 * tau) / ((1 + k1 * tau) * (1 + k2 * tau))

# CSTRにおける最適空間時間 tau_max と最大収率 Y_B_max の理論計算
tau_max = 1 / np.sqrt(k1 * k2)
Y_B_max = 1 / (1 + np.sqrt(k2 / k1))**2

# 3. 選択率 (Selectivity) S_B の計算 (t=0での0/0除算エラーを防ぐ処理、または CSTRの解析解 1 / (1 + k2*tau))
S_B = np.ones_like(tau)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]

# --- tau_max における X_A と S_B の計算 ---
X_A_taumax = (k1 * tau_max) / (1 + k1 * tau_max)
S_B_taumax = Y_B_max / X_A_taumax

print(f"=== CSTR at tau_max = {tau_max:.3f} ===")
print(f"  Conversion (X_A): {X_A_taumax:.3f}")
print(f"  Yield      (Y_B): {Y_B_max:.3f}")
print(f"  Selectivity (S_B): {S_B_taumax:.3f}")
#-----------------------------------------------

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(tau, X_A, label=r'Conversion ($X_A$)', color='blue', linewidth=2)
plt.plot(tau, Y_B, label=r'Yield ($Y_B$)', color='green', linewidth=2)
plt.plot(tau, S_B, label=r'Selectivity ($S_B$)', color='orange', linewidth=2, linestyle='--')

# 最適空間時間 tau_max の明示（縦線とアノテーション）
plt.axvline(x=tau_max, color='red', linestyle=':', linewidth=1.5, label=r'$\tau_{\max}$')
plt.scatter([tau_max], [Y_B_max], color='green', zorder=5)
plt.scatter([tau_max], [X_A_taumax], color='blue', zorder=5)
plt.scatter([tau_max], [S_B_taumax], color='orange', zorder=5)

# アノテーションに tau_max 時の X_A, Y_B, S_B をすべて記載
annotation_text = (
    f'CSTR Max Yield Point ($\tau_{{max}} = {tau_max:.2f}$)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A(\tau_{{max}}) = {X_A_taumax:.2f}$\n'
    f'  $S_B(\tau_{{max}}) = {S_B_taumax:.2f}$'
)

plt.annotate(annotation_text,
             xy=(tau_max, Y_B_max),
             xytext=(tau_max + 0.3, Y_B_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

# グラフの装飾（英語化）
plt.title(r'CSTR Consecutive Reaction ($A \to B \to C$) Analysis ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel(r'Space Time ($\tau$)', fontsize=12)
plt.ylabel('Fraction / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.1)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
````

# 講義ノート：非等温回分反応器における逐次反応（$A \to B \to C, E_1 > E_2$）の数学的モデリングと解析手順

本講義ノートでは、逐次反応 $A \xrightarrow{k_1} B \xrightarrow{k_2} C$ を対象とし、反応器の温度が時間とともに変化する非等温回分反応器（Non-isothermal Batch Reactor）の設計方程式（物質収支式およびエンタルピー収支式）の導出と、$E_1 > E_2$（第1段階の活性化エネルギーが第2段階よりも大きい）の条件を活かした解析・最適化の手順を解説します。

---

## 1. 物質収支式（Mass Balance Equations）の導出

定容系（反応液の体積 $V = V_0 = \text{一定}$）の回分反応器を仮定します。各成分の反応速度はアレニウスの式に従うため、速度定数 $k_1, k_2$ は温度 $T$ の関数となります。

$$k_1(T) = A_1 \exp\left(-\frac{E_1}{RT}\right), \quad k_2(T) = A_2 \exp\left(-\frac{E_2}{RT}\right)$$

前回の一般式 $\frac{dC_i}{dt} = \sum \nu_{i,j} r_j$ に基づき、各成分の物質収支式は以下のように立式されます。

1. **成分 $A$ の収支:**

$$\frac{dC_A}{dt} = -r_1 = -k_1(T) C_A$$


2. **成分 $B$ の収支（生成と消費の競合）:**

$$\frac{dC_B}{dt} = r_1 - r_2 = k_1(T) C_A - k_2(T) C_B$$


3. **成分 $C$ の収支:**

$$\frac{dC_C}{dt} = r_2 = k_2(T) C_B$$



---

## 2. エンタルピー収支式（Energy / Enthalpy Balance Equations）の導出

非等温系では温度 $T$ が変数となるため、温度のの時間変化 $\frac{dT}{dt}$ を記述するエネルギー保存則（熱収支式）が必要になります。

### (1) 閉ざされた系のエネルギー保存則

熱力学の第一法則より、外部との仕事のやり取り（容積仕事）を無視できる液相系では、内部エネルギー $U$ の変化は流入・流出がないため熱の出入りおよび反応熱に等しくなります。

$$\frac{dU}{dt} = \dot{Q} + \sum_{j=1}^{M} (-\Delta H_j) V r_j$$

* $\dot{Q}$：外部のジャケット等から加えられる（または奪われる）伝熱速度 $[\mathrm{J/s}]$
* $\Delta H_j$：$j$ 番目の反応の標準反応エンタルピー（発熱反応の場合は負：$\Delta H < 0$）

### (2) 温度に関する微分方程式への変換

内部エネルギー $U$ およびエンタルピー $H$ は温度と組成の関数です。液相の密度 $\rho$ および比熱 $C_p$ が一定であると仮定し、全系の熱容量を用いて温度 $T$ の変化率 $\frac{dT}{dt}$ に書き換えます。

$$V \rho C_p \frac{dT}{dt} = \dot{Q} + V \left[ (-\Delta H_1) r_1 + (-\Delta H_2) r_2 \right]$$

ここで、外部からの伝熱速度 $\dot{Q}$ は、伝熱係数 $U_h$、伝熱面積 $A_h$、およびジャケット温度 $T_j$ を用いて $\dot{Q} = U_h A_h (T_j - T)$ と表されます。これを代入すると、非等温回分反応器のエンタルピー収支式（温度の微分方程式）が得られます。

$$\rho C_p \frac{dT}{dt} = U_h a (T_j - T) + (-\Delta H_1) k_1(T) C_A + (-\Delta H_2) k_2(T) C_B$$

（※ $a = A_h / V$ は単位体積あたりの伝熱面積）

---

## 3. $E_1 > E_2$ 条件を活かす解析と最適化の手順

活性化エネルギーの関係が $E_1 > E_2$ である場合、温度変化に対する反応速度の感度が異なります。

* **$E_1 > E_2$ の物理的意味:** 温度を上げると、副反応 $B \to C$ よりも主反応 $A \to B$ の方がより急激に加速されます。

この特性を最大限に利用し、中間体 $B$ の収率を極限まで高めるための解析・最適化手順は以下の通りです。

### ステップ 1：物性値および運動論パラメータの定義

* 各種パラメータ（$A_1, A_2, E_1, E_2, \Delta H_1, \Delta H_2, \rho, C_p, U_h a$）を実験データ等から設定します。

### ステップ 2：連立常微分方程式（ODE）の構築

以下の4元連立非線形微分方程式システムを構築します。

1. $\frac{dC_A}{dt} = -k_1(T) C_A$
2. $\frac{dC_B}{dt} = k_1(T) C_A - k_2(T) C_B$
3. $\frac{dC_C}{dt} = k_2(T) C_B$
4. $\frac{dT}{dt} = \frac{1}{\rho C_p} \left[ U_h a (T_j - T) + (-\Delta H_1) k_1(T) C_A + (-\Delta H_2) k_2(T) C_B \right]$

### ステップ 3：最適温度プロファイル（$T_j(t)$ または $T(t)$）の設計

* **反応前半（高温運転）:** $E_1 > E_2$ であるため、あえて反応器温度を高めに設定し、主反応速度 $k_1$ を極限まで引き上げて短時間で原料 $A$ を消費させます。
* **反応後半（急冷・低温化）:** 目的物質 $B$ が最大濃度に近づいた瞬間、ジャケット温度 $T_j$ を急激に下げて系全体の温度を低下させます。これにより、相対的に $E_2$ が小さくても温度低下によって $k_2$ が急速に抑制され、$B \to C$ への過剰反応がブロックされます。

### ステップ 4：数値シミュレーションによる検証

* Python等の数値計算ライブラリ（`scipy.integrate.solve_ivp` など）を用いて微分方程式を解き、等温運転時と非等温最適制御時での $Y_B$ の最大値を比較・評価します。

---

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# --- 1. パラメータの設定 ---
R = 8.314  # 気体定数 [J/(mol*K)]
T_room = 298.15  # 基準温度（室温） [K]

# 活性化エネルギーの設定 (E1 > E2 の条件)
E1 = 55000.0  # 主反応 A -> B の活性化エネルギー [J/mol]
E2 = 45000.0  # 副反応 B -> C の活性化エネルギー [J/mol]

# 室温で k1=1.0, k2=0.5 になるように頻度因子 A1, A2 を逆算決定
k1_room = 1.0
k2_room = 0.5
A1 = k1_room * np.exp(E1 / (R * T_room))
A2 = k2_room * np.exp(E2 / (R * T_room))

# 反応熱 (発熱反応) [J/mol]
dH1 = -50000.0
dH2 = -40000.0

# 物性値・熱伝達パラメータ
rho_Cp = 4180.0  # 体積あたりの熱容量 [J/(L*K)]
Ua = 200.0  # 総括伝熱係数×伝熱面積/体積 [J/(L*s*K)]

# --- ★ ジャケット温度を一定に設定 ---
T_j_const = 350.0  # ジャケット温度 [K] (一定)


# --- 2. 4元連立常微分方程式の定義 (y = [C_A, C_B, C_C, T]) ---
def nonisothermal_batch_constant_jacket(t, y):
  CA, CB, CC, T = y

  # アレニウスの式による温度依存の速度定数
  k1 = A1 * np.exp(-E1 / (R * T))
  k2 = A2 * np.exp(-E2 / (R * T))

  # 反応速度
  r1 = k1 * CA
  r2 = k2 * CB

  # 物質収支 (濃度変化)
  dCA_dt = -r1
  dCB_dt = r1 - r2
  dCC_dt = r2

  # エンタルピー収支 (温度変化) ※ジャケット温度 T_j_const は一定
  dT_dt = (
      Ua * (T_j_const - T) + (-dH1) * r1 + (-dH2) * r2
  ) / rho_Cp

  return [dCA_dt, dCB_dt, dCC_dt, dT_dt]


# --- 3. 初期条件と時間設定 ---
CA0 = 1.0  # Aの初期濃度 [mol/L]
CB0 = 0.0
CC0 = 0.0
#T0 = T_room  # 初期温度（室温 298.15 K）
T0 = T_j_const
y0 = [CA0, CB0, CC0, T0]

t_span = (0, 1.5)
t_eval = np.linspace(t_span[0], t_span[1], 500)

# --- 4. 数値計算の実行 (solve_ivp) ---
sol = solve_ivp(
    nonisothermal_batch_constant_jacket,
    t_span,
    y0,
    t_eval=t_eval,
    method='RK45',
)

# --- 5. 転化率・収率・選択率の算出 ---
CA = sol.y[0]
CB = sol.y[1]
T_hist = sol.y[3]

X_A = 1.0 - (CA / CA0)  # 転化率
Y_B = CB / CA0  # 収率

S_B = np.ones_like(X_A)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]  # 選択率

# --- 6. 最適操作時間 t_max の自動検出 ---
max_idx = np.argmax(Y_B)
t_max = sol.t[max_idx]
Y_B_max = Y_B[max_idx]
X_A_tmax = X_A[max_idx]
S_B_tmax = S_B[max_idx]
T_tmax = T_hist[max_idx]

print(
    f"=== 非等温回分反応器 (ジャケット温度一定: {T_j_const:.1f} K) の結果 ==="
)
print(f"  最適操作時間 (t_max): {t_max:.3f} s")
print(f"  最大収率 (Y_B_max):   {Y_B_max:.3f}")
print(f"  時の転化率 (X_A):     {X_A_tmax:.3f}")
print(f"  時の選択率 (S_B):     {S_B_tmax:.3f}")
print(f"  時の反応器温度 (T):   {T_tmax:.1f} K")

# --- 7. 2段グラフ描画 ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# 上段：転化率、収率、選択率の時間変化
ax1.plot(sol.t, X_A, label=r'Conversion ($X_A$)', color='blue', lw=2)
ax1.plot(sol.t, Y_B, label=r'Yield ($Y_B$)', color='green', lw=2)
ax1.plot(
    sol.t,
    S_B,
    label=r'Selectivity ($S_B$)',
    color='orange',
    lw=2,
    linestyle='--',
)
ax1.axvline(
    x=t_max, color='red', linestyle=':', linewidth=1.5, label=r'$t_{\max}$'
)
ax1.scatter([t_max], [Y_B_max], color='green', zorder=5)

annotation_text = (
    f'Max Yield Point ($t_{{max}} = {t_max:.2f}$ s)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A = {X_A_tmax:.2f}$'
)
ax1.annotate(
    annotation_text,
    xy=(t_max, Y_B_max),
    xytext=(t_max + 0.1, Y_B_max - 0.25),
    arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
    fontsize=10,
    color='darkred',
    bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='red', lw=1, alpha=0.9),
)
ax1.set_ylabel('Fraction / Ratio (-)', fontsize=11)
ax1.set_title(
    r'Non-isothermal Batch Reactor (Constant Jacket Temp $T_j = '
    f'{T_j_const:.1f}$ K)',
    fontsize=13,
)
ax1.set_ylim(-0.05, 1.1)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend(fontsize=10, loc='lower right')

# 下段：反応器温度とジャケット温度の推移
ax2.plot(sol.t, T_hist, label='Reactor Temp ($T$)', color='purple', lw=2)
ax2.axhline(
    y=T_j_const,
    color='gray',
    linestyle='--',
    lw=1.5,
    label=f'Jacket Temp ($T_j = {T_j_const:.1f}$ K)',
)
ax2.axvline(x=t_max, color='red', linestyle=':', linewidth=1.5)
ax2.set_xlabel('Reaction Time ($t$) [s]', fontsize=11)
ax2.set_ylabel('Temperature [K]', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.legend(fontsize=10, loc='upper right')

plt.tight_layout()
plt.show()
````

````python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# --- 1. パラメータの設定 ---
R = 8.314  # 気体定数 [J/(mol*K)]
T_room = 298.15  # 基準温度（室温） [K]

# 活性化エネルギーの設定 (E1 > E2 の条件)
E1 = 55000.0  # 主反応 A -> B の活性化エネルギー [J/mol]
E2 = 45000.0  # 副反応 B -> C の活性化エネルギー [J/mol]

# 室温で k1=1.0, k2=0.5 になるように頻度因子 A1, A2 を逆算決定
k1_room = 1.0
k2_room = 0.5
A1 = k1_room * np.exp(E1 / (R * T_room))
A2 = k2_room * np.exp(E2 / (R * T_room))

# 反応熱 (発熱反応を想定) [J/mol]
dH1 = -50000.0
dH2 = -40000.0

# 物性値・熱伝達パラメータ
rho_Cp = 4180.0  # 体積あたりの熱容量 [J/(L*K)]
Ua = 200.0  # 総括伝熱係数×伝熱面積/体積 [J/(L*s*K)]


# --- 2. ジャケット温度 T_j のプログラム制御（急冷クエンチング） ---
def get_jacket_temp(t):
  if t < 0.4:
    return 360.0  # 反応前半：高温にして主反応を加速
  else:
    return 280.0  # 反応後半：急冷して副反応 B -> C をフリーズ


# --- 3. 4元連立常微分方程式の定義 (y = [C_A, C_B, C_C, T]) ---
def nonisothermal_batch(t, y):
  CA, CB, CC, T = y

  # アレニウスの式による温度依存の速度定数
  k1 = A1 * np.exp(-E1 / (R * T))
  k2 = A2 * np.exp(-E2 / (R * T))

  # 反応速度
  r1 = k1 * CA
  r2 = k2 * CB

  # ① 物質収支式 (濃度変化)
  dCA_dt = -r1
  dCB_dt = r1 - r2
  dCC_dt = r2

  # ジャケット温度の取得
  T_j = get_jacket_temp(t)

  # ② エンタルピー収支式 (温度変化)
  dT_dt = (
      Ua * (T_j - T) + (-dH1) * r1 + (-dH2) * r2
  ) / rho_Cp

  return [dCA_dt, dCB_dt, dCC_dt, dT_dt]


# --- 4. 初期条件と時間設定 ---
CA0 = 1.0  # Aの初期濃度 [mol/L]
CB0 = 0.0
CC0 = 0.0
T0 = T_room  # 初期温度（室温 298.15 K）
y0 = [CA0, CB0, CC0, T0]

t_span = (0, 1.2)
t_eval = np.linspace(t_span[0], t_span[1], 500)

# --- 5. 数値計算の実行 (solve_ivp) ---
sol = solve_ivp(nonisothermal_batch, t_span, y0, t_eval=t_eval, method='RK45')

# --- 6. 転化率・収率・選択率の算出 ---
CA = sol.y[0]
CB = sol.y[1]
T_hist = sol.y[3]

X_A = 1.0 - (CA / CA0)  # 転化率
Y_B = CB / CA0  # 収率

S_B = np.ones_like(X_A)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]  # 選択率

# --- 7. 最適操作時間 t_max の自動検出 ---
max_idx = np.argmax(Y_B)
t_max = sol.t[max_idx]
Y_B_max = Y_B[max_idx]
X_A_tmax = X_A[max_idx]
S_B_tmax = S_B[max_idx]
T_tmax = T_hist[max_idx]

print("=== 非等温回分反応器のシミュレーション結果 ===")
print(f"  最適操作時間 (t_max): {t_max:.3f} s")
print(f"  最大収率 (Y_B_max):   {Y_B_max:.3f}")
print(f"  時の転化率 (X_A):     {X_A_tmax:.3f}")
print(f"  時の選択率 (S_B):     {S_B_tmax:.3f}")
print(f"  時の反応器温度 (T):   {T_tmax:.1f} K")

# --- 8. 2段グラフ描画 ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# 上段：転化率、収率、選択率の時間変化
ax1.plot(sol.t, X_A, label=r'Conversion ($X_A$)', color='blue', lw=2)
ax1.plot(sol.t, Y_B, label=r'Yield ($Y_B$)', color='green', lw=2)
ax1.plot(
    sol.t,
    S_B,
    label=r'Selectivity ($S_B$)',
    color='orange',
    lw=2,
    linestyle='--',
)
ax1.axvline(
    x=t_max, color='red', linestyle=':', linewidth=1.5, label=r'$t_{\max}$'
)
ax1.scatter([t_max], [Y_B_max], color='green', zorder=5)

annotation_text = (
    f'Max Yield Point ($t_{{max}} = {t_max:.2f}$ s)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A = {X_A_tmax:.2f}$'
)
ax1.annotate(
    annotation_text,
    xy=(t_max, Y_B_max),
    xytext=(t_max + 0.12, Y_B_max - 0.25),
    arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
    fontsize=10,
    color='darkred',
    bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='red', lw=1, alpha=0.9),
)
ax1.set_ylabel('Fraction / Ratio (-)', fontsize=11)
ax1.set_title(
    r'Non-isothermal Batch Reactor: $X_A, Y_B, S_B$ Evolution ($E_1 > E_2$)',
    fontsize=13,
)
ax1.set_ylim(-0.05, 1.1)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend(fontsize=10, loc='lower right')

# 下段：反応器温度とジャケット温度の推移
T_j_history = [get_jacket_temp(t) for t in sol.t]
ax2.plot(sol.t, T_hist, label='Reactor Temp ($T$)', color='purple', lw=2)
ax2.plot(
    sol.t,
    T_j_history,
    label='Jacket Temp ($T_j$)',
    color='gray',
    lw=1.5,
    linestyle='--',
)
ax2.axvline(x=t_max, color='red', linestyle=':', linewidth=1.5)
ax2.set_xlabel('Reaction Time ($t$) [s]', fontsize=11)
ax2.set_ylabel('Temperature [K]', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.legend(fontsize=10, loc='upper right')

plt.tight_layout()
plt.show()
````

---

逐次反応 $A \to B \to C$ における選択率 $S_B(t)$ の時間変化の式を以下に導出します。

### 1. 選択率の定義

選択率 $S_B$ は、「反応した原料 $A$ のうち、目的物質 $B$ に変化した割合」として定義されます。


$$S_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{反応した } A \text{ の物質量}} = \frac{C_B(t)}{C_{A0} - C_A(t)}$$

### 2. 各濃度の時間変化式の代入

回分反応器における各成分の濃度 $C_A(t)$ および $C_B(t)$ は以下の通りです。


$$C_A(t) = C_{A0} e^{-k_1 t}$$

$$C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

これらを $S_B(t)$ の定義式に代入します。


$$S_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0} - C_{A0} e^{-k_1 t}}$$

### 3. 式の整理

分子・分母の $C_{A0}$ で約分します。


$$S_B(t) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 t} - e^{-k_2 t}}{1 - e^{-k_1 t}}$$

これが選択率 $S_B(t)$ の時間変化を表す一般式です。

---

### 補足：極限挙動の確認

この式から、時間の経過に伴う選択率の変化を読み取ることができます。

* **反応初期 ($t \to 0$ のとき):**
分子・分母ともに $0$ に収束するため、ロピタルの定理を適用すると $S_B \to 1$ となります。これは、反応初期には副反応 ($B \to C$) がほとんど起きていないためです。
* **反応長時間経過 ($t \to \infty$ のとき):**
$e^{-k_1 t} \to 0$ および $e^{-k_2 t} \to 0$ となるため、$S_B \to 0$ となります。これは、最終的にすべての物質が $C$ に変化してしまうためです。

逐次反応 $A \to B \to C$ における**転化率 $X_A$ の時間変化**および、転化率を基準とした反応解析を以下に解説・導出します。

---

## 1. 転化率の時間変化の導出

### (1) 定義の確認

転化率 $X_A(t)$ は、反応開始から時間 $t$ までに反応した原料 $A$ の割合として定義されます。


$$X_A(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### (2) 濃度の時間変化の代入

回分反応器における原料 $A$ の濃度は、一次反応の速度式より以下のように表されます。


$$C_A(t) = C_{A0} e^{-k_1 t}$$

これを転化率の定義式に代入します。


$$X_A(t) = 1 - \frac{C_{A0} e^{-k_1 t}}{C_{A0}} = 1 - e^{-k_1 t}$$

これが転化率 $X_A(t)$ の時間変化を表す一般式です。

### (3) 転化率の時間変化率（速度式）

転化率の時間に対する微分（速度）は次のようになります。


$$\frac{dX_A}{dt} = \frac{d}{dt}(1 - e^{-k_1 t}) = k_1 e^{-k_1 t} = k_1 (1 - X_A)$$

---

## 2. 転化率に関する重要な特徴

1. **第2段階の反応（$k_2$）の影響を受けない**
* 原料 $A$ が $B$ に変わる最初のステップは、独立した一次反応です。そのため、転化率 $X_A$ の式には $k_2$ や物質 $B, C$ の項が一切含まれません。すなわち、**転化率の挙動は $k_1$ のみで決まります。**


2. **時間の経過に伴う挙動**
* 反応初期 ($t = 0$): $X_A = 0$（未反応）
* 長時間経過 ($t \to \infty$): $X_A \to 1$（100%反応）



---

## 3. 発展：転化率を基準とした収率・選択率の解析

反応工学では、時間を変数にするのではなく、**「原料がどこまで反応したか（＝転化率 $X_A$）」を基準にして中間生成物 $B$ の収率や選択率を評価する**ことがよく行われます。

### (1) 収率 $Y_B$ と転化率 $X_A$ の関係

時間 $t$ を消去し、$X_A$ と $Y_B$ の関係を導出します。

* $e^{-k_1 t} = 1 - X_A$ より、 $t = -\frac{1}{k_1} \ln(1 - X_A)$
* これを $B$ の濃度式 $C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$ に代入します。
* $e^{-k_2 t} = (e^{-k_1 t})^{k_2 / k_1} = (1 - X_A)^{k_2 / k_1}$ と変形できるため、収率 $Y_B$ は次のように表されます。

$$Y_B = \frac{k_1}{k_2 - k_1} \left[ (1 - X_A) - (1 - X_A)^{\frac{k_2}{k_1}} \right]$$

### (2) 最適な転化率（$X_{A,\text{opt}}$）の考察

中間生成物 $B$ の収率を最大にする「最適な転化率」は、上式を $X_A$ で微分してゼロとおくことで求められます。

$$\frac{dY_B}{dX_A} = \frac{k_1}{k_2 - k_1} \left[ -1 + \frac{k_2}{k_1} (1 - X_A)^{\frac{k_2}{k_1} - 1} \right] = 0$$

これを解くと、収率が最大になるときの転化率 $X_{A,\text{opt}}$ が得られます。

$$X_{A,\text{opt}} = 1 - \left(\frac{k_1}{k_2}\right)^{\frac{k_1}{k_2 - k_1}}$$

* **実務的な意味:**
* 「原料を完全に反応させよう（転化率 $X_A \to 1$ にしよう）」として反応を続けすぎると、中間体 $B$ まで $C$ に変わってしまい、かえって $B$ の収率が低下します。
* したがって、反応を途中で打ち切るべき「最適な転化率の限界値」が存在することが、この解析からわかります。

#【付録】
### ② 経済性・生産性（コストパフォーマンス）の最適化

実際のプラント設計では、「収率の絶対値」だけでなく「時間あたりの生産性（Space-Time Yield）」や「経済性」が最適化の基準になります。

* **バッチ反応器の稼働サイクル:** 回分反応器では、反応に要する時間 $t_{\text{react}}$ だけでなく、原料の仕込み、生成物の取り出し、洗浄などの付帯時間（downtime: $t_{\text{down}}$）が必ず発生します。
* **時間あたりの生産量 ($P_{\text{eff}}$):**

$$P_{\text{eff}} = \frac{C_B(t)}{t + t_{\text{down}}}$$



理論上の最高収率が得られる時間 $t_{\max}$ まで反応させるよりも、少し手前（あるいは状況に応じた時間）で反応を打ち切る方が、1日あたりのバッチ回転数が上がり、トータルの生産量が大きくなる場合があります。

### ③ 原料コストと分離精製コストのバランス（トレードオフの克服）

* **転化率と選択率のトレードオフ:** 高い転化率 $X_A$ を狙うほど選択率 $S_B$ は低下します（未反応の $A$ は回収して再循環させればよいが、$C$ まで副生すると損失になる）。
* **分離・精製の負荷:** 未反応の $A$ や副生物 $C$ が多く混ざると、後段の分離・精製工程（蒸留や抽出など）のエネルギーコストが跳ね上がります。「反応工程の収率」だけでなく「プラント全体のトータルコスト（反応＋分離）」を最小化することが、真の経済的最適化となります。

---

## 5. まとめ

* 逐次反応 $A \to B \to C$ では、転化率 $X_A$ は時間とともに単調増加し、選択率 $S_B$ は単調減少する。
* 中間生成物 $B$ の収率 $Y_B$ には必ず極大値が存在し、速度定数から決まる**最適反応時間 $t_{\max}$** が存在する。
* 「最適化」の定義は状況によって異なり、単なる収率の最大化だけでなく、**生産性（時間あたりの効率）**や**分離コストを含めた経済性全体**を総合的に評価して決定される。

**はい、その通りです。**

体積流量 $v$ が一定（液相反応や、気相でも全モル数が変わらない・あるいは容積変化が無視できる場合）であれば、$F = v C$ が成り立つため、連続系と回分系（バッチ）の定義式は濃度 $C$ を用いた**完全に同じ数式**になります。

---

### なぜ同じ式になるのか（対応関係）

回分系のモル数 $N$ と、連続系のモル流量 $F$ は、それぞれ以下のように濃度 $C$ に置き換えられます。

* **回分系（バッチ）**: $N_A = V \cdot C_A$ （$V$ は反応液量）
* **連続系（流通）**: $F_A = v \cdot C_A$ （$v$ は体積流量）

これをそれぞれの定義式に代入すると、体積 $V$ や体積流量 $v$ が分子・分母で相殺されるため、濃度ベースの同じ式に帰着します。

#### 1. 収率（総括・供給原料基準: $Y_B$）

* **回分系:**

$$Y_B = \frac{N_B - N_{B0}}{N_{A0}} = \frac{v C_B - v C_{B0}}{v C_{A0}} = \frac{C_B - C_{B0}}{C_{A0}}$$


* **連続系:**

$$Y_B = \frac{F_B - F_{B0}}{F_{A0}} = \frac{v C_B - v C_{B0}}{v C_{A0}} = \frac{C_B - C_{B0}}{C_{A0}}$$



#### 2. 選択率（総括・反応原料基準: $S_{B/C}$ ※提示された表の $\Phi_B$ など）

※ご提示いただいた表の $\Phi_B$ の分母は、積（$F_{A0} F_A$）ではなく「反応した原料 $A$ の量（$F_{A0} - F_A$ または $N_{A0} - N_A$）」が正しい定義です。

* **回分系:**

$$\Phi_B = \frac{N_B - N_{B0}}{N_{A0} - N_A} = \frac{C_B - C_{B0}}{C_{A0} - C_A}$$


* **連続系:**

$$\Phi_B = \frac{F_B - F_{B0}}{F_{A0} - F_A} = \frac{C_B - C_{B0}}{C_{A0} - C_A}$$



このように、**体積（流量）が一定の条件下では、回分系のモル $N$ をすべて濃度 $C$（または連続系のモル流量 $F$）に置き換えて考えることができる**ため、実質的に同じ定義として扱って問題ありません。

# 講義資料：逐次反応における最適反応時間（空間時間）の探索

〜「転化率・収率・選択率」の定義の整理と工学的トレードオフ〜

## 1. 導入：なぜ指標の定義を厳密に区別するのか？

化学反応工学において、プロセスを評価する指標には「転化率」「収率」「選択率」があります。
本講義では、学生の誤解や混同を防ぐため、以下のように定義を厳格に切り分けて議論します。

* **分母に原料 $A$（供給量または反応量）が関与するものはすべて「収率」の範疇として捉える。**
* **選択率 ($S_{B/C}$) は、反応した原料が最終的に「目的物 $B$」と「副生物 $C$」のどちらにどれだけ分配されたかという「生成物の内訳（比率）」を表す指標として定義する。**

複合反応（逐次反応）では、「原料をどれだけ反応させるか（転化率）」と「どちらの生成物に変えるか（選択率 $S_{B/C}$）」がトレードオフの関係になります。これらを定量的に評価し、プラント全体の経済性を最大化する条件を探索します。

---

## 2. 逐次反応モデルの定義と各指標の定式化

一次の逐次反応を考えます。

$$A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$$

* 初期条件 ($t = 0$): 濃度 $C_A = C_{A0}$、$C_B = 0$、$C_C = 0$
* 任意の時間 $t$ における各成分の濃度：$C_A(t), C_B(t), C_C(t)$

> **💡 【重要なお断り：回分と連続におけるベースの違いについて】**
> * **回分反応器 (Batch):** 閉じた系であるため、**濃度 $C$**（またはモル数）の時間変化として直接定式化できます。
> * **連続反応器 (CSTR / PFR):** 流体の出入りがある開いた系であるため、厳密には**モル流量 $F$ $[\text{mol/s}]$** ベースの物質収支からスタートします。
> * *ただし、本講義および本資料の導入部では、見通しを良くするため「液相・体積一定（$v = v_0$）」の系を仮定し、$F_A = v_0 C_A$ および空間時間 $\tau = V/v_0$ を用いることで回分（時間 $t$）と完全に数学的アナロジーとして扱っています。*
>
>

### ① 転化率 ($X$)

投入した原料 $A$ のうち、何%が反応したか。


$$X(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### ② 収率 ($Y$)

投入した原料 $A$ 全体に対して、目的物 $B$ がどれだけ得られたか。


$$Y(t) = \frac{C_B(t)}{C_{A0}}$$


*(※分母に $A$ が入るため、プロセス全体の経済性・生産効率を表す指標となります)*

### ③ 選択率 ($S_{B/C}$)

反応した $A$ から生成した $B$ と $C$ の分配状況を表す指標として、**$B$ と $C$ の選択率 ($S_{B/C}$)** を導入し、それぞれの成分に対する選択率を $S_B$ および $S_C$ として定義します。

* **目的物 $B$ の選択率 ($S_B$):**

$$S_B = \frac{\text{生成した } B \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_B(t)}{C_{A0} - C_A(t)}$$


* **副生物 $C$ の選択率 ($S_C$):**

$$S_C = \frac{\text{生成した } C \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_C(t)}{C_{A0} - C_A(t)}$$



> **💡 重要な関係式:**
> モル保存則（$C_{A0} = C_A + C_B + C_C$）より、反応した $A$ はすべて $B$ または $C$ に分配されるため、**常に $S_B + S_C = 1$（または 100%）**が成り立ちます。
> また、収率・転化率・選択率の間には以下の関係が成り立ちます。
>
> $$Y = X \times S_B$$
>
>

---

## 3. 濃度変化の解析解と、指標の振る舞い

各成分の濃度変化は次のように求まります（$k_1 \neq k_2$ の場合）。

* $C_A(t) = C_{A0} e^{-k_1 t}$
* $C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$
* $C_C(t) = C_{A0} - C_A(t) - C_B(t)$

時間経過に伴う各指標の変化：

* **転化率 $X(t)$:** 時間とともに単調増加し、$1$（100%）に近づく。
* **選択率 $S_B$ と $S_C$:**
* 反応初期 ($t \approx 0$) は $S_B \approx 1$、$S_C \approx 0$ です。
* 時間が経つにつれて $B$ が $C$ に変化するため、**$S_B$ は単調に減少し、$S_C$ は単調に増加**します。


* **収率 $Y(t)$:** 「増える $X$」と「減る $S_B$」の積（$Y = X \times S_B$）であるため、**途中に必ず「山なりのピーク（極大値）」が現れます。**

---

## 4. 中間体 $B$ の最大収率を与える最適反応時間 $t_{\text{opt}}$ の導出

プラントの利益（目的物 $B$ の生産量）を最大化するためには、収率 $Y(t)$ が最大のところで反応を止めなければなりません。微分して $0$ と置くことで、最適反応時間 $t_{\text{opt}}$（またはPFRの最適空間時間 $\tau_{\text{opt}}$）が求まります。

$$t_{\text{opt}} = \frac{1}{k_1 - k_2} \ln\left(\frac{k_1}{k_2}\right)$$

> **⚠️ 工学的示唆:**
> この最適時点では、**転化率 $X$ は100%未満であり、副生物 $C$ への選択率 $S_C$ が一定割合存在した状態で反応を打ち切る**ことになります。
> 「原料を完全に使い切りたい（転化率100%）」という発想ではなく、あえて未反応 $A$ をリサイクルし、不要な $C$ への変換ロスを防ぐ設計がプラントの経済性を左右します。

## 5. 自習用演習問題と解説

### 【問題】

ある化学プラントで、液相の逐次反応 $A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$ が行われています。

* 反応速度定数：$k_1 = 2.0 \text{ h}^{-1}$、$k_2 = 0.5 \text{ h}^{-1}$
* 初濃度：$C_{A0} = 1.0 \text{ kmol/m}^3$

(1) 目的物 $B$ の収率 $Y$ が最大になる最適反応時間 $t_{\text{opt}}$ $[\text{h}]$ を求めよ。

(2) その最適時点における、**転化率 $X$、収率 $Y$、および $B$ と $C$ の選択率 ($S_B, S_C$)** の値をそれぞれ求めよ。

(3) もし誤って反応時間を $t = 2.0 \text{ h}$ まで延ばしてしまった場合、これらの指標はどう変化するか計算せよ。

---

### 【解答・解説】

**(1) 最適反応時間 $t_{\text{opt}}$ の計算**


$$t_{\text{opt}} = \frac{1}{2.0 - 0.5} \ln\left(\frac{2.0}{0.5}\right) = \frac{1}{1.5} \ln(4) \approx 0.924 \text{ h}$$

**(2) $t = 0.924 \text{ h}$ における各指標の計算**

* $C_A = 1.0 \times e^{-2.0 \times 0.924} \approx 0.1575 \text{ kmol/m}^3$
* $C_B = 1.0 \times \frac{2.0}{0.5 - 2.0} (e^{-1.848} - e^{-0.462}) \approx 0.630 \text{ kmol/m}^3$
* $C_C = 1.0 - 0.1575 - 0.630 = 0.2125 \text{ kmol/m}^3$

各指標を算定します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.1575}{1.0} = 0.8425 \quad (\mathbf{84.3 \%})$$


* **収率 $Y$:**
$$Y = \frac{C_B}{C_{A0}} = \frac{0.630}{1.0} = 0.630 \quad (\mathbf{63.0 \%})$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{C_B}{C_{A0} - C_A} = \frac{0.630}{0.8425} \approx 0.748 \quad (\mathbf{74.8 \%})$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.2125}{0.8425} \approx 0.252 \quad (\mathbf{25.2 \%}) \quad (\text{確認: } S_B + S_C = 100\%)$$



**(3) 反応時間を $t = 2.0 \text{ h}$ に延長した場合**

* $C_A(2.0) = e^{-4.0} \approx 0.0183 \text{ kmol/m}^3$
* $C_B(2.0) = -\frac{4}{3} (e^{-4.0} - e^{-1.0}) \approx 0.466 \text{ kmol/m}^3$
* $C_C(2.0) = 1.0 - 0.0183 - 0.466 = 0.5157 \text{ kmol/m}^3$

各指標を計算します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.0183}{1.0} \approx 0.982 \quad (\mathbf{98.2 \%}) \quad \text{【UP】}$$


* **収率 $Y$:**
$$Y = \frac{0.466}{1.0} \approx 0.466 \quad (\mathbf{46.6 \%}) \quad \text{【DOWN】}$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{C_B}{C_{A0} - C_A} = \frac{0.466}{0.982} \approx 0.475 \quad (\mathbf{47.5 \%}) \quad \text{【DOWN】}$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.5157}{0.982} \approx 0.525 \quad (\mathbf{52.5 \%}) \quad \text{【UP】}$$



> **結論:**
> 反応時間を延ばすと転化率 $X$ は約98%に上がりますが、その代わり副生物 $C$ の選択率 $S_C$ が50%を超えて急増し（$S_B$ は47.5%に激減）、結果として目的物 $B$ の収率 $Y$ も大きく低下します。
> このように、**「生成物の内訳を示す選択率 $S_{B/C}$」と「全体効率を示す収率 $Y$」の動きを切り分けて評価する**ことが、適切なプラント設計の基本となります。

ご指摘ありがとうございます。逐次反応（$A \xrightarrow{k_1} B \xrightarrow{k_2} C$）において、選択率は目的物 $B$ だけでなく、副生物 $C$ の選択率もあわせて「$B$ と $C$ の選択率」としてセットで評価することが実務上非常に重要です。

# 講義資料：逐次反応における最適反応時間（空間時間）の探索

〜「転化率・選択率・収率」の3つを同時に考えなければならない理由〜

## 1. 導入：「数学的等価性」と「工学的意義」のギャップ

化学反応工学の計算上、転化率 $X$、収率 $Y$、および目的物 $B$ の選択率 $S_B$ の間には、以下のマスター方程式が成り立ちます。

$$\text{収率 } (Y) = \text{転化率 } (X) \times \text{選択率 } (S_B)$$

数学的にはこのうち2つが分かれば残り1つは自動的に決まりますが、プラント設計の現場では**3つの異なる指標を常にセットで評価**します。

* **転化率 $X$（原料をどこまで使い切るか）：** 反応器の「大きさ（設備投資 CAPEX）」に直結する。
* **選択率 $S_B, S_C$（生成物の内訳）：** 反応した原料が「目的物 $B$」になったのか、「不要な副生物 $C$」になってしまったのかを示す。分離・精製コスト（OPEX）に直結する。
* **収率 $Y$（トータルでどれだけ目的物が得られたか）：** プラント全体の「経済性・利益率」に直結する。

---

## 2. 逐次反応モデルの定義と $X, S, Y$ の定式化

一次の逐次反応を考えます。

$$A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$$

* 初期条件 ($t = 0$): 濃度 $C_A = C_{A0}$、$C_B = 0$、$C_C = 0$
* 任意の時間 $t$ における各成分の濃度：$C_A(t), C_B(t), C_C(t)$

### ① 転化率 ($X$)

投入した $A$ のうち、反応した割合。


$$X(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### ② 収率 ($Y$)

投入した $A$ 全体に対して、目的物 $B$ が得られた割合。


$$Y(t) = \frac{C_B(t)}{C_{A0}}$$

### ③ $B$ と $C$ の選択率 ($S_B, S_C$)

反応した $A$ の総量（$= C_{A0} - C_A(t)$）のうち、それぞれがどこに変わったかの割合。

* **目的物 $B$ の選択率 ($S_B$):**

$$S_B(t) = \frac{\text{生成した } B \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_B(t)}{C_{A0} - C_A(t)} = \frac{Y(t)}{X(t)}$$


* **副生物 $C$ の選択率 ($S_C$):**

$$S_C(t) = \frac{\text{生成した } C \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_C(t)}{C_{A0} - C_A(t)}$$



> **💡 重要な関係式:**
> モル保存則（$C_{A0} = C_A + C_B + C_C$）より、反応した $A$ はすべて $B$ または $C$ になるため、**常に $S_B + S_C = 1$（または 100%）**が成り立ちます。つまり、$B$ の選択率が下がれば、その分だけ $C$ の選択率が必ず上がります。

---

## 3. 濃度変化の解析解と、指標の振る舞い

各成分の濃度変化は次のように求まります（$k_1 \neq k_2$ の場合）。

* $C_A(t) = C_{A0} e^{-k_1 t}$
* $C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$
* $C_C(t) = C_{A0} - C_A(t) - C_B(t)$

時間経過に伴う各指標の変化：

* **転化率 $X(t)$:** 時間とともに単調増加し、$1$（100%）に近づく。
* **選択率 $S_B(t)$ と $S_C(t)$:**
* 反応初期 ($t \approx 0$) は $S_B \approx 1$、$S_C \approx 0$ です。
* 時間が経つにつれて $B$ が $C$ に変わるため、**$S_B$ は単調に減少し、$S_C$ は単調に増加**します。


* **収率 $Y(t)$:** 「増える $X$」と「減る $S_B$」の積（$Y = X \times S_B$）であるため、**途中に必ず「山なりのピーク（極大値）」が現れます。**

---

## 4. 中間体 $B$ の最大収率を与える最適反応時間 $t_{\text{opt}}$ の導出

利益を最大化するためには、収率 $Y(t)$ が最大のところで反応を止めなければなりません。微分して $0$ と置くことで、最適反応時間 $t_{\text{opt}}$（またはPFRの最適空間時間 $\tau_{\text{opt}}$）が求まります。

$$t_{\text{opt}} = \frac{1}{k_1 - k_2} \ln\left(\frac{k_1}{k_2}\right)$$

> **⚠️ 工学的示唆:**
> この最適時点では、**転化率 $X$ は100%未満であり、副生物 $C$ の選択率 $S_C$ が一定割合（数％〜数十％）存在した状態で反応を打ち切る**ことになります。
> 「原料を完全に使い切りたい（転化率100%）」という発想を捨て、あえて未反応 $A$ をリサイクルし、不要な $C$ への変換ロス（選択率 $S_B$ の低下）を防ぐ設計がプラントの経済性を左右します。

## 5. 自習用演習問題と解説

### 【問題】

ある化学プラントで、液相の逐次反応 $A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$ が行われています。

* 反応速度定数：$k_1 = 2.0 \text{ h}^{-1}$、$k_2 = 0.5 \text{ h}^{-1}$
* 初濃度：$C_{A0} = 1.0 \text{ kmol/m}^3$

(1) 目的物 $B$ の収率 $Y$ が最大になる最適反応時間 $t_{\text{opt}}$ $[\text{h}]$ を求めよ。

(2) その最適時点における、**転化率 $X$、収率 $Y$、および $B$ と $C$ の選択率 ($S_B, S_C$)** の値をそれぞれ求めよ。

(3) もし誤って反応時間を $t = 2.0 \text{ h}$ まで延ばしてしまった場合、これらの指標はどう変化するか計算せよ。

---

### 【解答・解説】

**(1) 最適反応時間 $t_{\text{opt}}$ の計算**


$$t_{\text{opt}} = \frac{1}{2.0 - 0.5} \ln\left(\frac{2.0}{0.5}\right) = \frac{1}{1.5} \ln(4) \approx 0.924 \text{ h}$$

**(2) $t = 0.924 \text{ h}$ における各指標の計算**

* $C_A = 1.0 \times e^{-2.0 \times 0.924} \approx 0.1575 \text{ kmol/m}^3$
* $C_B = 1.0 \times \frac{2.0}{0.5 - 2.0} (e^{-1.848} - e^{-0.462}) \approx 0.630 \text{ kmol/m}^3$
* $C_C = 1.0 - 0.1575 - 0.630 = 0.2125 \text{ kmol/m}^3$

各指標を算定します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.1575}{1.0} = 0.8425 \quad (\mathbf{84.3 \%})$$


* **収率 $Y$:**
$$Y = \frac{C_B}{C_{A0}} = \frac{0.630}{1.0} = 0.630 \quad (\mathbf{63.0 \%})$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{Y}{X} = \frac{0.630}{0.8425} \approx 0.748 \quad (\mathbf{74.8 \%})$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.2125}{0.8425} \approx 0.252 \quad (\mathbf{25.2 \%}) \quad (\text{確認: } 74.8\% + 25.2\% = 100\%)$$



**(3) 反応時間を $t = 2.0 \text{ h}$ に延長した場合**

* $C_A(2.0) = e^{-4.0} \approx 0.0183 \text{ kmol/m}^3$
* $C_B(2.0) = -\frac{4}{3} (e^{-4.0} - e^{-1.0}) \approx 0.466 \text{ kmol/m}^3$
* $C_C(2.0) = 1.0 - 0.0183 - 0.466 = 0.5157 \text{ kmol/m}^3$

各指標を計算します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.0183}{1.0} \approx 0.982 \quad (\mathbf{98.2 \%}) \quad \text{【UP】}$$


* **収率 $Y$:**
$$Y = \frac{0.466}{1.0} \approx 0.466 \quad (\mathbf{46.6 \%}) \quad \text{【DOWN】}$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{Y}{X} = \frac{0.466}{0.982} \approx 0.475 \quad (\mathbf{47.5 \%}) \quad \text{【DOWN】}$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.5157}{0.9817} \approx 0.525 \quad (\mathbf{52.5 \%}) \quad \text{【UP】}$$



> **結論:**
> 反応時間を延ばすと転化率 $X$ は約98%に上がりますが、その代わり副生物 $C$ の選択率 $S_C$ が50%を超えて急増し（$S_B$ は47.5%に激減）、結果として目的物 $B$ の収率 $Y$ も大きく低下します。
> このように、**「$B$ と $C$ の選択率がどのようにトレードオフになっているか」**を追跡することが、プラント設計において極めて重要になります。
