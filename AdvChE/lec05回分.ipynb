## 1. 回分反応器
流入・流出がない反応器である。

$$\frac{dN_j}{dt}=r_jV$$

一定体積なら

$$\frac{dC_j}{dt}=r_j$$

## 2. ベクトル形式

$$\frac{d\mathbf C}{dt}=\mathbf S\mathbf r(\mathbf C)$$

量論行列 S と速度ベクトル r を用いて表現できる。

## 3. 逐次反応

$$A
\rightarrow B
\rightarrow C$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

## 4. 支配方程式

$$\frac{dC_A}{dt}=-k_1C_A$$
$$\frac{dC_B}{dt}=k_1C_A-k_2C_B$$
$$\frac{dC_C}{dt}=k_2C_B$$

## 5. 選択率
中間生成物 B を目的生成物とする場合、適切な反応時間で停止する必要がある。

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt

k1=1.0
k2=0.3

def f(t,y):
    CA,CB,CC=y
    return [-k1*CA, k1*CA-k2*CB, k2*CB]

sol=solve_ivp(f,[0,10],[1,0,0],dense_output=True)
t=np.linspace(0,10,400)
CA,CB,CC=sol.sol(t)

plt.plot(t,CA,label="A")
plt.plot(t,CB,label="B")
plt.plot(t,CC,label="C")
plt.legend();plt.grid();plt.show()
```

## 6. B最大条件

$$\frac{dC_B}{dt}=0$$

$$k_1C_A=k_2C_B$$

## 7. 並列反応

$$A
\rightarrow B$$
$$A
\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 8. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C}$$

$$S_{B/C}=\frac{k_B}{k_C}C_A^{m-n}$$

```python
import numpy as np
import matplotlib.pyplot as plt
CA=np.linspace(0.05,2,200)
S=1.5*CA
plt.plot(CA,S)
plt.xlabel("C_A")
plt.ylabel("S_B/C")
plt.grid()
plt.show()
```

## 9. 行列表現の例
逐次反応の量論行列


$$
S=
\begin{bmatrix}
-1 & 0\\
1 & -1\\
0 & 1
\end{bmatrix}
$$

速度ベクトル

$$
r=
\begin{bmatrix}
k_1C_A\\
k_2C_B
\end{bmatrix}
$$


```python
import numpy as np
S=np.array([[-1,0],[1,-1],[0,1]])
r=np.array([2.0,0.5])
print(S@r)
```
