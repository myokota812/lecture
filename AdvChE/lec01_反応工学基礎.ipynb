<h3>(Continuous Stirred Tank Reactor)

<figure style="text-align:center;">
<img

width="600">
</figure>

<details>
<summary>📝🖱️説明文はこちら</summary>

ここから、連続攪拌槽反応器、CSTRについて復習します。
CSTRは Continuous Stirred Tank Reactor の略で、日本語では連続攪拌槽反応器と呼ばれます。

まず個別の設計方程式を導出する前に、すべての反応器設計の出発点となる一般的な物質収支式を確認しましょう。

図の左側から成分 $i$ が流量 $F_{i0}$​ で流入し、右側へ $F_i$で流出しています。

一方、反応器内部では化学反応が起こるため、成分 $i$ が生成されたり消費されたりします。この反応による寄与が $r_iV$ です。

また、反応器内の成分量が時間とともに変化する場合、その変化速度が蓄積項 $dN_i/dt$ となります。

したがって、物質収支の基本形は
$$
\text{Accumulation} = \text{In} - \text{Out} + \text{Generation} - \text{Consumption}
$$
となります。

つまり、

「流入してくる量」から「流出していく量」を引き、さらに「反応による生成や消費」を加味したものが、反応器内に蓄積する量になるということです。

この式そのものは、CSTRだけではなく、回分反応器、PFR、さらには複雑な反応プロセスにも共通して適用できる、反応工学の最も重要な基本式です。

👉 この一般式に対して、CSTR特有の仮定である「完全混合」と「定常状態」を導入すると、皆さんがよく知っているCSTRの設計方程式が導出されます。

次のスライドでは、その導出過程を順を追って見ていきましょう。

【説明ここまで】
</details>

対象成分：$i = \mathrm{A}$の場合

<figure style="text-align:center;">
  <img
    src="https://raw.githubusercontent.com/myokota812/lecture/master/AdvChE/figure/cstr2.svg"
    width="500">
    <figcaption>
     図1. 連続攪拌槽反応器（CSTR）の模式図
    </figcaption>
</figure>

- $N_A$ ：反応器内の成分Aの物質量 [mol]
- $F_{A0}$ ：成分Aの流入モル流量 [mol/s]
- $F_A$ ：成分Aの流出モル流量 [mol/s]
- $r_A$ ：成分Aの反応速度 [mol/(m³·s)]生成物では正値,反応物では負値
- $V$ ：反応器体積 [m³]

◇◇◇

物質収支式

$$
\text{蓄積} = \text{流入} − \text{流出} + \text{反応による生成（または消費）}
$$
から
$$
\frac{dN_A}{dt}=F_{A0}-F_A+r_AV
$$

物理的意味

$$
\underbrace{\frac{dN_A}{dt}}_{\text{蓄積}} = \underbrace{F_{A0}}_{\text{流入}} - \underbrace{F_A}_{\text{流出}} + \underbrace{r_AV}_{\text{反応}}
$$


CSTRの定常状態では

定常状態では反応器内に物質は蓄積しないため、

$$
\frac{dN_A}{dt}=0
$$
となり、

$$
F_{A0}-F_A+r_AV=0
$$
あるいは

$$
F_{A0}-F_A=-r_AV
$$

したがって


$$V=\frac{F_{A0}-F_A}{-r_A}$$


がCSTRの基本設計方程式となる。

もしくは転化率$X$を用いる場合

$$X=\frac{F_{A0}-F_A}{F_{A0}}$$

より

$$V=\frac{F_{A0}X}{-r_A} \tag{1}$$

>図１だけをみて、自力で物質収支式を立式し、Eq.(1)まで導出できるようにしてください。

説明セリフ例（PFRの設計方程式導出）

ここでは、管型反応器であるPFR、Plug Flow Reactor の設計方程式を導出します。

CSTRの場合は反応器全体を1つの制御体積として扱いましたが、PFRでは少し考え方が異なります。

PFRでは流体が管内を押し出し流れ、軸方向に濃度や反応速度が変化するため、反応器全体を一度に扱うことができません。

そこで図に示すように、反応器内部の微小な体積要素 $dV$ を取り出して物質収支を考えます。

成分 $j$ のモル流量が $F_j$ で流入し、この微小区間で反応が進行することによって、流出時には $F_j+dF_j$ に変化します。

この微小体積内で成分 $j$ は生成あるいは消費されるため、その変化量が $dF_j$ として表れています。


</details>

---

定常状態を仮定すると蓄積項はゼロになりますので、

$$
\text{流入} - \text{流出} + \text{反応} = 0
$$
となります。

これを微小体積 $dV$ について書くと、

$$
F_j-(F_j+dF_j)+r_j\,dV=0
$$
が得られます。

整理すると、

$$
dF_j=r_j\,dV
$$
あるいは
$$
\frac{dF_j}{dV}=r_j
$$
となります。

これがPFRの微分形式の設計方程式です。

>👉 CSTRでは代数方程式が得られましたが、PFRでは反応器内部で濃度が連続的に変化するため、設計方程式は微分方程式になる点が重要です。この違いが、同じ転化率を達成する場合にPFRの方が一般に小さな反応器体積で済む理由につながります。




先程のCSTR同様成分$j$を$\mathrm{A}$に置き換えます。

$$\frac{dF_A}{dV}=r_A$$

したがって入口 $(V=0,\;F_A=F_{A0})$ から出口 $(V=V,\;F_A=F_{Af})$ まで積分して
$$
V=\int_{F_{A0}}^{F_{Af}}\frac{dF_A}{r_A}.
$$


通常は反応物 A について $r_A<0$ なので、

$$
V=\int_{F_{Af}}^{F_{A0}}\frac{dF_A}{-r_A}
$$

と書くことが多いです。

◇◇◇

転化率を導入すると

$$F_A=F_{A0}(1-X)$$

$$dF_A=-F_{A0}dX$$

したがって

$$dV=\frac{F_{A0}}{-r_A}dX$$

積分して

$$V=F_{A0}\int_0^X \frac{dX}{-r_A} \tag{2}$$

---

><h3><font color="Red">課題2</font></h3>

>図2だけをみて、自力で物質収支式を立式し、Eq.(2)まで導出できるようにしてください。ただし、図の記号の添字は$j$は最初から$A$に置き換えて結構です。

## CSTRとPFRの比較

- <h4> CSTR (Eq.1):</h4>

$$V=\frac{F_{A0}X}{-r_A}$$

- <h4>PFR(Eq. 2):</h4>

$$V=F_{A0}\int_0^X \frac{dX}{-r_A}$$



<details>
<summary>📝🖱️説明文はこちら</summary>

それでは、CSTRとPFRの設計方程式を並べて比較してみましょう。

まずCSTRの設計方程式は、Eq.(1)でした。この式の特徴は、反応器出口の反応速度 $r_A$そのものが装置体積を決定することです。

CSTRでは完全混合が成立するため、反応器内のどの場所でも濃度が同じになります。したがって反応速度も全体で一定となり、このような単純な代数式で表現できます。

一方、PFRの設計方程式はEq.(2)のように積分となります。

PFRでは流体が反応器内を移動するにつれて反応が進行するため、濃度も反応速度も位置によって変化します。

そのため、反応器全体を小さな区間に分割し、それぞれの区間の寄与を積み上げる必要があります。その結果、このような積分形式の設計方程式になります。
</details>

## 一次反応への適用

$$A\rightarrow B$$

$$-r_A=kC_A$$

$$C_A=C_{A0}(1-X)$$

- #### CSTR

$$
V=\frac{F_{A0}X}{kC_{A0}(1-X)} \tag{A1}
$$

- #### PFR

$$
V=\frac{F_{A0}}{kC_{A0}}\ln\left(\frac{1}{1-X}\right) \tag{A2}
$$

◇◇◇

>上記の一次反応における$-r_A=kC_A$をEq.(1)とEq.(2)に代入するとそれぞれEq.(A1),Eq.(A2)となることを確認してください。

## 空間時間

空間時間$\tau$は
$$
\tau = ~\frac{V}{v}
$$
で定義される、時間の次元を持つ量である。

ここで、$v\ [\mathrm{m}^3 \cdot \mathrm{s}^{-1}]$は溶液の体積流量。 
$$
F_{A0} = C_{A0} v
$$
だから

先程の設計方程式Eqs. (A1),(A2)を、空間時間に置き換えると

- <h4>CSTR:</h4>

$$\tau=\frac{X}{k(1-X)}$$

- <h4>PFR:</h4>

$$\tau=\frac{1}{k}\ln\left(\frac{1}{1-X}\right)$$



◇◇◇
<h3>（発展）</h3>

反応速度がn次反応の場合

$$
-r_A = kC_A^n
$$
で、液相・等密度系
$$
C_A=C_{A0}(1-X)
$$
を仮定すると、それぞれの設計方程式は次のようになります。


<h5>CSTR</h5>

$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \frac{X}{(1-X)^n} }
$$

<h5>PFR</h5>

$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$


<details>
<summary>📝🖱️参考（導出）</summary>

<h5> 1. CSTR</h5>

設計方程式
$$
V=\frac{F_{A0}X}{-r_A}
$$
に代入すると
$$
V = \frac{F_{A0}X} {k[C_{A0}(1-X)]^n}
$$
したがって
$$
\boxed{ V= \frac{F_{A0}}{kC_{A0}^{\,n}} \frac{X}{(1-X)^n} }
$$
となります。

空間時間

$$
\tau=\frac{V}{v_0}, \qquad F_{A0}=v_0C_{A0}
$$
を用いると
$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \frac{X}{(1-X)^n} }
$$

<h5>2. PFR</h5>

設計方程式
$$
V = F_{A0} \int_0^X \frac{dX}{-r_A}
$$
へ代入すると
$$
V = \frac{F_{A0}}{kC_{A0}^{\,n}} \int_0^X \frac{dX}{(1-X)^n}
$$
となります。

(1) $n\neq1$

積分すると
$$
\int (1-X)^{-n}dX = \frac{(1-X)^{\,1-n}}{n-1}
$$
したがって
$$
V = \frac{F_{A0}} {kC_{A0}^{\,n}} \left[ \frac{(1-X)^{1-n}-1} {n-1} \right]
$$
よって
$$
\boxed{ V= \frac{F_{A0}} {kC_{A0}^{\,n}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$
空間時間表示では
$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$

(2) $n=1$のときは先程と同じ




## 演習問題

1. 一次反応($A \rightarrow B \,\left(-r_A=kC_A \right) $  )で $X=0.8$ を達成するためのCSTRとPFRの空間時間を求めよ。求めた２つの空間時間を比較してCSTRとPFRのどちらが優れているか論じなさい。ただし、同じ$v$で操作した場合を前提にします。

2. 下記のpythonコードは、一次反応について、ＰＦＲとＣＴＲを比較します。出力されるグラフを見てそこから得られる知見を記述しなさい。


```python
#@title演習２のpythonコード
import numpy as np
import matplotlib.pyplot as plt

k = 1.0
CA0 = 1.0

X = np.linspace(0.001, 0.95, 500)

def tau_pfr(X, n, k=1.0, CA0=1.0):

    if np.isclose(n, 0):
        return X / (k / CA0)

    elif np.isclose(n, 1):
        return np.log(1/(1-X)) / k

    else:
        return ((1-X)**(1-n)-1)/(
            (n-1)*k*CA0**(n-1)
        )

def tau_cstr(X, n, k=1.0, CA0=1.0):

    if np.isclose(n, 0):
        return X/(k/CA0)

    return X/(
        k*CA0**(n-1)*(1-X)**n
    )

orders = [-1, 0,1,2]

fig, axes = plt.subplots(
    2, 2,
    figsize=(10,8),
    sharex=True
)

axes = axes.flatten()

for ax, n in zip(axes, orders):

    ax.plot(
        X,
        tau_pfr(X,n,k,CA0),
        lw=3,
        label="PFR"
    )

    ax.plot(
        X,
        tau_cstr(X,n,k,CA0),
        "--",
        lw=3,
        label="CSTR"
    )

    ax.set_title(f"n={n}")
    ax.grid(True)
    ax.legend()

    ax.set_xlabel("Conversion X")
    ax.set_ylabel(r"$\tau$")

plt.tight_layout()
plt.show()
```
