## 提出課題2

## 1. 問題設定

定常状態の管型反応器（PFR）において、次の液相の逐次反応が進行している。

$$2A\xrightarrow{r_1}B\xrightarrow{r_2}3C$$

各反応の速度式は以下のように与えられる。

$$r_1 = k_1 C_A^2,\qquad r_2 = k_2 C_B$$

なお、液相反応とし、反応器内の体積流量は一定とする。

| 項目 | 値 |
| --- | --- |
| Aの入口モル流量 $F_{A0}$ | $10\ \mathrm{mol/min}$ |
| Bの入口モル流量 $F_{B0}$ | $0\ \mathrm{mol/min}$ |
| Cの入口モル流量 $F_{C0}$ | $0\ \mathrm{mol/min}$ |
| 入口体積流量 $v$ | $2\ \mathrm{L/min}$ |
| **反応器容積 $V$**（空間時間 $\tau = 2.0\ \mathrm{min}$） | **$4.0\ \mathrm{L}$** |
| 反応速度定数 $k_1$ | $0.10\ \mathrm{L/(mol\cdot min)}$ |
| 反応速度定数 $k_2$ | $0.20\ \mathrm{min^{-1}}$ |

---

## 2. 設問

**(1)**
反応式数が$M$の複合反応をPFRで実施する場合、各成分$i$の物質収支から導出される設計方程式の一般式は次の通りである。

$$\frac{dF_i}{dV} = \sum_{j=1}^{M} \nu_{ij} r_j \quad (i = \text{A, B, C, } \dots)$$


本問題の場合は$M=2$.したがって$j=1,2$, $i = A, B, C$である。
- $F_i=C_i \cdot v$
- $\tau =V/v$
を用いて次の微分方程式を導きなさい。

$$\frac{dC_A}{d\tau} = -2 k_1 C_A^2 \tag{A}$$
$$\frac{dC_B}{d\tau} = k_1 C_A^2 - k_2 C_B \tag{B}$$
$$\frac{dC_C}{d\tau} = 3 k_2 C_B \tag{C}$$

**(2)**
微分方程式(A)式を解くことで、反応器出口におけるAのモル流量 $F_A$ を求め、さらに転化率 $X_A$ を求めなさい。


**(3)**
微分方程式(B)式（1階線形微分方程式）は初等関数の積分では解けないため、この微分方程式を数値解法することにします。下記にPythonコードを実行すると、$C_B$ vs $\tau$のグラフ、$C_B(\tau)$, $F_B(\tau)$が出力されます。その $F_B$ を使い、 $Y_B$ を求めなさい。<br>
参考までに、Pythonコードの中で微分方程式を定義している行は
~~~
# 微分方程式
def dCB_dtau(tau, CB):
    return (
        k1 * CA0**2 / (1 + 2 * k1 * CA0 * tau)**2
        - k2 * CB[0]
    )
~~~

です。これは
$$\frac{dC_{\text{B}}}{d\tau}  = \frac{k_1 C_{\text{A0}}^2}{(1 + 2 k_1 C_{\text{A0}} \tau)^2} - k_2 C_{\text{B}}$$
を記述しております。


**(4)**
反応器出口におけるCのモル流量 $F_C$ を求め、BのCに対する選択率 $S_{B/C}$ を求めなさい。


**(5) 考察**
本反応器をワンパス（1回通し）で運転する場合、中間生成物である B の選択率を高めようとして空間時間 $\tau$ を短くすると、未反応の原料 A が大量に残留してしまう。しかし、原料 A が無駄になっては経済的に成り立たない。この課題を克服し、「高い選択率（Bの確保）」と「原料の有効利用（原料ロスゼロ）」を両立させるために、反応器システム全体のプロセスフロー（配管や分離工程の配置）において、どのような工夫（改善策）が考えられるか。化学工学的な観点から自分のアイデアを提案し、そのメリットを説明しなさい。

```python
#@title微分方程式(B)の数値解法 
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# パラメータ
k1 = 0.10       # L/(mol min)
k2 = 0.20       # 1/min
CA0 = 5.0       # mol/L

# 微分方程式
def dCB_dtau(tau, CB):
    return (
        k1 * CA0**2 / (1 + 2 * k1 * CA0 * tau)**2
        - k2 * CB[0]
    )

# 初期条件と計算区間
tau_span = (0.0, 2.0)
CB0 = [0.0]     # 入口ではBが存在しない

# 数値解法
tau_eval = np.linspace(0.0, 2.0, 201)

sol = solve_ivp(
    dCB_dtau,
    tau_span,
    CB0,
    t_eval=tau_eval,
    method="RK45",
    rtol=1e-10,
    atol=1e-12
)

# 出口濃度
CB_out = sol.y[0, -1]

print(f"出口B濃度 CB = {CB_out:.6f} mol/L")
print(f"出口Bモル流量 FB = {CB_out * 2.0:.6f} mol/min")

# グラフ
plt.figure(figsize=(8, 5))
plt.plot(sol.t, sol.y[0], linewidth=2, label="$C_B$")
plt.xlabel("Space time $\\tau$ (min)")
plt.ylabel("$C_B$ (mol/L)")
plt.title("Numerical solution of the PFR equation")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

## 解答例

### (1) 微分方程式の導出

**【解答】**

問題の一般式において、$M=2$（反応1と反応2）です。
各成分の化学量論係数 $\nu_{ij}$（成分 $i$ の反応 $j$ における量論係数、生成物は正、反応物は負）は以下の通りです。

* 反応1 ($2\text{A} \xrightarrow{k_1} \text{B}$): $\nu_{\text{A}1} = -2$, $\nu_{\text{B}1} = +1$, $\nu_{\text{C}1} = 0$
* 反応2 ($\text{B} \xrightarrow{k_2} 3\text{C}$): $\nu_{\text{A}2} = 0$, $\nu_{\text{B}2} = -1$, $\nu_{\text{C}2} = +3$

一般式 $\frac{dF_i}{dV} = \sum_{j=1}^{2} \nu_{ij} r_j$ に代入すると、各成分のモル収支式は次のように記述されます。

$$\frac{dF_{\text{A}}}{dV} = -2 r_1 = -2 k_1 C_{\text{A}}^2$$

$$\frac{dF_{\text{B}}}{dV} = r_1 - r_2 = k_1 C_{\text{A}}^2 - k_2 C_{\text{B}}$$

$$\frac{dF_{\text{C}}}{dV} = 3 r_2 = 3 k_2 C_{\text{B}}$$

ここで、体積流量 $v$ が一定であるため $F_i = C_i \cdot v$ であり、また空間時間 $\tau = V / v$ より $dV = v \cdot d\tau$ と変形できます。
これらを各モル収支式に代入すると、$v$ が約分され、求める微分方程式が導出されます。

$$\frac{dC_{\text{A}}}{d\tau} = -2 k_1 C_{\text{A}}^2$$

$$\frac{dC_{\text{B}}}{d\tau} = k_1 C_{\text{A}}^2 - k_2 C_{\text{B}}$$

$$\frac{dC_{\text{C}}}{d\tau} = 3 k_2 C_{\text{B}}$$

---

### (2) Aのモル流量 $F_{\text{A}}$ および転化率 $X_{\text{A}}$

**【解答】**

1. **入口濃度の算出:**

$$C_{\text{A0}} = \frac{F_{\text{A0}}}{v} = \frac{10\ \text{mol/min}}{2\ \text{L/min}} = 5.0\ \text{mol/L}$$


2. **Aの微分方程式の解法（変数分離法）:**

$$\frac{dC_{\text{A}}}{C_{\text{A}}^2} = -2 k_1 d\tau$$



両辺を積分します（$\tau=0$ で $C_{\text{A}}=C_{\text{A0}}$）。

$$-\frac{1}{C_{\text{A}}} + \frac{1}{C_{\text{A0}}} = -2 k_1 \tau \implies C_{\text{A}}(\tau) = \frac{C_{\text{A0}}}{1 + 2 k_1 C_{\text{A0}} \tau}$$


3. **出口濃度・モル流量の計算:**
$\tau = 2.0\ \text{min}$, $k_1 = 0.10\ \text{L/(mol}\cdot\text{min)}$, $C_{\text{A0}} = 5.0\ \text{mol/L}$ を代入します。

$$C_{\text{A}} = \frac{5.0}{1 + 2 \times 0.10 \times 5.0 \times 2.0} = \frac{5.0}{1 + 2.0} = \frac{5.0}{3.0} \approx 1.67\ \text{mol/L}$$


$$F_{\text{A}} = C_{\text{A}} \cdot v = \frac{5}{3}\ \text{mol/L} \times 2\ \text{L/min} = \mathbf{3.33\ \text{mol/min}} \quad \left(\frac{10}{3}\ \text{mol/min}\right)$$


4. **転化率 $X_{\text{A}}$ の算出:**

$$X_{\text{A}} = \frac{F_{\text{A0}} - F_{\text{A}}}{F_{\text{A0}}} = \frac{10 - \frac{10}{3}}{10} = \mathbf{0.667 \quad (66.7\%)}$$



---

### (3) Bのモル流量 $F_{\text{B}}$ および収率 $Y_{\text{B}}$

**【解答】**

1. **1階線形微分方程式の整理:**
(2)より $C_{\text{A}}(\tau) = \frac{C_{\text{A0}}}{1 + 2 k_1 C_{\text{A0}} \tau}$ をBの微分方程式に代入すると、

$$\frac{dC_{\text{B}}}{d\tau} + k_2 C_{\text{B}} = \frac{k_1 C_{\text{A0}}^2}{(1 + 2 k_1 C_{\text{A0}} \tau)^2}$$


2. **解:**

与えられた各パラメータ（$k_1=0.10$, $k_2=0.20$, $C_{\text{A0}}=5.0$）において数値計算すると

~~~
出口B濃度 CB = 1.279223 mol/L
出口Bモル流量 FB = 2.558445 mol/min
~~~

$$Y_{\text{B}} = \frac{2F_{\text{B}}}{F_{\text{A0}}} = \frac{5.12}{10.0} = \mathbf{0.512 \quad (51.2\%)}$$

### (4) Cのモル流量 $F_{\text{C}}$ および選択率 $S_{\text{B/C}}$

**【解答】**


* 反応1：$2\text{A} \to \text{B}$ （進行度を$\xi_1$とおく）
* 反応2：$\text{B} \to 3\text{C}$ （進行度を$\xi_2$とおく）

そうすると各成分のモル流量は、空間時間 $\tau$ を変数とした反応進行度 $\xi_1, \xi_2$ を用いて以下のように表せます：

* $F_A = F_{A0} - 2\xi_1 \quad(1)$
* $F_B = F_{B0} + \xi_1 - \xi_2 \quad (\text{いま } F_{B0} = 0)\quad(2) $
* $F_C = F_{C0} + 3\xi_2 \quad (\text{いま } F_{C0} = 0) \quad(3)$


(1)式より $\xi_1 = \frac{F_{A0} - F_A}{2}$ です。
これを $F_B$ の式に代入すると：


$$F_B = \frac{F_{A0} - F_A}{2} - \xi_2$$


さらに、 $F_C = 3\xi_2$ より $\xi_2 = \frac{F_C}{3}$ なので、代入すると：


$$F_B = \frac{F_{A0} - F_A}{2} - \frac{F_C}{3}$$

#### 【この式を使った $F_C$ の計算】

* $F_{A0} = 10$
* $F_A = \frac{10}{3} \approx 3.333$
* $F_B = 2.558$ （与えられた値）

これらを代入：


$$2.558 = \frac{10 - 3.333}{2} - \frac{F_C}{3}$$

$$2.558 = 3.3335 - \frac{F_C}{3}$$

$$\frac{F_C}{3} = 3.3335 - 2.558 = 0.7755$$

$$F_C = 0.7755 \times 3 = \mathbf{2.3265\ \mathrm{mol/min}}$$



---



**【別解】**

1. **モル収支（元素収支）による C のモル流量 $F_{\text{C}}$ の算出:** <br>
A原子（単量体単位）の保存則より：

$$F_{\text{A0}} = F_{\text{A}} + 2 F_{\text{B}} + \frac{2}{3} F_{\text{C}}$$



これを $F_{\text{C}}$ について解きます。

$$F_{\text{C}} = \frac{3}{2} \left( F_{\text{A0}} - F_{\text{A}} - 2 F_{\text{B}} \right)$$



数値を代入します（$F_{\text{A0}}=10$, $F_{\text{A}}=3.33$, $F_{\text{B}}=2.56$）。

$$F_{\text{C}} = 1.5 \times (10 - 3.33 - 2.56*2)  = \mathbf{\ 2.325\text{mol/min}}$$


2. **選択率 $S_{\text{B/C}}$ の算出:**
BのCに対するモル選択率は、出口における両成分のモル流量の比として定義されます。

$$S_{\text{B/C}} = \frac{F_{\text{B}}}{F_{\text{C}}} = \frac{2.55}{2.325} \approx \mathbf{1.07}$$

### (5) 考察

改善策の提案：
反応器の出口に未反応の原料 A を回収・分離するプロセス（蒸留塔や膜分離など）を設置し、回収した A を全量（あるいは一部パージを除いて）反応器の入口へリサイクル（循環）するシステムを構築する。

メリット（理由）：
これによって、反応器単体の空間時間 $\tau$ はあえて短め（高選択率が得られる条件）に設定して運転しつつ、全体としての原料 A の最終的な消失率（実質的な限界転化率）を 100% に近づけることができる。結果として、「高い B の選択率」と「原料の無駄のない完全消費」を同時に達成できる。
