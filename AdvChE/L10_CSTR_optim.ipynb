### CSTR（等温）の最適化問題

### 逐次反応（ A\->D\->U）＋並行反応( 2A\->P)の設計式（復習）

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$\begin{array}{l} A\;\overset{k_1 }{\to} \;D\;\overset{k_2 }{\to} \;U\newline A+A\;\overset{k_3 }{\to} \;P \end{array}$$

定常状態における成分Aの物質収支式は

&#160;&#160;&#160;&#160;
$$C_{\textrm{A0}} -C_{\textrm{A}} +\tau r_{\textrm{A}} =0$$

ここで

&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{A}} =-k_1 C_{\textrm{A}} -2k_3 C_{\textrm{A}}^2$$

を代入して整理すると

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$2\tau k_3 C_{\textrm{A}}^2 +(1+k_1 \tau )C_{\textrm{A}} -C_{\textrm{A0}} =0$$

の２次方程式となり、解の公式から $C_{\textrm{A}}$ は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_{\textrm{A}} =\frac{-(1+k_1 \tau ).+\sqrt{(1+k_1 \tau )^2 +8k_3 \tau C_{\textrm{A0}} }}{4k_3 \tau }$  \-\-\- Eq. (1)

一方、成分Dの物質収支式は

&#160;&#160;&#160;&#160;
$$C_{\textrm{D0}} -C_{\textrm{D}} +\tau r_{\textrm{D}} =0$$

 で、 $r_{\textrm{D}}$ は

&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{D}} =k_1 C_{\textrm{A}} -k_2 \textrm{D}$$

これを整理すると

&#160;&#160;&#160;&#160;&#160; $C_{\textrm{D}} =\frac{k_1 \tau }{1+k_2 \tau }C_{\textrm{A}}$   \-\-\-\-  Eq. (2)

成分Uの物質収支式は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_{\textrm{U0}} -C_{\textrm{U}} +\tau r_{\textrm{U}} =0$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\textrm{U}} =k_2 C_{\textrm{D}}$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_{\mathrm{U}} =\tau k_2 C_{\mathrm{D}} =\frac{k_1 k_2 \tau^2 }{1+k_2 \tau }C_{\textrm{A}}$ \-\-\-  Eq. (3)

成分Pの物質収支式は

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$C_{\textrm{P0}} -C_{\textrm{P}} +\tau r_{\textrm{P}} =0$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160;
$$r_{\mathrm{P}} =k_3 C_{\mathrm{A}}^2$$

&#160;&#160;&#160;&#160;&#160;&#160;&#160;&#160; $C_{\mathrm{P}} =\tau k_3 C_{\mathrm{A}}^2$  \-\-\- Eq. (4)

前回の課題では max{ $C_{\mathrm{D}}$ }を考え目的関数としました。当然他の成分も反応によって、消費もしくは生成されますので、プロセス開発の立場からは更に他の因子も考える必要があります。

　ここで、Dを目的生成物、Pは不用な副生成物という前提のもとで、

- $C_{\mathrm{D}}$ を最大
- $C_{\mathrm{P}}$ を最小
- Dの選択性 $S_{\mathrm{D}} =\frac{C_{\mathrm{D}} }{C_{\mathrm{D}} +C_{\mathrm{U}} +C_{\mathrm{P}} }$ を最大
- Dの収率 $Y_{\mathrm{D}} =\frac{C_{\mathrm{D}} }{C_{\mathrm{A0}} }$ 　を最大

この４項目の要求をできるだけ満たす最適値（この場合は滞留時間 $\tau$ )を探索するということにします。 (複雑になるのでUは省略します）。

この要求を満たす目的関数を用意し、最適化をしてやればよいことになります。

そのためには、まず要求を定量的な式に表し（目的関数）それを、最大もしくは最小化する手法が考えられます。MATLABでは最適化コマンドが色々と用意されておりますが、今回は、fminconを使います。

のコマンドの各項目の意味を完結に示してください。（記入済みのところは解答例です。空欄のところを記入してください。

それでは、冒頭でしました、複合反応について、

逐次反応と並列反応のCSTRにおいて滞留時間 tau を1変数の設計変数として、以下の4指標を重み付きの単一目的（phi）にまとめて最適化します。

- CD（Dの濃度）        → 大きいほど良い（最大化）
- CP（Pの濃度：副生成物）→ 小さいほど良い（最小化）
- SD（選択性：組成基準） → 大きいほど良い（最大化）
- YD（収率）            → 大きいほど良い（最大化）

#### phi(tau) = \-wD\*CD + wP\*CP \- wS\*SD \- wY\*YD

fminconは最小化をしますので、最大化させる指標にはマイナス記号を付与します。また、それぞの指標には重みをつけております。下記のコードを実行してください。


```matlab
clc; clear;


%% 1) 問題設定（反応・運転パラメータ／探索範囲／重み）
CA0 = 1.0;    % [mol/L] 入口 A 濃度
k1  = 1.0;    % [1/h]   A -> D
k2  = 0.5;    % [1/h]   D -> U
k3  = 0.05;   % [L/mol/h] 2A -> P


TAU_MIN = 1e-3;     % tau 下限 [h]
TAU_MAX = 20.0;   % tau 上限 [h]（デモ用。現実的な範囲に調整推奨）


% ---- 重み（初期例）----
wD = 0.30;   % CD を重視（大きいほど CD 優先）
wP = 0.30;   % CP を抑制（大きいほど P 抑制を強化）
wS = 0.25;   % 選択性 SD を重視
wY = 0.15;   % 収率 YD を重視
% ※ 合計が 1 である必要はありません（相対比のため）。
%    ただし桁が極端に違うとスケールの影響が出るため、概ね同程度で調整すると良いです。


%% 2) 目的関数ハンドル
obj = @(tau) objectiveWeightedSumMulti(tau, CA0, k1, k2, k3, wD, wP, wS, wY);


%% 3) 境界・初期値・オプション
%% local minimum探索のため初期値は多点設定しました
lb = TAU_MIN; ub = TAU_MAX;
x0_list = [0.05, 0.2, 0.5, 1.0, 2.0, 5.0, 10.0];  % multi-start 初期点
opts = optimoptions('fmincon', 'Algorithm','interior-point', 'Display','off');


% ベスト解の格納用
best_tau = NaN; best_phi = Inf;
best_CD  = NaN; best_CP = NaN; best_SD = NaN; best_YD = NaN;


%% 4) 最適化（multi-start）
for x0 = x0_list
    [tau_star, phi_star] = fmincon(obj, x0, [],[],[],[], lb, ub, [], opts);
    % 指標を取得
    [~, CD, CP, SD, YD] = objectiveWeightedSumMulti(tau_star, CA0, k1, k2, k3, wD, wP, wS, wY);
    % ベスト更新
    if phi_star < best_phi
        best_phi = phi_star; best_tau = tau_star;
        best_CD = CD; best_CP = CP; best_SD = SD; best_YD = YD;
    end
end


%% 5) 結果表示
fprintf('Weighted-sum optimum (multi-metric):\n');
fprintf('  tau* = %.4f h\n', best_tau);
fprintf('  CD   = %.4f\n',   best_CD);
fprintf('  CP   = %.4f\n',   best_CP);
fprintf('  SD   = %.4f\n',   best_SD);
fprintf('  YD   = %.4f\n',   best_YD);
fprintf('  weights: wD=%.2f, wP=%.2f, wS=%.2f, wY=%.2f\n', wD, wP, wS, wY);


%% 6) （任意）プロット：phi, CD, CP, SD, YD の形を確認
doPlot = true;
if doPlot
    N = 400;
    taus = linspace(TAU_MIN, TAU_MAX, N);
    phi = zeros(1,N); CD = phi; CP = phi; SD = phi; YD = phi;


    for i = 1:N
        [phi(i), CD(i), CP(i), SD(i), YD(i)] = objectiveWeightedSumMulti(taus(i), CA0, k1, k2, k3, wD, wP, wS, wY);
    end


    figure('Color','w','Position',[50 80 1280 640]);
    tiledlayout(2,3,'TileSpacing','compact','Padding','compact');


    nexttile; plot(taus, phi, 'k','LineWidth',1.6); hold on;
    plot(best_tau, best_phi, 'ks','MarkerFaceColor','y','MarkerSize',8);
    grid on; xlabel('tau [h]'); ylabel('phi = -wD*CD + wP*CP - wS*SD - wY*YD');
    title('Objective (weighted sum)');


    nexttile; plot(taus, CD, 'b','LineWidth',1.6); hold on;
    plot(best_tau, best_CD, 'bo','MarkerFaceColor','b','MarkerSize',8);
    grid on; xlabel('tau [h]'); ylabel('C_D [mol/L]'); title('C_D(tau)');


    nexttile; plot(taus, CP, 'r','LineWidth',1.6); hold on;
    plot(best_tau, best_CP, 'ro','MarkerFaceColor','r','MarkerSize',8);
    grid on; xlabel('tau [h]'); ylabel('C_P [mol/L]'); title('C_P(tau)');


    nexttile; plot(taus, SD, 'm','LineWidth',1.6); hold on;
    plot(best_tau, best_SD, 'mo','MarkerFaceColor','m','MarkerSize',8);
    grid on; xlabel('tau [h]'); ylabel('S_D [-]'); title('S_D(tau) = CD/(CD+CU+CP)');


    nexttile; plot(taus, YD, 'g','LineWidth',1.6); hold on;
    plot(best_tau, best_YD, 'go','MarkerFaceColor','g','MarkerSize',8);
    grid on; xlabel('tau [h]'); ylabel('Y_D [-]'); title('Y_D(tau) = CD/CA0');


    % 空きタイルは使わない
end


%% 7) 関数定義（最後にまとめて配置）


function [phi, CD, CP, SD, YD] = objectiveWeightedSumMulti(tau, CA0, k1, k2, k3, wD, wP, wS, wY)
% 重み付き和の単一目的関数（4指標対応）：
%   phi = -wD*CD + wP*CP - wS*SD - wY*YD
% 返り値：
%   phi : 目的関数値（小さいほど良い）
%   CD  : Dの濃度
%   CP  : Pの濃度
%   SD  : 選択性（組成基準）
%   YD  : 収率


    [CA, CD, CU, CP] = cstrSteady(tau, CA0, k1, k2, k3);
    SD = CD / max(CD + CU + CP, 1e-12);   % S_D = CD / (CD + CU + CP)
    YD = CD / max(CA0, 1e-12);            % Y_D = CD / CA0


    phi = -wD*CD + wP*CP - wS*SD - wY*YD;
end


function [CA, CD, CU, CP] = cstrSteady(tau, CA0, k1, k2, k3)
% 等温CSTRの定常濃度（簡易式）
% Aの物質収支：
%   (CA0 - CA)/tau = k1*CA + 2*k3*CA^2
% ⇔  2*k3*CA^2 + (k1 + 1/tau)*CA - CA0/tau = 0（CAの二次方程式）


    a = 2*k3;
    b = (k1 + 1/tau);
    c = -CA0/tau;


    if a == 0
        % 2A->P が無いときは一次式に退化
        CA = CA0 / (1 + k1*tau);
    else
        disc = b*b - 4*a*c;
        if disc < 0, disc = 0; end             % 数値ガード
        CA = (-b + sqrt(disc)) / (2*a);        % 物理根（>=0）
        CA = max(CA, 0);
    end


    CD = (k1*tau*CA) / (1 + k2*tau);   % A->D 生成 / D->U 消費のバランス
    CU = k2*tau*CD;                    % D->U で U 生成
    CP = k3*tau*CA^2;                  % 2A->P（二次反応）
end

```

```matlab
%% fgoalattain_demo_plain.m
% 目的：
%   tau（滞留時間）を設計変数に、次の4指標を「目標到達法」で同時に満たすよう最適化します。
%     1) C_D（Dの濃度）        → 大きいほど良い（最大化）
%     2) C_P（Pの濃度：副生成物）→ 小さいほど良い（最小化）
%     3) S_D（選択性）         → 大きいほど良い（最大化）
%     4) Y_D（収率）           → 大きいほど良い（最大化）
%
% ポイント：
%   - fgoalattain は「最小化ソルバ」なので、最大化したい量は負符号にして
%     F(tau) = [ -C_D, C_P, -S_D, -Y_D ] を最小化の対象にします。


clear; clc; close all;
fprintf('=== fgoalattain demo (plain): start ===\n');


%% 1) 問題設定（反応パラメータ・範囲・初期点）
CA0 = 1.0;   % [mol/L] 入口A濃度
k1  = 1.0;   % [1/h]   A -> D
k2  = 0.5;   % [1/h]   D -> U
k3  = 0.05;  % [L/mol/h] 2A -> P


tau_min = 1e-3;   % [h]
tau_max = 20.0;   % [h]
tau0    = 0.5;    % 初期点 [h]


%% 2) 目的ベクトル F(tau) を定義（最大化したい量は負符号）
F = @(tau) objectiveVector(tau, CA0, k1, k2, k3);


%% 3) 目標（goal）と重み（weight）
% 例：CD>=0.30, CP<=0.05, SD>=0.70, YD>=0.70 を狙う
% → F の向きに合わせて goal = [ -0.30, 0.05, -0.70, -0.70 ]
goal   = [ -0.30, 0.005, -0.70, -0.70 ];
weight = [  1.00, 1.00,  1.00,  1.00 ];


% （任意）非線形制約：CP <= CPmax を厳守したいとき
use_nonlin_constraint = false;
CPmax_allow          = 0.05;   % 許容上限の例
nonlcon = [];
if use_nonlin_constraint
    nonlcon = @(tau) deal( cpInequality(tau, CA0, k1, k2, k3, CPmax_allow), [] );
end


%% 4) fgoalattain で最適化
opts = optimoptions('fgoalattain', 'Display','iter');  % 進捗表示
[tau_star, fval, attainfactor, exitflag, output] = ...
    fgoalattain(F, tau0, goal, weight, [],[],[],[], tau_min, tau_max, nonlcon, opts);


% 最適点の指標値を計算
[CDs, CPs, SDs, YDs] = computeMetrics(tau_star, CA0, k1, k2, k3);


fprintf('\n=== 結果 ===\n');
fprintf('tau* = %.6g h\n', tau_star);
fprintf('CD=%.4g, CP=%.4g, SD=%.4g, YD=%.4g\n', CDs, CPs, SDs, YDs);
fprintf('attainfactor = %.6g（小さいほど目標到達）\n', attainfactor);
fprintf('exitflag = %d\n', exitflag);
fprintf('=== fgoalattain demo (plain): end ===\n');


%% 5) 可視化
% 指標曲線（CD, CP, SD, YD）、目的ベクトル各成分、最大不満足度 delta(tau) を描画


N     = 400;
taus  = linspace(tau_min, tau_max, N);
CD    = zeros(1,N);
CP    = zeros(1,N);
SD    = zeros(1,N);
YD    = zeros(1,N);
Fvals = zeros(N,4);
delta = zeros(1,N);


for i = 1:N
    t = taus(i);
    Fvals(i,:)      = F(t);
    [CD(i), CP(i), SD(i), YD(i)] = computeMetrics(t, CA0, k1, k2, k3);
    % 最大不満足度 delta(tau) = max_i (F_i(tau) - goal_i) / weight_i
    delta(i)        = max( (Fvals(i,:) - goal) ./ max(weight, 1e-12) );
end
F_star     = F(tau_star);
delta_star = max( (F_star - goal) ./ max(weight, 1e-12) );


% (A) 指標曲線：CD, CP, SD, YD
figure('Color','w','Position',[50 80 1200 500]);
tiledlayout(2,2,'TileSpacing','compact','Padding','compact');


nexttile;
plot(taus, CD, 'b','LineWidth',1.6); hold on;
plot(tau_star, CDs, 'bo','MarkerFaceColor','b','MarkerSize',7);
yline(-goal(1), '--b', 'Goal(C_D)');
grid on; xlabel('tau [h]'); ylabel('C_D [mol/L]'); title('C_D(tau)');


nexttile;
plot(taus, CP, 'r','LineWidth',1.6); hold on;
plot(tau_star, CPs, 'ro','MarkerFaceColor','r','MarkerSize',7);
yline(goal(2), '--r', 'Goal(C_P)');
grid on; xlabel('tau [h]'); ylabel('C_P [mol/L]'); title('C_P(tau)');


nexttile;
plot(taus, SD, 'm','LineWidth',1.6); hold on;
plot(tau_star, SDs, 'mo','MarkerFaceColor','m','MarkerSize',7);
yline(-goal(3), '--m', 'Goal(S_D)');
grid on; xlabel('tau [h]'); ylabel('S_D [-]'); title('S_D(tau) = C_D/(C_D + C_U + C_P)');


nexttile;
plot(taus, YD, 'g','LineWidth',1.6); hold on;
plot(tau_star, YDs, 'go','MarkerFaceColor','g','MarkerSize',7);
yline(-goal(4), '--g', 'Goal(Y_D)');
grid on; xlabel('tau [h]'); ylabel('Y_D [-]'); title('Y_D(tau) = C_D/C_A0');


% (B) 目的ベクトル各成分（-CD, CP, -SD, -YD）＋目標ライン
figure('Color','w','Position',[50 600 1200 500]);
plot(taus, Fvals(:,1), 'b', 'LineWidth',1.5); hold on;
plot(taus, Fvals(:,2), 'r', 'LineWidth',1.5);
plot(taus, Fvals(:,3), 'm', 'LineWidth',1.5);
plot(taus, Fvals(:,4), 'g', 'LineWidth',1.5);
plot(tau_star, F_star(1), 'bo','MarkerFaceColor','b','MarkerSize',7);
plot(tau_star, F_star(2), 'ro','MarkerFaceColor','r','MarkerSize',7);
plot(tau_star, F_star(3), 'mo','MarkerFaceColor','m','MarkerSize',7);
plot(tau_star, F_star(4), 'go','MarkerFaceColor','g','MarkerSize',7);
yline(goal(1), '--b'); yline(goal(2), '--r'); yline(goal(3), '--m'); yline(goal(4), '--g');
grid on; xlabel('tau [h]'); ylabel('F_i(tau)');
title('Objective components: [-C_D, C_P, -S_D, -Y_D] and goals');
legend({'-C_D','C_P','-S_D','-Y_D','opt -C_D','opt C_P','opt -S_D','opt -Y_D'}, 'Location','best');


% (C) 最大不満足度 delta(tau)
figure('Color','w','Position',[1280 80 540 500]);
plot(taus, delta, 'k', 'LineWidth',1.8); hold on;
plot(tau_star, delta_star, 'ks', 'MarkerFaceColor','y', 'MarkerSize',8);
yline(0, '--', 'delta = 0 (all goals attained)');
grid on; xlabel('tau [h]');
ylabel('delta(tau) = max_i { (F_i(tau) - goal_i) / weight_i }');
title('Goal attainment: max dissatisfaction curve');
text(tau_star, delta_star, sprintf('  tau* = %.3g,  delta* = %.3g', tau_star, delta_star), 'Color','k');


% 不満足度の各成分（最適点で）
F_star = F(tau_star);
deltas_star = (F_star - goal) ./ max(weight, 1e-12);


fprintf('\n--- dissatisfaction components at tau* ---\n');
fprintf('delta_CD  = ( -CD - goal_CD )/wD  = %.6g\n', deltas_star(1));
fprintf('delta_CP  = (  CP - goal_CP )/wP  = %.6g\n', deltas_star(2));
fprintf('delta_SD  = ( -SD - goal_SD )/wS  = %.6g\n', deltas_star(3));
fprintf('delta_YD  = ( -YD - goal_YD )/wY  = %.6g\n', deltas_star(4));
fprintf('max delta = %.6g (attainfactor)\n', max(deltas_star));


%% 6) 関数定義（計算ロジックはここに集約）


function F = objectiveVector(tau, CA0, k1, k2, k3)
% 目的ベクトル F(tau) を返す。
%   F = [ -C_D(tau), C_P(tau), -S_D(tau), -Y_D(tau) ]
% ここで最大化したい量（C_D, S_D, Y_D）は負符号にして最小化ソルバに渡す。
    [CD, CP, SD, YD] = computeMetrics(tau, CA0, k1, k2, k3);
    F = [ -CD, CP, -SD, -YD ];
end


function [CD, CP, SD, YD] = computeMetrics(tau, CA0, k1, k2, k3)
% 出口指標（C_D, C_P, S_D, Y_D）を計算するユーティリティ。
%   - S_D は組成基準： S_D = C_D / (C_D + C_U + C_P)
%   - Y_D は収率：    Y_D = C_D / C_A0
    [CA, CD, CU, CP] = cstrSteady2(tau, CA0, k1, k2, k3);
    SD = CD / max(CD + CU + CP, 1e-12);
    YD = CD / max(CA0, 1e-12);
end


function [CA, CD, CU, CP] = cstrSteady2(tau, CA0, k1, k2, k3)
% 等温 CSTR の定常濃度を計算（簡易モデル）
% A の物質収支：
%   (CA0 - CA)/tau = k1*CA + 2*k3*CA^2
% ⇔  2*k3*CA^2 + (k1 + 1/tau)*CA - CA0/tau = 0（CA についての二次方程式）
    a = 2.0*k3;
    b = (k1 + 1.0/tau);
    c = -CA0/tau;


    if a == 0.0
        % 副反応 2A->P がない場合は一次式に退化
        CA = CA0 / (1.0 + k1*tau);
    else
        disc = b*b - 4*a*c;
        disc = max(disc, 0.0);                 % 数値ガード（負の丸め回避）
        CA = (-b + sqrt(disc)) / (2.0*a);      % 物理根（>=0）
        CA = max(CA, 0.0);
    end


    % D, U, P の出口濃度（定常）
    CD = (k1*tau*CA) / (1.0 + k2*tau);   % A->D 生成と D->U 消費のバランス
    CU = k2*tau*CD;                      % D->U で U が生成
    CP = k3*tau*CA^2;                    % 2A->P（二次反応）
end


function c = cpInequality(tau, CA0, k1, k2, k3, CPmax_allow)
% 非線形不等式制約： c(tau) <= 0
%   例：CP(tau) - CPmax_allow <= 0（CP をしきい値以下に）
    [~, CP, ~, ~] = computeMetrics(tau, CA0, k1, k2, k3);
    c = CP - CPmax_allow;
end

```
