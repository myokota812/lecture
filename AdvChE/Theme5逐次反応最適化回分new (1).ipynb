## 1. 複合反応の物質収支式一般式から回分反応器の設計方程式の導出

化学反応工学において、単一の反応だけでなく、複数の反応が同時に進行する複合反応（Complex Reaction）系を取り扱うことは極めて重要です。本資料では、一般的な複合反応系における物質収支式の一般式を出発点とし、**回分反応器（Batch Reactor）の設計方程式**がどのように導出されるかを体系的に解説します。

---

## 1.1. 複合反応における物質収支の一般表現

系内に $N$ 種類の化学種（$i = 1, 2, \dots, N$）が存在し、$M$ 個の独立した化学反応（$j = 1, 2, \dots, M$）が同時に進行している系を考えます。

### (1) 化学量論係数と反応速度の定義

$j$ 番目の反応式を一般的に次のように表します。

$$\sum_{i=1}^{N} \nu_{i,j} A_i = 0 \quad (j = 1, 2, \dots, M)$$

* $A_i$：化学種 $i$
* $\nu_{i,j}$：化学種 $i$ の化学量論係数（生成物は正、消費（原料）は負）
* $r_j$：$j$ 番目の反応の反応速度（単位体積・単位時間あたりの反応量、$[ \mathrm{mol/(m^3 \cdot s)} ]$）

### (2) 一般的な物質収支式（系全体の物質収支）

任意の反応器において、微小時間 $dt$ の間の化学種 $i$ の物質量 $n_i$ の変化は、「流出入による増減」と「反応による生成・消費」の和で表されます。

$$\frac{dn_i}{dt} = F_{i,\text{in}} - F_{i,\text{out}} + V \sum_{j=1}^{M} \nu_{i,j} r_j$$

* $n_i$：化学種 $i$ のモル数 $[\mathrm{mol}]$
* $F_{i,\text{in}}, F_{i,\text{out}}$：化学種 $i$ の流入・流出モル速度 $[\mathrm{mol/s}]$
* $V$：反応液の体積 $[\mathrm{m^3}]$

---

## 1.2. 回分反応器（Batch Reactor）への適用と設計方程式の導出

### (1) 回分系の基本仮定

回分反応器は閉ざされた系（Closed System）であるため、反応器外部との物質の出入りが一切ありません。

* 流入・流出項がゼロになります： $F_{i,\text{in}} = 0, \ F_{i,\text{out}} = 0$

したがって、上記の一般物質収支式は次のように簡略化されます。

$$\frac{dn_i}{dt} = V \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

これが、**複合反応における回分反応器の微分型設計方程式の一般形**です。

---

### (2) 濃度基準の設計方程式への変換（定容系の場合）

反応液の体積 $V$ が時間によらず一定（$V = V_0 = \text{一定}$、定容系）であると仮定します。
濃度 $C_i = \frac{n_i}{V_0}$ を用いて式を変形すると、両辺を $V_0$ で割ることで濃度に関する微分方程式が得られます。

$$\frac{dC_i}{dt} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

これが、回分反応器の設計や解析で最も一般的に使用される方程式です。

---


## 1.3. 具体例：逐次反応 $A \to B \to C$ への適用確認

これまでの一般式が、これまで扱ってきた逐次反応にどのように当てはまるかを確認します。

* 化学種 ($N=3$): $A$ ($i=1$), $B$ ($i=2$), $C$ ($i=3$)
* 反応 ($M=2$):
* 反応1 ($j=1$): $A \to B \quad (r_1 = k_1 C_A)$
* 反応2 ($j=2$): $B \to C \quad (r_2 = k_2 C_B)$



化学量論係数 $\nu_{i,j}$ をマトリクス形式で整理すると：

* $A$ について： $\nu_{1,1} = -1, \ \nu_{1,2} = 0$
* $B$ について： $\nu_{2,1} = +1, \ \nu_{2,2} = -1$
* $C$ について： $\nu_{3,1} = 0, \ \nu_{3,2} = +1$

定容系における一般式 $\frac{dC_i}{dt} = \sum_{j=1}^{M} \nu_{i,j} r_j$ に代入すると、おなじみの以下の3元連立微分方程式が完全に導出されます。

1. **成分 $A$:** $\frac{dC_A}{dt} = (-1)r_1 + (0)r_2 = -k_1 C_A$
2. **成分 $B$:** $\frac{dC_B}{dt} = (+1)r_1 + (-1)r_2 = k_1 C_A - k_2 C_B$
3. **成分 $C$:** $\frac{dC_C}{dt} = (0)r_1 + (+1)r_2 = k_2 C_B$

## 2. 転化率、収率、選択率

本講義では、次のような一次の不可逆逐次反応を考えます。

$$A \xrightarrow{k_1} B \xrightarrow{k_2} C$$

* $A$：原料
* $B$：目的とする中間生成物
* $C$：副生物（過剰反応による分解物など）
* $k_1, k_2$：それぞれの反応速度定数

初期の $A$ の濃度を $C_{A0}$ とし、体積変化は無視できる（定容系）と仮定します。また、反応開始時 ($t = 0$) において $B$ および $C$ は存在しないものとします。

---

## 2.1. 諸定義の確認

反応の進行度や目的物質の生産効率を評価するため、以下の3つの指標を定義します。

### (1) 転化率 (Conversion) $X_A$

反応した原料 $A$ の割合を示します。


$$X_A(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}} \tag{eqX}$$

### (2) 収率 (Yield) $Y_B$

仕込んだ原料 $A$ に対し、目的物質 $B$ がどれだけ得られたかの割合です。


$$Y_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{仕込んだ } A \text{ の初期物質量}} = \frac{C_B(t)}{C_{A0}} \tag{eqY}$$

### (3) 選択率 (Selectivity) $S_B$

反応した原料 $A$ のうち、どれだけが目的物質 $B$ に変化したかの割合です。


$$S_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{反応した } A \text{ の物質量}} = \frac{C_B(t)}{C_{A0} - C_A(t)} = \frac{Y_B(t)}{X_A(t)} \tag{eqS}$$

---

## 2.2. 定式化

回分反応器における各成分の物質収支式（微分方程式）は以下の通りです。

$$\frac{dC_A}{dt} = -k_1 C_A$$

$$\frac{dC_B}{dt} = k_1 C_A - k_2 C_B$$

$$\frac{dC_C}{dt} = k_2 C_B$$

これらを解くことで、各変数の時間変化を明らかにできます。

* **原料 $A$:**

$$C_A = C_{A0} e^{-k_1 t}\tag{1}$$


* **中間生成物 $B$ ($k_1 \neq k_2$ の場合):**

$$C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)\tag{2}$$


* **副生物 $C$:**

$$C_C = C_{A0} \left( 1 - \frac{k_2 e^{-k_1 t} - k_1 e^{-k_2 t}}{k_2 - k_1} \right)\tag{3}$$

---

### (1) 転化率 $X_A$ の時間変化

原料 $A$ の濃度は(1)式より、転化率は次のように表されます。


$$X_A(t) = 1 - e^{-k_1 t} \tag{4}$$

* **特徴:** 転化率の挙動は第2段階の反応速度定数 $k_2$ に依存せず、$k_1$ のみで決定されます。時間が経つにつれて $X_A \to 1$（100%）に漸近します。


### (2) 収率 $Y_B(t)$ の時間変化の導出

#### 1. 定義の確認

収率 $Y_B(t)$ は、「仕込んだ原料 $A$ の初期物質量に対する、生成した目的物質 $B$ の割合」として定義されます。定容系（体積変化がない系）を仮定すると、濃度を用いて次のように表されます。


$$Y_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{仕込んだ } A \text{ の初期物質量}} = \frac{C_B(t)}{C_{A0}}$$


回分反応器における$B$ の濃度 $C_B(t)$ はEq. (2)（$k_1 \neq k_2$ の場合）であるので、これを収率の定義式(eqY)に代入します。


$$Y_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0}}$$


分子・分母の $C_{A0}$ で約分すると、収率 $Y_B(t)$ の時間変化を表す一般式が得られます。


$$Y_B(t) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) \tag{5}$$

---
### 補足：挙動の特徴

* **反応初期 ($t = 0$ のとき):**
$e^0 = 1$ より、$Y_B(0) = \frac{k_1}{k_2 - k_1} (1 - 1) = 0$ となり、まだ $B$ は生成していません。
* **反応長時間経過 ($t \to \infty$ のとき):**
$e^{-k_1 t} \to 0, e^{-k_2 t} \to 0$ となるため、$Y_B(\infty) \to 0$ となります。最終的にすべての $B$ が $C$ に変化するため、収率はゼロに戻ります。
* **極大値の存在:**
時間が経つにつれて $Y_B(t)$ は一旦増加してピーク（極大値）を迎えた後、減少に転じます。このピークを与える時間が「最適反応時間 $t_{\max}$」となります。



---

### (3) 選択率 $S_B(t)$ の時間変化の導出

選択率の定義式(eqS)に $C_A(t)$ および $C_B(t)$ の式を代入します。


$$S_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0} - C_{A0} e^{-k_1 t}}$$

分子・分母の $C_{A0}$ で約分すると、選択率の時間変化を表す一般式が得られます。


$$S_B(t) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 t} - e^{-k_2 t}}{1 - e^{-k_1 t}} \tag{6}$$

* **挙動の傾向:**
* 反応初期 ($t \to 0$): 副反応 ($B \to C$) がほとんど起きていないため、$S_B \to 1$（100%）となります。
* 反応長時間経過 ($t \to \infty$): すべての物質が $C$ に変化してしまうため、$S_B \to 0$ となります。すなわち、選択率は時間とともに単調減少します。

## 3. なにをもって最適化とするか（最適化の基準）

化学工業やプラント運転において「最適化」を行う際、何を基準にするか（目的関数）が極めて重要になります。逐次反応における最適化には、主に以下の視点が存在します。

## 3.1. 目的物質 $B$ の収率（または生産量）の最大化

技術的・化学的な観点における最も一般的な最適化基準です。

* **課題:** 反応時間が短すぎると原料 $A$ が残ってしまい、長すぎると $B$ がさらに反応して $C$ になってしまうため、収率 $Y_B$ には必ず極大値が存在します。
* **最適反応時間 $t_{\max}$ の導出:**
$\frac{dC_B}{dt} = 0$ となる条件から、収率を最大にする時間が求められます。

$$t_{\max} = \frac{1}{k_2 - k_1} \ln\left(\frac{k_2}{k_1}\right) \tag{7}$$


* **最大収率 $Y_{B,\max}$:**

$$Y_{B,\max} = \left(\frac{k_1}{k_2}\right)^{\frac{k_2}{k_2 - k_1}} \tag{8}$$

---





一例を示します。任意のパラメーターを与え、式(4),(5),(6)をグラフにするとともに、Bの収率最大とした場合の、$t_\text{max}$と$Y_\text{B,max}$を図示するコード（Code1)を用意しました。実行してください。

```python
#@title CODE 1：Bの収率最大化
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 時間軸の設定 (t = 0 から 5 まで)
t = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A の計算（※タイポ修正: 1 - np.exp）
X_A = 1 - np.exp(-k1 * t)

# 2. 収率 (Yield) Y_B の計算
if k1 == k2:
    Y_B = k1 * t * np.exp(-k1 * t)
    t_max = 1 / k1
    Y_B_max = np.exp(-1)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))
    t_max = (1 / (k2 - k1)) * np.log(k2 / k1)
    Y_B_max = (k1 / k2) ** (k2 / (k2 - k1))

# 3. 選択率 (Selectivity) S_B の計算 (t=0での0/0除算エラーを防ぐ処理)
S_B = np.ones_like(t)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]

# --- 追加: t_max における X_A と S_B の計算 ---
X_A_tmax = 1 - np.exp(-k1 * t_max)
S_B_tmax = Y_B_max / X_A_tmax

print(f"=== at t_max = {t_max:.3f} ===")
print(f"  Conversion (X_A): {X_A_tmax:.3f}")
print(f"  Yield      (Y_B): {Y_B_max:.3f}")
print(f"  Selectivity (S_B): {S_B_tmax:.3f}")
#-----------------------------------------------

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(t, X_A, label=r'Conversion ($X_A$)', color='blue', linewidth=2)
plt.plot(t, Y_B, label=r'Yield ($Y_B$)', color='green', linewidth=2)
plt.plot(t, S_B, label=r'Selectivity ($S_B$)', color='orange', linewidth=2, linestyle='--')

# 最適反応時間 t_max の明示（縦線とアノテーション）
plt.axvline(x=t_max, color='red', linestyle=':', linewidth=1.5, label=r'$t_{\max}$')
plt.scatter([t_max], [Y_B_max], color='green', zorder=5)
plt.scatter([t_max], [X_A_tmax], color='blue', zorder=5)
plt.scatter([t_max], [S_B_tmax], color='orange', zorder=5)

# アノテーションに t_max 時の X_A, Y_B, S_B をすべて記載
annotation_text = (
    f'Max Yield Point ($t_{{max}} = {t_max:.2f}$)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A(t_{{max}}) = {X_A_tmax:.2f}$\n'
    f'  $S_B(t_{{max}}) = {S_B_tmax:.2f}$'
)

plt.annotate(annotation_text,
             xy=(t_max, Y_B_max),
             xytext=(t_max + 0.3, Y_B_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

# グラフの装飾（英語化）
plt.title(r'Consecutive Reaction ($A \to B \to C$) Analysis ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Reaction Time ($t$)', fontsize=12)
plt.ylabel('Fraction / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.1)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
```

## 3.2. 転化率、収率（および選択率）の最大化

逐次反応（$A \to B \to C$）において、転化率 $X_A$ と収率 $Y_B$ の**重み付き線形結合**を目的関数（Objective Function）として設定し、それを最大化するアプローチを解説します。

実際の化学プラントでは、「原料をどれだけ反応させたか（転化率 $X_A$）」と「目的物質がどれだけ得られたか（収率 $Y_B$）」の双方が経済性に直結します。未反応の $A$ は回収して再利用できますが、副生物 $C$ まで進んでしまうと回収が難しく損失になります。このトレードオフを評価するための経済モデルや最適化基準として、線形結合を用いた目的関数がよく使われます。

---

## 3.2.1. 目的関数の定義

転化率 $X_A(t)$ と収率 $Y_B(t)$ を用いて、目的関数 $J(t)$ を次のように定義します。

$$J(t) = w_1 X_A(t) + w_2 Y_B(t)$$

* $w_1$：転化率に対する重み（例：原料 $A$ を反応させること自体のインセンティブ、あるいは未反応 $A$ の回収コストに対するペナルティの兼ね合い）
* $w_2$：収率に対する重み（例：目的物質 $B$ の市場価値や販売価格）

ここで、$X_A(t) = 1 - e^{-k_1 t}$ および $Y_B(t) = \frac{k_1}{k_2 - k_1} (e^{-k_1 t} - e^{-k_2 t})$ を代入すると、目的関数は時間 $t$ の関数として次のように表されます。

$$J(t) = w_1 (1 - e^{-k_1 t}) + w_2 \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

---

## 3.2.2. 目的関数の最大化（最適反応時間の探索）

この目的関数 $J(t)$ を最大化する最適反応時間 $t_{\text{opt}}$ は、時間微分 $\frac{dJ}{dt} = 0$ となる条件から求められます。

$$\frac{dJ}{dt} = w_1 k_1 e^{-k_1 t} + w_2 \frac{k_1}{k_2 - k_1} \left( -k_1 e^{-k_1 t} + k_2 e^{-k_2 t} \right) = 0$$

この式を整理すると、単なる収率単体の最大化（$w_1 = 0$ の場合）とは異なり、**重みのバランス（$w_1 / w_2$）によって最適な反応時間が変化する**ことがわかります。

* $w_1$（転化率の重み）を大きくすると、原料をより多く消費する（長めの反応時間）方向に最適値がシフトします。
* $w_2$（収率の重み）を大きくすると、副生物 $C$ への過剰反応を嫌うため、収率のピークに近づき短めの反応時間で打ち切るようになります。

## 3.2.3. Pythonによる実装とグラフ描画コード

重み付き目的関数 $J(t)$ を計算し、その最大値と最適反応時間をプロットするPythonコードです。


```python
#@title CODE 2:目的関数の最大化
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 重み係数の設定（例: w1 = 0.3 (転化率の価値), w2 = 1.0 (目的物質Bの価値)）
w1 = 0.3
w2 = 0.7

# 時間軸の設定 (t = 0 から 5 まで)
t = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A と 収率 (Yield) Y_B の計算（※タイポ修正: 1 - np.exp）
X_A = 1 - np.exp(-k1 * t)
if k1 == k2:
    Y_B = k1 * t * np.exp(-k1 * t)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))

# 2. 重み付き目的関数 J(t) の計算
J = w1 * X_A + w2 * Y_B

# 3. 目的関数 J(t) を最大化する最適時間 t_opt の数値的な探索
max_index = np.argmax(J)
t_opt = t[max_index]
J_max = J[max_index]

# --- 追加: t_opt における X_A, Y_B, S_B の計算 ---
X_A_topt = 1 - np.exp(-k1 * t_opt)
if k1 == k2:
    Y_B_topt = k1 * t_opt * np.exp(-k1 * t_opt)
else:
    Y_B_topt = (k1 / (k2 - k1)) * (np.exp(-k1 * t_opt) - np.exp(-k2 * t_opt))

# 選択率 S_B の計算（t=0での除算エラーを防ぐ処理）
S_B_topt = Y_B_topt / X_A_topt if X_A_topt > 1e-6 else 1.0

print(f"=== at t_opt = {t_opt:.3f} ===")
print(f"  Conversion (X_A): {X_A_topt:.3f}")
print(f"  Yield      (Y_B): {Y_B_topt:.3f}")
print(f"  Selectivity (S_B): {S_B_topt:.3f}")
print(f"  Objective  (J):   {J_max:.3f}")
#-----------------------------------------------

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(t, X_A, label=r'Conversion ($X_A$)', color='blue', alpha=0.6, linestyle=':')
plt.plot(t, Y_B, label=r'Yield ($Y_B$)', color='green', alpha=0.6, linestyle=':')
plt.plot(t, J, label=rf'Objective Function $J(t) = {w1}X_A + {w2}Y_B$', color='purple', linewidth=2.5)

# 最適点 (t_opt, J_max) および各値のプロット
plt.axvline(x=t_opt, color='red', linestyle='--', linewidth=1.5, label=f'Optimal Time $t_{{opt}} = {t_opt:.2f}$')
plt.scatter([t_opt], [J_max], color='purple', zorder=5)
plt.scatter([t_opt], [X_A_topt], color='blue', zorder=5)
plt.scatter([t_opt], [Y_B_topt], color='green', zorder=5)

# アノテーションに t_opt 時の J, X_A, Y_B, S_B をすべて記載
annotation_text = (
    f'Optimal Time ($t_{{opt}} = {t_opt:.2f}$)\n'
    f'  $J_{{max}} = {J_max:.2f}$\n'
    f'  $X_A(t_{{opt}}) = {X_A_topt:.2f}$\n'
    f'  $Y_B(t_{{opt}}) = {Y_B_topt:.2f}$\n'
    f'  $S_B(t_{{opt}}) = {S_B_topt:.2f}$'
)

plt.annotate(annotation_text,
             xy=(t_opt, J_max),
             xytext=(t_opt + 0.3, J_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

# グラフの装飾
plt.title(r'Optimization using Weighted Objective Function ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Reaction Time ($t$)', fontsize=12)
plt.ylabel('Value / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.2)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
```

## 1. 複合反応の物質収支式一般式から管型反応器（PFR）の設計方程式の導出

化学反応工学において、単一の反応だけでなく、複数の反応が同時に進行する複合反応（Complex Reaction）系を取り扱うことは極めて重要です。本資料では、一般的な複合反応系における物質収支式の一般式を出発点とし、**管型反応器（Plug Flow Reactor: PFR）の設計方程式**がどのように導出されるかを体系的に解説します。

---

## 1.1. 複合反応における物質収支の一般表現

系内に $N$ 種類の化学種（$i = 1, 2, \dots, N$）が存在し、$M$ 個の独立した化学反応（$j = 1, 2, \dots, M$）が同時に進行している系を考えます。

### (1) 化学量論係数と反応速度の定義

$j$ 番目の反応式を一般的に次のように表します。

$$\sum_{i=1}^{N} \nu_{i,j} A_i = 0 \quad (j = 1, 2, \dots, M)$$

* $A_i$：化学種 $i$
* $\nu_{i,j}$：化学種 $i$ の化学量論係数（生成物は正、消費（原料）は負）
* $r_j$：$j$ 番目の反応の反応速度（単位体積・単位時間あたりの反応量、$[ \mathrm{mol/(m^3 \cdot s)} ]$）

### (2) 一般的な物質収支式（系全体の物質収支）

任意の反応器において、微小時間 $dt$ の間の化学種 $i$ の物質量 $n_i$ の変化は、「流出入による増減」と「反応による生成・消費」の和で表されます。

$$\frac{dn_i}{dt} = F_{i,\text{in}} - F_{i,\text{out}} + V \sum_{j=1}^{M} \nu_{i,j} r_j$$

* $n_i$：化学種 $i$ のモル数 $[\mathrm{mol}]$
* $F_{i,\text{in}}, F_{i,\text{out}}$：化学種 $i$ の流入・流出モル速度 $[\mathrm{mol/s}]$
* $V$：反応液の体積 $[\mathrm{m^3}]$

---

## 1.2. 管型反応器（PFR）への適用と設計方程式の導出

### (1) PFRの基本仮定

管型反応器（Plug Flow Reactor）は、流体が管内を乱れなくプラグ流（押し出し流れ）として流れる連続式反応器です。

* **定常状態（Steady State）**：時間の経過に伴う反応器内の状態変化がありません（$\frac{\partial}{\partial t} = 0$）。
* **軸方向の逆混合なし**：流れ方向（軸方向）の混ざり合いがなく、半径方向には完全に混合していると仮定します。

### (2) 微小体積要素 $dV$ における物質収支

<figure style="text-align:center;">
  <img
    src="https://raw.githubusercontent.com/myokota812/lecture/master/AdvChE/figure/pfr3.svg"
    width="800">
    <figcaption>
     図2. 管型反応器（PFR）の模式図
    </figcaption>
</figure>

PFRの入口から体積 $V$ 進んだ位置にある、微小な体積要素 $dV$ （または微小長さ $dz$）を考えます。この微小要素を定常状態で通過する化学種 $i$ の収支は、「流入 - 流出 + 反応による生成・消費 = 0」となります。

* 流入モル速度： $F_i(V)$
* 流出モル速度： $F_i(V + dV) = F_i(V) + \frac{dF_i}{dV} dV$
* 反応による生成・消費： $dV \sum_{j=1}^{M} \nu_{i,j} r_j$

これらを足し合わせると、次式が得られます。

$$F_i(V) - \left( F_i(V) + \frac{dF_i}{dV} dV \right) + dV \sum_{j=1}^{M} \nu_{i,j} r_j = 0$$

整理すると、PFRにおける**モル速度基準の微分型設計方程式の一般形**が得られます。

$$\frac{dF_i}{dV} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

---

### (3) 濃度基準の設計方程式への変換（定容系の場合）

体積流量 $v$ が反応を通じて一定（$v = v_0 = \text{一定}$、定容系）であると仮定します。
モル速度と濃度の関係 $F_i = C_i v_0$ を用いて式を変形すると、両辺を $v_0$ で割ることで、反応体積 $V$ に対する濃度変化の微分方程式が得られます。

$$\frac{dC_i}{d(V/v_0)} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad \text{あるいは} \quad \frac{dC_i}{d\tau} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

* $\tau = \frac{V}{v_0}$：空間時間（Residence Time / Space Time） $[\mathrm{s}]$

これが、PFRの設計や解析で最も一般的に使用される方程式です。（※形としては回分反応器の $t$ が PFRの $\tau$ に置き換わった形になります）

---

## 1.3. 具体例：逐次反応 $A \to B \to C$ への適用確認

これまでの一般式が、PFRにおける逐次反応にどのように当てはまるかを確認します。

* 化学種 ($N=3$): $A$ ($i=1$), $B$ ($i=2$), $C$ ($i=3$)
* 反応 ($M=2$):
* 反応1 ($j=1$): $A \to B \quad (r_1 = k_1 C_A)$
* 反応2 ($j=2$): $B \to C \quad (r_2 = k_2 C_B)$



化学量論係数 $\nu_{i,j}$ は以下の通りです。

* $A$ について： $\nu_{1,1} = -1, \ \nu_{1,2} = 0$
* $B$ について： $\nu_{2,1} = +1, \ \nu_{2,2} = -1$
* $C$ について： $\nu_{3,1} = 0, \ \nu_{3,2} = +1$

定容系におけるPFRの一般式 $\frac{dC_i}{d\tau} = \sum_{j=1}^{M} \nu_{i,j} r_j$ に代入すると、空間時間 $\tau$ に関する以下の3元連立微分方程式が完全に導出されます。

1. **成分 $A$:** $\frac{dC_A}{d\tau} = (-1)r_1 + (0)r_2 = -k_1 C_A$
2. **成分 $B$:** $\frac{dC_B}{d\tau} = (+1)r_1 + (-1)r_2 = k_1 C_A - k_2 C_B$
3. **成分 $C$:** $\frac{dC_C}{d\tau} = (0)r_1 + (+1)r_2 = k_2 C_B$

## 2. 転化率、収率、選択率

本講義では、次のような一次の不可逆逐次反応を、連続式反応器の一つである管型反応器（Plug Flow Reactor: PFR）で実施する系を考えます。

$$A \xrightarrow{k_1} B \xrightarrow{k_2} C$$

* $A$：原料
* $B$：目的とする中間生成物
* $C$：副生物（過剰反応による分解物など）
* $k_1, k_2$：それぞれの反応速度定数

反応器入口における $A$ の濃度を $C_{A0}$ とし、体積変化は無視できる（定容系）と仮定します。また、反応器入口（空間時間 $\tau = 0$）において $B$ および $C$ は存在しないものとします。

---

## 2.1. 諸定義の確認

反応の進行度や目的物質の生産効率を評価するため、以下の3つの指標を定義します（回分反応器の時間 $t$ が、PFRでは**空間時間 $\tau = \frac{V}{v_0}$** に置き換わります）。

### (1) 転化率 (Conversion) $X_A$

反応した原料 $A$ の割合を示します。

$$X_A(\tau) = \frac{C_{A0} - C_A(\tau)}{C_{A0}} = 1 - \frac{C_A(\tau)}{C_{A0}} \tag{eqX}$$

### (2) 収率 (Yield) $Y_B$

供給した原料 $A$ に対し、目的物質 $B$ がどれだけ得られたかの割合です。

$$Y_B(\tau) = \frac{\text{生成した } B \text{ のモル速度}}{\text{供給した } A \text{ の初期モル速度}} = \frac{C_B(\tau)}{C_{A0}} \tag{eqY}$$

### (3) 選択率 (Selectivity) $S_B$

反応した原料 $A$ のうち、どれだけが目的物質 $B$ に変化したかの割合です。

$$S_B(\tau) = \frac{\text{生成した } B \text{ のモル速度}}{\text{反応した } A \text{ のモル速度}} = \frac{C_B(\tau)}{C_{A0} - C_A(\tau)} = \frac{Y_B(\tau)}{X_A(\tau)} \tag{eqS}$$

---

## 2.2. 定式化

PFRにおける各成分の物質収支式（空間時間 $\tau$ に関する微分方程式）は以下の通りです。

$$\frac{dC_A}{d\tau} = -k_1 C_A$$

$$\frac{dC_B}{d\tau} = k_1 C_A - k_2 C_B$$

$$\frac{dC_C}{d\tau} = k_2 C_B$$

これらを解くことで、反応管の長さ（空間時間 $\tau$）に沿った各変数の濃度変化を明らかにできます。

* **原料 $A$:**

$$C_A = C_{A0} e^{-k_1 \tau}\tag{1}$$

* **中間生成物 $B$ ($k_1 \neq k_2$ の場合):**

$$C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)\tag{2}$$

* **副生物 $C$:**

$$C_C = C_{A0} \left( 1 - \frac{k_2 e^{-k_1 \tau} - k_1 e^{-k_2 \tau}}{k_2 - k_1} \right)\tag{3}$$

---

### (1) 転化率 $X_A$ の空間時間変化

原料 $A$ の濃度は(1)式より、転化率は次のように表されます。

$$X_A(\tau) = 1 - e^{-k_1 \tau} \tag{4}$$

* **特徴:** 転化率の挙動は第2段階の反応速度定数 $k_2$ に依存せず、$k_1$ のみで決定されます。空間時間 $\tau$ が大きくなる（反応管が長くなる、または流速が遅くなる）につれて $X_A \to 1$（100%）に漸近します。

### (2) 収率 $Y_B(\tau)$ の空間時間変化の導出

#### 1. 定義の確認

収率 $Y_B(\tau)$ は、「供給した原料 $A$ の初期濃度に対する、生成した目的物質 $B$ の濃度の割合」として定義されます。定容系を仮定すると、濃度を用いて次のように表されます。

$$Y_B(\tau) = \frac{C_B(\tau)}{C_{A0}}$$

PFRにおける $B$ の濃度 $C_B(\tau)$ は Eq. (2)（$k_1 \neq k_2$ の場合）であるので、これを収率の定義式(eqY)に代入します。

$$Y_B(\tau) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)}{C_{A0}}$$

分子・分母の $C_{A0}$ で約分すると、収率 $Y_B(\tau)$ の空間時間変化を表す一般式が得られます。

$$Y_B(\tau) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right) \tag{5}$$

---

### 補足：挙動の特徴

* **反応器入口 ($\tau = 0$ のとき):**
$e^0 = 1$ より、$Y_B(0) = \frac{k_1}{k_2 - k_1} (1 - 1) = 0$ となり、まだ $B$ は生成していません。
* **空間時間が非常に大きい場合 ($\tau \to \infty$ のとき):**
$e^{-k_1 \tau} \to 0, e^{-k_2 \tau} \to 0$ となるため、$Y_B(\infty) \to 0$ となります。最終的にすべての $B$ が $C$ に変化するため、収率はゼロに戻ります。
* **極大値の存在:**
空間時間 $\tau$ が大きくなるにつれて $Y_B(\tau)$ は一旦増加してピーク（極大値）を迎えた後、減少に転じます。このピークを与える空間時間が「最適空間時間 $\tau_{\max}$」となります。

---

### (3) 選択率 $S_B(\tau)$ の空間時間変化の導出

選択率の定義式(eqS)に $C_A(\tau)$ および $C_B(\tau)$ の式を代入します。

$$S_B(\tau) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 \tau} - e^{-k_2 \tau} \right)}{C_{A0} - C_{A0} e^{-k_1 \tau}}$$

分子・分母の $C_{A0}$ で約分すると、選択率の空間時間変化を表す一般式が得られます。

$$S_B(\tau) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 \tau} - e^{-k_2 \tau}}{1 - e^{-k_1 \tau}} \tag{6}$$

* **挙動の傾向:**
* **反応器入口付近 ($\tau \to 0$):** 副反応 ($B \to C$) がほとんど起きていないため、$S_B \to 1$（100%）となります。
* **空間時間が非常に大きい場合 ($\tau \to \infty$):** すべての物質が $C$ に変化してしまうため、$S_B \to 0$ となります。すなわち、選択率は空間時間とともに単調減少します。

---

## 3. なにをもって最適化とするか（最適化の基準）

化学工業やプラント運転におけるPFRの設計・運用において「最適化」を行う際、何を基準にするか（目的関数）が極めて重要になります。逐次反応における最適化には、主に以下の視点が存在します。

## 3.1. 目的物質 $B$ の収率（または生産量）の最大化

技術的・化学的な観点における最も一般的な最適化基準です。

* **課題:** 反応管の長さ（空間時間 $\tau$）が短すぎると原料 $A$ が残ってしまい、長すぎると $B$ がさらに反応して $C$ になってしまうため、収率 $Y_B$ には必ず極大値が存在します。
* **最適空間時間 $\tau_{\max}$ の導出:**
$\frac{dC_B}{d\tau} = 0$ となる条件から、収率を最大にする空間時間が求められます。

$$\tau_{\max} = \frac{1}{k_2 - k_1} \ln\left(\frac{k_2}{k_1}\right) \tag{7}$$

* **最大収率 $Y_{B,\max}$:**

$$Y_{B,\max} = \left(\frac{k_1}{k_2}\right)^{\frac{k_2}{k_2 - k_1}} \tag{8}$$

```python
#@title PFR
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 滞留時間（空間時間）の軸設定 (tau = 0 から 5 まで)
tau = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A の計算
X_A = 1 - np.exp(-k1 * tau)

# 2. 収率 (Yield) Y_B の計算
if k1 == k2:
    Y_B = k1 * tau * np.exp(-k1 * tau)
    tau_max = 1 / k1
    Y_B_max = np.exp(-1)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * tau) - np.exp(-k2 * tau))
    tau_max = (1 / (k2 - k1)) * np.log(k2 / k1)
    Y_B_max = (k1 / k2) ** (k2 / (k2 - k1))

# 3. 選択率 (Selectivity) S_B の計算
S_B = np.ones_like(tau)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]

# tau_max における値の計算
X_A_taumax = 1 - np.exp(-k1 * tau_max)
S_B_taumax = Y_B_max / X_A_taumax

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(tau, X_A, label=r'Conversion ($X_A$)', color='blue', linewidth=2)
plt.plot(tau, Y_B, label=r'Yield ($Y_B$)', color='green', linewidth=2)
plt.plot(tau, S_B, label=r'Selectivity ($S_B$)', color='orange', linewidth=2, linestyle='--')

# 最適滞留時間 tau_max の明示
plt.axvline(x=tau_max, color='red', linestyle=':', linewidth=1.5, label=r'Optimal $\tau_{\max}$')
plt.scatter([tau_max], [Y_B_max], color='green', zorder=5)

annotation_text = (
    f'PFR Max Yield Point\n'
    f'  $\tau_{{max}} = {tau_max:.2f}$\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A(\tau_{{max}}) = {X_A_taumax:.2f}$\n'
    f'  $S_B(\tau_{{max}}) = {S_B_taumax:.2f}$'
)

plt.annotate(annotation_text,
             xy=(tau_max, Y_B_max),
             xytext=(tau_max + 0.3, Y_B_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

plt.title(r'PFR Consecutive Reaction ($A \to B \to C$) Analysis ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Space Time / Residence Time ($\tau$)', fontsize=12)
plt.ylabel('Fraction / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.1)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
```

# 講義ノート：非等温回分反応器における逐次反応（$A \to B \to C, E_1 > E_2$）の数学的モデリングと解析手順

本講義ノートでは、逐次反応 $A \xrightarrow{k_1} B \xrightarrow{k_2} C$ を対象とし、反応器の温度が時間とともに変化する非等温回分反応器（Non-isothermal Batch Reactor）の設計方程式（物質収支式およびエンタルピー収支式）の導出と、$E_1 > E_2$（第1段階の活性化エネルギーが第2段階よりも大きい）の条件を活かした解析・最適化の手順を解説します。

---

## 1. 物質収支式（Mass Balance Equations）の導出

定容系（反応液の体積 $V = V_0 = \text{一定}$）の回分反応器を仮定します。各成分の反応速度はアレニウスの式に従うため、速度定数 $k_1, k_2$ は温度 $T$ の関数となります。

$$k_1(T) = A_1 \exp\left(-\frac{E_1}{RT}\right), \quad k_2(T) = A_2 \exp\left(-\frac{E_2}{RT}\right)$$

前回の一般式 $\frac{dC_i}{dt} = \sum \nu_{i,j} r_j$ に基づき、各成分の物質収支式は以下のように立式されます。

1. **成分 $A$ の収支:**

$$\frac{dC_A}{dt} = -r_1 = -k_1(T) C_A$$


2. **成分 $B$ の収支（生成と消費の競合）:**

$$\frac{dC_B}{dt} = r_1 - r_2 = k_1(T) C_A - k_2(T) C_B$$


3. **成分 $C$ の収支:**

$$\frac{dC_C}{dt} = r_2 = k_2(T) C_B$$



---

## 2. エンタルピー収支式（Energy / Enthalpy Balance Equations）の導出

非等温系では温度 $T$ が変数となるため、温度のの時間変化 $\frac{dT}{dt}$ を記述するエネルギー保存則（熱収支式）が必要になります。

### (1) 閉ざされた系のエネルギー保存則

熱力学の第一法則より、外部との仕事のやり取り（容積仕事）を無視できる液相系では、内部エネルギー $U$ の変化は流入・流出がないため熱の出入りおよび反応熱に等しくなります。

$$\frac{dU}{dt} = \dot{Q} + \sum_{j=1}^{M} (-\Delta H_j) V r_j$$

* $\dot{Q}$：外部のジャケット等から加えられる（または奪われる）伝熱速度 $[\mathrm{J/s}]$
* $\Delta H_j$：$j$ 番目の反応の標準反応エンタルピー（発熱反応の場合は負：$\Delta H < 0$）

### (2) 温度に関する微分方程式への変換

内部エネルギー $U$ およびエンタルピー $H$ は温度と組成の関数です。液相の密度 $\rho$ および比熱 $C_p$ が一定であると仮定し、全系の熱容量を用いて温度 $T$ の変化率 $\frac{dT}{dt}$ に書き換えます。

$$V \rho C_p \frac{dT}{dt} = \dot{Q} + V \left[ (-\Delta H_1) r_1 + (-\Delta H_2) r_2 \right]$$

ここで、外部からの伝熱速度 $\dot{Q}$ は、伝熱係数 $U_h$、伝熱面積 $A_h$、およびジャケット温度 $T_j$ を用いて $\dot{Q} = U_h A_h (T_j - T)$ と表されます。これを代入すると、非等温回分反応器のエンタルピー収支式（温度の微分方程式）が得られます。

$$\rho C_p \frac{dT}{dt} = U_h a (T_j - T) + (-\Delta H_1) k_1(T) C_A + (-\Delta H_2) k_2(T) C_B$$

（※ $a = A_h / V$ は単位体積あたりの伝熱面積）

---

## 3. $E_1 > E_2$ 条件を活かす解析と最適化の手順

活性化エネルギーの関係が $E_1 > E_2$ である場合、温度変化に対する反応速度の感度が異なります。

* **$E_1 > E_2$ の物理的意味:** 温度を上げると、副反応 $B \to C$ よりも主反応 $A \to B$ の方がより急激に加速されます。

この特性を最大限に利用し、中間体 $B$ の収率を極限まで高めるための解析・最適化手順は以下の通りです。

### ステップ 1：物性値および運動論パラメータの定義

* 各種パラメータ（$A_1, A_2, E_1, E_2, \Delta H_1, \Delta H_2, \rho, C_p, U_h a$）を実験データ等から設定します。

### ステップ 2：連立常微分方程式（ODE）の構築

以下の4元連立非線形微分方程式システムを構築します。

1. $\frac{dC_A}{dt} = -k_1(T) C_A$
2. $\frac{dC_B}{dt} = k_1(T) C_A - k_2(T) C_B$
3. $\frac{dC_C}{dt} = k_2(T) C_B$
4. $\frac{dT}{dt} = \frac{1}{\rho C_p} \left[ U_h a (T_j - T) + (-\Delta H_1) k_1(T) C_A + (-\Delta H_2) k_2(T) C_B \right]$

### ステップ 3：最適温度プロファイル（$T_j(t)$ または $T(t)$）の設計

* **反応前半（高温運転）:** $E_1 > E_2$ であるため、あえて反応器温度を高めに設定し、主反応速度 $k_1$ を極限まで引き上げて短時間で原料 $A$ を消費させます。
* **反応後半（急冷・低温化）:** 目的物質 $B$ が最大濃度に近づいた瞬間、ジャケット温度 $T_j$ を急激に下げて系全体の温度を低下させます。これにより、相対的に $E_2$ が小さくても温度低下によって $k_2$ が急速に抑制され、$B \to C$ への過剰反応がブロックされます。

### ステップ 4：数値シミュレーションによる検証

* Python等の数値計算ライブラリ（`scipy.integrate.solve_ivp` など）を用いて微分方程式を解き、等温運転時と非等温最適制御時での $Y_B$ の最大値を比較・評価します。

---

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.integrate import solve_ivp

# --- 1. パラメータの設定 ---
R = 8.314  # 気体定数 [J/(mol*K)]
T_room = 298.15  # 基準温度（室温） [K]

# 活性化エネルギーの設定 (E1 > E2 の条件)
E1 = 55000.0  # 主反応 A -> B の活性化エネルギー [J/mol]
E2 = 45000.0  # 副反応 B -> C の活性化エネルギー [J/mol]

# 室温で k1=1.0, k2=0.5 になるように頻度因子 A1, A2 を逆算決定
k1_room = 1.0
k2_room = 0.5
A1 = k1_room * np.exp(E1 / (R * T_room))
A2 = k2_room * np.exp(E2 / (R * T_room))

# 反応熱 (発熱反応) [J/mol]
dH1 = -50000.0
dH2 = -40000.0

# 物性値・熱伝達パラメータ
rho_Cp = 4180.0  # 体積あたりの熱容量 [J/(L*K)]
Ua = 200.0  # 総括伝熱係数×伝熱面積/体積 [J/(L*s*K)]

# --- ★ ジャケット温度を一定に設定 ---
T_j_const = 350.0  # ジャケット温度 [K] (一定)


# --- 2. 4元連立常微分方程式の定義 (y = [C_A, C_B, C_C, T]) ---
def nonisothermal_batch_constant_jacket(t, y):
  CA, CB, CC, T = y

  # アレニウスの式による温度依存の速度定数
  k1 = A1 * np.exp(-E1 / (R * T))
  k2 = A2 * np.exp(-E2 / (R * T))

  # 反応速度
  r1 = k1 * CA
  r2 = k2 * CB

  # 物質収支 (濃度変化)
  dCA_dt = -r1
  dCB_dt = r1 - r2
  dCC_dt = r2

  # エンタルピー収支 (温度変化) ※ジャケット温度 T_j_const は一定
  dT_dt = (
      Ua * (T_j_const - T) + (-dH1) * r1 + (-dH2) * r2
  ) / rho_Cp

  return [dCA_dt, dCB_dt, dCC_dt, dT_dt]


# --- 3. 初期条件と時間設定 ---
CA0 = 1.0  # Aの初期濃度 [mol/L]
CB0 = 0.0
CC0 = 0.0
#T0 = T_room  # 初期温度（室温 298.15 K）
T0 = T_j_const
y0 = [CA0, CB0, CC0, T0]

t_span = (0, 1.5)
t_eval = np.linspace(t_span[0], t_span[1], 500)

# --- 4. 数値計算の実行 (solve_ivp) ---
sol = solve_ivp(
    nonisothermal_batch_constant_jacket,
    t_span,
    y0,
    t_eval=t_eval,
    method='RK45',
)

# --- 5. 転化率・収率・選択率の算出 ---
CA = sol.y[0]
CB = sol.y[1]
T_hist = sol.y[3]

X_A = 1.0 - (CA / CA0)  # 転化率
Y_B = CB / CA0  # 収率

S_B = np.ones_like(X_A)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]  # 選択率

# --- 6. 最適操作時間 t_max の自動検出 ---
max_idx = np.argmax(Y_B)
t_max = sol.t[max_idx]
Y_B_max = Y_B[max_idx]
X_A_tmax = X_A[max_idx]
S_B_tmax = S_B[max_idx]
T_tmax = T_hist[max_idx]

print(
    f"=== 非等温回分反応器 (ジャケット温度一定: {T_j_const:.1f} K) の結果 ==="
)
print(f"  最適操作時間 (t_max): {t_max:.3f} s")
print(f"  最大収率 (Y_B_max):   {Y_B_max:.3f}")
print(f"  時の転化率 (X_A):     {X_A_tmax:.3f}")
print(f"  時の選択率 (S_B):     {S_B_tmax:.3f}")
print(f"  時の反応器温度 (T):   {T_tmax:.1f} K")

# --- 7. 2段グラフ描画 ---
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# 上段：転化率、収率、選択率の時間変化
ax1.plot(sol.t, X_A, label=r'Conversion ($X_A$)', color='blue', lw=2)
ax1.plot(sol.t, Y_B, label=r'Yield ($Y_B$)', color='green', lw=2)
ax1.plot(
    sol.t,
    S_B,
    label=r'Selectivity ($S_B$)',
    color='orange',
    lw=2,
    linestyle='--',
)
ax1.axvline(
    x=t_max, color='red', linestyle=':', linewidth=1.5, label=r'$t_{\max}$'
)
ax1.scatter([t_max], [Y_B_max], color='green', zorder=5)

annotation_text = (
    f'Max Yield Point ($t_{{max}} = {t_max:.2f}$ s)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A = {X_A_tmax:.2f}$'
)
ax1.annotate(
    annotation_text,
    xy=(t_max, Y_B_max),
    xytext=(t_max + 0.1, Y_B_max - 0.25),
    arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
    fontsize=10,
    color='darkred',
    bbox=dict(boxstyle='round,pad=0.3', fc='white', ec='red', lw=1, alpha=0.9),
)
ax1.set_ylabel('Fraction / Ratio (-)', fontsize=11)
ax1.set_title(
    r'Non-isothermal Batch Reactor (Constant Jacket Temp $T_j = '
    f'{T_j_const:.1f}$ K)',
    fontsize=13,
)
ax1.set_ylim(-0.05, 1.1)
ax1.grid(True, linestyle=':', alpha=0.7)
ax1.legend(fontsize=10, loc='lower right')

# 下段：反応器温度とジャケット温度の推移
ax2.plot(sol.t, T_hist, label='Reactor Temp ($T$)', color='purple', lw=2)
ax2.axhline(
    y=T_j_const,
    color='gray',
    linestyle='--',
    lw=1.5,
    label=f'Jacket Temp ($T_j = {T_j_const:.1f}$ K)',
)
ax2.axvline(x=t_max, color='red', linestyle=':', linewidth=1.5)
ax2.set_xlabel('Reaction Time ($t$) [s]', fontsize=11)
ax2.set_ylabel('Temperature [K]', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.7)
ax2.legend(fontsize=10, loc='upper right')

plt.tight_layout()
plt.show()
```

---

逐次反応 $A \to B \to C$ における選択率 $S_B(t)$ の時間変化の式を以下に導出します。

### 1. 選択率の定義

選択率 $S_B$ は、「反応した原料 $A$ のうち、目的物質 $B$ に変化した割合」として定義されます。


$$S_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{反応した } A \text{ の物質量}} = \frac{C_B(t)}{C_{A0} - C_A(t)}$$

### 2. 各濃度の時間変化式の代入

回分反応器における各成分の濃度 $C_A(t)$ および $C_B(t)$ は以下の通りです。


$$C_A(t) = C_{A0} e^{-k_1 t}$$

$$C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

これらを $S_B(t)$ の定義式に代入します。


$$S_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0} - C_{A0} e^{-k_1 t}}$$

### 3. 式の整理

分子・分母の $C_{A0}$ で約分します。


$$S_B(t) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 t} - e^{-k_2 t}}{1 - e^{-k_1 t}}$$

これが選択率 $S_B(t)$ の時間変化を表す一般式です。

---

### 補足：極限挙動の確認

この式から、時間の経過に伴う選択率の変化を読み取ることができます。

* **反応初期 ($t \to 0$ のとき):**
分子・分母ともに $0$ に収束するため、ロピタルの定理を適用すると $S_B \to 1$ となります。これは、反応初期には副反応 ($B \to C$) がほとんど起きていないためです。
* **反応長時間経過 ($t \to \infty$ のとき):**
$e^{-k_1 t} \to 0$ および $e^{-k_2 t} \to 0$ となるため、$S_B \to 0$ となります。これは、最終的にすべての物質が $C$ に変化してしまうためです。

逐次反応 $A \to B \to C$ における**転化率 $X_A$ の時間変化**および、転化率を基準とした反応解析を以下に解説・導出します。

---

## 1. 転化率の時間変化の導出

### (1) 定義の確認

転化率 $X_A(t)$ は、反応開始から時間 $t$ までに反応した原料 $A$ の割合として定義されます。


$$X_A(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### (2) 濃度の時間変化の代入

回分反応器における原料 $A$ の濃度は、一次反応の速度式より以下のように表されます。


$$C_A(t) = C_{A0} e^{-k_1 t}$$

これを転化率の定義式に代入します。


$$X_A(t) = 1 - \frac{C_{A0} e^{-k_1 t}}{C_{A0}} = 1 - e^{-k_1 t}$$

これが転化率 $X_A(t)$ の時間変化を表す一般式です。

### (3) 転化率の時間変化率（速度式）

転化率の時間に対する微分（速度）は次のようになります。


$$\frac{dX_A}{dt} = \frac{d}{dt}(1 - e^{-k_1 t}) = k_1 e^{-k_1 t} = k_1 (1 - X_A)$$

---

## 2. 転化率に関する重要な特徴

1. **第2段階の反応（$k_2$）の影響を受けない**
* 原料 $A$ が $B$ に変わる最初のステップは、独立した一次反応です。そのため、転化率 $X_A$ の式には $k_2$ や物質 $B, C$ の項が一切含まれません。すなわち、**転化率の挙動は $k_1$ のみで決まります。**


2. **時間の経過に伴う挙動**
* 反応初期 ($t = 0$): $X_A = 0$（未反応）
* 長時間経過 ($t \to \infty$): $X_A \to 1$（100%反応）



---

## 3. 発展：転化率を基準とした収率・選択率の解析

反応工学では、時間を変数にするのではなく、**「原料がどこまで反応したか（＝転化率 $X_A$）」を基準にして中間生成物 $B$ の収率や選択率を評価する**ことがよく行われます。

### (1) 収率 $Y_B$ と転化率 $X_A$ の関係

時間 $t$ を消去し、$X_A$ と $Y_B$ の関係を導出します。

* $e^{-k_1 t} = 1 - X_A$ より、 $t = -\frac{1}{k_1} \ln(1 - X_A)$
* これを $B$ の濃度式 $C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$ に代入します。
* $e^{-k_2 t} = (e^{-k_1 t})^{k_2 / k_1} = (1 - X_A)^{k_2 / k_1}$ と変形できるため、収率 $Y_B$ は次のように表されます。

$$Y_B = \frac{k_1}{k_2 - k_1} \left[ (1 - X_A) - (1 - X_A)^{\frac{k_2}{k_1}} \right]$$

### (2) 最適な転化率（$X_{A,\text{opt}}$）の考察

中間生成物 $B$ の収率を最大にする「最適な転化率」は、上式を $X_A$ で微分してゼロとおくことで求められます。

$$\frac{dY_B}{dX_A} = \frac{k_1}{k_2 - k_1} \left[ -1 + \frac{k_2}{k_1} (1 - X_A)^{\frac{k_2}{k_1} - 1} \right] = 0$$

これを解くと、収率が最大になるときの転化率 $X_{A,\text{opt}}$ が得られます。

$$X_{A,\text{opt}} = 1 - \left(\frac{k_1}{k_2}\right)^{\frac{k_1}{k_2 - k_1}}$$

* **実務的な意味:**
* 「原料を完全に反応させよう（転化率 $X_A \to 1$ にしよう）」として反応を続けすぎると、中間体 $B$ まで $C$ に変わってしまい、かえって $B$ の収率が低下します。
* したがって、反応を途中で打ち切るべき「最適な転化率の限界値」が存在することが、この解析からわかります。

#【付録】
### ② 経済性・生産性（コストパフォーマンス）の最適化

実際のプラント設計では、「収率の絶対値」だけでなく「時間あたりの生産性（Space-Time Yield）」や「経済性」が最適化の基準になります。

* **バッチ反応器の稼働サイクル:** 回分反応器では、反応に要する時間 $t_{\text{react}}$ だけでなく、原料の仕込み、生成物の取り出し、洗浄などの付帯時間（downtime: $t_{\text{down}}$）が必ず発生します。
* **時間あたりの生産量 ($P_{\text{eff}}$):**

$$P_{\text{eff}} = \frac{C_B(t)}{t + t_{\text{down}}}$$



理論上の最高収率が得られる時間 $t_{\max}$ まで反応させるよりも、少し手前（あるいは状況に応じた時間）で反応を打ち切る方が、1日あたりのバッチ回転数が上がり、トータルの生産量が大きくなる場合があります。

### ③ 原料コストと分離精製コストのバランス（トレードオフの克服）

* **転化率と選択率のトレードオフ:** 高い転化率 $X_A$ を狙うほど選択率 $S_B$ は低下します（未反応の $A$ は回収して再循環させればよいが、$C$ まで副生すると損失になる）。
* **分離・精製の負荷:** 未反応の $A$ や副生物 $C$ が多く混ざると、後段の分離・精製工程（蒸留や抽出など）のエネルギーコストが跳ね上がります。「反応工程の収率」だけでなく「プラント全体のトータルコスト（反応＋分離）」を最小化することが、真の経済的最適化となります。

## 5. まとめ

* 逐次反応 $A \to B \to C$ では、転化率 $X_A$ は時間とともに単調増加し、選択率 $S_B$ は単調減少する。
* 中間生成物 $B$ の収率 $Y_B$ には必ず極大値が存在し、速度定数から決まる**最適反応時間 $t_{\max}$** が存在する。
* 「最適化」の定義は状況によって異なり、単なる収率の最大化だけでなく、**生産性（時間あたりの効率）**や**分離コストを含めた経済性全体**を総合的に評価して決定される。

**はい、その通りです。**

体積流量 $v$ が一定（液相反応や、気相でも全モル数が変わらない・あるいは容積変化が無視できる場合）であれば、$F = v C$ が成り立つため、連続系と回分系（バッチ）の定義式は濃度 $C$ を用いた**完全に同じ数式**になります。

### なぜ同じ式になるのか（対応関係）

回分系のモル数 $N$ と、連続系のモル流量 $F$ は、それぞれ以下のように濃度 $C$ に置き換えられます。

* **回分系（バッチ）**: $N_A = V \cdot C_A$ （$V$ は反応液量）
* **連続系（流通）**: $F_A = v \cdot C_A$ （$v$ は体積流量）

これをそれぞれの定義式に代入すると、体積 $V$ や体積流量 $v$ が分子・分母で相殺されるため、濃度ベースの同じ式に帰着します。

#### 1. 収率（総括・供給原料基準: $Y_B$）

* **回分系:**

$$Y_B = \frac{N_B - N_{B0}}{N_{A0}} = \frac{v C_B - v C_{B0}}{v C_{A0}} = \frac{C_B - C_{B0}}{C_{A0}}$$


* **連続系:**

$$Y_B = \frac{F_B - F_{B0}}{F_{A0}} = \frac{v C_B - v C_{B0}}{v C_{A0}} = \frac{C_B - C_{B0}}{C_{A0}}$$



#### 2. 選択率（総括・反応原料基準: $S_{B/C}$ ※提示された表の $\Phi_B$ など）

※ご提示いただいた表の $\Phi_B$ の分母は、積（$F_{A0} F_A$）ではなく「反応した原料 $A$ の量（$F_{A0} - F_A$ または $N_{A0} - N_A$）」が正しい定義です。

* **回分系:**

$$\Phi_B = \frac{N_B - N_{B0}}{N_{A0} - N_A} = \frac{C_B - C_{B0}}{C_{A0} - C_A}$$


* **連続系:**

$$\Phi_B = \frac{F_B - F_{B0}}{F_{A0} - F_A} = \frac{C_B - C_{B0}}{C_{A0} - C_A}$$



このように、**体積（流量）が一定の条件下では、回分系のモル $N$ をすべて濃度 $C$（または連続系のモル流量 $F$）に置き換えて考えることができる**ため、実質的に同じ定義として扱って問題ありません。

# 講義資料：逐次反応における最適反応時間（空間時間）の探索

〜「転化率・収率・選択率」の定義の整理と工学的トレードオフ〜

## 1. 導入：なぜ指標の定義を厳密に区別するのか？

化学反応工学において、プロセスを評価する指標には「転化率」「収率」「選択率」があります。
本講義では、学生の誤解や混同を防ぐため、以下のように定義を厳格に切り分けて議論します。

* **分母に原料 $A$（供給量または反応量）が関与するものはすべて「収率」の範疇として捉える。**
* **選択率 ($S_{B/C}$) は、反応した原料が最終的に「目的物 $B$」と「副生物 $C$」のどちらにどれだけ分配されたかという「生成物の内訳（比率）」を表す指標として定義する。**

複合反応（逐次反応）では、「原料をどれだけ反応させるか（転化率）」と「どちらの生成物に変えるか（選択率 $S_{B/C}$）」がトレードオフの関係になります。これらを定量的に評価し、プラント全体の経済性を最大化する条件を探索します。

---

## 2. 逐次反応モデルの定義と各指標の定式化

一次の逐次反応を考えます。

$$A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$$

* 初期条件 ($t = 0$): 濃度 $C_A = C_{A0}$、$C_B = 0$、$C_C = 0$
* 任意の時間 $t$ における各成分の濃度：$C_A(t), C_B(t), C_C(t)$

> **💡 【重要なお断り：回分と連続におけるベースの違いについて】**
> * **回分反応器 (Batch):** 閉じた系であるため、**濃度 $C$**（またはモル数）の時間変化として直接定式化できます。
> * **連続反応器 (CSTR / PFR):** 流体の出入りがある開いた系であるため、厳密には**モル流量 $F$ $[\text{mol/s}]$** ベースの物質収支からスタートします。
> * *ただし、本講義および本資料の導入部では、見通しを良くするため「液相・体積一定（$v = v_0$）」の系を仮定し、$F_A = v_0 C_A$ および空間時間 $\tau = V/v_0$ を用いることで回分（時間 $t$）と完全に数学的アナロジーとして扱っています。*
>
>

### ① 転化率 ($X$)

投入した原料 $A$ のうち、何%が反応したか。


$$X(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### ② 収率 ($Y$)

投入した原料 $A$ 全体に対して、目的物 $B$ がどれだけ得られたか。


$$Y(t) = \frac{C_B(t)}{C_{A0}}$$


*(※分母に $A$ が入るため、プロセス全体の経済性・生産効率を表す指標となります)*

### ③ 選択率 ($S_{B/C}$)

反応した $A$ から生成した $B$ と $C$ の分配状況を表す指標として、**$B$ と $C$ の選択率 ($S_{B/C}$)** を導入し、それぞれの成分に対する選択率を $S_B$ および $S_C$ として定義します。

* **目的物 $B$ の選択率 ($S_B$):**

$$S_B = \frac{\text{生成した } B \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_B(t)}{C_{A0} - C_A(t)}$$


* **副生物 $C$ の選択率 ($S_C$):**

$$S_C = \frac{\text{生成した } C \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_C(t)}{C_{A0} - C_A(t)}$$



> **💡 重要な関係式:**
> モル保存則（$C_{A0} = C_A + C_B + C_C$）より、反応した $A$ はすべて $B$ または $C$ に分配されるため、**常に $S_B + S_C = 1$（または 100%）**が成り立ちます。
> また、収率・転化率・選択率の間には以下の関係が成り立ちます。
>
> $$Y = X \times S_B$$
>
>

---

## 3. 濃度変化の解析解と、指標の振る舞い

各成分の濃度変化は次のように求まります（$k_1 \neq k_2$ の場合）。

* $C_A(t) = C_{A0} e^{-k_1 t}$
* $C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$
* $C_C(t) = C_{A0} - C_A(t) - C_B(t)$

時間経過に伴う各指標の変化：

* **転化率 $X(t)$:** 時間とともに単調増加し、$1$（100%）に近づく。
* **選択率 $S_B$ と $S_C$:**
* 反応初期 ($t \approx 0$) は $S_B \approx 1$、$S_C \approx 0$ です。
* 時間が経つにつれて $B$ が $C$ に変化するため、**$S_B$ は単調に減少し、$S_C$ は単調に増加**します。


* **収率 $Y(t)$:** 「増える $X$」と「減る $S_B$」の積（$Y = X \times S_B$）であるため、**途中に必ず「山なりのピーク（極大値）」が現れます。**

---

## 4. 中間体 $B$ の最大収率を与える最適反応時間 $t_{\text{opt}}$ の導出

プラントの利益（目的物 $B$ の生産量）を最大化するためには、収率 $Y(t)$ が最大のところで反応を止めなければなりません。微分して $0$ と置くことで、最適反応時間 $t_{\text{opt}}$（またはPFRの最適空間時間 $\tau_{\text{opt}}$）が求まります。

$$t_{\text{opt}} = \frac{1}{k_1 - k_2} \ln\left(\frac{k_1}{k_2}\right)$$

> **⚠️ 工学的示唆:**
> この最適時点では、**転化率 $X$ は100%未満であり、副生物 $C$ への選択率 $S_C$ が一定割合存在した状態で反応を打ち切る**ことになります。
> 「原料を完全に使い切りたい（転化率100%）」という発想ではなく、あえて未反応 $A$ をリサイクルし、不要な $C$ への変換ロスを防ぐ設計がプラントの経済性を左右します。

## 5. 自習用演習問題と解説

### 【問題】

ある化学プラントで、液相の逐次反応 $A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$ が行われています。

* 反応速度定数：$k_1 = 2.0 \text{ h}^{-1}$、$k_2 = 0.5 \text{ h}^{-1}$
* 初濃度：$C_{A0} = 1.0 \text{ kmol/m}^3$

(1) 目的物 $B$ の収率 $Y$ が最大になる最適反応時間 $t_{\text{opt}}$ $[\text{h}]$ を求めよ。

(2) その最適時点における、**転化率 $X$、収率 $Y$、および $B$ と $C$ の選択率 ($S_B, S_C$)** の値をそれぞれ求めよ。

(3) もし誤って反応時間を $t = 2.0 \text{ h}$ まで延ばしてしまった場合、これらの指標はどう変化するか計算せよ。

---

### 【解答・解説】

**(1) 最適反応時間 $t_{\text{opt}}$ の計算**


$$t_{\text{opt}} = \frac{1}{2.0 - 0.5} \ln\left(\frac{2.0}{0.5}\right) = \frac{1}{1.5} \ln(4) \approx 0.924 \text{ h}$$

**(2) $t = 0.924 \text{ h}$ における各指標の計算**

* $C_A = 1.0 \times e^{-2.0 \times 0.924} \approx 0.1575 \text{ kmol/m}^3$
* $C_B = 1.0 \times \frac{2.0}{0.5 - 2.0} (e^{-1.848} - e^{-0.462}) \approx 0.630 \text{ kmol/m}^3$
* $C_C = 1.0 - 0.1575 - 0.630 = 0.2125 \text{ kmol/m}^3$

各指標を算定します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.1575}{1.0} = 0.8425 \quad (\mathbf{84.3 \%})$$


* **収率 $Y$:**
$$Y = \frac{C_B}{C_{A0}} = \frac{0.630}{1.0} = 0.630 \quad (\mathbf{63.0 \%})$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{C_B}{C_{A0} - C_A} = \frac{0.630}{0.8425} \approx 0.748 \quad (\mathbf{74.8 \%})$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.2125}{0.8425} \approx 0.252 \quad (\mathbf{25.2 \%}) \quad (\text{確認: } S_B + S_C = 100\%)$$



**(3) 反応時間を $t = 2.0 \text{ h}$ に延長した場合**

* $C_A(2.0) = e^{-4.0} \approx 0.0183 \text{ kmol/m}^3$
* $C_B(2.0) = -\frac{4}{3} (e^{-4.0} - e^{-1.0}) \approx 0.466 \text{ kmol/m}^3$
* $C_C(2.0) = 1.0 - 0.0183 - 0.466 = 0.5157 \text{ kmol/m}^3$

各指標を計算します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.0183}{1.0} \approx 0.982 \quad (\mathbf{98.2 \%}) \quad \text{【UP】}$$


* **収率 $Y$:**
$$Y = \frac{0.466}{1.0} \approx 0.466 \quad (\mathbf{46.6 \%}) \quad \text{【DOWN】}$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{C_B}{C_{A0} - C_A} = \frac{0.466}{0.982} \approx 0.475 \quad (\mathbf{47.5 \%}) \quad \text{【DOWN】}$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.5157}{0.982} \approx 0.525 \quad (\mathbf{52.5 \%}) \quad \text{【UP】}$$



> **結論:**
> 反応時間を延ばすと転化率 $X$ は約98%に上がりますが、その代わり副生物 $C$ の選択率 $S_C$ が50%を超えて急増し（$S_B$ は47.5%に激減）、結果として目的物 $B$ の収率 $Y$ も大きく低下します。
> このように、**「生成物の内訳を示す選択率 $S_{B/C}$」と「全体効率を示す収率 $Y$」の動きを切り分けて評価する**ことが、適切なプラント設計の基本となります。

# 講義資料：逐次反応における最適反応時間（空間時間）の探索

〜「転化率・選択率・収率」の3つを同時に考えなければならない理由〜

## 1. 導入：「数学的等価性」と「工学的意義」のギャップ

化学反応工学の計算上、転化率 $X$、収率 $Y$、および目的物 $B$ の選択率 $S_B$ の間には、以下のマスター方程式が成り立ちます。

$$\text{収率 } (Y) = \text{転化率 } (X) \times \text{選択率 } (S_B)$$

数学的にはこのうち2つが分かれば残り1つは自動的に決まりますが、プラント設計の現場では**3つの異なる指標を常にセットで評価**します。

* **転化率 $X$（原料をどこまで使い切るか）：** 反応器の「大きさ（設備投資 CAPEX）」に直結する。
* **選択率 $S_B, S_C$（生成物の内訳）：** 反応した原料が「目的物 $B$」になったのか、「不要な副生物 $C$」になってしまったのかを示す。分離・精製コスト（OPEX）に直結する。
* **収率 $Y$（トータルでどれだけ目的物が得られたか）：** プラント全体の「経済性・利益率」に直結する。

---

## 2. 逐次反応モデルの定義と $X, S, Y$ の定式化

一次の逐次反応を考えます。

$$A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$$

* 初期条件 ($t = 0$): 濃度 $C_A = C_{A0}$、$C_B = 0$、$C_C = 0$
* 任意の時間 $t$ における各成分の濃度：$C_A(t), C_B(t), C_C(t)$

### ① 転化率 ($X$)

投入した $A$ のうち、反応した割合。


$$X(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}}$$

### ② 収率 ($Y$)

投入した $A$ 全体に対して、目的物 $B$ が得られた割合。


$$Y(t) = \frac{C_B(t)}{C_{A0}}$$

### ③ $B$ と $C$ の選択率 ($S_B, S_C$)

反応した $A$ の総量（$= C_{A0} - C_A(t)$）のうち、それぞれがどこに変わったかの割合。

* **目的物 $B$ の選択率 ($S_B$):**

$$S_B(t) = \frac{\text{生成した } B \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_B(t)}{C_{A0} - C_A(t)} = \frac{Y(t)}{X(t)}$$


* **副生物 $C$ の選択率 ($S_C$):**

$$S_C(t) = \frac{\text{生成した } C \text{ のモル数}}{\text{反応した } A \text{ のモル数}} = \frac{C_C(t)}{C_{A0} - C_A(t)}$$



> **💡 重要な関係式:**
> モル保存則（$C_{A0} = C_A + C_B + C_C$）より、反応した $A$ はすべて $B$ または $C$ になるため、**常に $S_B + S_C = 1$（または 100%）**が成り立ちます。つまり、$B$ の選択率が下がれば、その分だけ $C$ の選択率が必ず上がります。

---

## 3. 濃度変化の解析解と、指標の振る舞い

各成分の濃度変化は次のように求まります（$k_1 \neq k_2$ の場合）。

* $C_A(t) = C_{A0} e^{-k_1 t}$
* $C_B(t) = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$
* $C_C(t) = C_{A0} - C_A(t) - C_B(t)$

時間経過に伴う各指標の変化：

* **転化率 $X(t)$:** 時間とともに単調増加し、$1$（100%）に近づく。
* **選択率 $S_B(t)$ と $S_C(t)$:**
* 反応初期 ($t \approx 0$) は $S_B \approx 1$、$S_C \approx 0$ です。
* 時間が経つにつれて $B$ が $C$ に変わるため、**$S_B$ は単調に減少し、$S_C$ は単調に増加**します。


* **収率 $Y(t)$:** 「増える $X$」と「減る $S_B$」の積（$Y = X \times S_B$）であるため、**途中に必ず「山なりのピーク（極大値）」が現れます。**

---

## 4. 中間体 $B$ の最大収率を与える最適反応時間 $t_{\text{opt}}$ の導出

利益を最大化するためには、収率 $Y(t)$ が最大のところで反応を止めなければなりません。微分して $0$ と置くことで、最適反応時間 $t_{\text{opt}}$（またはPFRの最適空間時間 $\tau_{\text{opt}}$）が求まります。

$$t_{\text{opt}} = \frac{1}{k_1 - k_2} \ln\left(\frac{k_1}{k_2}\right)$$

> **⚠️ 工学的示唆:**
> この最適時点では、**転化率 $X$ は100%未満であり、副生物 $C$ の選択率 $S_C$ が一定割合（数％〜数十％）存在した状態で反応を打ち切る**ことになります。
> 「原料を完全に使い切りたい（転化率100%）」という発想を捨て、あえて未反応 $A$ をリサイクルし、不要な $C$ への変換ロス（選択率 $S_B$ の低下）を防ぐ設計がプラントの経済性を左右します。

## 5. 自習用演習問題と解説

### 【問題】

ある化学プラントで、液相の逐次反応 $A \xrightarrow{k_1} B (\text{目的物}) \xrightarrow{k_2} C (\text{副生物})$ が行われています。

* 反応速度定数：$k_1 = 2.0 \text{ h}^{-1}$、$k_2 = 0.5 \text{ h}^{-1}$
* 初濃度：$C_{A0} = 1.0 \text{ kmol/m}^3$

(1) 目的物 $B$ の収率 $Y$ が最大になる最適反応時間 $t_{\text{opt}}$ $[\text{h}]$ を求めよ。

(2) その最適時点における、**転化率 $X$、収率 $Y$、および $B$ と $C$ の選択率 ($S_B, S_C$)** の値をそれぞれ求めよ。

(3) もし誤って反応時間を $t = 2.0 \text{ h}$ まで延ばしてしまった場合、これらの指標はどう変化するか計算せよ。

---

### 【解答・解説】

**(1) 最適反応時間 $t_{\text{opt}}$ の計算**


$$t_{\text{opt}} = \frac{1}{2.0 - 0.5} \ln\left(\frac{2.0}{0.5}\right) = \frac{1}{1.5} \ln(4) \approx 0.924 \text{ h}$$

**(2) $t = 0.924 \text{ h}$ における各指標の計算**

* $C_A = 1.0 \times e^{-2.0 \times 0.924} \approx 0.1575 \text{ kmol/m}^3$
* $C_B = 1.0 \times \frac{2.0}{0.5 - 2.0} (e^{-1.848} - e^{-0.462}) \approx 0.630 \text{ kmol/m}^3$
* $C_C = 1.0 - 0.1575 - 0.630 = 0.2125 \text{ kmol/m}^3$

各指標を算定します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.1575}{1.0} = 0.8425 \quad (\mathbf{84.3 \%})$$


* **収率 $Y$:**
$$Y = \frac{C_B}{C_{A0}} = \frac{0.630}{1.0} = 0.630 \quad (\mathbf{63.0 \%})$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{Y}{X} = \frac{0.630}{0.8425} \approx 0.748 \quad (\mathbf{74.8 \%})$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.2125}{0.8425} \approx 0.252 \quad (\mathbf{25.2 \%}) \quad (\text{確認: } 74.8\% + 25.2\% = 100\%)$$



**(3) 反応時間を $t = 2.0 \text{ h}$ に延長した場合**

* $C_A(2.0) = e^{-4.0} \approx 0.0183 \text{ kmol/m}^3$
* $C_B(2.0) = -\frac{4}{3} (e^{-4.0} - e^{-1.0}) \approx 0.466 \text{ kmol/m}^3$
* $C_C(2.0) = 1.0 - 0.0183 - 0.466 = 0.5157 \text{ kmol/m}^3$

各指標を計算します。

* **転化率 $X$:**
$$X = \frac{1.0 - 0.0183}{1.0} \approx 0.982 \quad (\mathbf{98.2 \%}) \quad \text{【UP】}$$


* **収率 $Y$:**
$$Y = \frac{0.466}{1.0} \approx 0.466 \quad (\mathbf{46.6 \%}) \quad \text{【DOWN】}$$


* **$B$ の選択率 $S_B$:**
$$S_B = \frac{Y}{X} = \frac{0.466}{0.982} \approx 0.475 \quad (\mathbf{47.5 \%}) \quad \text{【DOWN】}$$


* **$C$ の選択率 $S_C$:**
$$S_C = \frac{C_C}{C_{A0} - C_A} = \frac{0.5157}{0.9817} \approx 0.525 \quad (\mathbf{52.5 \%}) \quad \text{【UP】}$$



> **結論:**
> 反応時間を延ばすと転化率 $X$ は約98%に上がりますが、その代わり副生物 $C$ の選択率 $S_C$ が50%を超えて急増し（$S_B$ は47.5%に激減）、結果として目的物 $B$ の収率 $Y$ も大きく低下します。
> このように、**「$B$ と $C$ の選択率がどのようにトレードオフになっているか」**を追跡することが、プラント設計において極めて重要になります。
