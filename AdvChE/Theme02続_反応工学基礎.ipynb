## 空間時間

空間時間$\tau$は
$$
\tau = ~\frac{V}{v}
$$
で定義される、時間の次元を持つ量である。

ここで、$v\ [\mathrm{m}^3 \cdot \mathrm{s}^{-1}]$は溶液の体積流量。 
$$
F_{A0} = C_{A0} v
$$
だから

先程の設計方程式Eqs. (A1),(A2)を、空間時間に置き換えると

- <h4>CSTR:</h4>

$$\tau=\frac{X}{k(1-X)}$$

- <h4>PFR:</h4>

$$\tau=\frac{1}{k}\ln\left(\frac{1}{1-X}\right)$$



◇◇◇
<h3>（発展）</h3>

反応速度がn次反応の場合

$$
-r_A = kC_A^n
$$
で、液相・等密度系
$$
C_A=C_{A0}(1-X)
$$
を仮定すると、それぞれの設計方程式は次のようになります。


<h5>CSTR</h5>

$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \frac{X}{(1-X)^n} }
$$

<h5>PFR</h5>

$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$


<details>
<summary>📝🖱️参考（導出）</summary>

<h5> 1. CSTR</h5>

設計方程式
$$
V=\frac{F_{A0}X}{-r_A}
$$
に代入すると
$$
V = \frac{F_{A0}X} {k[C_{A0}(1-X)]^n}
$$
したがって
$$
\boxed{ V= \frac{F_{A0}}{kC_{A0}^{\,n}} \frac{X}{(1-X)^n} }
$$
となります。

空間時間

$$
\tau=\frac{V}{v_0}, \qquad F_{A0}=v_0C_{A0}
$$
を用いると
$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \frac{X}{(1-X)^n} }
$$

<h5>2. PFR</h5>

設計方程式
$$
V = F_{A0} \int_0^X \frac{dX}{-r_A}
$$
へ代入すると
$$
V = \frac{F_{A0}}{kC_{A0}^{\,n}} \int_0^X \frac{dX}{(1-X)^n}
$$
となります。

(1) $n\neq1$

積分すると
$$
\int (1-X)^{-n}dX = \frac{(1-X)^{\,1-n}}{n-1}
$$
したがって
$$
V = \frac{F_{A0}} {kC_{A0}^{\,n}} \left[ \frac{(1-X)^{1-n}-1} {n-1} \right]
$$
よって
$$
\boxed{ V= \frac{F_{A0}} {kC_{A0}^{\,n}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$
空間時間表示では
$$
\boxed{ \tau= \frac{1}{kC_{A0}^{\,n-1}} \, \frac{(1-X)^{1-n}-1} {n-1} }
$$

(2) $n=1$のときは先程と同じ
