### 1. PFR/リサイクルプロセス

逐次反応　$\text{A} \to \text{B} \to \text{C}$における$\text{B}$の収率改善に、リサイクルが有効かを考えます。

まずは、ブロックフローダイアグラム（BFD）を示します。

このBFD（Block Flow Diagram）は、**「原料が入り、反応し、生成物を分離し、一部を再び反応器へ戻す」化学プロセス全体の流れ**を表したものです。

- FRESH FEED : 原料Aの流入, $F_{A1}$
- MIXER: 新規原料Aとリサイクル流を混合, $F_{A2} = F_{A1} + F_R$
- PFR INLET：$F_{\text{A2}}$でPFRに送液
- PFR:逐次反応（等温）
- PFR OUTLET: 反応により最大3成分の液が排出$F_{A3},\  F_{B3},\  F_{C3}$
- SEPARATOR:
  - 製品$\text B$は完全に分離するものとして$F_{\text P}=F_{B3}$
  - 反応成分$\text A$ の分離・回収効率を $\eta_A$とし、$F_\text R = F_{A3} \cdot\eta_A$でリサイクルへ。
  - 未回収$\text A$と$\text C$は系外へ排出されます。$F_{\text{PURGE}}=F_{A3} \cdot(1-\eta_A) + F_{C3}$

  ---







### 2. 命題の確認

>「リサイクルをした場合、原料転化率$X_\text A$と製品収率 $Y_\text B$が改善されるか？」

---
### 2.1 用語の確認
今回のように反応器に分離装置とリサイクルなどを追加した場合、プロセスとなります。本プロセスにおける転化率と収率は、これまでの反応器だけを考慮したものと異なります。
 - プロセス全体：総括転化率、総括収率
 - 反応器だけ：単パス転化率、単パス収率

 と呼ぶことにします。

### 2.2 A転化率

(1) 定義の確認

* **総括転化率**：

$$X_{\mathrm{overall}} = \frac{F_{A1} - F_{AP}}{F_{A1}}$$



（※ $F_{AP}$ は、図には示しておりませんが、パージストリーム（系外へ排出される流れ）に含まれる未反応Aのモル流量です）
* **1パスあたりの転化率**：

$$X_{\mathrm{pass}} = \frac{F_{A2} - F_{A3}}{F_{A2}}$$



ここから、反応器出口のAの流量 $F_{A3}$ は次のように変形できます。

$$F_{A3} = (1 - X_{\mathrm{pass}}\ ) F_{A2} \tag{1}$$



---

(2) パージされるAの量 $F_{AP}$

セパレータに流入した未反応のA ($F_{A3}$) のうち、リサイクルされる割合が $\eta_A$ です。したがって、**リサイクルされずにパージ（系外へ排出）される割合は $(1 - \eta_A)$** となります。

よって、パージ中のAの流量 $F_{AP}$ は次のように表せます。


$$F_{AP} = (1 - \eta_A) F_{A3}$$


ここに先ほどの (1)式 を代入します。


$$F_{AP} = (1 - \eta_A) (1 - X_{\mathrm{pass}}\ ) F_{A2} \tag{2}$$

---

(3) 総括転化率の分子 ($F_{A1} - F_{AP}$) を計算する

詳細は【参考】をみてください。物質収支式から$F_{A2}$は次のようになります。

$$
F_{A2}
=
\frac{F_{A1}}
{1-\eta_A(1-X_{\mathrm{pass}}\ )} \tag{3}
$$


これと 式(2) を使って、総括転化率の分子である $(F_{A1} - F_{AP})$ を計算します。


$$
\begin{aligned}
F_{A1} - F_{AP} &= F_{A2} \left[ 1 - \eta_A (1 - X_{\mathrm{pass\ }}\ ) \right] - (1 - \eta_A) (1 - X_{\mathrm{pass\ }}\ ) F_{A2}\\
&= F_{A2} \left[ 1 - \eta_A (1 - X_{\mathrm{pass}}\ ) - (1 - \eta_A) (1 - X_{\mathrm{pass}}\ ) \right]\\
&=F_{A2} X_{\mathrm{pass}}
\end{aligned}
$$

---

(4) 最終的な式

総括転化率の定義式に、分子（$F_{A2} X_{\mathrm{pass}}\ $）と分母（式(3)で表した $F_{A1}$）を代入し,最終的に次の式が得られます。

$$
\begin{aligned}
X_{\mathrm{overall}} &= \frac{F_{A1} - F_{AP}}{F_{A1}} \\
&= \frac{X_{\mathrm{pass}}}{1 - \eta_A (1 - X_{\mathrm{pass}}\ )}
\end{aligned} \tag{4}
$$

ここで思い出してください。
逐次反応$A\to B\to C$での、Aの単パス転化率とBの単パス収率は
$$
X_{\mathrm{pass}}
=
1-
e^{-k_1\tau}
$$


でしたね（講義資料Theme6参照）。
これを代入すれば、解析的に計算を実行することができます。




### 2.3 B収率

* **総括収率**：

Bは全量製品として回収されるので、総括収率は

$$
Y_{\mathrm{overall}}
=
\frac{F_\text P}{F_\text{A1}}
$$

* **1パスあたりの収率**：
$$
Y_{\mathrm{pass}}
=
\frac{F_{B3}}{F_{A2}}
$$

今回のBFDの場合、PFRで生成した$B$は,すべて製品になる特殊な例ですので$F_{B3} = F_P$。従って、$Y_{\mathrm{overall}}\ $と$Y_{\mathrm{pass}}\ $には

$$
Y_{\mathrm{overall}}
=
\frac{F_\text{A2}}{F_\text{A1}} Y_{\mathrm{pass}}
$$

の関係があります。
ここで、先ほどの式(3)の関係式を使うと、最終的に

$$
Y_{\mathrm{overall}}
=
\frac{Y_{\mathrm{pass}}}
{1-\eta_A(1-X_{\mathrm{pass}}\ )} \tag{5}
$$

回収率の式(4)と同じようにスッキリした形にできます。

これと
逐次反応$A\to B\to C$での、Aの単パス転化率とBの単パス収率の式
$$
X_{\mathrm{pass}}
=
1-
e^{-k_1\tau}
$$
$$
Y_{\mathrm{pass}}
=
\frac{k_1}{k_2-k_1}
\left(
e^{-k_1\tau}
-
e^{-k_2\tau}
\right)
$$
を使うと解析的に計算を実行することができます。

---

### 【参考】$F_{A2}$の導出
セパレーターでAを $\eta_A$ の割合で回収するので、

$$
F_\text R=\eta_A F_{A3}
$$
さらに、

$$
F_{A3}
=
F_{A2}(1-X_{\mathrm{pass}}\ )
$$

なので、

$$
F_\text R
=
\eta_A F_{A2}(1-X_{\mathrm{pass}}\ )
$$

となります。

次にMixerの収支

$$
F_{A2}=F_{A1}+F_\text R
$$

に代入すると、

$$
F_{A2}
=
F_{A1}
+
\eta_A F_{A2}(1-X_{\mathrm{pass}}\ )
$$

です。

整理すると、


$$
F_{A2}
=
\frac{F_{A1}}
{1-\eta_A(1-X_{\mathrm{pass}}\ )}
$$





となります。

---

```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# 1. パラメータの設定（等温・一定）
# ============================================================

# 反応：A → B → C
k1 = 1.0       # A → B の速度定数 [1/s]
k2 = 0.5       # B → C の速度定数 [1/s]

# Fresh Feed
F_A0 = 1.0     # 新鮮原料Aのモル流量 [mol/s]

# SeparatorにおけるAの回収率
eta_A = 0.95   # Aの回収率（95%）

# 空間時間（滞留時間）
tau_list = np.linspace(0.01, 5.0, 300)


# ============================================================
# 2. Single-pass PFR の計算
# ============================================================

# Aの単パス転化率
X_pass = 1 - np.exp(-k1 * tau_list)

# Bの単パス収率
if k1 == k2:
    Y_pass = k1 * tau_list * np.exp(-k1 * tau_list)
else:
    Y_pass = (
        k1 / (k2 - k1)
        * (np.exp(-k1 * tau_list) - np.exp(-k2 * tau_list))
    )


# ============================================================
# 3. プロセス全体の物質収支（資料の導出式を使用）
# ============================================================

# 分母共通項： 1 - eta_A * (1 - X_pass)
denom = 1.0 - eta_A * (1.0 - X_pass)

# 総括転化率（資料の式(4)）
Overall_Conversion_A = X_pass / denom

# 総括収率（資料の式(5)）
Overall_Yield_B = Y_pass / denom


# ============================================================
# 4. グラフ描画
# ============================================================

plt.figure(figsize=(10, 6))

# Single-pass
plt.plot(
    tau_list,
    X_pass * 100,
    label='Single-Pass Conversion ($X_{pass}$)',
    linestyle='-.',
    linewidth=2
)

plt.plot(
    tau_list,
    Y_pass * 100,
    label='Single-Pass Yield of B ($Y_{pass}$)',
    linestyle='--',
    linewidth=2
)

# Overall
plt.plot(
    tau_list,
    Overall_Conversion_A * 100,
    label='Overall Conversion ($X_{overall}$)',
    linewidth=2.5
)

plt.plot(
    tau_list,
    Overall_Yield_B * 100,
    label='Overall Yield of B ($Y_{overall}$)',
    linewidth=2.5
)

plt.xlabel(
    r'Space Time / Residence Time ($\tau$ [s])',
    fontsize=12
)

plt.ylabel(
    'Percentage [%]',
    fontsize=12
)

plt.title(
    'Comparison: Single-Pass vs. Overall (Recycle System)',
    fontsize=14
)

plt.grid(
    True,
    linestyle=':',
    alpha=0.7
)

plt.legend(
    fontsize=11,
    loc='upper right'
)

plt.ylim(-5, 105)
plt.tight_layout()
plt.show()


# ============================================================
# 5. Overall B収率が最大となる条件
# ============================================================

max_idx = np.argmax(Overall_Yield_B)

print(
    f"=== Recycle System Optimal Point "
    f"(at tau = {tau_list[max_idx]:.3f} s) ==="
)
print(
    f"  Single-Pass Conversion (X_pass) "
    f": {X_pass[max_idx]*100:.1f}%"
)
print(
    f"  Single-Pass Yield (Y_pass) "
    f": {Y_pass[max_idx]*100:.1f}%"
)
print(
    f"  Overall Conversion (X_overall) "
    f": {Overall_Conversion_A[max_idx]*100:.1f}%"
)
print(
    f"  Overall Yield with Recycle (Y_overall) "
    f": {Overall_Yield_B[max_idx]*100:.1f}%"
)
print(
    "--------------------------------------------------"
)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ============================================================
# 1. パラメータの設定（等温・一定）
# ============================================================

# 反応：A → B → C
k1 = 1.0       # A → B の速度定数 [1/s]
k2 = 0.5       # B → C の速度定数 [1/s]

# Fresh Feed
F_A0 = 1.0     # 新鮮原料Aのモル流量 [mol/s]

# SeparatorにおけるAの回収率
eta_A = 0.95   # Aの回収率（95%）

# 空間時間（滞留時間）
tau_list = np.linspace(0.01, 5.0, 300)


# ============================================================
# 2. Single-pass PFR
# ============================================================

# PFRでのAの単パス転化率
#
# F_A2 → PFR → F_A3
#
# X_pass = (F_A2 - F_A3) / F_A2

X_pass = 1 - np.exp(-k1 * tau_list)


# ------------------------------------------------------------
# PFR出口におけるBの単パス収率
#
# Y_pass = F_B3 / F_A2
# ------------------------------------------------------------

if k1 == k2:
    Y_pass = k1 * tau_list * np.exp(-k1 * tau_list)

else:
    Y_pass = (
        k1 / (k2 - k1)
        * (np.exp(-k1 * tau_list)
           - np.exp(-k2 * tau_list))
    )


# ============================================================
# 3. Recycleを含むプロセス全体の物質収支
# ============================================================

# ------------------------------------------------------------
# Mixerの物質収支
#
# F_A2 = F_A0 + F_3
#
# また、
#
# F_3 = eta_A * F_A3
#
# F_A3 = F_A2 (1 - X_pass)
#
# したがって
#
# F_A2 =
# F_A0 / [1 - eta_A (1 - X_pass)]
# ------------------------------------------------------------

F_A2 = F_A0 / (
    1.0 - eta_A * (1.0 - X_pass)
)


# ------------------------------------------------------------
# PFR出口の未反応A
#
# F_A3 = F_A2 (1 - X_pass)
# ------------------------------------------------------------

F_A3 = F_A2 * (1.0 - X_pass)


# ------------------------------------------------------------
# PFR出口のB
#
# F_B3 = F_A2 Y_pass
#
# BはSeparatorで全量Productへ回収される
# ------------------------------------------------------------

F_B3 = F_A2 * Y_pass


# ------------------------------------------------------------
# SeparatorからRecycleされるA
#
# F_3 = eta_A F_A3
# ------------------------------------------------------------

F_3 = eta_A * F_A3


# ------------------------------------------------------------
# WASTEへ排出される未回収A
#
# F_A,W = (1 - eta_A) F_A3
# ------------------------------------------------------------

F_A_W = (1.0 - eta_A) * F_A3


# ------------------------------------------------------------
# PFR出口のC
#
# F_A2 = F_A3 + F_B3 + F_C3
#
# より
#
# F_C3 = F_A2 - F_A3 - F_B3
# ------------------------------------------------------------

F_C3 = F_A2 - F_A3 - F_B3


# ------------------------------------------------------------
# WASTE STREAM
#
# WASTEには
#
#   ・未回収A
#   ・C
#
# が入る。
#
# F_W3 = F_A,W + F_C3
# ------------------------------------------------------------

F_W3 = F_A_W + F_C3


# ============================================================
# 4. Overallのプロセス指標
# ============================================================

# ------------------------------------------------------------
# Overall B Yield
#
# Bは全量Productとして回収されるので
#
# Y_overall = F_B3 / F_A0
# ------------------------------------------------------------

Overall_Yield_B = F_B3 / F_A0


# ------------------------------------------------------------
# Overall A Conversion
#
# 系外へ出る未反応AはWASTE中のF_A,W
#
# X_overall
# = (F_A0 - F_A,W) / F_A0
# ------------------------------------------------------------

Overall_Conversion_A = (
    F_A0 - F_A_W
) / F_A0


# ============================================================
# 5. グラフ描画
# ============================================================

plt.figure(figsize=(10, 6))

# Single-pass
plt.plot(
    tau_list,
    X_pass * 100,
    label='Single-Pass Conversion ($X_{pass}$)',
    linestyle='-.',
    linewidth=2
)

plt.plot(
    tau_list,
    Y_pass * 100,
    label='Single-Pass Yield of B ($Y_{pass}$)',
    linestyle='--',
    linewidth=2
)


# Overall
plt.plot(
    tau_list,
    Overall_Conversion_A * 100,
    label='Overall Conversion ($X_{overall}$)',
    linewidth=2.5
)

plt.plot(
    tau_list,
    Overall_Yield_B * 100,
    label='Overall Yield of B ($Y_{overall}$)',
    linewidth=2.5
)


plt.xlabel(
    r'Space Time / Residence Time ($\tau$ [s])',
    fontsize=12
)

plt.ylabel(
    'Percentage [%]',
    fontsize=12
)

plt.title(
    'Comparison: Single-Pass vs. Overall (Recycle System)',
    fontsize=14
)

plt.grid(
    True,
    linestyle=':',
    alpha=0.7
)

plt.legend(
    fontsize=11,
    loc='upper right'
)

plt.ylim(-5, 105)

plt.tight_layout()

plt.show()


# ============================================================
# 6. Overall B収率が最大となる条件
# ============================================================

max_idx = np.argmax(Overall_Yield_B)

print(
    f"=== Recycle System Optimal Point "
    f"(at tau = {tau_list[max_idx]:.3f} s) ==="
)

print(
    f"  Single-Pass Conversion (X_pass) "
    f": {X_pass[max_idx]*100:.1f}%"
)

print(
    f"  Single-Pass Yield (Y_pass) "
    f": {Y_pass[max_idx]*100:.1f}%"
)

print(
    f"  Overall Conversion (X_overall) "
    f": {Overall_Conversion_A[max_idx]*100:.1f}%"
)

print(
    f"  Overall Yield with Recycle (Y_overall) "
    f": {Overall_Yield_B[max_idx]*100:.1f}%"
)

print(
    "--------------------------------------------------"
)
```

##【予備】

今回のノート（ブロック図）で使われている変数（$F_{A1}, F_{A2}, F_{A3}, F_R, \eta_A, X_{\mathrm{pass}}$）を使って、「回収率（$\eta_A$）」**と、一般に化学工学で「リサイクル比（循環比）」と呼ばれる**「リサイクル率 ($R = F_R / F_{A1}$）」の関係式をきれいに導出することができます。

---

### 1. 使用する定義の確認（ノートより）

* **新鮮原料に対するリサイクル率（リサイクル比 $R$）**：

$$R = \frac{F_R}{F_{A1}}$$


* **回収率（$\eta_A$）**：セパレータでの未反応Aの回収効率


$$F_R = \eta_A F_{A3} \quad \text{--- (①)}[cite: 2]$$


* **1パスあたりの転化率から得られる反応器出口のA ($F_{A3}$)**：

$$F_{A3} = F_{A2} (1 - X_{\mathrm{pass}}) \quad \text{--- (②)}[cite: 2]$$


* **ミキサー（Mixer）の物質収支**：

$$F_{A2} = F_{A1} + F_R \quad \text{--- (③)}[cite: 2]$$



---

### 2. 関係式の導出ステップ

#### ① $F_R$ を $F_{A1}$ の式で表す

まず、①式に②式を代入します。


$$F_R = \eta_A F_{A3} = \eta_A F_{A2} (1 - X_{\mathrm{pass}})$$

次に、③式の $F_{A2} = F_{A1} + F_R$ をここに代入します。


$$F_R = \eta_A (F_{A1} + F_R) (1 - X_{\mathrm{pass}})$$

#### ② $F_R$ について整理する

右辺を展開して、 $F_R$ の項を左辺にまとめます。


$$F_R = \eta_A F_{A1} (1 - X_{\mathrm{pass}}) + \eta_A F_R (1 - X_{\mathrm{pass}})$$

$$F_R \left[ 1 - \eta_A (1 - X_{\mathrm{pass}}) \right] = \eta_A F_{A1} (1 - X_{\mathrm{pass}})$$

#### ③ リサイクル率 $R$ の形にする

両辺を $F_{A1}$ で割ると、左辺にリサイクル率 $R \left( = \frac{F_R}{F_{A1}} \right)$ が現れます。


$$R = \frac{F_R}{F_{A1}} = \frac{\eta_A (1 - X_{\mathrm{pass}})}{1 - \eta_A (1 - X_{\mathrm{pass}})}$$

---

### 3. 相互の変換式（まとめ）

#### A. 回収率 $\eta_A$ と 1パス転化率 $X_{\mathrm{pass}}$ から「リサイクル率 $R$」を求める式

$$R = \frac{\eta_A (1 - X_{\mathrm{pass}})}{1 - \eta_A (1 - X_{\mathrm{pass}})}$$

* **意味**：回収率 $\eta_A$ を高くするほど、あるいは1パス転化率 $X_{\mathrm{pass}}$ が低くなって未反応物が増えるほど、新鮮原料に対して何倍ものリサイクル量 ($R$) が必要になることを表しています。

#### B. 逆に「回収率 $\eta_A$」をリサイクル率 $R$ で表す式

上の式を $\eta_A$ について解きほぐすと、次のように変形することもできます。


$$\eta_A = \frac{R}{(1 + R) (1 - X_{\mathrm{pass}})}$$

* **意味**：プロセス全体で特定のリサイクル率 $R$ を実現するために、セパレータがどれだけの回収率 $\eta_A$ を持たなければならないかの目安になります。
