# 並列反応の速度式と微分方程式

定容積回分反応系において、次の並列一次反応をおこないます。

$$
\begin{aligned}
A &\xrightarrow{r_1} B\\
A &\xrightarrow{r_2} C
\end{aligned}
\tag{1}
$$

ここで、各反応の反応速度を

$$
r_1=k_1C_A
$$

$$
r_2=k_2C_A
$$

とする。また、反応器内は完全混合されており、温度は一定であるとします。

**初期条件**は

$$
C_A(0)=C_{A0},\qquad
C_B(0)=0,\qquad
C_C(0)=0
$$

とする。

---

>## 問題1　支配方程式の立式
>
>各成分について物質収支を行い、連立微分方程式を導きなさい。
>
>$\text A$,$\text B$, $\text C$ の濃度変化を表す微分方程式
>
>$$
\frac{dC_A}{dt}=\\
\frac{dC_B}{dt}=\\
\frac{dC_C}{dt}=\\
$$
>
>を $r_1,r_2$ を用いて表しなさい。
>
>続けて、
>$$
r_1=k_1C_A,\qquad r_2=k_2C_A
$$
>
>を代入した連立微分方程式も示しなさい。

---

## 解答例　問１


定容積回分反応器における各成分の物質収支（単位体積・単位時間あたりの変化量）は、「生成速度の総和 - 消費速度の総和」で表されます。

* **$A$ の物質収支:**
$A$ は反応1と反応2の両方で消費されるため、

$$\frac{dC_A}{dt} = -r_1 - r_2$$


* **$B$ の物質収支:**
$B$ は反応1によって生成されるため、

$$\frac{dC_B}{dt} = r_1$$


* **$C$ の物質収支:**
$C$ は反応2によって生成されるため、

$$\frac{dC_C}{dt} = r_2$$



これらに $r_1 = k_1 C_A, \quad r_2 = k_2 C_A$ を代入すると、以下の連立微分方程式が得られます。

$$\begin{aligned} \frac{dC_A}{dt} &= -(k_1 + k_2)C_A \\ \frac{dC_B}{dt} &= k_1 C_A \\ \frac{dC_C}{dt} &= k_2 C_A \end{aligned}$$

---



# 問題2　解析解を求める

>作成した連立微分方程式を初期条件（冒頭表示）のもとで解きなさい。

---

## 解答例 問2

### 1. $C_A(t)$ の導出

$A$ に関する微分方程式は単独で解くことができます。


$$\frac{dC_A}{dt} = -(k_1 + k_2)C_A$$

変数分離法を用いて積分します。初期条件 $C_A(0) = C_{A0}$ を用います。


$$\int_{C_{A0}}^{C_A} \frac{dC_A}{C_A} = -\int_0^t (k_1 + k_2) dt$$

$$\ln\left(\frac{C_A}{C_{A0}}\right) = -(k_1 + k_2)t$$

したがって、$C_A(t)$ は次のように求められます。


$$C_A(t) = C_{A0} e^{-(k_1 + k_2)t}$$


### 2. $C_B(t)$ の導出

$B$ に関する微分方程式に求めた $C_A(t)$ を代入します。


$$\frac{dC_B}{dt} = k_1 C_{A0} e^{-(k_1 + k_2)t}$$

初期条件 $C_B(0) = 0$ のもとで両辺を $0$ から $t$ まで積分します。


$$C_B(t) = \int_0^t k_1 C_{A0} e^{-(k_1 + k_2)\tau} d\tau$$

$$C_B(t) = k_1 C_{A0} \left[ \frac{-1}{k_1 + k_2} e^{-(k_1 + k_2)\tau} \right]_0^t$$

整理すると、


$$C_B(t) = \frac{k_1 C_{A0}}{k_1 + k_2} \left( 1 - e^{-(k_1 + k_2)t} \right)$$


### 3. $C_C(t)$ の導出

$C$ についても同様に、$C_A(t)$ を代入して解きます。


$$\frac{dC_C}{dt} = k_2 C_{A0} e^{-(k_1 + k_2)t}$$

初期条件 $C_C(0) = 0$ のもとで積分します。


$$C_C(t) = \int_0^t k_2 C_{A0} e^{-(k_1 + k_2)\tau} d\tau$$

整理すると、


$$C_C(t) = \frac{k_2 C_{A0}}{k_1 + k_2} \left( 1 - e^{-(k_1 + k_2)t} \right)$$

以上まとめると

$$\begin{aligned} C_A(t) &= C_{A0} e^{-(k_1 + k_2)t} \\ C_B(t) &= \frac{k_1 C_{A0}}{k_1 + k_2} \left( 1 - e^{-(k_1 + k_2)t} \right) \\ C_C(t) &= \frac{k_2 C_{A0}}{k_1 + k_2} \left( 1 - e^{-(k_1 + k_2)t} \right) \end{aligned}$$

## 解答例 問3

```python
#@title 逐次反応の解析解と数値解の比較
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# --- 1. パラメータと初期条件の設定 ---
k1 = 0.2   # 反応速度定数 k1 [1/min]
k2 = 0.1   # 反応速度定数 k2 [1/min]
CA0 = 1.0  # 初期濃度 C_A(0) [mol/L]

t_span = (0, 20)            # 計算時間範囲 [0, 20]
t_eval = np.linspace(0, 20, 200) # グラフ描画用の時間点

# --- 2. 微分方程式の定義（数値計算用） ---
#def reaction_system(t, C):
#    CA, CB, CC = C
#    dCA_dt = -k1 * CA
#    dCB_dt = k1 * CA - k2 * CB
#    dCC_dt = k2 * CB
#    return [dCA_dt, dCB_dt, dCC_dt]

def reaction_system(t, C):
    CA, CB, CC = C
    dCA_dt = -(k1 + k2) * CA
    dCB_dt = k1 * CA
    dCC_dt = k2 * CA
    return [dCA_dt, dCB_dt, dCC_dt]

# --- 3. 数値解の計算 (Scipy solve_ivp) ---
C0 = [CA0, 0.0, 0.0]  # 初期条件 [CA0, CB0, CC0]
sol = solve_ivp(reaction_system, t_span, C0, t_eval=t_eval, method='RK45')

# --- 4. 解析解の計算 ---
#t_exact = t_eval
#CA_exact = CA0 * np.exp(-k1 * t_exact)
#CB_exact = (k1 * CA0 / (k2 - k1)) * (np.exp(-k1 * t_exact) - np.exp(-k2 * t_exact))
#CC_exact = CA0 * (1 - (k2 / (k2 - k1)) * np.exp(-k1 * t_exact) + (k1 / (k2 - k1)) * np.exp(-k2 * t_exact))

# --- 4. 解析解の計算 ---
t_exact = t_eval
CA_exact = CA0 * np.exp(-(k1 + k2) * t_exact)
CB_exact = (k1 * CA0 / (k1 + k2)) * (1 - np.exp(-(k1 + k2) * t_exact))
CC_exact = (k2 * CA0 / (k1 + k2)) * (1 - np.exp(-(k1 + k2) * t_exact))



# --- 5. 比較グラフの描画 ---
plt.figure(figsize=(10, 6))

# 数値解のプロット（点）
plt.plot(sol.t, sol.y[0], 'o', label='$C_A$ (Numerical)', markersize=4, color='tab:blue')
plt.plot(sol.t, sol.y[1], 's', label='$C_B$ (Numerical)', markersize=4, color='tab:orange')
plt.plot(sol.t, sol.y[2], '^', label='$C_C$ (Numerical)', markersize=4, color='tab:green')

# 解析解のプロット（実線）
plt.plot(t_exact, CA_exact, '-', label='$C_A$ (Analytical)', color='navy', linewidth=1.5)
plt.plot(t_exact, CB_exact, '-', label='$C_B$ (Analytical)', color='darkorange', linewidth=1.5)
plt.plot(t_exact, CC_exact, '-', label='$C_C$ (Analytical)', color='darkgreen', linewidth=1.5)

# グラフ装飾
plt.xlabel('Time $t$ [min]', fontsize=12)
plt.ylabel('Concentration [mol/L]', fontsize=12)
plt.title('Comparison between Numerical and Analytical Solutions', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# 表示
plt.show()
```
