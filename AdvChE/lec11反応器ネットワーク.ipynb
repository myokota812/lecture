## 1. 反応ネットワーク
A→B, A→C, B→D, C→D

## 2. 量論行列
$$S=\begin{bmatrix}
-1&-1&0&0\\
1&0&-1&0\\
0&1&0&-1\\
0&0&1&1
\end{bmatrix}$$

## 3. 速度ベクトル
$$r_1=k_1C_A$$
$$r_2=k_2C_A$$
$$r_3=k_3C_B$$
$$r_4=k_4C_C$$

## 4. 行列表現
$$\frac{d\mathbf C}{dt}=\mathbf S\mathbf r$$

```python
import numpy as np
S=np.array([[-1,-1,0,0],[1,0,-1,0],[0,1,0,-1],[0,0,1,1]])
print(S)
print("rank=",np.linalg.matrix_rank(S))
```

## 5. 数値シミュレーション

```python
import numpy as np
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
k1=1.0;k2=0.5;k3=0.4;k4=0.2
def f(t,y):
 CA,CB,CC,CD=y
 r1=k1*CA;r2=k2*CA;r3=k3*CB;r4=k4*CC
 return [-r1-r2,r1-r3,r2-r4,r3+r4]
sol=solve_ivp(f,[0,15],[1,0,0,0],dense_output=True)
t=np.linspace(0,15,300)
CA,CB,CC,CD=sol.sol(t)
plt.plot(t,CA,label="A")
plt.plot(t,CB,label="B")
plt.plot(t,CC,label="C")
plt.plot(t,CD,label="D")
plt.legend();plt.grid();plt.show()
```

## 6. 可逆反応
$$A\rightleftharpoons B$$

$$r=k_fC_A-k_rC_B$$

## 7. 平衡条件
$$k_fC_A=k_rC_B$$

$$K=\frac{k_f}{k_r}=\frac{C_B}{C_A}$$

```python
import numpy as np
kf=2.0; kr=0.5
K=kf/kr
print("Equilibrium constant=",K)
```
