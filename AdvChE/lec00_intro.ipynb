本講義では、化学プロセスにおける反応装置の設計法について勉強します。

学部の「反応工学」の講義では、各種反応器について、設計方程式を立式し、そこから体積を求める方法を勉強したと思います。このことは非常に大切ですが、それは出発点にすぎないということを、本講義を通じて理解してもらえればと思っております。
本講義では、
まず反応速度を定式化し、必要な反応器容量を求め、その後に熱移動、流動状態、触媒内拡散、さらには多相系やプロセス全体の最適化へと発展していきます（予定です）。

反応工学の学習内容を大まかに整理すると次のようになります。

| 項目 | 内容 | 学習キーワード |
|--------|--------|--------|
| 1. 反応を数式化する | 反応速度式を構築し、速度定数や反応機構を求める。反応器設計の基礎となる。 | 反応速度論、速度解析、反応機構、Arrhenius式 |
| 2. 反応器サイズを決める | 設計方程式から必要な反応器体積や滞留時間を計算する。 | 物質収支、回分反応器、CSTR、PFR、Levenspielプロット |
| 3. 熱を扱う | 発熱・吸熱反応における温度変化や除熱性能を評価する。 | エネルギー収支、多重定常状態、熱暴走 |
| 4. 実際の流れを考慮する | 理想反応器とのずれを評価する。 | 滞留時間分布（RTD）、非理想流動、混合 |
| 5. 触媒を考慮する | 触媒粒子内部への物質移動や反応との競合を評価する。 | 有効係数、内部拡散、Thiele数、触媒設計 |
| 6. 多相系へ拡張する | 気液系、気固系、液固系などの反応を扱う。 | 吸収、反応吸収、流動層、スラリー反応器 |
| 7. システムとして最適化する | 分離・リサイクルを含めたプロセス全体を設計する。 | PBR、反応分離、リサイクル、プロセス最適化 |
| 8. 実機設計へ展開する | 装置仕様や運転条件を決定し、工業化へつなげる。 | スケールアップ、安全設計、運転解析 |


  

---

プロセス設計における **設計方程式**  は、

$$
V=F_{A_0}\int_0^X\frac{dX}{-r_A}
$$

のような式から反応器体積を求めるところまでで終わりません。

実際のプラント設計では、

1. 反応速度式を決定する
2. 反応器体積を求める
3. 熱設計を行う
4. 機械設計を行う
5. 計装・安全設計を行う
6. 機器仕様書としてまとめる

という流れで具体的な装置設計へと展開されます。

したがって、反応工学は単なる体積計算ではなく、

**「化学反応を実際の工業装置として実現するための総合工学」**

といえます。

反応器仕様書には、プロセス条件、機械仕様、伝熱仕様、計装仕様、安全仕様などが記載されています。

# 反応工学の全体構造

反応工学は大きく三つの階層に分けられます。

## 第1層：反応速度論

「どれくらい速く反応するのか」を扱う。

主な内容は、

- 速度式
- 反応次数
- Arrhenius式
- 反応機構
- 速度定数の推定

です。

ここで得られた速度式が、後続のすべての設計の出発点となります。

---

## 第2層：反応器設計

「どのような反応器をどれだけの大きさで作るか」を扱う。

主な内容は、

- 物質収支
- 設計方程式
- 空間時間
- 滞留時間
- Levenspielプロット
- 反応器の比較

である。

ここでは求めた速度式を設計に利用します。

---

## 第3層：反応プロセス設計

「製品を最も効率よく作るシステムをどう構築するか」を扱います。

主な内容は、

- 分離操作
- リサイクル
- エネルギー回収
- 多段反応器
- プロセス最適化

です。

実際の化学プラント設計では、この段階まで考慮してはじめて経済的なプロセスとなります。

---

# 講義の最終目標

本講義の最終目標は、

$$
\text{速度式}
\rightarrow
\text{設計方程式}
\rightarrow
\text{反応器設計}
\rightarrow
\text{プロセス設計}
$$

という一連の流れを理解することです。

単に設計方程式を解けるようになるだけでなく、

- どの反応器を選ぶべきか
- どの条件で運転すべきか
- どのように選択率を向上させるか
- 熱暴走をどう防ぐか
- リサイクルをどう設計するか

を論理的に判断できるようになることを目指します。

---

# 反応工学が扱う典型的な問題

- 指定した転化率を達成するにはどれだけの反応器容積が必要か
- CSTRとPFRのどちらが有利か
- 発熱反応で熱暴走は起こらないか
- 目的生成物の選択率を最大化するにはどうすればよいか
- 触媒内部の拡散は無視できるか
- リサイクルを導入すると生産性はどう変わるか
- 分離工程まで含めると最適条件は何か

これらの問題を体系的に解決することが反応工学の目的である。
