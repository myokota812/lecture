$$r_B=3C_A^2$$

$$r_C=2C_A$$

### 解答

## 問題3 転化率

一次反応について

$$\tau=\frac1k\ln\frac1{1-X}$$

を用いて X=0.8, k=0.5 min^{-1} の空間時間を求めよ。

```python
import numpy as np
k=0.5
X=0.8
tau=(1/k)*np.log(1/(1-X))
print("tau=",tau)
```

## 問題4 逐次反応

$$A\rightarrow B\rightarrow C$$

B最大条件を求めよ。

### 解答
$$k_1C_A=k_2C_B$$

```python
import numpy as np
k1=1.0;k2=0.3
t=np.linspace(0,10,1000)
CB=k1/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
idx=np.argmax(CB)
print("Optimal time=",t[idx])
print("Maximum CB=",CB[idx])
```

## 問題5 リサイクル反応器

$$X_o=\frac{(1+R)X_s}{1+RX_s}$$

```python
R=3
Xs=0.5
Xo=(1+R)*Xs/(1+R*Xs)
print("Overall conversion=",Xo)
```

## 問題6 Arrhenius式

$$k=k_0e^{-E/RT}$$

```python
import numpy as np
R=8.314
E=80000
k0=1e10
T=500
k=k0*np.exp(-E/(R*T))
print("k=",k)
```
