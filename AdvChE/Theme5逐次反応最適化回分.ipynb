## 1. 複合反応の物質収支式一般式から回分反応器の設計方程式の導出

化学反応工学において、単一の反応だけでなく、複数の反応が同時に進行する複合反応（Complex Reaction）系を取り扱うことは極めて重要です。本資料では、一般的な複合反応系における物質収支式の一般式を出発点とし、**回分反応器（Batch Reactor）の設計方程式**がどのように導出されるかを体系的に解説します。

---

## 1.1. 複合反応における物質収支の一般表現

系内に $N$ 種類の化学種（$i = 1, 2, \dots, N$）が存在し、$M$ 個の独立した化学反応（$j = 1, 2, \dots, M$）が同時に進行している系を考えます。

### (1) 化学量論係数と反応速度の定義

$j$ 番目の反応式を一般的に次のように表します。

$$\sum_{i=1}^{N} \nu_{i,j} A_i = 0 \quad (j = 1, 2, \dots, M)$$

* $A_i$：化学種 $i$
* $\nu_{i,j}$：化学種 $i$ の化学量論係数（生成物は正、消費（原料）は負）
* $r_j$：$j$ 番目の反応の反応速度（単位体積・単位時間あたりの反応量、$[ \mathrm{mol/(m^3 \cdot s)} ]$）

### (2) 一般的な物質収支式（系全体の物質収支）

任意の反応器において、微小時間 $dt$ の間の化学種 $i$ の物質量 $n_i$ の変化は、「流出入による増減」と「反応による生成・消費」の和で表されます。

$$\frac{dn_i}{dt} = F_{i,\text{in}} - F_{i,\text{out}} + V \sum_{j=1}^{M} \nu_{i,j} r_j$$

* $n_i$：化学種 $i$ のモル数 $[\mathrm{mol}]$
* $F_{i,\text{in}}, F_{i,\text{out}}$：化学種 $i$ の流入・流出モル速度 $[\mathrm{mol/s}]$
* $V$：反応液の体積 $[\mathrm{m^3}]$

---

## 1.2. 回分反応器（Batch Reactor）への適用と設計方程式の導出

### (1) 回分系の基本仮定

回分反応器は閉ざされた系（Closed System）であるため、反応器外部との物質の出入りが一切ありません。

* 流入・流出項がゼロになります： $F_{i,\text{in}} = 0, \ F_{i,\text{out}} = 0$

したがって、上記の一般物質収支式は次のように簡略化されます。

$$\frac{dn_i}{dt} = V \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

これが、**複合反応における回分反応器の微分型設計方程式の一般形**です。

---

### (2) 濃度基準の設計方程式への変換（定容系の場合）

反応液の体積 $V$ が時間によらず一定（$V = V_0 = \text{一定}$、定容系）であると仮定します。
濃度 $C_i = \frac{n_i}{V_0}$ を用いて式を変形すると、両辺を $V_0$ で割ることで濃度に関する微分方程式が得られます。

$$\frac{dC_i}{dt} = \sum_{j=1}^{M} \nu_{i,j} r_j \quad (i = 1, 2, \dots, N)$$

これが、回分反応器の設計や解析で最も一般的に使用される方程式です。

---


## 1.3. 具体例：逐次反応 $A \to B \to C$ への適用確認

これまでの一般式が、これまで扱ってきた逐次反応にどのように当てはまるかを確認します。

* 化学種 ($N=3$): $A$ ($i=1$), $B$ ($i=2$), $C$ ($i=3$)
* 反応 ($M=2$):
* 反応1 ($j=1$): $A \to B \quad (r_1 = k_1 C_A)$
* 反応2 ($j=2$): $B \to C \quad (r_2 = k_2 C_B)$



化学量論係数 $\nu_{i,j}$ をマトリクス形式で整理すると：

* $A$ について： $\nu_{1,1} = -1, \ \nu_{1,2} = 0$
* $B$ について： $\nu_{2,1} = +1, \ \nu_{2,2} = -1$
* $C$ について： $\nu_{3,1} = 0, \ \nu_{3,2} = +1$

定容系における一般式 $\frac{dC_i}{dt} = \sum_{j=1}^{M} \nu_{i,j} r_j$ に代入すると、おなじみの以下の3元連立微分方程式が完全に導出されます。

1. **成分 $A$:** $\frac{dC_A}{dt} = (-1)r_1 + (0)r_2 = -k_1 C_A$
2. **成分 $B$:** $\frac{dC_B}{dt} = (+1)r_1 + (-1)r_2 = k_1 C_A - k_2 C_B$
3. **成分 $C$:** $\frac{dC_C}{dt} = (0)r_1 + (+1)r_2 = k_2 C_B$

## 2. 転化率、収率、選択率

本講義では、次のような一次の不可逆逐次反応を考えます。

$$A \xrightarrow{k_1} B \xrightarrow{k_2} C$$

* $A$：原料
* $B$：目的とする中間生成物
* $C$：副生物（過剰反応による分解物など）
* $k_1, k_2$：それぞれの反応速度定数

初期の $A$ の濃度を $C_{A0}$ とし、体積変化は無視できる（定容系）と仮定します。また、反応開始時 ($t = 0$) において $B$ および $C$ は存在しないものとします。

---

## 2.1. 諸定義の確認

反応の進行度や目的物質の生産効率を評価するため、以下の3つの指標を定義します。

### (1) 転化率 (Conversion) $X_A$

反応した原料 $A$ の割合を示します。


$$X_A(t) = \frac{C_{A0} - C_A(t)}{C_{A0}} = 1 - \frac{C_A(t)}{C_{A0}} \tag{eqX}$$

### (2) 収率 (Yield) $Y_B$

仕込んだ原料 $A$ に対し、目的物質 $B$ がどれだけ得られたかの割合です。


$$Y_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{仕込んだ } A \text{ の初期物質量}} = \frac{C_B(t)}{C_{A0}} \tag{eqY}$$

### (3) 選択率 (Selectivity) $S_B$

反応した原料 $A$ のうち、どれだけが目的物質 $B$ に変化したかの割合です。


$$S_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{反応した } A \text{ の物質量}} = \frac{C_B(t)}{C_{A0} - C_A(t)} = \frac{Y_B(t)}{X_A(t)} \tag{eqS}$$

---

## 2.2. 定式化

回分反応器における各成分の物質収支式（微分方程式）は以下の通りです。

$$\frac{dC_A}{dt} = -k_1 C_A$$

$$\frac{dC_B}{dt} = k_1 C_A - k_2 C_B$$

$$\frac{dC_C}{dt} = k_2 C_B$$

これらを解くことで、各変数の時間変化を明らかにできます。<br>
導出は実施済みなので省略。

* **原料 $A$:**

$$C_A = C_{A0} e^{-k_1 t}\tag{1}$$


* **中間生成物 $B$ ($k_1 \neq k_2$ の場合):**

$$C_B = C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)\tag{2}$$


* **副生物 $C$:**

$$C_C = C_{A0} \left( 1 - \frac{k_2 e^{-k_1 t} - k_1 e^{-k_2 t}}{k_2 - k_1} \right)\tag{3}$$

---

### (1) 転化率 $X_A$ の時間変化

原料 $A$ の濃度は(1)式より、転化率は次のように表されます。


$$X_A(t) = 1 - e^{-k_1 t} \tag{4}$$

* **特徴:** 転化率の挙動は第2段階の反応速度定数 $k_2$ に依存せず、$k_1$ のみで決定されます。時間が経つにつれて $X_A \to 1$（100%）に漸近します。


### (2) 収率 $Y_B(t)$ の時間変化の導出

#### 1. 定義の確認

収率 $Y_B(t)$ は、「仕込んだ原料 $A$ の初期物質量に対する、生成した目的物質 $B$ の割合」として定義されます。定容系（体積変化がない系）を仮定すると、濃度を用いて次のように表されます。


$$Y_B(t) = \frac{\text{生成した } B \text{ の物質量}}{\text{仕込んだ } A \text{ の初期物質量}} = \frac{C_B(t)}{C_{A0}}$$


回分反応器における$B$ の濃度 $C_B(t)$ はEq. (2)（$k_1 \neq k_2$ の場合）であるので、これを収率の定義式(eqY)に代入します。


$$Y_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0}}$$


分子・分母の $C_{A0}$ で約分すると、収率 $Y_B(t)$ の時間変化を表す一般式が得られます。


$$Y_B(t) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) \tag{5}$$

---
### 補足：挙動の特徴

* **反応初期 ($t = 0$ のとき):**
$e^0 = 1$ より、$Y_B(0) = \frac{k_1}{k_2 - k_1} (1 - 1) = 0$ となり、まだ $B$ は生成していません。
* **反応長時間経過 ($t \to \infty$ のとき):**
$e^{-k_1 t} \to 0, e^{-k_2 t} \to 0$ となるため、$Y_B(\infty) \to 0$ となります。最終的にすべての $B$ が $C$ に変化するため、収率はゼロに戻ります。
* **極大値の存在:**
時間が経つにつれて $Y_B(t)$ は一旦増加してピーク（極大値）を迎えた後、減少に転じます。このピークを与える時間が「収率の最適反応時間 $t_{\max}$」となります。



---

### (3) 選択率 $S_B(t)$ の時間変化の導出

選択率の定義式(eqS)に $C_A(t)$ および $C_B(t)$ の式を代入します。


$$S_B(t) = \frac{C_{A0} \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)}{C_{A0} - C_{A0} e^{-k_1 t}}$$

分子・分母の $C_{A0}$ で約分すると、選択率の時間変化を表す一般式が得られます。


$$S_B(t) = \frac{k_1}{k_2 - k_1} \cdot \frac{e^{-k_1 t} - e^{-k_2 t}}{1 - e^{-k_1 t}} \tag{6}$$

* **挙動の傾向:**
* 反応初期 ($t \to 0$): 副反応 ($B \to C$) がほとんど起きていないため、$S_B \to 1$（100%）となります。
* 反応長時間経過 ($t \to \infty$): すべての物質が $C$ に変化してしまうため、$S_B \to 0$ となります。すなわち、選択率は時間とともに単調減少します。

## 3. なにをもって最適化とするか（最適化の基準）

化学工業やプラント運転において「最適化」を行う際、何を基準にするか（目的関数）が極めて重要になります。逐次反応における最適化には、主に以下の視点が存在します。

## 3.1. 目的物質 $B$ の収率（または生産量）の最大化

技術的・化学的な観点における最も一般的な最適化基準です。

* **課題:** 反応時間が短すぎると原料 $A$ が残ってしまい、長すぎると $B$ がさらに反応して $C$ になってしまうため、収率 $Y_B$ には必ず極大値が存在します。
* **最適反応時間 $t_{\max}$ の導出:**
$\frac{dC_B}{dt} = 0$ となる条件から、収率を最大にする時間が求められます。

$$t_{\max} = \frac{1}{k_2 - k_1} \ln\left(\frac{k_2}{k_1}\right) \tag{7}$$


* **最大収率 $Y_{B,\max}$:**

$$Y_{B,\max} = \left(\frac{k_1}{k_2}\right)^{\frac{k_2}{k_2 - k_1}} \tag{8}$$

【注意】(7),(8)式の導出は本日の課題とも連動しますので、省略します。課題で解いてみてください。

---





一例を示します。任意のパラメーターを与え、式(4),(5),(6)をグラフにするとともに、Bの収率最大とした場合の、$t_\text{max}$と$Y_\text{B,max}$を図示するコード（Code1)を用意しました。実行してください。

```python
#@title CODE 1：Bの収率最大化
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 時間軸の設定 (t = 0 から 5 まで)
t = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A の計算（※タイポ修正: 1 - np.exp）
X_A = 1 - np.exp(-k1 * t)

# 2. 収率 (Yield) Y_B の計算
if k1 == k2:
    Y_B = k1 * t * np.exp(-k1 * t)
    t_max = 1 / k1
    Y_B_max = np.exp(-1)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))
    t_max = (1 / (k2 - k1)) * np.log(k2 / k1)
    Y_B_max = (k1 / k2) ** (k2 / (k2 - k1))

# 3. 選択率 (Selectivity) S_B の計算 (t=0での0/0除算エラーを防ぐ処理)
S_B = np.ones_like(t)
nonzero = X_A > 1e-6
S_B[nonzero] = Y_B[nonzero] / X_A[nonzero]

# --- 追加: t_max における X_A と S_B の計算 ---
X_A_tmax = 1 - np.exp(-k1 * t_max)
S_B_tmax = Y_B_max / X_A_tmax

print(f"=== at t_max = {t_max:.3f} ===")
print(f"  Conversion (X_A): {X_A_tmax:.3f}")
print(f"  Yield      (Y_B): {Y_B_max:.3f}")
print(f"  Selectivity (S_B): {S_B_tmax:.3f}")
#-----------------------------------------------

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(t, X_A, label=r'Conversion ($X_A$)', color='blue', linewidth=2)
plt.plot(t, Y_B, label=r'Yield ($Y_B$)', color='green', linewidth=2)
plt.plot(t, S_B, label=r'Selectivity ($S_B$)', color='orange', linewidth=2, linestyle='--')

# 最適反応時間 t_max の明示（縦線とアノテーション）
plt.axvline(x=t_max, color='red', linestyle=':', linewidth=1.5, label=r'$t_{\max}$')
plt.scatter([t_max], [Y_B_max], color='green', zorder=5)
plt.scatter([t_max], [X_A_tmax], color='blue', zorder=5)
plt.scatter([t_max], [S_B_tmax], color='orange', zorder=5)

# アノテーションに t_max 時の X_A, Y_B, S_B をすべて記載
annotation_text = (
    f'Max Yield Point ($t_{{max}} = {t_max:.2f}$)\n'
    f'  $Y_{{B,\\max}} = {Y_B_max:.2f}$\n'
    f'  $X_A(t_{{max}}) = {X_A_tmax:.2f}$\n'
    f'  $S_B(t_{{max}}) = {S_B_tmax:.2f}$'
)

plt.annotate(annotation_text,
             xy=(t_max, Y_B_max),
             xytext=(t_max + 0.3, Y_B_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

# グラフの装飾（英語化）
plt.title(r'Consecutive Reaction ($A \to B \to C$) Analysis ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Reaction Time ($t$)', fontsize=12)
plt.ylabel('Fraction / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.1)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
```

## 3.2. 転化率、収率（および選択率）の最大化

逐次反応（$A \to B \to C$）において、転化率 $X_A$ と収率 $Y_B$ の**重み付き線形結合**を目的関数（Objective Function）として設定し、それを最大化するアプローチを解説します。

実際の化学プラントでは、「原料をどれだけ反応させたか（転化率 $X_A$）」と「目的物質がどれだけ得られたか（収率 $Y_B$）」の双方が経済性に直結します。未反応の $A$ は回収して再利用できますが、副生物 $C$ まで進んでしまうと回収が難しく損失になります。このトレードオフを評価するための経済モデルや最適化基準として、線形結合を用いた目的関数がよく使われます。

---

## 3.2.1. 目的関数の定義

転化率 $X_A(t)$ と収率 $Y_B(t)$ を用いて、目的関数 $J(t)$ を次のように定義します。

$$J(t) = w_1 X_A(t) + w_2 Y_B(t)$$

* $w_1$：転化率に対する重み（例：原料 $A$ を反応させること自体のインセンティブ、あるいは未反応 $A$ の回収コストに対するペナルティの兼ね合い）
* $w_2$：収率に対する重み（例：目的物質 $B$ の市場価値や販売価格）

ここで、$X_A(t) = 1 - e^{-k_1 t}$ および $Y_B(t) = \frac{k_1}{k_2 - k_1} (e^{-k_1 t} - e^{-k_2 t})$ を代入すると、目的関数は時間 $t$ の関数として次のように表されます。

$$J(t) = w_1 (1 - e^{-k_1 t}) + w_2 \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

---

## 3.2.2. 目的関数の最大化（最適反応時間の探索）

この目的関数 $J(t)$ を最大化する最適反応時間 $t_{\text{opt}}$ は、時間微分 $\frac{dJ}{dt} = 0$ となる条件から求められます。

$$\frac{dJ}{dt} = w_1 k_1 e^{-k_1 t} + w_2 \frac{k_1}{k_2 - k_1} \left( -k_1 e^{-k_1 t} + k_2 e^{-k_2 t} \right) = 0$$

この式を整理すると、単なる収率単体の最大化（$w_1 = 0$ の場合）とは異なり、**重みのバランス（$w_1 / w_2$）によって最適な反応時間が変化する**ことがわかります。

* $w_1$（転化率の重み）を大きくすると、原料をより多く消費する（長めの反応時間）方向に最適値がシフトします。
* $w_2$（収率の重み）を大きくすると、副生物 $C$ への過剰反応を嫌うため、収率のピークに近づき短めの反応時間で打ち切るようになります。

## 3.2.3. Pythonによる実装とグラフ描画コード

重み付き目的関数 $J(t)$ を計算し、その最大値と最適反応時間をプロットするPythonコードです。


```python
#@title CODE 2:目的関数の最大化
import matplotlib.pyplot as plt
import numpy as np

# 速度定数の設定
k1 = 1.0  # A -> B の速度定数
k2 = 0.5  # B -> C の速度定数

# 重み係数の設定（例: w1 = 0.3 (転化率の価値), w2 = 1.0 (目的物質Bの価値)）
w1 = 0.3
w2 = 0.7

# 時間軸の設定 (t = 0 から 5 まで)
t = np.linspace(0, 5, 500)

# 1. 転化率 (Conversion) X_A と 収率 (Yield) Y_B の計算（※タイポ修正: 1 - np.exp）
X_A = 1 - np.exp(-k1 * t)
if k1 == k2:
    Y_B = k1 * t * np.exp(-k1 * t)
else:
    Y_B = (k1 / (k2 - k1)) * (np.exp(-k1 * t) - np.exp(-k2 * t))

# 2. 重み付き目的関数 J(t) の計算
J = w1 * X_A + w2 * Y_B

# 3. 目的関数 J(t) を最大化する最適時間 t_opt の数値的な探索
max_index = np.argmax(J)
t_opt = t[max_index]
J_max = J[max_index]

# --- 追加: t_opt における X_A, Y_B, S_B の計算 ---
X_A_topt = 1 - np.exp(-k1 * t_opt)
if k1 == k2:
    Y_B_topt = k1 * t_opt * np.exp(-k1 * t_opt)
else:
    Y_B_topt = (k1 / (k2 - k1)) * (np.exp(-k1 * t_opt) - np.exp(-k2 * t_opt))

# 選択率 S_B の計算（t=0での除算エラーを防ぐ処理）
S_B_topt = Y_B_topt / X_A_topt if X_A_topt > 1e-6 else 1.0

print(f"=== at t_opt = {t_opt:.3f} ===")
print(f"  Conversion (X_A): {X_A_topt:.3f}")
print(f"  Yield      (Y_B): {Y_B_topt:.3f}")
print(f"  Selectivity (S_B): {S_B_topt:.3f}")
print(f"  Objective  (J):   {J_max:.3f}")
#-----------------------------------------------

# グラフ描画
plt.figure(figsize=(10, 6))
plt.plot(t, X_A, label=r'Conversion ($X_A$)', color='blue', alpha=0.6, linestyle=':')
plt.plot(t, Y_B, label=r'Yield ($Y_B$)', color='green', alpha=0.6, linestyle=':')
plt.plot(t, J, label=rf'Objective Function $J(t) = {w1}X_A + {w2}Y_B$', color='purple', linewidth=2.5)

# 最適点 (t_opt, J_max) および各値のプロット
plt.axvline(x=t_opt, color='red', linestyle='--', linewidth=1.5, label=f'Optimal Time $t_{{opt}} = {t_opt:.2f}$')
plt.scatter([t_opt], [J_max], color='purple', zorder=5)
plt.scatter([t_opt], [X_A_topt], color='blue', zorder=5)
plt.scatter([t_opt], [Y_B_topt], color='green', zorder=5)

# アノテーションに t_opt 時の J, X_A, Y_B, S_B をすべて記載
annotation_text = (
    f'Optimal Time ($t_{{opt}} = {t_opt:.2f}$)\n'
    f'  $J_{{max}} = {J_max:.2f}$\n'
    f'  $X_A(t_{{opt}}) = {X_A_topt:.2f}$\n'
    f'  $Y_B(t_{{opt}}) = {Y_B_topt:.2f}$\n'
    f'  $S_B(t_{{opt}}) = {S_B_topt:.2f}$'
)

plt.annotate(annotation_text,
             xy=(t_opt, J_max),
             xytext=(t_opt + 0.3, J_max - 0.25),
             arrowprops=dict(facecolor='red', shrink=0.05, width=1, headwidth=6),
             fontsize=10, color='darkred',
             bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="red", lw=1, alpha=0.9))

# グラフの装飾
plt.title(r'Optimization using Weighted Objective Function ($k_1=' + str(k1) + r', k_2=' + str(k2) + r'$)', fontsize=14)
plt.xlabel('Reaction Time ($t$)', fontsize=12)
plt.ylabel('Value / Ratio (-)', fontsize=12)
plt.ylim(-0.05, 1.2)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11, loc='upper right')
plt.tight_layout()

plt.show()
```

### 1. 「1%の改善・悪化」で何が変わるか？

* **年間生産量**: $10,000 \text{ トン} = 10,000,000 \text{ kg}$
* **1%の変動**:

$$10,000,000 \text{ kg} \times 0.01 = 100,000 \text{ kg} \text{（100トン）}$$



収率がわずか1%変わるだけで、年間で **100トン（10万kg）** 分の製品が多く手に入る、あるいは失われることになります。

### 2. 金額換算の具体例

化学工業における製品の単価は物質によってさまざまですが、代表的なスケールで計算すると以下のようになります。

#### パターンA：機能性化学品・工業用素材（単価：1kgあたり 1,000円 の場合）

少し付加価値の高い化学品や樹脂などの場合です。

* 計算：$100,000 \text{ kg} \times 1,000 \text{ 円/kg} = 100,000,000 \text{ 円}$
* **結果**: **年間 1億円** の売上（または原料ロスの金額）の差になります。

#### パターンB：高機能なファインケミカルズ（単価：1kgあたり 5,000円 の場合）

電子材料や特殊な添加剤などの場合です。

* 計算：$100,000 \text{ kg} \times 5,000 \text{ 円/kg} = 500,000,000 \text{ 円}$
* **結果**: **年間 5億円** のインパクトになります。

#### パターンC：医薬品中間体（※規模を少し小さくし、年間1,000トン生産の場合）

医薬品の中間体は非常に高価であり、1kgあたり数万円することが珍しくありません（ここでは 1kgあたり 20,000円 と仮定）。

* 年間生産量 1,000トン の 1% ＝ $10 \text{ トン} = 10,000 \text{ kg}$
* 計算：$10,000 \text{ kg} \times 20,000 \text{ 円/kg} = 200,000,000 \text{ 円}$
* **結果**: 生産量が小さくても、**年間 2億円** もの売上増減に直結します。

---
