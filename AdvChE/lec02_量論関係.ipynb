## 1. 化学量論式

$$
aA+bB 
\rightarrow cC+dD$$

化学量論式は反応物と生成物の量的関係を表す。

## 2. 一般化量論係数

$$\nu_A=-a$$
$$\nu_B=-b$$
$$\nu_C=c$$
$$\nu_D=d$$

$$\sum_j \nu_j A_j = 0$$

## 3. 反応進行度

$$dN_j=\nu_j d\xi$$

$$N_j=N_{j0}+\nu_j\xi$$

## 4. 巨視的量論関係

$$
\frac{N_A-N_{A0}}{-a}
=
\frac{N_B-N_{B0}}{-b}
=
\frac{N_C-N_{C0}}{c}
=
\frac{N_D-N_{D0}}{d}
=
\xi
$$

## 5. 反応速度の量論関係（微視的量論関係）

$$r_A=-ar$$
$$r_B=-br$$
$$r_C=cr$$
$$r_D=dr$$

$$
\frac{r_A}{-a}
=
\frac{r_B}{-b}
=
\frac{r_C}{c}
=
\frac{r_D}{d}
=
r
$$

$$r_j=\nu_j r$$

## 6. 例題

$$A+3B\rightarrow2C+D$$

$$r_A=-r$$
$$r_B=-3r$$
$$r_C=2r$$
$$r_D=r$$

## 7. ベクトル表現

$$
\mathbf r_s=
\begin{bmatrix}
r_A\\r_B\\r_C\\r_D
\end{bmatrix}
=r
\begin{bmatrix}
\nu_A\\\nu_B\\\nu_C\\\nu_D
\end{bmatrix}
$$

## 8. 複合反応と量論ベクトル

逐次反応

$$A\rightarrow B\rightarrow C$$

$$
\boldsymbol{\nu}_1=
\begin{bmatrix}
-1\\1\\0
\end{bmatrix}
$$

$$
\boldsymbol{\nu}_2=
\begin{bmatrix}
0\\-1\\1
\end{bmatrix}
$$

## 9. 量論行列

$$
\mathbf S=
\begin{bmatrix}
-1 & 0\\
1 & -1\\
0 & 1
\end{bmatrix}
$$

## 10. 量論行列による物質収支

$$\mathbf N=\mathbf N_0+\mathbf S\boldsymbol\xi$$

## 11. 速度ベクトル

$$
\mathbf r=
\begin{bmatrix}
r_1\\r_2
\end{bmatrix}
$$

$$\mathbf r_s=\mathbf S\mathbf r$$

## 12. 反応速度式との結合 逐次反応の例

$$A\rightarrow B\rightarrow C$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

$$\frac{d\mathbf C}{dt}=\mathbf S\mathbf r$$

$$\frac{dC_A}{dt}=-r_1$$
$$\frac{dC_B}{dt}=r_1-r_2$$
$$\frac{dC_C}{dt}=r_2$$

```python
import numpy as np
S=np.array([[-1,0],[1,-1],[0,1]])
print(S)
```

```python
import numpy as np
import matplotlib.pyplot as plt
k1=1.0;k2=0.3;CA0=1.0
t=np.linspace(0,10,400)
CA=CA0*np.exp(-k1*t)
CB=k1*CA0/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
CC=CA0-CA-CB
plt.plot(t,CA,label="A")
plt.plot(t,CB,label="B")
plt.plot(t,CC,label="C")
plt.legend();plt.grid();plt.show()
```

## 量論行列のランクと独立反応数
ランクは一次独立な量論ベクトルの数であり、独立反応数を与える。

### 例1
$$S=\begin{bmatrix}-1&0\\1&-1\\0&1\end{bmatrix}$$

$$\mathrm{rank}(S)=2$$

### 例2
$$S=\begin{bmatrix}-1&0&-1\\1&-1&0\\0&1&1\end{bmatrix}$$

$$\nu_3=\nu_1+\nu_2$$

$$\mathrm{rank}(S)=2$$

## 並列反応の量論行列
$$A\rightarrow B$$
$$A\rightarrow C$$

$$S=\begin{bmatrix}-1&-1\\1&0\\0&1\end{bmatrix}$$

$$\frac{dC_A}{dt}=-r_1-r_2$$
$$\frac{dC_B}{dt}=r_1$$
$$\frac{dC_C}{dt}=r_2$$

## 逐次反応の量論行列
$$A\rightarrow B\rightarrow C$$

$$S=\begin{bmatrix}-1&0\\1&-1\\0&1\end{bmatrix}$$

$$r_1=k_1C_A$$
$$r_2=k_2C_B$$

$$\frac{dC}{dt}=Sr$$

## Python演習1: ランク計算

```python
import numpy as np
S=np.array([[-1,0],[1,-1],[0,1]])
print(S)
print(np.linalg.matrix_rank(S))
```

## Python演習2: 並列反応

```python
import numpy as np
S=np.array([[-1,-1],[1,0],[0,1]])
print(S)
print(np.linalg.matrix_rank(S))
```

## Python演習3: 成分速度

```python
import numpy as np
S=np.array([[-1,0],[1,-1],[0,1]])
r=np.array([3,1])
print(S@r)
```
