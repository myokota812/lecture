問1. 一次反応($A \rightarrow B \,\left(-r_A=kC_A \right) $  )で $X=0.8$ を達成するためのCSTRとPFRの空間時間を求めよ。求めた２つの空間時間を比較してCSTRとPFRのどちらが優れているか論じなさい。ただし、同じ$v$で操作した場合を前提にします。

問2. 反応
$$
A \rightarrow \text{Products}
$$

の速度式が$n$次

$$
-r_A=kC_A^n
$$
の場合、反応次数によってＰＦＲとＣＴＲの優位性がどううなるかを比較します。出力されるグラフを見てそこから得られる知見を記述しなさい。



```python
#@title演習２のpythonコード
import numpy as np
import matplotlib.pyplot as plt

k = 1.0
CA0 = 1.0

X = np.linspace(0.001, 0.95, 500)

def tau_pfr(X, n, k=1.0, CA0=1.0):

    if np.isclose(n, 0):
        return X / (k / CA0)

    elif np.isclose(n, 1):
        return np.log(1/(1-X)) / k

    else:
        return ((1-X)**(1-n)-1)/(
            (n-1)*k*CA0**(n-1)
        )

def tau_cstr(X, n, k=1.0, CA0=1.0):

    if np.isclose(n, 0):
        return X/(k/CA0)

    return X/(
        k*CA0**(n-1)*(1-X)**n
    )

orders = [-1, 0,1,2]

fig, axes = plt.subplots(
    2, 2,
    figsize=(10,8),
    sharex=True
)

axes = axes.flatten()

for ax, n in zip(axes, orders):

    ax.plot(
        X,
        tau_pfr(X,n,k,CA0),
        lw=3,
        label="PFR"
    )

    ax.plot(
        X,
        tau_cstr(X,n,k,CA0),
        "--",
        lw=3,
        label="CSTR"
    )

    ax.set_title(f"n={n}")
    ax.grid(True)
    ax.legend()

    ax.set_xlabel("Conversion X")
    ax.set_ylabel(r"$\tau$")

plt.tight_layout()
plt.show()
```
