# <span style="color:rgb(213,80,0)">CSTRにおける複合反応安定判別</span>


$$\begin{array}{l} A\;\overset{k_1 }{\to} \;B\newline 2A\;\overset{k_2 }{\to} \;P \end{array}$$

について、数値的に定常状態を求め、固有値で安定性を判別します。

　まず、各成分の設計方程式

$$\begin{array}{l} \frac{dC_A }{dt}=\frac{F}{V}\left(C_{Ain} -C_A \right)-k_1 C_A -2k_2 C_A^2 ,\newline \frac{dC_B }{dt}=-\frac{F}{V}C_B +k_1 C_A ,\newline \frac{dC_P }{dt}=-\frac{F}{V}C_P +k_2 C_A^2 . \end{array}$$

次に、Jacobianを導出します。

一般に ３変数の ODE 系

 $ $ \begin{array}{l} \frac{d\mathbf{C}}{dt}=\mathbf{f}(\mathbf{C}),~~\mathbf{C}=\left\lbrack \begin{array}{rr} C_{\textrm{A}} \newline C_{\textrm{B}} \newline C_{\textrm{P}}  \end{array}\right\rbrack \\
\end{array} $ $

に対して、Jacobian $J$ は

$$J=\frac{\partial \mathbf{f}}{\partial \mathbf{C}}=\left\lbrack \begin{array}{rrr} \frac{\partial f_{\textrm{A}} }{\partial C_{\textrm{A}} } & \frac{\partial f_{\textrm{A}} }{\partial C_{\textrm{B}} } & \frac{\partial f_{\textrm{A}} }{\partial C_{\textrm{P}} }\newline \frac{\partial f_{\textrm{B}} }{\partial C_{\textrm{A}} } & \frac{\partial f_{\textrm{B}} }{\partial C_{\textrm{B}} } & \frac{\partial f_{\textrm{B}} }{\partial C_{\textrm{P}} }\newline \frac{\partial f_{\textrm{P}} }{\partial C_{\textrm{A}} } & \frac{\partial f_{\textrm{P}} }{\partial C_{\textrm{B}} } & \frac{\partial f_{\textrm{P}} }{\partial C_{\textrm{P}} } \end{array}\right\rbrack$$

となります。

ここで各要素の偏微分は、冒頭の設計方程式から次のように求まります。

$$\begin{array}{l} f_A (C_A ,C_B ,C_P )=\frac{F}{V}\,(C_{Ain} -C_A )-k_1 C_A -2k_2 C_A^2 \newline ~~\frac{\partial f_A }{\partial C_A }=-\frac{F}{V}-k_1 -4k_2 C_A \newline ~~\frac{\partial f_A }{\partial C_B }=0\newline ~~\frac{\partial f_A }{\partial C_A }=0\newline  \end{array}$$

$$\begin{array}{l} f_B (C_A ,C_B ,C_P )=-\frac{F}{V}C_B +k_1 C_A \newline ~~\frac{\partial f_B }{\partial C_A }=k_1 \newline ~~\frac{\partial f_B }{\partial C_B }=-\frac{F}{V}\newline ~~\frac{\partial f_B }{\partial C_P }=0\newline  \end{array}$$

$$\begin{array}{l} f_P (C_A ,C_B ,C_P )=-\frac{F}{V}C_P +k_2 C_A^2 \newline ~~\frac{\partial f_P }{\partial C_A }=2k_2 C_A \newline ~~\frac{\partial f_P }{\partial C_B }=0\newline ~~\frac{\partial f_P }{\partial C_P }=-\frac{F}{V} \end{array}$$

したがってJacobianは

$$\mathbf{J}=\left\lbrack \begin{array}{ccc} -F/V-k_1 -4k_2 C_A  & 0 & 0\newline k_1  & -F/V & 0\newline 2k_2 C_A  & 0 & -F/V \end{array}\right\rbrack$$

固有値 $\lambda$ は

$$\det (\mathbf{J}-\lambda \mathbf{I})=0$$

を満たす数です。一般的にはこれを計算するわけですが、幸運なことに、行列 $\mathbf{J}$ は下三角行列になっているため、固有値は

$$\lambda_1 =-F/V-k_1 -4k_2 C_A$$

$$\lambda_2 =-\frac{F}{V}$$

でことが、すぐにわかります。

#### 安定条件はすべての固有値の実部が負である

したがって

###
$$\lambda_1 =-F/V-k_1 -4k_2 C_A <0$$

####
$$\lambda_2 =-F/V<0$$

は自明であるので、

$$\begin{array}{l} A\;\overset{k_1 }{\to} \;B\newline 2A\;\overset{k_2 }{\to} \;P \end{array}$$

のCSTRは常に安定していることがわかります。

CSTRの物質収支式からヤコビ行列とその固有値を解析的に導出するMATLABコード


```matlab
%% CSTR複合反応系のヤコビ行列と固有値（シンボリック解析）
```

```matlab
clear; clc;


%% 記号変数の定義
syms F V k1 k2 CA CB CP CAin real


%% 反応速度式（ODE）
fA =  F/V*(CAin - CA) - k1*CA - 2*k2*CA^2;   % dCA/dt
fB = -F/V*CB + k1*CA;                        % dCB/dt
fP = -F/V*CP + k2*CA^2;                      % dCP/dt


% ベクトル形式でまとめる
C = [CA; CB; CP];
f = [fA; fB; fP];


disp('--- 反応速度式 f(C) ---');
```

```matlab
pretty(f)
```

```matlab


%%  ヤコビ行列の自動導出
J = jacobian(f, C);
J = simplify(J);


disp('--- ヤコビ行列 J = ∂f/∂C ---');
```

```matlab
pretty(J)
```

```matlab


%%  固有値の解析的導出
syms lambda real
I = eye(3);
char_eq = simplify(det(J - lambda*I));   % 特性方程式
disp('--- 特性方程式 det(J - λI) = 0 ---');
```

```matlab
pretty(char_eq)
```

```matlab


% 固有値を解析的に解く
lambda_sol = simplify(solve(char_eq == 0, lambda));
```

```matlab
disp('--- 固有値（解析解） ---');
```


（参考２）固有値計算


```matlab
%% ================================
% CSTR 複合反応の定常状態と安定性解析
% A -> B, 2A -> P
% ================================


clear; clc; close all;


%% パラメータ
F = 1;       % 流量 [L/min]
V = 1;       % 体積 [L]
CAin = 1;    % 流入濃度 [mol/L]
k1 = 1;      % A->B 反応速度
k2 = 1;      % 2A->P 反応速度


%% 定常状態の方程式（fsolve用）
fun = @(C) [ F/V*(CAin - C(1)) - k1*C(1) - 2*k2*C(1)^2;
             -F/V*C(2) + k1*C(1);
             -F/V*C(3) + k2*C(1)^2 ];


% 初期値を複数用意して複数解を探索
C0_list = [0.1, 0.5, 0.9];
steady_states = [];


for C0 = C0_list
    C_sol = fsolve(fun, [C0, 0, 0], optimoptions('fsolve','Display','off'));
    steady_states = [steady_states; C_sol];
end


% 重複を削除
steady_states = unique(round(steady_states,6),'rows');


disp('定常状態 [CA CB CP] =');
disp(steady_states);


%% 固有値による安定性判定
disp('--- 固有値と安定性 ---');
for i=1:size(steady_states,1)
    CA = steady_states(i,1);
    CB = steady_states(i,2);
    CP = steady_states(i,3);

    % Jacobian
    J = [ -F/V - k1 - 4*k2*CA, 0, 0;
           k1, -F/V, 0;
           2*k2*CA, 0, -F/V ];

    eigJ = eig(J);

    fprintf('定常点 %d: [CA CB CP] = [%g %g %g]\n', i, CA, CB, CP);
    fprintf('  固有値: %s\n', mat2str(eigJ',3));
    if all(real(eigJ)<0)
        fprintf('  安定\n');
    else
        fprintf('  不安定\n');
    end
end


%% 可視化
figure; hold on; grid on;
for i=1:size(steady_states,1)
    plot3(steady_states(i,1), steady_states(i,2), steady_states(i,3), 'ro', 'MarkerSize',8, 'LineWidth',2);
end
xlabel('[A]'); ylabel('[B]'); zlabel('[P]');
title('CSTR定常状態の可視化');
view(45,30);

```


（参考３）CSTR複合反応系の時間応答シミュレーションコード例


```matlab


clear; clc; close all;


%% パラメータ設定
F  = 0.2;     % 流量 [L/min]
V  = 1.0;     % 反応槽体積 [L]
k1 = 1.0;     % 1次反応速度定数 [1/min]
k2 = 5.0;     % 2次反応速度定数 [L/(mol·min)]
CAin = 1.0;   % 流入濃度 [mol/L]


%%  初期条件
CA0 = 1.0;    % 初期濃度 [A]
CB0 = 0.0;    % 初期濃度 [B]
CP0 = 0.0;    % 初期濃度 [P]
C0 = [CA0 CB0 CP0];


%% 計算時間設定
tspan = [0 10];   % 時間範囲 [min]


%% ODE の定義
dCSTR = @(t,C) [
    F/V*(CAin - C(1)) - k1*C(1) - 2*k2*C(1)^2;  % dCA/dt
   -F/V*C(2) + k1*C(1);                         % dCB/dt
   -F/V*C(3) + k2*C(1)^2                        % dCP/dt
];


%% 数値積分（ode45）
[t, C] = ode45(dCSTR, tspan, C0);


CA = C(:,1); CB = C(:,2); CP = C(:,3);


%% 結果のプロット
figure; hold on; grid on;
plot(t, CA, 'r-', 'LineWidth',2);
plot(t, CB, 'b--', 'LineWidth',2);
plot(t, CP, 'g-.', 'LineWidth',2);
xlabel('時間 t [min]');
ylabel('濃度 [mol/L]');
title('CSTRにおける各成分の時間変化');
legend('[A]', '[B]', '[P]');
```
