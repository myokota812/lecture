## 1. Arrhenius式
$$k=k_0 e^{-E/RT}$$

$$\ln k = \ln k_0 - \frac{E}{RT}$$

## 2. 活性化エネルギー

$$E=-R\frac{d(\ln k)}{d(1/T)}$$

温度感受性を表す重要なパラメータである。

```python
import numpy as np
import matplotlib.pyplot as plt
R=8.314
E=80000
k0=1e10
T=np.linspace(300,800,300)
k=k0*np.exp(-E/(R*T))
plt.semilogy(T,k)
plt.xlabel("Temperature [K]")
plt.ylabel("k")
plt.grid()
plt.show()
```

## 3. エネルギー収支

定常状態では

$$\text{流入熱}-\text{流出熱}+\text{反応熱}+Q=0$$

## 4. 非等温PFR

物質収支
$$\frac{dF_A}{dV}=r_A$$

エネルギー収支
$$\frac{dT}{dV}=\frac{(-\Delta H)r_A+q}{\sum_iF_iC_{pi}}$$

## 5. 非等温CSTR

$$F_{A0}-F_A+r_AV=0$$

$$\sum F_{in}C_p(T_0-T)+(-\Delta H)r_AV+Q=0$$

## 6. 発熱反応と吸熱反応

発熱反応: $$\Delta H<0$$

吸熱反応: $$\Delta H>0$$

## 7. 多重定常状態

発熱CSTRでは熱発生曲線と熱除去曲線の交点が複数存在し得る。

```python
import numpy as np
import matplotlib.pyplot as plt
T=np.linspace(300,800,300)
Qgen=(T-300)*0.04*(800-T)
Qrem=5+0.5*(T-300)
plt.plot(T,Qgen,label="Heat generation")
plt.plot(T,Qrem,label="Heat removal")
plt.legend()
plt.grid()
plt.show()
```

## 8. ホットスポット

固定床反応器では発熱反応により温度ピークが生じる場合がある。

## 9. 温度と選択率

並列反応

$$A\rightarrow B$$
$$A\rightarrow C$$

$$k_B=k_{0B}e^{-E_B/RT}$$
$$k_C=k_{0C}e^{-E_C/RT}$$

## 10. 選択率

$$S_{B/C}=\frac{k_B}{k_C}
=\frac{k_{0B}}{k_{0C}}
\exp\left[-\frac{E_B-E_C}{RT}\right]$$

活性化エネルギー差により温度が選択率へ影響する。

```python
import numpy as np
import matplotlib.pyplot as plt
R=8.314
EB=90000
EC=70000
T=np.linspace(300,1000,300)
S=np.exp(-(EB-EC)/(R*T))
plt.plot(T,S)
plt.xlabel("T [K]")
plt.ylabel("Relative selectivity")
plt.grid()
plt.show()
```

## 演習問題
1. Arrhenius式を導出せよ。
2. Arrheniusプロットから活性化エネルギーを求める方法を説明せよ。
3. 発熱CSTRで多重定常状態が生じる理由を説明せよ。
4. 活性化エネルギーの異なる並列反応で温度が選択率へ与える影響を説明せよ。

## 解答
1. ln(k)と1/Tの関係を用いる。
2. 傾き=-E/R。
3. 熱発生と熱除去の非線形交点が複数存在するため。
4. Eの大きい反応ほど温度上昇の影響を強く受ける。
