## 1. 感度解析
$$S_p=\frac{p}{y}\frac{dy}{dp}$$

設計変数が出力へ与える影響を定量化する。

```python
import numpy as np
k=np.linspace(0.1,2,100)
X=1-np.exp(-k)
S=k*np.exp(-k)/X
print("Sensitivity range:",S.min(),S.max())
```

## 2. パラメータ推定
一次反応
$$C_A=C_{A0}e^{-kt}$$

実験データから k を推定する。

```python
import numpy as np
from scipy.optimize import curve_fit

def model(t,k):
    return np.exp(-k*t)

t=np.array([0,1,2,3,4])
CA=np.array([1.0,0.61,0.37,0.23,0.14])
popt,_=curve_fit(model,t,CA,p0=[0.5])
print("Estimated k=",popt[0])
```

## 4. 制約付き最適化
目的関数
$$J=f(x)$$

制約
$$g(x)\le0$$

```python
from scipy.optimize import minimize

f=lambda x: (x[0]-2)**2
cons=({'type':'ineq','fun':lambda x: x[0]})
res=minimize(f,[0.5],constraints=cons)
print(res.x)
```

## 5. 逐次反応への応用
$$A\rightarrow B\rightarrow C$$

B収率を最大とする条件を探索する。

```python
import numpy as np
k1=1.0;k2=0.3
t=np.linspace(0,10,1000)
CB=k1/(k2-k1)*(np.exp(-k1*t)-np.exp(-k2*t))
idx=np.argmax(CB)
print("Optimal time =",t[idx])
print("Maximum yield =",CB[idx])
```
