## 1. 複合反応とは

逐次反応: $$\mathrm{A}\rightarrow \mathrm{B}\rightarrow \mathrm{C}$$

並列反応: $$\mathrm{A}\rightarrow \mathrm{B}\\ \mathrm{A}\rightarrow \mathrm{C}$$

◇◇◇

実際の反応では、一つの反応だけが起こるとは限りません。複数の反応が関係する系を複合反応と呼びます。<br>
代表例として、上に示した逐次反応と並列反応があります。このことは前回の講義でも触れました。<br>
本講義では、それぞれの反応速度式の立て方と濃度変化の解析方法を学びます。

---

## 2. 逐次反応

$$ \text{A} \xrightarrow{r_1} \text{B} \xrightarrow{r_2} \rm C \tag{1}$$

$$r_1=k_1C_A \tag{2}$$
$$r_2=k_2C_B \tag{3}$$

◇◇◇

まず、逐次反応を考えます。<br>
この反応系では、$\text{A}$が反応して$\text B$を生成し、さらに生成した$\text{B}$が反応して$\text{C}$になります。<br>
それぞれの反応速度を一次反応と仮定すると、<br>
第1段階および第2段階の速度はそれぞれ、(2),(3)式で与えられます。<br>

この反応では$\text{B}$が中間生成物となるため、$\text{B}$は生成される一方で消費もされます。<br>
そのため、$\text{B}$の濃度は一般に一度増加した後、減少するという特徴的な挙動を示します。<br>
これが単純反応との大きな違いです。

---


### 2.1 物質収支式（支配方程式）



それでは、**回分反応器**を例にして、物質収支式を導出してみます。<br>
反応式が$j$ある場合の、成分$i$の物質収支式は
$$\frac{dN_i}{dt} = r_i V$$
$$
r_i=\sum_j \nu_{ij}R_j \tag{4}
$$

となることです。<br>
- $R_j$：反応 $j$ の進行速度
- $\nu_{ij}$：反応 $j$における成分 $i$ の量論係数

**体積一定**の場合
$$\frac{dC_i}{dt} = r_i $$


- $\mathbf{A}$は反応1でのみ消費されるため、
$$
\frac{dC_A}{dt}=-k_1C_A
$$

- $\mathbf B$は、反応1で生成される一方、反応2で消費されます。<br>
したがって、
$$
\frac{dC_B}{dt}=k_1C_A-k_2C_B
$$

この式の第1項が生成速度、第2項が消費速度を表しています。<br>
つまり$\text{B}$の濃度変化は「生成量と消費量の差」で決まります。

- $\mathbf C$は反応2でのみ生成されるため、

$$
\frac{dC_C}{dt}=k_2C_B
$$

以上が、同じ装置内で同時進行しますので、式の上では(5)式の連立微分方程式で表現できます。

$$
\left\{ 
    \begin{aligned} 
        \frac{dC_A}{dt} &= -k_1C_A\\ 
        \frac{dC_B}{dt} &= k_1C_A-k_2C_B \tag{5}\\ 
        \frac{dC_C}{dt} &= k_2C_B
    \end{aligned}
\right. 
$$
となります。


---



### 2.2. 解法


それでは、連立微分方程式(5)を解いてみましょう。ここではあえて３種類の解法を紹介します。<br>
**2.2.1直接解法**は学部の微分方程式レベルです。解説と一緒に解きながらみてください。<br>
**2.2.2行列による解法**は線形代数の固有値と固有行列の知識が最低限必要になりますので、興味のある人だけみてください。

---

<h3>2.2.1 直接微分方程式を解く方法【必読】</h3>

初期条件を $t = 0$ において $C_A(0) = C_{A0}$, $C_B(0) = 0$, $C_C(0) = 0$ として各成分の濃度 $C_A(t)$, $C_B(t)$, $C_C(t)$ を解析的に求めます。

**1. $C_A(t)$ の導出**

第1式は**変数分離形**です。

$$\frac{dC_A}{C_A} = -k_1 dt$$

両辺を積分して初期条件を適用すると：

$$C_A(t) = C_{A0} e^{-k_1 t}$$



**2. $C_B(t)$ の導出**

定数変化法（微分方程式の標準的な解法）を用いて、第2式の $C_B(t)$ について解く手順を詳しく解説します。<br>
解く対象の微分方程式：


$$\frac{dC_B}{dt} + k_2 C_B = k_1 C_{A0} e^{-k_1 t}$$

**Step 1: 斉次方程式の一般解を求める**

まず、右辺を $0$ と置いた斉次方程式（同次方程式）を解きます。

$$\frac{dC_B}{dt} + k_2 C_B = 0$$

変数分離を行います。

$$\frac{dC_B}{C_B} = -k_2 dt$$

両辺を積分します。

$$\ln \vert{}C_B\vert{} = -k_2 t + C_1 \implies C_B(t) = A e^{-k_2 t}$$

（ここで $A$ は任意の定数です）

◇◇◇

**Step 2: 任意定数 $A$ を関数 $A(t)$ に置き換える（定数変化法）**

定数 $A$ を時間の関数 $A(t)$ と仮定し、求める非斉次方程式の特解の形を次のように置きます。

$$C_B(t) = A(t) e^{-k_2 t} \tag{6}$$

これを微分します（積の微分公式を使用）。

$$\frac{dC_B}{dt} = \frac{dA(t)}{dt} e^{-k_2 t} - k_2 A(t) e^{-k_2 t} \tag{7}$$

◇◇◇

**Step 3: 元の微分方程式に代入する**

(6)式と(7)式の右辺を元の式 $\frac{dC_B}{dt} + k_2 C_B = k_1 C_{A0} e^{-k_1 t}$ に代入します。

$$\left( \frac{dA(t)}{dt} e^{-k_2 t} - k_2 A(t) e^{-k_2 t} \right) + k_2 \left( A(t) e^{-k_2 t} \right) = k_1 C_{A0} e^{-k_1 t}$$

左辺の $k_2 A(t) e^{-k_2 t}$ の項同士が打ち消し合って消去されます。

$$\frac{dA(t)}{dt} e^{-k_2 t} = k_1 C_{A0} e^{-k_1 t}$$

両辺に $e^{k_2 t}$ を掛けます。

$$\frac{dA(t)}{dt} = k_1 C_{A0} e^{(k_2 - k_1)t}$$

◇◇◇

**Step 4: $A(t)$ を求める**

両辺を $t$ で積分します。ここからは $k_1 \neq k_2$ と $k_1 = k_2$ で場合分けします。

* **$k_1 \neq k_2$ の場合：**

$$A(t) = \int k_1 C_{A0} e^{(k_2 - k_1)t} dt = \frac{k_1 C_{A0}}{k_2 - k_1} e^{(k_2 - k_1)t} + B$$

（$B$ は積分定数）

これを $C_B(t) = A(t) e^{-k_2 t}$ に代入すると、一般解が得られます。

$$C_B(t) = \left( \frac{k_1 C_{A0}}{k_2 - k_1} e^{(k_2 - k_1)t} + B \right) e^{-k_2 t} = \frac{k_1 C_{A0}}{k_2 - k_1} e^{-k_1 t} + B e^{-k_2 t}$$

* **$k_1 = k_2$ の場合：**

積分因子による解法、ロピタルの定理など色々ありますが、手っ取り早いのは

$$\frac{dA(t)}{dt} = k_1 C_{A0} e^{(k_2 - k_1)t}$$
を使う方法です。$k_1=k_2$のとき

$$\frac{dA(t)}{dt} = k_1 C_{A0} \implies A(t) = k_1 C_{A0} t + B$$

$$C_B(t) = (k_1 C_{A0} t + B) e^{-k_1 t}$$

◇◇◇

**Step 5: 初期条件 $C_B(0) = 0$ を適用する**

* **$k_1 \neq k_2$ の場合：**

$$C_B(0) = \frac{k_1 C_{A0}}{k_2 - k_1} + B = 0 \implies B = -\frac{k_1 C_{A0}}{k_2 - k_1}$$

これを戻すと、目的の解が得られます。

$$C_B(t) = \frac{k_1 C_{A0}}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right)$$

* **$k_1 = k_2$ の場合：**

$$C_B(0) = B = 0 \implies C_B(t) = k_1 C_{A0} t e^{-k_1 t}$$

◇◇◇



**3. $C_C(t)$ の導出**

物質収支の関係式 $C_A(t) + C_B(t) + C_C(t) = C_{A0}$（または第3式を積分）より、$C_C(t) = C_{A0} - C_A(t) - C_B(t)$ として求めます。

* $k_1 \neq k_2$ の場合：

$$C_C(t) = C_{A0} \left[ 1 - \frac{k_2}{k_2 - k_1} e^{-k_1 t} + \frac{k_1}{k_2 - k_1} e^{-k_2 t} \right]$$

* $k_1 = k_2 = k$ の場合：

$$C_C(t) = C_{A0} \left[ 1 - (1 + kt) e^{-kt} \right]$$

<h3>2.2.2 行列微分方程式を解く方法【選択】</h3>

連立微分方程式（４）式を反応速度に基づく量論行列で表していきます。<br>

$$\frac{d}{dt} \begin{bmatrix} C_A\\ C_B\\ C_C \end{bmatrix} = \begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix} \begin{bmatrix} C_A\\ C_B\\ C_C \end{bmatrix}$$


まず濃度ベクトルと係数行列を$\mathbf C$と$\mathbf K$とおきます。
$$
\mathbf C= \begin{bmatrix} C_A\\ C_B\\ C_C \end{bmatrix}
$$
$$
\mathbf K = \begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix}
$$

そうすると

 $$\frac{d\mathbf{C}}{dt} = \mathbf K \mathbf{C}$$
とおけます。 


---

### 解法の手順

#### ステップ 1: 係数行列 $\mathbf{K}$ の固有値を求める

行列 $\mathbf{K}$ の固有値 $\lambda$ は、特性方程式 $\det(\mathbf{K} - \lambda \mathbf{I}) = 0$ を解くことで得られます。

ここで、$\mathbf{K}$ は以下のような**下三角行列**（対角成分より上がすべて 0 の行列）です。


$$\mathbf{K} = \begin{bmatrix} -k_1 & 0 & 0 \\ k_1 & -k_2 & 0 \\ 0 & k_2 & 0 \end{bmatrix}$$

下三角行列の行列式の値は**対角成分の積**になるため、特性方程式は非常にシンプルになります。


$$\det(\mathbf{K} - \lambda \mathbf{I}) = (-k_1 - \lambda)(-k_2 - \lambda)(0 - \lambda) = 0$$

したがって、3つの固有値は一目で次のように求まります。

* $\lambda_1 = -k_1$
* $\lambda_2 = -k_2$
* $\lambda_3 = 0$

---

#### ステップ 2: 各固有値に対する固有ベクトルを求める

求めたそれぞれの固有値 $\lambda_i$ に対し、$(\mathbf{K} - \lambda_i \mathbf{I})\mathbf{v}_i = \mathbf{0}$ を満たす固有ベクトル $\mathbf{v}_i = \begin{bmatrix} v_{i1} & v_{i2} & v_{i3} \end{bmatrix}^T$ を求めます。
（※ここでは反応速度定数 $k_1 \neq k_2$ かつ $k_1, k_2 > 0$ と仮定します）

1. **$\lambda_1 = -k_1$ のとき:**

$$\begin{bmatrix} 0 & 0 & 0 \\ k_1 & k_1 - k_2 & 0 \\ 0 & k_2 & k_1 \end{bmatrix} \begin{bmatrix} v_{11} \\ v_{12} \\ v_{13} \end{bmatrix} = \mathbf{0}$$



これを解くと、例えば次のような固有ベクトルが得られます。

$$\mathbf{v}_1 = \begin{bmatrix} (k_2 - k_1) \\ k_1 \\ -k_2 \end{bmatrix} \quad (\text{定数倍の任意性あり})$$


2. **$\lambda_2 = -k_2$ のとき:**

$$\mathbf{v}_2 = \begin{bmatrix} 0 \\ k_2 - k_1 \\ k_2 \end{bmatrix}$$


3. **$\lambda_3 = 0$ のとき:**

$$\mathbf{v}_3 = \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix}$$



---

#### ステップ 3: 一般解を構成する

求めた固有値 $\lambda_i$ と固有ベクトル $\mathbf{v}_i$ を用いて、同次線形微分方程式の一般解は、それぞれの固有モードの線形結合（足し合わせ）として次のように表されます。

$$\mathbf{C}(t) = c_1 \mathbf{v}_1 e^{\lambda_1 t} + c_2 \mathbf{v}_2 e^{\lambda_2 t} + c_3 \mathbf{v}_3 e^{\lambda_3 t}$$

これを成分ごとに書き下すと、以下のようになります。


$$\mathbf{C}(t) = c_1 \begin{bmatrix} k_2 - k_1 \\ k_1 \\ -k_2 \end{bmatrix} e^{-k_1 t} + c_2 \begin{bmatrix} 0 \\ k_2 - k_1 \\ k_2 \end{bmatrix} e^{-k_2 t} + c_3 \begin{bmatrix} 0 \\ 0 \\ 1 \end{bmatrix} e^{0 \cdot t}$$


（ここで $c_1, c_2, c_3$ は初期条件によって決まる積分定数です）

---

#### ステップ 4. 初期条件 $t = 0$ を代入して定数 $c_1, c_2, c_3$ を決定する

初期条件 $\mathbf{C}(0) = \begin{bmatrix} C_{A0} \\ 0 \\ 0 \end{bmatrix}$ （つまり $C_A(0) = C_{A0}, C_B(0) = 0, C_C(0) = 0$）を代入します。
なお、$e^0 = 1$ です。

#### ① $c_1$ の決定（第1成分より）

$$C_A(0) = c_1 (k_2 - k_1) = C_{A0}$$


より、


$$c_1 = \frac{C_{A0}}{k_2 - k_1}$$

#### ② $c_2$ の決定（第2成分より）

$$C_B(0) = c_1 k_1 + c_2 (k_2 - k_1) = 0$$


ここに①で求めた $c_1$ を代入します。


$$\left(\frac{C_{A0}}{k_2 - k_1}\right) k_1 + c_2 (k_2 - k_1) = 0$$

$$c_2 (k_2 - k_1) = -\frac{k_1 C_{A0}}{k_2 - k_1}$$


両辺を $(k_2 - k_1)$ で割ると、


$$c_2 = -\frac{k_1 C_{A0}}{(k_2 - k_1)^2}$$

#### ③ $c_3$ の決定（第3成分より）

$$C_C(0) = -c_1 k_2 + c_2 k_2 + c_3 = 0$$

$$c_3 = k_2 (c_1 - c_2)$$


ここに①と②で求めた $c_1, c_2$ を代入します。


$$c_3 = k_2 \left( \frac{C_{A0}}{k_2 - k_1} - \left( -\frac{k_1 C_{A0}}{(k_2 - k_1)^2} \right) \right)$$


括弧内を通分して整理します。


$$c_3 = k_2 C_{A0} \left( \frac{(k_2 - k_1) + k_1}{(k_2 - k_1)^2} \right) = k_2 C_{A0} \left( \frac{k_2}{(k_2 - k_1)^2} \right) = \frac{k_2^2 C_{A0}}{(k_2 - k_1)^2}$$

---

#### ステップ 5. 求めた定数を各成分の式に代入する

決定した $c_1, c_2, c_3$ を各成分の式に代入して整理します。

#### ■ $C_A(t)$ の導出

$$C_A(t) = c_1 (k_2 - k_1) e^{-k_1 t} = \left(\frac{C_{A0}}{k_2 - k_1}\right) (k_2 - k_1) e^{-k_1 t} = C_{A0} e^{-k_1 t}$$

#### ■ $C_B(t)$ の導出

$$C_B(t) = c_1 k_1 e^{-k_1 t} + c_2 (k_2 - k_1) e^{-k_2 t}$$

$$C_B(t) = \left(\frac{C_{A0}}{k_2 - k_1}\right) k_1 e^{-k_1 t} + \left(-\frac{k_1 C_{A0}}{(k_2 - k_1)^2}\right) (k_2 - k_1) e^{-k_2 t}$$

$$C_B(t) = \frac{k_1 C_{A0}}{k_2 - k_1} e^{-k_1 t} - \frac{k_1 C_{A0}}{k_2 - k_1} e^{-k_2 t}$$


共通因数の $\frac{k_1}{k_2 - k_1} C_{A0}$ で括ります。


$$C_B(t) = \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) C_{A0}$$

#### ■ $C_C(t)$ の導出

$$C_C(t) = -c_1 k_2 e^{-k_1 t} + c_2 k_2 e^{-k_2 t} + c_3$$

$$C_C(t) = -\left(\frac{C_{A0}}{k_2 - k_1}\right) k_2 e^{-k_1 t} + \left(-\frac{k_1 C_{A0}}{(k_2 - k_1)^2}\right) k_2 e^{-k_2 t} + \frac{k_2^2 C_{A0}}{(k_2 - k_1)^2}$$


全体を $C_{A0}$ で括り、整理します。


$$C_C(t) = \left[ \frac{k_2^2}{(k_2 - k_1)^2} - \frac{k_2}{k_2 - k_1} e^{-k_1 t} - \frac{k_1 k_2}{(k_2 - k_1)^2} e^{-k_2 t} \right] C_{A0}$$

※質量保存則 $C_A(t) + C_B(t) + C_C(t) = C_{A0}$（すなわち $C_C(t) = C_{A0} - C_A(t) - C_B(t)$）を用いると、よりシンプルに次のように書くこともできます。


$$C_C(t) = \left[ 1 - e^{-k_1 t} - \frac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) \right] C_{A0} = \left[ 1 - \frac{k_2}{k_2 - k_1} e^{-k_1 t} + \frac{k_1}{k_2 - k_1} e^{-k_2 t} \right] C_{A0}$$



#### ステップ 6. 最終結果（ベクトル形式）

以上の結果をベクトル形式でまとめると、求めたい式が得られます。

$$\mathbf{C}(t) = \begin{bmatrix} C_A(t) \\ C_B(t) \\ C_C(t) \end{bmatrix} = \begin{bmatrix} e^{-k_1 t} \\ \dfrac{k_1}{k_2 - k_1} \left( e^{-k_1 t} - e^{-k_2 t} \right) \\ 1 - \dfrac{k_2}{k_2 - k_1} e^{-k_1 t} + \dfrac{k_1}{k_2 - k_1} e^{-k_2 t} \end{bmatrix} C_{A0}$$

<h3>2.2.3 数値計算により解く方法【必読】</h3>

数値的に解く代表的な2つの手法、**オイラー法（Euler Method）** と **ルンゲ＝クッタ法（Runge-Kutta Method: 通称RK法）** について説明します。

#### 1. 数値計算の基本アイデア

微分 $\frac{dC}{dt}$ は「時間 $t$ の微小な変化に対する濃度の変化率（＝グラフの傾き）」です。
数値計算では、時間を刻み幅 $\Delta t$（例えば $0.01$ 秒ごとなど）で区切り、**「現在の状態から少し先の状態を予測する」** 作業を繰り返します。

* $t_n$ : $n$ ステップ目の時間
* $C_{A, n}, C_{B, n}, C_{C, n}$ : $n$ ステップ目の各成分の濃度
* $\Delta t$ : 時間の刻み幅（ステップサイズ）


#### 2. オイラー法（Euler Method）

最もシンプルで直感的な方法です。**「現在の傾きが、次の $\Delta t$ 時間ずっと続く」** と仮定して計算します。

$$\mathbf{C}_{n+1} \approx \mathbf{C}_n + \left( \frac{d\mathbf{C}}{dt} \right)_n \Delta t$$


 **更新式**

微分方程式の右辺をそれぞれ関数 $f_A, f_B, f_C$ とおくと、次のステップの濃度は以下のように計算できます。

$$\begin{aligned} C_{A, n+1} &= C_{A, n} + \Delta t \cdot \left( -k_1 C_{A, n} \right) \\ C_{B, n+1} &= C_{B, n} + \Delta t \cdot \left( k_1 C_{A, n} - k_2 C_{B, n} \right) \\ C_{C, n+1} &= C_{C, n} + \Delta t \cdot \left( k_2 C_{B, n} \right) \end{aligned}$$

#### 特徴

* **メリット**: 式が単純でプログラムが書きやすい。
* **デメリット**: 誤差が大きく溜まりやすい（精度が低い）。$\Delta t$ をかなり小さくしないと、解が発振・発散して壊れることがあります。


#### 3. ルンゲ＝クッタ法（特に最も一般的な 4次RK法: RK4）

オイラー法の「最初の傾きだけで突っ走る」弱点を克服するために、**「途中の傾きもいくつか試して、平均をとって進む」** 高精度な方法です。4次ルンゲ＝クッタ法（RK4）は科学技術計算で最も広く使われています。

#### 計算の手順

（成分 $A, B, C$ をまとめてベクトル $\boldsymbol{C} = [C_A, C_B, C_C]^T$ として考えます）

1つのステップ（$t_n \rightarrow t_n + \Delta t$）を進めるために、4つの傾き $k_1, k_2, k_3, k_4$（※反応速度定数の $k_1, k_2$ と混同しないよう、ここでは傾きベクトル $\boldsymbol{m}_1, \boldsymbol{m}_2, \boldsymbol{m}_3, \boldsymbol{m}_4$ と呼びます）を計算します。

1. **$\boldsymbol{m}_1$（スタート地点の傾き）**
現在の状態 $\boldsymbol{C}_n$ での微分値を計算します。
2. **$\boldsymbol{m}_2$（中間地点での傾き 1）**
$\boldsymbol{m}_1$ を使って半ステップ（$\frac{\Delta t}{2}$）進んだ場所での微分値を計算します。
3. **$\boldsymbol{m}_3$（中間地点での傾き 2）**
$\boldsymbol{m}_2$ を使って半ステップ（$\frac{\Delta t}{2}$）進んだ場所での微分値を計算します。
4. **$\boldsymbol{m}_4$（ゴール地点での傾き）**
$\boldsymbol{m}_3$ を使って1ステップ（$\Delta t$）進んだ場所での微分値を計算します。

最後に、これらを重み付け平均（中間の傾きを重視）して次の値を決定します。

$$\boldsymbol{C}_{n+1} = \boldsymbol{C}_n + \frac{\Delta t}{6} \left( \boldsymbol{m}_1 + 2\boldsymbol{m}_2 + 2\boldsymbol{m}_3 + \boldsymbol{m}_4 \right)$$

### 特徴

* **メリット**: 非常に精度が高く、オイラー法よりも大きな刻み幅 $\Delta t$ で正確に計算できる。
* **デメリット**: 1ステップの更新のために4回右辺を計算する必要があるため、コードが少し複雑になる。

### Pythonによる実装例




```python
#@title数値計算手法の比較
import matplotlib.pyplot as plt
import numpy as np

# パラメータ設定
k1 = 0.5  # A -> B の反応速度定数
k2 = 0.2  # B -> C の反応速度定数

# 初期条件
C_A0 = 1.0  # 初期濃度 A
C_B0 = 0.0  # 初期濃度 B
C_C0 = 0.0  # 初期濃度 C

# 時間の設定
t_start = 0.0
t_end = 20.0
dt = 0.1  # 刻み幅
steps = int((t_end - t_start) / dt)
t_eval = np.linspace(t_start, t_end, steps + 1)


# 右辺の微分方程式を定義する関数
def derivatives(C):
    CA, CB, CC = C
    dCA_dt = -k1 * CA
    dCB_dt = k1 * CA - k2 * CB
    dCC_dt = k2 * CB
    return np.array([dCA_dt, dCB_dt, dCC_dt])


# --- 1. オイラー法 ---
C_euler = np.zeros((steps + 1, 3))
C_euler[0] = [C_A0, C_B0, C_C0]

for n in range(steps):
    C_euler[n + 1] = C_euler[n] + dt * derivatives(C_euler[n])

# --- 2. 4次ルンゲ＝クッタ法（RK4） ---
C_rk4 = np.zeros((steps + 1, 3))
C_rk4[0] = [C_A0, C_B0, C_C0]

for n in range(steps):
    C_n = C_rk4[n]

    m1 = derivatives(C_n)
    m2 = derivatives(C_n + 0.5 * dt * m1)
    m3 = derivatives(C_n + 0.5 * dt * m2)
    m4 = derivatives(C_n + dt * m3)

    C_rk4[n + 1] = C_n + (dt / 6.0) * (m1 + 2 * m2 + 2 * m3 + m4)

# グラフ描画 (RK4の結果を表示)
plt.figure(figsize=(8, 5))
plt.plot(t_eval, C_rk4[:, 0], label="C_A (RK4)", color="blue")
plt.plot(t_eval, C_rk4[:, 1], label="C_B (RK4)", color="orange")
plt.plot(t_eval, C_rk4[:, 2], label="C_C (RK4)", color="green")

# オイラー法との比較用（点線で描画）
plt.plot(t_eval, C_euler[:, 0], "--", label="C_A (Euler)", alpha=0.6)
plt.plot(t_eval, C_euler[:, 1], "--", label="C_B (Euler)", alpha=0.6)
plt.plot(t_eval, C_euler[:, 2], "--", label="C_C (Euler)", alpha=0.6)

plt.xlabel("Time t")
plt.ylabel("Concentration C")
plt.title("Sequential Reaction A -> B -> C")
plt.grid(True)
plt.legend()
plt.show()
```

### Pythonコード（数値解法と解析解の比較グラフ）

解析解と数値計算の結果を比較するPythonコードを用意しました。<br>
下に示したコードの▷印をクリックすると、計算が開始され比較のグラフが表示されます。

解析解と数値解がほぼ一致していることを確認できれば結構です。


---



```python
#@title 逐次反応の解析解と数値解の比較
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# --- 1. パラメータと初期条件の設定 ---
k1 = 0.5   # 反応速度定数 k1 [1/min]
k2 = 0.2   # 反応速度定数 k2 [1/min]
CA0 = 1.0  # 初期濃度 C_A(0) [mol/L]

t_span = (0, 20)            # 計算時間範囲 [0, 20]
t_eval = np.linspace(0, 20, 200) # グラフ描画用の時間点

# --- 2. 微分方程式の定義（数値計算用） ---
def reaction_system(t, C):
    CA, CB, CC = C
    dCA_dt = -k1 * CA
    dCB_dt = k1 * CA - k2 * CB
    dCC_dt = k2 * CB
    return [dCA_dt, dCB_dt, dCC_dt]

# --- 3. 数値解の計算 (Scipy solve_ivp) ---
C0 = [CA0, 0.0, 0.0]  # 初期条件 [CA0, CB0, CC0]
sol = solve_ivp(reaction_system, t_span, C0, t_eval=t_eval, method='RK45')

# --- 4. 解析解の計算 ---
t_exact = t_eval
CA_exact = CA0 * np.exp(-k1 * t_exact)
CB_exact = (k1 * CA0 / (k2 - k1)) * (np.exp(-k1 * t_exact) - np.exp(-k2 * t_exact))
CC_exact = CA0 * (1 - (k2 / (k2 - k1)) * np.exp(-k1 * t_exact) + (k1 / (k2 - k1)) * np.exp(-k2 * t_exact))

# --- 5. 比較グラフの描画 ---
plt.figure(figsize=(10, 6))

# 数値解のプロット（点）
plt.plot(sol.t, sol.y[0], 'o', label='$C_A$ (Numerical)', markersize=4, color='tab:blue')
plt.plot(sol.t, sol.y[1], 's', label='$C_B$ (Numerical)', markersize=4, color='tab:orange')
plt.plot(sol.t, sol.y[2], '^', label='$C_C$ (Numerical)', markersize=4, color='tab:green')

# 解析解のプロット（実線）
plt.plot(t_exact, CA_exact, '-', label='$C_A$ (Analytical)', color='navy', linewidth=1.5)
plt.plot(t_exact, CB_exact, '-', label='$C_B$ (Analytical)', color='darkorange', linewidth=1.5)
plt.plot(t_exact, CC_exact, '-', label='$C_C$ (Analytical)', color='darkgreen', linewidth=1.5)

# グラフ装飾
plt.xlabel('Time $t$ [min]', fontsize=12)
plt.ylabel('Concentration [mol/L]', fontsize=12)
plt.title('Comparison between Numerical and Analytical Solutions', fontsize=14)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(fontsize=11)
plt.tight_layout()

# 表示
plt.show()
```
