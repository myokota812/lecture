## リサイクル比
$$R=\frac{F_R}{F_0}$$

$$F_M=(1+R)F_0$$

## 総合転化率
$$X_o=\frac{(1+R)X_s}{1+RX_s}$$

```python
import numpy as np
R=np.linspace(0,10,11)
Xs=0.4
Xo=(1+R)*Xs/(1+R*Xs)
for r,x in zip(R,Xo):
    print(f"R={r:.0f}, Xo={x:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
R=np.linspace(0,10,200)
Xs=0.4
Xo=(1+R)*Xs/(1+R*Xs)
plt.plot(R,Xo)
plt.xlabel("Recycle ratio")
plt.ylabel("Overall conversion")
plt.grid()
plt.show()
```

## パージ比
$$f=\frac{F_P}{F_R+F_P}$$

## 反応分離システム
Reactor → Separator → Recycle

## 演習問題
R=3, Xs=0.5 のとき総合転化率を求めよ。

## 解答
$$X_o=\frac{(1+3)0.5}{1+3\times0.5}=0.8$$
