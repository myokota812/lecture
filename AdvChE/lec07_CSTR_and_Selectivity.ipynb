## 1. CSTRとは
完全混合槽型反応器である。

特徴
- 槽内濃度は出口濃度と等しい
- 定常状態で運転する

## 2. 定常物質収支

$$蓄積=流入-流出+反応=0$$

## 3. CSTR設計方程式

$$F_{A0}-F_A+r_AV=0$$

したがって

$$V=\frac{F_{A0}X}{(-r_A)_{out}}$$

## 4. 空間時間

$$\tau=\frac{V}{v_0}$$

一次反応

$$-r_A=kC_A$$

より

$$\tau=\frac{X}{k(1-X)}$$

## 5. PFRとの比較

PFR

$$\tau=\frac1k\ln\frac1{1-X}$$

CSTR

$$\tau=\frac{X}{k(1-X)}$$

同一転化率では一般に

$$\tau_{CSTR}>\tau_{PFR}$$

## 6. 並列反応

$$A\rightarrow B$$
$$A\rightarrow C$$

$$r_B=k_BC_A^m$$
$$r_C=k_CC_A^n$$

## 7. 瞬間選択率

$$S_{B/C}=\frac{r_B}{r_C}
=\frac{k_B}{k_C}C_A^{m-n}$$

## 8. Levenspiel流判定

m > n ：高濃度有利 → PFR・回分有利

m < n ：低濃度有利 → CSTR有利

## 9. 例題1

$$r_B=3C_A^2$$
$$r_C=2C_A$$

$$S_{B/C}=\frac32 C_A$$

高濃度有利 → PFR有利

## 10. 例題2

$$r_B=2C_A$$
$$r_C=5C_A^3$$

$$S_{B/C}=\frac{2}{5}C_A^{-2}$$

低濃度有利 → CSTR有利

## 11. 逐次反応

$$A\rightarrow B\rightarrow C$$

中間生成物Bの生成では最適滞留時間が存在する。

```python
import numpy as np
import matplotlib.pyplot as plt
k=0.5
X=np.linspace(0.01,0.95,300)
tau_pfr=(1/k)*np.log(1/(1-X))
tau_cstr=X/(k*(1-X))
plt.plot(X,tau_pfr,label="PFR")
plt.plot(X,tau_cstr,label="CSTR")
plt.xlabel("Conversion")
plt.ylabel("Space time")
plt.legend();plt.grid();plt.show()
```
