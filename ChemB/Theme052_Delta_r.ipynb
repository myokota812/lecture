## 2. ルシャトリエの法則（復習）

> 突然ですが、以下の反応式をみて、温度と圧力の影響をルシャトリエの法則で考察してみてください。

$$\text{N}_2(\text{気}) + 3\text{H}_2(\text{気}) \rightleftarrows 2\text{NH}_3(\text{気}) \quad (\Delta H^\circ = -92.2 \text{ kJ/mol})$$

この反応は以下の2つの大きな特徴を持ちます。

1. **発熱反応**である（$\Delta H < 0$、右向きに進むと熱を放出する）。
2. **気体分子の総数が減少する**反応である（左辺 $1 + 3 = 4\text{ mol}$ $\rightarrow$ 右辺 $2\text{ mol}$）。

---

### 2.1. 温度を上げると平衡はどちら向きに移動しますか

みなさんはどのように解答されますか！

<details>
<summary>説明</summary>

* **外部からの操作:** 温度を上げる（加熱する）、または下げる（冷却する）。
* **系の応答（ルシャトリエの原理）:**
* 温度を**上げる**と、系は余分な熱を逃がそうとするため、熱を吸収する方向（吸熱方向）へ平衡が移動する。
* 逆に温度を**下げる**と、失われた熱を補おうとするため、熱を放出する方向（発熱方向）へ平衡が移動する。

したがって

アンモニア合成は**発熱反応**であるため、熱を放出する方向は「正反応（右向き：アンモニアができる方向）」である。
したがって、

* **低温にするほど** $\rightarrow$ 平衡は右（生成物側）に傾き、アンモニアの平衡収率が高くなる。
* **高温にするほど** $\rightarrow$ 平衡は左（反応物側）に傾き、アンモニアの収率は低下する。

> 💡 **発展（工学的ジレンマ）:** 熱力学的には「低温」が圧倒的に有利であるが、実際には低温にすると反応速度（分子が反応するスピード）が極めて遅くなるため、触媒を用いつつ「400〜500℃」という妥協点（最適温度）で工業運転が行われる。このジレンマの根本的な原因はエネルギー障壁ですね。後ほど
</details>

### 2.2. 圧力上げるとどうなりますか

これも自問自答してください。

<details>
<summary>説明</summary>
* **外部からの操作:** 圧力を上げる（圧縮する）、または下げる（膨張させる）。
* **系の応答（ルシャトリエの原理）:**
* 圧力を**上げる**と、系はその高圧を和らげるため、**気体の分子の総数を減らす方向**へ平衡が移動する。
* 圧力を**下げる**と、圧力を補うため、**気体の分子の総数を増やす方向**へ平衡が移動する。

反応前後の気体分子のモル数を比較すると、

* 左辺（反応物）: $1\text{ (N}_2\text{)} + 3\text{ (H}_2\text{)} = \mathbf{4 \text{ mol}}$
* 右辺（生成物）: $2\text{ (NH}_3\text{)} = \mathbf{2 \text{ mol}}$

となり、右側の方が分子数が少ない。
したがって、

* **高圧にするほど** $\rightarrow$ 分子数を減らそうとして平衡は右（生成物側）に大きく傾き、アンモニアの収率が向上する。
* **低圧にするほど** $\rightarrow$ 平衡は左（反応物側）に傾く。

> 💡 **発展（工学的アプローチ）:** 分子数が減少するこの反応では、高圧であればあるほど有利になるため、実際のプラントでは 100〜300気圧という高圧環境を利用して効率よくアンモニアを生産している。

まとめ

アンモニア合成反応における収率を最大化するための条件を、ルシャトリエの原理から導くと以下のようになる。

1. **温度条件:** 熱を放出する発熱反応であるため、「低温」が有利。
2. **圧力条件:** 分子数が減少する反応であるため、「高圧」が有利。

</details>

## 3. 熱力学的な考察（本題）

### 3.1. 導入：「ルシャトリエ」だけでは物足りない！

 ルシャトリエの法則では反応式をみて
* 「温度を下げると発熱反応だから右に進む」「圧力を上げると分子数が減るから右に進む」という方向の予測（定性的）はできます。。

「では、
 - **どれくらいの温度**を境にして左右へのの平衡移動が変わるのか？」

 これらを数値（kJ/mol）で白黒はっきりさせるのが **$\Delta G$（ギブズ自由エネルギー）** です。

---

### 3.2. なぜその方向へ進むのか？ ギブズ自由エネルギー ($\Delta G$) の正体

「なぜ低温が好きなのか」「なぜ高圧が良いのか」の本当の原因（エネルギーの収支）を明かすのが、ギブズ自由エネルギーの変化を表すマスター方程式です。

$$\Delta G = \Delta H - T \Delta S$$

化学反応は、$\Delta G$ がマイナスになる方向（エネルギーの谷に向かって）へ自発的に進みます。

### ① エンタルピー項 ($\Delta H$) の役割

* 結合ができるときに放出されるエネルギー（発熱）。
* $\Delta H < 0$ であれば、持っているエネルギーの底が下がるため、**無条件で「生成物側に行きたがる（マイナスの寄与）」**。

### ② エントロピー項 ($T \Delta S$) の役割

* 物質の乱雑さ、自由度の指標。
* 気体が固体になるような反応では $\Delta S < 0$ となり、$-T\Delta S$ 全体としては「プラス（生成物側にとって不利・逆らいたい力）」として働きます。

### ③ 温度 $T$ が勝敗を決めるメカニズム

* **低温のとき ($T$ が小さい):**
* $T\Delta S$ の影響がかき消され、発熱項である **$\Delta H$ のマイナスのパワーが勝利**します。結果、$\Delta G < 0$ となり生成物が安定します。


* **高温のとき ($T$ が大きい):**
* 温度 $T$ が大きくなると、$T\Delta S$ のプラスの寄与がどんどん膨らみ、$\Delta H$ のマイナスを打ち消してしまいます。その結果、$\Delta G > 0$（または平衡が逆転）となり、高温では生成物が不利（分解が有利）になります。

## 問1

四酸化二窒素の分解反応は次式で表される。


$$\text{N}_2\text{O}_4(\text{気}) \rightleftarrows 2\text{NO}_2(\text{気}) \quad (\Delta H^\circ > 0, \ \Delta S^\circ > 0)$$


この反応の標準エンタルピー変化 $\Delta H^\circ$ は正であり、分子数が増加するため標準エントロピー変化 $\Delta S^\circ$ も正である。ギブズ自由エネルギー変化 $\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ$ の観点から、以下の問いに答えよ。

1. 室温付近の低温において、$\Delta G^\circ$ の符号は正（ $>0$ ）と負（ $<0$ ）のどちらになると予想されるか。また、そのと
きの反応の自発性について述べよ。
2. 温度 $T$ を十分に高くしたとき、$\Delta G^\circ$ の符号はどう変化するか。また、それに伴って平衡はどちらの方向に移動するか説明せよ。

## 問2


一酸化炭素と水蒸気から水素を生成する水性ガスシフト反応は次式で表される。


$$\text{CO}(\text{気}) + \text{H}_2\text{O}(\text{気}) \rightleftarrows \text{CO}_2(\text{気}) + \text{H}_2(\text{気}) \quad (\Delta H^\circ < 0, \ \Delta S^\circ \approx 0)$$


この反応は発熱反応であり（$\Delta H^\circ < 0$）、反応前後の気体分子のモル数が等しいためエントロピー変化 $\Delta S^\circ$ はほぼゼロである。この系における $\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ$ の挙動について考察せよ。

## 問3

炭酸カルシウムの熱分解反応は次式で表される。


$$\text{CaCO}_3(\text{固}) \rightleftarrows \text{CaO}(\text{固}) + \text{CO}_2(\text{気}) \quad (\Delta H^\circ > 0, \ \Delta S^\circ > 0)$$


この反応は極めて大きな吸熱反応（$\Delta H^\circ \gg 0$）であり、気体が発生するためエントロピー変化も大きい（$\Delta S^\circ \gg 0$）。室温では $\text{CaCO}_3$ は安定に存在するが、高温に加熱すると分解が進行する。この現象を $\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ$ の符号変化に基づいて説明せよ。

---

#### 【解答・解説のポイント】

## 問1

1. **答え:** $\Delta G^\circ > 0$（正）。反応物（$\text{N}_2\text{O}_4$）側が熱力学的に安定であり、正反応は自発的に進行しない。
* *解説:* $\Delta H^\circ > 0$（吸熱）であるため、低温では $-TS$ 項よりも $\Delta H^\circ$ のプラスの寄与が勝り、$\Delta G^\circ > 0$ となるため。


2. **答え:** $\Delta G^\circ < 0$（負）に逆転し、正反応（分解方向）が自発的に進行するようになる（平衡は右に傾く）。
* *解説:* $T$ を高くすると $-TS^\circ$ 項（マイナスの寄与）の絶対値が大きくなり、$\Delta H^\circ$ を上回るため、$\Delta G^\circ$ が $0$ を下回って符号が反転するから。


## 問2

1. **答え:** $\Delta S^\circ \approx 0$ であるため、温度 $T$ を変化させても $-TS^\circ$ 項の寄与が小さく、$\Delta G^\circ$ は主にエンタルピー項 $\Delta H^\circ$ によって決定される。
2. **答え:** $\Delta H^\circ < 0$ であるため、幅広い温度範囲で $\Delta G^\circ < 0$ となり、正反応（水素生成）が自発的に進みやすい。ただし、温度が高くなるとわずかな $-TS$ 項や熱力学的制限により $\Delta G^\circ$ が $0$ に近づくため、より負の値を維持できる「低温」のほうが $\Delta G^\circ$ の観点から生成系に有利である。


## 問3

1. **答え:**
* **室温付近:** $\Delta H^\circ$ が非常に大きなプラスである一方、$-TS^\circ$ 項のマイナス効果が小さいため、$\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ > 0$ となり、分解は自発的に起きない（$\text{CaCO}_3$ が安定）。
* **高温加熱時:** $T$ が十分に高くなると、大きな $\Delta S^\circ$ とのかけ算である $-TS^\circ$ 項のマイナスが $\Delta H^\circ$ のプラスを上回る。これにより $\Delta G^\circ < 0$ へと符号が逆転し、高温でのみ熱分解が自発的に進行するようになる。
