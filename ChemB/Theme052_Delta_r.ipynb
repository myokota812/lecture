
# 化学反応の熱力学的考察

## 1. はじめに（本時のテーマ）

本講義では、工業的にも極めて重要なハーバー・ボッシュ法によるアンモニア合成反応を題材に、ルシャトリエの原理を用いて「温度」**と**「圧力」が平衡に与える影響を熱力学的に考察する。

## 2. ルシャトリエの法則（復習）

> 突然ですが、以下の反応式をみて、温度と圧力の影響をルシャトリエの法則で考察してみてください。

$$\text{N}_2(\text{気}) + 3\text{H}_2(\text{気}) \rightleftarrows 2\text{NH}_3(\text{気}) \quad (\Delta H^\circ = -92.2 \text{ kJ/mol})$$

この反応は以下の2つの大きな特徴を持つ。

1. **発熱反応**である（$\Delta H < 0$、右向きに進むと熱を放出する）。
2. **気体分子の総数が減少する**反応である（左辺 $1 + 3 = 4\text{ mol}$ $\rightarrow$ 右辺 $2\text{ mol}$）。

### 2.1. 温度変化の影響の考察（発熱反応と平衡）

みなさんはどのように解答されますか！

<summary>説明</summary>

* **外部からの操作:** 温度を上げる（加熱する）、または下げる（冷却する）。
* **系の応答（ルシャトリエの原理）:**
* 温度を**上げる**と、系は余分な熱を逃がそうとするため、熱を吸収する方向（吸熱方向）へ平衡が移動する。
* 逆に温度を**下げる**と、失われた熱を補おうとするため、熱を放出する方向（発熱方向）へ平衡が移動する。

したがって

アンモニア合成は**発熱反応**であるため、熱を放出する方向は「正反応（右向き：アンモニアができる方向）」である。
したがって、

* **低温にするほど** $\rightarrow$ 平衡は右（生成物側）に傾き、アンモニアの平衡収率が高くなる。
* **高温にするほど** $\rightarrow$ 平衡は左（反応物側）に傾き、アンモニアの収率は低下する。

> 💡 **発展（工学的ジレンマ）:** 熱力学的には「低温」が圧倒的に有利であるが、実際には低温にすると反応速度（分子が反応するスピード）が極めて遅くなるため、触媒を用いつつ「400〜500℃」という妥協点（最適温度）で工業運転が行われる。このジレンマの根本的な原因はエネルギー障壁ですね。後ほど

### 2.2. 圧力変化の影響の考察（分子数の変化と平衡）

これも自問自答してください。

<details>
<summary>説明</summary>
* **外部からの操作:** 圧力を上げる（圧縮する）、または下げる（膨張させる）。
* **系の応答（ルシャトリエの原理）:**
* 圧力を**上げる**と、系はその高圧を和らげるため、**気体の分子の総数を減らす方向**へ平衡が移動する。
* 圧力を**下げる**と、圧力を補うため、**気体の分子の総数を増やす方向**へ平衡が移動する。

反応前後の気体分子のモル数を比較すると、

* 左辺（反応物）: $1\text{ (N}_2\text{)} + 3\text{ (H}_2\text{)} = \mathbf{4 \text{ mol}}$
* 右辺（生成物）: $2\text{ (NH}_3\text{)} = \mathbf{2 \text{ mol}}$

となり、右側の方が分子数が少ない。
したがって、

* **高圧にするほど** $\rightarrow$ 分子数を減らそうとして平衡は右（生成物側）に大きく傾き、アンモニアの収率が向上する。
* **低圧にするほど** $\rightarrow$ 平衡は左（反応物側）に傾く。

> 💡 **発展（工学的アプローチ）:** 分子数が減少するこの反応では、高圧であればあるほど有利になるため、実際のプラントでは 100〜300気圧という高圧環境を利用して効率よくアンモニアを生産している。

まとめ

アンモニア合成反応における収率を最大化するための条件を、ルシャトリエの原理から導くと以下のようになる。

1. **温度条件:** 熱を放出する発熱反応であるため、「低温」が有利。
2. **圧力条件:** 分子数が減少する反応であるため、「高圧」が有利。

</details>

* **連動する挙動:** スライダーを動かすと、温度や圧力の変化に応じて $H, S, G$ の値が変化し、それに伴って左右の粒子数（$N$ と $M$）がリアルタイムで増減する。

### 3.2. アプリのコード概説


#### ① 「温度と圧力」から「どれくらいアンモニアができるか」を計算する仕組み

アプリのスライダーで温度や圧力を変えたとき、裏側では次の3つのステップで「今の条件だと、どれくらいの割合でアンモニアになるか」を計算しています。

1. **エネルギーの基本を計算する（$\Delta G^\circ$）**
* 「この温度のとき、アンモニアを作るのにどれくらいエネルギーが必要か（あるいは放出されるか）」を、おなじみの式 $\Delta G^\circ = \Delta H^\circ - T \Delta S^\circ$ で計算しています。
* （※エンタルピー $H$ とエントロピー $S$ のバランスを見ています）


2. **「できやすさ（平衡定数 $K_p$）」を出す**
* 計算した自由エネルギーをもとに、「今、どれくらいアンモニアになりたがっているか」の度合いを示す数字（平衡定数 $K_p$）を求めます。これが大きいほど、アンモニアができやすい状態です。


3. **ぴったりの「転化率 ($X_{eq}$)」を探す（グリッドサーチ）**
* 「窒素と水素がこれだけあって、圧力がこれくらいなら、最終的に何％がアンモニアに変わるはずだ」という数式（質量作用の法則）があります。
* アプリはこの数式を満たす割合（転化率 $X_{eq}$）を、$0.001$ 刻みでしらみつぶしに（コンピューターの高速な計算で）探して当て出しています。


#### ② 画面の「分子アニメーション」が連動して動く仕組み

アプリの左右にある四角い枠（キャンバス）の中では、コンピューターが計算した結果に合わせて、分子の絵をリアルタイムで動かしています。

* **反応物の数 ($N$) と生成物の数 ($M$) の連動:**
* 計算されたアンモニアの割合（$X_{eq}$）に応じて、
* **左側のカード:** まだ反応していないガス（窒素や水素）の数 $(1 - X_{eq})$
* **右側のカード:** できあがったアンモニアの数 $(X_{eq})$


* が、自動的に増えたり減ったりするようにプログラミングされています。


* **温度による動きの激しさ:**
* スライダーで温度（$T$）を高くすると、プログラム内のスピード係数（`tempRatio`）が大きくなります。
* これによって、画面の中の分子たちが「ビュンビュン」と激しく跳ね回るようになり、「温度が高い＝分子が大きな熱エネルギーを持っている」という状態を視覚的に表現しています。

### 3.3. 熱力学的考察：$G = H - TS$ が決める化学平衡　【重要】

ルシャトリエの原理は「現象の定性的な説明」を与えてくれましたが、本講義ではこれを**ギブズ自由エネルギー $G$** の定量的な変化として捉えることを主目的とします。

化学反応が自発的にどちらの方向に進むかは、系のギブズ自由エネルギー変化 $\Delta G$ が**負 ($\Delta G < 0$)** になるかどうかで決定されます。

$$\Delta G = \Delta H - T\Delta S$$

この式をアンモニア合成反応（$\text{N}_2 + 3\text{H}_2 \rightleftarrows 2\text{NH}_3$）に適用して、各項の意味を説明します。

### ① エンタルピー項（$H$）の寄与：なぜ発熱なのか？

* アンモニア合成の標準エンタルピー変化 $\Delta H^\circ \approx -92.2\text{ kJ/mol}$ は大きな負の値（**発熱**）である。
* これは、生成物である $\text{NH}_3$ の結合エネルギーの総和が、反応物 $\text{N}_2$ と $\text{H}_2$ の結合を断ち切るのに必要なエネルギーよりも大きいため、余剰なエネルギーが外部に放出されることを意味する。
* したがって、エネルギーの「低い（安定な）」状態へ向かおうとする力が、生成物側への駆動力となる。

### ② エントロピー項（$S$）の寄与：乱雑さのペナルティ

* 反応前後の気体分子のモル数を比較すると、左辺 $4\text{ mol}$ から右辺 $2\text{ mol}$ へと**分子数が減少**している。
* 分子数が減るということは、空間的な「乱雑さ（エントロピー $S$）」が小さくなることを意味する。つまり、エントロピー変化 $\Delta S^\circ$ は負の値（$\Delta S^\circ \approx -198.7\text{ J/(mol}\cdot\text{K)}$）となる。
* $\Delta S < 0$ であるため、$-TS$ の項は正（プラス）となり、温度 $T$ が高くなるほど $\Delta G$ を引き上げる（＝反応を不利にする）方向に働く。

### ③ 温度 $T$ による主導権の逆転

* **低温時 ($T$ が小さい):**
$-TS$ の寄与が小さいため、エンタルピー項 $\Delta H$（$-92.2\text{ kJ/mol}$）の寄与が勝り、$\Delta G < 0$ となって**生成物（アンモニア）側が熱力学的に圧倒的優勢**になる。
* **高温時 ($T$ が大きい):**
温度が高くなるにつれて $-TS$ の項（プラスの寄与）が大きくなり、$\Delta G$ が正に転じる。そのため、熱力学的には反応物（窒素と水素）側が安定（熱解離が優勢）になり、アンモニアは得られなくなる。

---

### 3.4. まとめ

本講義で学んだポイントは以下の通りである。

1. **アプリによる視覚化:** 温度・圧力を変化させると、$G = H - TS$ の値と連動して分子アニメーション上の粒子数 $N, M$ が変化し、熱力学の平衡状態を直感的に捉えることができる。
2. **ルシャトリエの原理の裏付け:** 「低温・高圧が有利」という性質は、単なる経験則ではなく、発熱反応（$\Delta H < 0$）と分子数減少に伴うエントロピー低下を包括するギブズ自由エネルギー最小化の法則（$\Delta G = \Delta H - TS$）の現れに他ならない。

>どこかに記述したと思いますが、熱力学的には「アンモニア合成は低温ほどよい」ということになります。？？？が頭に浮かんだ人もいますよね。工業的には、何百℃という高温で製造しています。この矛盾はどこからくるかというと、エネルギー障壁（速度論）が極めて高いというこです。
>
>だからと言って、熱力学的結論（低温優位）が間違っているとは限りません。$\Delta G < 0$なので、自発的に反応しますという方向性は間違えではないのですが、気の遠くほど長年をようすると思います。

一酸化炭素と水蒸気から水素を生成する水性ガスシフト反応は次式で表される。

この反応は発熱反応であり（$\Delta H^\circ < 0$）、反応前後の気体分子のモル数が等しいためエントロピー変化 $\Delta S^\circ$ はほぼゼロである。この系における $\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ$ の挙動について考察せよ。

## 問3

炭酸カルシウムの熱分解反応は次式で表される。


$$\text{CaCO}_3(\text{固}) \rightleftarrows \text{CaO}(\text{固}) + \text{CO}_2(\text{気}) \quad (\Delta H^\circ > 0, \ \Delta S^\circ > 0)$$


この反応は極めて大きな吸熱反応（$\Delta H^\circ \gg 0$）であり、気体が発生するためエントロピー変化も大きい（$\Delta S^\circ \gg 0$）。室温では $\text{CaCO}_3$ は安定に存在するが、高温に加熱すると分解が進行する。この現象を $\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ$ の符号変化に基づいて説明せよ。

---

#### 【解答・解説のポイント】

## 問1

1. **答え:** $\Delta G^\circ > 0$（正）。反応物（$\text{N}_2\text{O}_4$）側が熱力学的に安定であり、正反応は自発的に進行しない。
* *解説:* $\Delta H^\circ > 0$（吸熱）であるため、低温では $-TS$ 項よりも $\Delta H^\circ$ のプラスの寄与が勝り、$\Delta G^\circ > 0$ となるため。


2. **答え:** $\Delta G^\circ < 0$（負）に逆転し、正反応（分解方向）が自発的に進行するようになる（平衡は右に傾く）。
* *解説:* $T$ を高くすると $-TS^\circ$ 項（マイナスの寄与）の絶対値が大きくなり、$\Delta H^\circ$ を上回るため、$\Delta G^\circ$ が $0$ を下回って符号が反転するから。


## 問2

1. **答え:** $\Delta S^\circ \approx 0$ であるため、温度 $T$ を変化させても $-TS^\circ$ 項の寄与が小さく、$\Delta G^\circ$ は主にエンタルピー項 $\Delta H^\circ$ によって決定される。
2. **答え:** $\Delta H^\circ < 0$ であるため、幅広い温度範囲で $\Delta G^\circ < 0$ となり、正反応（水素生成）が自発的に進みやすい。ただし、温度が高くなるとわずかな $-TS$ 項や熱力学的制限により $\Delta G^\circ$ が $0$ に近づくため、より負の値を維持できる「低温」のほうが $\Delta G^\circ$ の観点から生成系に有利である。


## 問3

1. **答え:**
* **室温付近:** $\Delta H^\circ$ が非常に大きなプラスである一方、$-TS^\circ$ 項のマイナス効果が小さいため、$\Delta G^\circ = \Delta H^\circ - T\Delta S^\circ > 0$ となり、分解は自発的に起きない（$\text{CaCO}_3$ が安定）。
* **高温加熱時:** $T$ が十分に高くなると、大きな $\Delta S^\circ$ とのかけ算である $-TS^\circ$ 項のマイナスが $\Delta H^\circ$ のプラスを上回る。これにより $\Delta G^\circ < 0$ へと符号が逆転し、高温でのみ熱分解が自発的に進行するようになる。
