## 1. 有効に利用できるエネルギーという考え方


1人の人が

$$
\longrightarrow
$$

だけの力を持っているとします。

何人か集まって、ロープにつながれた重たい荷物を運ぼうとしています。

# 全員が同じ方向に引く場合

みんなで左向きに引きます。

$$
\longleftarrow \longleftarrow \longleftarrow \longleftarrow \longleftarrow
$$

すると力が足し合わされて荷物は動きます。

### この状態の特徴

- 力がそろっている
- 力を有効利用できる
- 仕事ができる

つまり

> エネルギーを取り出しやすい状態

です。

# 全員がバラバラに引く場合

$$
\leftarrow\quad\rightarrow\quad\uparrow\quad\downarrow\quad\nearrow\quad\swarrow
$$

相殺され、ほとんど動きません。

### この状態の特徴

- 力そのものは存在する
- しかし相殺される
- 荷物は動かない

つまり

> エネルギーはもっているが仕事として利用できない状態

です。

# エンタルピー $H$

ここで、全員が出している力の総量が

$$
H
$$

に相当します。

例えば6人がそれぞれ10 Nで引いているなら$H=60$です。

重要なのは、**整列していてもバラバラでも総量は変わらない** $H=60$であることです。

# エントロピー $S$

力の向きのバラバラさを表す量を考えます。

これがエントロピーです。

## 整列

$$
\leftarrow\ \leftarrow\ \leftarrow\ \leftarrow\ \leftarrow\ \leftarrow
$$

エントロピー小

$$
S\text{ 小}
$$

## バラバラ

$$
\leftarrow\ \rightarrow\ \uparrow\ \downarrow\ \nearrow\ \swarrow
$$

エントロピー大

$$
S\text{ 大}
$$

エントロピーは「乱雑さ」の尺度です。

# ギブス自由エネルギー $G$

私たちが本当に欲しいのは荷物を動かせる力です。

実際に利用できる力を

$$
G
$$

と考えます。

## 整列状態

総力 H = 60

利用可能 G = 55

損失 TS = 5

## バラバラ状態

総力 H = 60

利用可能 G = 10

損失 TS = 50

ここで

$$
TS
$$

は、バラバラさのために利用できなくなった部分を表します。

少しこじつけですが、温度$T$が高いほど乱雑になりやすいとイメージしてください。


$$
H = G + TS
$$

$$
\text{総エネルギー} = \text{利用できるエネルギー} + \text{利用できないエネルギー}
$$

並べ替えて


$$
G = H - TS
$$

## 2. 「利用できるエネルギー」から「変化のポテンシャル」への言い換え

物理化学や熱力学において、$G$（ギブス自由エネルギー）の本質は「その系が自発的に変化するために使える、残されたエネルギーの余裕（ポテンシャル）」です。

綱引きのたとえの言葉を少しだけアップデートして、次のように繋げます。

> 「荷物を動かす仕事」だけでなく、**「状態を変化させる（自ら進んで現象を起こす）ための原動力」**としても $G$ は働いています。
> 自然界は、より楽な方、より自然な方へ向かおうとします。そのときに「まだ変化を起こす余力（ポテンシャル）がどれだけ残っているか」が $G$ です。

---

### 2.1.  具体例：氷 $\rightarrow$ 水 の相変化



### ステップ①：分子の視点で $H$ と $S$ を見直す

水分子の集まりを、先ほどの綱引きになぞらえます。

**氷（カチッと整列した状態）**
* 分子同士がしっかり手をつないで安定している $\rightarrow$ **エンタルピー $H$ は低い**（エネルギー的に落ち着いている）
* 分子は身動きが取れない $\rightarrow$ **エントロピー $S$ は小さい**


**水（バラバラに動き回る状態）**
* 分子同士の結合がほどけて自由 $\rightarrow$ **エンタルピー $H$ は高い**（エネルギー的には少し損をしている、バラバラにするためのコストがかかっている）
* 分子が自由に動き回る $\rightarrow$ **エントロピー $S$ は大きい**



### ステップ②：温度 $T$ という「ざわつき（揺らぎ）」の導入

ここで、綱引きのたとえに出てきた「温度 $T$」が効いてきます。

* 温度が高い $\quad=\quad$ 周囲からの熱エネルギーによって、分子が激しく揺さぶられる状態（バラバラになりたがる勢い）。
* つまり、$TS$ という項は、「温度のせいで、バラバラ（エントロピー大）でいたいという要求がどれくらい強まるか」を表すコスト（損失）になります。

### ステップ③：ギブス自由エネルギー $G = H - TS$ で勝負が決まる

自然界は、トータルの自由エネルギー $G$ が最も低くなる状態（安定な状態）に落ち着こうとします。ここで、氷と水の $G$ のバランスを温度ごとに見ます。

>参考までに氷、水、水蒸気のギブス自由エネルギー$G$を計算してグラフにしてみました。

````python
#@tile H2O三態のギブス自由エネルギーの温度依存性
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. 物性値
# ==========================================

T0 = 298.15       # 基準温度 [K]
Tm = 273.15       # 融点 [K]
Tb = 373.15       # 沸点 [K]

# 定圧モル熱容量 [J/(mol K)]
Cp_ice = 37.6
Cp_water = 75.3
Cp_vapor = 33.6

# 相転移エンタルピー [J/mol]
dH_fus = 6010.0
dH_vap = 40650.0

# 液体水の標準生成エンタルピー [J/mol]
H_water_0 = -285830.0

# 液体水の標準モルエントロピー [J/(mol K)]
S_water_0 = 69.91


# ==========================================
# 2. 各相の基準エンタルピー
# ==========================================

# 融点での融解エンタルピーを満たすように設定
H_ice_0 = (
    H_water_0
    - dH_fus
    - (Cp_water - Cp_ice) * (T0 - Tm)
)

# 沸点での蒸発エンタルピーを満たすように設定
H_vapor_0 = (
    H_water_0
    + dH_vap
    + (Cp_vapor - Cp_water) * (T0 - Tb)
)


# ==========================================
# 3. 各相の基準エントロピー
# ==========================================

# 融点で G_ice = G_water となるように設定
S_ice_0 = (
    S_water_0
    + Cp_water * np.log(Tm / T0)
    - dH_fus / Tm
    + Cp_ice * np.log(T0 / Tm)
)

# 沸点で G_water = G_vapor となるように設定
S_vapor_0 = (
    S_water_0
    + Cp_water * np.log(Tb / T0)
    + dH_vap / Tb
    + Cp_vapor * np.log(T0 / Tb)
)


# ==========================================
# 4. H(T), S(T), G(T)
# ==========================================

def H(T, H0, Cp):
    """モルエンタルピー [J/mol]"""
    return H0 + Cp * (T - T0)

def S(T, S0, Cp):
    """モルエントロピー [J/(mol K)]"""
    return S0 + Cp * np.log(T / T0)

def G(T, H0, S0, Cp):
    """モルギブズ自由エネルギー [J/mol]"""
    return H(T, H0, Cp) - T * S(T, S0, Cp)


# ==========================================
# 5. 温度範囲と計算
# ==========================================

T = np.linspace(240, 400, 500)

G_ice = G(T, H_ice_0, S_ice_0, Cp_ice)
G_water = G(T, H_water_0, S_water_0, Cp_water)
G_vapor = G(T, H_vapor_0, S_vapor_0, Cp_vapor)


# ==========================================
# 6. グラフ
# ==========================================

plt.figure(figsize=(10, 6))

plt.plot(T, G_ice / 1000, label="Ice", linewidth=2.5)
plt.plot(T, G_water / 1000, label="Liquid water", linewidth=2.5)
plt.plot(T, G_vapor / 1000, label="Water vapor", linewidth=2.5)

plt.axvline(Tm, color="gray", linestyle="--", alpha=0.7)
plt.axvline(Tb, color="gray", linestyle="--", alpha=0.7)

plt.xlabel("Temperature [K]")
plt.ylabel("Standard molar Gibbs energy [kJ/mol]")
plt.title("Gibbs Free Energy of Water in Three Phases")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()


# ==========================================
# 7. 融点・沸点での確認
# ==========================================

for temp, label in [(Tm, "Melting point"), (Tb, "Boiling point")]:
    print(f"\n{label}: T = {temp:.2f} K")
    print(f"G_ice   = {G(temp, H_ice_0, S_ice_0, Cp_ice)/1000:.3f} kJ/mol")
    print(f"G_water = {G(temp, H_water_0, S_water_0, Cp_water)/1000:.3f} kJ/mol")
    print(f"G_vapor = {G(temp, H_vapor_0, S_vapor_0, Cp_vapor)/1000:.3f} kJ/mol")
````





1. **低温（$0^\circ\text{C}$ 未満：冷凍庫の中）**
* 温度 $T$ が低いので、$-TS$ の影響は小さい。
* 勝負の決め手は $H$（結びつきの強さ）。
* 結びつきが強い「氷」の方が $G$ が低くなる $\rightarrow$ **氷のままでいる**


2. **常温（$0^\circ\text{C}$ 超：常温）**
* 温度 $T$ が高くなると、$-TS$（バラバラになりたい効果）のマイナスが大きくなる。
* 「氷は $H$ が小さくて有利だけど、高温では $TS$（バラバラのコスト）が大きすぎて不利になりすぎる！」
* 結果として、トータルの $G$ は「水」の方が低くなる $\rightarrow$ **自発的に水に溶ける（融解）**

3. 水蒸気は説明省略



---

> **まとめのフレーズ**
> * **エンタルピー ($H$)** ＝「結びつきの強さ（まとまろうとする力）」
> * **エントロピー ($S$)** ＝「バラバラになろうとする乱雑さ」
> * **温度 ($T$)** ＝「その乱雑さをどれくらい推し進めるかの倍率」
> 
> 
> この $H$（まとまりたい）と $TS$（バラバラになりたい）の**せめぎ合いの結果（差し引きの余力）**が、ギブス自由エネルギー $G = H - TS$ です。
> 自然界は常に、**「 $G$ がより低い状態（より安定な状態）」**に向かって進みます。だから、温度が変わると氷が水になったり、化学反応が進んだりするのです。



---
状態の安定性（どの相が一番居心地が良いか）を $G$ の値の大小で決めるステップから、いよいよ「ある状態から別の状態へ、自発的に変化が起きるかどうか」を $\Delta G$（デルタG）で判定するステップへ進めます！

## 3.$\Delta G$ で読み解く「自発的な変化」の優位性

### 3.1. 導入：なぜ「 $\Delta$（デルタ）」が必要なのか？

* **これまでの復習（静的）：**
* ある温度 $T$ において、「どの状態の $G$ が最も低いか」を見ることで、そのとき最も安定な状態（氷なのか、水なのか、気体なのか）を決めました。




* **ここからの新しい視点（動的）：**
* 「今、氷のままでいるけど、本当に水に変われるの？」
* 「この化学反応は、勝手に進むの？ それとも勝手には進まないの？」
* このように、「ある状態から別の状態へ変化するとき、その現象が自発的に起きるかどうか」を知るためには、状態そのものの $G$ ではなく、変化の前後での差分（変化量 $\Delta G$）を計算する必要があります。



---

### 3.2. $\Delta G$、$\Delta H$、$\Delta S$ の定義

変化の前後（「反応前＝始状態」から「反応後＝終状態」）の差分を $\Delta$（デルタ）で表します。

#### **$\Delta G$ （ギブス自由エネルギーの変化量）**
* $\Delta G = G_{\text{後}} - G_{\text{前}}$
* **意味：** この変化が自発的に起きるかどうかの**総合的な判定基準**。


#### **$\Delta H$ （エンタルピーの変化量）**
* $\Delta H = H_{\text{後}} - H_{\text{前}}$
* **意味：** 発熱・吸熱の度合い（結合の安定性の変化）。マイナスなら発熱（安定な結合ができた）。


#### **$\Delta S$ （エントロピーの変化量）**
* $\Delta S = S_{\text{後}} - S_{\text{前}}$
* **意味：** 乱雑さ・自由度の変化。プラスならよりバラバラになった。



そして、これらを結ぶ基本関係式（温度 $T$ は一定とする）がこちらです：


$$\Delta G = \Delta H - T\Delta S$$

---

### 3.3. 自発変化のルール：なぜ $\Delta G < 0$ なのか？

自然界のすべての自発的な現象（高いところから水が落ちる、熱いお湯が冷めるなど）は、「全体としてエネルギーがより安定（低い方）へ向かおうとする」性質があります。

* **$\Delta G < 0$ （マイナス）の場合**
* 後の状態の $G$ 方が小さいため、**「勝手に（自発的に）進む」**！


* **$\Delta G > 0$ （プラス）の場合**
* 後の状態の方が $G$ が大きくなってしまうため、**「勝手には進まない」**（外からエネルギーを入れないと起きない）。



> **イメージのたとえ：**
> * ギブス自由エネルギー $G$ は「山の高さ（位置エネルギー）」のようなものです。
> * ボールが坂道を**勝手に転がり下りる**とき、高さの差（$\Delta G$）は**マイナス**になります。自然界は常に $\Delta G < 0$ の方向へ流れます。
> 
>

## 4. 化学変化（および相変化）の優位性を論じる

では、ある化学変化や相変化（例：氷 $\rightarrow$ 水）が起きるかどうかは、どのように決まるのでしょうか？
方程式 $\Delta G = \Delta H - T\Delta S$ の右辺にある「2つのライバル（$\Delta H$ と $T\Delta S$）」の力関係で決まります。

### 勝負を決める2つの因子

1. **エネルギー的安定さを好む力 ($\Delta H$)**
* 物体同士がくっついて安定しようとする（マイナスだと有利）。


2. **乱雑さを好む力 ($T\Delta S$)**
* より自由に、バラバラに広がろうとする（これが大きいと $\Delta G$ を下げる強力な武器になる）。

>### 4 つのパターン一覧
>1. **$\Delta H < 0$ かつ $\Delta S > 0$** $\Rightarrow$ **常に $\Delta G < 0$** （どんな温度でも自発的）
>2. **$\Delta H > 0$ かつ $\Delta S < 0$** $\Rightarrow$ **常に $\Delta G > 0$** （どんな温度でも進まない）
>3. **$\Delta H < 0$ かつ $\Delta S < 0$（発熱＆秩序化）** $\Rightarrow$ **低温でのみ自発的**（例：水が凍って氷になる）
>4. **$\Delta H > 0$ かつ $\Delta S > 0$（吸熱＆乱雑化）** $\Rightarrow$ **高温でのみ自発的**（例：氷が溶けて水になる）

ある温度（例えば $0^\circ\text{C} = 273.15\text{K}$）で氷から水へ変わる瞬間、何が起きているでしょうか？

* 氷点下（例: $250\text{K}$付近）の場合


* グラフを見ると、氷（Ice）の $G$ が一番下にあります。


* ここから「水」になろうとすると、$G$ の高い方へ逆らって登ることになるため、変化の差分は $\Delta G > 0$ となり、**水には変わらない（氷のままが安定）**。


* **$0^\circ\text{C}$ を超えた場合（例: $280\text{K}$以上）**
* 温度が上がると、水（Liquid water）の $G$ の方が氷よりも下（低く）になります。


* すると、氷から水へ変わるとき、$G$ が高いところから低いところへ下るため、**$\Delta G < 0$ となり、氷は自発的に水に溶け出す（融解が起きる）**！



> **まとめ**
> 「化学変化や相変化の優位性とは、**『最終的に $\Delta G$ をマイナスにできるかどうか』**で決まります。
> 物質は、温度 $T$ という環境が変わったとき、自分の $\Delta H$ と $\Delta S$ のバランスを計算し、**『今、俺の $\Delta G$ はマイナスになれるか？』**を判断して、自ら進むべき方向（変化）を選んでいるのです。」

# 12. エンタルピーの立場で考える

氷の分子は規則正しく並んでいます。

```text
● ● ● ●
● ● ● ●
● ● ● ●
```

水になるには水素結合の一部を壊す必要があります。

そのため熱を吸収します。

$$
\Delta H > 0
$$

エンタルピーだけ見ると

```text
氷 → 水
```

は不利です。

---

# 13. エントロピーの立場で考える

氷では分子はほぼ固定されています。

```text
● ● ● ●
● ● ● ●
● ● ● ●
```

水になると自由に動けます。

```text
●   ●  ●

  ●

      ●
```

状態数が増えるため

$$
\Delta S > 0
$$

です。

綱引きでいえば

```text
整列
↓
バラバラ
```

です。

---

# 14. ギブス自由エネルギーで考える

自然現象は

$$
\Delta G=\Delta H-T\Delta S
$$

で決まります。

氷の融解では

$$
\Delta H > 0
$$

かつ

$$
\Delta S > 0
$$

です。

つまり

```text
エネルギー的には不利

しかし

乱雑さ的には有利
```

という綱引きになります。

---

# 15. 低温の場合

低温では

$$
T\Delta S
$$

が小さいため

$$
\Delta H
$$

の影響が勝ちます。

その結果

$$
\Delta G > 0
$$

となり、氷が安定です。

---

# 16. 高温の場合

高温では

$$
T\Delta S
$$

が大きくなります。

やがて

$$
T\Delta S > \Delta H
$$

となり、

$$
\Delta G < 0
$$

になります。

その結果、氷は自然に溶けて水になります。

---

# 17. 融点では何が起こっているのか

融点では

$$
\Delta G = 0
$$

です。

したがって

$$
\Delta H = T\Delta S
$$

となります。

これは

```text
エネルギーを下げたい力

＝

自由に散らばりたい力
```

がちょうど釣り合った状態です。

そのため

```text
氷 ⇔ 水
```

が共存できます。

---

# 綱引きとの対応

```text
氷
↓
規則正しい
↓
エントロピー小
```

```text
水
↓
自由に動ける
↓
エントロピー大
```

```text
氷→水

ΔH > 0（エネルギー的には損）

ΔS > 0（乱雑さ的には得）
```

したがって

$$
\Delta G=\Delta H-T\Delta S
$$

で勝敗が決まります。

---

# まとめ

```text
氷が溶けるためにはエネルギーが必要である
（ΔH > 0）

しかし分子は自由になれる
（ΔS > 0）

低温ではエネルギーの不利が勝つ

高温では自由になりたがる効果が勝つ
```

その勝敗を決める審判が

$$
\Delta G=\Delta H-T\Delta S
$$

です。
