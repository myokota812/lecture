# 解答例

## 問1

沸点上昇の式

$$
\Delta T = K_b m
$$

ここで

$$
m:\ \text{質量モル濃度}
\qquad
\left[
\frac{\mathrm{mol\!-\!溶質}}
{\mathrm{kg\!-\!溶媒}}
\right]
$$

である。

質量モル濃度は

$$
m
=
\frac{\dfrac{0.698}{M}}
{63.88\times10^{-3}}
$$

より，

$$
0.145
=
2.53
\times
\frac{0.698}
{63.88\times10^{-3}\times M}
$$

したがって，

$$
M
=
\frac{2.53\times0.698}
{63.88\times10^{-3}\times0.145}
$$

$$
M
=
190.7
$$

$$
\boxed{M \simeq 191}
$$

# 発展：エントロピーから「ラウールの法則」を導く

## 1. 化学ポテンシャル（$\mu$）とは何か？（「分子の逃げやすさ」）

熱力学では、物質の「その場所での居心地の悪さ」や「そこから逃げ出したい度合い」を化学ポテンシャル（符号：$\mu$）という指標で表します。

* **化学ポテンシャルが高い（$\mu$ が大）：** 居心地が悪いため、他の場所へ逃げたがる（蒸発しやすい、反応しやすい）。
* **化学ポテンシャルが低い（$\mu$ が小）：** 居心地が良いため、その場にとどまりたがる（安定）。

液体（水）と気体（水蒸気）がバランスをとっている（平衡状態）とき、それぞれの化学ポテンシャルは等しくなっています。


$$\mu_{\text{liquid}} = \mu_{\text{vapor}}$$

---

## 2. 砂糖を混ぜると、なぜ液体の化学ポテンシャルが下がるのか？

前回の議論で、砂糖を混ぜるとエントロピー（$S$）が増大し、液体の自由エネルギー（$G$）が低下することを学びました。

化学ポテンシャル $\mu$ は、「物質を1mol増やしたときの自由エネルギーの変化量」です。したがって、液相の自由エネルギーが下がったということは、**水分子の化学ポテンシャル $\mu_{\text{liquid}}$ そのものも下がる**ことを意味します。

> **直感的なイメージ：**
> 砂糖を混ぜて分子の配置が自由になり（エントロピー増大）、液体の居心地が良くなった結果、水分子の「逃げたい度合い（＝化学ポテンシャル）」がグッと下がります。

---

## 3. ラウールの法則の数式的な導出

化学ポテンシャルの低下が、いかにして「蒸気圧降下（ラウールの法則）」を引き起こすのかを数式で追ってみましょう。

### ① 溶液中の溶媒（水）の化学ポテンシャル

純粋な液体（純水）の化学ポテンシャルを $\mu_1^*$ とします。これに溶質（砂糖）を溶かしたとき、溶媒のモル分率（全体のなかで水分子が占める割合）を $x_1$ とすると、化学ポテンシャルはエントロピーの項（ボルツマンの原理に由来する項）によって次のように表されます。


$$\mu_1 = \mu_1^* + RT \ln x_1$$

* ($R$: 気体定数, $T$: 絶対温度, $x_1 < 1$ なので $\ln x_1$ はマイナスになり、$\mu_1$ は $\mu_1^*$ より低くなります)

### ② 気体（蒸気）側の化学ポテンシャル

一方、気体（蒸気）側の化学ポテンシャルは、そのときの蒸気圧 $P_1$ を用いて次のように表されます。


$$\mu_1(\text{g}) = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1}{P^{\circ}}$$


*(※ $\mu_1^{\circ}$ は標準状態での化学ポテンシャル、$P^{\circ}$ は標準圧力)*

### ③ 平衡状態（液＝気）からの導出

液体と気体が平衡にあるとき、両者の化学ポテンシャルは等しくなります ($\mu_1 = \mu_1(\text{g})$)。


$$\mu_1^* + RT \ln x_1 = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1}{P^{\circ}}$$

ここで、純水の場合（溶質を混ぜていないとき、$x_1 = 1, P_1 = P_1^*$）の平衡関係を考えると、


$$\mu_1^* = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1^*}{P^{\circ}}$$


となります。これを先ほどの式に代入して整理すると、綺麗な関係式が導かれます。

$$\mu_1^* + RT \ln x_1 = \mu_1^* + RT \ln \left(\frac{P_1}{P_1^*}\right)$$

両辺から $\mu_1^*$ と $RT$ を引いて整理すると、


$$\ln x_1 = \ln \left(\frac{P_1}{P_1^*}\right)$$


対数を外すと、ついに**ラウールの法則**が得られます！


$$P_1 = x_1 P_1^*$$

---

## 4. まとめ：すべては「エントロピー」から繋がっている

* 砂糖を混ぜる $\rightarrow$ **配置のエントロピー増大**
* 液体の自由エネルギーが下がり、**化学ポテンシャル $\mu_1$ が低下**
* 逃げ出したい勢いが弱まるため、気体側の圧力（蒸気圧 $P_1$）も押し下げられる（**ラウールの法則**の成立）
