## 1. エネルギーの移動：熱と仕事

### 🔬 内部エネルギー $U$ の正体

フラスコやビーカーの中に入っている分子たちは、激しく動き回ったり、分子同士で手をつないだり（結合したり）しています。
- 動き回るエネルギー = **運動エネルギー**
- 結合（手をつなぐ）エネルギー ＝ **ポテンシャルエネルギー（位置エネルギー）**
で、その総和を内部エネルギー$U$といいます。

$$\text{内部エネルギー } U = \text{（運動エネルギーの総和）} + \text{（ポテンシャルエネルギーの総和）}$$

>補足 💡 化学反応で何が起きているのか？
>
>化学反応が起きて物質が変化するとき：
>
>1. 分子の手をつなぎ直す（化学結合の組み換え）が起きます。
>2. これにより**ポテンシャルエネルギー**が大きく変化します。
>3. その結果、余ったエネルギーが外へあふれ出たり（発熱反応）、足りない分を外から奪ったり（吸熱反応）します。これが、第4回で学ぶ「熱（エンタルピー変化）」の正体です。
>
>このミクロなイメージが頭に入っていれば、これからの熱力学の数式やエネルギーダイアグラムが理解しやすいと思います。

### 🔬 内部エネルギーを変化させるには

この内部エネルギーを増減させる方法は2つあります。
 1. **熱を加える（または奪う）** ＝ $q$
 2. **外から力仕事をする（またはされる）** ＝ $w$
 
 つまり

最初にフラスコ（系）の中にある分子たちの内部エネルギーが **$U_1$** で<br>
そこに外から熱が入り（$+q$）、外から仕事が加えられた（$+w$）<br>
その結果、内部エネルギーの残高が変化して **$U_2$** になったとき<br>

$$U_1 + q + w = U_2$$
ゴールからスタートを引いたものを「変化量 $\Delta U$」とおくと。
$$U_2 - U_1 = \Delta U$$
したがって

## $$\Delta U = q + w$$

これが熱力学の第一法則とよばれるものです。

* **解説と気づき：**

### 熱力学第一法則（First Law of Thermodynamics）
系の内部エネルギーの変化 $\Delta U$ は、系に与えられた **熱量（$q$）** と系になされた **仕事（$w$）** の和に等しい（エネルギー保存則）。

$$\Delta U = q + w$$

**内部エネルギーを増加させるためには、熱 ($q$) を加えれば良いのですが、残念ながら、加えた熱エネルギーはすべて内部エネルギーに消費されるわけではありません。

$$\Delta U \neq q$$

与えられたエネルギーの一部は膨張（ピストンを押し上げる）仕事（$-P\Delta V$）に使われて外に逃げてしまし、中に残る正味のエネルギー変化（$\Delta U$）はそのぶん目減りします。

### $$\Delta U = q - P \Delta V$$

この式が大切です。ただし、定圧下であることを認識しておいてください。

この式を書き換えると

#### $$ q = \Delta U +  P \Delta V$$

この式を言葉で表現すれば、「与えた熱エネルギーは内部エネルギーの上昇と膨張仕事に分散されてしまうということです。

  > 実験室のフラスコ（定圧条件）では、気体の膨張による仕事（$-P \Delta V$）が勝手について回ります。その面>倒な仕事をまとめて、**「私たちが測りたい本当の熱量（$q_p$）」**だけを取り出すために生まれたのが、次の>「エンタルピー ($H$)」という概念です。

## 2. エンタルピー（Enthalpy, $H$）の定義

定圧条件下で出入りする熱量を直接表す利便性の高い**状態量**として、エンタルピー $H$ を定義します。

$$H = U + PV$$

定圧下（$P = \text{const}$）におけるエンタルピー変化 $\Delta H$：
$$\Delta H = \Delta U + P \Delta V = q_p$$

### 発熱反応と吸熱反応
- **発熱反応（Exothermic Reaction）：** 
  - 体系から周囲へ熱が放出される。 
  - $\Delta H < 0$ （生成物のエンタルピーが反応物より低い）
- **吸熱反応（Endothermic Reaction）：** 
  - 周囲から体系へ熱が吸収される。 
  - $\Delta H > 0$ （生成物のエンタルピーが反応物より高い）

```python
# Pythonによる発熱反応と吸熱反応のエネルギーダイヤグラムの可視化
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.5))

# 発熱反応 (Exothermic)
ax1.plot([0, 1], [100, 100], 'b-', linewidth=3, label='Reactants')
ax1.plot([2, 3], [30, 30], 'r-', linewidth=3, label='Products')
ax1.annotate('', xy=(2.5, 30), xytext=(2.5, 100),
             arrowprops=dict(arrowstyle='->', lw=2, color='black'))
ax1.text(2.6, 65, '$\Delta H < 0$\n(Exothermic)', fontsize=12, color='red')
ax1.set_title('Exothermic Reaction')
ax1.set_ylabel('Enthalpy $H$')
ax1.set_xticks([])
ax1.set_ylim(0, 130)
ax1.grid(True, alpha=0.3)

# 吸熱反応 (Endothermic)
ax2.plot([0, 1], [30, 30], 'b-', linewidth=3, label='Reactants')
ax2.plot([2, 3], [100, 100], 'r-', linewidth=3, label='Products')
ax2.annotate('', xy=(2.5, 100), xytext=(2.5, 30),
             arrowprops=dict(arrowstyle='->', lw=2, color='black'))
ax2.text(2.6, 65, '$ \Delta H > 0$\n(Endothermic)', fontsize=12, color='blue')
ax2.set_title('Endothermic Reaction')
ax2.set_xticks([])
ax2.set_ylim(0, 130)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 3.1. 熱化学方程式（反応エンタルピー）の書き方

化学反応に伴う熱の出入りは、従来の「＝」を使った式ではなく、反応エンタルピー（$\Delta H$）を用いて表します。

### 基本ルール

1. **矢印を使う**：物質の変化を **$\rightarrow$** でつなぎます（「＝」は使いません）。
2. **状態を書く**：物質の化学式の後ろに、状態をカッコ書きで添えます。
* 気体：$(g)$
* 液体：$(l)$
* 固体：$(s)$ ※黒鉛などは $\text{C}(黒鉛)$ のように書くこともあります。


3. **エンタルピー変化 ($\Delta H$) を添える**：式の右側に、単位（$\text{kJ/mol}$）とともに $\Delta H$ の値を書きます。
* **発熱反応**：まわりに熱を放出する $\rightarrow$ **$\Delta H$ はマイナス（$< 0$）**
* **吸熱反応**：まわりから熱を吸収する $\rightarrow$ **$\Delta H$ はプラス（$> 0$）**



> **【例：水素の燃焼】**
>
> $\text{H}_2(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{H}_2\text{O}(l) \quad \Delta H = -286\text{ kJ/mol}$
>
> （水素1 molが燃焼して液体用水ができるとき、286 kJの熱を放出する）<br>
>注意：エンタルピーの単位はJ/molです。基本的に反応物質のうち主役となる物質を1molとするため、反応式の量論係数が非整数もでてきます。

---

###  3.2. ヘスの法則（総エンタルピー保存の法則）とは？

化学反応のエンタルピー変化（出入りする熱の総量）は、**「反応の最初（出発物）」と「最後（生成物）」が同じであれば、途中でどのような経路を通っても常に一定である**という法則です。

* **何が便利なの？**
実験で直接測るのが難しい反応の熱も、すでにわかっている他の反応の式をパズルのように足し引きすることで、計算だけで求めることができます。

---

### 3.3. ヘスの法則を使った計算のコツ

方程式を**連立方程式やパズルのように扱う**のがコツです。

1. **式を何倍かする**
* 例：式全体を2倍にしたら、$\Delta H$ の値も **2倍** にする。


2. **式を逆向きにする**
* 例：左辺と右辺をひっくり返したら、$\Delta H$ の**符号を反転**させる（プラス $\iff$ マイナス）。


3. **式を足し引きする**
* 求めたい式になるように複数の式を足すと、途中にしかない物質（中間物質など）が相殺されて消えていきます。それに合わせて $\Delta H$ もそのまま足し引きします。



---

ヘスの法則は「目的地（ゴール）さえ同じなら、遠回りしようが近道しようが、消費するエネルギーの合計は同じ」というイメージで捉えると、とても分かりやすくなります！

### 例題１

次の熱化学方程式を利用して、**一酸化炭素 $\text{CO}(g)$ の生成エンタルピー**（または生成熱）を求めよ。

1. $\text{C}(黒鉛) + \text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_1 = -394\text{ kJ/mol}$
2. $\text{CO}(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_2 = -283\text{ kJ/mol}$

---

#### 解説

求めたいのは「一酸化炭素 $\text{CO}$ の生成エンタルピー」なので、以下の反応の $\Delta H$ を求めます。

* **目標の式**：$\text{C}(黒鉛) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{CO}(g) \quad \Delta H$

ヘスの法則（総エンタルピー保存の法則）より、化学反応のエンタルピー変化は経路によらず一定であるため、式をそのまま足し引きして $\Delta H$ を計算できます。

* 式①はそのまま：
$$\text{C}(黒鉛) + \text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_1 = -394\text{ kJ/mol}$$
* 式②を逆向きにする（＝両辺を入れ替えて、$\Delta H$ の符号を反転させる）：
$$\text{CO}_2(g) \rightarrow \text{CO}(g) + \frac{1}{2}\text{O}_2(g) \quad -\Delta H_2 = +283\text{ kJ/mol}$$

これら２つの式を足し合わせると、$\text{CO}_2$ が相殺されて目標の式が得られます。


$$\Delta H = \Delta H_1 - \Delta H_2 = -394 - (-283) = -111\text{ kJ/mol}$$

### 例題２

次の熱化学方程式を利用して、**エタノール $\text{C}_2\text{H}_5\text{OH}(l)$ の生成エンタルピー**を求めよ。

1. $\text{C}(黒鉛) + \text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_1 = -394\text{ kJ/mol}$
2. $\text{H}_2(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{H}_2\text{O}(l) \quad \Delta H_2 = -286\text{ kJ/mol}$
3. $\text{C}_2\text{H}_5\text{OH}(l) + 3\text{O}_2(g) \rightarrow 2\text{CO}_2(g) + 3\text{H}_2\text{O}(l) \quad \Delta H_3 = -1368\text{ kJ/mol}$

---

#### 解説

求めたいのは「エタノールの生成エンタルピー」なので、以下の反応の $\Delta H$ を求めます。

* **目標の式**：$2\text{C}(黒鉛) + 3\text{H}_2(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{C}_2\text{H}_5\text{OH}(l) \quad \Delta H$

与えられた式を組み合わせて目標の式を作ります。

1. **式①を 2倍** する（炭素を2つにするため）
2. **式②を 3倍** する（水素を3つにするため）
3. **式③を逆向き**にする（エタノールを右辺に移すため）

それぞれのエンタルピー変化は以下のようになります。

* ① $\times 2$：$\Delta H = 2 \times (-394) = -788\text{ kJ/mol}$
* ② $\times 3$：$\Delta H = 3 \times (-286) = -858\text{ kJ/mol}$
* ③を逆向き：$\Delta H = -(-1368) = +1368\text{ kJ/mol}$

これらを足し合わせます（①$\times 2$ ＋ ②$\times 3$ － ③）：


$$\Delta H = (-788) + (-858) - (-1368)$$

$$\Delta H = -1646 + 1368 = -278\text{ kJ/mol}$$

したがって、エタノールの生成エンタルピーは $-278\text{ kJ/mol}$ と求まります。

## ステップ1：目標の式の「定義」を言葉にする

求められているのは「エタノール $\text{C}_2\text{H}_5\text{OH}(l)$ の生成エンタルピー」です。<br>
ここで「生成」の定義を思い出します。

* **生成（生成エンタルピー）の定義**：
**成分元素の単体**から、目的の化合物**1 mol**が生成するときの反応。

つまり、作らなければならない目標の式は：<br>
> **（成分元素の単体たち）$\rightarrow 1 \text{C}_2\text{H}_5\text{OH}(l)$**

という形になります。

---

## ステップ2：右辺と左辺を埋める

1. **右辺を決める**：
目的物であるエタノールを**係数1**で右辺に置きます。

$$\dots \rightarrow \text{C}_2\text{H}_5\text{OH}(l)$$


2. **左辺の元素を抜き出す**：
$\text{C}_2\text{H}_5\text{OH}$ に含まれる元素は **$\text{C}$（炭素）**、**$\text{H}$（水素）**、**$\text{O}$（酸素）** の3つです。
3. **それぞれを「最も安定な単体」の形で左辺に書く**：
* 炭素の単体 $\rightarrow \text{C}(黒鉛)$
* 水素の単体 $\rightarrow \text{H}_2(g)$
* 酸素の単体 $\rightarrow \text{O}_2(g)$


これを左辺に並べます。

$$\text{C}(黒鉛) + \text{H}_2(g) + \text{O}_2(g) \rightarrow \text{C}_2\text{H}_5\text{OH}(l)$$


4. **右辺（エタノール1 mol）に合わせて、左辺の係数を合わせる**：
* **$\text{C}$**：右辺に2個あるので $\rightarrow$ **$2\text{C}$**
* **$\text{H}$**：右辺に6個（$5 + 1$）あるので $\rightarrow$ **$3\text{H}_2$**
* **$\text{O}$**：右辺に1個あるので $\rightarrow$ **$\frac{1}{2}\text{O}_2$**



---

### 完成した目標の式

$$\mathbf{2\text{C}(黒鉛) + 3\text{H}_2(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{C}_2\text{H}_5\text{OH}(l) \quad \Delta H}$$

---

## ステップ3：目標の式と「与えられた式」を照らし合わせる

目標の式ができたら、あとは与えられた1〜3の式と「特有の物質（その式にしか出てこない物質）」に着目して、何倍・逆向きにするかを決定します。

* **$\text{C}(黒鉛)$ に注目**：
目標の式の左辺に **$2\text{C}$** が必要 $\rightarrow$ 式1を **2倍** する。
* **$\text{H}_2(g)$ に注目**：
目標の式の左辺に **$3\text{H}_2$** が必要 $\rightarrow$ 式2を **3倍** する。
* **$\text{C}_2\text{H}_5\text{OH}(l)$ に注目**：
目標の式の右辺に **$1\text{C}_2\text{H}_5\text{OH}$** が必要 $\rightarrow$ 式3はエタノールが左辺にあるので逆向き（$-1$倍）にする。

（※$\text{O}_2$ や $\text{CO}_2$ など複数の式に登場する物質は、他の物質の係数を合わせれば自然と数が合うので、最初は無視して大丈夫です！）

## まとめ：目標の式がわからないときのチェクリスト

1. **問われている用語の定義を確認する**
* **生成**：構成元素の「単体」$\rightarrow$ 化合物 1 mol
* **燃焼**：物質 1 mol $+$ $\text{O}_2$ $\rightarrow$ 完全燃焼の生成物（$\text{CO}_2$ や $\text{H}_2\text{O}$）


2. **主役の物質を「係数1」で配置する**
3. **両辺の原子の数を合わせる（分数ができてもOK）**

このステップを通せば、どんな複雑な問題でも目標の式を機械的に導き出せます！

## 4. 標準生成エンタルピー（$\Delta H_f^\circ$）から反応エンタルピーを求める方法

### 標準生成エンタルピーについて

- **定義：** 標準状態（$1\text{ atm}, 25^\circ\text{C}$）において、化合物をその構成元素の最も安定な単体から合成するときのエンタルピー変化。
- **基準：** 最も安定な元素単体（例：$\text{O}_2(g), \text{C}(\text{graphite}), \text{H}_2(g)$ など）の標準生成エンタルピーは**ゼロ**と規定します。
  $$\Delta H_f^\circ (\text{単体}) = 0$$

### 標準生成エンタルピーを用いた反応エンタルピー（$\Delta H_{\text{rxn}}^\circ$）の算出公式
$$
\Delta H_{\text{rxn}}^\circ = \sum n \cdot \Delta H_f^\circ (\text{生成物}) - \sum m \cdot \Delta H_f^\circ (\text{反応物})
$$
（「生成物の生成エンタルピーの和」から「反応物の生成エンタルピーの和」を引く）

標準生成エンタルピー（Standard Enthalpy of Formation）は、化学の熱力学や計算において非常に重要な基準となる数値です。

文字通り分解すると、「標準状態」**のもとで、**「成分元素の単体からその化合物が生成する」ときのエンタルピー変化を意味します。

---

## 1. 言葉の意味を分解して理解する

### ①「生成」とは？

先ほど確認した通り、「成分元素の最も安定な単体」から、その化合物 **1 mol** が作られる反応のことです。

* 例：水（$\text{H}_2\text{O}$）の生成反応

$$\text{H}_2(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{H}_2\text{O}(l)$$



### ②「標準」とは？

化学の実験は温度や圧力によってエネルギーの値が微妙に変わるため、世界共通の「基準（標準状態）」を定めています。

* **温度**：$25^\circ\text{C}$ ($298.15\text{ K}$)
* **圧力**：$1\text{ bar}$ （約 $10^5\text{ Pa}$、1気圧に近い値）

この標準状態で測定されたエンタルピー変化であることを示すために、記号の右肩に **$\circ$（度マーク）** をつけて $\Delta H^\circ$ のように表します（生成エンタルピーの場合は $\Delta H^\circ_f$ と書くこともあります）。

---

## 2. 超重要なルール：「安定な単体の標準生成エンタルピーは 0」

ここが理解の大きなポイントです！
**「すでに安定な単体であるもの」を、さらに単体から作ろうとしても、エネルギーの出入りは起こりません。**

そのため、**すべての元素の「最も安定な単体」の標準生成エンタルピーは、一律で $0\text{ kJ/mol}$** と定義されています。

* **例**：
* 炭素の最も安定な単体：黒鉛 $\rightarrow \Delta H^\circ = 0\text{ kJ/mol}$
* 酸素の単体：酸素分子 $\text{O}_2(g)$ $\rightarrow \Delta H^\circ = 0\text{ kJ/mol}$
* 水素の単体：水素分子 $\text{H}_2(g)$ $\rightarrow \Delta H^\circ = 0\text{ kJ/mol}$



---

## 3. なぜ標準生成エンタルピーが便利なのか？（一発計算の公式）

これまでのヘスの法則の計算では、式をパズルのように足し引きしていました。しかし、世の中のすべての物質の「標準生成エンタルピーのデータ一覧表」**があれば、パズルを解かなくても**「生成物の値」と「反応物の値」を足し引きするだけで、どんな反応のエンタルピーでも計算できるようになります。

### 計算の公式

任意の化学反応において、その反応エンタルピー $\Delta H^\circ$ は次の式で求められます。

$$\Delta H^\circ = \sum \Delta H^\circ_f(\text{生成物}) - \sum \Delta H^\circ_f(\text{反応物})$$

日本語で覚えるときはこうです：

> **「（生成物の標準生成エンタルピーの総和）ー（反応物の標準生成エンタルピーの総和）」**

### なぜこの引き算で求められるのか？（イメージ）

ヘスの法則の本質は「どのルートを通ってもエネルギーの総和は同じ」ということです。

1. 一度、すべての反応物を「バラバラの単体（エネルギー基準 $= 0$）」まで分解する。
2. そこから、目的の生成物を「組み立てる」。

この「バラバラにする（反応物の逆）」と「組み立てる（生成物）」を合わせたものが、上の引き算の公式です。この公式を使えば、複雑な方程式のパズルを組まなくても、表の数値を代入するだけで一発で答えが求まるため、大学受験や実務で非常に重宝されます。

---

メタン $\text{CH}_4$ の燃焼反応を例にして、標準生成エンタルピーのデータから**反応エンタルピー（$\Delta H^\circ$）を計算する手順**をわかりやすく解説します。

---

### 1. 対象の化学反応式を確認する

メタンの完全燃焼（生成する水が液体の場合）の化学反応式は次のようになります。

$$\text{CH}_4(g) + 2\text{O}_2(g) \rightarrow \text{CO}_2(g) + 2\text{H}_2\text{O}(l)$$

この反応が起きたときに、全体でどれだけの熱（エンタルピー変化 $\Delta H^\circ$）が出入りするのかを計算します。

---

### 2. 使う公式を確認する

標準生成エンタルピーのデータ（表）がある場合、複雑なパズル（式の足し引き）をしなくても、次の公式に当てはめるだけで一発で計算できます。

$$\Delta H^\circ = \sum \Delta H^\circ_f(\text{生成物}) - \sum \Delta H^\circ_f(\text{反応物})$$

日本語で覚えるときはこうです：

> **「（生成物の生成エンタルピーの総和） ー （反応物の生成エンタルピーの総和）」**

---

### 3. 具体的な数値を用意する

計算に必要な各物質の標準生成エンタルピー（$25^\circ\text{C}$、$1\text{ bar}$ での値の一例）を用意します。

* $\text{CH}_4(g)$ の生成エンタルピー：$-75\text{ kJ/mol}$
* $\text{CO}_2(g)$ の生成エンタルピー：$-394\text{ kJ/mol}$
* $\text{H}_2\text{O}(l)$ の生成エンタルピー：$-286\text{ kJ/mol}$
* $\text{O}_2(g)$ の生成エンタルピー：$0\text{ kJ/mol}$
* ※酸素 $\text{O}_2$ は「最も安定な単体」なので、生成エンタルピーは必ず **$0\text{ kJ/mol}$** になります（試験や問題文に書かれていないことも多いので注意です！）。



---

### 4. 公式に代入して計算する

先ほどの公式に、化学反応式の「係数（モル数）」をかけながら数値を代入します。

#### ① 生成物側の総和を計算する

生成物は $\text{CO}_2$ が 1 mol と $\text{H}_2\text{O}$ が 2 mol です。


$$\sum \Delta H^\circ_f(\text{生成物}) = (1 \times -394) + (2 \times -286)$$

$$= -394 - 572 = -966\text{ kJ}$$

#### ② 反応物側の総和を計算する

反応物は $\text{CH}_4$ が 1 mol と $\text{O}_2$ が 2 mol です。


$$\sum \Delta H^\circ_f(\text{反応物}) = (1 \times -75) + (2 \times 0)$$

$$= -75\text{ kJ}$$

#### ③ 引き算をする（生成物 ー 反応物）

$$\Delta H^\circ = (\text{生成物の総和}) - (\text{反応物の総和})$$

$$\Delta H^\circ = (-966) - (-75)$$

$$\Delta H^\circ = -966 + 75 = -891\text{ kJ/mol}$$

---

### 💡 答え

**反応エンタルピー：$\Delta H^\circ = -891\text{ kJ/mol}$**
（※メタン 1 mol が燃焼すると、 $891\text{ kJ}$ の熱を放出する発熱反応であることがわかります）

---

### まとめ：なぜこの引き算で解けるのか？

先ほどのエネルギー図を思い出してください。

* すべての物質の材料である「単体（一番上のてっぺん）」を基準（$0$）にします。
* 単体から**反応物**ができるまでの高さと、単体から**生成物**ができるまでの高さの「差」を計算しているのが、この引き算（生成物 ー 反応物）の本質です。

この「生成物 － 反応物」の公式を覚えておけば、どのような物質の燃焼や反応でもスムーズにエンタルピーを計算できるようになります！

## 5. 提出課題

### 【問題1】ヘスの法則を用いた計算
以下の熱化学反応式を利用して、一酸化炭素 $\text{CO}(g)$ の標準生成エンタルピー $\Delta H_f^\circ [\text{CO}(g)]$ を求めなさい。

1. $\text{C}(\text{graphite}) + \text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_1^\circ = -393.5 \text{ kJ}$
2. $\text{CO}(g) + \frac{1}{2}\text{O}_2(g) \rightarrow \text{CO}_2(g) \quad \Delta H_2^\circ = -283.0 \text{ kJ}$

---

### 【問題2】標準生成エンタルピーからの燃焼熱計算
プロパンガス $\text{C}_3\text{H}_8(g)$ の完全燃焼反応における標準反応エンタルピー $\Delta H_{\text{r}}^\circ$ を求めなさい。

$$\text{C}_3\text{H}_8(g) + 5\text{O}_2(g) \rightarrow 3\text{CO}_2(g) + 4\text{H}_2\text{O}(l)$$

ただし、各物質の標準生成エンタルピー $\Delta H_f^\circ$ は以下の通りとします：
- $\Delta H_f^\circ [\text{C}_3\text{H}_8(g)] = -103.8 \text{ kJ/mol}$
- $\Delta H_f^\circ [\text{CO}_2(g)] = -393.5 \text{ kJ/mol}$
- $\Delta H_f^\circ [\text{H}_2\text{O}(l)] = -285.8 \text{ kJ/mol}$

*(下の計算セルを使ってプログラムを実行し、数値を検証してみましょう)*

```python
# 【問題1・問題2の解答プログラム】

print("=== 問題1の解答 ===")
# 目的反応: C(graphite) + 1/2 O2(g) -> CO(g)
# 目的反応 = (1)式 - (2)式
dH1 = -393.5  # kJ
dH2 = -283.0  # kJ

dH_f_CO = dH1 - dH2
print(f"CO(g) の標準生成エンタルピー dH_f°[CO] = {dH_f_CO:.1f} kJ/mol")

print("\n=== 問題2の解答 ===")
# dH_rxn = [ 3*dH_f(CO2) + 4*dH_f(H2O) ] - [ 1*dH_f(C3H8) + 5*dH_f(O2) ]
dH_f_C3H8 = -103.8   # kJ/mol
dH_f_CO2  = -393.5   # kJ/mol
dH_f_H2O  = -285.8   # kJ/mol
dH_f_O2   = 0.0      # 単体なので 0 kJ/mol

products_sum = 3 * dH_f_CO2 + 4 * dH_f_H2O
reactants_sum = 1 * dH_f_C3H8 + 5 * dH_f_O2

dH_rxn = products_sum - reactants_sum
print(f"プロパンの燃焼エンタルピー dH_rxn° = {dH_rxn:.1f} kJ/mol")
```
