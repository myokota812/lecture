## 1. 凝縮相（Condensed Phases）とは？





気体、液体、固体の微視的（ミクロ）な比較：

| 状態 | 分子間の距離 | 分子の運動 | 形状と体積 | 主導的なエネルギー |
| :--- | :--- | :--- | :--- | :--- |
| **気体 (Gas)** | 非常に大きい | 自由で自由奔放 | 容器に依存 | 熱運動エネルギー（$k_B T$） |
| **液体 (Liquid)** | 近接（接触） | 滑り合うような運動 | 体積一定・形は可変 | 熱運動 $\approx$ 分子間力 |
| **固体 (Solid)** | 近接（固定） | 振動運動のみ | 体積・形状ともに一定 | 分子間力・結合力 |

- 液体と固体をあわせて**凝縮相（Condensed Phase）**と呼びます。
- 凝縮相の性質を決定づける主な要因は、分子同士を引きつける**「分子間力（Intermolecular Forces）」**です。

◇◇◇

<details>
<summary>説明</summary>

それでは、1つ目のテーマ『凝縮相（Condensed Phases）』とは何かについて解説します。

まず、私たちがよく知る物質の三態、つまり気体・液体・固体の特徴を、ミクロな視点から表で比較してみましょう。

1つ目の**気体**は、分子同士の距離が非常に大きく離れており、分子は自由に飛び回っています。形状や体積は入れる容器によって変わり、その状態を支配しているのは主に『熱運動エネルギー（$k_B T$）』です。

2つ目の**液体**になると、分子同士は隣り合うように接触します。位置関係を保ちつつ互いに滑り合うように運動するため、体積は一定ですが形は自由に変化します。このとき、熱運動と分子間力の影響力はほぼ拮抗した状態（$熱運動 \approx 分子間力$）になります。

3つ目の**固体**では、分子は互いにがっしりと固定されます。運動もその場での振動運動のみに限られ、体積・形状ともに一定となります。ここでは分子同士を引きつける『分子間力』や『結合力』が支配的です。


このように分子どうしが接近して存在する『液体』と『固体』を合わせて『凝縮相（Condensed Phase）』と呼びます。

そして、この凝縮相の性質を決定づける最も大きな要因が、分子同士を引きつける『分子間力（Intermolecular Forces）』です。

次のスライドからは、この分子間力についてさらに詳しく見ていきましょう。」

<details>
<summary>説明</summary>

それでは、『分子の双極子と双極子モーメント』について解説します。

まずスライド左側をご覧ください。塩化水素（$\text{HCl}$）を例に挙げています。
原子には、共有電子対を引き寄せる強さを示す『電気陰性度』という値があります。水素（$\text{H}$）は $2.1$、塩素（$\text{Cl}$）は $3.0$ であり、塩素の方が電子を引き寄せる力が強くなっています。

この電気陰性度の差によって電子雲が塩素側に偏り、水素側がわずかにプラス（$\delta+$）、塩素側がわずかにマイナス（$\delta-$）という『部分電荷』を持ちます。このように正と負の極が生じる現象を『双極子』と呼びます。
そして、この電荷の偏りの大きさと向きを表したベクトル量が『双極子モーメント』です。数式では $\mu = q \times d$、つまり『電荷量 $q$』と『距離 $d$』の積で表されます。

次にスライド右側をご覧ください。分子全体で見たときに双極子モーメントが存在するかどうかで、分子は『極性分子』と『無極性分子』に分けられます。

たとえば水分子（$\text{H}_2\text{O}$）の場合、それぞれの $\text{O-H}$ 結合に結合モーメント（$\mu_1, \mu_2$）が生じます。水分子は折れ線構造をしているため、これらのベクトルを合成すると相殺されず、上向きに『正味の双極子モーメント（$\mu_{\text{net}} \neq 0$）』が残ります。このように分子全体として電荷の偏りがあるものを『極性分子』と呼びます（$\text{H}_2\text{O}$ や $\text{HCl}$ など）。

一方で、二酸化炭素（$\text{CO}_2$）や四塩化炭素（$\text{CCl}_4$）を見てみましょう。それぞれの結合には極性がありますが、直線型や正四面体型といった対称な分子構造をしているため、結合モーメントのベクトル和が完全に打ち消し合ってゼロ（$\mu_{\text{net}} = 0$）になります。このように分子全体では偏りがなくなるものを『無極性分子』と呼びます。

結合の極性だけでなく、**『分子の立体構造（ベクトルの和）』によって分子全体の極性が決まる** という点をしっかり押さえておきましょう。

</details>

## 2. 分子間力の分類とエネルギーの比較

<figure style="text-align:center;">
<img
src="https://raw.githubusercontent.com/myokota812/lecture/master/ChemB/figure/intermoleforce.svg"
width="800">
<figcaption>
図1 主要な分子間力の概念図
</figcaption>
</figure>





### (1) 分散力（Dispersion force / London force）
- **対象：** すべての分子（無極性分子にも働く唯一の力）
- **起因：** 電子雲の瞬間的な偏りによる**誘発双極子（Induced dipole）**。
- **特徴：** 分子量（分子の電子数）が大きいほど、また分子の表面積が広いほど極性化（**分極率**）しやすくなり、分散力は強くなる。

### (2) 双極子-双極子相互作用（Dipole-Dipole forces）
- **対象：** 永久双極子を持つ極性分子
- **起因：** 分子内の正電荷（$\delta+$）と負電荷（$\delta-$）の静電的引力。

### (3) 双極子－誘起双極子相互作用（Dipole-Induced Dipole forces）
図は省略
- **対象：** 極性分子と無極性分子の間
- **起因：** 極性分子の永久双極子が近接する無極性分子の電子雲を歪ませ、誘起双極子を生じさせることで発生する。
  例： $\mathrm{H_2O-O_2}$、$\mathrm{HCl-Ar}$ など。
- **特徴：** 無極性分子の分極率が大きいほど強くなる。

### (4) 水素結合（Hydrogen Bonding）
- **対象：** $\text{F-H}$, $\text{O-H}$, $\text{N-H}$ 結合を持つ分子
- **起因：** 電気陰性度が非常に高い原子（$\text{F, O, N}$）に結合した $\text{H}$ 原子が持つ非常に強い正電荷と、隣接分子の孤立電子対との強固な相互作用。
- **強度：** $10 \sim 40 \text{ kJ/mol}$（他の分子間力より際立って強い）。
- **結果：** $\text{H}_2\text{O}, \text{NH}_3, \text{HF}$ の沸点が同族元素の水素化物の中で異常に高くなる原因。

◇◇◇

上記のうち、非水素結合系はまとめて、ファンデルワールス力とよんでいます。

一般的な分類は次のようになります。

$$
\begin{cases} \text{ファンデルワールス力} \begin{cases} \text{配向力（双極子－双極子相互作用）}\\ \text{誘起力（双極子－誘起双極子相互作用）}\\ \text{分散力（ロンドン分散力）} \end{cases}\\ \text{水素結合} \end{cases}$$




◇◇◇

<details>
<summary>説明</summary>

続いて、凝縮相を理解するうえで不可欠な『分子間力の分類とエネルギーの比較』について見ていきましょう。代表的な分子間力として、主に以下の3つが挙げられます。

1つ目は、分散力（ロンドン力）です。
これはすべての分子に働きます。特に、極性を持たない『無極性分子』同士に働く唯一の分子間力である点が重要です。
原因は、電子雲が瞬間的に揺らぐことで生じる『誘発双極子』です。
特徴として、分子量（つまり分子の中の電子の数）が大きいほど、あるいは分子の表面積が広いほど分極しやすくなり（分極率が高くなり）、分散力も強くなります。

2つ目は、**双極子-双極子相互作用**です。
こちらは、もともと電荷の偏りがある極性分子（永久双極子を持つ分子）の間に働きます。
分子内の正の電荷（$\delta+$）と、隣の分子の負の電荷（$\delta-$）が静電気的に引きつけ合うことで生じます。

そして3つ目が、**水素結合**です。
これは、フッ素（F）、酸素（O）、窒素（N）という電気陰性度が非常に高い原子に水素（H）が結合した分子で生じます。
F, O, N に引っ張られたH原子は強い正の電荷（$\delta+$）を帯び、これが隣接する分子の非共有電子対（孤立電子対）と強力に相互作用します。
その強度は $10 \sim 40 \text{ kJ/mol}$ と、他の分子間力に比べて際立って強いのが特徴です。
この強力な水素結合が存在するため、水（$\text{H}_2\text{O}$）、アンモニア（$\text{NH}_3$）、フッ化水素（$\text{HF}$）の沸点は、同じ族の水素化合物と比べて異常に高くなります。

以上の3つの力の仕組みと強さの違いをしっかり押さえておきましょう。

</details>

## 3. 蒸気圧と相平衡（Phase Equilibrium）

### 蒸発速度と凝縮速度
- **気化（Evaporation）：** 分子運動エネルギーが分子間力を振り切って気相へ飛び出す現象。
- **気液平衡：** 単位時間あたりの「蒸発速度」と「凝縮速度」が等しくなった動的平衡状態。
- **蒸気圧（Vapor Pressure）：** 気液平衡状態において気体が示す圧力。**温度が高くなるほど急激に高くなる**。

<details>
<summary>説明</summary>
ここでは、蒸発・凝縮・蒸気圧の関係を理解していきましょう。

まず、液体の分子は常に運動しています。その中には十分な運動エネルギーを持つ分子もあり、分子間力を振り切って液体表面から気体として飛び出します。この現象を**蒸発（気化）**といいます。

一方で、気体中を飛び回っている分子が再び液体表面に戻る現象を凝縮といいます。

左の図を見てください。

蒸発によって液体から分子が飛び出し、同時に気体分子の一部は液体へ戻っています。
密閉容器内では時間が経つと気体分子の数が増え、凝縮も起こりやすくなります。
そしてある時点で、

『単位時間あたりに蒸発する分子数』と『単位時間あたりに凝縮する分子数』

が等しくなります。
この状態を気液平衡といいます。
見た目には変化していないように見えますが、実際には分子が絶えず行き来しているため、これは動的平衡です。


次に中央の図です。

気液平衡状態で気体分子が容器の壁や液面に衝突すると圧力を生じます。

このとき、液体と平衡状態にある蒸気が示す圧力を

蒸気圧（vapor pressure）

と呼びます。

蒸気圧は物質によって異なり、また温度によって大きく変化します。

右のグラフを見てください。

温度が上昇すると分子運動が激しくなり、蒸発しやすくなります。

その結果、蒸気圧は温度とともに急激に増加します。

例えば水では、100 ℃付近で蒸気圧が約101 kPaになります。

これは大気圧とほぼ同じ値です。

ここで重要なのが沸騰との関係です。

液体の蒸気圧が外部の圧力、つまり大気圧と等しくなると液体内部でも気泡が成長できるようになります。

この状態が沸騰です。

したがって、

『蒸気圧 = 外圧』となる温度が沸点

ということになります。

まとめると、

液体では蒸発と凝縮が同時に起こっている。 ↓
 蒸発速度と凝縮速度が等しくなると気液平衡になる。 ↓
 そのときの蒸気の圧力が蒸気圧。 ↓
 温度が高いほど蒸気圧は大きくなる。 ↓
 蒸気圧が外圧に達すると沸騰する。

</details>

### クラウジウス・クラペロンの式（Clausius-Clapeyron Equation）
蒸気圧 $P$ の温度 $T$ 依存性を表す重要な式：

$$\ln P = -\frac{\Delta H_{\text{vap}}}{R T} + C$$

または2つの温度条件 ($T_1, P_1$) および ($T_2, P_2$) において：

$$\ln \left( \frac{P_2}{P_1} \right) = -\frac{\Delta H_{\text{vap}}}{R} \left( \frac{1}{T_2} - \frac{1}{T_1} \right)$$

- $\Delta H_{\text{vap}}$: 蒸発エンタルピー（蒸発熱） $[\text{J/mol}]$
- $R$: 気体定数 $[8.314 \text{ J/(K}\cdot\text{mol)}]$
- $T$: 絶対温度 $[\text{K}]$

◇◇◇

<details>
<summary>説明</summary>
ここでは、蒸気圧の温度依存性を定量的に表す式であるクラウジウス・クラペイロンの式を学びます。
前のスライドで見たように、温度が高くなるほど液体分子の運動は激しくなり、蒸発しやすくなります。
その結果、蒸気圧は急激に増加します。
では、その増加のしかたを数式で表すとどうなるでしょうか。
それを与えるのがこの式です。

$$
\ln P= -\frac{\Delta H_{\rm vap}}{RT}+C
$$

$P$ :蒸気圧
$T$ :絶対温度
$R$ :気体定数
$\Delta H_{\rm vap}$ :蒸発エンタルピー

蒸発エンタルピーとは、液体1 molを気体にするのに必要なエネルギーです。つまり分子間力が強い液体ほど $\Delta H_{\rm vap}$ は大きくなります。

この式を見ると、
$\ln P$
は
$\frac1T$
に対して直線関係であることがわかります。
つまり横軸に $1/T$、縦軸に $\ln P$ を取ると直線になります。

その傾きは
$-\frac{\Delta H_{\rm vap}}{R}$
です。

したがって実験で蒸気圧を測定すれば、蒸発エンタルピーを求めることができます。

また実際には定数 $C$ を求めるよりも、2つの温度条件を比較する形をよく使います。

$$
\ln \left( \frac{P_2}{P_1} \right) = -\frac{\Delta H_{\rm vap}}{R} \left( \frac1{T_2} -\frac1{T_1} \right)
$$

この式を使うと、ある温度での蒸気圧が分かっていれば、別の温度での蒸気圧を計算できます。

重要なポイントは、*** 蒸発熱が大きいつまり分子間力が強いほど蒸気圧は低くなり***、

逆に

蒸発熱が小さい
分子間力が弱い
ほど蒸気圧は高くなる

ということです。


</details>

おまけ：式の導出

「クラペイロンの式」と「クラウジウス・クラペイロンの式」は実は少し違います。

あなたのスライドの式は、

ln⁡P=−ΔHvapRT+C\ln P=-\frac{\Delta H_{\rm vap}}{RT}+C

なので、これはクラウジウス・クラペイロン式です。

その元になるのがクラペイロン式です。

1. クラペイロン式はどこから来るのか

相平衡（気液平衡や固液平衡）では

Gliquid=GvaporG_{\rm liquid}=G_{\rm vapor}

が成立しています。

ここでギブズエネルギーの全微分

dG=VdP−SdTdG=VdP-SdT

を用います。

平衡状態が少し変化しても

dGliquid=dGvapordG_{\rm liquid}=dG_{\rm vapor}

なので

VldP−SldT=VgdP−SgdTV_l dP-S_l dT = V_g dP-S_g dT

となります。

整理すると

(Vg−Vl)dP=(Sg−Sl)dT(V_g-V_l)dP = (S_g-S_l)dT

です。

したがって

dPdT=ΔSΔV\frac{dP}{dT} = \frac{\Delta S}{\Delta V}

を得ます。

さらに、

ΔH=TΔS\Delta H=T\Delta S

であることから

ΔS=ΔHT\Delta S = \frac{\Delta H}{T}

です。

代入すると

dPdT=ΔHTΔV\boxed{ \frac{dP}{dT} = \frac{\Delta H} {T\Delta V} }

これがクラペイロンの式です。

2. 蒸発の場合

気液平衡では

ΔH=ΔHvap\Delta H = \Delta H_{\rm vap} ΔV=Vg−Vl\Delta V = V_g-V_l

です。

すると

dPdT=ΔHvapT(Vg−Vl)\frac{dP}{dT} = \frac{\Delta H_{\rm vap}} {T(V_g-V_l)}

となります。

3. クラウジウスの近似

蒸気の体積は液体の体積より圧倒的に大きいので

Vg≫VlV_g \gg V_l

として

Vg−Vl≃VgV_g-V_l \simeq V_g

と近似します。

さらに蒸気を理想気体とみなすと

PVg=RTPV_g=RT

より

Vg=RTPV_g=\frac{RT}{P}

です。

これをクラペイロン式に代入すると

dPdT=ΔHvapT(RT/P)\frac{dP}{dT} = \frac{\Delta H_{\rm vap}} {T(RT/P)} =ΔHvapPRT2= \frac{\Delta H_{\rm vap}P} {RT^2}

となります。

両辺を PP で割ると

1PdPdT=ΔHvapRT2\frac1P\frac{dP}{dT} = \frac{\Delta H_{\rm vap}} {RT^2}

左辺は

d(ln⁡P)dT\frac{d(\ln P)}{dT}

なので

d(ln⁡P)=ΔHvapRT2dTd(\ln P) = \frac{\Delta H_{\rm vap}} {RT^2}dT

です。

積分すると

ln⁡P=−ΔHvapRT+C\ln P = -\frac{\Delta H_{\rm vap}}{RT} +C
ln⁡P=−ΔHvapRT+C\boxed{ \ln P = -\frac{\Delta H_{\rm vap}}{RT} +C }

これがスライドに載っているクラウジウス・クラペイロン式です。

授業では最後に

「蒸発熱が大きい水は蒸気圧が低い。
 蒸発熱が小さいジエチルエーテルは蒸気圧が高い。
 この違いが沸点の違いにつながる。」

とまとめると、蒸気圧 → 沸点 → 分子間力の流れが綺麗につながります。

```python
# クラウジウス・クラペロンの式に基づいた水の蒸気圧曲線のプロット
R = 8.314        # J/(mol K)
dH_vap = 40.7e3  # 水の蒸発熱 約 40.7 kJ/mol
T_boil = 373.15  # 100 °C = 373.15 K (このとき 1 atm = 101.325 kPa)
P0 = 101.325     # kPa

T_degC = np.linspace(0, 120, 200)
T_K = T_degC + 273.15

# ln(P / P0) = - dH_vap / R * (1/T - 1/T_boil)
P = P0 * np.exp(- (dH_vap / R) * ((1 / T_K) - (1 / T_boil)))

plt.figure(figsize=(7, 4))
plt.plot(T_degC, P, color='red', linewidth=2, label='Water Vapor Pressure')
plt.axhline(101.325, color='black', linestyle='--', label='1 atm (101.325 kPa)')
plt.axvline(100, color='gray', linestyle=':')

plt.title('Vapor Pressure Curve of Water')
plt.xlabel('Temperature (°C)')
plt.ylabel('Vapor Pressure (kPa)')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

## 4. 状態図（Phase Diagram）の解釈

圧力 $P$ 対 温度 $T$ の二次元平面上に、気相・液相・固相の安定領域を示した図。

### 状態図の重要キーワード
1. **蒸気圧曲線（気液境界）：** 沸点の圧力依存性を示す。
2. **融解曲線（固液境界）：** 融点の圧力依存性を示す。
   - **水（$\text{H}_2\text{O}$）の特異性：** 水素結合による粗い結晶構造（氷）のため、氷の密度は水より小さい。したがって融解曲線の傾きが**負（左上がり）**になる（超高圧をかけると氷が溶ける）。一般的な物質は正の傾き。
3. **昇華曲線（固気境界）：** 固体が直接気化する境目。
4. **三重点（Triple Point）：** 気・液・固の3相が同時に共存・平衡にある唯一の温度・圧力点。
5. **臨界点（Critical Point）：** 気液境界の終端。これを超える高圧・高温では超臨界流体（Supercritical fluid）となり、気体と液体の区別がなくなる。

### 【問題1】分子間力の強さと沸点
次の物質群を、沸点が高い順に並べ替え、その理由を分子間力の観点から説明しなさい。
$$\text{CH}_4, \quad \text{CH}_3\text{OH}, \quad \text{CH}_3\text{CH}_2\text{OH}$$

### 【問題2】クラウジウス・クラペロンの式の応用計算
ある液体 $A$ の標準沸点（$1.00 \text{ atm} = 101.3 \text{ kPa}$ における沸点）は $80.0^\circ\text{C}$ ($353.15 \text{ K}$) です。この液体の蒸発エンタルピー $\Delta H_{\text{vap}}$ が $32.0 \text{ kJ/mol}$ であるとき、$25.0^\circ\text{C}$ ($298.15 \text{ K}$) における蒸気圧 $P_1 \text{ (kPa)}$ を求めなさい。（気体定数 $R = 8.314 \text{ J/(K}\cdot\text{mol)}$ とする）

*(下の計算セルを使ってプログラムで数値解を求め、確認してみましょう)*

```python
# 【問題2の解答プログラム】
import numpy as np

# 与えられた定数
R = 8.314             # J/(mol K)
dH_vap = 32.0 * 1000  # kJ/mol -> J/mol に変換

T2 = 80.0 + 273.15    # K (沸点)
P2 = 101.3            # kPa

T1 = 25.0 + 273.15    # K

# Clausius-Clapeyron 式:
# ln(P2 / P1) = - (dH_vap / R) * (1/T2 - 1/T1)
# -> ln(P1 / P2) = (dH_vap / R) * (1/T2 - 1/T1)
# -> P1 = P2 * exp( (dH_vap / R) * (1/T2 - 1/T1) )

P1 = P2 * np.exp((dH_vap / R) * ((1 / T2) - (1 / T1)))

print(f"25°C における蒸気圧 P1 = {P1:.2f} kPa")
```

### 【問題1 解説】
**順序：** $\text{CH}_3\text{CH}_2\text{OH} > \text{CH}_3\text{OH} > \text{CH}_4$

- **理由：**
  - $\text{CH}_4$（メタン）は無極性分子であり、分子間に働くのは弱い**分散力**のみであるため、沸点は最も低い（$-161.5^\circ\text{C}$）。
  - $\text{CH}_3\text{OH}$（メタノール）と $\text{CH}_3\text{CH}_2\text{OH}$（エタノール）は、ともに $-\text{OH}$ 基を持つため強い**水素結合**を形成し、メタンより著しく沸点が高い。
  - エタノールはメタノールよりもアルキル鎖が長く分子量・表面積が大きいため、**分散力も加算されて**より強い分子間力が働き、沸点が最も高くなる（エタノール：$78.3^\circ\text{C}$、メタノール：$64.7^\circ\text{C}$）。
