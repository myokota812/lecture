### 問4

実験で得られた蒸気圧データ

|--------------|------------------|
| 300 | 8.0 |
| 320 | 18.0 |
| 340 | 36.5 |
| 360 | 68.0 |

このデータを用いて Clausius–Clapeyron プロット

$$
\ln P
\quad \text{vs.} \quad
\frac{1}{T}
$$

を作成し，蒸発エンタルピーを求めなさい。

# 解答1

$$\ln(P_2/P_1)=-\frac{\Delta H}{R}(1/T_2-1/T_1)$$

計算より $$P_2 \approx 99\,kPa$$

# 解答3

$$T_b \approx347\,K \approx74^\circ C$$

```python
import numpy as np
import matplotlib.pyplot as plt
T=np.array([300,320,340,360])
P=np.array([8.0,18.0,36.5,68.0])
x=1/T
y=np.log(P)
coef=np.polyfit(x,y,1)
print(-coef[0]*8.314/1000,'kJ/mol')
```
