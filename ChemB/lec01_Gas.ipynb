> 気体分子が容器の壁に衝突すると、分子は壁を押し、壁からは反作用を受けて跳ね返されます。このとき、個々の分子が壁に与える力はごくわずかですが、**天文学的な数の分子が、一秒間に何度も何度も壁に衝突し続ける**ことになります。

> 
> 
> つまり、私たちがマクロ（巨視的）な視点で観測している『気体の圧力』とは、実は**『無数の気体分子が容器の壁に連綿とぶつかり続けることで生じる、力（力積）の単位面積あたりの平均値』**にほかなりません。
> 温度が高くなると分子の運動速度が上がり、壁にぶつかる勢いや回数が増えるため、圧力が上がる——というボイル・シャルルの法則のイメージも、この分子運動論モデルからすんなり理解できるかと思います。

- 単位面積あたりに加わる力： $P = \frac{F}{A}$
- **SI単位：** パスカル（$\text{Pa} = \text{N/m}^2$）

### 気圧の各種単位の換算
標準大気圧（$1\text{ atm}$）の定義：
$$1\text{ atm} = 101,325\text{ Pa} = 101.325\text{ kPa} = 1.01325\text{ bar}$$
$$1\text{ atm} = 760\text{ mmHg} = 760\text{ Torr}$$

◇◇◇
<details>
<summary>説明文</summary>

それでは、圧力の定義と各種単位の換算について確認していきましょう。   まず、圧力 $P$（Pressure）は、単位面積あたりに加わる垂直な力 $F$ として $P = \frac{F}{A}$ と定義されます。SI単位系ではパスカル（$\text{Pa}$）が使われ、$1\text{ Pa}$ は $1\text{ N/m}^2$ と等しくなります。   続いて、気圧でよく使われる各種単位の換算についてです。
基準となる標準大気圧、$1\text{ atm}$（エイ・ティー・エム）の定義を見てみましょう。   $1\text{ atm}$ は、パスカルで表すと $101,325\text{ Pa}$、キロパスカルにすると $101.325\text{ kPa}$ です。また、気象学などで使われるバール（$\text{bar}$）を用いると $1.01325\text{ bar}$ となります。   さらに、トリチェリの水銀柱の実験に由来する単位として、$1\text{ atm}$ は水銀柱 $760\text{ mm}$ 分に相当するため、$760\text{ mmHg}$、あるいは $760\text{ Torr}$（トル）とも表されます。   計算問題などで単位を変換する場面がよく出てきますので、この相互関係をしっかりと頭に入れておいてください。

## 2. 理想気体の法則（気体の状態方程式）

実験的法則の統合：
1. **ボイルの法則:** $V \propto \frac{1}{P}$ （$T, n$ が一定）
2. **シャルルの法則:** $V \propto T$ （$P, n$ が一定、温度 $T$ は絶対温度 $\text{K}$）
3. **アボガドロの法則:** $V \propto n$ （$T, P$ が一定）

◇◇◇
<details>
<summary>説明文</summary>

高校の復習になります。これまでに発見されてきた気体に関する3つの実験的法則をひとつに統合していきます。

まず1つ目はボイルの法則です。温度 $T$ と物質量 $n$ が一定のとき、体積 $V$ は圧力 $P$ に反比例します。

2つ目はシャルルの法則です。圧力 $P$ と物質量 $n$ が一定のとき、体積 $V$ は絶対温度 $T$（単位はケルビン K）に比例します。

そして3つ目がアボガドロの法則です。温度 $T$ と圧力 $P$ が一定のとき、体積 $V$ は気体の物質量 $n$ に比例します。
</details>

---



### 理想気体の状態方程式（Ideal Gas Law）
$$PV = nRT$$

- $P$: 圧力 $(\text{Pa})$
- $V$: 体積 $(\text{m}^3)$
- $n$: 物質量 $(\text{mol})$
- $R$: 気体定数 $= 8.314\text{ J/(K}\cdot\text{mol)} = 0.08206\text{ L}\cdot\text{atm/(K}\cdot\text{mol)}$
- $T$: 絶対温度 $(\text{K})$

◇◇◇

<details>
<summary>説明</summary>

それでは、気体の挙動を記述する最も基本となる **『理想気体の状態方程式』** を確認します。

式は **$PV = nRT$** ですね。
計算問題で最もミスが起きやすいのは『単位の組み合わせ』ですので、ここを重点的に見ていきます。

* 圧力 $P$ の単位はパスカル


* 体積 $V$ の単位は立方メートル


* 物質量 $n$ の単位はモル


* 温度 $T$ の単位は、セルシウス度（℃）ではなく絶対温度のケルビン（$\mathrm{K}$）を使用します。



そして、比例定数である $R$ は**気体定数**と呼ばれます。
$\mathrm{SI}$標準単位系（$\mathrm{Pa}, \mathrm{m^3}$）を使う場合は **$8.314$** を用いますが、化学などで「リットル（$\mathrm{L}$）」や「気圧（$\mathrm{atm}$）」を使って計算する場合は **$0.08206$** を使います。

扱う問題の単位設定に応じて、気体定数 $R$ の値を正しく使い分けるよう意識してください。

</details>

---

```python
# PythonによるP-V等温線の可視化（ボイル・シャルルの法則の確認）
import numpy as np
import matplotlib.pyplot as plt

R = 8.314  # 気体定数 J/(K mol)
n = 1.0    # 1 mol
V = np.linspace(0.005, 0.05, 100) # 体積 m^3

temperatures = [200, 300, 400] # 各絶対温度 (K)

plt.figure(figsize=(7, 4))
for T in temperatures:
    P = (n * R * T) / V
    plt.plot(V * 1000, P / 1000, label=f'T = {T} K') # L と kPa に変換

plt.title('P-V Isotherms for Ideal Gas (n = 1 mol)')
plt.xlabel('Volume (L)')
plt.ylabel('Pressure (kPa)')
plt.grid(True)
plt.legend()
plt.show()
```

## 3. 混合気体とダルトンの分圧の法則

全圧 $P_{\text{total}}$ は、各成分気体が単独で存在する仮定のときの圧力（分圧 $P_i$）の和に等しい。

$$P_{\text{total}} = P_A + P_B + P_C + \dots = \sum P_i$$

### モル分率（Mole Fraction）
成分 $A$ のモル分率 $x_A$：
$$x_A = \frac{n_A}{n_{\text{total}}}$$

分圧と全圧の関係：
$$P_A = x_A \times P_{\text{total}}$$

◇◇◇

<details>
<summary>説明</summary>
それでは、気体が混ざった状態、つまり**混合気体**の振る舞いについて見ていきましょう。

ここで登場するのが『ダルトンの分圧の法則』です。
イメージとしては、容器の中に気体 $A, B, C$ が混ざっているとき、容器全体の圧力（全圧 $P_{\text{total}}$）は、それぞれの気体が壁を押す力（分圧 $P_i$）を足し合わせたものになる、という法則です。

では、特定の気体 $A$ だけがどれくらいの圧力を担っているか（分圧 $P_A$）を知りたいときはどうすればよいでしょうか。

そこで使うのがスライド中央にある『モル分率』です。
これは、気体全体の分子の数のうち、成分 $A$ の分子が占める『割合』のことです。

この割合がわかれば、一番下の式にあるように、**『全圧にモル分率（存在割合）を掛けるだけで、成分 $A$ の分圧が求められる』** ということになります。

例えば、空気に占める窒素の割合が約80%なら、窒素の分圧は全圧の約80%になる、ということですね。
</details>

## 4. ダルトンの分圧の法則とアマガの部分体積の法則

width="800">
<figcaption>
混合気体の全圧と分圧

### 1. $P_A V = n_A RT, \quad P_B V = n_B RT$

**（同じ「全体積 $V$」の中に、それぞれの気体が同居していると考える）**

* **イメージ:** 容積 $V$ の**頑丈な密閉容器**に、気体Aと気体Bが混ざって入っている状態です。
* **特徴:**
* 気体Aも気体Bも、容器全体（体積 $V$）を自由に変形せず飛び回っています。
* したがって、**体積 $V$ と温度 $T$ は共通**です。
* このとき、気体A単独が容器の壁を押す圧力が **分圧 $P_A$**、気体B単独が押す圧力が **分圧 $P_B$** となります。
* 全圧 $P_{\text{total}}$ はそれぞれの分圧の和になります（**ダルトンの分圧の法則**: $P_{\text{total}} = P_A + P_B$）。



> **日常の例え:** 部屋（体積 $V$）の中に大人（$n_A$）と子供（$n_B$）が同居している状態。壁を叩く力の合計が全圧です。

---

### 2. $P_{\text{total}} V_A = n_A RT, \quad P_{\text{total}} V_B = n_B RT$

**（同じ「全圧 $P_{\text{total}}$」のもとで、それぞれの気体が領域を分け合っていると考える）**

* **イメージ:** 大気圧やピストンなどで**全体圧力を $P_{\text{total}}$ に保ったまま**、気体Aと気体Bを仕切りで別々に分けた（または集めた）状態です。
* **特徴:**
* **圧力 $P_{\text{total}}$ と温度 $T$ は共通**です。
* 気体Aだけを集めて圧力 $P_{\text{total}}$ にしたときの体積が **$V_A$**、気体Bだけを集めて圧力 $P_{\text{total}}$ にしたときの体積が **$V_B$** となります（これを「分圧体積」や「成分体積」と呼びます）。
* 全体積 $V$ はそれぞれの部分体積の和になります（**アマガの部分体積の法則**: $V = V_A + V_B$）。



> **日常の例え:** 全体で1気圧に保たれた風船の中で、気体Aが占めるスペース（$V_A$）と気体Bが占めるスペース（$V_B$）を足すと、風船全体の大きさ（$V$）になるという考え方です。

---

### まとめ（2つの違いの比較）

| 項目 | $P_A V = n_A RT$ の組 | $P_{\text{total}} V_A = n_A RT$ の組 |
| --- | --- | --- |
| **注目する法則** | **ダルトンの分圧の法則** | **アマガの部分体積の法則** |
| **共通（固定）のもの** | 体積 $V$, 温度 $T$ | 圧力 $P_{\text{total}}$, 温度 $T$ |
| **足し算の形** | $P_{\text{total}} = P_A + P_B$ （全圧＝分圧の和） | $V = V_A + V_B$ （全体積＝分体積の和） |
| **モル比との関係** | $\frac{P_A}{P_{\text{total}}} = \frac{n_A}{n_{\text{total}}}$ | $\frac{V_A}{V} = \frac{n_A}{n_{\text{total}}}$ |



両者とも、割り算をすると **「モル比 ＝ 圧力比 ＝ 体積比」** になるという本質は同じです。

<h3>ダルトンの法則とアマガの法則を使う問題例</h3>

---

> **【例題１】**  
> $1\text{ atm} \ (1.01325 \times 10^5\text{ Pa})$ の大気中における $\text{N}_2$ と $\text{O}_2$ の分圧を求めよ。
> 但し、空気中の主な成分とその体積百分率は次の通りとする。
> $\text{N}_2: 78.08\%$, $\text{O}_2: 20.95\%$, $\text{Ar}: 0.93\%$, $\text{Others}: 0.04\%$

atm単位で求めるのであれば、即答できるはずです。解き終えたら、解答例で確認してください。

◇◇◇

<details>
<summary>解答例</summary>

アマガの法則および状態方程式により、理想気体の混合物において「体積百分率（体積比）」は「モル分率（モル比）」および「分圧比」と等しくなります。

したがって、各気体の分圧は **全圧 × 体積比** で計算できます。

この問題の場合は全圧が1 atmなので、体積分率がそのまま解答になります。

#### 1. 窒素 $\text{N}_2$ の分圧 $P_{\text{N}_2}$

$$\begin{align*} P_{\text{N}_2} &= 1.01325 \times 10^5\text{ Pa} \times \frac{78.08}{100} \\ &= 0.79114559... \times 10^5\text{ Pa} \\ &\approx \mathbf{7.911 \times 10^4\text{ Pa}} \quad (\text{または } \mathbf{0.7808\text{ atm}}) \end{align*}$$

#### 2. 酸素 $\text{O}_2$ の分圧 $P_{\text{O}_2}$

$$\begin{align*} P_{\text{O}_2} &= 1.01325 \times 10^5\text{ Pa} \times \frac{20.95}{100} \\ &= 0.212275875 \times 10^5\text{ Pa} \\ &\approx \mathbf{2.123 \times 10^4\text{ Pa}} \quad (\text{または } \mathbf{0.2095\text{ atm}}) \end{align*}$$

</details>



---
> **【例題2】**  
> 容積 $3.0\text{ L}$ の容器 A に $2.0 \times 10^5\text{ Pa}$ の窒素 $\text{N}_2$ が入っており、容積 $2.0\text{ L}$ の容器 B に $3.0 \times 10^5\text{ Pa}$ の酸素 $\text{O}_2$ が入っている。コックを開けて両方の容器をつなぎ、温度を一定に保って十分時間を置いた。
> 1. 混合後の窒素 $\text{N}_2$ と酸素 $\text{O}_2$ の分圧をそれぞれ求めよ。
> 2. 混合気体の全圧を求めよ。
> 
> 

◇◇◇

<details>
<summary> 解答例</summary>

コックを開けた後の全体積は $V = 3.0\text{ L} + 2.0\text{ L} = 5.0\text{ L}$ となります。
ボイルの法則 $P_1 V_1 = P_2 V_2$ （温度・物質量が一定のとき、圧力と体積は反比例する）を用いて、各気体が $5.0\text{ L}$ 全体に広がったときの分圧を計算します。

#### 1. 各気体の分圧

* **窒素 $\text{N}_2$ の分圧 $P_{\text{N}_2}$:**

$$P_{\text{N}_2} \times 5.0\text{ L} = 2.0 \times 10^5\text{ Pa} \times 3.0\text{ L}$$

$$P_{\text{N}_2} = \frac{2.0 \times 10^5 \times 3.0}{5.0} = \mathbf{1.2 \times 10^5\text{ Pa}}$$

* **酸素 $\text{O}_2$ の分圧 $P_{\text{O}_2}$:**

$$P_{\text{O}_2} \times 5.0\text{ L} = 3.0 \times 10^5\text{ Pa} \times 2.0\text{ L}$$

$$P_{\text{O}_2} = \frac{3.0 \times 10^5 \times 2.0}{5.0} = \mathbf{1.2 \times 10^5\text{ Pa}}$$

#### 2. 混合気体の全圧 $P_{\text{total}}$

ダルトンの分圧の法則（$P_{\text{total}} = P_{\text{N}_2} + P_{\text{O}_2}$）より、各分圧の和を求めます。

$$\begin{align*} P_{\text{total}} &= 1.2 \times 10^5\text{ Pa} + 1.2 \times 10^5\text{ Pa} \\ &= \mathbf{2.4 \times 10^5\text{ Pa}} \end{align*}$$

</details>

## 5. 実在気体とファンデルワールスの状態方程式

**理想気体からのズレ（高圧・低温条件で顕著）：**
- 分子自身の体積が存在する → 有効体積の減少（$-nb$）

<details>
<summary>説明</summary>
さて、ここからは『理想気体』から『実在気体』へと一歩足を踏み入れましょう。
理想気体では『気体分子自身の大きさはゼロ』と仮定していましたね。しかし、現実に存在する分子には、当然ながら決まった大きさ（体積）が存在します。スライドの図をご覧ください。左側に並んでいる緑色の粒々が気体分子です。分子自身に大きさがあるということは、他の分子がその場所に侵入できない領域ができるということです。これを『排除体積（Excluded Volume）』と呼びます。

分子がモル数 $n$ だけあれば、全分子が占める体積の合計は $nb$ と表せます。

では、分子が実際に自由に動き回れる空間はどれくらいでしょうか？容器全体の体積を $V$ とすると、分子自身が占めてしまっている排除体積 $nb$ の分だけ、自由に動き回れるスペースが狭くなりますよね。これが、図の右側に示されている緑色の背景部分、

『有効体積（Free Volume）』です。式で表すと $(V - nb)$ となります。

つまり、状態方程式の体積 $V$ の部分を、実在気体ではこの『有効体積 $(V - nb)$』に置き換えてあげる必要があるわけです。
</details>

---



- 分子間力が働く → 容器壁を押す圧力の減少（$-\frac{an^2}{V^2}$）

<figure style="text-align:center;">
<img
src="https://raw.githubusercontent.com/myokota812/lecture/master/ChemB/figure/realgas_P.svg"
width="800">
<figcaption>
</figure>

◇◇◇

<details>
<summary>説明</summary>

続いて、実在気体のもう一つの大きな特徴である『分子間力（引力）』による影響を見ていきましょう。

気体の『圧力』というのは、分子が容器の壁にドカンと衝突したときの衝撃の強さです。

まず、左側の**理想気体**をご覧ください。
理想気体では『分子間力はゼロ』と仮定していますので、壁に向かって飛んでいく青い分子は、周囲の分子から何の邪魔も受けません。そのため、ノーブレーキの最大スピードで壁にドカンと強く衝突します。これが理想的な圧力 $P_{\text{ideal}}$ です。

一方で、右側の**実在気体**はどうでしょうか？
実在気体では、分子同士の間に互いに引き合う『分子間力（引力）』が働いています。
壁に衝突しようとする青い分子に注目すると、まさに壁にぶつかる直前、**後ろにいる緑色の分子たちから後ろ向きにグッと引き戻される力**を受けます。

いわば、**衝突直前にブレーキがかけられるような状態**になるわけですね。
その結果、衝突の勢いが弱まり、実際に測定される圧力 $P_{\text{real}}$ は、理想的な圧力 $P_{\text{ideal}}$ よりも小さくなってしまいます（$P_{\text{real}} < P_{\text{ideal}}$）。

ここで一番大切なポイントです！

実際に観測される圧力 $P_{\text{real}}$ はブレーキのせいで目減りしているので、理想気体の状態方程式（$P V = nRT$）にそのまま当てはめることができません。
本来の『理想的な圧力』に戻してあげるためには、引力によって減ってしまった圧力ぶん、つまり **$a \left(\frac{n}{V}\right)^2$ を足してあげる補正**が必要になります。

したがって、圧力の項は **$\left(P + a\frac{n^2}{V^2}\right)$** と補正されるわけです。」
</details>

---





### ファンデルワールスの状態方程式（van der Waals Equation）
$$\left( P + a \frac{n^2}{V^2} \right)(V - nb) = nRT$$

- $a$: 分子間力に関する補正係数
- $b$: 分子自身の体積に関する補正係数

◇◇◇


<details>
<summary>説明</summary>
それではここまでに学んだ『体積の補正』**と**『圧力の補正』の2つを合体させてみましょう！

これが、実在気体の振る舞いを精度よく表す代表的な式、『ファンデルワールスの状態方程式』です。

式を左から順番に確認してみましょう。

まず、1つ目のカッコに入っている **$\left( P + a\frac{n^2}{V^2} \right)$** です。
覚えていますね？分子同士の引き合い（分子間力）によって壁を押す圧力が弱まってしまうため、目減りした分をプラスしてあげた『圧力の補正項』です。
ここで使われている **$a$** は、**分子間力の強さに関する補正係数**です。引き合う力が強い気体ほど、この $a$ の値が大きくなります。

次に、2つ目のカッコの **$(V - nb)$** です。
こちらは分子自身が占有する体積（排除体積）を引き算して、分子が自由に飛び回れるスペースを表した『体積の補正項』でしたね。
ここに出てくる **$b$** は、**分子自身の大きさ（体積）に関する補正係数**です。分子のサイズが大きい気体ほど、$b$ の値も大きくなります。

そして、これらをかけ合わせたものが右辺の **$nRT$** と等しくなります。

理想気体の状態方程式 $PV = nRT$ の **$P$ に圧力補正**を、**$V$ に体積補正**を組み込むことで、私たちが生きている現実世界の実在気体の性質をみごとに数式化できた、というわけです。

ちなみに、定数 $a$ や $b$ の値は気体の種類（酸素、窒素、二酸化炭素など）によってそれぞれ異なりますので、問題で与えられた数値を当てはめて計算していくことになります。」
</details>

---


<h3>理想気体の状態方程式とファンデルワールス状態方程式の比較</h3>
比較の前に、与えられたファンデルワールスの状態方程式：

$$\left( P + a \frac{n^2}{V^2} \right)(V - nb) = nRT$$

この式の両辺を $n$（モル数）で割る（または項を変形する）ことで、$V_m = \frac{V}{n}$ （モル体積）を用いた形式に書き換えます。

◇◇◇

<details><summary> 変形のステップ</summary>

1. **第1項の括弧内を変形:**

$$a \frac{n^2}{V^2} = a \left(\frac{n}{V}\right)^2 = \frac{a}{\left(\frac{V}{n}\right)^2} = \frac{a}{V_m^2}$$

2. **第2項の括弧から $n$ を括り出す:**

$$(V - nb) = n \left(\frac{V}{n} - b\right) = n (V_m - b)$$

3. **元の式に代入:**

$$\left( P + \frac{a}{V_m^2} \right) \cdot n (V_m - b) = nRT$$

4. **両辺を $n$ で割る:**

$$\left( P + \frac{a}{V_m^2} \right)(V_m - b) = RT$$
</details>

◇◇◇

### 最終形

$$\left( P + \frac{a}{V_m^2} \right)(V_m - b) = RT$$

この形にすることで、物質量 $n$ に依存しない「モルあたりの状態方程式」として扱うことができます。
$V_m$が大きいほど、分子間距離が長くなる、すなわち相互作用が働かない、と見ることができます。

ちなみに、理想気体の状態方程式は

$$
P V_m =RT
$$

---

上述の通り、モル体積 $V_m$ が小さくなると分子間距離が縮まり、分子間力（パラメータ $a$）や分子自体の大きさ（パラメータ $b$）といった実在気体の特性が無視できなくなります。この相互作用の現れ方は、気体の種類（分子のサイズや極性）によって大きく異なります。ここでは、分子間力が非常に小さく理想気体に近い性質を持つ**水素（$\text{H}_2$）と、比較的分化間力が大きく理想気体からのズレが顕著に現れる二酸化炭素（$\text{CO}_2$）**を例に取り、それぞれの $P-V_m$ 等温線を下図のように出力しました。

```python
#@title 理想状態方程式とファンデルワールス状態方程式の比較：水素分子
import matplotlib.pyplot as plt
import numpy as np

# ファンデルワールスの状態方程式のパラメータ (H2) (SI単位)
a = 2.476e-2  # Pa·m^6/mol^2 (CO2の約1/15と非常に小さい)
b = 2.661e-5  # m^3/mol     (CO2より小さい)
R = 8.314  # J/(mol·K)
T = 273.15  # 温度 (K)

# Vmの範囲を定義 (m³/mol)
Vm_min = 0.05e-3  # 0.05 L/mol
Vm_max = 1.0e-3  # 1.0 L/mol
num_points = 500
Vm_values = np.linspace(Vm_min, Vm_max, num_points)

# 状態方程式による圧力計算
P_values_vdw = (R * T) / (Vm_values - b) - a / (Vm_values**2)
P_values_ideal = (R * T) / Vm_values

# グラフのプロット
plt.plot(Vm_values, P_values_vdw, label="Van der Waals (H2)", color="red")
plt.plot(
    Vm_values, P_values_ideal, label="Ideal Gas", color="blue", linestyle="--"
)
plt.xlabel("Vm (m³/mol)")
plt.ylabel("P (Pa)")
plt.title("Figure1 H2 P-Vm: Van der Waals vs Ideal Gas")
plt.legend()
plt.grid(True)
plt.ylim(0, 120 * 1.01325e5)

plt.show()
```

**テーマ：** ミクロな分子の運動（壁への衝突と力積）からマクロな状態方程式（$PV = nRT$）を導き、「温度の正体」を解明する。

## 1. 物理モデルの設定

* 一辺の長さが $L$ の立方形の容器（容積 $V = L^3$）を考える。
* この中に、質量 $m$ の気体分子が $N$ 個入っている。
* **仮定：**
* 分子の大きさは無視できる（点質量）。
* 分子間力ははたらかない。
* 壁との衝突は完全弾性衝突（反発係数 $e = 1$）。



---

## 2. PART 1：気体分子運動論による圧力 $P$ の基本導出

### ステップ 1：1個の分子が1回の衝突で壁に与える力積

$x$ 軸に垂直な壁 $A_1$ に、速度 $\vec{v} = (v_x, v_y, v_z)$ の分子が衝突する場合を考えます。

* 衝突前後で $y, z$ 成分の速度は変化せず、$x$ 成分のみ $v_x \rightarrow -v_x$ に反転します。
* 分子の運動量の変化：

$$\Delta p_x = (-m v_x) - (m v_x) = -2 m v_x$$


* 作用・反作用の法則より、**壁が受け取る力積** $I$ は分子の運動量変化の反対符号となります：

$$I = 2 m v_x$$



### ステップ 2：1個の分子が壁に及ぼす平均の力 $F_1$

分子が壁 $A_1$ に衝突後、反対側の壁にぶつかって再び壁 $A_1$ に戻ってくるまでの距離は $2L$ です。

* 衝突の間隔 $\Delta t$：

$$\Delta t = \frac{2L}{v_x}$$


* 1秒間（単位時間）あたりの衝突回数 $f$：

$$f = \frac{1}{\Delta t} = \frac{v_x}{2L}$$


* 1個の分子が壁に及ぼす1秒間の力積の和（＝平均の力 $F_1$）：

$$F_1 = I \times f = (2 m v_x) \times \left(\frac{v_x}{2L}\right) = \frac{m v_x^2}{L}$$



### ステップ 3：全 $N$ 個の分子による圧力 $P$

容器内の全 $N$ 個の分子が壁 $A_1$ に及ぼす力 $F$ は、各分子の $x$ 方向速度の二乗和になります。

$$F = \frac{m}{L} \sum_{i=1}^{N} v_{xi}^2 = \frac{m}{L} \cdot N \overline{v_x^2}$$

ここで、三次元空間において分子運動は等方的（ランダム）なので、速度の二乗平均 $\overline{v^2}$ について：


$$\overline{v_x^2} = \overline{v_y^2} = \overline{v_z^2} = \frac{1}{3}\overline{v^2}$$

よって、全分子が押す力 $F$ は：


$$F = \frac{N m \overline{v^2}}{3L}$$

圧力 $P$ は「単位面積あたりの力（$P = \frac{F}{L^2}$）」であり、容積 $V = L^3$ であるため：

$$\mathbf{P = \frac{F}{L^2} = \frac{N m \overline{v^2}}{3 L^3} = \frac{N m \overline{v^2}}{3V}}$$

両辺に $V$ を掛けると、ミクロな力学モデルから導かれる**分子運動論の基本式**が得られます。

$$\mathbf{PV = \frac{1}{3} N m \overline{v^2}} \quad \text{--- (①式)}$$

---

## 3. PART 2：絶対温度 $T$ と平均運動エネルギー $\bar{E}$ の証明

ここから、ミクロな①式と、マクロな状態方程式を結びつけます。

### ステップ 1：①式の中に「平均運動エネルギー」を作る

分子1個あたりの平均運動エネルギーを $\bar{E} = \frac{1}{2}m\overline{v^2}$ とおきます。
①式の右辺に $\frac{1}{2}$ を作るために、変形します。

$$PV = \frac{2}{3} N \left( \frac{1}{2} m \overline{v^2} \right) = \frac{2}{3} N \bar{E} \quad \text{--- (②式)}$$

### ステップ 2：理想気体の状態方程式を変形する

マクロな実験事実から得られた**理想気体の状態方程式**は以下の通りです。

$$PV = nRT$$

物質量 $n\text{ [mol]}$ は、全分子数 $N$ とアボガドロ定数 $N_\text{A}$ を用いて $n = \frac{N}{N_\text{A}}$ と表せるので：

$$PV = \frac{N}{N_\text{A}} RT = N \left( \frac{R}{N_\text{A}} \right) T$$

ここで、**分子1個あたりの気体定数**として **ボルツマン定数 $k_\text{B}$** を定義します。

$$k_\text{B} \equiv \frac{R}{N_\text{A}} \approx 1.38 \times 10^{-23} \text{ J/K}$$

これを用いると、状態方程式は次のように書き直せます。

$$PV = N k_\text{B} T \quad \text{--- (③式)}$$

### ステップ 3：②式と③式を結合する（完成）

ミクロの式（②式）とマクロの式（③式）の左辺（$PV$）は同じものですから、右辺同士を結びつけます。

$$\frac{2}{3} N \bar{E} = N k_\text{B} T$$

両辺の全分子数 $N$ を消去して $\bar{E}$ について整理すると、目的の公式が導出されます。

$$\mathbf{\bar{E} = \frac{1}{2}m\overline{v^2} = \frac{3}{2}k_\text{B}T}$$

---

## 4. 物理的な重要性

### ① 質量 $m$（気体の種類）の消滅

* 式の右辺 $\frac{3}{2}k_\text{B}T$ には、分子の質量 $m$ や原子量・分子量が一切含まれていません。
* **結論：** 気体の種類（重い・軽い）に関わらず、**温度 $T$ が同じであれば、分子1個あたりの平均運動エネルギー $\bar{E}$ は完全に等しい**。

### ②「絶対温度」の微視的な正体

* 人間がマクロに感じている「温度」とは、微視的には「分子の熱運動の激しさ（平均運動エネルギー）そのもの」を測定しているに過ぎません。

### ③ 分子の速さ（二乗平均速度）の違い

* 運動エネルギー（$\frac{1}{2}m\overline{v^2}$）が同じであるため、質量 $m$ が大きい（重い）分子ほど、実際の飛ぶ速さは遅くなります。

$$\sqrt{\overline{v^2}} = \sqrt{\frac{3k_\text{B}T}{m}} = \sqrt{\frac{3RT}{M}} \quad (\text{ただし } M = m N_\text{A} \text{ はモル質量})$$

## 📝 課題1〜課題3 模範解答＆解説シート

### 課題1：【シミュレーション観察】ミクロな動きとマクロな量の観察（解答）

#### 1. 温度 $T = 300\text{ K}$、質量比 $16\text{倍}$ での観察結果

* **平均運動エネルギー $\bar{E}$ の比較:**
**等しい**（重い分子・軽い分子ともに数値は同じ値を示す）。
* **平均速度 $\sqrt{\overline{v^2}}$ の比較:**
軽い分子の方が速く、数値はおよそ **$4\text{倍}$** となる。
* **1秒間の壁への衝突頻度の比較:**
軽い分子の方が飛び交うスピードが速いため、**衝突頻度（1秒間の回数）も約 $4\text{倍}$** 多くなる。

#### 2. 温度を $300\text{ K} \rightarrow 600\text{ K}$（2倍）にしたときの変化

* **平均運動エネルギー $\bar{E}$:**
温度 $T$ に比例するため **$2\text{倍}$** になる。
* **平均速度 $\sqrt{\overline{v^2}}$:**
運動エネルギーが $2\text{倍}$ になるため、速度は **$\sqrt{2}\text{倍}$（約 $1.41\text{倍}$）** になる。

### 課題2：【数式導出・理由の証明】観察結果を理論式から説明する（解答）

#### 1. 質量比による「速さの比」の証明

**【証明】**

同じ温度 $T$ において、重い分子（質量 $m_H$）と軽い分子（質量 $m_L$）の平均運動エネルギーは等しい。

$$\bar{E}_H = \bar{E}_L$$

$$\frac{1}{2} m_H \overline{v_H^2} = \frac{1}{2} m_L \overline{v_L^2}$$

両辺の $\frac{1}{2}$ を消去し、速度の二乗比について整理する。

$$\frac{\overline{v_L^2}}{\overline{v_H^2}} = \frac{m_H}{m_L}$$

条件より $m_H = 16 m_L$ であるから、$\frac{m_H}{m_L} = 16$ となる。両辺の平方根（ルート）をとると、

$$\frac{\sqrt{\overline{v_L^2}}}{\sqrt{\overline{v_H^2}}} = \sqrt{\frac{m_H}{m_L}} = \sqrt{16} = 4$$

したがって、軽い分子の二乗平均速度は重い分子の **$4\text{倍}$** になる。（証明終）

---

#### 2. 温度と「速さ」の関係の導出

**【証明】**

分子1個あたりの平均運動エネルギーの式 $\frac{1}{2} m \overline{v^2} = \frac{3}{2} k_\text{B} T$ （2次元の場合は $k_\text{B} T$）より、二乗平均速度 $\sqrt{\overline{v^2}}$ について解くと、

$$\sqrt{\overline{v^2}} = \sqrt{\frac{3 k_\text{B} T}{m}}$$

質量 $m$ およびボルツマン定数 $k_\text{B}$ は定数であるため、二乗平均速度は**絶対温度の平方根 $\sqrt{T}$ に比例**する。

$$\sqrt{\overline{v^2}} \propto \sqrt{T}$$

温度を $T$ から $2T$（2倍）に変化させたとき、

$$\sqrt{\overline{v_\text{after}^2}} = \sqrt{\frac{3 k_\text{B} (2T)}{m}} = \sqrt{2} \cdot \sqrt{\frac{3 k_\text{B} T}{m}} = \sqrt{2} \sqrt{\overline{v_\text{before}^2}}$$

したがって、平均速度は **$\sqrt{2}\text{倍}$（約 $1.41\text{倍}$）** になる。（証明終）

### 課題3：【発展考察】「なぜ質量が違っても圧力 $P$ は同じになるのか？」（解答）

**【解答例】**

軽い分子は、重い分子に比べて移動速度が速いため**壁への衝突頻度が大きく**なります。しかし、同じ温度であれば容器全体の**圧力 $P$ は全く同じ**になります。その理由は以下の通りです。

分子が壁に及ぼす平均の力 $F_1$ は、「1回の衝突で与える力積 $I$」と「1秒間の衝突回数 $f$」の積で表されます。

$$F_1 = I \times f = (2 m v_x) \times \left(\frac{v_x}{2L}\right) = \frac{m v_x^2}{L}$$

質量が $\frac{1}{16}$ の軽い分子の場合：

1. **1回の力積 $I$：** 質量 $m$ が $\frac{1}{16}$ ですが、速度 $v_x$ が $4\text{倍}$ になるため、$I = 2 \cdot \left(\frac{1}{16} m\right) \cdot (4 v_x) = \mathbf{\frac{1}{4}\text{倍}}$ に減少します。（1回のインパクトは弱い）
2. **1秒間の衝突回数 $f$：** 速度 $v_x$ が $4\text{倍}$ になるため、壁と壁を往復する時間が短くなり、$f = \frac{4 v_x}{2L} = \mathbf{4\text{倍}}$ に増加します。（衝突回数は多い）

これらを掛け合わせると、

$$F_1 = \left( \frac{1}{4} I \right) \times (4 f) = I \times f$$

となり、**質量 $m$ が減少した影響（$\frac{1}{4}$）と、速さが増加した影響（$4\text{倍}$）が相殺**されます。

分子運動論の基本式 $PV = \frac{1}{3} N m \overline{v^2} = \frac{2}{3} N \bar{E}$ より、圧力 $P$ は分子の質量 $m$ 個別には依存せず、**全分子の平均運動エネルギーの和（＝絶対温度 $T$）のみによって決まる**ため、同じ温度・同じ容器であれば重い分子と軽い分子の圧力は等しくなります。

## 💡 指導・解説のポイント（講評用メモ）

* **二乗平均速度と平均運動エネルギーの混同を防ぐ:**
生徒が誤りがちなポイントとして「温度が2倍になれば速さも2倍になる」という直感的な勘違いがあります。「エネルギーは2倍だが、速度はルートがつくので $\sqrt{2}\text{倍}$ になる」という関係性を確認させてください。
* **ミクロ（力積と頻度）とマクロ（圧力）の相殺関係:**
課題3の「1回の力積は $\frac{1}{4}$ になるが、回数が $4\text{倍}$ になるので相殺される」という説明は、数式の代入だけで終わらせず、物理的な意味（パンチの軽さを手数でカバーしているイメージ）として定着させるのに非常に効果的です。
