| 濃度単位 | 定義式 | 単位 | 特徴・用途 |

① 気液平衡とは？

液体の表面では、

- 液体 → 気体（蒸発）
- 気体 → 液体（凝縮）

が同時に起こっています。<br>
最初は蒸発のほうが多いですが、容器を閉じておくと蒸気が増え、凝縮も起こるようになります。<br>
やがて

$$\text{蒸発速度} = \text{凝縮速度}$$

となります。この状態を気液平衡といいます。<br>
「蒸発が止まった」のではなく、蒸発と凝縮が同じ速さで起こっている状態です。

② 蒸気圧とは？

気液平衡になったとき、液面の上（気相）に存在する蒸気が示す圧力を蒸気圧といいます。<br>
言い換えると、蒸発した分子が容器の壁や液面にぶつかることで生じる圧力です。<br>

ポイント<br>
- 温度が高いほど蒸気圧は大きい
- 物質によって蒸気圧は異なる

例えば、<br>
- エタノールは蒸気圧が高い
- 水はそれより低い

ため、エタノールのほうが蒸発しやすくなります。

③ 蒸気圧と温度の関係

温度が上がると、分子の運動が激しくなります。<br>
その結果、

- 蒸発する分子が増える
- 蒸気圧が上昇する<br>
ことになります。

水の場合、

- 20 ℃ → 約2.3 kPa
- 50 ℃ → 約12 kPa
- 100 ℃ → 約101 kPa

まで蒸気圧が増加します。<br>
100 ℃では蒸気圧が大気圧（約101 kPa）に達します。<br>
なぜ水は100 ℃で沸騰するのか？<br>
液体内部には大気圧がかかっています。蒸気圧が大気圧より小さいと、内部にできた気泡はつぶれてしまいます。<br>
しかし、<br>

$$ \text{蒸気圧} = \text{大気圧}$$

になると、気泡が成長できるようになります。<br>
この状態が **沸騰**です。

</details>

## 液体に溶質を加えると

液体表面から分子が飛び出すと、気体になります。溶媒成分を$\text{A}$とおくと（図１中のsolventを$\text{A}$とおきかえてください）,

$$
\mathrm{A(l)\rightleftharpoons A(g)}
$$

揮発しない溶質を加えると、液面に存在する溶媒分子の割合が小さくなります。<br>
その結果、

$$
P_A=x_A P_A^* \tag{1}
$$

となります。

- $P_A$：溶液中の溶媒の蒸気圧
- $P_A^*$：純溶媒の蒸気圧
- $x_A$：溶媒のモル分率

これが**ラウールの法則**です。


溶質を$B$とすると
$$
x_A=1-x_B
$$
になりますので、

$$
P_A=(1-x_B) P_A^* \tag{1'}
$$

(1')式にしたがって、$P_A$を溶質モル分率$x_B$に対してグラフにすると、直線となることは明らかですね。
下のコードを実行して確認してください。


````python
# ラウールの法則による蒸気圧降下

import numpy as np
import matplotlib.pyplot as plt

# 純溶媒Aの蒸気圧 [kPa]
P_A_star = 100.0

# 溶質Bのモル分率
x_B = np.linspace(0, 0.5, 101)

# 溶媒Aのモル分率
x_A = 1 - x_B

# ラウールの法則
P_A = x_A * P_A_star

# 描画
plt.figure(figsize=(7, 4))

plt.plot(
    x_B,
    P_A,
    color="blue",
    linewidth=2,
    label=r"$P_A=x_A P_A^*$"
)

plt.xlabel(r"Solute mole fraction $x_B$")
plt.ylabel(r"Vapor pressure $P_A$ [kPa]")
plt.title("Raoult's Law: Vapor-Pressure Lowering")

plt.xlim(0, 0.5)
plt.ylim(40, 105)

plt.grid(True, alpha=0.3)
plt.legend()

plt.tight_layout()
plt.show()
````

### **テーマ：なぜ、砂糖を混ぜると水の蒸気圧が下がるのか？**

---

### **1. まずは高校までの「おさらい（直感的なイメージ）」**

* **教科書の説明：**
* 純水なら液面は水分子だけ $\rightarrow$ 自由に飛び出せる。
* 砂糖水なら液面に水分子と砂糖分子が混ざる $\rightarrow$ 水分子が飛び出す「隙間」が減る。


* **イメージ：**
> 「水分子の割合が減る $\rightarrow$ 蒸発できる水分子が減る $\rightarrow$ 蒸気圧が下がる」


* *👉 これで「なるほど」と思えますが、実はもっと根本的な物理のルールが隠されています。*

---

### **2. 大学レベルの視点：主役は「分子の気ままさ（エントロピー）」**

本質的な原因は、「溶質（砂糖）を混ぜると、分子の並べ方のバリエーション（自由度）が爆発的に増えるから」です。

#### **【視覚的イメージ：分子の並び方の違い】**

* **純水の場合（パターンが少ない）**
```text
┌─────────┐
│ ○ ○ ○ ○ │
│ ○ ○ ○ ○ │
│ ○ ○ ○ ○ │
└─────────┘
水分子の並び方は「ほぼ1通り」 → エントロピー 小

```


* **砂糖水を混ぜた場合（パターンが莫大！）**
```text
┌─────────┐
│ ○ ○ □ ○ │
│ □ ○ ○ ○ │
│ ○ ○ ○ □ │
└─────────┘
○ : 水分子 □ : 砂糖分子

並べ方②

┌─────────┐
│ ○ □ ○ ○ │
│ ○ ○ ○ □ │
│ □ ○ ○ ○ │
└─────────┘


並べ方③

┌─────────┐
│ □ ○ ○ ○ │
│ ○ ○ □ ○ │
│ ○ □ ○ ○ │
└─────────┘
・・・

莫大な数の配置が存在
~~~

水分子と砂糖分子の組み合わせで、莫大な数の配置パターンが存在します。

## 3. なぜ「並び方の多さ」が蒸気圧を下げるのか？**

物理の世界には、「確率が高い状態（＝色々な並び方ができる状態）ほど安定で、自然にその状態になりたがる」という大原則があります。

1. **エントロピー（配置の多様さ）の増大**
* 砂糖を混ぜることで、分子の並び方の数 $W$ が増え、**エントロピー（$S$）が大きく増大**します。


2. **自由エネルギーの低下（液相の勝ち）**
* 物理の世界の安定さは、自由エネルギー $G = H - TS$（※温度 $T$ とエントロピー $S$ が効いてくる）で決まります。
* エントロピー $S$ が大きくなると、$G$ の値が小さくなります。
* 「液体でいる状態（液相）」の居心地がものすごく良くなる（＝安定化する）ため、わざわざエネルギーを使って気体になって飛び出す必要がなくなります。



---




---

## 本論にもどります



ラウールの法則では、

$$
P_A=x_AP_A^*
$$

です。<br>
つまり、蒸気圧は**溶媒のモル分率**によって決まります。<br>
溶質を増やすと、

$$
x_A\downarrow
$$

なので、

$$
P_A\downarrow
$$

となります。<br>
したがって、

$$
\boxed{
\text{溶質を加える}
\rightarrow
\text{溶媒のモル分率低下}
\rightarrow
\text{蒸気圧低下}
}
$$

です。<br>
これが**蒸気圧降下 $\Delta P$**です。

---
## 蒸気圧低下についてもう少し

$$
\Delta P = \text{(純溶媒の蒸気圧)} - \text{(溶液の蒸気圧)}
$$

$$
\boxed{\Delta P = P_A^{*}-P_A}
$$


で定義します。



ラウールの法則を代入すると
$$
P_A=x_A P_A^*
$$
なので、
$$
\Delta P = P_A^*-x_A P_A^* = P_A^*(1-x_A)
$$
さらに
$$
x_B=1-x_A
$$
より
$$
\boxed{ \Delta P=P_A^*x_B }
$$
となります。


### 具体例

純水の蒸気圧が
$$
P_A^*=100\ \mathrm{kPa}
$$
溶液の蒸気圧が

$$
P_A=90\ \mathrm{kPa}
$$
なら
$$
\Delta P = 100-90 = 10\ \mathrm{kPa}
$$
です。

## 3. 沸点上昇

ラウールの法則は、単に「蒸気圧が下がる」という話だけではありません。

例えば1 atmで沸騰する場合、

$$
P_{\mathrm{vap}}=P_{\mathrm{ext}}（外圧）
$$

となる必要があります。<br>
ところが、溶質を入れるとラウールの法則によって蒸気圧が下がります。<br>
そのため、<br>
**同じ1 atmに到達するためには、温度をさらに上げる必要があります。**<br>
したがって、

$$
\boxed{
\text{ラウールの法則}
\rightarrow
\text{蒸気圧降下}
\rightarrow
\text{沸点上昇}
}
$$

というつながりがあります。

---

$$
\boxed{
\begin{array}{c}
\text{溶質を加える}\\
\downarrow\\
x_{\rm solvent}\downarrow\\
\downarrow\\
P_{\rm solvent}=x_{\rm solvent}P_{\rm solvent}^*
\downarrow\\
\text{蒸気圧低下}\\
\downarrow\\
\text{沸騰するにはさらに加熱が必要}\\
\downarrow\\
\boxed{\text{沸点上昇}}
\end{array}}
$$

この流れを理解すると、**「ラウールの法則」と「束一性」がバラバラの知識ではなく、一本につながります。**

◇◇◇

沸点上昇は、希薄溶液において、

$$
\boxed{\Delta T_b=K_bm}
$$

と表されます。これは良くご存知だと思います。

- $K_b$：モル沸点上昇定数
- $m$：質量モル濃度 [mol/kg]

参考までに、下のコードを実行すると水の沸点上昇のグラフが表示されます。

````python
import numpy as np
import matplotlib.pyplot as plt

Kb = 0.512       # 水 [K kg/mol]
m = np.linspace(0, 2.0, 100)
Tb0 = 100.0

dTb = Kb * m
Tb = Tb0 + dTb

plt.figure(figsize=(7, 4))
plt.plot(m, Tb, linewidth=2)
plt.xlabel("Molality [mol/kg]")
plt.ylabel("Boiling point [°C]")
plt.title("Boiling-point elevation of water")
plt.grid(True)
plt.show()
````

## 4. 希薄溶液の凝固点降下式

十分に希薄な理想溶液では、凝固点降下は質量モル濃度に比例します。

$$\boxed{\Delta T_f=K_fm}$$

- $\Delta T_f$：凝固点降下（$\rm K$ または ℃の温度差）
- $K_f$：溶媒ごとに決まるモル凝固点降下定数（$\rm K\, \rm{kg}\, \rm{mol}^{-1}$）
- $m$：質量モル濃度（$\rm {mol}\, \rm{kg}^{-1}$）

凝固点は

$$
T_f=T_f^0-\Delta T_f
$$

となります。

◇◇◇

電解質など、溶液中で複数の粒子に分かれる溶質では、理想的にはファントホッフ係数 \(i\) を使って

$$\boxed{\Delta T_f=iK_fm}$$

と表します。凝固点降下は、溶質の種類そのものより、溶液中に存在する溶質粒子の数に強く依存します。これが束一性の特徴です。

例えば


### グルコース

$$
\mathrm{C_6H_{12}O_6}
$$

水中でほぼ電離しません。<br>
したがって、

$$
i\approx1
$$

です。

### NaCl

$$
\mathrm{NaCl\rightarrow Na^+ + Cl^-}
$$

理想的に完全電離すると、

$$
i\approx2
$$

です。
同じ0.10 mol/Lでも、理想的にはNaClのほうが約2倍の粒子を生じます。

# 発展：エントロピーから「ラウールの法則」を導く

## 1. 化学ポテンシャル（$\mu$）とは何か？（「分子の逃げやすさ」）

熱力学では、物質の「その場所での居心地の悪さ」や「そこから逃げ出したい度合い」を化学ポテンシャル（符号：$\mu$）という指標で表します。

* **化学ポテンシャルが高い（$\mu$ が大）：** 居心地が悪いため、他の場所へ逃げたがる（蒸発しやすい、反応しやすい）。
* **化学ポテンシャルが低い（$\mu$ が小）：** 居心地が良いため、その場にとどまりたがる（安定）。

液体（水）と気体（水蒸気）がバランスをとっている（平衡状態）とき、それぞれの化学ポテンシャルは等しくなっています。


$$\mu_{\text{liquid}} = \mu_{\text{vapor}}$$

---

## 2. 砂糖を混ぜると、なぜ液体の化学ポテンシャルが下がるのか？

前回の議論で、砂糖を混ぜるとエントロピー（$S$）が増大し、液体の自由エネルギー（$G$）が低下することを学びました。

化学ポテンシャル $\mu$ は、「物質を1mol増やしたときの自由エネルギーの変化量」です。したがって、液相の自由エネルギーが下がったということは、**水分子の化学ポテンシャル $\mu_{\text{liquid}}$ そのものも下がる**ことを意味します。

> **直感的なイメージ：**
> 砂糖を混ぜて分子の配置が自由になり（エントロピー増大）、液体の居心地が良くなった結果、水分子の「逃げたい度合い（＝化学ポテンシャル）」がグッと下がります。

## 3. ラウールの法則の数式的な導出

化学ポテンシャルの低下が、いかにして「蒸気圧降下（ラウールの法則）」を引き起こすのかを数式で追ってみましょう。

### ① 溶液中の溶媒（水）の化学ポテンシャル

純粋な液体（純水）の化学ポテンシャルを $\mu_1^*$ とします。これに溶質（砂糖）を溶かしたとき、溶媒のモル分率（全体のなかで水分子が占める割合）を $x_1$ とすると、化学ポテンシャルはエントロピーの項（ボルツマンの原理に由来する項）によって次のように表されます。


$$\mu_1 = \mu_1^* + RT \ln x_1$$

* ($R$: 気体定数, $T$: 絶対温度, $x_1 < 1$ なので $\ln x_1$ はマイナスになり、$\mu_1$ は $\mu_1^*$ より低くなります)

### ② 気体（蒸気）側の化学ポテンシャル

一方、気体（蒸気）側の化学ポテンシャルは、そのときの蒸気圧 $P_1$ を用いて次のように表されます。


$$\mu_1(\text{g}) = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1}{P^{\circ}}$$


*(※ $\mu_1^{\circ}$ は標準状態での化学ポテンシャル、$P^{\circ}$ は標準圧力)*

### ③ 平衡状態（液＝気）からの導出

液体と気体が平衡にあるとき、両者の化学ポテンシャルは等しくなります ($\mu_1 = \mu_1(\text{g})$)。


$$\mu_1^* + RT \ln x_1 = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1}{P^{\circ}}$$

ここで、純水の場合（溶質を混ぜていないとき、$x_1 = 1, P_1 = P_1^*$）の平衡関係を考えると、


$$\mu_1^* = \mu_1^{\circ}(\text{g}) + RT \ln \frac{P_1^*}{P^{\circ}}$$


となります。これを先ほどの式に代入して整理すると、綺麗な関係式が導かれます。

$$\mu_1^* + RT \ln x_1 = \mu_1^* + RT \ln \left(\frac{P_1}{P_1^*}\right)$$

両辺から $\mu_1^*$ と $RT$ を引いて整理すると、


$$\ln x_1 = \ln \left(\frac{P_1}{P_1^*}\right)$$


対数を外すと、ついに**ラウールの法則**が得られます！


$$P_1 = x_1 P_1^*$$

---

## 4. まとめ：すべては「エントロピー」から繋がっている

* 砂糖を混ぜる $\rightarrow$ **配置のエントロピー増大**
* 液体の自由エネルギーが下がり、**化学ポテンシャル $\mu_1$ が低下**
* 逃げ出したい勢いが弱まるため、気体側の圧力（蒸気圧 $P_1$）も押し下げられる（**ラウールの法則**の成立）
