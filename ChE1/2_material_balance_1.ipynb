<details>
<summary>説明文</summary>
はじめに質量保存の法則を確認しましょう。化学反応とは、原子の組み替えが起こることですが、たとえ原子の組み替えが起こっても、ある原子が消失したり、生まれたり、ということは核反応を伴わない限り起こりません。
従って、通常の化学反応であるかぎり、反応の前後で物質の総量が変化することはあり得ないと考えて問題ありません。当然と言えばそれまでですが、化学工学ではそれを数式にして表します。
いま、メタンが反応容器に4 molと酸素4 molが投入されたとしましょう。中央の容器で燃焼が起こり、CO2が2 mol、水が4 mol発生したとしましょう。未反応メタンも2 mol残ります。そしてこの発生ガスの半分を系外にパージしたとします。
このとき、マスバランスは原子に注目する考え方と、化合物に注目する考え方とがあります。
系に入る元素は、炭素原子4 mol, 酸素原子8mol,　水素原子16 molです。これらは、半分が蓄積され、半分が系外に出されます。従って系への蓄積量は、炭素原子2 mol, 酸素原子4 mol、水素原子8 molです。
一方、化合物に注目すると、元素と異なり容器の中で生成したり消費されたりもします。従って、inとoutの差だけでなく、容器内での生成と消費をマスバランスに加える必要があります。
例えばメタンの蓄積量は、入量が4 mol, 出量が1 mol, 反応での生成量は0, 反応での消費量は2 molですから、蓄積量は、4-1+0-2=1となり、1 molが蓄積されます。
このように、一見愚直とも思えるような式ですが、このような式を連立して立てることが、成分量や濃度などを計算していく上で必須になるのです。

以下の反応系において、各反応で量論係数が異なる場合を考えます。

$$
\text{A} \xrightarrow{r_1} \text{B} + 2\text{C} \\
\text{B} + \text{C} \xrightarrow{r_2} \text{D} + \text{E} \\
\text{C} \xrightarrow{r_3} 2\text{E}
$$

この場合、反応の詳細は次のようになります。

1. **反応 $r_1$**：Aが反応してBとCを生成します。  
   $\text{A} \rightarrow \text{B} + 2\text{C}$  
   量論係数は $\nu_{1A} = -1$、$\nu_{1B} = +1$、$\nu_{1C} = +2$ です。

2. **反応 $r_2$**：BとCが反応してDとEを生成します。  
   $\text{B} + \text{C} \rightarrow \text{D} + \text{E}$
   量論係数は $\nu_{2B} = -1$、$\nu_{2C} = -1$、$\nu_{2D} = +1$、$\nu_{2E} = +1$ です。

3. **反応 $r_3$**：Cが反応してEを生成します。  
   $\text{C} \rightarrow 2\text{E}$  
   量論係数は $\nu_{3C} = -1$、$\nu_{3E} = +2$ です。

### 物質収支式の立式

反応系全体における物質収支式を立てます。各成分について物質収支を考えます。

物質収支式は次の形式になります：

$$
n_{i1} - n_{i2} + \sum_j P_j \nu_{ji} = 0
$$

ここで、$n_{i1}$ は反応器に入る成分 $i$ の物質量、$n_{i2}$ は反応器から出る成分 $i$ の物質量、$P_j$ は反応 $j$ において消費される限定物質の物質量、$\nu_{ji}$ は反応 $j$ における成分 $i$ の化学量論係数です。

### 1. 成分Aの物質収支

成分Aは反応 $r_1$ でBとCに変換されます。このため、Aの物質収支式は次のようになります：

$$
n_{A1} - n_{A2} + P_1 \nu_{1A} + P_2 \nu_{2A} + P_3 \nu_{3A} = 0
$$

ここで：
- $n_{A1}$ は反応器に入る成分Aの物質量。
- $n_{A2}$ は反応器から出る成分Aの物質量。
- 反応 $r_1$ では、$\nu_{1A} = -1$ でAは消費されます。
- 反応 $r_2$ では、Aは関与しないため、$\nu_{2A} = 0$ です。
- 反応 $r_3$ では、Aは関与しないため、$\nu_{3A} = 0$ です。

したがって、Aの物質収支式は次のようになります：

$$
n_{A1} - n_{A2} - P_1 = 0
$$

### 2. 成分Bの物質収支

Bは反応 $r_1$ で生成され、反応 $r_2$ で消費されます。このため、Bの物質収支式は次のようになります：

$$
n_{B1} - n_{B2} + P_1 \nu_{1B} + P_2 \nu_{2B} + P_3 \nu_{3B} = 0
$$

ここで：
- $n_{B1}$ は反応器に入る成分Bの物質量。
- $n_{B2}$ は反応器から出る成分Bの物質量。
- 反応 $r_1$ では、$\nu_{1B} = +1$ でBは生成されます。
- 反応 $r_2$ では、$\nu_{2B} = -1$ でBは消費されます。
- 反応 $r_3$ では、Bは関与しないため、$\nu_{3B} = 0$ です。

したがって、Bの物質収支式は次のようになります：

$$
n_{B1} - n_{B2} + P_1 - P_2 = 0
$$

### 3. 成分Cの物質収支

Cは反応 $r_1$ で生成され、反応 $r_2$ で消費され、さらに反応 $r_3$ でも消費されます。このため、Cの物質収支式は次のようになります：

$$
n_{C1} - n_{C2} + P_1 \nu_{1C} + P_2 \nu_{2C} + P_3 \nu_{3C} = 0
$$

ここで：
- $n_{C1}$ は反応器に入る成分Cの物質量。
- $n_{C2}$ は反応器から出る成分Cの物質量。
- 反応 $r_1$ では、$\nu_{1C} = +2$ でCは生成されます。
- 反応 $r_2$ では、$\nu_{2C} = -1$ でCは消費されます。
- 反応 $r_3$ では、$\nu_{3C} = -1$ でCは消費されます。

したがって、Cの物質収支式は次のようになります：

$$
n_{C1} - n_{C2} + 2P_1 - P_2 - P_3 = 0
$$

### 4. 成分Dの物質収支

Dは反応 $r_2$ で生成されます。このため、Dの物質収支式は次のようになります：

$$
n_{D1} - n_{D2} + P_1 \nu_{1D} + P_2 \nu_{2D} + P_3 \nu_{3D} = 0
$$

ここで：
- $n_{D1}$ は反応器に入る成分Dの物質量。
- $n_{D2}$ は反応器から出る成分Dの物質量。
- 反応 $r_1$ では、Dは関与しないため、$\nu_{1D} = 0$ です。
- 反応 $r_2$ では、$\nu_{2D} = +1$ でDは生成されます。
- 反応 $r_3$ では、Dは関与しないため、$\nu_{3D} = 0$ です。

したがって、Dの物質収支式は次のようになります：

$$
n_{D1} - n_{D2} + P_2 = 0
$$

### 5. 成分Eの物質収支

Eは反応 $r_2$ で生成され、反応 $r_3$ でも生成されます。このため、Eの物質収支式は次のようになります：

$$
n_{E1} - n_{E2} + P_1 \nu_{1E} + P_2 \nu_{2E} + P_3 \nu_{3E} = 0
$$

ここで：
- $n_{E1}$ は反応器に入る成分Eの物質量。
- $n_{E2}$ は反応器から出る成分Eの物質量。
- 反応 $r_1$ では、Eは関与しないため、$\nu_{1E} = 0$ です。
- 反応 $r_2$ では、$\nu_{2E} = +1$ でEは生成されます。
- 反応 $r_3$ では、$\nu_{3E} = +2$ でEは生成されます。

したがって、Eの物質収支式は次のようになります：

$$
n_{E1} - n_{E2} + P_2 + 2P_3 = 0
$$

次に

- 反応(A):CH$_3$OH + (1/2)O$_2$ $\longrightarrow$ HCOH + H$_2$O

- 反応(B):CH$_3$OH $\longrightarrow$ HCOH + H$_2$

- 反応(C):CH$_3$OH + (3/2)O$_2$ $\longrightarrow$ CO$_2$ + 2H$_2$O

の反応系について物質収支式:

$$
n_{i1}-n_{i2}+ \sum_{j} P_j \nu_{ji} =0 \ \ ... (A)
$$

を考えます。$i$は成分, $j$は反応式の種類になります。この場合はA,B,Cの3種類です。量論係数は$ \nu_{Ai}$,$ \nu_{Bi}$,$ \nu_{Ci}$の３種類となり、これを先ほどの表に追加した表が下になります。

| 成分$i$ | 入量[mol] | 出量[mol] |$ \nu_{Ai}$ |$ \nu_{Bi}$ |$ \nu_{Ci}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH$_3$OH | 100 | C |-1|-1 |-1 |  
| O$_2$ | 27.3 | 9.79 |-1/2 |- |-3/2 |
| N$_2$ | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |1 |1 |- |
| H$_2$ | 0 | 40.24 |- |1 |- |
| CO$_2$ | 0 | 2.64 |- |- |1 |
| H$_2$O | 0 | H |1 |- |2 |

さらに4,5,6列目を$P_j \nu_{ji}$にする。記号を簡単にするため
反応Aによって消費される限定成分の物質量$P_A$を$x$,同じく$P_B$を$y$、$P_C$を$z$とする。

| 成分$i$ | 入量[mol] | 出量[mol] |$ \nu_{Ai}P_{A}$ |$ \nu_{Bi}P_{B}$ |$ \nu_{Ci}P_{C}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH$_3$OH | 100 | C |-1$x$|-1$y$ |-1$z$ |  
| O$_2$ | 27.3 | 9.79 |-1/2$x$ |- |-3/2 $z$|
| N$_2$ | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |1$x$ |1$y$ |- |
| H$_2$ | 0 | 40.24 |- |1$y$ |- |
| CO$_2$ | 0 | 2.64 |- |- |1 $z$|
| H$_2$O | 0 | H |1$x$ |- |2 $z$|

このような表にすると各成分の物質収支は行ごとに並べれば良い。つまり

* (i=CH3OH): $100-C-x-y-z=0$.
* (i=O2):    $27.3-9.79-(1/2)x-(3/2)z=0$
* (i=HCOH): $0-E+x+y=0$
* (i=H2): $0-40.24+y=0$
* (i=CO2): $0-2.64+z=0$
* (i=H2O): $0-H+x+2z=0$

未知の変数は６個、式の数が６つなので、この連立方程式を解けば解が求まる。
