<h1> 反応を伴う物質収支</h1>


化学反応を伴うプロセスの物質収支についてもう少し勉強します。
今回は混乱をさけるために、連続操作に限定します。

化学工学において最も重要な装置の一つである反応器では、
反応物が消費され、生成物が新たに生じます。

例えば、
$$
\mathbf{A} + \mathbf{B} \longrightarrow  \mathbf{C}
$$
という反応では、$\mathbf{A}$と$\mathbf{B}$は減少し、$\mathbf{C}$は増加します。

このような場合でも物質保存則は成立しているため、
各成分について生成・消費を考慮した物質収支を立てることができます。

これから学ぶ反応器の物質収支は、
化学プラント設計の基礎となる重要な考え方です。

まずは簡単な反応器の例を用いて、
反応器出口の組成や流量をどのように求めるかを学びましょう。

---




まずは、例題を通して、反応器の物質収支のとり方を勉強します。

---
<h3>例題1</h3>

$\text{H}_2\text{S}$を触媒を用いて空気で部分酸化する次の連続プロセスを考える。

$$ \text{H}_2\text{S}+ \frac{1}{2}\text{O}_2 \longrightarrow \text{S}+\text{H}_2\text{O}$$

$$ (\text{H}_2\text{S}, \text{O}_2, \text{N}_2) \longrightarrow \boxed{反応器} \longrightarrow (\text{S}, \text{H}_2\text{O}, \text{H}_2\text{S}, \text{O}_2, \text{N}_2)
$$

反応器に$\text{H}_2\text{S}$と$\text{O}_2$が **量論比** で供給され、$\text{H}_2\text{S}$の転化率が96%のとき、$\text{H}_2\text{S}$ 100 mol/h に対する反応器出口の各成分の量を求めよ。酸素源としては空気を用い、空気は窒素と酸素の２成分とし、その物質量比は79:21とする。

<h3>物質収支式の一般式</h3>
単一反応で、単純な場合は、先程の解答例のように逐次計算することができますが、反応が複雑な複合反応になってくると、混乱する場合があります。そのような場合、以下の方法を理解しておくと便利です。

複数種類の反応式から構成される、複合反応のうち$j$番目の化学反応を伴う物質収支式は
$$
F_{in,i}-F_{out,i}+ \sum_{j} \dot{\xi_j} \nu_{ji} =0 \ \ \tag{A}
$$
で表されます。ここで$i$は化学種です。

* $F_{in,i}$:反応器に入る成分$i$の物質量速度 [mol/h] 
* $F_{out,i}$:反応器から出る成分$i$の物質量速度[mol/h]
* $\dot{\xi_j}$:反応$j$によって消費される限定物質の物質量速度[mol/h] （反応進行度）
* $\nu_{ji}$:反応$j$における成分$i$の化学量論係数 

注）:反応進行度$\xi$について
* 回分反応器では反応進行度 $\xi$ [mol] 
* 連続流反応器では単位時間あたりの反応量 $\dot{\xi}$ [mol/h] 

---

---
### 補足：

式(A)は、実は
$$
\text{流入}-\text{流出}+\text{生成}-\text{消費}=0
$$

の「生成−消費」を化学量論で表しただけです。

導出を順を追って説明します。

◇◇◇

### 1. 一般物質収支から出発

定常状態の反応器では

$$
\text{流入} -\text{流出} +\text{生成} -\text{消費} =0
$$
が成り立ちます。

成分 $i$ について書けば

$$
F_{in,i}-F_{out,i} +R_i =0 \tag{1}
$$
ここで
$$
R_i= (\text{生成量})-(\text{消費量})
$$
です。

### 2. 単一反応の場合

#### 反応1

$$
A+B\rightarrow C
$$
を考えます。

化学量論係数を

$\nu_A=-1,\qquad \nu_B=-1,\qquad \nu_C=+1$

とします。

反応が $\dot{\xi}$ [mol/h] 進行したとすると、
単位時間あたりに
* Aは $\xi$ mol 消費
* Bは $\xi$ mol 消費
* Cは $\xi$ mol 生成

されます。

つまり

$$
R_i=\nu_i \dot{\xi}
$$
と書けますので


$$
R_A=(-1)\dot{\xi}=-\dot{\xi}, \quad R_B=(-1)\dot{\xi}=-\dot{\xi}, \quad R_C=(+1)\dot{\xi}=+\dot{\xi}
$$
です。

式(1)へ代入すると
$$
\begin{aligned}
F_{in,A}-F_{out,A} - \dot{\xi} &=0\\
F_{in,B}-F_{out,B} - \dot{\xi} &=0\\
F_{in,C}-F_{out,C} + \dot{\xi} &=0\\
\end{aligned}
$$

となります。

◇◇◇

#### 反応例2


$$
2A+B\rightarrow 3C
$$
なら

$$
\nu_A=-2,\quad \nu_B=-1,\quad \nu_C=+3
$$
です。

変化量は
$$
A:\quad -2\dot{\xi} \\ 
B:\quad -\dot{\xi} \\
C:\quad +3 \dot{\xi}
$$
です。

よって

$$
\begin{aligned}
F_{in,A}-F_{out,A} - 2 \dot{\xi} &=0\\
F_{in,B}-F_{out,B} - \dot{\xi} &=0\\
F_{in,C}-F_{out,C} + 3 \dot{\xi} &=0\\
\end{aligned}
$$

これをまとめて１つの式で表現すると(A)式になるということです。
$$
\boxed{
F_{in,i}-F_{out,i}+ \sum_{j} \dot{\xi_j} \nu_{ji} =0 \ \ \tag{A}
}
$$

---
### 3. 複数の反応がある場合

次に複合反応

$$
A\rightarrow B\\
B\rightarrow C
$$

を考えます。

それぞれの反応進行度を
$\xi_1,\quad \xi_2$
とします。

$\boxed{反応1}$
$$
A\rightarrow B
$$
の化学量論係数は

$\nu_{1A}=-1, \quad \nu_{1B}=+1, \quad \nu_{1C}=0$

です。

$\boxed{反応2}$

$$
B\rightarrow C
$$
の化学量論係数は

$$
\nu_{2A}=0, \quad \nu_{2B}=-1,\quad \nu_{2C}=+1
$$
です。

◇◇◇


### 4. 各反応の寄与を足し合わせる

ここで成分Bについて考えます。

反応1では

$+\xi_1$

だけ生成される。

反応2では

$-\xi_2$

だけ消費される。

したがって
$$
R_B=\xi_1-\xi_2
$$

です。

化学量論係数を使うと
$$
R_B=\xi_1\nu_{1B}+ \xi_2\nu_{2B}
$$
となります。

同様に任意の成分 $i$ について
$$
R_i = \xi_1\nu_{1i} + \xi_2\nu_{2i} +\cdots
$$
となるので
$$
R_i=\sum_j \xi_j\nu_{ji} \tag{2}
$$
が得られます。


### 4. 一般物質収支へ代入

式(1)
$$
F_{in,i}-F_{out,i}+R_i=0
$$
へ式(2)を代入すると
$$
F_{in,i} - F_{in, i} + \sum_j \xi_j\nu_{ji} = 0
$$
となり、
$$
\boxed{F_{in,i}-F_{out,i} +\sum_j \xi _j\nu_{ji} = 0 }
$$
が得られます。

これが式(A)です。

  <summary>
  **(A)式をもう少し詳しく知りたい人はみてください。**  
  </summary>





以下の反応系において、各反応で量論係数が異なる場合を考えます。

$$
\begin{aligned}
\text{A} &\xrightarrow{r_1} \text{B} + 2\text{C} \\
\text{B} + \text{C} &\xrightarrow{r_2} \text{D} + \text{E} \\
\text{C} &\xrightarrow{r_3} 2\text{E}
\end{aligned}
$$

この場合、反応の詳細は次のようになります。

1. **反応 $r_1$**：Aが反応してBとCを生成します。  
   $\text{A} \rightarrow \text{B} + 2\text{C}$  
   量論係数は $\nu_{1A} = -1$、$\nu_{1B} = +1$、$\nu_{1C} = +2$ です。

2. **反応 $r_2$**：BとCが反応してDとEを生成します。  
   $\text{B} + \text{C} \rightarrow \text{D} + \text{E}$
   量論係数は $\nu_{2B} = -1$、$\nu_{2C} = -1$、$\nu_{2D} = +1$、$\nu_{2E} = +1$ です。

3. **反応 $r_3$**：Cが反応してEを生成します。  
   $\text{C} \rightarrow 2\text{E}$  
   量論係数は $\nu_{3C} = -1$、$\nu_{3E} = +2$ です。

### 物質収支式の立式

反応系全体における物質収支式を立てます。各成分について物質収支を考えます。

物質収支式は次の形式になります：

$$
F_{in,i} - F_{out,i} + \sum_j \xi_j \nu_{ji} = 0
$$

ここで、$F_{in,i}$ は反応器に入る成分 $i$ の物質量、$F_{out,i}$ は反応器から出る成分 $i$ の物質量、$\xi_j$ は反応 $j$ において消費される限定物質の物質量、$\nu_{ji}$ は反応 $j$ における成分 $i$ の化学量論係数です。

### 1. 成分Aの物質収支

成分Aは反応 $r_1$ でBとCに変換されます。このため、Aの物質収支式は次のようになります：

$$
F_{in,A} - F_{out,A} + \xi_1 \nu_{1A} + \xi_2 \nu_{2A} + \xi_3 \nu_{3A} = 0
$$

ここで：
- $F_{in,A}$ は反応器に入る成分Aの物質量。
- $F_{out,A}$ は反応器から出る成分Aの物質量。
- 反応 $r_1$ では、$\nu_{1A} = -1$ でAは消費されます。
- 反応 $r_2$ では、Aは関与しないため、$\nu_{2A} = 0$ です。
- 反応 $r_3$ では、Aは関与しないため、$\nu_{3A} = 0$ です。

したがって、Aの物質収支式は次のようになります：

$$
F_{in,A} - F_{out,A} - \xi_1 = 0
$$

### 2. 成分Bの物質収支

Bは反応 $r_1$ で生成され、反応 $r_2$ で消費されます。このため、Bの物質収支式は次のようになります：

$$
F_{in,B} - F_{out,B} + \xi_1 \nu_{1B} + \xi_2 \nu_{2B} + \xi_3 \nu_{3B} = 0
$$

ここで：
- $F_{in,B}$ は反応器に入る成分Bの物質量。
- $F_{out,B}$ は反応器から出る成分Bの物質量。
- 反応 $r_1$ では、$\nu_{1B} = +1$ でBは生成されます。
- 反応 $r_2$ では、$\nu_{2B} = -1$ でBは消費されます。
- 反応 $r_3$ では、Bは関与しないため、$\nu_{3B} = 0$ です。

したがって、Bの物質収支式は次のようになります：

$$
F_{in,B} - F_{out,B} + \xi_1 - \xi_2 = 0
$$

### 3. 成分Cの物質収支

Cは反応 $r_1$ で生成され、反応 $r_2$ で消費され、さらに反応 $r_3$ でも消費されます。このため、Cの物質収支式は次のようになります：

$$
F_{in,C} - F_{out,C} + \xi_1 \nu_{1C} + \xi_2 \nu_{2C} + \xi_3 \nu_{3C} = 0
$$

ここで：
- $F_{in,C}$ は反応器に入る成分Cの物質量。
- $F_{out,C}$ は反応器から出る成分Cの物質量。
- 反応 $r_1$ では、$\nu_{1C} = +2$ でCは生成されます。
- 反応 $r_2$ では、$\nu_{2C} = -1$ でCは消費されます。
- 反応 $r_3$ では、$\nu_{3C} = -1$ でCは消費されます。

したがって、Cの物質収支式は次のようになります：

$$
F_{in,C} - F_{out,C} + 2\xi_1 - \xi_2 - \xi_3 = 0
$$

### 4. 成分Dの物質収支

Dは反応 $r_2$ で生成されます。このため、Dの物質収支式は次のようになります：

$$
F_{in,D} - F_{out,D} + \xi_1 \nu_{1D} + \xi_2 \nu_{2D} + \xi_3 \nu_{3D} = 0
$$

ここで：
- $F_{in,D}$ は反応器に入る成分Dの物質量。
- $F_{out,D}$ は反応器から出る成分Dの物質量。
- 反応 $r_1$ では、Dは関与しないため、$\nu_{1D} = 0$ です。
- 反応 $r_2$ では、$\nu_{2D} = +1$ でDは生成されます。
- 反応 $r_3$ では、Dは関与しないため、$\nu_{3D} = 0$ です。

したがって、Dの物質収支式は次のようになります：

$$
F_{in,D} - F_{out,D} + \xi_2 = 0
$$

### 5. 成分Eの物質収支

Eは反応 $r_2$ で生成され、反応 $r_3$ でも生成されます。このため、Eの物質収支式は次のようになります：

$$
F_{in,E} - F_{out,E} + \xi_1 \nu_{1E} + \xi_2 \nu_{2E} + \xi_3 \nu_{3E} = 0
$$

ここで：
* $F_{in,E}$ は反応器に入る成分Eの物質量。
* $F_{out,E}$ は反応器から出る成分Eの物質量。
* 反応 $r_1$ では、Eは関与しないため、$\nu_{1E} = 0$ です。
* 反応 $r_2$ では、$\nu_{2E} = +1$ でEは生成されます。
* 反応 $r_3$ では、$\nu_{3E} = +2$ でEは生成されます。

したがって、Eの物質収支式は次のようになります：

$$
F_{in,E} - F_{out,E} + \xi_2 + 2\xi_3 = 0
$$

<h3>例題１の解法２</h3>

再び例題１にもどります。

(A)式をベースとした解き方をいかに示します。

◇◇◇


##### 1. 反応進行度 $\xi$ を求める 

本問題では反応が1つだけ($j=1$)なので、

$$
F_{in,i}-F_{out,i}+ \xi \nu_i=0 
$$

と書けます。

ここで $\xi$ は反応によって消費された限定物質の量で,限定物質を $\mathrm{H_2S}$ とすると、転化率は

$$
X= \frac{\text{反応した }\mathrm{H_2S}} {\text{供給した }\mathrm{H_2S}}
$$

より

$$
\xi=X\,F_{\mathrm{in, H_2S}}
$$
したがって

$$
\xi=0.96\times100 =96\ \mathrm{mol/h}
$$
です。

◇◇◇


反応

$$
\mathrm{H_2S} +\frac12\mathrm{O_2} \rightarrow \mathrm{S} +\mathrm{H_2O}
$$

に対して，

$$
\mathbf{F}_{out} = \mathbf{F}_{in} + \boldsymbol{\nu}\xi
$$

と書くと，

$$
\mathbf{F}_{out} = \begin{bmatrix} F_{\mathrm{H_2S}}\\ F_{\mathrm{O_2}}\\ F_{\mathrm{N_2}}\\ F_{\mathrm{S}}\\ F_{\mathrm{H_2O}} \end{bmatrix}_{out} = \begin{bmatrix} 100\\ 50\\ \dfrac{79}{21}\times 50\\ 0\\ 0 \end{bmatrix} + \begin{bmatrix} -1\\ -\dfrac12\\ 0\\ 1\\ 1 \end{bmatrix}\xi .
$$



$$
\xi = 100\times0.96 = 96\ \mathrm{mol/h}
$$
を代入すると

$$
\mathbf{F}_{out} = \begin{bmatrix} 100\\ 50\\ 188.1\\ 0\\ 0 \end{bmatrix} + 96 \begin{bmatrix} -1\\ -\dfrac12\\ 0\\ 1\\ 1 \end{bmatrix} = \begin{bmatrix} 4\\ 2\\ 188.1\\ 96\\ 96 \end{bmatrix}.
$$
と求まります。

5. 結果

反応器出口組成は

| 成分               | 出口流量 \[mol/h] |
| ---------------- | ------------: |
| $\mathrm{H_2S}$ |           4.0 |
| $\mathrm{O_2}$  |           2.0 |
| $\mathrm{N_2}$  |         188.1 |
| $\mathrm{S}$     |          96.0 |
| $\mathrm{H_2O}$ |          96.0 |



ポイント

この解法の特徴は、

1. 転化率から反応進行度 $\xi$ を求める
2. 各成分について式(A) 
$F_{in,i}-F_{out,i}+ \xi \nu_i=0 $
をそのまま適用する
3. 未知量は各成分の出口流量 $F_{out,i}$ のみになる

という点です。

複数反応系でも **式(A)** 

を用いるだけで同様に解けるため、一般性の高い物質収支式です。簡単な反応系（実プロセスでは皆無ですが）の場合は解法１でもいいかもしれません。

<h3>解答例</h3>

以下の表の値を求めることになる。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |

| CH<sub>3</sub>OH | 100 | C |

| HCOH | 0 | E |
| H<sub>2</sub> | 0 | F |
| CO<sub>2</sub> | 0 | G |

- 仕様からわかるもの（A,B,D,F,G）
  - A,B:仕様2.3から
    - A=100 $\times$ 1.3 $\times$ 0.21 = 27.3 mol/h
    - B=100 $\times$ 1.3 $\times$ 0.79 = 102.7 mol/h

  - D,F,G:B=102.7 molと仕様2.2から
    - D=102.7 $\times$ (6.3/66.1)  = 9.79 mol/h
    - F=102.7 $\times$ (25.9/66.1)  = 40.24 mol/h
    - G=102.7 $\times$ (1.7/66.1)  = 2.64 mol/h

---




次に
$$ \begin{aligned} 
\text{反応(A):} \quad & \text{CH}_3\text{OH} + \frac{1}{2}\text{O}_2 \longrightarrow \text{HCHO} + \text{H}_2\text{O} \\ 
\text{反応(B):} \quad & \text{CH}_3\text{OH} \longrightarrow \text{HCHO} + \text{H}_2 \\ \text{反応(C):} \quad & \text{CH}_3\text{OH} + \frac{3}{2}\text{O}_2 \longrightarrow \text{CO}_2 + 2\text{H}_2\text{O} 
\end{aligned} 
$$



の反応系について物質収支式:

$$
n_{i1}-n_{i2}+ \sum_{j} P_j \nu_{ji} =0 \ \ ... (A)
$$

を考えます。$i$は成分, $j$は反応式の種類になります。この場合$j$はA,B,Cの3種類です。量論係数は$ \nu_{Ai}$,$ \nu_{Bi}$,$ \nu_{Ci}$の３種類となり、これを先ほどの表に追加した表が下になります。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |$ \nu_{Ai}$ |$ \nu_{Bi}$ |$ \nu_{Ci}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH<sub>3</sub>OH | 100 | C |-1|-1 |-1 |  
| O<sub>2</sub> | 27.3 | 9.79 |-1/2 |- |-3/2 |
| N<sub>2</sub> | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |1 |1 |- |
| H<sub>2</sub> | 0 | 40.24 |- |1 |- |
| CO<sub>2</sub> | 0 | 2.64 |- |- |1 |
| H<sub>2</sub>O | 0 | H |1 |- |2 |

さらに4,5,6列目を$P_j \nu_{ji}$にする。記号を簡単にするため
- 反応Aによって消費される限定成分の物質量$P_A$を$x$
- $P_B$を$y$
- $P_C$を$z$
とする。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |$ \nu_{Ai}P_{A}$ |$ \nu_{Bi}P_{B}$ |$ \nu_{Ci}P_{C}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH<sub>3</sub>OH  | 100 | C |$-1x$|$-1y$ |$-1z$ |  
| O<sub>2</sub> | 27.3 | 9.79 |$-1/2x$ |- |$-3/2 z$|
| N<sub>2</sub> | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |$1x$ |$1y$ |- |
| H<sub>2</sub> | 0 | 40.24 |- |$1y$ |- |
| CO<sub>2</sub> | 0 | 2.64 |- |- |$1 z$|
| H<sub>2</sub>O| 0 | H |$1x$ |- |$2 z$|

このような表にすると各成分の物質収支は行ごとに並べれば良い。つまり

* (i=CH3OH): $100-C-x-y-z=0$.
* (i=O2):    $27.3-9.79-(1/2)x-(3/2)z=0$
* (i=HCOH): $0-E+x+y=0$
* (i=H2): $0-40.24+y=0$
* (i=CO2): $0-2.64+z=0$
* (i=H2O): $0-H+x+2z=0$

未知の変数は６個、式の数が６つなので、この連立方程式を解けば解が求まる。


* $C=100-x-y-z$.
* $9.79= 27.3-(1/2)x-(3/2)z$
* $E= 0 +x+y $
* $40.24 = y$
* $2.64 = z$
* $H =  x+2z$

\begin{bmatrix} 
C\\ 9.79\\ 102.7\\ E\\ 40.24\\ 2.64\\ H 
\end{bmatrix} = \begin{bmatrix} 100\\ 27.3\\ 102.7\\ 0\\ 0\\ 0\\ 0 \end{bmatrix}+ \begin{bmatrix}-1&-1&-1\\ -\frac12&0&-\frac32\\ 0&0&0\\ 1&1&0\\ 0&1&0\\ 0&0&1\\ 1&0&2 \end{bmatrix} \begin{bmatrix} x\\ y\\ z \end{bmatrix}

```python
import numpy as np

# 未知数ベクトル
# u = [C, E, H, x, y, z]^T

A = np.array([
    [1, 0, 0, 1, 1, 1],      # C + x + y + z = 100
    [0, 0, 0, 0.5, 0, 1.5],  # 0.5x + 1.5z = 27.3 - 9.79
    [0, 1, 0, -1, -1, 0],    # E - x - y = 0
    [0, 0, 0, 0, 1, 0],      # y = 40.24
    [0, 0, 0, 0, 0, 1],      # z = 2.64
    [0, 0, 1, -1, 0, -2]     # H - x - 2z = 0
])

b = np.array([
    100,
    27.3 - 9.79,
    0,
    40.24,
    2.64,
    0
])

# 方程式を解く
u = np.linalg.solve(A, b)

# 結果を取り出す
C, E, H, x, y, z = u

# 出力
print(f"C = {C:.2f}")
print(f"E = {E:.2f}")
print(f"H = {H:.2f}")
print(f"x = {x:.2f}")
print(f"y = {y:.2f}")
print(f"z = {z:.2f}")
```
