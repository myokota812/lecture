### 提出課題3：未反応Aおよび溶媒Sのリサイクル

課題2と同じ

$$
A \rightarrow P
$$

の反応プロセスを考える。

課題2では未反応Aおよび溶媒Sを廃棄していたが，今回はそれらを全量リサイクルするものとする。

なお，

- 分離器で回収された溶媒Sは全量を原料調整工程へリサイクルする。
- リサイクル溶媒流量は 200 kg/h とする。
- 溶媒損失はないものとし，新規の溶媒供給は行わない。
- 原料Aは，新規供給分 100 kg/h と分離器で回収されたリサイクル分を合わせて反応器へ供給する。

反応器内でのAの単流転化率は80%である。

反応器出口は分離器へ送られ，分離器では

- 製品Pを全量回収する
- 未反応Aを全量回収し原料調整工程へリサイクルする
- 溶媒Sを全量回収し原料調整工程へリサイクルする

ものとする。

プロセスは連続・定常状態で運転されている。

---

(1) 次の単位操作を含むBFDを作図せよ。

- 原料調整
- 反応器

反応器におけるAの単流転化率の定義式を書け。

(4)　プロセス全体を検査体積として，Aの物質収支式を書け。
