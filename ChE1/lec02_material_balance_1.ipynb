<h2> 2. プロセスの操作方式 </h2>

化学プロセスは操作方法によって、

* 回分操作（Batch Process）
* 半回分操作（Semi-batch Process）
* 連続操作（Continuous Process）

の3種類に大別されます。
BFD（Block Flow Diagram）を作成する際には、まず対象プロセスがどの運転方式であるかを理解することが重要です。

</figcaption>

<details>
<summary>説明</summary>
<h3> 回分操作（Batch Process）</h3>

原料を一度に装置へ投入し、反応や処理を行った後、生成物を取り出す操作。
処理中は原則として原料や生成物の出入りはない。
#### 特徴
* 原料をまとめて処理
* 多品種少量生産に適する
* 操作が柔軟
* 生産量は比較的小さい

◇◇◇

<h3> 半回分操作（Semi-batch Process）</h3>

反応中に原料または生成物の一部が出入りする操作である。
回分操作と連続操作の中間的な運転方式と考えられる。

#### 特徴
* 発熱反応の制御がしやすい
* 危険反応に利用される
* 反応速度の調整が可能

#### 利用例
* 重合反応
* ニトロ化反応
* 医薬品合成

#### なぜ使うのか

例えば反応<br>
A + B → P<br>
が非常に発熱性の場合、<br>
Bを一度に入れると温度が急上昇する。<br>
そこでBを少しずつ供給することで反応を安全に制御する。


◇◇◇

<h3> 連続操作（Continuous Process）</h3>

原料を連続的に供給し、生成物を連続的に取り出す操作。
化学工業では最も一般的な運転方式。

#### 特徴
* 大量生産に適する
* 生産効率が高い
* 品質が安定しやすい
* 定常運転を行うことが多い


#### 定常状態

連続運転では、
装置内の蓄積量が時間とともに変化しない状態を定常状態（Steady State）という。

流入量 = 流出量

となることが多い。<br>
プロセス設計や物質収支計算では定常状態を仮定する場合が多い。

◇◇◇
</details>

<h3> 【補足】 BFDでの表現</h3>

BFDでは通常、

~~~
[混合]
   ↓
[反応]
   ↓
[分離]
~~~

のように単位操作のつながりを表現します。

ただし同じBFDであっても、

* 回分プロセス
* 半回分プロセス
* 連続プロセス

では**矢印**の意味が異なります。

例えば、

A → 混合 → 反応 → 分離 → P


というBFDは、

回分反応にも連続反応にも用いられますが、

* 回分の場合：矢印は操作手順
* 連続の場合：矢印は物質の移動

のようにみてください。

<h2>3. ブロックフローダイアグラム（BFD）</h2>

<h3>3.1 ブロックフローダイアグラムとは</h3>

化学プロセスやバイオプロセスは、多くの装置によって構成されています。プロセス全体を理解しやすくするために、各装置を「単位操作（Unit Operation）」として表し、それらを矢印線で接続した図を **ブロックフローダイアグラム（Block Flow Diagram: BFD）** と呼びます。

BFDはプロセス設計の初期段階で作成され、原料がどのような工程を経て製品になるのかを大まかに示します。

◇ ◇ ◇
<h3>3.2 単位操作の表現</h3>

BFDでは、単位操作を四角形や円で表します。

<figure style="text-align:center;">
<img
src="https://raw.githubusercontent.com/myokota812/lecture/master/ChE1/figures/unit.png"
width="300">
</figure>

<details><summary>説明</summary>
ここでは、BFD（ブロックフロー図）における単位操作の表現方法について説明します。<br>
 BFDでは、各装置や工程を「単位操作」として四角形や円で表します。<br>
 図のように、左側から原料やエネルギーなどの入量が単位操作に入り、その内部で処理が行われ、右側から出量として取り出されます。<br>
 このように単位操作を簡略化して表現することで、複雑なプロセス全体の流れを分かりやすく把握することができます。<br>
 実際の化学プロセスでは、混合、加熱、冷却、蒸留、ろ過といった様々な単位操作が組み合わされて構成されています。

◇ ◇ ◇
<h3>3.3 単位操作の例</h3>

* Mixer（混合）
* Reactor（反応）
* Separator（分離）
* Distillation Column（蒸留塔）
* Heat Exchanger（熱交換器）

<figure style="text-align:center;">
<img
src="https://raw.githubusercontent.com/myokota812/lecture/master/ChE1/figures/unit_exmpl.png"
width="300">
</figure>

<details><summary>説明</summary>
ここでは、化学プロセスでよく用いられる代表的な単位操作の例を紹介します。<br>
 主な単位操作には、混合を行うMixer、化学反応を行うReactor、成分を分離するSeparator、蒸留によって成分を分けるDistillation Column、そして熱のやり取りを行うHeat Exchangerがあります。<br>
上の図を見てみましょう。<br>
 一番上の混合操作では、成分Aと成分Bが装置に入り、混合された状態で流出します。<br>
 真ん中の反応操作では、成分Aと成分Bが反応器に入り、反応によって成分Cが生成されます。また、反応しなかった成分は未反応成分として出口に現れます。<br>
 一番下の分離操作では、混合物として入ってきた成分Aと成分Bが、それぞれ別の流れとして取り出されます。<br>
このように、単位操作は「入ってくる流れ」と「出ていく流れ」に着目して表現されます。<br>
実際の化学プラントは、これらの単位操作を組み合わせることで構成されています。<br>

単位操作間を移動させる物質の流れ（ストリーム）の書き方については、次に説明します。

◇ ◇ ◇
<h3> 3.4 ストリームの表現</h3>

単位操作（Unit Operation）間を結ぶ線は ストリーム（Stream） と呼ばれ、物質またはエネルギーの流れを表します。

流れの方向は矢印で示します。

原料 ───► 反応器 ───► 生成物

**基本ルール**
- 1本のストリームには流れ方向を示す矢印を付ける
- 矢印は流体の移動方向を表す
- 一般的にはプロセスを左から右へ描く

**ストリームに記載される情報(BFDの場合)**

通常は次の情報のみ記載します。

- 物質名
- 流量


例
~~~
Toluene Feed
10,000 kg/h
~~~
* PFDの場合は

Flow Rate : 10,000 kg/h<br>
Temperature : 35°C<br>
Pressure : 40 bar

Composition:<br>
Toluene 99.5 wt%<br>
Benzene 0.5 wt%


のように、より詳細な情報を記載します。

<h3>物質収支</h3>

<figure style="text-align:center;">
<img
src="https://raw.githubusercontent.com/myokota812/lecture/master/ChE1/figures/matebala1.png"
width="300">
<figcaption>
図1 物質収支の基礎
</figcaption>
</figure>

◇◇◇

* 全成分：（系の蓄積量）＝（系の入量）ー（系の出量）

* 成分$j$:（j成分の蓄積量）＝（j成分の入量）ー（ j成分の出量）＋（ j成分の生成量）ー（ j成分の消費量）

<details><summary>説明</summary>
ここでは、化学工学や環境工学の解析の基本となる物質収支について説明します。<br>
まず、中央の円を「系（システム）」と考えてください。この系の中には物質が存在しており、左側から物質が入量として流入し、右側から出量として流出します。<br>
物質収支の基本的な考え方は、<br>
「系の中にどれだけ物質が蓄積されたか」<br>
を計算することです。<br>
全成分、つまり系全体の物質量について考えると、<br>
蓄積量=入量−出量<br>
となります。<br>
つまり、入ってくる量が出ていく量より多ければ系内の物質は増加し、逆に出ていく量が多ければ系内の物質は減少します。<br>
次に、特定の成分jについて考えてみましょう。<br>
例えば反応器内で化学反応が起こる場合、その成分は反応によって新たに作られたり、逆に消費されたりします。<br>
そのため成分収支は<br>
成分の蓄積量=成分の入量−成分の出量+成分の生成量−成分の消費量<br>
となります。<br>
つまり、成分収支では単なる流入・流出だけでなく、系内部での生成や消費も考慮する必要があるという点が重要です。<br>

今後学ぶ反応器設計や分離操作の解析は、すべてこの物質収支の考え方を基礎として進めていきます。

$$
b=200\times1.10=220\ \mathrm{mol}
$$

$$
c=220\times\frac{79}{21}=827.6\ \mathrm{mol}
$$
となります。

<h5> 3. 表の作成について</h5>

先程の物質収支の表はICE表をつかいました。<br>

せっかく、物質収支の一般式
$$
（j\text{成分の蓄積量}）＝（j\text{成分の入量}）ー（ j\text{成分の出量}）＋（ j\text{成分の生成量}）ー（ j\text{成分の消費量}）\tag{A}
$$
を勉強しましたので、これに合わせた表を示していきます。<br>
回分操作特有の複雑な話です。連続の場合は混乱しません。<br>
◇◇◇<br>
(A)式は回分にも適応させることができますが、その場合は注意が必要です。<br>
回分の場合は流入、流出はありませんので<br>
回分の物質収支式
$$
（j\text{成分の蓄積量}）＝（j\text{成分の生成量}）ー（ j\text{成分の消費量}）
$$
となります。<br>
しかし厳密に言えば、この収支式は基本的に物質量の速度（微分型）になります。<br>
成分 $j$ について書けば
$$
\frac{dN_j}{dt} = F_{j,\text{in}} - F_{j,\text{out}} + R_{j,\text{gen}} - R_{j,\text{cons}}
$$
です。<br>
ここで、

- $N_j$：反応器内の $j$ 成分量 [mol]
- $F_{j,\text{in}}$：流入速度 [mol/s]
- $F_{j,\text{out}}$：流出速度 [mol/s]
- $R_{j,\text{gen}}$：生成速度 [mol/s]
- $R_{j,\text{cons}}$：消費速度 [mol/s]


これを時間で積分したのがICE表と考えてください。


ただし



回分反応器では反応中に流入も流出もありません。

$$
F_{j,\text{in}}=0\\ 
F_{j,\text{out}}=0
$$
したがって
$$
\frac{dN_j}{dt} = R_{j,\text{gen}} - R_{j,\text{cons}}
$$
となります。<br>
つまり、
$$
成分の蓄積速度 = 成分の正味生成速度
$$
です。

◇◇◇

例：A → B

反応
$$
A \xrightarrow{r} B
$$

を回分反応器で行うとき、

- A成分:　Aは消費されるので

$$
\frac{dN_A}{dt}=-rV
$$

- B成分:　Bは生成されるので
$$
\frac{dN_B}{dt}=rV
$$
となります。<br>
ここで
- $r$ ：反応速度 $[\mathrm{mol/(m^3\,s)}]$
- $V$ ：反応液体積 $[\mathrm{m^3}]$

ICEではなく、(A)式を回分に適用した式に準拠した表に整理すると

| 成分  | 蓄積量 \[mol] | 生成量 \[mol] | 消費量 \[mol] |
| --- | ---------: | ---------: | ---------: |

| O₂  |       -180 |          0 |        180 |
| CO₂ |        +90 |         90 |          0 |
| H₂O |       +180 |        180 |          0 |
| N₂  |          0 |          0 |          0 |

のようになり、反応前の仕込み量を書く欄がなく、わかりにくくなります。

このように回分の場合はICE表を作成するほうが、目的に合致することがあります。

```python
# prompt: 100=D+W, 100*0.2=0.95D+0.005W
# の連立方程式を行列で解く

import numpy as np

# 連立方程式を記述
# 100 = D + W
# 20 = 0.95D + 0.005W

# 行列Aとベクトルbを作成
A = np.array([[1, 1], [0.95, 0.005]])
b = np.array([100, 20])

# 連立方程式を解く
x = np.linalg.solve(A, b)

# 解を出力
print("D =", x[0])
print("W =", x[1])
```
