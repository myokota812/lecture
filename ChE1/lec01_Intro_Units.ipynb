# 1. 化学工学とは何か

## 1.1 化学だけでは「工場」を作れない

化学反応そのものを理解することは重要です。

例えば、

$$
A \rightarrow B
$$

という反応について、

- どの温度で反応するか
- どのくらい速く反応するか
- どのような触媒が必要か

を調べるのは化学・反応科学の重要な仕事です。

しかし、工業的に大量のBを製造するには、それだけでは不十分です。

例えば、

- 原料を何 kg/h 供給するか
- 反応器の大きさをいくつにするか
- どのくらいの熱を加えるか
- 圧力損失はどの程度か
- 生成物から未反応原料をどう分離するか
- 原料をどの程度リサイクルするか
- どの程度のエネルギーを消費するか

まで考える必要があります。

このように、**化学現象を工業的なプロセスとして成立させる**ことが化学工学の大きな役割です。

混合 → 反応 → 冷却 → 結晶化 → 固液分離 → 洗浄 → 乾燥 → 包装

となっており、医薬品やファインケミカルに典型的な回分晶析プロセスです。

　皆さんがビーカースケールで目的物質を合成をするときには大まかに

~~~
反応試薬の調整→混合 → 反応 → 分離・生成 →目的物質
~~~
のフローになると思います。化学プラントでも、その流れは同じです。しかし、大量生産が必要な場合はスケールアップをして、製造する必要があります。そのため、化学プラントではひとつひとつの装置が大きくなっております。さらに、規模が大きいので、装置間の運搬は、すべてポンプを備えた配管を通じて行われます。


化学工学では、単に化学反応そのものを扱うだけでなく、以下のような現象や装置設計を対象とします。

- 物質移動（蒸留、吸収、抽出、乾燥など）
- 熱移動（加熱、冷却、熱交換器の設計）
- 反応工学（反応器の設計・運転）
- 流体工学（配管やポンプ内の流れ）
- プロセス設計・制御（プラント全体の最適化）
- 環境・エネルギー工学（排水処理、省エネルギー、CO₂削減など）

## 2.物理量と単位 ## 

化学プロセスを設計・解析する際には、物質収支やエネルギー収支などの計算が欠かせません。そして、それらの計算の基礎となるのが物理量と単位です。

ここからは、質量、長さ、時間、温度、圧力などの物理量と、その表し方である単位について学んでいきます。

### 物理量： 長さ、質量、時間、… ###
### 物理量の表し方： 数値 $\times$単位 ###

<details>
<summary>[説明文]</summary>
すでに高校や教養課程の物理の授業や基礎物理化学の授業で習った人が多いと思いますが、私たちが物質の特性を比較して論ずるとき、共通の物差しが必要になります。ここでいう共通の物差しを単位と言います。
例えば質量を表したいとき、基準になる物差しとして、私たちはキログラム原器というものを用いていました。私たちが物質の質量を議論するとき、私たちはその物質の質量を、1kgの定義の何倍であるかで表現します。このように、全ての物理量には単位があり、基準となる量の何倍という形で表現されます。
すなわち、物理量は単位と数から成り立ちます。例外として、比重や分子量を挙げることができます。これらはいずれも相対値で無次元です。物理量そのものではなく、二つの物理量の比ということです。
</details>

スライドにあるように、軸ラベルの書き方には『物理量 / 単位』とスラッシュで区切る方法（上図）と、『物理量 [単位]』のようにカッコで囲む方法（下図）があります。

結論から言うと、**学術論文や国際標準（IUPACなど）では、上の『物理量 / 単位』という表記が推奨されています**。

なぜなら、数学的には『物理量 ＝ 数値 × 単位』だからです。
例えば、$m = 5\ \text{kg}$ という式があるとします。両辺を $\text{kg}$ で割ると、$m / \text{kg} = 5$ となり、右辺は純粋な『数値（無次元量）』になりますよね。
グラフの軸に並んでいる『1, 2, 3, 4, 5』という数字は単位を持たないただの数値ですので、軸ラベルを『$m / \text{kg}$』と書くことで、『この軸に並んでいる数字は $m$ を $\text{kg}$ で割った数値ですよ』という数学的に正しい意味になります。

日本の教科書や実務では下の角カッコ『$[ \text{kg} ]$』を使う表記も広く使われていますが、国際標準はスラッシュ表記『$m / \text{kg}$』であることを知っておいてください。

---

### 【要点を絞った短めのセリフ】

「グラフや表の軸ラベルの表記についてです。

上は『スラッシュ表記』、下は『カッコ表記』ですが、国際基準や論文などで推奨されるのは上の『スラッシュ表記』です。

物理量は『数値 × 単位』で表されます。軸に書かれている『1, 2, 3…』という数字は無次元の数値なので、物理量 $m$ を単位 $\text{kg}$ で割った『$m / \text{kg}$』という形にすることで、数学的に矛盾のない正しい表記になります。

下の角カッコ表記も一般的によく見かけますが、専門的なレポートや論文ではぜひ上のスラッシュ表記をマスターしてください。」


----
上のグラフをみてください。グラフの横軸は質量ですが、グラフそのものには質量が目盛ってあるわけではなく、数が目盛ってあります。質量を数に変換するためには、単位質量で割る必要があります。すなわち、横軸は質量ではなく、質量を1kgで割った数が目盛ってあるのです。質量を表す量記号は、mと約束されています。mはkgの単位をもっています。横軸に質量、m/kgと書いてあるのは、mをkgで割って数とした測定量を目盛っているという意味になります。

### 物理量の加成性

*   $\underbrace{1\  \text{m} + 3\  \text{cm}}=1.03\  \text{m}$  
同一次元
*   $\underbrace{1\  \text{m} + 3\  \text{kg}}=???$  
異次元

<details>
<summary>[説明文]</summary>
これも当然ですが、単位の異なる物理量に加成性はありません。1 mと3 cmは一見単位が異なるようにもみえますが、同じ物理量ですから足すことは可能です。1 mと5kgを足すことができないのは当たり前ですね。
それでは次に進みましょう。

---
### 次元の一致 ###

~~~
物理量を表す数式では、すべての項の次元は等しくなければならない。
~~~

* 次元一致例:ベルヌーイの定理（エネルギー保存則）
$$
\frac{1}{2}u^2+\frac{P}{\rho} + gZ = \text{constant}
$$
$$
\begin{align*}
u&: \text{流速} [\text{m}\ \text{s}^{-1}] \\
P&: \text{圧力} [\text{Pa}=\text{kg}\ \text{m}^{-1}\ \text{s}^{-2}]\\
\rho&: \text{密度}\  [\text{kg}\ \text{m}^{-3}]\\
g&:\text{重力加速度} \\
Z&: \text{高さ}\  [\text{m}]\\
\end{align*}
$$

<details>
<summary>[説明文]</summary>
先ほど単位の異なる物理量に加成性は成り立たないと言いました。
ここに例示したベルヌーイの定理の式、これは流体の持つエネルギーが流速と圧力および位置エネルギーの３つの項で表され、その総和は一定であることを示す式ですが、３つの項の単位は、すべてm2/s2でそろっています。
したがって、３つの項には加成性が成り立ちます。

---
*   これはどうでしょう？（実験式等）
$$
C_p=28.11 + 0.197 \times 10^{-2} T+0.48 \times 10^{-5} T^2 -1.97 \times 10^{-9} T^3
$$

<details>
<summary>[説明文]</summary>

$T$ の次元は [$\mathrm{K}$]  
$T^2$ の次元は [$\mathrm{K}^2$]  
$T^3$ の次元は [$\mathrm{K}^3$]  

です。

加算される項はすべて同じ次元でなければならないので、

$0.197\times10^{-2}$

は実際には

$0.197\times10^{-2} \frac{[C_p]}{\mathrm{K}}$

という次元を持っています。と考えてください。

同様に

$0.48\times10^{-5}$
は

$0.48\times10^{-5} \frac{[C_p]}{\mathrm{K}^2}$ 

という次元を持っています。

つまり、実験式には、単位的に理解しかねるものが多いですが、

「係数を無次元数とみなすと各項の次元が一致しない」

あるいは

「係数ごとに異なる単位を持たせて無理やり合わせている経験式」
と考えてください。

単位換算の基本的な考え方

図３は同じ量を異なる単位系で表した関係

$$
12X = 10Y
$$

を表しています。

この式から、

$$
1=\frac{12X}{10Y}
=\frac{10Y}{12X}
$$


が成り立ちます。したがって（例えば）

- $200X$を$Y$単位系に変換する場合

$$
\begin{aligned}
200X &= 200X \times \frac{10Y}{12X}\\
     &=\frac{200\times10}{12}Y\\
     &= 166.7Y
\end{aligned}
$$

- $200Y$を$X$単位系に変換する場合


$$
\begin{aligned}
200Y&= 200Y \times \frac{12X}{10Y}\\
&=\frac{200\times12}{10}X\\ 
&= 240X
\end{aligned}
$$

◇◇◇


実際の単位換算への適用

例えば、

$$
1\,\mathrm{ft}
=
304.8\,\mathrm{mm}
$$


が与えられているとき、

$$
1=
\frac{1\,\mathrm{ft}}
{304.8\,\mathrm{mm}}
$$


であるから、

$$
200\,\mathrm{mm}
=
200\,\mathrm{mm}
\times
\frac{1\,\mathrm{ft}}
{304.8\,\mathrm{mm}}
=
0.656\,\mathrm{ft}
$$


となります。

## 3. 次元

### 3.1. 次元とは

次元とは、物理量の本質的な種類を表したものです。

例えば、

| 物理量              | 単位の例      | 次元         |
| ---------------- | --------- | ---------- |
| 長さ (Length)      | m, cm, km | $$L$$      |
| 質量 (Mass)        | kg, g     | $$M$$      |
| 時間 (Time)        | s, min, h | $$T$$      |
| 温度 (Temperature) | K, ℃      | $$\Theta$$ |


単位は表現方法ですが，

- $1\,\mathrm{m}$
- $100\,\mathrm{cm}$
- $0.001\,\mathrm{km}$

は異なる単位で表されていても，すべて長さの次元$L$です。

---

### 3.2. 化学工学で重要な基本次元

化学工学では主に次の基本次元を用います。

| 次元 | 意味 |
|------|------|
| $M$ | 質量 (Mass) |
| $L$ | 長さ (Length) |
| $T$ | 時間 (Time) |
| $\Theta$ | 温度 (Temperature) |
| $N$ | 物質量 (Amount of substance) |

---

### 3.3. 導出次元

多くの物理量は基本次元の組み合わせで表されます。

- 面積：$L^2$
- 体積：$L^3$
- 速度：$LT^{-1}$
- 加速度：$LT^{-2}$
- 密度：$ML^{-3}$
- モル濃度：$NL^{-3}$

---

### 3.4. 次元解析の考え方

物理法則は，式の両辺で次元が一致しなければならないというルールがあります。

### 例1

移動距離$s$, 速度$u$, 時間$t$において

$$
s=ut
$$

左辺の次元は

$$
[s]=L
$$

右辺の次元は

$$
[u][t]
=
(LT^{-1})(T)
=
L
$$

となります。

左右の次元が一致するので，この式は次元的に正しい式です。

### 例2

$$
s=u+t
$$

左辺の次元は

$$
[s]=L
$$

右辺は

$$
[u]+[t]
=
LT^{-1}+T
$$

となります。

異なる次元同士は加算できないため，

$$
LT^{-1}+T
$$

は意味を持ちません。

したがって，

$$
s=u+t
$$

は次元的に誤った式です。

---

### 3.5. 化学工学での利用例

### 物質収支

物質収支の基本式は

$$
\text{入力}
-
\text{出力}
+
\text{生成}
=
\text{蓄積}
$$

です。

各項は

$$
\mathrm{kg/h}
$$

または

$$
\mathrm{mol/s}
$$

など，同じ次元・同じ単位系で揃える必要があります。

### エネルギー収支

エネルギー収支の一例は

$$
Q-W+\sum H_{\mathrm{in}}
=
\sum H_{\mathrm{out}}
$$

です。

各項は

$$
\mathrm{J/s}
$$

または

$$
\mathrm{kW}
$$

などの同一単位で表す必要があります。

### 3.6. 無次元量

次元を持たない量を**無次元量（dimensionless number）**といいます。

### レイノルズ数

$$
Re
=
\frac{\rho uD}{\mu}
$$

各物理量の次元は

$$
[\rho]
=
ML^{-3}
$$

$$
[u]
=
LT^{-1}
$$

$$
[D]
=
L
$$

$$
[\mu]
=
ML^{-1}T^{-1}
$$

であるため，

$$
[Re]
=
\frac{
(ML^{-3})
(LT^{-1})
(L)
}{
ML^{-1}T^{-1}
}
=
1
$$

となります。

したがって，$Re$は無次元量です。

---

### プラントル数（Prandtl number）

**動粘性係数（運動量の拡散）と温度拡散率（熱の拡散）の比**を表す無次元数です。流体内の「運動量の伝わりやすさ」と「熱の伝わりやすさ」の相関を示します。

$$Pr = \frac{C_p \mu}{k} = \frac{\nu}{\alpha}$$

* $Pr$ ： プラントル数（Prandtl number） $[-]$
* $C_p$ ： 定圧比熱 $[\mathrm{J/(kg\cdot K)}]$
* $\mu$ ： 粘性係数 $[\mathrm{Pa\cdot s}]$ または $[\mathrm{kg/(m\cdot s)}]$
* $k$ ： 熱伝導率 $[\mathrm{W/(m\cdot K)}]$
* $\nu$ ： 動粘性係数 $\left( \nu = \frac{\mu}{\rho} \right)$ $[\mathrm{m^2/s}]$
* $\alpha$ ： 温度拡散率（熱拡散率） $\left( \alpha = \frac{k}{\rho C_p} \right)$ $[\mathrm{m^2/s}]$
* $\rho$ ： 流体の密度 $[\mathrm{kg/m^3}]$

---

### シャーウッド数（Sherwood number）

**対流による物質移動と分子拡散による物質移動の比**を表す無次元数です。熱伝達におけるヌセルト数（$Nu$）に相当する無次元パラメータです。

$$Sh = \frac{k_c L}{D}$$

* $Sh$ ： シャーウッド数（Sherwood number） $[-]$
* $k_c$ ： 物質移動係数 $[\mathrm{m/s}]$
* $L$ ： 基準長さ（代表長さ） $[\mathrm{m}]$
* $D$ ： 拡散係数（体積拡散係数） $[\mathrm{m^2/s}]$

## 発展：次元解析によるレイノルズ数（$Re$）の導出

## 1. 導入と目的

流体の流れ（管内流や物体周囲の流れ）を支配する物理現象を解析する際、どのような物理量がどの程度影響しているかをまとめる手法として次元解析（Dimensional Analysis）**があります。
流体力学において最も重要な無次元数の一つである**レイノルズ数（Reynolds Number, $Re$）を、バッキンガムの$\pi$定理を用いて基礎から導出します。

---

## 2. 現象を支配する物理量と次元

流れに関する力学的現象（例えば流体内に置かれた物体にかかる力や流れの状態）は、主に以下の4つの物理量によって決定されると仮定します。

| 物理量 | 記号 | SI単位 | 基本次元 |
| --- | --- | --- | --- |
| 流速 | $U$ | $\text{m/s}$ | $[L T^{-1}]$ |
| 代表長さ | $L$ | $\text{m}$ | $[L]$ |
| 流体の密度 | $\rho$ | $\text{kg/m}^3$ | $[M L^{-3}]$ |
| 流体の粘性係数 | $\mu$ | $\text{Pa}\cdot\text{s}$ ($\text{kg/(m}\cdot\text{s)}$) | $[M L^{-1} T^{-1}]$ |

※ **基本次元**：質量 $[M]$、長さ $[L]$、時間 $[T]$ の3項目。

---

## 3. バッキンガムの $\pi$ 定理の適用

### ステップ 1：物理量の数 $n$ と基本次元の数 $k$ の特定

* 物理量の数： $n = 4$ （$U, L, \rho, \mu$）
* 基本次元の数： $k = 3$ （$[M], [L], [T]$）

定理より、独立した無次元数（$\pi$グループ）の個数は、


$$\text{無次元数の個数} = n - k = 4 - 3 = 1$$


となり、**1つの無次元数 $\pi_1$** が構成できることがわかります。

---

### ステップ 2：繰り返し変数の選定

基本次元 $[M],[L],[T]$ を網羅するように、互いに独立した $k=3$ 個の物理量を繰り返し変数として選択します。
ここでは **$U, L, \rho$** を選択します。

---

### ステップ 3：無次元数 $\pi_1$ の指数方程式の作成

残りの物理量である粘性係数 $\mu$ に対し、指数 $a, b, c$ を用いて無次元量 $\pi_1$ を定義します。

$$\pi_1 = U^a \cdot L^b \cdot \rho^c \cdot \mu^1$$

各物理量を基本次元に置き換えます。

$$\pi_1 = [L T^{-1}]^a \cdot [L]^b \cdot [M L^{-3}]^c \cdot [M L^{-1} T^{-1}]^1$$

同類項をまとめると：

$$\pi_1 = [M]^{c+1} \cdot [L]^{a+b-3c-1} \cdot [T]^{-a-1}$$

---

### ステップ 4：指数の連立方程式と解法

$\pi_1$ が無次元（$[M]^0 [L]^0 [T]^0$）となるために、各次元の指数はすべて 0 になる必要があります。

1. **$[M]$ について：** $c + 1 = 0 \implies \mathbf{c = -1}$
2. **$[T]$ について：** $-a - 1 = 0 \implies \mathbf{a = -1}$
3. **$[L]$ について：** $a + b - 3c - 1 = 0$
* $a, c$ を代入： $(-1) + b - 3(-1) - 1 = 0$
* $-1 + b + 3 - 1 = 0 \implies \mathbf{b = -1}$



したがって、指数は $(a, b, c) = (-1, -1, -1)$ と求まります。

---

## 4. 導出結果と物理的意味

求めた指数を $\pi_1$ の式に代入します。

$$\pi_1 = U^{-1} \cdot L^{-1} \cdot \rho^{-1} \cdot \mu^1 = \frac{\mu}{\rho U L}$$

この無次元量の逆数をとったものを、レイノルズ数（Reynolds Number, $Re$）と定義します。

$$Re = \frac{1}{\pi_1} = \frac{\rho U L}{\mu} = \frac{U L}{\nu}$$

※ $\nu = \frac{\mu}{\rho}$ は動粘性係数 ($\text{m}^2/\text{s}$)。

---

### 補足：レイノルズ数の物理的意味

レイノルズ数は、流れの中での**慣性力**と**粘性力**の比を表しています。

* **慣性力** $\approx \rho U^2 L^2$ （流れが運動を続けようとする力）
* **粘性力** $\approx \mu U L$ （流体の粘りけによるブレーキ力）

$$\frac{\text{慣性力}}{\text{粘性力}} = \frac{\rho U^2 L^2}{\mu U L} = \frac{\rho U L}{\mu} = Re$$

* **$Re$ が小さい場合：** 粘性力が支配的（層流：規則正しく滑らかな流れ）
* **$Re$ が大きい場合：** 慣性力が支配的（乱流：乱れた渦を伴う流れ）
