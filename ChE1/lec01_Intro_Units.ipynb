## 1. 化学工学とは ##

化学はもとより、数学、物理・生物学などの基礎知識を工業生産に応用し, **原料から有用な製品を効率的かつ安全に製造するための学問** です。
　職業としては、 化学製品を作るための工場や設備（化学プラントといいます）を設計するプラントエンジニア、**化学プラント** において、製造条件,コストなどを勘案して化学プロセスを最適化するプロセスエンジニアなどがあります。皆さんの先輩は国内外で活躍しています。

このPFDの主要単位操作は、

混合 → 反応 → 冷却 → 結晶化 → 固液分離 → 洗浄 → 乾燥 → 包装

となっており、医薬品やファインケミカルに典型的な回分晶析プロセスです。

　皆さんがビーカースケールで目的物質を合成をするときには大まかに

~~~
反応試薬の調整→混合 → 反応 → 分離・生成 →目的物質
~~~
のフローになると思います。化学プラントでも、その流れは同じです。しかし、大量生産が必要な場合はスケールアップをして、製造する必要があります。そのため、化学プラントではひとつひとつの装置が大きくなっております。さらに、規模が大きいので、装置間の運搬は、すべてポンプを備えた配管を通じて行われます。


化学工学では、単に化学反応そのものを扱うだけでなく、以下のような現象や装置設計を対象とします。

- 物質移動（蒸留、吸収、抽出、乾燥など）
- 熱移動（加熱、冷却、熱交換器の設計）
- 反応工学（反応器の設計・運転）
- 流体工学（配管やポンプ内の流れ）
- プロセス設計・制御（プラント全体の最適化）
- 環境・エネルギー工学（排水処理、省エネルギー、CO₂削減など）

化学工学の特徴は、特定の化学物質や製品に限定されず、「物質・エネルギー・情報の流れ」を共通原理として扱う点にあります。そのため、化学産業だけでなく、石油・食品・医薬品・半導体・電池・環境分野など幅広い産業で活用されています。

化学工学の目的は、製品を大量に、安全に、低コストで、環境負荷を抑えながら生産することにあります。

それでは、化学工学の基礎を勉強していきましょう。

## 2.物理量と単位 ## 

化学プロセスを設計・解析する際には、物質収支やエネルギー収支などの計算が欠かせません。そして、それらの計算の基礎となるのが物理量と単位です。

ここからは、質量、長さ、時間、温度、圧力などの物理量と、その表し方である単位について学んでいきます。

### 物理量： 長さ、質量、時間、… ###
### 物理量の表し方： 数値 $\times$単位 ###

<details>
<summary>[説明文]</summary>
すでに高校や教養課程の物理の授業や基礎物理化学の授業で習った人が多いと思いますが、私たちが物質の特性を比較して論ずるとき、共通の物差しが必要になります。ここでいう共通の物差しを単位と言います。
例えば質量を表したいとき、基準になる物差しとして、私たちはキログラム原器というものを用いていました。私たちが物質の質量を議論するとき、私たちはその物質の質量を、1kgの定義の何倍であるかで表現します。このように、全ての物理量には単位があり、基準となる量の何倍という形で表現されます。
すなわち、物理量は単位と数から成り立ちます。例外として、比重や分子量を挙げることができます。これらはいずれも相対値で無次元です。物理量そのものではなく、二つの物理量の比ということです。
</details>

続いて、グラフや表における物理量の表記方法について確認します。

スライドにあるように、軸ラベルの書き方には『物理量 / 単位』とスラッシュで区切る方法（上図）と、『物理量 [単位]』のようにカッコで囲む方法（下図）があります。

結論から言うと、**学術論文や国際標準（IUPACなど）では、上の『物理量 / 単位』という表記が推奨されています**。

なぜなら、数学的には『物理量 ＝ 数値 × 単位』だからです。
例えば、$m = 5\ \text{kg}$ という式があるとします。両辺を $\text{kg}$ で割ると、$m / \text{kg} = 5$ となり、右辺は純粋な『数値（無次元量）』になりますよね。
グラフの軸に並んでいる『1, 2, 3, 4, 5』という数字は単位を持たないただの数値ですので、軸ラベルを『$m / \text{kg}$』と書くことで、『この軸に並んでいる数字は $m$ を $\text{kg}$ で割った数値ですよ』という数学的に正しい意味になります。

日本の教科書や実務では下の角カッコ『$[ \text{kg} ]$』を使う表記も広く使われていますが、国際標準はスラッシュ表記『$m / \text{kg}$』であることを知っておいてください。

---

### 【要点を絞った短めのセリフ】

「グラフや表の軸ラベルの表記についてです。

上は『スラッシュ表記』、下は『カッコ表記』ですが、国際基準や論文などで推奨されるのは上の『スラッシュ表記』です。

物理量は『数値 × 単位』で表されます。軸に書かれている『1, 2, 3…』という数字は無次元の数値なので、物理量 $m$ を単位 $\text{kg}$ で割った『$m / \text{kg}$』という形にすることで、数学的に矛盾のない正しい表記になります。

下の角カッコ表記も一般的によく見かけますが、専門的なレポートや論文ではぜひ上のスラッシュ表記をマスターしてください。」


----
上のグラフをみてください。グラフの横軸は質量ですが、グラフそのものには質量が目盛ってあるわけではなく、数が目盛ってあります。質量を数に変換するためには、単位質量で割る必要があります。すなわち、横軸は質量ではなく、質量を1kgで割った数が目盛ってあるのです。質量を表す量記号は、mと約束されています。mはkgの単位をもっています。横軸に質量、m/kgと書いてあるのは、mをkgで割って数とした測定量を目盛っているという意味になります。

### 物理量の加成性

*   $\underbrace{1\  \text{m} + 3\  \text{cm}}=1.03\  \text{m}$  
同一次元
*   $\underbrace{1\  \text{m} + 3\  \text{kg}}=???$  
異次元

<details>
<summary>[説明文]</summary>
これも当然ですが、単位の異なる物理量に加成性はありません。1 mと3 cmは一見単位が異なるようにもみえますが、同じ物理量ですから足すことは可能です。1 mと5kgを足すことができないのは当たり前ですね。
それでは次に進みましょう。

---
### 次元の一致 ###

~~~
物理量を表す数式では、すべての項の次元は等しくなければならない。
~~~

* 次元一致例:ベルヌーイの定理（エネルギー保存則）
$$
\frac{1}{2}u^2+\frac{P}{\rho} + gZ = \text{constant}
$$
$$
\begin{align*}
u&: \text{流速} [\text{m}\ \text{s}^{-1}] \\
P&: \text{圧力} [\text{Pa}=\text{kg}\ \text{m}^{-1}\ \text{s}^{-2}]\\
\rho&: \text{密度}\  [\text{kg}\ \text{m}^{-3}]\\
g&:\text{重力加速度} \\
Z&: \text{高さ}\  [\text{m}]\\
\end{align*}
$$

<details>
<summary>[説明文]</summary>
先ほど単位の異なる物理量に加成性は成り立たないと言いました。
ここに例示したベルヌーイの定理の式、これは流体の持つエネルギーが流速と圧力および位置エネルギーの３つの項で表され、その総和は一定であることを示す式ですが、３つの項の単位は、すべてm2/s2でそろっています。
したがって、３つの項には加成性が成り立ちます。

---
*   これはどうでしょう？（実験式等）
$$
C_p=28.11 + 0.197 \times 10^{-2} T+0.48 \times 10^{-5} T^2 -1.97 \times 10^{-9} T^3
$$

<details>
<summary>[説明文]</summary>

$T$ の次元は [$\mathrm{K}$]  
$T^2$ の次元は [$\mathrm{K}^2$]  
$T^3$ の次元は [$\mathrm{K}^3$]  

です。

加算される項はすべて同じ次元でなければならないので、

$0.197\times10^{-2}$

は実際には

$0.197\times10^{-2} \frac{[C_p]}{\mathrm{K}}$

という次元を持っています。と考えてください。

同様に

$0.48\times10^{-5}$
は

$0.48\times10^{-5} \frac{[C_p]}{\mathrm{K}^2}$ 

という次元を持っています。

つまり、実験式には、単位的に理解しかねるものが多いですが、

「係数を無次元数とみなすと各項の次元が一致しない」

あるいは

「係数ごとに異なる単位を持たせて無理やり合わせている経験式」
と考えてください。

単位換算の基本的な考え方

図３は同じ量を異なる単位系で表した関係

$$
12X = 10Y
$$

を表しています。

この式から、

$$
1=\frac{12X}{10Y}
=\frac{10Y}{12X}
$$


が成り立ちます。したがって（例えば）

- $200X$を$Y$単位系に変換する場合

$$
\begin{aligned}
200X &= 200X \times \frac{10Y}{12X}\\
     &=\frac{200\times10}{12}Y\\
     &= 166.7Y
\end{aligned}
$$

- $200Y$を$X$単位系に変換する場合


$$
\begin{aligned}
200Y&= 200Y \times \frac{12X}{10Y}\\
&=\frac{200\times12}{10}X\\ 
&= 240X
\end{aligned}
$$

◇◇◇


実際の単位換算への適用

例えば、

$$
1\,\mathrm{ft}
=
304.8\,\mathrm{mm}
$$


が与えられているとき、

$$
1=
\frac{1\,\mathrm{ft}}
{304.8\,\mathrm{mm}}
$$


であるから、

$$
200\,\mathrm{mm}
=
200\,\mathrm{mm}
\times
\frac{1\,\mathrm{ft}}
{304.8\,\mathrm{mm}}
=
0.656\,\mathrm{ft}
$$


となります。

### 例3: 気体定数の換算 ###

> $R$=8.314 $\text{J}\cdot\text{K}^{-1} \text{mol}^{-1}$の単位を$\text{L}\cdot\text{atm}\cdot\text{K}^{-1} \cdot \text{mol}^{-1}$に変換。

>1 atm =$1.01325 \times 10^{5} \text{Pa}$とする。

<details>
<summary>解答例</summary>

$$
R=8.314\  \text{J}\cdot\text{K}^{-1} \text{mol}^{-1} =8.314\  \text{Pa}\cdot\ \text{m}^3 \cdot \text{K}^{-1} \text{mol}^{-1} \times \frac{1}{1.01325 \times 10^5\text{Pa}} \times \frac{1000 \text{L}}{1 \text{m}^3}=0.08205 \text{L}\cdot\text{atm}\cdot\text{K}^{-1}\cdot\text{mol}^{-1}
$$

</details>

1.   $1 \text{ft} =304.8\, \text{mm}$が与えられている場合、$200\,\text{mm}$は何$\text{ft}$か。
2.   $1 \text{cal}=4.18\,\text{J}$が与えられている場合、$100\,\text{kJ}$は何$\text{kcal}$か。
3.   $1 \mathrm{m}^3 = 10^6\, \text{cm}^3$が与えられている場合、$20 \,\text{cm}^3$は何$\text{m}^3$か。






<details>
<summary>解答例</summary>

1. $200\,\mathrm{mm}$ は何 $\mathrm{ft}$ か

与えられている換算関係は

$$
1\,\mathrm{ft}=304.8\,\mathrm{mm}
$$

である。$200\,\mathrm{mm}$ を $\mathrm{ft}$ に変換するため，換算係数を用いて

$$
200\,\mathrm{mm}
=
200\,\mathrm{mm}
\times
\frac{1\,\mathrm{ft}}{304.8\,\mathrm{mm}}
=
0.656\,\mathrm{ft}
$$
---

2. $100\,\mathrm{kJ}$ は何 $\mathrm{kcal}$ か

$$
100\,\mathrm{kJ}
=
100\,\mathrm{kJ}
\times
\frac{1\,\mathrm{kcal}}{4.18\,\mathrm{kJ}}
=23.9\,\mathrm{kcal}
$$


---

3. $20\,\mathrm{cm^3}$ は何 $\mathrm{m^3}$ か


$$
20\,\mathrm{cm^3}
=
20\,\mathrm{cm^3}
\times
\frac{1\,\mathrm{m^3}}{10^6\,\mathrm{cm^3}}
=\frac{20}{10^6}\,\mathrm{m^3}
=2.0\times10^{-5}\,\mathrm{m^3}
$$

である。

</details>

## 3. 次元

### 3.1. 次元とは

次元とは、物理量の本質的な種類を表したものです。

例えば、

| 物理量              | 単位の例      | 次元         |
| ---------------- | --------- | ---------- |
| 長さ (Length)      | m, cm, km | $$L$$      |
| 質量 (Mass)        | kg, g     | $$M$$      |
| 時間 (Time)        | s, min, h | $$T$$      |
| 温度 (Temperature) | K, ℃      | $$\Theta$$ |


単位は表現方法ですが，

- $1\,\mathrm{m}$
- $100\,\mathrm{cm}$
- $0.001\,\mathrm{km}$

は異なる単位で表されていても，すべて長さの次元$L$です。

---

### 3.2. 化学工学で重要な基本次元

化学工学では主に次の基本次元を用います。

| 次元 | 意味 |
|------|------|
| $M$ | 質量 (Mass) |
| $L$ | 長さ (Length) |
| $T$ | 時間 (Time) |
| $\Theta$ | 温度 (Temperature) |
| $N$ | 物質量 (Amount of substance) |

---

### 3.3. 導出次元

多くの物理量は基本次元の組み合わせで表されます。

- 面積：$L^2$
- 体積：$L^3$
- 速度：$LT^{-1}$
- 加速度：$LT^{-2}$
- 密度：$ML^{-3}$
- モル濃度：$NL^{-3}$

---

### 3.4. 次元解析の考え方

物理法則は，式の両辺で次元が一致しなければならないというルールがあります。

### 例1

$$
s=ut
$$

左辺の次元は

$$
[s]=L
$$

右辺の次元は

$$
[u][t]
=
(LT^{-1})(T)
=
L
$$

となります。

左右の次元が一致するので，この式は次元的に正しい式です。

### 例2

$$
s=u+t
$$

左辺の次元は

$$
[s]=L
$$

右辺は

$$
[u]+[t]
=
LT^{-1}+T
$$

となります。

異なる次元同士は加算できないため，

$$
LT^{-1}+T
$$

は意味を持ちません。

したがって，

$$
s=u+t
$$

は次元的に誤った式です。

---

### 3.5. 化学工学での利用例

### 物質収支

物質収支の基本式は

$$
\text{入力}
-
\text{出力}
+
\text{生成}
=
\text{蓄積}
$$

です。

各項は

$$
\mathrm{kg/h}
$$

または

$$
\mathrm{mol/s}
$$

など，同じ次元・同じ単位系で揃える必要があります。

### エネルギー収支

エネルギー収支の一例は

$$
Q-W+\sum H_{\mathrm{in}}
=
\sum H_{\mathrm{out}}
$$

です。

各項は

$$
\mathrm{J/s}
$$

または

$$
\mathrm{kW}
$$

などの同一単位で表す必要があります。

---

### 3.6. 無次元量

次元を持たない量を

**無次元量（dimensionless number）**

といいます。

### レイノルズ数

$$
Re
=
\frac{\rho uD}{\mu}
$$

各物理量の次元は

$$
[\rho]
=
ML^{-3}
$$

$$
[u]
=
LT^{-1}
$$

$$
[D]
=
L
$$

$$
[\mu]
=
ML^{-1}T^{-1}
$$

であるため，

$$
[Re]
=
\frac{
(ML^{-3})
(LT^{-1})
(L)
}{
ML^{-1}T^{-1}
}
=
1
$$

となります。

したがって，

$$
Re
$$

は無次元量です。

### プラントル数

$$
Pr
=
\frac{C_p\mu}{k}
$$

### シャーウッド数

$$
Sh
=
\frac{k_cL}{D}
$$

これらの無次元数は，

- 流れ
- 伝熱
- 物質移動

の現象を整理し，相関式を構築するための重要な指標です。
