### ＢＦＤの概説

図に示す連続発酵プロセスについて考える。

原料としてスクロース供給流（ストリーム1）、水供給流（ストリーム2）および酵母供給流（ストリームY）が混合器へ供給される。

**混合器**では反応は起こらず、各供給流は完全に混合された後、反応器へ送られる（ストリーム3）。

**反応器**では酵母の作用により、スクロースが発酵してエタノールと二酸化炭素を生成する。反応器内では次の反応のみが進行するものとする。

$$
\mathrm{C_{12}H_{22}O_{11}+H_2O 
\rightarrow 4C_2H_5OH+4CO_2}
$$


反応器出口流（ストリーム4）は蒸留塔へ送られる。一方、生成した二酸化炭素は気体としてベント流（ストリーム7）からすべて排出されるものとする。

**蒸留塔**ではエタノールの分離・回収を行う。塔頂からはエタノールを主成分とする製品流（ストリーム6）が得られ、塔底からは未反応スクロース、水、酵母および回収されなかったエタノールを含む缶底液（ストリーム5）が排出される。

#### 仮定

本プロセスについて、以下の仮定を置く。

- 1.プロセス全体は定常状態で運転されている。したがって各成分について蓄積はゼロであり、入力+生成=出力+消費の物質収支が成り立つ     
- 2.混合器および蒸留塔内では化学反応は起こらない。    
- 3.発酵反応は反応器内でのみ進行する。    
- 4.酵母は反応に関与しない不活性成分として扱う。  
- 5.生成した二酸化炭素はすべてストリーム7から排出される。     
- 6.スクロースおよび酵母は蒸留塔中で揮発しない。

必要に応じて転化率 $X$、反応進行度流量 $\dot{\xi}$、エタノール回収率 $R$、留出液中のエタノール質量分率 $x_E$ を用いる。

### 記号一覧

| 記号 | 意味 | 値・単位 |
|--------|--------|--------|
| $\dot{m}_{j,i}$ | ストリーム $j$ 中の成分 $i$ の質量流量 | kg/h |
| $\dot{m}_{j}$ | ストリーム $j$ の全質量流量 | kg/h |
| $\dot{m}_Y$ | 混合器へ供給される酵母流量（ストリームy） | kg/h |
| $\dot{\xi}$ | 発酵反応の反応進行度流量 | mol/h |
| $X$ | 反応器におけるスクロース転化率<br>$X=\frac{\dot{m}_{3,S}-\dot{m}_{4,S}}{\dot{m}_{3,S}}=\dot m_{3,S}-\dot{\xi} M_S$ | 0.95 |
| $f_Y$ | 酵母添加率 | 0.05 |
| $R$ | エタノール回収率$=\frac{\dot{m}_{6E}}{\dot{m}_{4E}}$ | 0.99 |
| $x_E$ | 留出液(ストリーム6)中のエタノール質量分率<br>$x_E=\frac{\dot m_{6,E}}{\dot m_{6,E} + \dot m_{6,W}}$ | 0.956 |
| $M_S$ | スクロースの分子量 | 342.30 g/mol |
| $M_W$ | 水の分子量 | 18.015 g/mol |
| $M_E$ | エタノールの分子量 | 46.07 g/mol |
| $M_C$ | 二酸化炭素の分子量 | 44.01 g/mol |

## 問題 

### 問1

発酵(反応）器出口ストリーム4について、物質収支をとり、

$\dot{m}_{4S},\quad \dot{m}_{4W},\quad \dot{m}_{4E},\quad \dot{m}_{4C}$

を$\dot{m}_{1S},X$(および分子量$M_S, M_W, M_E, M_C$のいずれか）を用いて表せ。

問2

蒸留塔について、

$\dot{m}_{6E}$
および
$\dot{m}_{6W}$

を

$\dot{m}_{1S},\dot{m}_{2W},X,R,x_E$

を用いて表せ。

問3

次の条件で各流量を求めよ。

$\dot{m}_{1S}=3750\ \rm{kg/h}$

$\dot{m}_{2W}=4400\ \rm{kg/h}$

$ X=0.90, \quad R=0.95, \quad x_E=0.95$

### 解説・解答例

問1
* 混合器での物質収支    
　混合器内では反応は起こらないため、各成分はそのまま通過する。

$$
\begin{aligned}
\dot m_{3,S}&=\dot m_{1,S}\\
\dot m_{3,W}&=\dot m_{2,W}\\
\dot m_{3,Y}&=\dot m_Y
\end{aligned}
$$
* 反応器での物質収支
    - スクロース
    $$
    \begin{aligned}
    \dot{m}_{4,S} &= \dot{m}_{3,S}-X \dot{m}_{3,S}\\
    &= (1-X) \dot{m}_{1,S}
    \end{aligned}
    $$
    - 水

    反応式より
    $M_S=342\ {\rm kg}$のショ糖に対し$M_W=18\ {\rm kg}$の水が消費される。
    したがって
    $$\dot{m}_{4,W} = \dot{m}_{2,W} - \frac{M_W}{M_S}Xm_{1,S}$$

    - エタノール
    
    反応式より $M_S=342\ {\rm kg}$のショ糖から$4\times M_E\ {\rm kg}$のエタノールが生成する。

    したがって

    $$\dot{m}_{4,E} = 4\frac{M_E}{M_S}Xm_{1,S}$$

    - 二酸化炭素

    $$\dot{m}_{4,C} = 4 \times \frac{M_C}{M_S}Xm_{1,S}$$

問2

- エタノール回収率

    定義より
    $$R = \frac{\dot{m}_{6,E}}{\dot{m}_{4,E}}$$

    したがって

$$
\boxed{ \dot{m}_{6,E} = 4RX \frac{M_E}{M_S} \dot{m}_{1,S}} 
$$

- 留出液組成

    定義より

$$
x_E = \frac{\dot{m}_{6,E}} {\dot{m}_{6,E}+\dot{m}_{6,W}}
$$

これを変形すると
$$
\dot{m}_{6,W} = \frac{1-x_E}{x_E} \dot{m}_{6,E}
$$
したがって

$$
\boxed{ \dot{m}_{6,W} = 4RX\frac{1-x_E}{x_E}  \left( \frac{M_E}{M_S} m_{1S} \right) }
$$

### 各単位操作における成分物質収支
#### 混合器 (Mixer)
混合器では、

- スクロース供給流(ストリーム1)
- 水供給流(ストリーム2)
- 酵母供給流(ストリームy)

を混合する。

混合器内では反応は起こらないため、各成分はそのまま通過する。

$$
\begin{aligned}
\dot m_{3,S}&=\dot m_{1,S}\\
\dot m_{3,W}&=\dot m_{2,W}\\
\dot m_{3,Y}&=\dot m_Y
\end{aligned}
$$


なお酵母供給量は、スクロースおよび水の供給量の合計に対する一定割合として与える。
$$
\dot m_Y=f_Y(\dot m_{1,S}+\dot m_{2,W})
$$
◇◇◇

### 反応器 (Reactor)
スクロースを限定物質とし、転化率（反応率）を $X$ とする。
$$
X= \frac{\dot{\xi}M_s}{\dot{m}_{3S}}
$$
反応進行度速度は
$$
\dot{\xi}=X\frac{\dot m_{3,S}}{M_S}
$$
で与えられる。

反応式
$$
\mathrm{C_{12}H_{22}O_{11}+H_2O 
\rightarrow 4C_2H_5OH+4CO_2}
$$
より、

スクロース 1 mol 消費   
水 1 mol 消費   
エタノール 4 mol 生成       
二酸化炭素 4 mol 生成   

である。

したがって反応器出口の成分流量は

- スクロース
$$
\dot m_{4,S}=\dot m_{3,S}-\dot{\xi} M_S
$$
もしくは転化率$X$を用いると
$$
\dot m_{4,S}=(1-X)\dot m_{3,S}
$$
- 水
$$
\dot m_{4,W}=\dot m_{3,W}-\dot{\xi} M_W
$$
- エタノール
$$
\dot m_{4,E}=4\dot{\xi} M_E
$$
- 二酸化炭素
$$
\dot m_{7,C}=4\dot{\xi} M_C
$$
- 酵母
$$
\dot m_{4,Y}=\dot m_{3,Y}
$$
酵母は不活性成分として扱い、反応には関与しない。

また、生成した二酸化炭素はすべてベント流 S7 から排出されるものとする。


***
### 分離器（蒸留塔） (Distillation Column)
蒸留塔では、低沸点成分に富む凝縮液が塔頂（留出液：ストリーム6)から、高沸点成分に富む凝縮液が塔底部（缶出液：ストリーム5）から、それぞれ排出される。

#### 物質収支

$$
\begin{aligned}
 \dot{m}_{4S} &= \dot{m}_{5S} \\ 
 \dot{m}_{4W} &= \dot{m}_{5W} + \dot{m}_{6W} \\ 
 \dot{m}_{4E} &= \dot{m}_{5E} + \dot{m}_{6E} 
\end{aligned}
$$

- エタノール回収率 $R$
$$
R= \frac{\dot m_{6,E}}{\dot m_{4,E}}
$$
- 留出液中のエタノール質量分率 $x_E$​
$$
x_E= \frac{\dot m_{6,E}}{\dot m_{6,E} + \dot m_{6,W}}
$$

によって規定される。

最終的に求めたいのは、
$\boxed{\dot m_{6,E}}$と$\boxed{\dot m_{6,W}}$

したがって、
$$
\begin{aligned}
\dot m_{6,E}&=R \dot m_{4,E}\\
\dot m_{6,W}&=\frac{1-x_E}{x_E}R\dot m_{4,E}
\end{aligned}
$$

◇◇◇

$$
\dot m_{4,E}=4\dot{\xi} M_E
$$

反応進行度速度は
$$
\dot{\xi}=X\frac{\dot m_{3,S}}{M_S}
$$
だから

$$
\dot m_{4,E}=4X\frac{M_E}{M_S} \dot m_{1,S}
$$


最終的に

$$
\begin{aligned}
\dot m_{6,E}&=4RX\frac{M_E}{M_S} \dot m_{1,S}\\
\dot m_{6,W}&=4\frac{1-x_E}{x_E}RX\frac{M_E}{M_S} \dot m_{1,S}
\end{aligned}
$$

演習問題
発酵プロセスの物質収支

ショ糖水溶液を発酵させてエタノールを製造し、その後蒸留塔でエタノールを回収するプロセスを考える。

発酵反応は

$$
\mathrm{C_{12}H_{22}O_{11}+H_2O \rightarrow 4C_2H_5OH+4CO_2}
$$

で表される。

蒸留塔では、

- エタノール回収率：$R$
- 留出液中エタノール質量分率：$x_E$

によって操作条件が規定される。

与えられる変数
$m_{1S},\quad m_{2W},\quad X,\quad R,\quad x_E$

ここで

$m_{1S}$：ショ糖供給量 [kg/h]       
$m_{2W}$：水供給量 [kg/h]       
$X$：ショ糖転化率 [-]       
$R$：エタノール回収率 [-]       
$x_E$：留出液中エタノール質量分率 [-]       

である。

また、

$M_S=342,\quad M_W=18,\quad M_E=46,\quad M_C=44$

とする。

問1

発酵器出口流れ4について、

$\dot{m}_{4S},\quad \dot{m}_{4W},\quad \dot{m}_{4E},\quad \dot{m}_{4C}$

を求める式を導け。

問2

蒸留塔について、

$\dot{m}_{6E}$
および
$\dot{m}_{6W}$

を

$\dot{m}_{1S},\dot{m}_{2W},X,R,x_E$

を用いて表せ。

問3

次の条件で各流量を求めよ。

$\dot{m}_{1S}=3750\ \rm{kg/h}$

$\dot{m}_{2W}=4400\ \rm{kg/h}$

$ X=0.90, \quad R=0.95, \quad x_E=0.95$



---
### 解説・解答例

問1

#### ショ糖収支

未反応ショ糖は

$$\dot{m}_{4S} = (1-X) \dot{m}_{1S}$$

#### 水収支

反応式より$M_S=342\ {\rm kg}$のショ糖に対し$M_W=18\ {\rm kg}$の水が消費される。

したがって

$$\dot{m}_{4W} = \dot{m}_{2W} - \frac{M_W}{M_S}Xm_{1S}$$

エタノール生成量

反応式より

$342\ {\rm kg}$

のショ糖から

$4\times 46 = 184\ {\rm kg}$

のエタノールが生成する。

したがって

$\dot{m}_{4E} = \frac{184}{342}Xm_{1S}$

二酸化炭素生成量

同様に

$4\times 44 = 176\ {\rm kg}$

より

$\dot{m}_{4C} = \frac{176}{342}Xm_{1S}$

問2
エタノール回収率

定義より

$$R = \frac{\dot{m}_{6E}}{\dot{m}_{4E}}$$

したがって

$$
\boxed{ \dot{m}_{6E} = R \left( \frac{184}{342} Xm_{1S} \right) }
$$

留出液組成

定義より

$$
x_E = \frac{\dot{m}_{6E}} {\dot{m}_{6E}+\dot{m}_{6W}}
$$

これを変形すると
$$
\dot{m}_{6W} = \frac{1-x_E}{x_E} \dot{m}_{6E}
$$
したがって

$$
\boxed{ \dot{m}_{6W} = \frac{1-x_E}{x_E} R \left( \frac{184}{342} Xm_{1S} \right) }
$$

学生への確認事項

- 転化率 XX が増加するとエタノール生成量はどう変化するか。
- 回収率 RR を高くすると缶出液中のエタノール量はどう変化するか。
- 留出液組成 xEx_E を高くすると留出液中の水量 m6Wm_{6W} はどう変化するか。
- 蒸留塔の自由度はなぜ RR と xEx_E の2つで規定できるのか。

これらを考察させると、自由度解析と物質収支の理解が深まります。
