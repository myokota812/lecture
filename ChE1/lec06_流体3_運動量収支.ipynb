# 直管の圧力損失と流体の基礎

## 1. 層流の場合：ハーゲン・ポアズイユの式の導出

円形断面を持つ管の中を、流体が乱れなくスムーズに流れる状態を層流（Laminar Flow）**と呼びます。この層流において、圧力差がどのように流量を生み出すかを理論的に表したのが**ハーゲン・ポアズイユの式（Hagen-Poiseuille equation）です。

### ① 導出のステップ（微小リングの力のつりあい）

半径 $R$ の円管の中心軸を $z$ 軸とし、中心からの距離を $r$ とします。管内を流れる微小な円筒状の流体要素（半径 $r$、長さ $dz$）を考えます。

定常流において、この流体要素に働く力はつりあっています。

1. **圧力による力**：

$$(p - (p + dp)) \cdot \pi r^2 = -\frac{dp}{dz} dz \cdot \pi r^2$$


2. **粘性摩擦力（せん断応力 $\tau$ による力）**：

$$\tau \cdot 2\pi r \, dz$$



これらがつりあうため、次式が成り立ちます。


$$-\frac{dp}{dz} \pi r^2 = 2\pi r \tau \implies \tau = -\frac{r}{2} \frac{dp}{dz}$$

### ② 速度分布の導出

ニュートンの粘性法則（$\tau = -\mu \frac{dv}{dr}$、$\mu$ は粘度）を代入すると、


$$\mu \frac{dv}{dr} = \frac{r}{2} \frac{dp}{dz} \implies \frac{dv}{dr} = \frac{1}{2\mu} \frac{dp}{dz} r$$

管壁（$r = R$）で速度がゼロ（$v = 0$）になる境界条件を用いて積分すると、管内の流速分布 $v(r)$ が放物線を描くことが分かります。


$$v(r) = \frac{1}{4\mu} \left( -\frac{dp}{dz} \right) (R^2 - r^2)$$


さらに
管の長さが $L$ で、入口の圧力が $P_1$、出口の圧力が $P_2$ であるとき、圧力は管の軸方向（$z$ 方向に沿って）直線的に降下するとみなせます。このとき、一定の圧力勾配 $-\frac{dp}{dz}$ は全体の圧力差 $\Delta P = P_1 - P_2$ を用いて次のように表されます。

$$-\frac{dp}{dz} = \frac{P_1 - P_2}{L} = \frac{\Delta P}{L}$$

これを先ほどの速度分布の式 $v(r)$ に代入します。

$$v(r) = \frac{1}{4\mu} \frac{\Delta P}{L} (R^2 - r^2)$$

この式のカッコ内を $R^2$ でくくると


$$v(r) = \frac{R^2 \Delta P}{4\mu L} \left( 1 - \frac{r^2}{R^2} \right) \tag{1}$$

### 💡 この形で表すメリット

* **中心流速との関係がわかりやすい**：
管の中心（$r = 0$）のとき、後半のカッコ内が $(1 - 0) = 1$ となるため、中心流速 $v_{\max}$ は次のようにシンプルに表せます。

$$v_{\max} = \frac{R^2 \Delta P}{4\mu L}$$



これを用いると、全体の速度分布は $v(r) = v_{\max} \left( 1 - \frac{r^2}{R^2} \right)$ と綺麗な放物線としてスッキリ表現できます。

* **物理的意味**：この式から、管の中心（$r = 0$）で流速が最大になり、管壁（$r = R$）に近づくにつれて速度が 0 になる美しい放物線状の速度分布（Poiseuille velocity profile）を描いていることがわかります。

### 2. 体積流量 $Q$ を求めるための積分

管の断面全体を流れる総流量（体積流量 $Q$）を求めるには、中心から管壁までの微小なリング状の面積 $dA = 2\pi r \, dr$ について、速度 $v(r)$ を全断面で積分（足し合わせ）します。

$$Q = \int_{0}^{R} v(r) \cdot (2\pi r \, dr)$$

ここに先ほどの $v(r)$ の式を代入します。

$$Q = \int_{0}^{R} \frac{1}{4\mu} \frac{\Delta P}{L} (R^2 - r^2) \cdot 2\pi r \, dr$$

---

### 3. 積分の計算

定数（変化しない項）を積分の外に出します。

$$Q = \frac{2\pi}{4\mu} \frac{\Delta P}{L} \int_{0}^{R} (R^2 r - r^3) \, dr$$

カッコの中をそれぞれ $r$ について積分します（$\int R^2 r \, dr = \frac{1}{2}R^2 r^2$、$\int r^3 \, dr = \frac{1}{4}r^4$）。

$$Q = \frac{\pi \Delta P}{2\mu L} \left[ \frac{1}{2} R^2 r^2 - \frac{1}{4} r^4 \right]_{0}^{R}$$

上限 $R$ と下限 $0$ を代入して計算します。  

$$
\begin{aligned}
Q &= \frac{\pi \Delta P}{2\mu L} \left( \frac{1}{2} R^4 - \frac{1}{4} R^4 \right)\\
&= \frac{\pi R^4 \Delta P}{8\mu L}
\end{aligned} \tag{2}$$




---

### 結論：ハーゲン・ポアズイユの式

こうして、圧力勾配が一定の定常層流において、流量 $Q$ が次式で表されるという**ハーゲン・ポアズイユの式**が完全に導かれます。

$$Q = \frac{\pi R^4 \Delta P}{8\mu L}$$


* **物理的意味**：流量 $Q$ は管の**半径の4乗 ($R^4$)** に比例するため、管が少し細くなるだけで流量が激減し、同じ流量を流すためには膨大な圧力差が必要になります。

---

## 2. 一般：ダルシー・ワイスバッハの式

実際の配管内では、層流だけでなく、渦を伴う乱流（Turbulent Flow）**が発生することが多くあります。層流・乱流を問わず、あらゆる直管の圧力損失を統一的に評価できるように拡張されたのが**ダルシー・ワイスバッハの式（Darcy-Weisbach equation）です。

### ① 基本方程式

長さ $L$、直径 $D$（$= 2R$）の直管において、流体の密度を $\rho$、平均流速を $v$ とするとき、発生する圧力損失 $\Delta P$ は次式で表されます。

$$\Delta P = f \cdot \frac{L}{D} \cdot \frac{\rho v^2}{2}$$

* $\Delta P$：管摩擦による圧力損失 $[\mathrm{Pa}]$
* $f$：**管摩擦係数** $[-]$
* $L$：配管長 $[\mathrm{m}]$
* $D$：管の内径 $[\mathrm{m}]$
* $\rho$：流体の密度 $[\mathrm{kg/m^3}]$
* $v$：平均流速 $[\mathrm{m/s}]$

### ② 各領域における管摩擦係数 $f$ の決まり方

ダルシー・ワイスバッハの式における肝は「管摩擦係数 $f$」であり、これは**レイノルズ数 $\text{Re} = \frac{\rho v D}{\mu}$** および管内面の粗さ（ざらざら度）によって変化します。

1. **層流領域 ($\text{Re} < 2300$)**：
ハーゲン・ポアズイユの式とダルシー・ワイスバッハの式を比較することで、理論的に $f$ が一意に定まります。

$$f = \frac{64}{\text{Re}}$$


2. **乱流領域 ($\text{Re} \ge 2300$)**：
慣性力が支配的となり、管壁の粗さ（絶対粗度 $\varepsilon$）も影響を与えるため、コールブルークの式（Colebrook's equation）やその近似式（スワミ・ジェインの式など）を用いて算出します。

$$f = \left[ -1.875 \log_{10}\left( \frac{\varepsilon / D}{3.7} + \frac{5.74}{\text{Re}^{0.9}} \right) \right]^{-2}$$



### ③ まとめ

* **層流**では粘性が支配的であり、ハーゲン・ポアズイユの式によってきれいに記述されます。
* **乱流・一般の流路**においては、管の摩擦や速度の二乗に比例する損失を考慮した**ダルシー・ワイスバッハの式**を用いることで、プラントや空調、給水配管など実務上のあらゆる圧力損失を正確に見積もることができます。

## 2. 微小シェル要素の設定
ここで考えてもらいたい運動量は３種類あります。

- ①流体の流れによる運動量（運動量ですので(質量)$\times$ (速度)がなじみがあると思いますが、対象とする流体は連続体のため、質量の代わりに密度を用います。)
- ②粘性による運動量の移動
- ③圧力による運動量の移動

丸数字は図1中の数字と対応しています。

#### ① $z$ 方向の対流（流れ）による運動量の流出入

$z$ 面における単位面積・単位時間あたりの運動量移動量（運動量流束といいます）を考えます。

$$
\text{運動量流束} \frac{\text{kg}\, \text{m/s}}{\text{m}^2 \text{s}}= \text{質量流束} \frac{\text{kg}}{\text{m}^2 \text{s}}\times \text{線速度} \frac{\text{m}}{\text{s}}
$$
ですので
$$
\text{運動量流束} =\rho u_z \cdot u_z = \rho u_z^2 (注2)
$$ です。

>(注2):$\text{質量流束}=\rho A u_z/A$のためこのように表せます。

これに運動量が移動する方向にあたる面の面積$A$をかけるて作用面の運動量を表し、それで収支をとっていきます。


$z=0$ 面（入口）からの流入量:

$$\left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=0}$$


$z=L$ 面（出口）からの流出量:

$$\left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=L}$$

◇◇◇

#### ② 粘性による$r$ 方向の運動量の流出入

粘性による運動量の移動というとイメージがつきにくいと思います。

後ほど、導出しますが、円管内の流体は速度分布があります。中心が一番はやく、壁に近づくほど遅くなっていきます。この速度差のために、運動量は速度の早い中心から壁方向に移動しています。このことを念頭においてください。

式で表すと以下のようになります。（図１中②の矢印の向きを気をつけてください）

$r$ 面（内側）からの流入量:

$$\left. (2\pi r L \cdot \tau_{rz}) \right\vert{}_r$$


$r+\Delta r$ 面（外側）への流出量:

$$\left. (2\pi r L \cdot \tau_{rz}) \right\vert{}_{r+\Delta r}$$

◇◇◇

#### ③ 端面に作用する圧力による運動量の移動（表面力）

圧力による運動量移動は、圧力が流体要素の表面を押して運動量を伝達する現象です。圧力は等方的に作用するため、上流面と下流面では圧力力の向きは逆になることに注意してください。すでに勉強したとおり両者の差が流体を加速する正味の力となります。


$$
z=0 面: P_0 \cdot (2\pi r \Delta r)\\
z=L 面: P_L \cdot (2\pi r \Delta r)
$$
---

### 全運動量収支の一般式

定常状態仮定のため、以上の総和は

$$\text{(流入項)} - \text{(流出項)} + \text{(圧力項)} = 0$$

となります。

$$
\underbrace{\left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=0} + \left. (2\pi r L \tau_{rz}) \right\vert{}_r}_{\text{In (流入)}}  - \underbrace{\left( \left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=L} + \left. (2\pi r L \tau_{rz}) \right\vert{}_{r+\Delta r} \right)}_{\text{Out (流出)}}  + \underbrace{(P_0 - P_L)(2\pi r \Delta r)}_{\text{圧力}} = 0　\tag{1}
$$

---

流体が「非圧縮性流体（密度 $\rho$ が一定）」**であり、かつ**「発達した定常流れ（Fully Developed Flow）」である場合、速度分布 $u_z$ は $z$ 方向に変化しません（$\frac{\partial u_z}{\partial z} = 0$）。

したがって、$z=0$ と $z=L$ での流速は完全に一致します：

$$u_z\vert{}_{z=0} = u_z\vert{}_{z=L}$$

これにより、対流による運動量の流入率と流出率が等しくなり、相殺されて消去されます。

$$\left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=0} - \left. (2\pi r \Delta r \cdot \rho u_z^2) \right\vert{}_{z=L} = 0$$

結果として、

$$\left. (2\pi r L \tau_{rz}) \right\vert{}_r - \left. (2\pi r L \tau_{rz}) \right\vert{}_{r+\Delta r} + (P_0 - P_L)(2\pi r \Delta r) = 0$$

両辺を $2\pi L \Delta r$ で割り、整理します。

$$\frac{\left. (r \tau_{rz}) \right\vert{}_{r+\Delta r} - \left. (r \tau_{rz}) \right\vert{}_r}{\Delta r} = \frac{(P_0 - P_L)}{L} r$$

$\Delta r \to 0$ の極限をとると、左辺は微分の定義そのものになります。

$$\frac{d}{dr}(r \tau_{rz}) = \left( \frac{\Delta P}{L} \right) r \quad \left( \text{ただし } \Delta P = P_0 - P_L \right)$$

### 運動量流束 $\tau_{rz}$ の導出

両辺を $r$ で積分します。

$$r \tau_{rz} = \left( \frac{\Delta P}{2L} \right) r^2 + C_1$$

$$\tau_{rz} = \left( \frac{\Delta P}{2L} \right) r + \frac{C_1}{r}$$

**境界条件 1**：円管中心（$r=0$）で $\tau_{rz}$ は有限の値をとる必要があるため、$C_1 = 0$。言い換えれば、このままだと$r=0$のときせん断応力は無限大になってしまう。

したがって

$$\tau_{rz} = \left( \frac{\Delta P}{2L} \right) r$$

---

## 5. 速度分布 $u_z(r)$ の導出

### ニュートンの粘性法則の適用

運動量流束 $\tau_{rz}$ は、速度勾配を用いて次のように表されます（高流速から低流速への運動量拡散）。

$$\tau_{rz} = -\mu \frac{du_z}{dr}$$

これを前式の $\tau_{rz}$ に代入します。

$$-\mu \frac{du_z}{dr} = \left( \frac{\Delta P}{2L} \right) r$$

$$\frac{du_z}{dr} = -\left( \frac{\Delta P}{2\mu L} \right) r$$

### 積分による速度分布の導出

両辺を $r$ で積分します。

$$u_z(r) = -\left( \frac{\Delta P}{4\mu L} \right) r^2 + C_2$$

**境界条件 2**：管壁（$r=R$）では「滑りなし条件（No-slip condition）」より $u_z(R) = 0$。

$$0 = -\left( \frac{\Delta P}{4\mu L} \right) R^2 + C_2 \implies C_2 = \left( \frac{\Delta P}{4\mu L} \right) R^2$$

これを代入して整理すると、速度分布の式が得られます。

$$u_z(r) = \frac{\Delta P R^2}{4\mu L} \left[ 1 - \left( \frac{r}{R} \right)^2 \right] \tag{1}$$

<h2>【付録2】流体の支配方程式から求める方法</h2>

流体力学の問題を解くときに必ず使う支配方程式（ナビエストークス式）から、層流の速度分布を求める方法を解説します。興味のある人はみてください。

<details>

ニュートンの運動方程式

$$
\boxed{\sum F=ma}
$$

を流体に適用します。

流体に働く代表的な力は、

1. 圧力による力
2. 粘性による力
3. 重力

です。

これらを一般化したものがNavier–Stokes方程式です。

非圧縮性ニュートン流体では、

$$
\boxed{
\rho\left(
\frac{\partial\boldsymbol u}{\partial t}
+\boldsymbol u\cdot\nabla\boldsymbol u
\right)
=
-\nabla P+\mu\nabla^2\boldsymbol u+\rho\boldsymbol g
} \tag{1}
$$

です。ここで$\nabla$は

$$\nabla = \left( \frac{\partial}{\partial x}, \frac{\partial}{\partial y}, \frac{\partial}{\partial z} \right) $$
で定義される微分演算子です。

つまり(1)式は$x,y,z$の３次元の式であり、これ以降の取り扱いを簡便にするため１次元にします。

3次元のナビエ・ストークス方程式を、$x$ 方向の**1次元**（速度を $u(x, t)$、空間座標を $x$ のみとする）に書き直すと、次のようになります。

$$\boxed{ \rho\left( \frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} \right) = -\frac{\partial P}{\partial x} +\mu \frac{\partial^2 u}{\partial x^2} +\rho g_x } \tag{2}$$

---
## (2)式の理解のための説明


### 全体の構造： $F = ma$ の形をしている

式全体を書き直すと、次のようになります。

$$\text{（左辺）} = \text{（右辺の力の合計）}$$

* **左辺** ＝ 単位体積あたりの「慣性力（加速度を生み出す効果）」
* **右辺** ＝ その流体が周囲から受けている「3種類の力の合計」


### 左辺：慣性項（加速度の表現）

$$\rho \left( \frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} \right)$$

* **$\rho$（密度）：** 流体の「重さ（質量）」です。重いほど動かすのに大きな力が必要です。
* **$\frac{\partial u}{\partial t}$（時間変化項）：** 「時間の経過とともに、その場のスピードがどう変わるか」を表します（例：アクセルを踏んでスピードが上がるとき）。
* **$u \frac{\partial u}{\partial x}$（移流項・空間変化項）：** 「場所移動によってスピードが変わるか」を表します（例：川幅が狭くなって、流れているうちに自然と水流が速くなるとき.この場合も立派な加速度です）。


### 右辺①：圧力勾配力（おしつける力）

$$-\frac{\partial P}{\partial x}$$

* **意味：** 「圧力の差（勾配）によって、流体が押し出される力」です。
* **イメージ：** 注射器のピストンを想像してください。圧力の高い方から低い方へと流体は押されます。マイナスがついているのは、「圧力が高いところから低いところ（＋の方向）へ向かって押し流される」からです。


### セクション②：粘性項（摩擦・ドロドロさによる抵抗）

$$\mu \frac{\partial^2 u}{\partial x^2}$$

* **$\mu$（粘度）：** ハチミツや油のような「ドロドロさ（粘り気）」です。
* **意味：** 「周囲の速い流れや遅い流れから受ける、摩擦（ブレーキやアクセル）の力」です。
* **イメージ：** 周囲の壁や遅い流れに引っ張られたり、逆に速い流れに引っ張ったりして、速度のムラを均そうとする力（運動量の拡散）です。


### 右辺③：外力（重力など）

$$\rho g_x$$

* **$g_x$：** $x$ 方向にかかる重力の加速度（成分）です。
* **意味：** 「重力などの外部から直接引っ張られる力」です。
* **イメージ：** 滑り台の上で、重力によって下に引きずり下ろされるような力です。

## 層流の速度分布式の導出

ナビエストークス式（1)式から円管内層流流れの、半径方向の速度分布を導出してみます。参考書の2.20式に相当します。

### 1. 物理モデルの設定

* **対象：** 2枚の無限に広い静止した平らな壁（$y = -R$ と $y = R$）の間に、粘性流体が流れている状態。
* **座標系：**
* $x$ 方向：流体の流れる方向（水平方向）
* $y$ 方向：壁に垂直な方向（高さ方向）


* **仮定：**
1. **定常状態：** 時間変化なし ($\frac{\partial}{\partial t} = 0$)
2. **1次元流れ：** 流れは $x$ 方向のみ、$u = u(y)$ のみで変化（$v = 0$）
3. **完全発達流：** 下流に行っても速度分布が変わらない ($\frac{\partial u}{\partial x} = 0$)
4. **圧力勾配**：** 圧力勾配は一定 ($\frac{\partial p}{\partial x} = \alpha$)
5. **重力：** 重力項は０



### 2. 支配方程式の簡略化

先ほどの1次元ナビエ・ストークス方程式（$x$ 方向）からスタートします。

$$\rho \left( \frac{\partial u}{\partial t} + u \frac{\partial u}{\partial x} \right) = -\frac{\partial P}{\partial x} + \mu \frac{\partial^2 u}{\partial y^2} + \rho g_x \tag{1}$$

上記の仮定（定常、1次元、重力無視）を代入すると、左辺の慣性項はすべてゼロになります。

$$0 = -\alpha+ \mu \frac{d^2 u}{dy^2}$$

これを変形すると、**2階の常微分方程式**が得られます。

$$\frac{d^2 u}{dy^2} = \frac{\alpha}{\mu} \tag{2}$$



### 3. 微分方程式(2)の解法（2回積分する）

#### ① 1回目の積分（速度の傾きを求める）

$y$ で両辺を積分します。

$$\frac{du}{dy} = \frac{\alpha}{\mu} y + C_1$$


（$C_1$ は積分定数）

#### ② 2回目の積分（速度分布を求める）

もう一度 $y$ で積分します。


$$u(y) = \frac{\alpha}{2\mu}  y^2 + C_1 y + C_2$$


（$C_2$ は積分定数）



### 4. 境界条件の適用と定数の決定

壁面において、流体は壁にぴったり張り付いて動かないという「粘着条件（no-slip condition）」を適用します。

* **境界条件：** $y = -R$ および $y = R$ において、速度 $u = 0$

1. **$y = R$ を代入：**

$$0 = \frac{\alpha}{2\mu} R^2 + C_1 R + C_2$$


2. **$y = -R$ を代入：**

$$0 = \frac{\alpha}{2\mu} R^2 - C_1 R + C_2$$



この2つの式を連立させると、

$$
C_1 = 0\\
C_2 = -\frac{\alpha}{2\mu} R^2$$

が求まります。



### 5. 完成：速度分布式

求まった $C_1, C_2$ を元の式に代入して整理すると、放物線状の速度分布を表す式が導かれます。

$$\boxed{ u(y) = -\frac{\alpha}{2\mu}  (R^2 - y^2) }$$

なお、流れ方向($x$)に圧力は減少するので、圧力勾配$\alpha= \frac{dP}{dx}$は負の値であるため(3)式の$-\frac{\alpha}{2\mu} > 0$である。マイナスがついたまま嫌な場合は$-\alpha=\beta$とおいて

$$\boxed{ u(y) = \frac{\beta}{2\mu}  (R^2 - y^2) }$$

とするといいでしょう。

ここで
$\beta = - \frac{dP}{dx}$で一定としているため、直線の傾きつまり、入口の圧力$P_0$、出口圧力$P_L$,長さ$L$とすると

$$
\beta = \frac{dP}{dx} = \frac{P_0 - P_L}{L}
$$

とおけば

$$ u(y) = \frac{P_0 - P_L}{2\mu L}  (R^2 - y^2) $$

> **💡 物理的な意味**
> * 中心 ($y = 0$) で速度が最大になり、壁面 ($y = \pm h$) に向かって速度がゼロになる**「放物線（二次曲線）の速度プロファイル」**が、直交座標系から綺麗に導き出されました！
> 
>分母が４ではなく２になってますが、これはNSを直交座標系の１次元で解いたからです。円筒座標系にして、解けば係数4となります。
