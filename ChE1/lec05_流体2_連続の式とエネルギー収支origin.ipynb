# 1. まず「収支」を思い出そう

第1回で、化学工学の基本として次の式を学びました。

$$
\boxed{
\text{蓄積}
=
\text{流入}
-
\text{流出}
+
\text{生成}
-
\text{消費}
}
$$

この考え方は、物質収支だけでなく、流体にもそのまま使えます。<br>
流体の流れを考えるとき、

> 「ある領域にどれだけの質量が入り、どれだけ出ていくか」

を考えればよいのです。<br>
ただし、化学反応がない場合は、

$$
\text{生成}=\text{消費}=0
$$

となります。<br>
すると、

$$
\boxed{
\text{蓄積}
=
\text{流入}
-
\text{流出}
}
$$

となります。

# 2. 流体の物質収支

第４回でも説明済みですが、もう一度。

## 2.1 制御体積を考える

流体の収支では、ある領域を「系」として切り出します。<br>
例えば配管の一部分を考えます。

$$
\boxed{
\text{入口}
\rightarrow
\boxed{\text{制御体積}}
\rightarrow
\text{出口}
}
$$

入口から質量流量

$$
\dot m_1
$$

が入り、出口から

$$
\dot m_2
$$

が出るとします。<br>
制御体積内に存在する質量を $M$ とすると、

$$
\frac{dM}{dt}
=
\dot m_1-\dot m_2
$$

です。

これが流体に対する基本的な物質収支です。


# 2.3. 定常流れと連続の式

定常状態では、制御体積内の質量が時間によって変化しません。

したがって、

$$
\frac{dM}{dt}=0
$$

です。よって、

$$
\dot m_1=\dot m_2
$$

となります。質量流量は、

$$
\dot m=\rho Q
$$

でした。また、

$$
Q=Au
$$

なので、

$$
\boxed{
\dot m=\rho Au
}
$$

です。したがって、定常流れでは、

$$
\boxed{
\rho_1A_1u_1
=
\rho_2A_2u_2
}
$$

となります。<br>
これが **連続の式（continuity equation）** の基本形です。


---
# 2.4. 非圧縮性流体の場合

液体では、圧力変化による密度変化が小さいことが多く、

$$
\rho_1\approx\rho_2
$$

と考えられます。<br>
この場合、

$$
\rho A_1u_1=\rho A_2u_2
$$

なので、$\rho$ が約分され、

$$
\boxed{
A_1u_1=A_2u_2
}
$$

となります。
さらに、

$$
Q=Au
$$

なので、

$$
\boxed{
Q_1=Q_2
}
$$

です。

---

## 重要な意味

非圧縮性流体の定常流れでは、

> **管が細くなれば流速が速くなる**

ということです。

$$
A\downarrow
\quad\Rightarrow\quad
u\uparrow
$$

逆に管が太くなれば、

$$
A\uparrow
\quad\Rightarrow\quad
u\downarrow
$$

となります。


---
# 2.5. 連続の式の例題

## 問題

水が定常的に配管を流れている。

入口の配管径(内径)が$D_1=0.10\ \mathrm{m}$<br>
出口の配管径が$D_2=0.05\ \mathrm{m}$<br>
であり、入口速度が$u_1=1.0\ \mathrm{m/s}$<br>
である。

出口速度 $u_2$ を求めなさい。

---
解答後、答え合わせをしてください。

<details>
<summary> 解答</summary>

非圧縮性流体なので、

$$
A_1u_1=A_2u_2
$$

です。
円管では、

$$
A=\frac{\pi D^2}{4}
$$

なので、

$$
\frac{\pi D_1^2}{4}u_1
=
\frac{\pi D_2^2}{4}u_2
$$

より、

$$
u_2
=
u_1\left(\frac{D_1}{D_2}\right)^2
$$

です。

したがって、

$$
u_2
=
1.0
\left(\frac{0.10}{0.05}\right)^2
$$

$$
\boxed{
u_2=4.0\ \mathrm{m/s}
}
$$

です。
配管径が半分になると、断面積は $1/4$ になります。
したがって、同じ流量を保つために流速は4倍になります。
</details>

# 2.6. 連続の式のもう一つの見方

連続の式は、

$$
A_1u_1=A_2u_2
$$

と覚えるだけではなく、

> **「質量が途中で突然増えたり消えたりしない」**

という物質収支から出てきた式だと理解してください。
つまり、

$$
\boxed{
\text{物質収支}
\rightarrow
\text{質量保存}
\rightarrow
\text{連続の式}
}
$$

です。
これは化学工学で非常に重要な考え方です。
第2・3回で学んだ物質収支が、流体力学でもそのまま使われています。

# 3. エネルギー収支
## 3.1 概説
物質収支では、

> 「どれだけ物質が入って、どれだけ出るか」

を考えました。

次は、

> 「流体がどれだけエネルギーを持っているか」

を考えます。

流体の機械的なエネルギーとして、まず次の3つを考えます。

### ① 圧力によるエネルギー

$$
\frac{P}{\rho}
$$

### ② 運動エネルギー

$$
\frac{u^2}{2}
$$

### ③ 位置エネルギー

$$
gz
$$

これらを合わせると、

$$
\boxed{
\frac{P}{\rho}
+
\frac{u^2}{2}
+
gz
}
$$

となります。

これは**単位質量あたりの機械的エネルギー**です。

>【確認しておいて】上の四角で囲んだ式の各項の単位はすべて$\text{J/kg}$と表せることを確認しておいてください。
>直接解析すると$\mathrm{m^2/s^2}$となりますが、これが$\text{J/kg}$と変換されることを確認してくださいということです。

---
## 3.2. それぞれのエネルギーを理解する

### (1) 圧力による項

$$
\frac{P}{\rho}
$$

圧力 $P$ が大きいほど、流体を押し動かす能力が大きくなります。


### (2) 運動エネルギー

$$
\frac{u^2}{2}
$$

速度 $u$ が大きいほど運動エネルギーが大きくなります。<br>
速度は二乗で効くことに注意してください。<br>
例えば速度が2倍になると、

$$
u^2\rightarrow4u^2
$$

なので、運動エネルギーは4倍になります。

### (3) 位置エネルギー

$$
gz
$$

高さ $z$ が高いほど位置エネルギーが大きくなります。<br>
つまり、流体が高いところにあること自体がエネルギーを持っているということです。


---
# 3.3. ベルヌーイの式

**理想的な** 流れについて、

- 定常流れ
- 非圧縮性流体
- 粘性によるエネルギー損失を無視
- ポンプなどによる外部仕事を無視

とすると、

$$
\boxed{
\frac{P}{\rho}
+
\frac{u^2}{2}
+
gz
=
\text{一定}
}
$$

となります。

入口1と出口2で書けば、

$$
\boxed{
\frac{P_1}{\rho}
+
\frac{u_1^2}{2}
+
gz_1
=
\frac{P_2}{\rho}
+
\frac{u_2^2}{2}
+
gz_2
} \tag{1}
$$

です。

これが**ベルヌーイの式**です。

$$

\text{一定}
$$

ということです。

# 4.次回への接続： ベルヌーイの法則の拡張：
### 実務計算のため

ここが非常に重要です。<br>
ベルヌーイの式は理想化された式です。しかし実際の流体には粘度があり、<br>
そのため、流体が配管を流れると、摩擦によって機械的エネルギーが失われます。

したがって、実際には、

$$
\boxed{
\text{入口のエネルギー}
>
\text{出口のエネルギー}
}
$$

となることがあります。この損失を**圧力損失**あるいは**エネルギー損失**として扱います。

◇◇◇

逆にエネルギーが加わることもあります。
例えば、低い位置から高い位置へ液体を送るとします。

流体を高い位置へ送るには、外部からエネルギーを与える必要があります。<br>
そこでポンプによる仕事を加えます。

理想化すると、

$$
\boxed{
\frac{P_1}{\rho}
+\frac{u_1^2}{2}
+gz_1
+w_p
=
\frac{P_2}{\rho}
+\frac{u_2^2}{2}
+gz_2
}
$$

と表せます。

ここで $w_p$ は、単位質量あたりにポンプから与えられるエネルギーです。



---
以上のように

実際の配管では、

- 配管摩擦
- エルボ
- バルブ
- 急拡大
- 急縮小
- 熱交換器
- その他の流動抵抗

によってエネルギーが失われます。

したがって一般的には、

$$
\boxed{
\frac{P_1}{\rho }
+\frac{u_1^2}{2}
+g z_1
+g h_p
=
\frac{P_2}{\rho }
+\frac{u_2^2}{2}
+gz_2
+gh_L
} \tag{2}
$$


と書きます。

ここで、

- $h_p$：ポンプによる揚程
- $h_L$：損失水頭

です。

この式は、化学工学における流体計算の非常に重要な基本式です。
