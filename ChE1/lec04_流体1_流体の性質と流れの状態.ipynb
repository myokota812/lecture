# 1. 流体とは何か

## 1.1 固体と流体

物質は大きく、

- 固体
- 液体
- 気体

などの状態をとります。

化学工学で「流体」と呼ぶのは、基本的には

> **液体と気体**

です。

流体は、固体とは異なり、力を受けると連続的に変形して流れます。

例えば水を容器に入れると、容器の形に合わせて水の形が変わります。

一方、固体のブロックは、通常は一定の形を保ちます。

---

## 1.2 化学工学で流体を扱う理由

化学プラントでは、

$$
\text{タンク}
\rightarrow
\text{配管}
\rightarrow
\text{ポンプ}
\rightarrow
\text{熱交換器}
\rightarrow
\text{反応器}
$$

のように、流体を次々に移動させます。

したがって、

> **どのくらい流れるか**

だけではなく、

> **どのくらいの圧力が必要か**

> **どのくらいの速度で流れるか**

> **どのような流れ方をするか**

を知る必要があります。

# 2. 流体の基本的な物性

流体の流れを考えるとき、特に重要なのが次の物性です。

| 物性 | 記号 | SI単位 | 意味 |
|---|---:|---:|---|
| 密度 | $\rho$ | kg/m³ | 単位体積あたりの質量 |
| 粘度 | $\mu$ | Pa·s | 流れにくさ・内部摩擦の程度 |
| 動粘度 | $\nu$ | m²/s | $\mu/\rho$ |
| 圧力 | $P$ | Pa | 単位面積あたりの力 |

特に今回重要なのは、

$$
\boxed{\rho,\quad \mu,\quad P}
$$

です。


# 3. 密度

## 3.1 密度の定義

密度は、

> **単位体積あたりの質量**

です。

$$
\boxed{
\rho=\frac{m}{V}
}
$$

したがって、

$$
m=\rho V
$$

となります。

SI単位は、

$$
\boxed{\mathrm{kg/m^3}}
$$

です。

---

## 3.2 例

水の密度を近似的に

$$
\rho=1000\ \mathrm{kg/m^3}
$$

とします。

水

$$
V=0.20\ \mathrm{m^3}
$$

の質量は、

$$
m=\rho V
$$

より、

$$
m=(1000)(0.20)
$$

したがって、

$$
\boxed{m=200\ \mathrm{kg}}
$$

です。

密度は、質量流量と体積流量を変換するときにも重要です。


```python
# 密度から質量を計算
rho = 1000       # kg/m^3
V = 0.20         # m^3

m = rho * V

print(f"質量 = {m:.1f} kg")
```

# 4. 質量流量と体積流量

化学工学では、流量を非常によく使います。

## 4.1 体積流量

単位時間に流れる体積を体積流量といいます。

$$
\boxed{
Q=\frac{dV}{dt}
}
$$

単位は、

$$
\mathrm{m^3/s}
$$

です。

---

## 4.2 質量流量

単位時間に流れる質量は、

$$
\boxed{
\dot m=\frac{dm}{dt}
}
$$

です。

単位は、

$$
\mathrm{kg/s}
$$

です。

---

## 4.3 密度による関係

$$
m=\rho V
$$

なので、時間で微分すると、

$$
\boxed{
\dot m=\rho Q
}
$$

となります。

これは流体計算で非常に重要な式です。

例えば、

$$
\rho=1000\ \mathrm{kg/m^3}
$$

$$
Q=0.010\ \mathrm{m^3/s}
$$

なら、

$$
\dot m
=
1000\times0.010
=
\boxed{10\ \mathrm{kg/s}}
$$

です。

# 5. 圧力

## 5.1 圧力の定義

圧力は、

> **単位面積あたりに作用する力**

として定義されます。

$$
\boxed{
P=\frac{F}{A}
}
$$

SI単位は Pa（パスカル）です。

$$
1\ \mathrm{Pa}
=
1\ \mathrm{N/m^2}
$$

です。

---

## 5.2 なぜ配管で圧力が重要なのか

流体を配管に流すには、一般に圧力差が必要です。

例えば、

$$
P_1>P_2
$$

なら、

$$
\text{高圧側}\rightarrow\text{低圧側}
$$

へ流体を流すことができます。

逆に、配管の途中で圧力が失われると、流体を所定の流量で送るためにポンプなどで圧力を補う必要があります。

# 6. 粘度とは何か

## 6.1 「流れにくさ」を表す物性

水と蜂蜜を比べると、蜂蜜の方が流れにくいでしょう。

このような「流れにくさ」に関係する物性が**粘度**です。

粘度が大きい流体ほど、流れにくくなります。

例えば一般的に、

- 水：比較的小さい粘度
- 油：水より大きいことが多い
- 蜂蜜：非常に大きい粘度

となります。

---

## 6.2 粘度は「内部摩擦」と関係する

流体の中で、隣り合う部分が異なる速度で動いているとします。

例えば、

$$
\boxed{\text{速い層}}
$$
と
$$
\boxed{\text{遅い層}}
$$

が接していると、その間には速度差があります。

この速度差を小さくしようとする抵抗が、流体の粘性による抵抗です。

# 8. 配管の流れを考える

ここからは、断面が円形の配管を考えます。

配管断面積を $A$、平均流速を $u$ とします。

体積流量は、

$$
\boxed{
Q=Au
}
$$

です。

円管の直径を $D$ とすると、

$$
A=\frac{\pi D^2}{4}
$$

なので、

$$
\boxed{
Q=\frac{\pi D^2}{4}u
}
$$

となります。

したがって、平均流速は、

$$
\boxed{
u=\frac{4Q}{\pi D^2}
}
$$

です。

---

## 重要なポイント

同じ流量なら、配管が細いほど流速は速くなります。

$$
u\propto\frac{1}{D^2}
$$

です。

つまり、配管径は流速・圧力損失・流れの状態などに大きく影響します。


```python
import math

Q = 0.010      # m^3/s
D = 0.050      # m

A = math.pi * D**2 / 4
u = Q / A

print(f"断面積 A = {A:.6f} m^2")
print(f"平均流速 u = {u:.3f} m/s")
```

ここで重要な注意があります。

配管内の流体は、断面全体で同じ速度で流れているとは限りません。

そのため、

$$
u=u(r)
$$

ただし、流量計算では断面平均速度

$$
\boxed{
u_{\mathrm{avg}}=\frac{Q}{A}
}
$$

を使うことができます。

今回の「平均流速」は、この意味です。

# 10. 層流と乱流

## 10.1 層流

流体が比較的規則正しく流れる状態を**層流（laminar flow）**といいます。

イメージとしては、

$$
\rightarrow\rightarrow\rightarrow\rightarrow
$$

のように、流体の層が大きく乱れずに流れます。

層流では、粘性の影響が非常に重要です。

---

## 10.2 乱流

一方、流れが複雑に変動し、速度や方向が時間的・空間的に不規則になる状態を**乱流（turbulent flow）**といいます。

イメージとして、

$$
\rightarrow\nearrow\rightarrow\searrow\rightarrow
$$

のように複雑な運動を伴います。

ただし、「乱流＝完全にランダム」という意味ではありません。

乱流には平均的な流れ方向があり、その周囲に速度変動が存在します。

# 11. レイノルズ数

## 11.1 なぜ無次元数が必要なのか

流れの状態は、

- 流速
- 配管径
- 密度
- 粘度

などによって変わります。

しかし、それぞれの物理量を個別に比較するより、

> **これらを一つの無次元数にまとめる**

と流れを整理しやすくなります。

それが**レイノルズ数（Reynolds number）**です。

円管内流れでは、

$$
\boxed{
Re=\frac{\rho uD}{\mu}
}
$$

です。

また、

$$
\nu=\frac{\mu}{\rho}
$$

なので、

$$
\boxed{
Re=\frac{uD}{\nu}
}
$$

とも書けます。


## 11.2 レイノルズ数の意味

レイノルズ数は単なる計算用の数字ではありません。

概念的には、

$$
\boxed{
Re
\sim
\frac{\text{慣性の効果}}
{\text{粘性の効果}}
}
$$

と考えることができます。

- $Re$ が小さい  
  → 粘性の影響が相対的に大きい
- $Re$ が大きい  
  → 慣性の影響が相対的に大きい

という意味です。

これが、流れの状態を考えるときに重要になります。


## 11.3 次元を確認する

レイノルズ数

$$
Re=\frac{\rho uD}{\mu}
$$

が本当に無次元なのか確認してみましょう。

$$
[\rho]
=
\mathrm{kg/m^3}
$$

$$
[u]=\mathrm{m/s}
$$

$$
[D]=\mathrm{m}
$$

したがって分子は、

$$
\frac{\mathrm{kg}}{\mathrm{m^3}}
\frac{\mathrm{m}}{\mathrm{s}}
\mathrm{m}
=
\frac{\mathrm{kg}}{\mathrm{m\,s}}
$$

です。

粘度は、

$$
[\mu]
=
\frac{\mathrm{kg}}{\mathrm{m\,s}}
$$

なので、

$$
[Re]
=
\frac{\mathrm{kg/(m\,s)}}
{\mathrm{kg/(m\,s)}}
=
\boxed{1}
$$

となります。

つまりレイノルズ数は**単位を持たない無次元数**です。

# 12. 円管内流れにおける流れの分類

円管内の流れでは、一般的な目安として、

$$
\boxed{Re<2300}
$$

なら層流、

$$
\boxed{Re>4000}
$$

なら乱流、

$$
\boxed{2300\lesssim Re\lesssim4000}
$$

では遷移領域、

と扱います。

ただし、これは**円管内の十分に発達した流れに対する代表的な目安**です。

流れの形状、入口条件、曲がり、障害物などによって状況は変化します。

したがって、

> 「Re = 2300を境に自然界で急に流れが変わる」

と考えるのではなく、

> **レイノルズ数を使って流れの状態を分類するための代表的な基準**

と理解してください。


---
# 13. レイノルズ数を計算してみよう

## 例題

水が円管を流れている。

$$
\rho=1000\ \mathrm{kg/m^3}
$$

$$
\mu=1.0\times10^{-3}\ \mathrm{Pa\cdot s}
$$

$$
u=1.0\ \mathrm{m/s}
$$

$$
D=0.010\ \mathrm{m}
$$

レイノルズ数を求めなさい。

---

### 解答

$$
Re=\frac{\rho uD}{\mu}
$$

なので、

$$
Re
=
\frac{
(1000)(1.0)(0.010)
}{
1.0\times10^{-3}
}
$$

したがって、

$$
\boxed{
Re=1.0\times10^4
}
$$

です。

これは $4000$ より大きいので、円管内の代表的な基準では**乱流領域**です。


```python
# レイノルズ数の計算
rho = 1000.0       # kg/m^3
mu = 1.0e-3        # Pa s
u = 1.0            # m/s
D = 0.010          # m

Re = rho * u * D / mu

print(f"Re = {Re:.2e}")

if Re < 2300:
    print("目安：層流")
elif Re > 4000:
    print("目安：乱流")
else:
    print("目安：遷移領域")
```

# 14. 何を変えると流れが変わるのか

レイノルズ数

$$
Re=\frac{\rho uD}{\mu}
$$

を見ると、

### 流速 $u$ を大きくする

$$
u\uparrow
\Rightarrow
Re\uparrow
$$

→ 乱流になりやすい。

### 配管径 $D$ を大きくする

$$
D\uparrow
\Rightarrow
Re\uparrow
$$

→ 乱流になりやすい。

### 密度 $\rho$ を大きくする

$$
\rho\uparrow
\Rightarrow
Re\uparrow
$$

→ 乱流になりやすい。

### 粘度 $\mu$ を大きくする

$$
\mu\uparrow
\Rightarrow
Re\downarrow
$$

→ 層流になりやすい。

したがって、

> **速く流すほど、太い配管ほど、粘度が小さいほど、乱流になりやすい**

と覚えるとよいでしょう。

# 15. 温度と粘度

流体の粘度は温度によって変化します。

液体では一般に、

$$
T\uparrow
\Rightarrow
\mu\downarrow
$$

となる傾向があります。

つまり液体は温度を上げると流れやすくなることが多いです。

一方、気体では一般に逆の傾向があり、

$$
T\uparrow
\Rightarrow
\mu\uparrow
$$

となる傾向があります。

これは化学プラント設計で重要です。

例えば液体を加熱すると、

- 粘度が低下
- 流れやすくなる
- 圧力損失が変化する
- レイノルズ数が変化する

という影響が生じます。

したがって、

> **物性値は「一定の数字」とは限らず、温度・圧力などの条件に依存する**

ということを意識してください。

# 16. 層流と乱流の比較

| 項目 | 層流 | 乱流 |
|---|---|---|
| 流れ | 比較的規則的 | 複雑な変動を伴う |
| 粘性の影響 | 大きい | 相対的に小さい |
| $Re$ | 小さい | 大きい |
| 混合 | 比較的弱い | 強い |
| 速度変動 | 小さい | 大きい |
| 解析 | 比較的扱いやすい | 複雑 |

ここで重要なのは、

**乱流だから「悪い」わけではない**

ということです。

乱流には、

- 混合を促進する
- 熱伝達を促進する
- 物質移動を促進する

という利点があります。

一方で、

- 圧力損失が大きくなる
- ポンプ動力が増える

などの影響があります。

したがって、化学工学では「層流か乱流か」を知るだけでなく、

> **その流れを利用して、どのような装置性能が得られるか**

まで考えます。

# 21. 発展：同じ流量でも流れは変わる

同じ流量

$$
Q=0.010\ \mathrm{m^3/s}
$$

を流す場合でも、配管径が違えば平均流速は変わります。

$$
u=\frac{4Q}{\pi D^2}
$$

したがって、配管径を小さくすると流速が急激に大きくなります。

さらに、

$$
Re=\frac{\rho uD}{\mu}
$$

なので、流速変化は流れの状態にも影響します。

このように、

$$
\boxed{
Q
\rightarrow
D
\rightarrow
u
\rightarrow
Re
\rightarrow
\text{流れの状態}
}
$$

というつながりがあります。

これは今後の流体計算で非常に重要な考え方です。


```python
# 配管径を変えたときの平均流速とReを比較
import math

rho = 1000.0
mu = 1.0e-3
Q = 0.010

for D in [0.02, 0.03, 0.05, 0.08, 0.10]:
    A = math.pi * D**2 / 4
    u = Q / A
    Re = rho * u * D / mu
    print(f"D={D:.3f} m : u={u:7.3f} m/s, Re={Re:8.0f}")
```
