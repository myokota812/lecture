<h3>提出課題1</h3>


以下の2つの化学反応が同時に起きる反応器を考えます。関与する物質は **A, B, C, D** の4種類です。

$$\begin{aligned} A &\xrightarrow{} 2B + C \qquad (\xi_1)\\ B + C &\xrightarrow{} D \qquad (\xi_2) \end{aligned}$$

このとき、各成分（A, B, C, D）の物質収支式を導出し、以下の形の**行列方程式**で表しなさい。

$$\mathbf{F}_{\mathrm{out}} = \mathbf{F}_{\mathrm{in}} + \boldsymbol{\nu}\, \boldsymbol{\xi}$$

<h3>提出課題１解答例</h3>
### 1. 各反応における各成分の量論係数の整理
 
 
 * **反応1 ($\xi_1$):** $A \rightarrow 2B + C$
 * $A: -1$
 * $B: +2$
 * $C: +1$
 * $D: 0$
 
 
 * **反応2 ($\xi_2$):** $B + C \rightarrow D$
 * $A: 0$
 * $B: -1$
 * $C: -1$
 * $D: +1$
 
 
 
 
 ### 2. 物質収支の連立方程式
 
 
 各成分の出口流量は以下のようになります。
 
 $$ \left\{  \begin{aligned}  F_{\mathrm{out},A} &= F_{\mathrm{in},A} - \xi_1\\   F_{\mathrm{out},B} &= F_{\mathrm{in},B} + 2\xi_1 - \xi_2\\   F_{\mathrm{out},C} &= F_{\mathrm{in},C} + \xi_1 - \xi_2\\   F_{\mathrm{out},D} &= F_{\mathrm{in},D} + \xi_2  \end{aligned}  \right. $$
 
 
 ### 3. 行列方程式による表現
 
 
 これをベクトルの足し算と化学量論行列 $\boldsymbol{\nu}$ の積に落とし込みます。
 $$ \begin{bmatrix}  F_{\mathrm{out},A}\\  F_{\mathrm{out},B}\\  F_{\mathrm{out},C}\\  F_{\mathrm{out},D}  \end{bmatrix}  =  \begin{bmatrix}  F_{\mathrm{in},A}\\  F_{\mathrm{in},B}\\  F_{\mathrm{in},C}\\  F_{\mathrm{in},D}  \end{bmatrix}  +  \begin{bmatrix}  -1 & 0\\  2 & -1\\  1 & -1\\  0 & 1  \end{bmatrix}  \begin{bmatrix}  \xi_1\\  \xi_2  \end{bmatrix} $$
> 
> 
> **【ポイント】**
> 物質が4種類（行が4）、反応が2つ（列が2）なので、化学量論行列 $\boldsymbol{\nu}$ は **$4 \times 2$ の非正方行列** になります。このように、反応や物質の数が変わっても、同じルール（$\mathbf{F}_{\mathrm{out}} = \mathbf{F}_{\mathrm{in}} + \boldsymbol{\nu}\boldsymbol{\xi}$）で綺麗にシステムを記述できるのがこの手法の最大の強みです！

<h3>提出課題２解答例</h3>

以下の表の値を求めることになる。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |
| ---- | :----: | :----: |
| CH<sub>3</sub>OH | 100 | C |  
| O<sub>2</sub> | A | D |
| N<sub>2</sub> | B | B |
| HCOH | 0 | E |
| H<sub>2</sub> | 0 | F |
| CO<sub>2</sub> | 0 | G |
| H<sub>2</sub>O | 0 | H |



- 仕様からわかるもの（A,B,D,F,G）
  - A,B:仕様2.3から
    - A=100 $\times$ 1.3 $\times$ 0.21 = 27.3 mol/h
    - B=100 $\times$ 1.3 $\times$ 0.79 = 102.7 mol/h

  - D,F,G:B=102.7 molと仕様2.2から
    - D=102.7 $\times$ (6.3/66.1)  = 9.79 mol/h
    - F=102.7 $\times$ (25.9/66.1)  = 40.24 mol/h
    - G=102.7 $\times$ (1.7/66.1)  = 2.64 mol/h

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |
| ---- | :----: | :----: |
| CH<sub>3</sub>OH | 100 | C |  
| O<sub>2</sub> | 27.3 | 9.79 |
| N<sub>2</sub> | 102.7 | 102.7 |
| HCOH | 0 | E |
| H<sub>2</sub> | 0 | 40.24 |
| CO<sub>2</sub> | 0 | 2.64 |

---




次に
$$ \begin{aligned} 
\text{反応(A):} \quad & \text{CH}_3\text{OH} + \frac{1}{2}\text{O}_2 \longrightarrow \text{HCHO} + \text{H}_2\text{O} \\ 
\text{反応(B):} \quad & \text{CH}_3\text{OH} \longrightarrow \text{HCHO} + \text{H}_2 \\ \text{反応(C):} \quad & \text{CH}_3\text{OH} + \frac{3}{2}\text{O}_2 \longrightarrow \text{CO}_2 + 2\text{H}_2\text{O} 
\end{aligned} 
$$



の反応系について物質収支式:

$$
n_{i1}-n_{i2}+ \sum_{j} P_j \nu_{ji} =0 \ \ ... (A)
$$

を考えます。$i$は成分, $j$は反応式の種類になります。この場合$j$はA,B,Cの3種類です。量論係数は$ \nu_{Ai}$,$ \nu_{Bi}$,$ \nu_{Ci}$の３種類となり、これを先ほどの表に追加した表が下になります。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |$ \nu_{Ai}$ |$ \nu_{Bi}$ |$ \nu_{Ci}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH<sub>3</sub>OH | 100 | C |-1|-1 |-1 |  
| O<sub>2</sub> | 27.3 | 9.79 |-1/2 |- |-3/2 |
| N<sub>2</sub> | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |1 |1 |- |
| H<sub>2</sub> | 0 | 40.24 |- |1 |- |
| CO<sub>2</sub> | 0 | 2.64 |- |- |1 |
| H<sub>2</sub>O | 0 | H |1 |- |2 |

さらに4,5,6列目を$P_j \nu_{ji}$にする。記号を簡単にするため
- 反応Aによって消費される限定成分の物質量$P_A$を$x$
- $P_B$を$y$
- $P_C$を$z$
とする。

| 成分$i$ | 入量[mol/h] | 出量[mol/h] |$ \nu_{Ai}P_{A}$ |$ \nu_{Bi}P_{B}$ |$ \nu_{Ci}P_{C}$|
| ---- | :----: | :----: |:----: |:----: |:----: |
| CH<sub>3</sub>OH  | 100 | C |$-1x$|$-1y$ |$-1z$ |  
| O<sub>2</sub> | 27.3 | 9.79 |$-1/2x$ |- |$-3/2 z$|
| N<sub>2</sub> | 102.7 | 102.7 |- |-|-|
| HCOH | 0 | E |$1x$ |$1y$ |- |
| H<sub>2</sub> | 0 | 40.24 |- |$1y$ |- |
| CO<sub>2</sub> | 0 | 2.64 |- |- |$1 z$|
| H<sub>2</sub>O| 0 | H |$1x$ |- |$2 z$|

このような表にすると各成分の物質収支は行ごとに並べれば良い。つまり

* (i=CH3OH): $100-C-x-y-z=0$.
* (i=O2):    $27.3-9.79-(1/2)x-(3/2)z=0$
* (i=HCOH): $0-E+x+y=0$
* (i=H2): $0-40.24+y=0$
* (i=CO2): $0-2.64+z=0$
* (i=H2O): $0-H+x+2z=0$

未知の変数は６個、式の数が６つなので、この連立方程式を解けば解が求まる。


* $C=100-x-y-z$.
* $9.79= 27.3-(1/2)x-(3/2)z$
* $E= 0 +x+y $
* $40.24 = y$
* $2.64 = z$
* $H =  x+2z$

\begin{bmatrix} 
C\\ 9.79\\ 102.7\\ E\\ 40.24\\ 2.64\\ H 
\end{bmatrix} = \begin{bmatrix} 100\\ 27.3\\ 102.7\\ 0\\ 0\\ 0\\ 0 \end{bmatrix}+ \begin{bmatrix}-1&-1&-1\\ -\frac12&0&-\frac32\\ 0&0&0\\ 1&1&0\\ 0&1&0\\ 0&0&1\\ 1&0&2 \end{bmatrix} \begin{bmatrix} x\\ y\\ z \end{bmatrix}

```python
import numpy as np

# 未知数ベクトル
# u = [C, E, H, x, y, z]^T

A = np.array([
    [1, 0, 0, 1, 1, 1],      # C + x + y + z = 100
    [0, 0, 0, 0.5, 0, 1.5],  # 0.5x + 1.5z = 27.3 - 9.79
    [0, 1, 0, -1, -1, 0],    # E - x - y = 0
    [0, 0, 0, 0, 1, 0],      # y = 40.24
    [0, 0, 0, 0, 0, 1],      # z = 2.64
    [0, 0, 1, -1, 0, -2]     # H - x - 2z = 0
])

b = np.array([
    100,
    27.3 - 9.79,
    0,
    40.24,
    2.64,
    0
])

# 方程式を解く
u = np.linalg.solve(A, b)

# 結果を取り出す
C, E, H, x, y, z = u

# 出力
print(f"C = {C:.2f}")
print(f"E = {E:.2f}")
print(f"H = {H:.2f}")
print(f"x = {x:.2f}")
print(f"y = {y:.2f}")
print(f"z = {z:.2f}")
```
