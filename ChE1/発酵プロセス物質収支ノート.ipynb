# 発酵プロセスの物質収支


# 詳細 (Details)

すべての収支は**定常状態**で行われるものとする。したがって各成分について蓄積はゼロであり、

$$
\text{入力}+\text{生成}=\text{出力}+\text{消費}
$$

が成り立つ。

## 記号一覧

| 記号 | 意味 | 値・単位 |
|--------|--------|--------|
| $\dot{m}_{j,i}$ | 流れ $j$ 中の成分 $i$ の質量流量 | kg/h |
| $\dot{m}_{j}$ | 流れ $j$ の全質量流量 | kg/h |
| $\dot{m}_Y$ | 混合器へ供給される酵母流量（流れSY） | kg/h |
| $\xi$ | 発酵反応の反応進行度 | mol/h |
| $X$ | 反応器におけるスクロース転化率 | 0.95 |
| $f_Y$ | 酵母添加率 | 0.05 |
| $R$ | エタノール回収率 | 0.99 |
| $x_E$ | 留出液中のエタノール質量分率 | 0.956 |
| $M_S$ | スクロースの分子量 | 342.30 g/mol |
| $M_W$ | 水の分子量 | 18.015 g/mol |
| $M_E$ | エタノールの分子量 | 46.07 g/mol |
| $M_C$ | 二酸化炭素の分子量 | 44.01 g/mol |

## 発酵反応
反応器では次の発酵反応が進行する。
$$
\mathrm{C_{12}H_{22}O_{11}+H_2O 
\rightarrow 4C_2H_5OH+4CO_2}
$$

## 混合器 (Mixer)
混合器では、

- スクロース供給流
- 水供給流
- 酵母供給流

を混合する。

酵母供給量は、スクロースおよび水の供給量の合計に対する一定割合として与えられる。
$$
\dot m_Y=f_Y(\dot m_{1,S}+\dot m_{2,W})
$$
混合器内では反応は起こらないため、各成分はそのまま通過する。
$$
\dot m_{3,S}=\dot m_{1,S}
$$

$$
\dot m_{3,W}=\dot m_{2,W}
$$

$$
\dot m_{3,Y}=\dot m_Y
$$

## 反応器 (Reactor)
スクロースを制限反応物とし、転化率を XXX とする。

反応進行度は
$$
\xi=X\frac{\dot m_{3,S}}{M_S}
$$
で与えられる。

反応式より、

スクロース 1 mol 消費
水 1 mol 消費
エタノール 4 mol 生成
二酸化炭素 4 mol 生成

である。

したがって反応器出口の成分流量は

- スクロース
$$
\dot m_{4,S}=(1-X)\dot m_{3,S}
$$
- 水
$$
\dot m_{4,W}=\dot m_{3,W}-\xi M_W
$$
- エタノール
$$
\dot m_{4,E}=4\xi M_E
$$
- 二酸化炭素
$$
\dot m_{7,C}=4\xi M_C
$$
- 酵母
$$
\dot m_{4,Y}=\dot m_{3,Y}
$$
酵母は不活性成分として扱い、反応には関与しない。

また、生成した二酸化炭素はすべてベント流 S7 から排出されるものとする。
***
## 蒸留塔 (Distillation Column)
蒸留塔は、

- エタノール回収率 RRR
- 留出液中のエタノール質量分率 xEx_ExE​

によって規定される。

### エタノール収支
$$
\dot m_{5,E}=R\dot m_{4,E}
$$

### 留出液中の水
$$
\dot m_{5,W}=\dot m_{5,E}\frac{1-x_E}{x_E}
$$
スクロースおよび酵母は不揮発性と仮定する。

したがって、缶底液流量は
$$
\dot m_{6,i}=\dot m_{4,i}-\dot m_{5,i}
$$
で与えられる。

すなわち、蒸留塔入口から留出液として除かれなかった成分はすべて缶底液へ移行する。

## 全体物質収支

プロセス全体については、
$$
\dot m_1+\dot m_2+\dot m_Y=\dot m_5+\dot m_6+\dot m_7
$$
が成り立つ。

この式は計算結果の妥当性を確認するための全体物質収支式である。
